## 1. Introduction to Regular Expressions

### 1.1 What Regular Expressions Are

A regular expression (regex, or "regexp") is a small, specialized pattern-matching language for describing *shapes* of text — not specific text, but the general structure a valid piece of text must follow. Where an ordinary string search asks "does this exact sequence of characters appear?", a regular expression asks a far more powerful question: "does any sequence of characters appear here that follows this particular structural rule?"

**Real-world analogy:** A normal string search is like looking for the exact phrase "January 5, 2026" inside a document. A regular expression is like asking "find me anything that looks like a date" — a single pattern that matches "January 5, 2026", "03/14/2025", or "2026-01-05" alike, because it describes the *shape* of a date (digits, separators, month names) rather than one literal instance of one.

### 1.2 Why Regular Expressions Exist

Regular expressions exist because an enormous amount of real-world text — log files, phone numbers, emails, product codes, HTML fragments, CSV cells, user input — follows recognizable but variable structural patterns that cannot be searched for with plain equality checks or even `str.find()`. Regex gives a program a compact, standardized way to describe "text that looks like this" once, instead of writing custom character-by-character loops for every new format encountered.

### 1.3 What Problems Regex Solves

| Problem | How Regex Helps |
|---|---|
| Pattern matching | Confirming a piece of text follows a particular structure, e.g. "is this a valid-looking email?" |
| Text searching | Locating occurrences of a pattern inside a larger body of text, even when the exact surrounding text is unknown. |
| Text extraction | Pulling specific pieces of matched text — a date, an ID, an amount — out of a larger string. |
| Text validation | Checking user input or data against an expected format before accepting it. |
| Text replacement | Systematically rewriting all instances of a pattern into a new form. |
| Data cleaning | Stripping out unwanted characters, normalizing whitespace, standardizing inconsistent formats across a dataset. |

### 1.4 Normal String Searching vs Regular Expression Searching

````python
text = 'Contact: ada@example.com or grace@example.org'

# Normal string searching - only finds an EXACT literal substring
print('ada@example.com' in text)   # True, but only for this exact address

# Regular expression searching - finds anything matching the SHAPE of an email
import re
print(re.findall(r'[\w.]+@[\w.]+', text))
````

| Code Line | Explanation |
|---|---|
| `'ada@example.com' in text` | Plain substring membership can only ever confirm the presence of the *exact* text given; it has no concept of "an email-shaped thing". |
| `re.findall(r'[\w.]+@[\w.]+', text)` | The regex pattern describes the general shape of an email address (word characters and dots, an `@`, more word characters and dots), and finds every substring in `text` that matches that shape, regardless of the specific address — producing `['ada@example.com', 'grace@example.org']`. |

### 1.5 Regex in Real-World Applications

Regular expressions are pervasive across professional Python development: validating form input on a web backend, extracting structured fields from unstructured log lines, cleaning inconsistent text columns in a Data Analytics pipeline, tokenizing or normalizing text before feeding it into an NLP model, searching source code, and parsing simple, well-defined text formats that don't warrant a full parser. This handbook builds toward exactly these applications, covered in depth from Part 33 onward, after establishing the full syntax and the `re` module's mechanics.

### 1.6 A First Complete Example

````python
import re

log_line = '2026-01-05 14:32:10 ERROR Failed to connect to database'
match = re.search(r'\d{4}-\d{2}-\d{2}', log_line)
if match:
    print('Date found:', match.group())
````

| Code Line | Explanation |
|---|---|
| `r'\d{4}-\d{2}-\d{2}'` | A pattern meaning "four digits, a hyphen, two digits, a hyphen, two digits" — the general shape of an ISO-style date, matching any specific date written that way. |
| `re.search(...)` | Scans `log_line` for the first place this shape occurs (Part 12). |
| `match.group()` | Retrieves the actual matched text, `'2026-01-05'` (Part 29). |

## 2. Introduction to Python's re Module

### 2.1 What the re Module Is

`re` is Python's built-in standard-library module providing regular-expression support; like `os` and `json`, it requires only `import re` and needs no separate installation. It exposes both convenient module-level functions and, for repeated use, a `Pattern` object produced by `re.compile()` (Part 30).

### 2.2 Regex Patterns as Strings

A regex pattern is written as an ordinary Python string — almost always a **raw string** (Part 28) — whose characters are interpreted by the `re` module according to regex syntax rather than as literal text alone. `r'\d+'` is a pattern string meaning "one or more digits"; it is not itself a match, only a description of what to look for.

### 2.3 Match Objects (Preview)

Most `re` functions that locate a match return a **Match object** (never the matched text directly, except for `findall()` in some cases), which bundles together the matched text, its position, and any captured groups. Match objects are covered fully in Part 29.

### 2.4 Compiled Patterns (Preview)

`re.compile(pattern)` pre-processes a pattern string into a reusable `Pattern` object, which can then call the same operations (`.search()`, `.match()`, `.findall()`, etc.) as methods, avoiding the small overhead of re-parsing the same pattern string repeatedly. Covered fully in Part 30.

### 2.5 Overview of the Major Functions

| Function | Purpose | Returns |
|---|---|---|
| `re.match(pattern, string)` | Checks for a match only at the **start** of the string. | Match object or `None` |
| `re.search(pattern, string)` | Checks for a match **anywhere** in the string, returns the first one found. | Match object or `None` |
| `re.fullmatch(pattern, string)` | Checks whether the **entire** string matches the pattern. | Match object or `None` |
| `re.findall(pattern, string)` | Finds **every** non-overlapping match. | List of strings (or tuples, with groups) |
| `re.finditer(pattern, string)` | Finds every match, one at a time. | Iterator of Match objects |
| `re.split(pattern, string)` | Splits the string wherever the pattern matches. | List of strings |
| `re.sub(pattern, repl, string)` | Replaces matches with replacement text. | New string |
| `re.subn(pattern, repl, string)` | Like `sub()`, but also reports how many replacements were made. | Tuple of (new string, count) |
| `re.compile(pattern)` | Pre-compiles a pattern for reuse. | Pattern object |

### 2.6 A Quick Tour

````python
import re

text = 'Order #4521, Order #7788, Order #1029'

print(re.match(r'Order', text))        # matches at the very start
print(re.search(r'#\d+', text))        # first match anywhere
print(re.findall(r'#\d+', text))       # every match
print(list(re.finditer(r'#\d+', text)))  # every match, as an iterator
print(re.split(r',\s*', text))         # split on comma + optional space
print(re.sub(r'#\d+', '#REDACTED', text))  # replace every match
````

Each of these functions is explored individually, with full syntax, edge cases, and worked examples, in Parts 11 through 18.

## 3. Literal Characters

### 3.1 The Simplest Regex: Exact Text

The simplest possible regex pattern is a sequence of ordinary characters that must appear exactly as written, in exactly that order — no special meaning at all, behaving just like a plain substring search.

````python
import re
text = 'I am learning data science and AI'
print(re.search(r'data', text))
print(re.search(r'Data', text))
````

| Code Line | Explanation |
|---|---|
| `re.search(r'data', text)` | Matches, since the literal substring `'data'` appears in `text`. |
| `re.search(r'Data', text)` | Returns `None`; regex matching is **case-sensitive by default** — `'Data'` with a capital D does not match `'data'` — unless the `re.IGNORECASE` flag (Part 31) is used. |

### 3.2 Matching Words and Phrases

````python
import re
print(re.search(r'Python', 'I love Python programming'))
print(re.search(r'machine learning', 'I study machine learning models'))
````

Multi-word literal patterns match the exact sequence of characters given, spaces included; `r'machine learning'` requires exactly one space between the words, and would not match `'machine  learning'` (two spaces) or `'machine-learning'`.

### 3.3 Matching Sequences of Literal Characters

````python
import re
print(re.findall(r'AI', 'AI is transforming Data Analytics, AI Engineering, and AI research'))
````

`findall()` (Part 14) locates every non-overlapping occurrence of a literal pattern, here returning `['AI', 'AI', 'AI']` — three separate matches.

### 3.4 What Is and Is Not Matched

| Pattern | Text | Match? | Reason |
|---|---|---|---|
| `r'cat'` | `'concatenate'` | Yes | `'cat'` appears as a substring, regardless of surrounding characters. |
| `r'cat'` | `'Cat'` | No | Case-sensitive by default. |
| `r'cat'` | `'ca t'` | No | The literal sequence `c`, `a`, `t` does not appear consecutively. |

**Common Mistake:** Assuming a literal pattern only matches whole words. `r'cat'` matches inside `'concatenate'`, `'category'`, and `'scattered'` just as readily as inside the standalone word `'cat'`; whole-word matching requires word boundaries (`\b`, Part 26).

## 4. Special Characters and Metacharacters

### 4.1 Overview

Metacharacters are characters that carry special meaning in a regex pattern rather than matching themselves literally. Understanding this set of characters — and being able to instantly recognize when one needs escaping — is the single most foundational skill in reading and writing regex.

### 4.2 The Complete Metacharacter Reference Table

| Character | Meaning | Covered In |
|---|---|---|
| `.` | Matches any character except newline (by default). | Part 5 |
| `^` | Anchors to the start of the string (or line, with `re.MULTILINE`). | Part 6 |
| `$` | Anchors to the end of the string (or line, with `re.MULTILINE`). | Part 6 |
| `*` | Zero or more of the preceding element. | Part 9 |
| `+` | One or more of the preceding element. | Part 9 |
| `?` | Zero or one of the preceding element; also marks lazy quantifiers. | Parts 9-10 |
| `{ }` | Specifies an exact count or range of repetitions. | Part 9 |
| `[ ]` | Defines a character class (a set of acceptable characters). | Part 7 |
| `\` | Escapes a metacharacter, or introduces a predefined class (`\d`, `\w`, `\s`) or assertion (`\b`). | Parts 8, 26, 27 |
| `|` | Alternation — matches one of several alternatives. | Part 23 |
| `( )` | Defines a group, for capturing or logical grouping. | Parts 19-21 |

### 4.3 Escaping a Metacharacter to Match It Literally

````python
import re
print(re.search(r'3.14', 'Pi is approximately 3.14'))     # '.' matches ANY character here
print(re.search(r'3\.14', 'Pi is approximately 3.14'))    # '\.' matches a literal dot only
print(re.search(r'3\.14', 'The value 3x14 is different'))  # correctly does NOT match
````

| Code Line | Explanation |
|---|---|
| `r'3.14'` | Because `.` is a metacharacter meaning "any character", this pattern would also incorrectly match `'3x14'`, `'3-14'`, or `'3 14'` — not just a literal decimal point. |
| `r'3\.14'` | Prefixing `.` with a backslash escapes it, telling the engine to match a literal dot character only. |

### 4.4 Common Mistakes With Metacharacters

- Forgetting that `.`, `*`, `+`, `?`, `(`, `)`, `[`, `]`, `{`, `}`, `^`, `$`, `|`, and `\` all carry special meaning, and using them unescaped when a literal match was intended — for example, searching for a literal `$5.00` price with the unescaped pattern `r'$5.00'`, which actually anchors to end-of-string (`$`) and treats `.` as "any character".
- Over-escaping characters that have no special meaning outside a character class, like escaping a literal hyphen `-` outside of `[ ]`, which is harmless but unnecessary.
- Forgetting that inside a character class (`[...]`), most metacharacters — including `.`, `*`, `+`, `?` — lose their special meaning automatically and do not need escaping (Part 7).

## 5. The Dot . Metacharacter

### 5.1 What . Matches

By default, `.` matches exactly one occurrence of **any character except a newline** (`\n`).

````python
import re
print(re.findall(r'c.t', 'cat cot cut c t c\tt'))
````

This returns `['cat', 'cot', 'cut', 'c t', 'c\tt']` — `.` matches the letter in each case, a space, and even a tab character, since all of these are "any character" other than a newline.

### 5.2 What . Does Not Match by Default

````python
import re
print(re.search(r'c.t', 'c\nt'))   # None - '.' does not match a newline by default
````

### 5.3 re.DOTALL — Making . Match Newlines Too

````python
import re
text = 'Start\nEnd'
print(re.search(r'Start.End', text))              # None
print(re.search(r'Start.End', text, re.DOTALL))    # matches, '.' now includes \n
````

The `re.DOTALL` flag (also `re.S`, Part 31) changes `.`'s behavior so it matches every character, including newlines — essential when a pattern must span multiple lines of text, such as matching content between two markers across line breaks.

### 5.4 Matching Unknown or Arbitrary Characters

````python
import re
print(re.findall(r'gr.y', 'grey and gray are both correct spellings'))
````

`.` is commonly used exactly like this: to match a single character whose specific value does not matter, or is genuinely unknown, while everything around it must match exactly.

### 5.5 Common Mistake: Overusing . Instead of a Character Class

````python
import re
# Overly permissive - '.' matches ANYTHING, including digits, symbols, etc.
print(re.findall(r'gr.y', 'gr3y gr!y grzy'))

# More precise - only matches an actual letter
print(re.findall(r'gr[a-z]y', 'gr3y gr!y grzy'))
````

Using `.` when a more specific character class (Part 7) is actually intended is one of the most common regex mistakes, since `.` matches indiscriminately, including digits, punctuation, and whitespace that the pattern's author likely never intended to accept.

## 6. Anchors

### 6.1 ^ — Start of String

````python
import re
print(re.search(r'^Data', 'Data Science is great'))
print(re.search(r'^Science', 'Data Science is great'))
````

`^` matches the position at the very beginning of the string (position zero) rather than matching any character itself; a pattern beginning with `^` only succeeds if the rest of the pattern matches starting from that exact position.

### 6.2 $ — End of String

````python
import re
print(re.search(r'great$', 'Data Science is great'))
print(re.search(r'Data$', 'Data Science is great'))
````

`$` matches the position at the very end of the string (or just before a trailing newline), analogous to `^` but for the opposite end.

### 6.3 Searching Anywhere vs Matching the Beginning vs Matching Completely

````python
import re
text = 'user123'

print(bool(re.search(r'\d+', text)))               # anywhere - True (finds digits inside)
print(bool(re.match(r'\d+', text)))                 # beginning only - False (starts with a letter)
print(bool(re.fullmatch(r'[a-z]+\d+', text)))        # entire string - True
````

| Function/Anchor | Question It Answers |
|---|---|
| `re.search(pattern, s)` | Does this pattern occur *anywhere* in the string? |
| `re.match(pattern, s)` | Does this pattern occur starting at *position 0*? |
| `re.fullmatch(pattern, s)` | Does this pattern account for the *entire* string, start to end? |
| `re.search(r'^pattern$', s)` | Equivalent to `fullmatch()` for single-line strings — explicit anchors on both ends. |

### 6.4 re.MULTILINE — Anchoring Per Line

````python
import re
text = 'apple\nbanana\ncherry'
print(re.findall(r'^\w+', text))                 # only the very first line, by default
print(re.findall(r'^\w+', text, re.MULTILINE))    # every line's start
````

| Code Line | Explanation |
|---|---|
| `re.findall(r'^\w+', text)` | Without `re.MULTILINE`, `^` only matches the true start of the whole string, so only `'apple'` is found. |
| `re.findall(r'^\w+', text, re.MULTILINE)` | With `re.MULTILINE` (also `re.M`, Part 31), `^` and `$` additionally match immediately after and before every `\n`, so this finds `['apple', 'banana', 'cherry']` — the start of every line. |

### 6.5 Validation Use Case: Anchors Prevent Partial Matches

````python
import re
def is_valid_username(name):
    return bool(re.fullmatch(r'[a-zA-Z0-9_]{3,16}', name))

print(is_valid_username('ada_99'))
print(is_valid_username('ada 99'))
print(is_valid_username('a!'))
````

**Common Mistake:** Using `re.search(r'[a-zA-Z0-9_]{3,16}', name)` for validation instead of `fullmatch()` (or `search()` with both `^` and `$`). `search()` alone would return a match even for the invalid input `'$$$ada_99$$$'`, because it only checks that the pattern occurs *somewhere* inside the string, not that the *entire* string conforms to it — a serious and common validation bug.

## 7. Character Classes

### 7.1 Basic Character Classes — [abc]

````python
import re
print(re.findall(r'[aeiou]', 'regular expressions are powerful'))
````

`[abc]` matches exactly **one** character that is any one of `a`, `b`, or `c` — a set of acceptable single characters, not a sequence. `[aeiou]` matches any single vowel.

### 7.2 Character Ranges — [a-z], [A-Z], [0-9]

````python
import re
print(re.findall(r'[a-z]+', 'Data Science 2026'))
print(re.findall(r'[A-Z]+', 'Data Science 2026'))
print(re.findall(r'[0-9]+', 'Data Science 2026'))
````

A hyphen inside a character class, between two characters, defines a **range** covering every character from the first to the second according to their underlying code point ordering — `[a-z]` is every lowercase letter, `[0-9]` every digit.

### 7.3 Combining Multiple Ranges

````python
import re
print(re.findall(r'[a-zA-Z]+', 'Version2 Release26'))
print(re.findall(r'[a-zA-Z0-9]+', 'Version2 Release26'))
````

Multiple ranges (and individual characters) can be combined inside a single character class; `[a-zA-Z0-9]` matches any single letter (either case) or digit.

### 7.4 Negated Character Classes — [^abc]

````python
import re
print(re.findall(r'[^0-9]+', 'abc123def456'))
````

A `^` as the very **first** character inside `[ ]` negates the class, meaning "match any single character that is *not* one of these" — `[^0-9]` matches any character that is not a digit, here returning `['abc', 'def']`.

**Common Mistake:** Confusing the anchor `^` (Part 6, matches a *position*) with the negation `^` (matches based on *exclusion*, but still consumes exactly one character); they share the same symbol but mean something entirely different depending on whether they appear at the start of a pattern/line or as the first character inside `[ ]`.

### 7.5 Metacharacters Lose Their Special Meaning Inside [ ]

````python
import re
print(re.findall(r'[.?!]', 'Wait... really? Yes!'))
````

Inside a character class, `.`, `*`, `+`, `?`, and most other metacharacters are treated as ordinary literal characters automatically — `[.?!]` matches a literal period, question mark, or exclamation point, with no escaping required. (A small number of characters, like `]`, `^`, and `-`, still need care depending on their position within the class.)

### 7.6 Character Classes Reference Table

| Pattern | Matches |
|---|---|
| `[abc]` | `a`, `b`, or `c` |
| `[a-z]` | Any lowercase letter |
| `[A-Z]` | Any uppercase letter |
| `[0-9]` | Any digit |
| `[a-zA-Z]` | Any letter, either case |
| `[a-zA-Z0-9]` | Any letter or digit |
| `[^abc]` | Any character *except* `a`, `b`, or `c` |
| `[.,!?]` | Any one of these literal punctuation marks |

## 8. Predefined Character Classes

### 8.1 Overview

Python's `re` module provides several shorthand escape sequences for extremely common character classes, saving the need to write out `[0-9]` or `[a-zA-Z0-9_]` repeatedly.

| Shorthand | Meaning | Equivalent |
|---|---|---|
| `\d` | Any digit | `[0-9]` (plus additional Unicode digits by default) |
| `\D` | Any non-digit | `[^0-9]` (plus additional Unicode exclusions) |
| `\w` | Any "word" character | `[a-zA-Z0-9_]` (plus Unicode letters by default) |
| `\W` | Any non-word character | `[^a-zA-Z0-9_]` |
| `\s` | Any whitespace character | Space, tab, newline, carriage return, form feed |
| `\S` | Any non-whitespace character | Everything except the above |

### 8.2 \d and \D

````python
import re
print(re.findall(r'\d+', 'Order 4521 shipped on 2026-01-05'))
print(re.findall(r'\D+', 'Order 4521 shipped on 2026-01-05'))
````

### 8.3 \w and \W

````python
import re
print(re.findall(r'\w+', "user_name-42 isn't valid!"))
````

`\w` matches letters, digits, and underscore, but **not** hyphens, apostrophes, or other punctuation — this example returns `['user_name', '42', 'isn', 't', 'valid']`, since the hyphen and apostrophe both break a `\w+` run.

### 8.4 \s and \S

````python
import re
print(re.split(r'\s+', 'data   science\tand\nAI'))
````

`\s+` matches one or more whitespace characters of any kind — spaces, tabs, newlines — making it far more robust than splitting on a literal space alone when input formatting is inconsistent.

### 8.5 Unicode Considerations

By default in Python 3, `\d`, `\w`, and `\s` are **Unicode-aware**: `\w` matches accented and non-Latin letters (like `é` or `Я`) in addition to ASCII letters, and `\d` matches other Unicode decimal digit characters, not only `0`–`9`. The `re.ASCII` flag (Part 31) restricts these shorthands back to ASCII-only behavior when that stricter matching is specifically required (for example, validating that a field contains only ASCII digits for a legacy system).

````python
import re
print(re.findall(r'\w+', 'café résumé'))
print(re.findall(r'\w+', 'café résumé', re.ASCII))
````

### 8.6 Comparing Predefined Classes With Explicit Character Classes

````python
import re
print(re.findall(r'[0-9]+', '42 apples'))   # explicit - ASCII digits only
print(re.findall(r'\d+', '42 apples'))       # Unicode-aware by default
````

`[0-9]` is always strictly ASCII digits, regardless of flags, and is a common, deliberate choice for validating identifiers that must be plain ASCII; `\d` is more concise but Unicode-aware by default, which matters for internationalized text input.

## 9. Quantifiers

### 9.1 Overview

A quantifier controls how many times the immediately preceding element (a single character, a character class, or a group) may repeat for the pattern to match.

| Quantifier | Meaning |
|---|---|
| `*` | Zero or more |
| `+` | One or more |
| `?` | Zero or one |
| `{n}` | Exactly n |
| `{n,}` | n or more |
| `{n,m}` | Between n and m, inclusive |

### 9.2 * — Zero or More

````python
import re
print(re.findall(r'ab*c', 'ac abc abbc abbbc'))
````

`ab*c` matches `a`, followed by zero or more `b`s, followed by `c` — matching all of `'ac'`, `'abc'`, `'abbc'`, and `'abbbc'`.

### 9.3 + — One or More

````python
import re
print(re.findall(r'ab+c', 'ac abc abbc abbbc'))
````

`ab+c` requires **at least one** `b`, so `'ac'` (zero `b`s) no longer matches, while the rest still do.

### 9.4 ? — Zero or One

````python
import re
print(re.findall(r'colou?r', 'color colour colouur'))
````

`colou?r` matches `'color'` (zero `u`) and `'colour'` (one `u`), but not `'colouur'` (two `u`s) — useful for optional single characters, like handling both American and British spellings.

### 9.5 {n} — Exact Repetitions

````python
import re
print(re.findall(r'\d{4}', '2026 26 20266'))
````

`\d{4}` matches exactly four consecutive digits; note it still matches the first four digits of `'20266'` (`'2026'`), since regex does not require the surrounding text to stop there unless anchored.

### 9.6 {n,} — Minimum Repetitions

````python
import re
print(re.findall(r'\d{3,}', '12 123 1234 12345'))
````

`\d{3,}` matches three or more consecutive digits, with no upper limit.

### 9.7 {n,m} — Range of Repetitions

````python
import re
print(re.findall(r'\d{2,4}', '1 12 123 1234 12345'))
````

`\d{2,4}` matches between two and four digits; applied to `'12345'`, it greedily takes the first four digits, `'1234'`.

### 9.8 Quantifiers Applied to Groups

````python
import re
print(re.findall(r'(ab)+', 'ababab abc'))
````

A quantifier immediately following a group `(...)` (Part 19) applies to the *entire group* as a unit, not just the single preceding character — `(ab)+` matches one or more repetitions of the whole sequence `'ab'`.

### 9.9 Quantifiers Reference Table

| Pattern | Meaning | Example Match |
|---|---|---|
| `a*` | zero or more `a` | `''`, `'a'`, `'aaa'` |
| `a+` | one or more `a` | `'a'`, `'aaa'` (not `''`) |
| `a?` | zero or one `a` | `''`, `'a'` (not `'aa'`) |
| `a{3}` | exactly three `a` | `'aaa'` |
| `a{2,}` | two or more `a` | `'aa'`, `'aaaa'` |
| `a{2,4}` | two to four `a` | `'aa'`, `'aaa'`, `'aaaa'` |

## 10. Greedy vs Non-Greedy Matching

### 10.1 Greedy Matching (The Default)

By default, every quantifier is **greedy**: it consumes as many characters as possible while still allowing the overall pattern to match, then backtracks (gives characters back) only if that greedy consumption prevents the rest of the pattern from matching.

````python
import re
html = '<b>bold</b> and <i>italic</i>'
print(re.findall(r'<.+>', html))
````

`.+` greedily consumes as much as it can, and since `>` appears multiple times in the string, the engine ends up matching from the very first `<` all the way to the very *last* `>`, producing the single, unexpectedly long match `'<b>bold</b> and <i>italic</i>'` — almost certainly not what was intended.

### 10.2 Non-Greedy (Lazy) Matching

````python
import re
html = '<b>bold</b> and <i>italic</i>'
print(re.findall(r'<.+?>', html))
````

Adding a `?` immediately after a quantifier (`*?`, `+?`, `??`, `{n,m}?`) makes it **lazy**: it consumes as *few* characters as possible while still allowing the rest of the pattern to match, expanding only when forced to. `<.+?>` now correctly returns `['<b>', '</b>', '<i>', '</i>']`, stopping at the very first `>` each time.

### 10.3 Lazy Quantifier Reference

| Greedy | Lazy | Meaning |
|---|---|---|
| `*` | `*?` | Zero or more, as few as possible |
| `+` | `+?` | One or more, as few as possible |
| `?` | `??` | Zero or one, prefer zero |
| `{n,m}` | `{n,m}?` | Between n and m, as few as possible |

### 10.4 Practical Example: Extracting Quoted Text

````python
import re
text = 'She said "hello" and then "goodbye" to everyone'
print(re.findall(r'"(.+)"', text))     # greedy - one giant match
print(re.findall(r'"(.+?)"', text))    # lazy - each quoted phrase separately
````

The greedy version spans from the first `"` to the very last `"`, incorrectly capturing `'hello" and then "goodbye'` as one block; the lazy version correctly isolates `['hello', 'goodbye']` as two separate matches.

### 10.5 Practical Example: Multiple Delimited Records

````python
import re
records = 'START:1234:END START:5678:END'
print(re.findall(r'START:(.+?):END', records))
````

Lazy quantifiers are the standard tool whenever a pattern must stop at the *nearest* closing delimiter rather than the last one found anywhere in the string — extremely common when parsing repeated, delimited records.

### 10.6 Why a Regex May Match More Text Than Expected

The root cause is almost always a greedy quantifier (typically `.*` or `.+`) combined with a delimiter that also appears later in the string; the fix is either switching to the corresponding lazy quantifier, or, often more robustly, replacing `.` with a more specific character class that explicitly excludes the delimiter itself, such as `[^"]+` instead of `.+?` for quoted text — which is also frequently faster (Part 40).

````python
import re
text = 'She said "hello" and then "goodbye" to everyone'
print(re.findall(r'"([^"]+)"', text))
````

## 11. re.match()

### 11.1 Syntax and Behavior

````python
re.match(pattern, string, flags=0)
````

`re.match()` attempts to match the pattern only at the **very start** (position 0) of the string. If the pattern does not match beginning at that exact position, `re.match()` returns `None`, **even if the pattern would match somewhere later in the string**.

````python
import re
print(re.match(r'\d+', '2026 is the year'))
print(re.match(r'\d+', 'The year is 2026'))
````

The second call returns `None`, since the string does not *start* with digits, even though digits appear later in it.

### 11.2 The Match Object or None

`re.match()` (like `search()` and `fullmatch()`) returns either a Match object (Part 29) if successful, or the singleton `None` if not — always check with `if match:` or `if match is not None:` before calling `.group()` on the result, since calling a method on `None` raises `AttributeError`.

````python
import re
match = re.match(r'\d+', 'no digits here')
if match:
    print(match.group())
else:
    print('No match at the start')
````

### 11.3 re.match() vs re.search()

````python
import re
text = 'Hello, World! 2026'
print(re.match(r'World', text))     # None - 'World' is not at position 0
print(re.search(r'World', text))    # matches - found further into the string
````

### 11.4 re.match() vs re.fullmatch()

````python
import re
print(re.match(r'\d{4}', '2026-01-05'))       # matches - only checks the start
print(re.fullmatch(r'\d{4}', '2026-01-05'))    # None - the whole string isn't just 4 digits
````

`match()` only anchors the *beginning*; anything can follow after the matched portion. `fullmatch()` (Part 13) requires the pattern to account for the entire string.

## 12. re.search()

### 12.1 Syntax and Behavior

````python
re.search(pattern, string, flags=0)
````

`re.search()` scans through the string looking for the **first** location where the pattern matches, anywhere at all — not just at the start.

````python
import re
text = 'The meeting is scheduled for 2026-01-05 at noon'
match = re.search(r'\d{4}-\d{2}-\d{2}', text)
print(match.group())
````

### 12.2 Match Objects and Positions

````python
import re
match = re.search(r'\d{4}-\d{2}-\d{2}', 'Date: 2026-01-05')
print(match.start())
print(match.end())
print(match.span())
````

`start()` and `end()` give the index positions the match occupies within the original string, and `span()` returns both as a tuple — covered fully alongside every other Match object attribute in Part 29.

### 12.3 search() Only Finds the First Match

````python
import re
text = 'IDs: 101, 202, 303'
print(re.search(r'\d+', text).group())   # only '101'
print(re.findall(r'\d+', text))           # every ID
````

To find *every* occurrence rather than just the first, `findall()` (Part 14) or `finditer()` (Part 15) is required instead.

## 13. re.fullmatch()

### 13.1 Syntax and Behavior

````python
re.fullmatch(pattern, string, flags=0)
````

`re.fullmatch()` succeeds only if the pattern matches the **entire** string, from the first character to the last, with nothing left over on either end — the strictest of the three core matching functions.

````python
import re
print(re.fullmatch(r'\d{5}', '12345'))
print(re.fullmatch(r'\d{5}', '12345 '))
print(re.fullmatch(r'\d{5}', '123456'))
````

Only the first call succeeds; a trailing space or an extra digit both cause `fullmatch()` to fail, since neither leaves the entire string accounted for by the pattern.

### 13.2 Difference From match() and search()

````python
import re
text = 'ID-4521'
print(bool(re.match(r'ID-\d+', text)))      # True - matches from the start
print(bool(re.search(r'ID-\d+', text)))     # True - found somewhere
print(bool(re.fullmatch(r'ID-\d+', text)))  # True - accounts for everything

text2 = 'ID-4521-EXTRA'
print(bool(re.match(r'ID-\d+', text2)))      # True - still matches the start
print(bool(re.fullmatch(r'ID-\d+', text2)))  # False - '-EXTRA' is left unmatched
````

### 13.3 Best Use Cases: Whole-Input Validation

````python
import re

def is_valid_product_code(code):
    return bool(re.fullmatch(r'[A-Z]{3}-\d{4}', code))

print(is_valid_product_code('ABC-1234'))
print(is_valid_product_code('ABC-1234-X'))
print(is_valid_product_code('prefix ABC-1234'))
````

`fullmatch()` is the correct, idiomatic choice whenever the goal is confirming that an *entire* piece of input — an ID, a numeric code, a structured field — conforms to a format, rather than merely containing something that looks like it somewhere inside a larger string.

## 14. re.findall()

### 14.1 Syntax and Behavior

````python
re.findall(pattern, string, flags=0)
````

`re.findall()` returns a list of every non-overlapping match of the pattern in the string, scanning left to right.

````python
import re
text = 'Prices: $12, $45, $7'
print(re.findall(r'\$\d+', text))
````

### 14.2 Return Type Without Groups

When the pattern contains no capturing groups, `findall()` returns a list of the full matched **strings**.

````python
import re
print(re.findall(r'\d+', 'a1 b22 c333'))
````

### 14.3 Return Type With One Group

````python
import re
print(re.findall(r'\$(\d+)', 'Prices: $12, $45, $7'))
````

When the pattern contains **exactly one** capturing group, `findall()` returns a list of just that group's matched text for each match — here, the dollar sign is part of the match but excluded from the returned strings, giving `['12', '45', '7']`.

### 14.4 Return Type With Multiple Groups

````python
import re
text = 'Ada:95 Grace:88'
print(re.findall(r'(\w+):(\d+)', text))
````

When the pattern contains **multiple** capturing groups, `findall()` returns a list of **tuples**, one tuple per match, containing each group's text in order: `[('Ada', '95'), ('Grace', '88')]`.

### 14.5 Empty Results

````python
import re
print(re.findall(r'\d+', 'no numbers here'))
````

A pattern with no matches simply returns an empty list `[]`, never `None` — no special check is needed beyond an ordinary falsy/truthy check or `len()`.

### 14.6 Data Cleaning / Data Analytics Example

````python
import re

reviews = [
    "Great product, 5 stars! Would buy again for $29.99",
    "Not worth it, 2 stars, paid $49.99",
]
for review in reviews:
    prices = re.findall(r'\$(\d+\.\d{2})', review)
    ratings = re.findall(r'(\d) stars?', review)
    print(f'Prices: {prices}, Rating: {ratings}')
````

## 15. re.finditer()

### 15.1 Syntax and Behavior

````python
re.finditer(pattern, string, flags=0)
````

`re.finditer()` returns an **iterator** of Match objects, one per match, rather than a list of strings — every piece of information a Match object carries (position, groups, span) remains available for each match, which `findall()` discards down to just the matched text (or group text).

````python
import re
text = 'Order #101 shipped, Order #202 pending'
for match in re.finditer(r'#(\d+)', text):
    print(match.group(), 'at position', match.start())
````

### 15.2 Memory Efficiency for Large Text

Because `finditer()` yields matches lazily, one at a time, rather than building the entire result list up front, it is the more memory-efficient choice when scanning very large text (a multi-gigabyte log file, for instance) for a pattern that may occur an enormous number of times — directly analogous to the line-by-line file-reading pattern from the File Handling handbook's Part 20.

### 15.3 findall() vs finditer()

| Aspect | findall() | finditer() |
|---|---|---|
| Return type | List (eagerly built) | Iterator (lazy) |
| Access to position (`start`/`end`) | No | Yes, via the Match object |
| Access to full Match object | No | Yes |
| Memory use on huge input | Higher (full list in memory) | Lower (one match at a time) |
| Typical use | Quick extraction of matched text/groups | Anything needing position, span, or further Match object detail |

### 15.4 Extracting Groups From finditer()

````python
import re
text = 'Ada:95 Grace:88 Alan:76'
for match in re.finditer(r'(?P<name>\w+):(?P<score>\d+)', text):
    print(match.group('name'), '->', match.group('score'))
````

(Named groups are introduced fully in Part 20; this previews their natural pairing with `finditer()` for readable extraction.)

## 16. re.split()

### 16.1 Syntax and Behavior

````python
re.split(pattern, string, maxsplit=0, flags=0)
````

`re.split()` splits a string wherever the pattern matches, returning a list of the pieces in between, analogous to `str.split()` but using a regex pattern as the delimiter instead of a fixed literal string.

````python
import re
print(re.split(r',\s*', 'apple, banana,cherry,  date'))
````

### 16.2 Splitting on Multiple Possible Delimiters

````python
import re
print(re.split(r'[,;]\s*', 'apple, banana; cherry,date'))
````

A character class as the delimiter pattern allows splitting on any one of several different punctuation marks in a single call, something `str.split()` cannot do at all with one call.

### 16.3 Splitting on Whitespace

````python
import re
print(re.split(r'\s+', 'data   science\tand\nAI'))
````

`re.split(r'\s+', text)` is a robust way to split on "any amount of any kind of whitespace" — more forgiving than `str.split()` with no argument in cases where the exact whitespace pattern varies, though in the specific case of "any whitespace, any amount" plain `text.split()` already does this identically and is simpler; the regex version becomes valuable once the delimiter is more complex than pure whitespace.

### 16.4 maxsplit

````python
import re
print(re.split(r',\s*', 'a, b, c, d', maxsplit=2))
````

`maxsplit` limits the number of splits performed, leaving the remainder of the string intact after that many splits have occurred — here producing `['a', 'b', 'c, d']`, stopping after two splits.

### 16.5 Capturing Groups in the Split Pattern

````python
import re
print(re.split(r'(,)', 'a,b,c'))
print(re.split(r',', 'a,b,c'))
````

If the pattern used for splitting contains a capturing group, the text that matched the delimiter itself is **included** in the result list, interleaved with the surrounding pieces — `['a', ',', 'b', ',', 'c']` — which is a deliberate and occasionally very useful feature for preserving the delimiters that were split on.

### 16.6 Data Processing Example

````python
import re
raw = 'Name: Ada Age: 36 City: London'
fields = re.split(r'\s*\w+:\s*', raw)
print([f for f in fields if f])
````

## 17. re.sub()

### 17.1 Syntax and Behavior

````python
re.sub(pattern, repl, string, count=0, flags=0)
````

`re.sub()` returns a new string with every match of the pattern replaced by `repl`, leaving the original string unmodified (strings are immutable in Python).

````python
import re
text = 'Contact us at 555-1234 or 555-5678'
print(re.sub(r'\d{3}-\d{4}', 'XXX-XXXX', text))
````

### 17.2 Replacement Strings and Backreferences

````python
import re
text = 'Ada Lovelace, Grace Hopper'
print(re.sub(r'(\w+) (\w+)', r'\2 \1', text))
````

Within the `repl` string, `\1`, `\2`, and so on refer to the text captured by the corresponding numbered group in the pattern (Part 19), allowing the replacement to reuse pieces of what was matched — here swapping first and last names.

### 17.3 Replacement Functions

````python
import re
def add_currency(match):
    amount = float(match.group(1))
    return f'${amount:,.2f}'

text = 'Total: 1200.5, Tax: 84'
print(re.sub(r'(\d+\.?\d*)', add_currency, text))
````

`repl` may be a **function** instead of a string; the function receives the Match object for each match and must return the string to substitute in its place — essential whenever the replacement needs computed logic rather than a fixed pattern of backreferences.

### 17.4 count — Limiting Replacements

````python
import re
text = 'one two three two one'
print(re.sub(r'two', '2', text, count=1))
````

`count` limits how many matches are replaced, working from the start of the string; `count=0` (the default) replaces every match.

### 17.5 Practical Example: Standardizing Phone Formats

````python
import re
raw_numbers = ['555.123.4567', '555-123-4567', '(555) 123 4567']
for n in raw_numbers:
    digits = re.sub(r'\D', '', n)
    formatted = re.sub(r'(\d{3})(\d{3})(\d{4})', r'(\1) \2-\3', digits)
    print(formatted)
````

### 17.6 Practical Example: Removing Unwanted Characters

````python
import re
messy = 'Pr0duct_C0de!! #4521 @@special'
cleaned = re.sub(r'[^a-zA-Z0-9\s]', '', messy)
print(cleaned)
````

### 17.7 Practical Example: Reformatting Data

````python
import re
dates = ['01/05/2026', '12/25/2025']
for d in dates:
    iso = re.sub(r'(\d{2})/(\d{2})/(\d{4})', r'\3-\1-\2', d)
    print(iso)
````

## 19. Capturing Groups

### 19.1 What (...) Does

Parentheses `(...)` create a **capturing group**: a sub-part of the overall pattern whose matched text is remembered separately and can be retrieved afterward, independent of the full match.

````python
import re
match = re.search(r'(\d{4})-(\d{2})-(\d{2})', 'Date: 2026-01-05')
print(match.group())
print(match.group(1))
print(match.group(2))
print(match.group(3))
````

### 19.2 Group Numbering

Groups are numbered left to right by the position of their **opening** parenthesis, starting at 1; `group(0)` (or plain `group()`) always refers to the entire match, not any individual group.

````python
import re
match = re.search(r'(\w+)@(\w+)\.(\w+)', 'ada@example.com')
print(match.group(0))
print(match.group(1))
print(match.group(2))
print(match.group(3))
````

### 19.3 groups() — Every Captured Group at Once

````python
import re
match = re.search(r'(\d{4})-(\d{2})-(\d{2})', '2026-01-05')
print(match.groups())
````

`groups()` returns a tuple of every captured group's text, in order — equivalent to manually calling `group(1)`, `group(2)`, `group(3)`, and so on, and collecting the results.

### 19.4 Nested Groups

````python
import re
match = re.search(r'((\d{4})-(\d{2}))-(\d{2})', '2026-01-05')
print(match.groups())
````

Groups may nest; numbering still follows the position of each opening parenthesis, so the outer group here is `1`, and the two groups nested inside it are `2` and `3`, with the final group `4`.

### 19.5 Practical Examples

````python
import re

# Extracting a structured phone number
match = re.search(r'\((\d{3})\) (\d{3})-(\d{4})', 'Call (555) 123-4567')
area, prefix, line = match.groups()
print(area, prefix, line)

# Extracting a full name into first and last
match = re.search(r'(\w+) (\w+)', 'Ada Lovelace')
first, last = match.groups()
print(first, last)
````

## 20. Named Groups

### 20.1 Syntax: (?P<name>...)

````python
import re
match = re.search(r'(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})', '2026-01-05')
print(match.group('year'))
print(match.group('month'))
````

A named group behaves exactly like a numbered capturing group for matching purposes, but can additionally be retrieved by a descriptive name instead of a positional number — `match.group('year')` is far more self-documenting than `match.group(1)`, especially in a pattern with many groups.

### 20.2 Why Named Groups Are Useful

- **Readability:** `match.group('email')` communicates intent immediately; `match.group(3)` does not, and requires counting parentheses in the pattern to verify.
- **Resilience to reordering:** if groups are added, removed, or reordered in the pattern later, code referencing them by name continues to work correctly, while code referencing them by number silently breaks.
- **Self-documenting patterns:** the pattern itself becomes a form of documentation about what each part represents.

### 20.3 groupdict()

````python
import re
match = re.search(r'(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})', '2026-01-05')
print(match.groupdict())
````

`groupdict()` returns every named group's captured text as a dictionary, mapping each group's name to its matched value — extremely convenient for turning a single match directly into structured data.

### 20.4 Realistic Data Extraction Example

````python
import re

log_line = '2026-01-05 14:32:10 ERROR Failed to connect to database'
pattern = re.compile(
    r'(?P<date>\d{4}-\d{2}-\d{2}) (?P<time>\d{2}:\d{2}:\d{2}) '
    r'(?P<level>\w+) (?P<message>.+)'
)
match = pattern.search(log_line)
if match:
    print(match.groupdict())
````

Named groups combined with `groupdict()` are the standard, idiomatic pattern for turning semi-structured text — log lines, structured filenames, form-like input — directly into a Python dictionary in one step, a technique used repeatedly throughout Part 37's log-analysis coverage.

## 21. Non-Capturing Groups

### 21.1 Syntax: (?:...)

````python
import re
print(re.findall(r'(?:Mr|Mrs|Dr)\. \w+', 'Dr. Smith and Mrs. Jones attended'))
````

A non-capturing group groups a sub-pattern together — for applying a quantifier to it, or for alternation (Part 23) — **without** creating a numbered (or named) capture that shows up in `groups()`, `group(n)`, or `findall()`'s tuple output.

### 21.2 Difference From Capturing Groups

````python
import re
# Capturing - the (?:...) prefix is captured as group 1
m1 = re.search(r'(Mr|Mrs|Dr)\. (\w+)', 'Dr. Smith')
print(m1.groups())

# Non-capturing - only the name is captured, as group 1
m2 = re.search(r'(?:Mr|Mrs|Dr)\. (\w+)', 'Dr. Smith')
print(m2.groups())
````

Using `(?:...)` for the title keeps the final `groups()` result focused only on the data actually needed (the name), rather than cluttering it with a group that was only ever needed for grouping/alternation purposes.

### 21.3 When Groups Are Needed Only for Grouping

````python
import re
print(re.findall(r'(?:ab)+', 'ababab xyz ababab'))
````

Here, the group exists purely so `+` can apply to the two-character sequence `'ab'` as a whole; since the matched repeated text itself is what's wanted (via `findall()`'s full-match behavior when there are no *capturing* groups), a non-capturing group keeps the return value as plain matched strings rather than an oddly nested tuple structure.

### 21.4 Capturing vs Non-Capturing — Comparison Table

| Aspect | Capturing `(...)` | Non-Capturing `(?:...)` |
|---|---|---|
| Groups for quantifiers/alternation | Yes | Yes |
| Appears in `groups()` / `group(n)` | Yes | No |
| Affects `findall()` return shape | Yes (switches to tuples with 2+ groups) | No |
| Slight performance benefit | No | Marginally yes (nothing to track) |
| Typical use | The data actually needs to be extracted | Grouping/alternation is structural only |

## 22. Backreferences

### 22.1 What a Backreference Is

A backreference lets a pattern refer back to whatever text a *previous* capturing group already matched **within the same match attempt**, requiring that same exact text to appear again — this is fundamentally different from a replacement-string backreference (Section 17.2), which reuses captured text in the *output*, not within the *matching pattern* itself.

### 22.2 Numeric Backreferences — \1, \2

````python
import re
text = 'the the quick fox jumps jumps over'
print(re.findall(r'\b(\w+) \1\b', text))
````

| Code Line | Explanation |
|---|---|
| `r'\b(\w+) \1\b'` | `(\w+)` captures a word as group 1; `\1` then requires that *exact same text* to appear again immediately after a space — detecting an accidentally repeated word. `\b` (Part 26) ensures whole-word matching on both sides. |

### 22.3 Named Backreferences — (?P=name)

````python
import re
text = 'the the quick fox jumps jumps over'
print(re.findall(r'\b(?P<word>\w+) (?P=word)\b', text))
````

`(?P=name)` is the named-group equivalent of `\1` — it refers back to whatever the group named `name` (defined earlier in the same pattern with `(?P<name>...)`) already matched.

### 22.4 Practical Example: Detecting Repeated Structured Components

````python
import re
# Detect an HTML-like tag that opens and closes with the SAME tag name
text = '<div>content</div> <span>content</span> <div>mismatched</span>'
print(re.findall(r'<(\w+)>.*?</\1>', text))
````

This correctly matches `<div>content</div>` and `<span>content</span>`, but not the mismatched final pair, because `\1` specifically requires the closing tag to repeat the *same* text the opening tag captured.

### 22.5 Backreferences vs Replacement Backreferences — A Common Confusion

| Context | Syntax | Meaning |
|---|---|---|
| Inside the **pattern** itself | `\1`, `(?P=name)` | Require the same text matched by group 1 (or `name`) to occur again, *during matching*. |
| Inside the **replacement string** of `sub()` | `\1`, `\g<name>` | Insert whatever text group 1 (or `name`) captured, *into the output*. |

These look nearly identical but serve entirely different roles depending on which side of the operation they appear on — a frequent point of confusion for learners moving between Part 17 and this part.

## 23. Alternation

### 23.1 The | Operator — OR Conditions

````python
import re
print(re.findall(r'cat|dog|bird', 'I have a cat, a dog, and a fish'))
````

`|` matches whichever of the alternatives given works at that position — `cat|dog|bird` matches any one of the three literal words.

### 23.2 Grouping With Alternation

````python
import re
print(re.findall(r'(?:Mr|Mrs|Dr)\.', 'Mr. Smith, Mrs. Jones, Dr. Lee'))
````

Without a group, `|` has very low precedence and applies to the *entire* surrounding pattern, not just the immediately adjacent piece — `Mr|Mrs\.` would actually mean "`Mr`, or `Mrs.`", not "`Mr.` or `Mrs.`"; wrapping the alternatives in a group (capturing or non-capturing) scopes the `|` correctly.

### 23.3 Order of Alternatives Matters

````python
import re
print(re.findall(r'\d{2}|\d{3}', '123 45'))
print(re.findall(r'\d{3}|\d{2}', '123 45'))
````

The regex engine tries alternatives **left to right** and stops at the first one that succeeds at the current position, even if a *later* alternative would have matched more text — `\d{2}|\d{3}` applied to `'123'` matches only `'12'` (the first alternative succeeds immediately), while `\d{3}|\d{2}` correctly prefers the longer match first. This ordering sensitivity is a common, subtle source of bugs.

### 23.4 Common Precedence Mistakes

````python
import re
# WRONG - intended "cat or dog", accidentally means "ca" then "t or dog"
print(re.findall(r'ca|t or dog', 'cat or dog'))

# RIGHT - alternation properly scoped
print(re.findall(r'cat|dog', 'cat or dog'))
````

**Common Mistake:** Forgetting that `|` splits the pattern at that exact character position across the *entire* expression on each side, not just the nearest word or group, unless parentheses explicitly limit its scope.

## 24. Lookahead Assertions

### 24.1 What a Lookahead Is — Zero-Width Assertions

A lookahead checks whether a given pattern would match at the current position **without actually consuming any characters** — it verifies a condition about what comes next, but the matched text itself does not include what the lookahead checked.

### 24.2 Positive Lookahead — (?=...)

````python
import re
text = 'Price: 100 USD, 250 EUR, 75 GBP'
print(re.findall(r'\d+(?= USD)', text))
````

`\d+(?= USD)` matches one or more digits, but only when they are immediately followed by the literal text `' USD'` — and critically, `' USD'` itself is **not** part of the returned match, since a lookahead consumes zero characters.

### 24.3 Negative Lookahead — (?!...)

````python
import re
text = 'Price: 100 USD, 250 EUR, 75 GBP'
print(re.findall(r'\d+(?! USD)', text))
````

`(?!...)` is the inverse: it requires that the given pattern does **not** match at the current position. This returns every number *not* immediately followed by `' USD'` — note it will actually match `'25'` and `'0'` as separate pieces of `'250'` too where the lookahead check re-evaluates partway through, a subtlety worth testing carefully with realistic input (Part 41).

### 24.4 Practical Password-Strength-Style Example

````python
import re
def has_letter_and_digit(password):
    return bool(re.search(r'(?=.*[a-zA-Z])(?=.*\d)', password))

print(has_letter_and_digit('abc123'))
print(has_letter_and_digit('abcdef'))
````

Multiple lookaheads can be chained at the same position to require several independent conditions simultaneously — a common technique for expressing "must contain a letter AND must contain a digit" without needing the letter and digit in any particular order.

## 25. Lookbehind Assertions

### 25.1 Positive Lookbehind — (?<=...)

````python
import re
text = 'Price: $100, Price: $250'
print(re.findall(r'(?<=\$)\d+', text))
````

`(?<=\$)\d+` matches digits, but only when immediately *preceded* by a literal `$`, which — like a lookahead — is not included in the returned match.

### 25.2 Negative Lookbehind — (?<!...)

````python
import re
text = '$100 discount, 100 regular'
print(re.findall(r'(?<!\$)\b\d+\b', text))
````

`(?<!\$)` requires that the given pattern does **not** immediately precede the current position — here isolating numbers that are *not* preceded by a dollar sign.

### 25.3 Python Regex Limitations on Lookbehind

Unlike lookaheads, Python's `re` module requires lookbehind patterns to be of **fixed width** — `(?<=\$)` (one fixed character) is valid, but `(?<=\$+)` or `(?<=\d{2,4})` (variable-length) raises an error, because the engine must know exactly how far to look back before attempting the match. The third-party `regex` module (not part of the standard library) lifts this restriction, but is outside the scope of the built-in `re` module this handbook focuses on.

````python
import re
try:
    re.compile(r'(?<=\d+)abc')
except re.error as e:
    print('Error:', e)
````

### 25.4 Combining Lookahead and Lookbehind

````python
import re
text = 'temp: 20C, 25F, 30C'
print(re.findall(r'(?<=: )\d+(?=C)', text))
````

Combining both directions allows extracting text that sits precisely between two markers, with neither marker included in the extracted result — an alternative to a capturing group that avoids the marker text ever needing to be discarded afterward.

## 26. Word Boundaries

### 26.1 \b — Word Boundary

````python
import re
print(re.findall(r'\bcat\b', 'cat concatenate cat scattered cat'))
````

`\b` is a zero-width assertion matching the position between a `\w` character and a non-`\w` character (or the very start/end of the string) — it does not match or consume any character itself, only a *position*. `\bcat\b` matches the standalone word `'cat'` but not `'cat'` as a substring inside `'concatenate'` or `'scattered'`, correctly returning only the three standalone occurrences.

### 26.2 \B — Non-Word Boundary

````python
import re
print(re.findall(r'\Bcat\B', 'concatenate cat scattered'))
````

`\B` matches everywhere `\b` does *not* — a position that is **not** a word boundary, useful for deliberately matching a substring only when it is *embedded inside* a larger word, the opposite use case from Section 26.1.

### 26.3 Whole-Word Searching

````python
import re
def contains_whole_word(text, word):
    return bool(re.search(rf'\b{re.escape(word)}\b', text))

print(contains_whole_word('The cat sat', 'cat'))
print(contains_whole_word('concatenate', 'cat'))
````

This is the standard, correct technique for "does this text contain this exact word" (as opposed to merely containing the substring anywhere), which plain `in` or `.find()` cannot express (`re.escape()` is covered in Part 27, needed here since `word` is a variable that could itself contain regex metacharacters).

### 26.4 Relationship to \w

`\b` is defined entirely in terms of `\w` (Part 8): a boundary exists at any point where a `\w` character sits next to a non-`\w` character, or at the very edge of the string next to a `\w` character. This means punctuation, whitespace, and the start/end of the string all count as boundaries, but the transition between two `\w` characters never does.

````python
import re
print(re.findall(r'\bpython\b', "python, Python's python-based"))
````

Note that the apostrophe in `"Python's"` and the hyphen in `"python-based"` both count as non-`\w` characters, so `\b` correctly identifies boundaries around `python` in all three cases here.

## 27. Escaping Characters

### 27.1 Matching Literal Metacharacters

Every character listed in Part 4's metacharacter table needs a backslash before it to be matched as a literal character rather than interpreted with its special regex meaning.

````python
import re
print(re.search(r'3\+4=7', 'The equation 3+4=7 is correct'))
print(re.search(r'50% off', '50% off everything'))
````

Note that `%` is *not* a metacharacter and needs no escaping; only the specific set in Part 4's table (`. ^ $ * + ? { } [ ] \ | ( )`) requires it.

### 27.2 re.escape()

````python
import re
user_input = 'C++ & Python (3.x)'
pattern = re.escape(user_input)
print(pattern)
print(re.search(pattern, 'I know C++ & Python (3.x) well'))
````

`re.escape()` automatically escapes every metacharacter in a given string, returning a pattern that matches that string **literally** — indispensable whenever a search pattern is built from external or user-supplied text whose content is not controlled or known in advance.

### 27.3 When re.escape() Is Useful

````python
import re

def find_literal_occurrences(text, literal_phrase):
    return re.findall(re.escape(literal_phrase), text)

print(find_literal_occurrences('Cost: $5.99 (was $7.99)', '$5.99'))
````

Without `re.escape()`, a literal search term like `'$5.99'` would be misinterpreted — `$` as an end-of-string anchor and `.` as "any character" — producing incorrect or no matches; wrapping any dynamically supplied literal search text in `re.escape()` is a standard defensive habit.

## 28. Raw Strings

### 28.1 Why Regex Patterns Are Written as Raw Strings

````python
print(r'\d+')
print('\d+')
````

A **raw string**, prefixed with `r`, tells Python not to process backslash escape sequences (`\n`, `\t`, `\\`, and so on) at all — every character, including backslashes, is taken completely literally. `r'\d+'` is exactly the two characters `\`, `d`, followed by `+`; the equivalent non-raw string `'\d+'` happens to produce the *same* result here only because `\d` is not a recognized Python escape sequence, so Python leaves it alone with a `DeprecationWarning` in some versions — but this is a fragile coincidence, not a rule to depend on.

### 28.2 The Double-Escaping Problem

````python
# Without a raw string, matching a literal backslash requires FOUR backslashes
pattern_normal = '\\\\d'
# With a raw string, only two are needed
pattern_raw = r'\\d'
print(pattern_normal == pattern_raw)
````

Regex itself already uses `\` to escape metacharacters or introduce shorthand classes; when that pattern is *also* written inside a normal (non-raw) Python string, Python's own string-escaping rules apply first, before the `re` module ever sees the string — meaning a literal backslash for regex purposes (`\\` in regex) would require **four** backslashes in a plain Python string (`'\\\\'`), since Python's string parser first collapses `\\\\` down to `\\`, which `re` then interprets as one literal backslash. Raw strings eliminate this entire layer of double-escaping.

````python
import re
# Matching a literal backslash character in the TEXT being searched
text = 'C:\\Users\\Ada'
print(re.findall(r'\\', text))     # raw string - clear and correct
print(re.findall('\\\\', text))    # equivalent, without raw string - much harder to read
````

### 28.3 Normal Strings vs Raw Strings — Comparison

| Pattern Intent | Raw String | Normal String Equivalent |
|---|---|---|
| One or more digits | `r'\d+'` | `'\\d+'` |
| A literal dot | `r'\.'` | `'\\.'` |
| A literal backslash | `r'\\'` | `'\\\\'` |
| Word boundary | `r'\b'` | `'\\b'` (note: `'\b'` alone is a **backspace** character, a genuine trap) |

### 28.4 A Genuine Trap: \b Without a Raw String

````python
print('\b')        # this is an actual backspace control character, NOT a word boundary!
print(r'\b')        # this is the two characters backslash and b, as intended for regex
````

`'\b'` (no `r` prefix) is a real, recognized Python escape sequence for the backspace control character — entirely unrelated to the regex word-boundary assertion — making this one of the most dangerous non-raw-string mistakes in regex, since the code runs without error but silently matches the wrong thing (or nothing at all).

### 28.5 Limitations of Raw Strings

A raw string still cannot end in a single unescaped backslash (`r'\'` is a syntax error, since Python's parser still uses the backslash to determine whether the closing quote is escaped), and raw strings do not disable *regex's* own escape processing — only Python's string-level escape processing; `r'\d'` is still interpreted by `re` as "any digit", exactly as intended.

## 29. Match Objects

### 29.1 Overview

Every successful `match()`, `search()`, or `fullmatch()` call (and each item yielded by `finditer()`) returns a Match object — a rich, structured result carrying far more than just the matched text.

### 29.2 Core Match Object Methods and Attributes

````python
import re
match = re.search(r'(?P<year>\d{4})-(?P<month>\d{2})', 'Report for 2026-01')

print(match.group())        # entire match
print(match.group(1))       # group 1's text
print(match.groups())       # all groups as a tuple
print(match.groupdict())    # named groups as a dict
print(match.start())        # start index of the whole match
print(match.end())          # end index of the whole match
print(match.span())         # (start, end) tuple
print(match.string)         # the original string that was searched
print(match.re)             # the compiled pattern object used
````

| Attribute/Method | Returns |
|---|---|
| `group()` / `group(0)` | The entire matched text. |
| `group(n)` | The text captured by group `n`. |
| `group('name')` | The text captured by the named group `name`. |
| `groups()` | A tuple of every numbered group's captured text. |
| `groupdict()` | A dict of every *named* group's captured text. |
| `start([n])` | Start index of the whole match, or of group `n`. |
| `end([n])` | End index of the whole match, or of group `n`. |
| `span([n])` | `(start, end)` tuple for the whole match, or group `n`. |
| `.string` | The original string the search was performed on. |
| `.re` | The compiled `Pattern` object that produced this match. |
| `.lastindex` | The integer index of the last group that actually matched (useful with alternation). |
| `.lastgroup` | The *name* of the last group that matched, if it was named. |

### 29.3 start(), end(), and span() for Specific Groups

````python
import re
match = re.search(r'(\d{4})-(\d{2})-(\d{2})', 'Date: 2026-01-05')
print(match.span(1))
print(match.span(2))
print(match.span(3))
````

Passing a group number to `start()`, `end()`, or `span()` gives the position of *that specific group's* match within the original string, rather than the overall match's position.

### 29.4 lastindex and lastgroup

````python
import re
pattern = re.compile(r'(?P<num>\d+)|(?P<word>[a-z]+)')
match = pattern.search('hello')
print(match.lastindex)
print(match.lastgroup)
````

With alternation (Part 23) between multiple groups, only one branch actually participates in any given match; `lastindex`/`lastgroup` identify which one, which is otherwise awkward to determine by checking each group for `None`.

### 29.5 Groups That Did Not Participate

````python
import re
match = re.search(r'(\d+)|([a-z]+)', 'hello')
print(match.groups())
print(match.group(1))
print(match.group(2))
````

A group that did not participate in the successful match (because a different alternative matched instead, or because the group itself was optional and skipped) yields `None` rather than an empty string, which must be handled explicitly (e.g. with `or ''`) before further string operations.

## 30. Compiled Regex Patterns

### 30.1 re.compile()

````python
import re
pattern = re.compile(r'\d{4}-\d{2}-\d{2}')
print(type(pattern))
````

`re.compile()` pre-parses a pattern string into a `Pattern` object, which exposes the same operations (`search`, `match`, `fullmatch`, `findall`, `finditer`, `sub`, `split`) as *methods* on the compiled object, called without needing to pass the pattern string again each time.

### 30.2 Reusing Compiled Patterns

````python
import re

date_pattern = re.compile(r'\d{4}-\d{2}-\d{2}')

logs = [
    '2026-01-05 server started',
    'no date here',
    '2026-01-06 server stopped',
]
for line in logs:
    match = date_pattern.search(line)
    if match:
        print(match.group())
````

### 30.3 Pattern Methods

````python
import re
pattern = re.compile(r'\d+')

print(pattern.search('abc 123'))
print(pattern.match('123 abc'))
print(pattern.fullmatch('123'))
print(pattern.findall('1 22 333'))
print(list(pattern.finditer('1 22 333')))
print(pattern.sub('#', '1 22 333'))
print(pattern.split('a1b22c333d'))
````

Every module-level function from Part 2's overview table (except `re.compile()` itself) has a corresponding method on a `Pattern` object, called as `pattern.search(string)` instead of `re.search(pattern, string)` — the pattern and the string simply swap roles from two separate arguments into an object and its method argument.

### 30.4 Readability Benefits

````python
import re

EMAIL_PATTERN = re.compile(r'[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}')

def extract_emails(text):
    return EMAIL_PATTERN.findall(text)
````

Naming a compiled pattern with a clear, uppercase constant-style name (following ordinary Python naming conventions for module-level constants) makes code that uses it elsewhere immediately clearer about *what* is being matched, without needing to mentally parse the raw pattern string at every call site.

### 30.5 Performance Benefits

Compiling a pattern once and reusing the resulting `Pattern` object avoids Python's internal pattern cache lookup (and, on a cache miss, full re-parsing) on every single call — this matters measurably when the same pattern is applied repeatedly, such as inside a loop processing thousands of lines, and is covered further alongside other performance considerations in Part 40.

````python
import re
import time

pattern_str = r'\d{3}-\d{4}'
text = 'Call 555-1234 now'

compiled = re.compile(pattern_str)

# Both work correctly; re module internally caches recently used
# uncompiled patterns too, but explicit compilation remains clearer
# and more reliably fast for patterns used heavily in a hot loop.
start = time.perf_counter()
for _ in range(100_000):
    compiled.search(text)
print('Compiled:', time.perf_counter() - start)
````

## 31. Regex Flags

### 31.1 Overview

Flags modify how a pattern is interpreted globally, without changing the pattern's text itself. They can be passed as an extra argument to any `re` function, passed to `re.compile()`, or combined with `|` when more than one is needed.

| Flag | Short Form | Purpose |
|---|---|---|
| `re.IGNORECASE` | `re.I` | Case-insensitive matching. |
| `re.MULTILINE` | `re.M` | `^`/`$` match at the start/end of every line, not just the whole string. |
| `re.DOTALL` | `re.S` | `.` also matches newline characters. |
| `re.VERBOSE` | `re.X` | Allows whitespace and comments in the pattern for readability. |
| `re.ASCII` | `re.A` | Restricts `\d`, `\w`, `\s`, `\b` to ASCII-only matching. |

### 31.2 re.IGNORECASE / re.I

````python
import re
print(re.findall(r'python', 'Python PYTHON python', re.IGNORECASE))
````

Case-insensitive matching is essential when the case of the source text cannot be relied upon — user input, free-form text, inconsistent data entry.

### 31.3 re.MULTILINE / re.M

Covered in depth in Section 6.4; changes the meaning of `^` and `$` to be per-line rather than whole-string.

### 31.4 re.DOTALL / re.S

Covered in depth in Section 5.3; makes `.` match newline characters too.

### 31.5 re.VERBOSE / re.X

Covered in full depth in Part 32; allows insignificant whitespace and `#` comments directly inside the pattern string for readability.

### 31.6 re.ASCII / re.A

Covered in Section 8.5; restricts Unicode-aware shorthand classes back to ASCII-only.

### 31.7 Combining Multiple Flags

````python
import re
pattern = re.compile(r'^error', re.IGNORECASE | re.MULTILINE)
text = 'INFO: started\nERROR: failed\nerror: retrying'
print(pattern.findall(text))
````

The `|` operator here is Python's ordinary bitwise OR (see the Operators handbook, Part 8), since flags are implemented as combinable bit values — `re.IGNORECASE | re.MULTILINE` activates both behaviors simultaneously.

### 31.8 Inline Flags

````python
import re
print(re.findall(r'(?i)python', 'Python PYTHON python'))
````

`(?i)` at the start of a pattern activates `re.IGNORECASE` for that pattern specifically, without needing to pass a separate flags argument — useful when a pattern is embedded in a context (like being passed as a plain string to a function that doesn't forward a `flags` parameter) where a separate argument isn't convenient. Scoped inline flags like `(?i:python)` restrict the effect to just part of the pattern in modern Python versions.

## 32. re.VERBOSE and Readable Regex

### 32.1 The Problem: Dense, Unreadable Patterns

````python
import re
# Dense and hard to read at a glance
pattern = re.compile(r'^(\d{3})-(\d{3})-(\d{4})$')
````

Even a moderately complex regex quickly becomes a wall of punctuation that is difficult to read, review, or modify confidently months later — a genuine, common maintainability problem in real codebases.

### 32.2 re.VERBOSE Basics

````python
import re
pattern = re.compile(r'''
    ^(\d{3})    # area code
    -(\d{3})    # exchange
    -(\d{4})$   # line number
''', re.VERBOSE)

print(pattern.match('555-123-4567').groups())
````

Under `re.VERBOSE` (`re.X`), whitespace within the pattern is ignored entirely (unless escaped or inside a character class), and `#` introduces a comment running to the end of the line — exactly like ordinary Python code, letting a complex pattern be broken across multiple lines, indented, and annotated.

### 32.3 Matching Literal Whitespace Under re.VERBOSE

````python
import re
pattern = re.compile(r'''
    \d+ \s      # a number followed by a literal space
    [a-z]+
''', re.VERBOSE)
print(pattern.search('42 apples'))
````

Because ordinary whitespace in the pattern is ignored under `re.VERBOSE`, matching an actual literal space requires an explicit escape sequence (`\s`, or `\ ` with a backslash) rather than a bare space character in the pattern text.

### 32.4 A Complete, Readable re.VERBOSE Example

````python
import re

EMAIL_PATTERN = re.compile(r'''
    ^                       # start of string
    (?P<local>[\w.+-]+)     # local part: letters, digits, ., +, -
    @                       # the literal @ symbol
    (?P<domain>[\w-]+)      # domain name
    \.                      # literal dot
    (?P<tld>[a-zA-Z]{2,})   # top-level domain, e.g. com, org
    $                       # end of string
''', re.VERBOSE)

match = EMAIL_PATTERN.fullmatch('ada.lovelace@example.com')
if match:
    print(match.groupdict())
````

This single pattern, laid out with `re.VERBOSE`, is dramatically easier to review, extend, and hand off to another developer than the equivalent single-line pattern would be — a direct, practical illustration of Part 43's best-practice recommendations.

### 32.5 When to Use re.VERBOSE

`re.VERBOSE` is most valuable for patterns with several distinct logical parts, patterns that will be revisited or maintained by others, and patterns complex enough that a reviewer would otherwise need to mentally "decode" them character by character; a short, simple pattern like `r'\d+'` gains little from it and is fine left as-is.

````python
import re
raw_names = ['  ada LOVELACE ', 'Grace   Hopper', 'alan turing!!']

def clean_name(name):
    name = re.sub(r'[^a-zA-Z\s]', '', name)
    name = re.sub(r'\s+', ' ', name).strip()
    return name.title()

print([clean_name(n) for n in raw_names])
````

### 33.8 Cleaning Product Codes

### 33.10 Regex With pandas

````python
import pandas as pd

df = pd.DataFrame({
    'email': ['ada@example.com', 'invalid-email', 'grace@example.org'],
    'phone': ['555-123-4567', '(555) 987-6543', 'no phone'],
})

print(df['email'].str.contains(r'^[\w.+-]+@[\w-]+\.\w+$', regex=True))
print(df['email'].str.extract(r'^([\w.+-]+)@'))
print(df['phone'].str.replace(r'\D', '', regex=True))
````

| Code Line | Explanation |
|---|---|
| `df['email'].str.contains(pattern, regex=True)` | Applies the regex to every row of the column, returning a boolean Series — the vectorized, per-row equivalent of calling `re.search()` in a loop. |
| `df['email'].str.extract(pattern)` | Extracts the first capturing group's match into a new column (or `NaN` where nothing matches), directly usable to build cleaned or derived columns. |
| `df['phone'].str.replace(pattern, repl, regex=True)` | The vectorized equivalent of `re.sub()` applied across an entire column at once. |

### 33.11 str.extractall()

````python
import pandas as pd

df = pd.DataFrame({'notes': ['call 555-1111 or 555-2222', 'contact 555-3333']})
print(df['notes'].str.extractall(r'(\d{3}-\d{4})'))
````

While `str.extract()` returns only the *first* match per row, `str.extractall()` returns *every* match per row, with a multi-index identifying which row and which occurrence each result came from — the pandas equivalent of choosing `finditer()`/`findall()` over `search()`.

## 34. Regex in Data Analytics

### 34.1 Extracting IDs From Mixed Records

````python
import re
records = ['Customer ID: CUST-00123', 'Order ID: ORD-99887', 'Invalid record']
ids = [re.search(r'[A-Z]+-\d+', r) for r in records]
print([m.group() if m else None for m in ids])
````

### 36.8 When to Combine Regex With Parsers, Schemas, or NLP/AI Tools

## 37. Regex for Log Analysis

### 37.1 Parsing a Standard Log Line

````python
import re

LOG_PATTERN = re.compile(
    r'(?P<date>\d{4}-\d{2}-\d{2}) (?P<time>\d{2}:\d{2}:\d{2}) '
    r'(?P<level>DEBUG|INFO|WARNING|ERROR|CRITICAL) (?P<message>.+)'
)

log_line = '2026-01-05 14:32:10 ERROR Failed to connect to database'
match = LOG_PATTERN.match(log_line)
if match:
    print(match.groupdict())
````

### 37.2 Extracting Timestamps

### 37.3 Filtering by Log Level

````python
import re

def filter_by_level(lines, level):
    pattern = re.compile(rf'\b{level}\b')
    return [line for line in lines if pattern.search(line)]

logs = [
    '2026-01-05 10:00:00 INFO Startup complete',
    '2026-01-05 10:01:15 ERROR Disk almost full',
    '2026-01-05 10:02:00 INFO Health check passed',
]
print(filter_by_level(logs, 'ERROR'))
````

### 37.4 Extracting User IDs From Logs

### 37.5 IP-Address-Shaped Extraction

````python
import re
line = '2026-01-05 10:06:00 INFO request from 192.168.1.42 to /api/v1/data'
match = re.search(r'\b(?:\d{1,3}\.){3}\d{1,3}\b', line)
print(match.group() if match else None)
````

**Important caveat:** this pattern matches text that has the *shape* of an IPv4 address (four dot-separated groups of one to three digits); it does not verify that each group is actually within the valid `0`–`255` range. A stricter pattern can enforce that numeric range directly, but for genuine IP-address validation, Python's built-in `ipaddress` module is the more correct and robust tool — regex is well-suited to *finding* IP-shaped text in a log stream, less so to *fully validating* it (Part 39 returns to this distinction).

### 37.7 Summarizing a Whole Log File

````python
import re
from collections import Counter

def summarize_log(path):
    level_counts = Counter()
    pattern = re.compile(r'\b(DEBUG|INFO|WARNING|ERROR|CRITICAL)\b')
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            match = pattern.search(line)
            if match:
                level_counts[match.group(1)] += 1
    return level_counts

# summarize_log('server.log')
````

This combines the streaming, memory-efficient file-reading pattern from the File Handling handbook's Part 20 with per-line regex matching — the standard shape of a production log-analysis script that must scale to arbitrarily large files.

## 39. Validation With Regex

### 39.1 Controlled-Format Validation: Usernames

````python
import re

def is_valid_username(username):
    return bool(re.fullmatch(r'[a-zA-Z][a-zA-Z0-9_]{2,15}', username))

print(is_valid_username('ada_99'))
print(is_valid_username('9ada'))
print(is_valid_username('ab'))
````

### 39.2 Validating Internal IDs and Product Codes

````python
import re

def is_valid_product_code(code):
    return bool(re.fullmatch(r'[A-Z]{2,4}-\d{4,6}', code))

print(is_valid_product_code('AB-1234'))
print(is_valid_product_code('ABCDE-1234'))
````

### 39.3 Validating Controlled Date Formats

````python
import re

def is_valid_iso_date_format(value):
    return bool(re.fullmatch(r'\d{4}-(0[1-9]|1[0-2])-(0[1-9]|[12]\d|3[01])', value))

print(is_valid_iso_date_format('2026-01-05'))
print(is_valid_iso_date_format('2026-13-40'))
print(is_valid_iso_date_format('2026-02-30'))
````

Note the final example: this pattern correctly rejects month `13` and day `40`, but it still accepts `2026-02-30`, since regex alone cannot easily encode "February never has more than 29 days" — a calendrical rule, not a structural/character-shape rule. This is precisely Section 39.5's point.

### 39.4 Validating Simple Phone Number Formats

````python
import re

def is_valid_us_phone(value):
    return bool(re.fullmatch(r'\(\d{3}\) \d{3}-\d{4}', value))

print(is_valid_us_phone('(555) 123-4567'))
print(is_valid_us_phone('555-123-4567'))
````

### 39.5 Format Validation vs Real-World Verification

**A regex match confirms only that the input's *shape* conforms to an expected pattern — it proves nothing whatsoever about whether the underlying data is real, current, or true.** `is_valid_us_phone('(555) 000-0000')` returns `True` even though that number may not be assigned to anyone; an email pattern matching `nobody@example.com` says nothing about whether that mailbox actually exists or accepts mail; a correctly shaped credit card number pattern says nothing about whether the card is active or has funds. Genuine verification — confirming a phone number is reachable, an email is deliverable, an account is real — requires an entirely separate step (sending a verification code, querying an authoritative external system), which regex cannot provide and should never be assumed to provide.

### 39.6 A Realistic Combined Validation Function

````python
import re

def validate_signup_form(username, email, zip_code):
    errors = []
    if not re.fullmatch(r'[a-zA-Z][a-zA-Z0-9_]{2,15}', username):
        errors.append('Invalid username format')
    if not re.fullmatch(r'[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}', email):
        errors.append('Invalid email format')
    if not re.fullmatch(r'\d{5}', zip_code):
        errors.append('Invalid ZIP code format')
    return errors

print(validate_signup_form('ada_99', 'ada@example.com', '02139'))
print(validate_signup_form('9!', 'not-an-email', 'ABCDE'))
````

## 40. Regex Security and Performance

### 40.1 Catastrophic Backtracking

Certain pattern shapes — most notoriously, **nested or overlapping quantifiers**, such as `(a+)+` or `(a*)*`, or two adjacent unbounded quantifiers that can match overlapping content, such as `(a+)(a+)` — cause the regex engine's backtracking search to explore an exponentially growing number of ways to divide the matched text among the repeated pieces when the overall match ultimately *fails*. On a short input this is invisible; on a moderately longer *non-matching* input, it can make the match attempt take seconds, minutes, or effectively forever.

````python
import re
import time

# DANGEROUS PATTERN - nested quantifiers
dangerous = re.compile(r'(a+)+b')

start = time.perf_counter()
dangerous.search('a' * 25 + '!')   # 25 a's followed by a character that can NEVER match 'b'
print(f'Took {time.perf_counter() - start:.3f} seconds for just 25 characters')
````

Each additional `a` in the input roughly *doubles* the time this pattern takes to fail, which is the hallmark of catastrophic backtracking — a genuine denial-of-service vector, commonly abbreviated **ReDoS** (Regular Expression Denial of Service), if such a pattern is ever applied to untrusted, attacker-controlled input.

### 40.2 Why This Happens (Conceptually)

The regex engine, when a match ultimately fails, backtracks and retries every possible way the inner quantifier could have divided up the matched characters among the outer quantifier's repetitions — for `(a+)+` matching a run of `a`s with no valid `b` at the end, the number of ways to partition `n` characters into one or more non-empty groups grows exponentially with `n`, and the engine tries all of them before giving up.

### 40.3 Untrusted Input and Pattern Complexity

The danger is not the pattern alone, nor the input alone — it is the *combination* of a vulnerable pattern shape applied to input whose length and content is not controlled by the program itself, such as a search box, an uploaded file, or an API parameter. A pattern that is perfectly safe on internally generated, bounded-length data can become a serious availability risk the moment it is applied directly to arbitrary user input.

### 40.4 Safer Regex Practices

| Practice | Why |
|---|---|
| Avoid nested quantifiers like `(a+)+`, `(a*)*`, `(a+)*` | These are the classic catastrophic-backtracking shape. |
| Avoid adjacent quantified groups that can match overlapping content, like `(\w+)(\w+)` | Creates many equivalent ways to split the input, multiplying backtracking. |
| Prefer specific character classes over `.` where possible | `[^"]+` cannot backtrack across a `"` the way `.+?` can be forced to; narrower classes reduce ambiguity. |
| Anchor patterns (`^`, `$`, `fullmatch()`) when validating whole input | Reduces the positions the engine must attempt matching from. |
| Limit input length before matching, for untrusted input | Bounds the worst case regardless of pattern shape. |
| Test patterns against deliberately adversarial input | Long, almost-matching strings are the ones that reveal backtracking blowups. |
| Consider a timeout or a safer alternative for genuinely untrusted, unbounded input | The standard `re` module has no built-in timeout; extreme cases may warrant the third-party `regex` module's timeout support, or restructuring the check entirely. |

### 40.5 Benchmarking Regex Performance

````python
import re
import timeit

pattern = re.compile(r'\d{3}-\d{4}')
text = 'Call 555-1234 for details' * 100

duration = timeit.timeit(lambda: pattern.search(text), number=10_000)
print(f'{duration:.4f} seconds for 10,000 searches')
````

`timeit` is the standard tool for measuring how long a given pattern actually takes against realistic sample input, which is the only reliable way to confirm a pattern's real-world performance rather than guessing from its apparent complexity.

### 40.6 Large Text and Performance

For scanning very large text (Part 37's log-analysis use case), `finditer()` (Part 15) over the whole text is typically far more efficient than manually splitting the text into lines first and re-searching each one, since it avoids the overhead of repeated, separate `search()` calls and lets the engine scan continuously.

## 41. Debugging Regex

### 41.1 A Professional Debugging Process

1. **Define the exact input** — write down concrete, realistic example strings, not a vague description.
2. **Define the expected output** — know precisely what should and should not match before writing the pattern.
3. **Start with a small, simple pattern** — get the core shape matching correctly before adding refinements.
4. **Test incrementally** — add one piece of complexity (a quantifier, a group, an anchor) at a time, re-testing after each addition.
5. **Inspect Match objects directly** — print `.group()`, `.span()`, and `.groups()` rather than only checking truthiness.
6. **Test edge cases explicitly** — empty strings, minimum-length input, maximum-length input, unexpected characters.
7. **Add complexity only when demonstrably necessary** — resist adding defensive pattern complexity for cases that were never actually observed in real data.

### 41.2 A Worked Debugging Example

````python
import re

# Goal: extract a price like "$1,234.56"
text = 'Total: $1,234.56'

# Attempt 1 - too simple, misses the comma
pattern = re.compile(r'\$\d+\.\d{2}')
print(pattern.search(text))   # matches only '$234.56' - wrong!

# Attempt 2 - account for an optional thousands separator
pattern = re.compile(r'\$\d{1,3}(?:,\d{3})*\.\d{2}')
print(pattern.search(text))   # correctly matches '$1,234.56'
````

Working through this incrementally — noticing the first attempt silently matched the *wrong* substring rather than failing outright — is a realistic illustration of why inspecting the actual matched text, not just whether a match object is truthy, matters at every debugging step.

### 41.3 Broken Regex Examples and Fixes

````python
import re

# BROKEN: forgot to escape the dot, matches too much
broken = re.compile(r'file.txt')
print(broken.search('fileXtxt'))   # incorrectly matches!

fixed = re.compile(r'file\.txt')
print(fixed.search('fileXtxt'))    # correctly does not match

# BROKEN: greedy quantifier grabs too much
broken2 = re.compile(r'<.*>')
print(broken2.findall('<a><b>'))   # one giant unwanted match

fixed2 = re.compile(r'<.*?>')
print(fixed2.findall('<a><b>'))    # correct: ['<a>', '<b>']
````

### 41.4 Using re.DEBUG-Style Inspection (Reading Group Structure)

````python
import re
pattern = re.compile(r'(?P<year>\d{4})-(?P<month>\d{2})')
print(pattern.groupindex)
````

`pattern.groupindex` on a compiled pattern reveals a dictionary mapping every *named* group to its numeric index — a quick way to confirm a pattern's group structure matches what the code expects, especially in a pattern complex enough that manually counting parentheses is error-prone.

### 41.5 Isolating Failures With Small Test Strings

````python
import re
pattern = re.compile(r'^\d{3}-\d{4}$')
test_cases = ['555-1234', '5551234', '555-12345', ' 555-1234', '555-1234 ']
for case in test_cases:
    print(repr(case), '->', bool(pattern.fullmatch(case)))
````

Building a small table of representative test strings — valid, invalid, and boundary cases — and checking each individually is far more reliable for isolating exactly where a pattern goes wrong than testing only against one realistic-looking sample string.

## 42. Common Regex Mistakes

### 42.1 Forgetting Raw Strings

**Wrong:**

````python
import re
pattern = re.compile('\d+\s\w+')
````

**Why It Is Wrong:** Without an `r` prefix, Python's own string-escaping rules process `\d`, `\s`, and `\w` first; these happen to pass through unchanged with only a `DeprecationWarning` in modern Python, but the pattern is fragile and, for sequences like `\b`, would be outright incorrect (Section 28.4).

**Correct Code:**

````python
pattern = re.compile(r'\d+\s\w+')
````

**Explanation:** A raw string guarantees every backslash reaches the `re` module exactly as written, with no risk of Python's string parser interfering.

### 42.2 Confusing match() and search()

**Wrong:**

````python
import re
if re.match(r'\d+', 'Total: 42'):
    print('Found a number')   # never prints - '\d+' isn't at position 0
````

**Correct Code:**

````python
if re.search(r'\d+', 'Total: 42'):
    print('Found a number')
````

**Explanation:** `match()` only checks the very start of the string (Part 11); `search()` checks anywhere (Part 12).

### 42.3 Incorrect Anchors

**Wrong:**

````python
import re
def is_valid_code(code):
    return bool(re.search(r'[A-Z]{3}\d{4}', code))  # matches ANYWHERE, not the whole thing

print(is_valid_code('XXXABC1234YYY'))   # incorrectly True
````

**Correct Code:**

````python
def is_valid_code(code):
    return bool(re.fullmatch(r'[A-Z]{3}\d{4}', code))
````

**Explanation:** Validation needs `fullmatch()` (or explicit `^`/`$`), not bare `search()` (Section 6.5).

### 42.4 Overusing .*

**Wrong:**

````python
import re
match = re.search(r'Name: (.*) Age: (.*)', 'Name: Ada Lovelace Age: 36 City: London')
print(match.groups())
````

**Why It Is Wrong:** The first `.*` greedily consumes as much as possible, including part of what was meant for the second group, producing unexpected boundaries between the two captures.

**Correct Code:**

````python
match = re.search(r'Name: (.*?) Age: (\d+)', 'Name: Ada Lovelace Age: 36 City: London')
print(match.groups())
````

**Explanation:** A lazy quantifier, or better, a more specific character class, prevents one greedy group from swallowing content intended for another part of the pattern.

### 42.5 Greedy Matching Problems (Recap)

Covered fully in Part 10; the fix is a lazy quantifier or a negated character class excluding the delimiter.

### 42.6 Incorrect Character Classes

**Wrong:**

````python
import re
# Intended: match a literal hyphen or the range a-z; instead creates confusing range depending on position
pattern = re.compile(r'[a-z-0-9]')
````

**Explanation:** A hyphen's meaning inside `[ ]` depends on its position; placing it at the very start or end of the class (`[-a-z0-9]` or `[a-z0-9-]`), or escaping it (`[a-z\-0-9]`), makes the intent to match a literal hyphen unambiguous.

### 42.7 Forgetting to Escape Metacharacters

Covered fully in Part 27; the fix is `re.escape()` for dynamic text, or manual escaping for known literal metacharacters.

### 42.8 Incorrect Grouping

**Wrong:**

````python
import re
print(re.findall(r'Mr|Mrs\.', 'Mr. Smith'))
````

**Why It Is Wrong:** `|` has very low precedence (Section 23.2); this pattern actually means "`Mr`, or `Mrs.`", not "`Mr.` or `Mrs.`" — so it incorrectly matches just `'Mr'` without the period.

**Correct Code:**

````python
print(re.findall(r'(?:Mr|Mrs)\.', 'Mr. Smith'))
````

### 42.9 Misunderstanding \b

**Wrong:**

````python
import re
print(re.findall(r'\b\d+\b', 'price: $100, discount: 10%'))
````

**Explanation:** This actually works correctly here (`\b` sits between a non-`\w` character like `$` and the digit), but a common related mistake is expecting `\b` to work the same way *inside* a character class (`[\b]`), where it instead means a literal backspace character, not a word boundary — a subtle, easy-to-miss trap.

### 42.10 Poor Alternation Order

Covered fully in Section 23.3; the fix is ordering longer/more specific alternatives before shorter/more general ones when they could otherwise "steal" a shorter match first.

### 42.11 Using Regex Where a Parser Is More Appropriate

Covered fully in Sections 35.6 and 38.5; the fix is recognizing genuinely nested/recursive structure (HTML, JSON, arbitrary code) and reaching for a dedicated parser instead of an increasingly elaborate regex.

### 42.12 Writing Unreadable Giant Patterns

**Wrong:**

## 43. Regex Design Best Practices

| # | Practice |
|---|---|
| 1 | Keep patterns as simple as the problem genuinely requires; resist adding speculative complexity. |
| 2 | Always use raw strings (`r'...'`) for regex patterns. |
| 3 | Compile patterns with `re.compile()` when reused, especially inside loops. |
| 4 | Use named groups (`(?P<name>...)`) where they meaningfully improve clarity over numbered groups. |
| 5 | Write automated tests covering valid input, invalid input, and boundary cases for any pattern used for validation. |
| 6 | Document complex patterns, ideally with `re.VERBOSE` and inline comments. |
| 7 | Prefer specific character classes over `.` wherever the acceptable characters are actually known. |
| 8 | Validate assumptions about input data against real, representative samples before finalizing a pattern. |
| 9 | Consider performance and backtracking risk (Part 40), especially for patterns applied to untrusted or unbounded input. |
| 10 | Use `fullmatch()` (or explicit anchors) for validation; use `search()`/`findall()`/`finditer()` for extraction. |
| 11 | Reach for a dedicated parser or schema validator instead of an ever-growing regex once the structure becomes genuinely nested or recursive. |
| 12 | Prefer non-capturing groups `(?:...)` when a group is needed only for structure, not for extracting its content. |

### 44.1 Project 1 — Email and Text Pattern Extractor

**Problem statement:** Build a tool that extracts every email address, phone number, and URL from an arbitrary block of text.

**Requirements:** Handle multiple emails/phones/URLs per document; ignore malformed near-matches; return structured results.

**Sample input:**

````python
sample_text = '''
Contact Ada at ada.lovelace@example.com or call 555-123-4567.
Visit https://example.com/careers for openings.
Backup contact: grace.hopper@example.org, phone (555) 987-6543.
'''
````

**Full code:**

````python
import re

EMAIL_RE = re.compile(r'[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}')
PHONE_RE = re.compile(r'\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}')
URL_RE = re.compile(r'https?://[^\s,]+')

def extract_contacts(text):
    return {
        'emails': EMAIL_RE.findall(text),
        'phones': PHONE_RE.findall(text),
        'urls': URL_RE.findall(text),
    }

result = extract_contacts(sample_text)
print(result)
````

**Line-by-line explanation:** `EMAIL_RE` matches the standard local-part/`@`/domain/TLD shape. `PHONE_RE` tolerates optional parentheses around the area code and any of a space, dot, or hyphen as a separator, covering the several common formats present in the sample. `URL_RE` matches `http`/`https` followed by any non-whitespace, non-comma run of characters (stopping at sentence-ending commas in this sample text).

**Expected output:** `{'emails': ['ada.lovelace@example.com', 'grace.hopper@example.org'], 'phones': ['555-123-4567', '(555) 987-6543'], 'urls': ['https://example.com/careers']}`

**Edge cases:** International phone formats are not covered by `PHONE_RE` and would need an additional pattern; a trailing period at the end of a sentence could be accidentally included in a URL match if not followed by whitespace.

**Possible improvements:** Add `re.IGNORECASE` awareness for scheme casing (`HTTPS://`), validate extracted emails further with `fullmatch()` against a stricter pattern, deduplicate results with a `set()`.

**Real-world applications:** Contact-scraping tools, customer-support ticket triage, CRM data enrichment.

### 44.2 Project 2 — Data Cleaning System Using Regex

**Problem statement:** Clean a raw customer dataset with inconsistent name casing, extra whitespace, and malformed phone numbers.

**Sample input:**

````python
raw_records = [
    {'name': '  ada LOVELACE ', 'phone': '555.123.4567'},
    {'name': 'grace   hopper', 'phone': '(555) 987-6543'},
    {'name': 'ALAN turing', 'phone': '5559998888'},
]
````

**Full code:**

````python
import re

def clean_name(name):
    name = re.sub(r'\s+', ' ', name).strip()
    return name.title()

def clean_phone(phone):
    digits = re.sub(r'\D', '', phone)
    if len(digits) == 10:
        return re.sub(r'(\d{3})(\d{3})(\d{4})', r'(\1) \2-\3', digits)
    return None

def clean_records(records):
    cleaned = []
    for r in records:
        cleaned.append({
            'name': clean_name(r['name']),
            'phone': clean_phone(r['phone']),
        })
    return cleaned

print(clean_records(raw_records))
````

**Expected output:** Every name properly cased with normalized whitespace; every phone number standardized to `(555) 123-4567` form.

**Edge cases:** A phone number with fewer or more than 10 digits after stripping non-digits returns `None` rather than a malformed result; an empty name string produces an empty cleaned name.

**Possible improvements:** Support international phone formats with a country-code prefix; flag records that fail cleaning for manual review instead of silently discarding.

### 44.3 Project 3 — Sales Transaction Description Analyzer

**Problem statement:** Categorize raw bank/POS transaction description strings by merchant type using pattern matching.

**Sample input:**

````python
transactions = [
    'POS PURCHASE 4521 AMAZON.COM WA',
    'ACH DEBIT NETFLIX.COM CA',
    'POS PURCHASE 8891 STARBUCKS #4521 NY',
    'ATM WITHDRAWAL BRANCH 002',
]
````

**Full code:**

````python
import re

CATEGORY_PATTERNS = {
    'Shopping': re.compile(r'AMAZON|WALMART|TARGET', re.IGNORECASE),
    'Streaming': re.compile(r'NETFLIX|SPOTIFY|HULU', re.IGNORECASE),
    'Food & Drink': re.compile(r'STARBUCKS|MCDONALD|CHIPOTLE', re.IGNORECASE),
    'Cash': re.compile(r'ATM WITHDRAWAL', re.IGNORECASE),
}

def categorize(description):
    for category, pattern in CATEGORY_PATTERNS.items():
        if pattern.search(description):
            return category
    return 'Uncategorized'

for t in transactions:
    print(t, '->', categorize(t))
````

### 44.4 Project 4 — Log File Analyzer

**Problem statement:** Parse a server log file, extracting timestamps, log levels, and messages, and summarize error frequency by hour.

**Sample input:**

````python
sample_log = '''2026-01-05 09:00:01 INFO Service started
2026-01-05 09:15:22 ERROR Connection timeout
2026-01-05 09:47:03 ERROR Disk write failed
2026-01-05 10:02:11 INFO Health check passed
2026-01-05 10:33:45 ERROR Connection timeout'''
````

**Full code:**

````python
import re
from collections import Counter

LOG_PATTERN = re.compile(
    r'(?P<date>\d{4}-\d{2}-\d{2}) (?P<hour>\d{2}):\d{2}:\d{2} '
    r'(?P<level>\w+) (?P<message>.+)'
)

def analyze_log(text):
    error_hours = Counter()
    for line in text.splitlines():
        match = LOG_PATTERN.match(line)
        if match and match.group('level') == 'ERROR':
            error_hours[match.group('hour')] += 1
    return error_hours

print(analyze_log(sample_log))
````

**Expected output:** `Counter({'09': 2, '10': 1})`.

**Edge cases:** Malformed lines that don't match `LOG_PATTERN` at all are silently skipped; multi-line messages (e.g. stack traces) would each need to be handled as a continuation of the previous entry, which this simple line-based approach does not address.

**Possible improvements:** Stream from an actual file with `finditer()`/line iteration for large files (Part 20 of the File Handling handbook); detect and merge multi-line stack traces.

### 44.5 Project 5 — Customer Data Validation Tool

**Problem statement:** Validate a batch of customer signup records against format rules for username, email, and ZIP code, reporting all issues per record.

**Sample input:**

````python
signups = [
    {'username': 'ada_99', 'email': 'ada@example.com', 'zip': '02139'},
    {'username': '9x', 'email': 'not-an-email', 'zip': 'ABCDE'},
]
````

**Full code:**

````python
import re

USERNAME_RE = re.compile(r'[a-zA-Z][a-zA-Z0-9_]{2,15}')
EMAIL_RE = re.compile(r'[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}')
ZIP_RE = re.compile(r'\d{5}')

def validate_signup(record):
    errors = []
    if not USERNAME_RE.fullmatch(record['username']):
        errors.append('Invalid username')
    if not EMAIL_RE.fullmatch(record['email']):
        errors.append('Invalid email')
    if not ZIP_RE.fullmatch(record['zip']):
        errors.append('Invalid ZIP code')
    return errors

for s in signups:
    result = validate_signup(s)
    print(s['username'], '->', result if result else 'Valid')
````

### 44.6 Project 6 — Web Text Cleaner

**Problem statement:** Clean raw scraped webpage text for downstream storage or analysis, stripping tags, entities, and excess whitespace.

**Sample input:**

````python
raw_html_text = '<div class="content"><p>Great deal!&nbsp;Only $19.99</p><br/></div>'
````

**Full code:**

````python
import re

def clean_web_text(raw):
    text = re.sub(r'<[^>]+>', ' ', raw)
    entities = {'&nbsp;': ' ', '&amp;': '&', '&lt;': '<', '&gt;': '>', '&quot;': '"'}
    for entity, replacement in entities.items():
        text = text.replace(entity, replacement)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

print(clean_web_text(raw_html_text))
````

**Expected output:** `'Great deal! Only $19.99'`

**Edge cases:** Numeric HTML entities (`&#39;`) are not covered by the fixed dictionary and would need a separate pattern (`&#\d+;`) plus `chr()` conversion; malformed or unclosed tags could cause `<[^>]+>` to over- or under-match.

**Possible improvements:** Use Python's `html.unescape()` for comprehensive entity decoding instead of a manual dictionary; combine with an actual HTML parser (Section 38.5) for anything beyond simple, flat fragments.

### 44.7 Project 7 — NLP Text Preprocessing Pipeline

**Problem statement:** Build a reusable text-cleaning pipeline preparing raw social-media-style text for a downstream NLP model.

**Sample input:**

````python
posts = [
    "Loving the new AI features!! Check it out: https://example.com/ai #excited",
    "This isn't what I expected... visit http://x.co/abc for details @user123",
]
````

**Full code:**

````python
import re

def preprocess_post(text):
    text = text.lower()
    text = re.sub(r'https?://\S+', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#(\w+)', r'\1', text)
    text = re.sub(r"[^a-z0-9\s']", ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

for p in posts:
    print(preprocess_post(p))
````

**Expected output:** URLs and @-mentions removed, hashtags kept as plain words, punctuation stripped, whitespace normalized, all lowercase.

**Edge cases:** Contractions (`isn't`) are deliberately preserved via the apostrophe in the final character class; a hashtag containing only punctuation after the `#` would be removed entirely along with the `#`.

**Possible improvements:** Add emoji handling; integrate with a proper tokenizer/lemmatizer for the next pipeline stage rather than treating this as the final preprocessing step.

### 44.8 Project 8 — Structured Information Extractor

**Problem statement:** Extract a structured record (name, ID, amount, date) from semi-consistent free-text order confirmations.

**Sample input:**

````python
confirmations = [
    'Order confirmed for Ada Lovelace (ID: CUST-4521) - Amount: $129.99 - Date: 2026-01-05',
    'Order confirmed for Grace Hopper (ID: CUST-7788) - Amount: $59.50 - Date: 2026-01-06',
]
````

**Full code:**

````python
import re

ORDER_PATTERN = re.compile(r'''
    Order\ confirmed\ for\ (?P<name>[\w\s]+?)\s+
    \(ID:\ (?P<customer_id>CUST-\d+)\)\s+
    -\ Amount:\ \$(?P<amount>\d+\.\d{2})\s+
    -\ Date:\ (?P<date>\d{4}-\d{2}-\d{2})
''', re.VERBOSE)

def extract_order(text):
    match = ORDER_PATTERN.search(text)
    return match.groupdict() if match else None

for c in confirmations:
    print(extract_order(c))
````

**Expected output:** A dictionary of `name`, `customer_id`, `amount`, and `date` for each confirmation string.

**Edge cases:** A customer name containing a parenthesis or extra punctuation would need a more tightly bounded `name` group; missing fields (e.g. no amount present) would simply return `None` for the whole match under `search()` with this pattern, requiring a fallback strategy.

**Possible improvements:** Make individual fields optional with `?` and default missing groups to `None` individually rather than failing the whole extraction; feed the pattern's named groups directly into a `pydantic` or `dataclass` model for downstream type safety.

## 45. Interview Questions

### 45.1 Beginner (Q1-Q15)

**Q1. What is a regular expression?** A pattern-matching language describing the *shape* of text rather than one exact literal string.

**Q2. What module does Python provide for regex?** The built-in `re` module.

**Q3. Difference between `re.match()` and `re.search()`?** `match()` only checks the start of the string; `search()` checks anywhere in the string.

**Q4. What does `re.findall()` return when the pattern has no groups?** A list of the full matched strings.

**Q5. What does `.` match by default?** Any single character except a newline.

**Q6. What is a raw string and why is it used for regex?** A string prefixed with `r` that disables Python's own backslash-escape processing, so patterns like `\d` reach `re` unmodified (Part 28).

**Q7. What does `\d` match?** Any digit (Unicode-aware by default).

**Q8. What does `\w` match?** Any word character: letters, digits, and underscore.

**Q9. What does `+` mean as a quantifier?** One or more of the preceding element.

**Q10. What does `?` mean as a quantifier?** Zero or one of the preceding element.

**Q11. What is the difference between `[abc]` and `(abc)`?** `[abc]` is a character class matching one of `a`, `b`, or `c`; `(abc)` is a group matching the literal three-character sequence `'abc'`.

**Q12. What does `re.IGNORECASE` do?** Makes matching case-insensitive.

**Q13. What does `^` mean at the start of a pattern?** Anchors the match to the start of the string (or line, under `re.MULTILINE`).

**Q14. What does an empty `re.findall()` result look like?** An empty list `[]`, never `None`.

**Q15. What is `re.sub()` used for?** Replacing every match of a pattern in a string with replacement text.

### 45.2 Intermediate (Q16-Q30)

**Q16. Difference between `findall()` and `finditer()`?** `findall()` eagerly returns a list of matched text/groups; `finditer()` lazily yields Match objects one at a time, preserving position and full group access (Part 15).

**Q17. What is a capturing group?** A `(...)` sub-pattern whose matched text is remembered and retrievable via `group(n)`/`groups()`.

**Q18. What is a non-capturing group and why use one?** `(?:...)` — groups for structure/alternation without adding to `groups()`; used when the sub-match itself isn't needed.

**Q19. What does `groupdict()` return?** A dictionary mapping every named group's name to its captured text.

**Q20. Greedy vs lazy quantifiers?** Greedy consumes as much as possible before backtracking; lazy (`*?`, `+?`, etc.) consumes as little as possible (Part 10).

**Q21. What does `re.compile()` do and why use it?** Pre-parses a pattern into a reusable `Pattern` object; improves readability and performance for repeated use (Part 30).

**Q22. What is `re.VERBOSE` for?** Allows whitespace and `#` comments inside a pattern for readability (Part 32).

**Q23. What is the difference between `\b` and `\B`?** `\b` matches a word boundary; `\B` matches everywhere that is *not* a word boundary.

**Q24. How do you match a literal `.`?** Escape it: `\.`.

**Q25. What does `re.escape()` do?** Escapes every regex metacharacter in a given string so it matches literally.

**Q26. What does `re.fullmatch()` check that `search()` does not?** That the *entire* string, not just part of it, conforms to the pattern.

**Q27. Why might `re.findall()` return tuples instead of strings?** When the pattern has more than one capturing group; each tuple holds every group's text for one match.

**Q28. What is a backreference?** A reference (`\1`, `(?P=name)`) inside a pattern requiring previously captured text to occur again.

**Q29. What flag makes `^`/`$` match per line instead of only the whole string?** `re.MULTILINE`.

**Q30. What flag makes `.` match newlines too?** `re.DOTALL`.

### 45.3 Advanced (Q31-Q45)

**Q31. What is a lookahead assertion, and is it zero-width?** `(?=...)`/`(?!...)`, checking what follows the current position without consuming characters — yes, zero-width.

**Q32. What is a lookbehind assertion, and what Python-specific limitation applies?** `(?<=...)`/`(?<!...)`; Python's `re` requires lookbehind patterns to be fixed-width, unlike lookaheads.

**Q33. Difference between positive and negative lookahead?** Positive requires the sub-pattern to match ahead; negative requires it to *not* match ahead.

**Q34. What does `\1` mean inside a replacement string passed to `re.sub()`?** Insert the text captured by group 1 into the output.

**Q35. What does `\1` mean inside the pattern itself (not the replacement)?** Require the text already captured by group 1 to occur again at that position during matching.

**Q36. Why does alternation order matter?** The engine tries alternatives left to right and stops at the first success, even if a later one would match more text (Section 23.3).

**Q37. What happens to a capturing group that didn't participate in a match (e.g. due to alternation)?** `group(n)` returns `None` for it.

**Q38. What is `lastindex` on a Match object?** The index of the last group that actually participated in the match.

**Q39. How would you validate that an entire string is exactly five digits?** `re.fullmatch(r'\d{5}', s)`.

**Q40. Why is `str.extract()` different from `str.extractall()` in pandas?** `extract()` returns only the first match per row; `extractall()` returns every match per row with a multi-index.

**Q41. What is the risk of `(a+)+` applied to non-matching input?** Catastrophic backtracking / potential ReDoS — exponential time as input length grows (Part 40).

**Q42. Why prefer `[^"]+` over `.+?` when matching up to a closing quote?** It cannot backtrack across a `"` at all, making it both more precise and typically faster.

**Q43. What does `pattern.groupindex` return on a compiled pattern?** A dict mapping named groups to their numeric index.

**Q44. Can Python regex fully validate arbitrary nested HTML?** No — regex matches only regular languages; nested/recursive structure requires a dedicated parser (Section 38.5).

**Q45. What is the difference between matching a phone number's *format* and *verifying* it is real?** Format validation confirms shape only; real verification requires an external check (e.g. sending an SMS code) that regex cannot perform (Section 39.5).

### 45.4 Expert / Debugging (Q46-Q52)

**Q46. Why might `re.findall(r'\d+(?! USD)', text)` return unexpected partial numbers?** A lookahead re-evaluates at every position, so it can match a *substring* of a longer number when the lookahead condition happens to hold partway through (Section 24.3).

**Q47. Given `re.search(r'<.+>', '<a><b>')`, what does it return and why?** The entire `'<a><b>'`, because `.+` is greedy and there is no reason to stop at the first `>` when a later one also allows the pattern to succeed.

**Q48. How do you fix a pattern that matches too much due to greediness?** Use a lazy quantifier, or better, a character class that explicitly excludes the delimiter.

**Q49. Why does `'\b'` without a raw string behave differently from `r'\b'`?** `'\b'` is Python's own backspace escape sequence; only `r'\b'` reaches `re` as the two characters `\` and `b`, interpreted as a word boundary (Section 28.4).

**Q50. Why might a lookbehind pattern raise an error in Python's `re` module?** Variable-length lookbehind is not supported; the sub-pattern inside `(?<=...)` must be fixed-width (Section 25.3).

**Q51. How would you debug a pattern that unexpectedly matches nothing?** Simplify the pattern down to its smallest working piece, then reintroduce complexity incrementally while testing against concrete sample strings at each step (Part 41).

**Q52. Why is `pattern.groupindex` useful when debugging a large pattern with many groups?** It shows exactly which named groups exist and their numeric positions, avoiding manual parenthesis-counting errors.

*Solution:* `re.findall(r'(?<=\$)\d+', text)`

**E30.** Debug: `re.compile(r'(?<=\d+)abc')` raises an error. Explain and fix.
*Solution:* Variable-width lookbehind isn't allowed in `re`; use a fixed width like `(?<=\d)abc` or restructure with a capturing group and `.group()` slicing instead.

**E31.** Write a `re.VERBOSE` pattern validating a simple product code (3 letters, hyphen, 4 digits) with inline comments.
*Solution:*

````python
re.compile(r'''
    ^[A-Z]{3}   # three letters
    -           # hyphen
    \d{4}$      # four digits
''', re.VERBOSE)
````

**E32.** Use `re.finditer()` to print the start position of every digit run in a string.
*Solution:* `for m in re.finditer(r'\d+', text): print(m.start())`

**E33.** Write a pattern that detects an immediately repeated word (e.g. `'the the'`) using a backreference.
*Solution:* `re.search(r'\b(\w+) \1\b', text)`

**E34.** Use `re.subn()` to redact all digits and report how many were redacted.
*Solution:* `re.subn(r'\d', '#', text)`

**E35.** Write a pattern using a non-capturing group for `'Mr'`/`'Mrs'`/`'Dr'` followed by a captured surname.
*Solution:* `re.search(r'(?:Mr|Mrs|Dr)\. (\w+)', text)`

**E36.** Given `'aaaa'`, explain why `re.match(r'(a+)+b', 'aaaa')` is slow to fail, and rewrite it safely.
*Solution:* Nested quantifiers cause catastrophic backtracking (Part 40); rewrite as `re.match(r'a+b', 'aaaa')`, since the inner grouping added nothing.

**E37.** Extract all key-value pairs like `key=value` from a query string into a dictionary.
*Solution:* `dict(re.findall(r'(\w+)=(\w+)', text))`

**E38.** Write a pattern validating a password requires both a letter and a digit, in any order, using lookaheads.
*Solution:* `re.search(r'(?=.*[a-zA-Z])(?=.*\d)', password)`

### 46.4 Level 4 — Professional

**E39.** Build a function that extracts every ISO date and every US-format date from a mixed document, returning both lists.
*Solution:*

````python
def extract_dates(text):
    iso = re.findall(r'\d{4}-\d{2}-\d{2}', text)
    us = re.findall(r'\d{2}/\d{2}/\d{4}', text)
    return iso, us
````

**E40.** Write a data-cleaning function that strips HTML tags, collapses whitespace, and lowercases input in one pass.
*Solution:*

````python
def clean(text):
    text = re.sub(r'<[^>]+>', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip().lower()
    return text
````

**E41.** Using `pandas`, add a boolean column flagging rows in `df['email']` that are validly formatted.
*Solution:* `df['valid_email'] = df['email'].str.fullmatch(r'[\w.+-]+@[\w-]+\.\w+')`

**E42.** Write a log-parsing function using named groups that returns `None` gracefully for lines that don't match, rather than raising an error.
*Solution:*

````python
def parse_log_line(line):
    m = re.match(r'(?P<date>\S+) (?P<level>\w+) (?P<msg>.+)', line)
    return m.groupdict() if m else None
````

**E45.** Given a large log file, write memory-efficient code (using `finditer()` and file streaming) to count how many lines contain the word `'ERROR'`.
*Solution:*

````python
count = 0
with open('app.log', 'r', encoding='utf-8') as f:
    for line in f:
        if re.search(r'\bERROR\b', line):
            count += 1
````

**E46.** Rewrite a catastrophically backtracking pattern `(\w+\s?)+@` into a safer equivalent.
*Solution:* `\w+(?:\s\w+)*@` avoids the nested-quantifier ambiguity of matching overlapping optional whitespace repeatedly.

**E47.** Write a function using `re.subn()` that normalizes all whitespace runs to a single space and reports how many runs were collapsed.
*Solution:* `re.subn(r'\s+', ' ', text)`

**E48.** Build a validator that checks a controlled internal ID format (`DEPT-####-YYYY`) and returns descriptive errors for each violated rule rather than a single boolean.
*Solution:*

````python
def validate_id(value):
    errors = []
    if not re.match(r'^[A-Z]+', value):
        errors.append('Must start with uppercase department code')
    if not re.search(r'-\d{4}-', value):
        errors.append('Must contain a 4-digit number segment')
    if not re.search(r'-\d{4}$', value):
        errors.append('Must end with a 4-digit year')
    return errors
````

**E49.** Explain, without running it, what `re.findall(r'(\d+)-(\d+)', '10-20 30-40')` returns, and why.
*Solution:* `[('10', '20'), ('30', '40')]` — two capturing groups means `findall()` returns a list of tuples, one tuple per match.

**E50.** Write a complete mini pipeline: read a list of raw transaction strings, extract amount and merchant with named groups, and return a list of dictionaries.
*Solution:*

````python
PATTERN = re.compile(r'(?P<merchant>[A-Z .]+) \$(?P<amount>\d+\.\d{2})')

def parse_transactions(lines):
    results = []
    for line in lines:
        m = PATTERN.search(line)
        if m:
            results.append(m.groupdict())
    return results
````

## 47. Complete Regex Cheat Sheet

### 47.1 Core Functions

| Function | Purpose |
|---|---|
| `re.match(p, s)` | Match at the start of `s` only. |
| `re.search(p, s)` | Find the first match anywhere in `s`. |
| `re.fullmatch(p, s)` | Match the entire string `s`. |
| `re.findall(p, s)` | List of every match (or group tuples). |
| `re.finditer(p, s)` | Iterator of Match objects. |
| `re.split(p, s)` | Split `s` wherever `p` matches. |
| `re.sub(p, repl, s)` | Replace matches with `repl`. |
| `re.subn(p, repl, s)` | Like `sub()`, plus a replacement count. |
| `re.compile(p)` | Precompile a pattern into a `Pattern` object. |
| `re.escape(s)` | Escape all metacharacters in `s` for literal matching. |

### 47.2 Pattern Syntax — Literals and Metacharacters

| Syntax | Meaning |
|---|---|
| `abc` | Literal characters, matched exactly. |
| `.` | Any character except newline (`re.DOTALL` includes newline). |
| `^` | Start of string (or line, with `re.MULTILINE`). |
| `$` | End of string (or line, with `re.MULTILINE`). |
| `\` | Escapes a metacharacter, or introduces a shorthand class/assertion. |

### 47.3 Character Classes

| Syntax | Meaning |
|---|---|
| `[abc]` | One of `a`, `b`, `c` |
| `[a-z]` | Any character in the range |
| `[^abc]` | Any character NOT `a`, `b`, or `c` |
| `\d` / `\D` | Digit / non-digit |
| `\w` / `\W` | Word character / non-word character |
| `\s` / `\S` | Whitespace / non-whitespace |

### 47.4 Quantifiers

| Syntax | Meaning |
|---|---|
| `*` / `*?` | Zero or more, greedy / lazy |
| `+` / `+?` | One or more, greedy / lazy |
| `?` / `??` | Zero or one, greedy / lazy |
| `{n}` | Exactly n |
| `{n,}` | n or more |
| `{n,m}` / `{n,m}?` | Between n and m, greedy / lazy |

### 47.5 Groups

| Syntax | Meaning |
|---|---|
| `(...)` | Capturing group |
| `(?P<name>...)` | Named capturing group |
| `(?:...)` | Non-capturing group |
| `\1`, `\2` | Backreference to numbered group (in pattern) |
| `(?P=name)` | Backreference to named group (in pattern) |
| `\1`, `\g<name>` | Reuse captured text in a `sub()` replacement string |

### 47.6 Alternation, Boundaries, Lookarounds

| Syntax | Meaning |
|---|---|
| `a\|b` | `a` or `b` |
| `\b` | Word boundary |
| `\B` | Non-word boundary |
| `(?=...)` | Positive lookahead |
| `(?!...)` | Negative lookahead |
| `(?<=...)` | Positive lookbehind (fixed-width only) |
| `(?<!...)` | Negative lookbehind (fixed-width only) |

### 47.7 Flags

| Flag | Short | Effect |
|---|---|---|
| `re.IGNORECASE` | `re.I` | Case-insensitive matching |
| `re.MULTILINE` | `re.M` | `^`/`$` match per line |
| `re.DOTALL` | `re.S` | `.` matches newlines too |
| `re.VERBOSE` | `re.X` | Ignore whitespace, allow `#` comments |
| `re.ASCII` | `re.A` | Restrict `\d`/`\w`/`\s`/`\b` to ASCII |

### 47.8 Match Object Quick Reference

| Method/Attribute | Returns |
|---|---|
| `.group()` / `.group(0)` | Full matched text |
| `.group(n)` / `.group('name')` | Group n's / named group's text |
| `.groups()` | Tuple of all numbered groups |
| `.groupdict()` | Dict of all named groups |
| `.start()` / `.end()` / `.span()` | Position(s) of the match |
| `.lastindex` / `.lastgroup` | Index/name of the last participating group |

### 47.9 Professional Practices Quick Reference

| Practice | Reminder |
|---|---|
| Always use raw strings | `r'\d+'`, never `'\d+'` |
| Compile reused patterns | `re.compile()` outside loops |
| Use `fullmatch()` for validation | Not bare `search()` |
| Avoid nested quantifiers | Prevents catastrophic backtracking |
| Prefer specific classes over `.` | `[^"]+` beats `.+?` when a delimiter is known |
| Use `re.VERBOSE` for complex patterns | Readability and maintainability |
| Test with realistic and adversarial input | Confirms correctness and performance |
| Reach for a parser when structure nests | Regex cannot reliably parse recursive structure |

## Conclusion

Regular expressions occupy a curious place in a programmer's toolkit: a small, dense, occasionally cryptic-looking syntax that, once genuinely internalized, becomes one of the fastest ways to answer an enormous range of everyday questions about text — does this look right, where does this occur, what's embedded inside this, and how do I rewrite every instance of it. The syntax in Parts 3 through 32 of this handbook is not merely a checklist to memorize, but a small, composable language: literal characters, character classes, quantifiers, groups, and assertions combine in predictable ways to describe arbitrarily specific shapes of text, from a single digit to a fully validated ISO date.

What separates regex used well from regex used poorly is rarely knowledge of more syntax — it is judgment about *when* to reach for it at all, and precisely how much of it a given problem actually needs. `fullmatch()` over `search()` for validation, a specific character class over a greedy `.+?`, a compiled pattern over a repeatedly re-parsed string, and, just as importantly, a real parser instead of an ever-expanding pattern once the underlying data genuinely nests — these decisions, covered throughout Parts 33 to 43, are what make the difference between a regex that quietly does its job for years and one that becomes an unreadable, fragile liability, or worse, a performance and security risk under adversarial input (Part 40).

Returning to the best-practices table in Part 43 and the one-page quick reference above, while keeping the reasoning behind each syntax element in mind, is the most direct path to genuine fluency with Python regular expressions — fluency that pays off directly across Data Analytics, Data Science, Machine Learning, AI Engineering, and everyday professional Python development alike.
