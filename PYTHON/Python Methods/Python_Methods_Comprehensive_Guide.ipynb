# 1. Introduction: What Is a Method?

## 1.1 Methods vs Functions

A **function** is an independent block of reusable code, defined with `def`, that is called by name on its own: `len(x)`, `print(x)`, `sum(x)`. It does not belong to any object.

A **method** is a function that is bound to an object and is accessed using dot notation on that object: `x.append(5)`, `"hello".upper()`, `my_dict.keys()`. A method always operates on the object it is called from, and Python automatically passes that object into the method as the first argument, conventionally named `self`.

This document deliberately does not re-teach built-in functions such as `len()`, `map()`, `filter()`, or `zip()` in depth (they are covered in a separate reference). Those functions appear here only where needed to demonstrate or explain a method.

| Aspect | Function | Method |
|---|---|---|
| Definition | `def greet(name):` | `def greet(self, name):` (inside a class) |
| Call syntax | `greet("Ada")` | `obj.greet("Ada")` |
| Belongs to an object? | No | Yes, always bound to an instance, class, or module |
| First argument | Whatever is passed | Automatically receives the object (`self`) for instance methods |
| Namespace | Global or local | Lives inside a class or a type's implementation |
| Example | `len("hi")` | `"hi".upper()` |

````python
def shout(text):
    return text.upper() + "!"

print(shout("hello"))
````

**Output:**
```
HELLO!
```

````python
text = "hello"
print(text.upper())
````

**Output:**
```
HELLO
```


| Code Line | Explanation |
|---|---|
| `def shout(text):` | Defines a standalone **function** named `shout` that takes one argument. |
| `text.upper()` | Calls the **method** `upper` that belongs to the string object `text`. Note there is no separate `upper(text)` function in Python; `upper` only exists as a method on string objects. |

## 1.2 How Method Calls Work

When you write `"hello".upper()`, Python:

1. Looks up the type of the object (`str`) to find the `upper` method definition.
2. Creates a **bound method** - a method object with the instance (`"hello"`) already attached as `self`.
3. Calls that bound method, automatically passing `"hello"` as the first argument.

````python
s = "hello"
bound = s.upper          # the bound method itself, not yet called
print(bound)              # shows it is a bound method object
print(bound())            # now it is called
print(str.upper(s))       # same result, called the unbound way through the class
````

**Output:**
```
<built-in method upper of str object at 0x...>
HELLO
HELLO
```


| Code Line | Explanation |
|---|---|
| `s.upper` (no parentheses) | Retrieves the method object bound to `s`, without calling it. Useful for passing a method as a callback, e.g. `map(str.upper, list_of_strings)`. |
| `str.upper(s)` | Calls the method through the class itself, explicitly passing the instance as the first argument. This proves a method is fundamentally a function that lives on the class and takes the instance as its first parameter. |

## 1.3 Objects, Classes, and Where Methods Live

Every value in Python is an object, and every object's type is a class (even `int`, `str`, and `list` are classes). Methods are defined once on the class and shared by every instance; Python does not duplicate the method code for each object.

````python
print(type(5))
print(type("hello"))
print(type([1, 2, 3]))
print((5).__class__ is int)
````

**Output:**
```
<class 'int'>
<class 'str'>
<class 'list'>
True
```


## 1.4 Mutable vs Immutable Objects - Why It Matters for Methods

This distinction affects how almost every method in this guide behaves, so it is introduced early and referenced throughout.

- **Immutable objects** (`str`, `int`, `float`, `bool`, `tuple`, `frozenset`, `bytes`, `complex`) cannot be changed after creation. Every method that appears to "modify" one of these actually returns a **new object**.
- **Mutable objects** (`list`, `dict`, `set`, `bytearray`) can be changed in place. Many of their methods modify the object directly and return `None`.

````python
name = "ada"
result = name.upper()
print(name)      # unchanged - strings are immutable
print(result)    # the new object

numbers = [3, 1, 2]
outcome = numbers.sort()
print(numbers)   # changed in place - lists are mutable
print(outcome)   # None - sort() returns nothing
````

**Output:**
```
ada
ADA
[1, 2, 3]
None
```


| Code Line | Explanation |
|---|---|
| `name.upper()` | Returns a brand-new string; `name` itself is untouched because strings cannot be mutated. |
| `numbers.sort()` | Rearranges the existing list object in memory and returns `None`, which is the standard convention for in-place mutating methods in Python. |

**Rule of thumb used throughout this guide:** if a method's name describes an action performed *on* the object with no obvious new value implied (`sort`, `append`, `reverse`, `update`, `clear`), it very likely mutates in place and returns `None`. If the method's name describes producing something (`sorted`-like conversions, `upper`, `split`, `union`), it very likely returns a new object and leaves the original unchanged.

# 2. Instance Methods, Class Methods, and Static Methods

## 2.1 Instance Methods

An **instance method** is the ordinary kind of method: it takes `self` as its first parameter and can read or change the state of that specific instance.

````python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):          # instance method
        self.balance += amount
        return self.balance

acc = Account("Ada", 100)
print(acc.deposit(50))
print(acc.balance)
````

**Output:**
```
150
150
```


| Code Line | Explanation |
|---|---|
| `def deposit(self, amount):` | `self` refers to the specific `Account` instance the method is called on, giving access to that instance's own `balance`. |
| `acc.deposit(50)` | Python automatically supplies `acc` as `self`; only `50` needs to be passed explicitly. |

## 2.2 Class Methods (`@classmethod`)

A **class method** receives the class itself (conventionally named `cls`) rather than an instance. It is defined with the `@classmethod` decorator. Class methods are commonly used for **alternative constructors** - ways to build an instance other than the default `__init__`.

````python
class Account:
    interest_rate = 0.02

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    @classmethod
    def from_string(cls, data):
        owner, balance = data.split(",")
        return cls(owner, float(balance))

    @classmethod
    def set_interest_rate(cls, rate):
        cls.interest_rate = rate

acc = Account.from_string("Grace,250")
print(acc.owner, acc.balance)

Account.set_interest_rate(0.05)
print(Account.interest_rate)
````

**Output:**
```
Grace 250.0
0.05
```


| Code Line | Explanation |
|---|---|
| `@classmethod` | Marks the method as a class method; Python will pass the class, not an instance, as the first argument. |
| `cls(owner, float(balance))` | Calling `cls(...)` inside the method builds a new instance of whichever class called it, which also makes the method work correctly for subclasses. |
| `Account.from_string(...)` | Class methods are usually called on the class itself, though calling them on an instance also works and still receives the class, not the instance. |

## 2.3 Static Methods (`@staticmethod`) and Their Relationship to Methods

A **static method** is defined inside a class using `@staticmethod`, but it receives **neither** `self` nor `cls`. It behaves like a plain function that has been namespaced inside the class purely for organizational purposes - it is grouped with related functionality but does not need, and cannot access, instance or class state.

````python
class TemperatureUtils:
    @staticmethod
    def celsius_to_fahrenheit(c):
        return c * 9 / 5 + 32

print(TemperatureUtils.celsius_to_fahrenheit(100))

temp_util = TemperatureUtils()
print(temp_util.celsius_to_fahrenheit(0))   # also callable from an instance
````

**Output:**
```
212.0
32.0
```


| Code Line | Explanation |
|---|---|
| `@staticmethod` | Tells Python this function needs no automatic `self`/`cls` argument at all. |
| `TemperatureUtils.celsius_to_fahrenheit(100)` | Called directly on the class, exactly like a regular function, just namespaced under `TemperatureUtils`. |

**Why call it a "method" at all if it does not use `self` or `cls`?** Because it is still accessed through the class/instance using dot notation (`obj.method()`), and Python still wraps it as a descriptor on the class. The defining feature of a static method is that it opts out of receiving the automatic first argument, while remaining part of the class's namespace and lookup (`dir()` will list it, `help()` will document it alongside the other methods).

## 2.4 Comparison Table

| Feature | Instance Method | Class Method | Static Method |
|---|---|---|---|
| Decorator | none | `@classmethod` | `@staticmethod` |
| First parameter | `self` (the instance) | `cls` (the class) | none |
| Can read/change instance state | Yes | No (no instance reference) | No |
| Can read/change class state | Yes (via `self.__class__` or the class name) | Yes (via `cls`) | No (unless it names the class explicitly) |
| Typical use | Normal object behavior | Alternative constructors, class-wide operations | Grouping utility logic that logically belongs with the class |
| Call example | `acc.deposit(50)` | `Account.from_string(data)` | `TemperatureUtils.celsius_to_fahrenheit(0)` |

# 3. Discovering and Inspecting Methods

## 3.1 Using `dir()`

`dir()` lists every attribute and method available on an object, including inherited and dunder methods.

````python
print(dir("hello"))
````

**Output:**
```
['__add__', '__class__', ..., 'capitalize', 'casefold', 'center', ...,
 'upper', 'zfill']
```

````python
# Filter out dunder methods to see only the "normal" public methods
public_methods = [m for m in dir("hello") if not m.startswith("__")]
print(public_methods)
````

**Output:**
```
['capitalize', 'casefold', 'center', 'count', 'encode', 'endswith',
 'expandtabs', 'find', 'format', 'format_map', 'index', 'isalnum', ...]
```


| Code Line | Explanation |
|---|---|
| `dir("hello")` | Returns an alphabetically sorted list of every name defined on the `str` class (and inherited from `object`), so it reveals every method the string in question can call. |
| `[m for m in dir("hello") if not m.startswith("__")]` | A list comprehension that filters out dunder methods so only the everyday public API remains. |

Because methods live on classes, knowing an object's type tells you exactly where to look up its methods.

## 3.4 Inspecting Method Signatures with the `inspect` Module

````python
import inspect

class Greeter:
    def greet(self, name, excited=False):
        punctuation = "!" if excited else "."
        return f"Hello, {name}{punctuation}"

sig = inspect.signature(Greeter.greet)
print(sig)
print(list(sig.parameters))
````

**Output:**
```
(self, name, excited=False)
['self', 'name', 'excited']
```


| Code Line | Explanation |
|---|---|
| `inspect.signature(Greeter.greet)` | Builds a `Signature` object describing the parameters `greet` expects, including defaults. |
| `sig.parameters` | An ordered mapping of parameter names to parameter details (kind, default value, annotation). |

# 4. String (`str`) Methods - A Comprehensive Reference

Strings are **immutable**. Every method on this list returns a **new string**; none of them change the original string object. This single fact governs every example below.

## 4.1 Case-Conversion Methods

````python
s = "Hello World"

print(s.upper())        # HELLO WORLD
print(s.lower())        # hello world
print(s.title())        # Hello World
print(s.capitalize())   # Hello world
print(s.swapcase())     # hELLO wORLD
print("STRASSE".casefold())  # strasse (like lower, but more aggressive for caseless matching)
````

**Output:**
```
HELLO WORLD
hello world
Hello World
Hello world
hELLO wORLD
strasse
```


| Code Line | Explanation |
|---|---|
| `s.title()` | Capitalizes the first letter of every word; note it can misfire on words with apostrophes, e.g. `"it's".title()` produces `"It'S"`. |
| `"STRASSE".casefold()` | `casefold()` is stronger than `lower()` for Unicode-aware equality checks (e.g. German "ß" becomes "ss"), and is the right choice for case-insensitive comparisons, not `lower()`. |

**Common mistake:** assuming `s.upper()` changes `s` itself.

````python
s = "hi"
s.upper()
print(s)   # still "hi" - the return value was discarded
````

**Output:**
```
hi
```


## 4.2 Searching and Testing Methods

````python
text = "The quick brown fox"

print(text.find("quick"))     # 4  (index where found)
print(text.find("slow"))      # -1 (not found, find() never raises)
print(text.index("quick"))    # 4  (same as find, but...)
try:
    text.index("slow")        # ...index() raises ValueError if not found
except ValueError as e:
    print("Error:", e)

print(text.count("o"))        # 2
print(text.startswith("The")) # True
print(text.endswith("fox"))   # True
````

**Output:**
```
4
-1
4
Error: substring not found
2
True
True
```


| Code Line | Explanation |
|---|---|
| `text.find("slow")` | Returns `-1` when the substring is absent, which can silently produce bugs if the result is used as an index without checking. |
| `text.index("slow")` | Behaves like `find()` but raises `ValueError` on failure, which is safer when a missing substring should stop the program. |

````python
print("abc123".isalnum())      # True - letters and digits only
print("abc".isalpha())         # True - letters only
print("123".isdigit())         # True
print("123".isdecimal())       # True
print("Ⅷ".isnumeric())         # True - Roman numeral counts as numeric, not digit
print("  ".isspace())          # True
print("Total".istitle())       # True
print("my_var".isidentifier()) # True - valid as a Python variable name
````

**Output:**
```
True
True
True
True
True
True
True
True
```


**Note on `isdigit`, `isdecimal`, `isnumeric`:** they overlap but are not identical. `isdecimal()` is strictest (only 0-9 style decimal digits), `isdigit()` also allows superscripts, and `isnumeric()` is broadest (also allows fractions and number words in some scripts). For validating typical user-entered numbers, `isdecimal()` is usually the most predictable choice.

## 4.3 Splitting, Joining, and Partitioning

````python
csv_line = "Ada,Grace,Katherine"
names = csv_line.split(",")
print(names)

sentence = "one two   three"
print(sentence.split())          # split() with no argument splits on any whitespace run

print(",".join(names))           # join is called on the separator, not the list

path = "folder/sub/file.txt"
print(path.rpartition("/"))      # splits around the LAST separator
print(path.partition("/"))       # splits around the FIRST separator

log = "line1\nline2\r\nline3"
print(log.splitlines())
````

**Output:**
```
['Ada', 'Grace', 'Katherine']
['one', 'two', 'three']
Ada,Grace,Katherine
('folder/sub', '/', 'file.txt')
('folder', '/', 'sub/file.txt')
['line1', 'line2', 'line3']
```


| Code Line | Explanation |
|---|---|
| `",".join(names)` | `join()` is a string method called on the separator, with the iterable of strings passed as the argument - a very common source of confusion for beginners who expect `names.join(",")`. |
| `path.rpartition("/")` | Returns a 3-tuple `(before, separator, after)` split at the last occurrence - ideal for isolating a filename from a path. |
| `log.splitlines()` | Splits on universal newlines (`\n`, `\r\n`, `\r`) without leaving the newline characters in the result, unlike `split("\n")` which would mishandle `\r\n`. |

**Common mistake:**

````python
names = "Ada,Grace,Katherine"
print(names.join(","))   # WRONG - joins the separator using each element of names as separator
````

**Output:**
```
,Ada,,Grace,,Katherine
```


## 4.4 Stripping and Padding

````python
raw = "   pad me   "
print(f"[{raw.strip()}]")     # [pad me]
print(f"[{raw.lstrip()}]")    # [pad me   ]
print(f"[{raw.rstrip()}]")    # [   pad me]

print("7".zfill(3))            # 007
print("42".rjust(5, "*"))      # ***42
print("42".ljust(5, "*"))      # 42***
print("42".center(6, "-"))     # -42---

print("a\tb\tc".expandtabs(4)) # replaces tabs with spaces up to the next multiple of 4
````

**Output:**
```
[pad me]
[pad me   ]
[   pad me]
007
***42
42***
-42---
a   b   c
```


| Code Line | Explanation |
|---|---|
| `"7".zfill(3)` | Pads a numeric-looking string with leading zeros to reach the target width; correctly handles a leading `+`/`-` sign by keeping it first. |
| `"42".center(6, "-")` | Pads with the given fill character on both sides; if the padding is odd, the extra character goes on the right. |

**Common mistake:** forgetting that `.strip()` only removes characters from the given set at the **edges**, not the middle.

````python
print("xxhelloxx".strip("x"))   # "hello" - fine
print("xxhexxllo".strip("x"))   # "hexxllo" - only outer x's removed
````

**Output:**
```
hello
hexxllo
```


## 4.5 Replacing and Translating

````python
sentence = "cats and cats are great"
print(sentence.replace("cats", "dogs"))          # replaces every occurrence
print(sentence.replace("cats", "dogs", 1))       # count=1 limits to the first occurrence

table = str.maketrans("aeiou", "AEIOU")
print("beautiful day".translate(table))

url = "https://example.com/"
print(url.removeprefix("https://"))
print(url.removesuffix("/"))
````

**Output:**
```
dogs and dogs are great
dogs and cats are great
bEAUtIfUl dAy
example.com/
https://example.com
```


| Code Line | Explanation |
|---|---|
| `str.maketrans("aeiou", "AEIOU")` | Builds a translation table mapping each character in the first string to the corresponding character in the second, for use with `translate()`. |
| `url.removeprefix(...)` / `removesuffix(...)` | Added in Python 3.9; safer than manual slicing because they do nothing (return the string unchanged) if the prefix/suffix is not actually present. |

## 4.6 Formatting

````python
template = "{name} scored {score:.1f}%"
print(template.format(name="Ada", score=91.567))

data = {"name": "Grace", "score": 88.2}
print(template.format_map(data))
````

**Output:**
```
Ada scored 91.6%
Grace scored 88.2%
```


| Code Line | Explanation |
|---|---|
| `template.format(...)` | Substitutes `{name}` and `{score:.1f}` placeholders using keyword arguments; `:.1f` formats the number to one decimal place. |
| `template.format_map(data)` | Same idea as `format()`, but takes a single mapping directly instead of `**kwargs`, which avoids copying the dictionary. |

## 4.8 Common Mistakes with String Methods

| Mistake | Why It Happens | Fix |
|---|---|---|
| Calling `s.upper()` and expecting `s` to change | Strings are immutable; all methods return new strings | Reassign: `s = s.upper()` |
| Using `.split(",")` on data with extra spaces | Leaves stray whitespace inside each piece | Strip each item: `[p.strip() for p in s.split(",")]` |
| Confusing `.find()` and `.index()` | Both find substrings, but behave differently on failure | Use `.find()` if a missing value should return `-1`; use `.index()` if it should raise |
| Calling `names.join(",")` instead of `",".join(names)` | `join()` is a method of the separator string, not the list | Always call `join` on the separator |
| Assuming `.strip()` removes characters anywhere in the string | It only trims from the two ends | Use `.replace()` for interior characters |

# 5. List (`list`) Methods - A Comprehensive Reference

Lists are **mutable**. Most list methods change the list in place and return `None`; a few (like indexing via slicing, which is not a method) return new lists, but the methods themselves mostly mutate.

## 5.1 Adding Elements

````python
fruits = ["apple", "banana"]
fruits.append("cherry")            # add one item to the end
print(fruits)

fruits.extend(["date", "fig"])     # add each item of an iterable
print(fruits)

fruits.insert(1, "avocado")        # insert at a specific index
print(fruits)
````

**Output:**
```
['apple', 'banana', 'cherry']
['apple', 'banana', 'cherry', 'date', 'fig']
['apple', 'avocado', 'banana', 'cherry', 'date', 'fig']
```


| Code Line | Explanation |
|---|---|
| `fruits.append("cherry")` | Adds exactly one element to the end; mutates in place; returns `None`. |
| `fruits.extend([...])` | Iterates over the argument and appends each element individually - different from `append`, which would add the whole list as one nested element. |
| `fruits.insert(1, "avocado")` | Shifts every element from index 1 onward one position to the right, an O(n) operation. |

**Common mistake:**

````python
fruits = ["apple"]
fruits.append(["banana", "cherry"])
print(fruits)          # nested list, probably not what was intended
````

**Output:**
```
['apple', ['banana', 'cherry']]
```


## 5.2 Removing Elements

````python
nums = [10, 20, 30, 20, 40]

nums.remove(20)          # removes the FIRST matching value
print(nums)

popped = nums.pop()      # removes and returns the LAST item by default
print(popped, nums)

popped_at = nums.pop(0)  # removes and returns item at a given index
print(popped_at, nums)

nums.clear()             # removes everything
print(nums)
````

**Output:**
```
[10, 30, 20, 40]
40 [10, 30, 20]
10 [30, 20]
[]
```


| Code Line | Explanation |
|---|---|
| `nums.remove(20)` | Removes only the first occurrence; raises `ValueError` if the value is not present. |
| `nums.pop()` | The only removal method that also **returns** the removed value, making it useful for stack-like (LIFO) behavior. |
| `nums.pop(0)` | Popping from the front works but is O(n) for a list; a `collections.deque` is more efficient for frequent front-removal. |

## 5.3 Searching, Counting, and Ordering

````python
letters = ["b", "a", "d", "c", "a"]

print(letters.index("d"))        # first index of "d"
print(letters.count("a"))        # how many times "a" appears

letters.sort()                    # in-place ascending sort
print(letters)

letters.sort(reverse=True)        # in-place descending sort
print(letters)

words = ["banana", "kiwi", "apple"]
words.sort(key=len)               # sort using a key function
print(words)

letters.reverse()                 # in-place reversal
print(letters)
````

**Output:**
```
2
2
['a', 'a', 'b', 'c', 'd']
['d', 'c', 'b', 'a', 'a']
['kiwi', 'apple', 'banana']
['a', 'a', 'b', 'c', 'd']
```


| Code Line | Explanation |
|---|---|
| `letters.sort()` | Sorts the list **in place** and returns `None` - a very common beginner mistake is writing `letters = letters.sort()`, which sets `letters` to `None`. |
| `words.sort(key=len)` | `key` accepts any callable applied to each element before comparing, here sorting by string length rather than alphabetically. |
| `letters.reverse()` | Reverses element order in place; unrelated to sorting - it does not sort in reverse, it just flips the current order. |

**Common mistake:**

````python
data = [3, 1, 2]
data = data.sort()
print(data)   # None - sort() returns None, this discards the list!
````

**Output:**
```
None
```

Correct approaches: call `data.sort()` on its own line, or use the built-in `sorted(data)` if you need a new sorted list without mutating the original (that built-in function is covered in the separate functions reference).

## 5.4 Copying

````python
original = [1, 2, [3, 4]]
shallow = original.copy()

shallow.append(5)
print(original, shallow)          # top-level change does not affect original

shallow[2].append(99)
print(original, shallow)          # but nested objects are SHARED
````

**Output:**
```
[1, 2, [3, 4]] [1, 2, [3, 4], 5]
[1, 2, [3, 4, 99]] [1, 2, [3, 4, 99], 5]
```


| Code Line | Explanation |
|---|---|
| `original.copy()` | Performs a **shallow copy**: a new outer list is created, but nested mutable objects (like the inner list) are shared references, not duplicated. |
| `shallow[2].append(99)` | Because the nested list is shared, mutating it through `shallow` is visible through `original` too - for a fully independent copy, use `copy.deepcopy()`. |

# 6. Tuple (`tuple`) Methods

Tuples are **immutable** and ordered. Because they cannot change size or content, they support only two methods (everything else that "modifies" a tuple, like slicing or concatenation with `+`, produces a brand-new tuple via operators, not methods).

````python
point = (3, 4, 3, 7, 3)

print(point.count(3))     # how many times 3 appears
print(point.index(7))     # first index of value 7

try:
    point.index(99)        # raises ValueError if not found
except ValueError as e:
    print("Error:", e)
````

**Output:**
```
3
3
Error: tuple.index(x): x not in tuple
```


| Code Line | Explanation |
|---|---|
| `point.count(3)` | Works exactly like `list.count()`; tuples support read-only operations because they never change. |
| `point.index(7)` | Returns the first matching index; like `list.index()`, it raises `ValueError` if the value is absent. |

**Why so few methods?** Any operation that would need to add, remove, sort, or reverse elements would require changing the tuple's contents, which immutability forbids. To "modify" a tuple, you must build a new one:

````python
t = (1, 2, 3)
t = t + (4,)          # concatenation creates a new tuple
t = t[:1] + (99,) + t[2:]  # "replacing" an element via slicing
print(t)
````

**Output:**
```
(1, 99, 3, 4)
```


## 6.1 Summary Table - All `tuple` Methods

| Method | Syntax | Purpose | Parameters | Return Value | Mutates Original? | Example |
|---|---|---|---|---|---|---|
| count | `t.count(x)` | Count occurrences of x | x | int | No | `(1,1,2).count(1)` -> `2` |
| index | `t.index(x, start, end)` | First index of x | x, start, end | int | No | `(1,2,3).index(3)` -> `2` |

# 7. Dictionary (`dict`) Methods

Dictionaries are **mutable**, ordered (insertion order preserved since Python 3.7), and store key-value pairs.

## 7.1 Accessing Values Safely

````python
person = {"name": "Ada", "age": 28}

print(person.get("name"))          # "Ada"
print(person.get("email"))         # None - no KeyError
print(person.get("email", "n/a"))  # "n/a" - custom default

try:
    person["email"]                 # direct indexing raises KeyError if missing
except KeyError as e:
    print("Error:", e)
````

**Output:**
```
Ada
None
n/a
Error: 'email'
```


| Code Line | Explanation |
|---|---|
| `person.get("email")` | Returns `None` instead of raising an error when the key is missing, which is safer than `person["email"]` for optional keys. |
| `person.get("email", "n/a")` | The second argument supplies a custom fallback value used only when the key is absent. |

## 7.2 Adding, Updating, and Removing

````python
person = {"name": "Ada"}

person.update({"age": 28, "city": "London"})   # merge another dict in
print(person)

value = person.setdefault("country", "UK")     # get value, or set+return default if missing
print(value, person)

removed = person.pop("city")                   # remove and return a value
print(removed, person)

person.pop("missing_key", "not found")         # safe pop with a default
print(person.pop("missing_key", "not found"))

last_pair = person.popitem()                   # remove and return the LAST inserted (key, value)
print(last_pair, person)

person.clear()
print(person)
````

**Output:**
```
{'name': 'Ada', 'age': 28, 'city': 'London'}
UK {'name': 'Ada', 'age': 28, 'city': 'London', 'country': 'UK'}
London {'name': 'Ada', 'age': 28, 'country': 'UK'}
not found
('country', 'UK') {'name': 'Ada', 'age': 28}
{}
```


| Code Line | Explanation |
|---|---|
| `person.update({...})` | Merges keys from the argument into `person`; existing keys are overwritten, new keys are added. |
| `person.setdefault("country", "UK")` | If `"country"` exists, returns its current value unchanged; if not, inserts it with the given default **and** returns that default - a single-line "insert if missing" pattern. |
| `person.popitem()` | Removes and returns the most recently inserted key-value pair as a tuple (LIFO order), useful for stack-like dict processing. |

## 7.3 Views: `keys()`, `values()`, `items()`

````python
prices = {"apple": 1.5, "banana": 0.5}

print(prices.keys())
print(prices.values())
print(prices.items())

for name, price in prices.items():
    print(f"{name}: ${price}")

prices["apple"] = 2.0   # views are LIVE - they reflect later changes
print(prices.values())
````

**Output:**
```
dict_keys(['apple', 'banana'])
dict_values([1.5, 0.5])
dict_items([('apple', 1.5), ('banana', 0.5)])
apple: $1.5
banana: $0.5
dict_values([2.0, 0.5])
```


| Code Line | Explanation |
|---|---|
| `prices.keys()` / `.values()` / `.items()` | Return special **view objects**, not plain lists - they are dynamic windows onto the dictionary that automatically reflect any later changes. |
| `prices["apple"] = 2.0` | Because `dict_values` is a live view, the earlier printed `values()` object shows the update without calling `.values()` again. |

## 7.4 Copying and Building

````python
original = {"a": 1, "b": {"nested": True}}
shallow = original.copy()
shallow["a"] = 99
print(original["a"], shallow["a"])          # independent at the top level

shallow["b"]["nested"] = False
print(original["b"]["nested"])              # nested dict is shared - shallow copy limitation

keys = ["x", "y", "z"]
new_dict = dict.fromkeys(keys, 0)
print(new_dict)
````

**Output:**
```
1 99
False
{'x': 0, 'y': 0, 'z': 0}
```


| Code Line | Explanation |
|---|---|
| `original.copy()` | Shallow copy again: top-level keys are independent, but nested mutable values are shared, same caveat as `list.copy()`. |
| `dict.fromkeys(keys, 0)` | A class method that builds a new dictionary from an iterable of keys, all mapped to the same default value - useful for initializing counters. |

**Common mistake:** using a mutable default with `fromkeys()`.

````python
grouped = dict.fromkeys(["a", "b"], [])
grouped["a"].append(1)
print(grouped)     # BOTH keys show the change - same list object is shared!
````

**Output:**
```
{'a': [1], 'b': [1]}
```

Fix: build each list separately, e.g. `{k: [] for k in ["a", "b"]}`.

# 8. Set (`set`) Methods

Sets are **mutable**, unordered, and store only unique, hashable elements.

## 8.1 Adding and Removing Elements

````python
colors = {"red", "green"}
colors.add("blue")
print(colors)

colors.add("red")             # adding a duplicate does nothing
print(colors)

colors.remove("green")        # raises KeyError if absent
print(colors)

colors.discard("purple")      # does nothing if absent, no error
print(colors)

popped = colors.pop()         # removes an ARBITRARY element (sets are unordered)
print(popped, colors)
````

**Output:**
```
{'red', 'green', 'blue'}
{'red', 'green', 'blue'}
{'red', 'blue'}
{'red', 'blue'}
red {'blue'}
```


| Code Line | Explanation |
|---|---|
| `colors.remove("green")` | Raises `KeyError` if the element does not exist - use this when the element's presence is guaranteed or the error is meaningful. |
| `colors.discard("purple")` | Silently does nothing if the element is absent - the safer choice when you are not sure the element exists. |
| `colors.pop()` | Because sets have no defined order, `pop()` removes and returns an arbitrary (implementation-defined) element, not necessarily the "first" or "last" one. |

## 8.2 Set Algebra

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.union(b))                 # all elements from both
print(a.intersection(b))          # elements in both
print(a.difference(b))            # in a but not in b
print(a.symmetric_difference(b))  # in exactly one of the two

print(a.issubset({1, 2, 3, 4, 5}))
print(a.issuperset({1, 2}))
print(a.isdisjoint({100, 200}))
````

**Output:**
```
{1, 2, 3, 4, 5, 6}
{3, 4}
{1, 2}
{1, 2, 5, 6}
True
True
True
```


| Code Line | Explanation |
|---|---|
| `a.union(b)` | Equivalent to the `|` operator; returns a **new** set combining unique elements of both. |
| `a.difference(b)` | Elements present in `a` but absent from `b`; order of operands matters, unlike `union` and `intersection`. |
| `a.isdisjoint({100, 200})` | Returns `True` if the two sets share no elements at all - faster than computing the full intersection just to check emptiness. |

## 8.3 In-Place Set Algebra (Mutating Versions)

````python
a = {1, 2, 3}
b = {3, 4, 5}

a.update(b)                        # in-place union
print(a)

a = {1, 2, 3}
a.intersection_update(b)           # keep only shared elements, in place
print(a)

a = {1, 2, 3}
a.difference_update(b)             # remove elements found in b, in place
print(a)

a = {1, 2, 3}
a.symmetric_difference_update(b)   # keep elements in exactly one set, in place
print(a)
````

**Output:**
```
{1, 2, 3, 4, 5}
{3}
{1, 2}
{1, 2, 4, 5}
```


| Code Line | Explanation |
|---|---|
| `a.update(b)` | The mutating counterpart of `union()` - modifies `a` in place instead of returning a new set. |
| `a.intersection_update(b)` | Keeps only elements common to both sets, discarding everything else from `a`. |

# 16. Iterator and Iterable Methods

## 16.1 Iterables vs Iterators

An **iterable** is anything that can produce an iterator (it defines `__iter__`). An **iterator** is an object that produces values one at a time (it defines both `__iter__` and `__next__`) and remembers its position between calls.

````python
numbers = [10, 20, 30]        # an iterable - has __iter__, but no __next__
it = iter(numbers)             # calling iter() on it produces an ITERATOR
print(type(it))

print(it.__next__())            # dunder form
print(next(it))                 # preferred: the next() built-in calls __next__ for you
print(next(it))

try:
    next(it)                     # exhausted - raises StopIteration
except StopIteration:
    print("No more items")
````

**Output:**
```
<class 'list_iterator'>
10
20
30
No more items
```


| Code Line | Explanation |
|---|---|
| `iter(numbers)` | Calls `numbers.__iter__()` internally, producing a fresh iterator object positioned before the first element. |
| `it.__next__()` | Advances the iterator by one position and returns the value there; calling it directly shows what `next()` does under the hood. |
| `next(it)` | The idiomatic way to advance an iterator - prefer this over calling `__next__()` directly, exactly as `len(x)` is preferred over `x.__len__()`. |
| `StopIteration` | The signal an iterator raises when there is nothing left; this is exactly what powers `for` loops behind the scenes. |

## 16.2 How a `for` Loop Uses These Methods Internally

````python
numbers = [1, 2, 3]

# What "for n in numbers:" does internally:
it = iter(numbers)
while True:
    try:
        n = next(it)
    except StopIteration:
        break
    print(n)
````

**Output:**
```
1
2
3
```


## 16.3 Writing a Custom Iterator Class

````python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self               # the object is its own iterator

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

for n in Countdown(3):
    print(n)
````

**Output:**
```
3
2
1
```


| Code Line | Explanation |
|---|---|
| `def __iter__(self): return self` | Makes the object itself the iterator, which is valid as long as it also implements `__next__`. |
| `def __next__(self):` | Called automatically on each loop iteration; raising `StopIteration` inside it is exactly how the loop knows to stop. |

## 16.4 Generator Methods: `send()`, `throw()`, `close()`

Generators (created with `yield`) are iterators with a few extra methods beyond plain `__next__`.

````python
def echo_generator():
    while True:
        received = yield
        print(f"Got: {received}")

gen = echo_generator()
next(gen)                  # prime the generator to the first yield
gen.send("hello")          # resume, passing a value INTO the generator
gen.send("world")
gen.close()                 # stop the generator early, raises GeneratorExit inside it
````

**Output:**
```
Got: hello
Got: world
```


| Code Line | Explanation |
|---|---|
| `next(gen)` | Advances execution to the first `yield`, which is required before `send()` can be used the first time. |
| `gen.send("hello")` | Resumes the generator and provides a value that becomes the result of the `yield` expression inside it - a two-way communication channel, unlike a plain iterator. |
| `gen.close()` | Raises `GeneratorExit` inside the generator at its current suspension point, allowing cleanup code (e.g. in a `finally` block) to run before it stops permanently. |

| Method | Syntax | Purpose | Parameters | Return Value | Mutates Original? | Example |
|---|---|---|---|---|---|---|
| `__iter__` | `iter(obj)` / `obj.__iter__()` | Return an iterator for obj | none | iterator | No | `iter([1,2])` |
| `__next__` | `next(it)` / `it.__next__()` | Get the next value | none | next value or raises StopIteration | Yes (advances internal state) | `next(iter([1,2]))` -> `1` |
| send | `gen.send(value)` | Resume generator, inject a value | value | next yielded value | Yes | `gen.send(5)` |

# 18. Special (Dunder) Methods Deep Dive

Dunder ("double underscore") methods are how Python hooks user-defined classes into its built-in syntax and functions. You rarely call them directly - you define them, and Python calls them for you when you use an operator, a built-in function, or a language construct.

## 18.1 Object Lifecycle: `__init__`, `__new__`, `__del__`

````python
class Widget:
    def __new__(cls, *args, **kwargs):
        print("1. __new__ creates the raw instance")
        instance = super().__new__(cls)
        return instance

    def __init__(self, name):
        print("2. __init__ initializes the instance")
        self.name = name

    def __del__(self):
        print(f"3. __del__ cleans up {self.name}")

w = Widget("gadget")
del w
````

**Output:**
```
1. __new__ creates the raw instance
2. __init__ initializes the instance
3. __del__ cleans up gadget
```


| Code Line | Explanation |
|---|---|
| `__new__` | The actual constructor - it creates and returns the raw object, before `__init__` ever runs. Almost never overridden except for immutable types or metaclasses. |
| `__init__` | Runs immediately after `__new__` returns; initializes attributes on the already-created instance. This is what people usually mean by "the constructor." |
| `__del__` | Called when the object is about to be garbage-collected; unreliable for critical cleanup (use context managers/`__exit__` instead), but useful for logging or optional cleanup. |

## 18.2 String Representation: `__str__` vs `__repr__`

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __str__(self):
        return f"({self.x}, {self.y})"                  # human-readable

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"          # unambiguous, developer-facing

p = Point(1, 2)
print(p)          # calls __str__
print(str(p))     # calls __str__
print(repr(p))    # calls __repr__
print([p])         # lists always use __repr__ for their elements, never __str__
````

**Output:**
```
(1, 2)
(1, 2)
Point(x=1, y=2)
[Point(x=1, y=2)]
```


| Code Line | Explanation |
|---|---|
| `__str__` | Used by `print()` and `str()`; should be readable and aimed at end users. |
| `__repr__` | Used by `repr()`, the interactive interpreter, and container display; should ideally be valid Python code that could recreate the object. If `__str__` is missing, Python falls back to `__repr__`. |

## 18.3 Container Protocol: `__len__`, `__getitem__`, `__setitem__`, `__delitem__`, `__contains__`

````python
class Roster:
    def __init__(self):
        self._students = []

    def __len__(self):
        return len(self._students)

    def __getitem__(self, index):
        return self._students[index]

    def __setitem__(self, index, value):
        self._students[index] = value

    def __delitem__(self, index):
        del self._students[index]

    def __contains__(self, name):
        return name in self._students

    def add(self, name):
        self._students.append(name)

roster = Roster()
roster.add("Ada")
roster.add("Grace")

print(len(roster))            # calls __len__
print(roster[0])               # calls __getitem__
roster[0] = "Katherine"        # calls __setitem__
print(roster[0])
print("Grace" in roster)       # calls __contains__
del roster[1]                   # calls __delitem__
print(len(roster))
````

**Output:**
```
2
Ada
Katherine
True
1
```


| Code Line | Explanation |
|---|---|
| `__len__` | Backs the `len()` built-in; must return a non-negative integer. |
| `__getitem__` | Backs `obj[index]` reads, and also enables iteration and slicing support if written to accept slice objects. |
| `__contains__` | Backs the `in` operator; without it, Python falls back to iterating the object with `__iter__`/`__getitem__` instead. |

## 18.4 Comparison Methods: `__eq__`, `__lt__`, `__le__`, `__gt__`, `__ge__`, `__ne__`, `__hash__`

````python
class Money:
    def __init__(self, cents):
        self.cents = cents

    def __eq__(self, other):
        return self.cents == other.cents

    def __lt__(self, other):
        return self.cents < other.cents

    def __repr__(self):
        return f"Money({self.cents})"

a, b = Money(500), Money(750)
print(a == b)
print(a < b)
print(sorted([b, a]))   # sorted() uses __lt__ internally
````

**Output:**
```
False
True
[Money(500), Money(750)]
```


| Code Line | Explanation |
|---|---|
| `__eq__` | Backs `==`; if you define it, Python automatically sets `__hash__` to `None` unless you also define `__hash__` explicitly, making instances unhashable by default (see note below). |
| `__lt__` | Backs `<`; also used internally by `sorted()` and `list.sort()` when comparing elements. |
| Using `functools.total_ordering` | If you define `__eq__` and one ordering method (e.g. `__lt__`), the `functools.total_ordering` class decorator can fill in the rest (`__le__`, `__gt__`, `__ge__`) automatically. |

**Important note:** defining `__eq__` without defining `__hash__` makes instances unhashable (cannot be put in a `set` or used as `dict` keys), because Python assumes mutable equality means the object's hash could change. If your object should remain hashable, define `__hash__` explicitly alongside `__eq__`.

## 18.5 Arithmetic Methods: `__add__`, `__sub__`, `__mul__`, and Friends

````python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1, v2 = Vector(1, 2), Vector(3, 4)
print(v1 + v2)      # calls __add__
print(v1 - v2)      # calls __sub__
print(v1 * 3)        # calls __mul__
````

**Output:**
```
Vector(4, 6)
Vector(-2, -2)
Vector(3, 6)
```


| Code Line | Explanation |
|---|---|
| `__add__(self, other)` | Called for `v1 + v2`; if `v1` does not know how to add `v2`, Python tries `v2.__radd__(v1)` as a fallback. |
| `__mul__(self, scalar)` | Here defined only for `Vector * number`; supporting `number * Vector` as well would require also implementing `__rmul__`. |

## 18.6 Callable Objects: `__call__`

````python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):
        return value * self.factor

double = Multiplier(2)
print(double(5))         # the instance itself is called like a function
print(callable(double))   # True - __call__ makes instances callable
````

**Output:**
```
10
True
```


## 18.7 Context Managers: `__enter__` and `__exit__`

````python
class ManagedResource:
    def __enter__(self):
        print("Acquiring resource")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Releasing resource")
        return False   # False means: do not suppress any exception

with ManagedResource() as r:
    print("Using resource")
````

**Output:**
```
Acquiring resource
Using resource
Releasing resource
```


| Code Line | Explanation |
|---|---|
| `__enter__` | Called at the start of the `with` block; its return value becomes whatever is bound by `as r`. |
| `__exit__` | Always called at the end of the block, even if an exception occurred inside it; returning `True` would suppress that exception, `False` lets it propagate. |

# 19. Method Inheritance, Overriding, and MRO

## 19.1 Method Inheritance

A subclass automatically has access to every method defined on its parent class(es), without rewriting any code.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"

class Dog(Animal):
    pass   # inherits everything from Animal, adds nothing new

d = Dog("Rex")
print(d.speak())    # inherited method, works unchanged
````

**Output:**
```
Rex makes a sound
```


## 19.2 Method Overriding

A subclass can redefine a method with the same name to change its behavior; Python always uses the most specific version found first.

````python
class Animal:
    def speak(self):
        return "Some generic animal sound"

class Cat(Animal):
    def speak(self):                 # overrides Animal.speak
        return "Meow"

class Kitten(Cat):
    def speak(self):                  # overrides Cat.speak
        base = super().speak()        # but still calls the parent's version too
        return f"{base} (but tiny)"

print(Animal().speak())
print(Cat().speak())
print(Kitten().speak())
````

**Output:**
```
Some generic animal sound
Meow
Meow (but tiny)
```


| Code Line | Explanation |
|---|---|
| `def speak(self):` (in `Cat`) | Same method name as in `Animal`; Python looks up `speak` on `Cat` first and finds it there, so `Animal.speak` is never reached for a `Cat` instance. |
| `super().speak()` | Explicitly calls the **next** class's version of the method in the MRO, letting you extend rather than completely replace inherited behavior. |

## 19.3 `super()` and Calling Parent Methods

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def summary(self):
        return f"{self.name}: ${self.salary}"

class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)     # delegate shared setup to the parent
        self.team_size = team_size

    def summary(self):
        base = super().summary()            # reuse the parent's formatting
        return f"{base}, manages {self.team_size} people"

m = Manager("Ada", 90000, 5)
print(m.summary())
````

**Output:**
```
Ada: $90000, manages 5 people
```


| Code Line | Explanation |
|---|---|
| `super().__init__(name, salary)` | Lets `Manager` reuse `Employee`'s initialization logic instead of duplicating `self.name = name` and `self.salary = salary`. |
| `super().summary()` | Reuses and extends the parent's method rather than overriding it from scratch - a common, clean pattern for cooperative subclassing. |

## 19.4 Method Resolution Order (MRO)

When a class inherits from multiple classes, Python needs a well-defined order to search for a method. This is the **Method Resolution Order**, computed with the C3 linearization algorithm, and inspectable via `__mro__` or `mro()`.

````python
class A:
    def greet(self):
        return "A"

class B(A):
    def greet(self):
        return "B"

class C(A):
    def greet(self):
        return "C"

class D(B, C):
    pass

print(D.__mro__)
print(D().greet())     # follows MRO: D -> B -> C -> A -> object
````

**Output:**
```
(<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)
B
```


| Code Line | Explanation |
|---|---|
| `class D(B, C):` | Diamond-shaped inheritance: both `B` and `C` inherit from `A`, and `D` inherits from both. |
| `D.__mro__` | Shows the exact linear order Python will search: `D`, then `B`, then `C`, then `A`, then `object` - never checking a class twice, and always checking a subclass before its parents. |
| `D().greet()` | Returns `"B"` because `B` appears before `C` in the MRO, even though both `B` and `C` define `greet`. |

**Why MRO solves the "diamond problem":** without a consistent order, it would be ambiguous whether `D` should use `B`'s or `C`'s version of a shared ancestor method. C3 linearization guarantees a single consistent order that respects each class's own inheritance list.

## 19.5 Abstract Methods

An **abstract method** is declared in a base class but has no real implementation there - it forces every subclass to provide its own. Python implements this using the `abc` module.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass                    # no implementation - subclasses MUST override this

    def describe(self):
        return f"This shape has area {self.area()}"

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

try:
    s = Shape()                 # cannot instantiate an abstract class directly
except TypeError as e:
    print("Error:", e)

sq = Square(4)
print(sq.describe())
````

**Output:**
```
Error: Can't instantiate abstract class Shape with abstract method area
Square: This shape has area 16
```


| Code Line | Explanation |
|---|---|
| `class Shape(ABC):` | Inheriting from `ABC` (Abstract Base Class) enables abstract-method enforcement. |
| `@abstractmethod` | Marks `area` as required; any subclass that fails to override it also cannot be instantiated. |
| `Shape()` | Fails immediately with `TypeError`, because `Shape` still has an unimplemented abstract method. |

## 19.6 Summary Table - Inheritance-Related Concepts

| Concept | Syntax | Purpose | Example Outcome |
|---|---|---|---|
| Inheritance | `class Sub(Base):` | Sub automatically gets Base's methods | `Dog(Animal)` gets `speak()` for free |
| Overriding | Redefine a method with the same name in the subclass | Replace or extend inherited behavior | `Cat.speak()` differs from `Animal.speak()` |
| `super()` | `super().method(...)` | Call the next class's version in the MRO | Reuse parent `__init__` logic |
| MRO | `Cls.__mro__` or `Cls.mro()` | Defines exact method lookup order for multiple inheritance | `D.__mro__` -> `(D, B, C, A, object)` |
| Abstract method | `@abstractmethod` inside an `ABC` subclass | Force subclasses to implement a method | Instantiating without override raises `TypeError` |

# 20. Properties, Getters, Setters, and Deleters

## 20.1 The Problem `@property` Solves

Without properties, controlling access to an attribute requires explicit getter/setter method calls (as in many other languages), which breaks Python's natural `obj.attribute` syntax.

````python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius       # convention: leading underscore = "internal"

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is not possible")
        self._celsius = value

    @celsius.deleter
    def celsius(self):
        print("Deleting celsius value")
        del self._celsius

    @property
    def fahrenheit(self):
        return self._celsius * 9 / 5 + 32

t = Temperature(25)
print(t.celsius)          # looks like plain attribute access, but calls the getter method
print(t.fahrenheit)        # a fully computed, read-only property

t.celsius = 30             # looks like plain assignment, but calls the setter method
print(t.celsius)

try:
    t.celsius = -300        # setter validation kicks in
except ValueError as e:
    print("Error:", e)

del t.celsius               # calls the deleter method
````

**Output:**
```
25
77.0
30
Error: Temperature below absolute zero is not possible
Deleting celsius value
```


| Code Line | Explanation |
|---|---|
| `@property` | Turns the `celsius` method into something accessed **without parentheses**, like a plain attribute, while still running actual code underneath. |
| `@celsius.setter` | Registers a setter method under the same property name; assigning `t.celsius = 30` calls this method with `value=30` instead of overwriting a plain attribute. |
| `@celsius.deleter` | Registers cleanup logic that runs when `del t.celsius` is used. |
| `fahrenheit` (no setter) | A property with only a getter is effectively **read-only** - attempting `t.fahrenheit = 100` raises `AttributeError`. |

## 20.2 Why Use Properties Instead of Plain Attributes?

| Without `@property` | With `@property` |
|---|---|
| `t.set_celsius(30)` - awkward, exposes implementation | `t.celsius = 30` - natural attribute syntax |
| Validation requires remembering to call the setter method everywhere | Validation happens automatically on every assignment |
| Changing internal storage later breaks all calling code | Internal representation can change freely; the public interface (`t.celsius`) stays the same |

## 20.3 Summary Table - Property-Related Decorators

| Decorator | Purpose | Triggered By |
|---|---|---|
| `@property` | Defines the getter | `obj.name` (read access) |
| `@name.setter` | Defines the setter | `obj.name = value` |
| `@name.deleter` | Defines the deleter | `del obj.name` |

# 21. Method Chaining and Fluent-Style Programming

**Method chaining** calls multiple methods in a single expression, one after another, because each method returns an object that itself has methods to call.

````python
text = "   Hello World   "
result = text.strip().lower().replace("world", "python")
print(result)
````

**Output:**
```
hello python
```


| Code Line | Explanation |
|---|---|
| `text.strip().lower().replace(...)` | Each string method returns a **new string**, so the next method can be called directly on that result without an intermediate variable. |

Chaining works naturally on strings because every method returns a new string. It does **not** work the same way on mutating list methods, because they return `None`:

````python
try:
    result = [3, 1, 2].sort().reverse()   # BROKEN: sort() returns None
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
Error: 'NoneType' object has no attribute 'reverse'
```


**To chain on a custom class, every chainable method must `return self`:**

````python
class QueryBuilder:
    def __init__(self, table):
        self.table = table
        self.filters = []
        self.order = None

    def where(self, condition):
        self.filters.append(condition)
        return self                      # returning self enables chaining

    def order_by(self, column):
        self.order = column
        return self

    def build(self):
        clause = f"SELECT * FROM {self.table}"
        if self.filters:
            clause += " WHERE " + " AND ".join(self.filters)
        if self.order:
            clause += f" ORDER BY {self.order}"
        return clause

query = (
    QueryBuilder("users")
    .where("age > 18")
    .where("active = true")
    .order_by("name")
    .build()
)
print(query)
````

**Output:**
```
SELECT * FROM users WHERE age > 18 AND active = true ORDER BY name
```


| Code Line | Explanation |
|---|---|
| `return self` | The key ingredient for fluent chaining: each method mutates the instance's internal state and then hands back the same instance so another method can be called immediately. |
| Parentheses around the whole chain | Allows the chained calls to be split across multiple lines for readability, which is a common style for fluent APIs like this one, pandas, or Django's QuerySets. |

# 22. Common Errors When Using Methods

## 22.1 `AttributeError` - Calling a Method That Doesn't Exist on That Type

````python
try:
    "hello".append("!")     # str has no append - that's a list method
except AttributeError as e:
    print("Error:", e)

try:
    [1, 2, 3].upper()        # list has no upper - that's a str method
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
Error: 'str' object has no attribute 'append'
Error: 'list' object has no attribute 'upper'
```


**Why this happens:** every object only has the methods defined on its own class (plus inherited ones). Mixing up which type a method belongs to is one of the most common beginner mistakes, especially between `str` and `list`, since both are sequences but support very different operations.

## 22.2 `TypeError` - Wrong Number or Type of Arguments

````python
try:
    "hello".find()             # find() requires at least one argument
except TypeError as e:
    print("Error:", e)

try:
    [1, 2].append(1, 2)         # append() takes exactly one argument
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
Error: find() takes at least 1 argument (0 given)
Error: list.append() takes exactly one argument (2 given)
```


## 22.3 `ValueError` - Right Type, Invalid Value

````python
try:
    [1, 2, 3].remove(99)        # 99 is a valid int, but not present in the list
except ValueError as e:
    print("Error:", e)

try:
    "abc".index("z")            # "z" is a valid string, but not found
except ValueError as e:
    print("Error:", e)
````

**Output:**
```
Error: list.remove(x): x not in list
Error: substring not found
```


## 22.4 Confusing Mutating and Non-Mutating Methods

Already covered in Sections 1, 4, and 5, but worth restating as the single most common category of "silent" bug: the code runs without error, but produces the wrong result because a method's return value was ignored or its in-place effect was not anticipated.

````python
name = "ada"
name.upper()          # return value discarded - bug, but no error raised
print(name)            # still "ada", unexpectedly
````

**Output:**
```

```


## 22.5 Using a Method Name Without Calling It

````python
s = "hello"
result = s.upper        # missing parentheses - this is the METHOD OBJECT, not its result
print(result)
print(type(result))
print(result())          # must actually call it
````

## 22.6 Summary Table - Common Method-Related Errors

| Error Type | Typical Cause | Example | Fix |
|---|---|---|---|
| AttributeError | Calling a method that belongs to a different type | `"hi".append(x)` | Use the correct type's method, or convert types first |
| TypeError | Wrong number/type of arguments to a method | `"a".find()` | Check the method's required parameters with `help()` |

## 23.2 Deduplicating While Preserving Order (List + Dict Methods)

````python
def dedupe_preserve_order(items):
    seen = {}
    for item in items:
        seen.setdefault(item, True)
    return list(seen.keys())

print(dedupe_preserve_order([3, 1, 2, 3, 1, 4]))
````

| Code Line | Explanation |

| `seen.setdefault(item, True)` | Because dictionaries preserve insertion order, using each item as a key and calling `setdefault` naturally deduplicates while keeping the first-seen order. |

## 23.3 Grouping Records by Category (Dict + List Methods)

````python
records = [
    {"name": "Ada", "team": "Backend"},
    {"name": "Grace", "team": "Frontend"},
    {"name": "Katherine", "team": "Backend"},
]

grouped = {}
for record in records:
    grouped.setdefault(record["team"], []).append(record["name"])

print(grouped)
````

**Output:**
```
{'Backend': ['Ada', 'Katherine'], 'Frontend': ['Grace']}
```


## 23.4 Building a Simple Cache with a Custom `__call__` Method

````python
class MemoizedSquare:
    def __init__(self):
        self.cache = {}

    def __call__(self, n):
        if n not in self.cache:
            print(f"Computing square of {n}")
            self.cache[n] = n * n
        return self.cache[n]

square = MemoizedSquare()
print(square(4))
print(square(4))   # served from cache, no "Computing" message
print(square(5))
````

**Output:**
```
Computing square of 4
16
16
Computing square of 5
25
```


## 23.5 Safe File Processing with Context Managers

````python
def count_lines(path):
    with open(path, "r") as f:
        return sum(1 for _ in f)      # iterates the file object line by line

with open("example.txt", "w") as f:
    f.writelines(["a\n", "b\n", "c\n"])

print(count_lines("example.txt"))
````

**Output:**
```
3
```

Methods are how Python objects expose behavior, and understanding them deeply means understanding far more than a list of names to memorize. The strongest habit to take from this guide is to always ask three questions before calling any method: What type of object is this? Does this method mutate the object in place, or return a new one?
