## 1. What is File Handling?

**File handling** is the process of creating, opening, reading, writing, updating, and managing files with a program.

In data science and AI engineering, file handling is extremely important because data often arrives as files such as:

- CSV files
- JSON files
- TXT files
- Log files
- Configuration files
- Exported database files
- Model configuration files
- Data generated by other applications

Python provides built in tools for working with these files.

### The basic file handling workflow

A typical workflow is:

**Open -> Read or Write -> Process -> Close**

Python's `with open(...)` pattern is preferred because Python automatically closes the file when the block finishes.

### Important idea

A file on your computer is persistent storage. A Python variable normally exists only while the program is running.

For example:

```text
Python variable -> temporary program memory
File            -> persistent storage
```

File handling allows us to move data between these two worlds.

# Part 1: Reading and Writing Text Files

## 3. Understanding `open()`

The general structure is:

```python
open(file_path, mode)
```

Common modes include:

| Mode | Meaning |
|---|---|
| `r` | Read |
| `w` | Write, replacing existing content |
| `a` | Append to existing content |
| `x` | Create a new file, failing if it already exists |
| `r+` | Read and write |
| `b` | Binary mode, commonly combined with another mode |

The most important distinction is between `w` and `a`:

- `w` can replace existing content.
- `a` adds new content to the end of the file.

````python
file_path = DATA_DIR / "employees.txt"

with open(file_path, "w", encoding="utf-8") as file:
    file.write("Ada, Data Scientist\n")
    file.write("Tunde, Data Analyst\n")
    file.write("Mariam, AI Engineer\n")

print(f"Created: {file_path}")
````

```python
file_path = DATA_DIR / "employees.txt"
```

**Line 3:**

```python
with open(file_path, "w", encoding="utf-8") as file:
```

Opens the file for writing.

- `file_path` tells Python which file to use.
- `"w"` means write mode.
- `encoding="utf-8"` tells Python how characters should be encoded.
- `as file` gives the opened file a variable name.
- `with` ensures the file is closed automatically.

**Line 4:**
```python
file.write("Ada, Data Scientist\n")
```

Writes a line into the file. `\n` represents a newline.

**Lines 5 and 6:**
Write two additional records.

**Line 8:**
Displays confirmation after the file is created.

### Important warning

If `employees.txt` already contains information, opening it with `"w"` will replace its existing content. Always choose the mode deliberately.

## 5. Reading the Entire File

The `read()` method reads the file content as one string.

````python
with open(file_path, "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

### Line by line breakdown

**Line 1:**
Opens `employees.txt` in read mode.

**Line 2:**
```python
content = file.read()
```
Reads all available content and stores it in the variable `content`.

The result is a single Python string.

**Line 4:**
Prints the content.

### When is `read()` useful?

It is useful when:

- The file is reasonably small.
- You need the entire content at once.
- You want to search or process the complete text.

For extremely large files, reading everything at once may consume too much memory. In those situations, process the file line by line.

## 6. Reading One Line at a Time

The `readline()` method reads one line at a time.

````python
with open(file_path, "r", encoding="utf-8") as file:
    first_line = file.readline()
    second_line = file.readline()

print("First line:", first_line.strip())
print("Second line:", second_line.strip())
````

### Line by line breakdown

`file.readline()` retrieves the next line from the file.

The first call gets the first line. The second call continues from where the first call stopped.

The `.strip()` method removes surrounding whitespace, including the newline character at the end of the line.

This is useful when you want controlled, sequential reading.

## 7. Reading All Lines into a List

The `readlines()` method returns the lines as a list.

````python
with open(file_path, "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)
````

### Interpretation

If the file contains three lines, `readlines()` produces a list containing three strings.

For example:

```python
[
    "Ada, Data Scientist\n",
    "Tunde, Data Analyst\n",
    "Mariam, AI Engineer\n"
]
```

Notice that the newline characters are normally retained.

### `read()` vs `readline()` vs `readlines()`

- `read()` -> entire content as one string.
- `readline()` -> next single line.
- `readlines()` -> all lines as a list.

There is also a very useful pattern:

```python
for line in file:
    ...
```

This processes the file one line at a time without first loading all lines into a list.

## 8. Iterating Through a File Line by Line

````python
with open(file_path, "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        print(f"{line_number}: {line.strip()}")
````

### Line by line breakdown

**`enumerate(file, start=1)`**

Adds a counter to each line. The first line therefore receives number `1`.

**`for line_number, line in ...`**

Each iteration gives us:

- `line_number` -> the current line number.
- `line` -> the current line's text.

**`line.strip()`**

Removes unwanted leading/trailing whitespace and the newline.

This pattern is particularly useful for log files, large text files, and data cleaning.

## 9. Appending to a File

Use `"a"` when you want to add content without replacing existing content.

````python
with open(file_path, "a", encoding="utf-8") as file:
    file.write("Chika, ML Engineer\n")

with open(file_path, "r", encoding="utf-8") as file:
    print(file.read())
````

### Interpretation

The first `with` block opens the file in append mode.

```python
"a"
```

moves writing to the end of the existing file.

The second block opens the file in read mode so that we can verify the new record.

### Key distinction

```text
w -> replace existing content
a -> preserve existing content and add to the end
```

This distinction is critical when working with logs and datasets.

## 10. Creating a File with `x`

The `x` mode means create a new file.

It is useful when you specifically want Python to fail if the file already exists.

````python
new_file = DATA_DIR / "new_file.txt"

if not new_file.exists():
    with open(new_file, "x", encoding="utf-8") as file:
        file.write("This file was created using x mode.\n")

print(new_file.read_text(encoding="utf-8"))
````

### Why check `exists()` first?

`x` raises `FileExistsError` if the target file already exists. The existence check makes this demonstration safe to rerun.

In real applications, you may intentionally use `x` without a prior check when an existing file should be treated as an error.

# Part 2: CSV Files

## 11. What is CSV?

CSV means **Comma Separated Values**.

A CSV file stores tabular data as rows and columns.

Example:

```text
name,department,salary
Ada,Data Science,85000
Tunde,Analytics,72000
Mariam,AI Engineering,95000
```

CSV is popular because it is:

- Simple
- Human readable
- Easy to exchange
- Supported by spreadsheets
- Supported by Python and data analysis libraries

Python's standard library includes the `csv` module.

## 12. Creating a CSV File

We will create a realistic employee dataset.

````python
import csv

employees = [
    ["name", "department", "salary", "years_experience"],
    ["Ada", "Data Science", 85000, 4],
    ["Tunde", "Analytics", 72000, 3],
    ["Mariam", "AI Engineering", 95000, 5],
    ["Chika", "Machine Learning", 90000, 4],
]

csv_path = DATA_DIR / "employees.csv"

with open(csv_path, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(employees)

print(f"Created: {csv_path}")
````

### Line by line breakdown

**`import csv`**

Imports Python's built in CSV module.

**`employees = [...]`**

Creates a list of lists. Each inner list represents one row.

The first inner list is the header row.

**`csv_path = DATA_DIR / "employees.csv"`**

Builds the path to the CSV file.

**`open(..., "w", newline="", encoding="utf-8")`**

Opens the file for writing.

`newline=""` is recommended when using Python's CSV module so that the CSV writer controls line endings correctly.

**`writer = csv.writer(file)`**

Creates a CSV writer object connected to the opened file.

**`writer.writerows(employees)`**

Writes all rows from the `employees` list into the file.

### `writerow()` vs `writerows()`

```python
writer.writerow(row)
```

writes one row.

```python
writer.writerows(rows)
```

writes multiple rows.

## 13. Reading a CSV with `csv.reader`

The simplest approach is to read rows as lists.

````python
with open(csv_path, "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
````

### Interpretation

`csv.reader(file)` creates an iterator that understands CSV structure.

Each `row` is returned as a list of values.

One important point: values read by `csv.reader` are strings.

For example:

```text
"85000"
```

is text, not the integer:

```python
85000
```

If numerical analysis is required, convert the appropriate fields to numeric types.

## 14. Reading CSV Rows as Dictionaries

`csv.DictReader` is often easier to use because column names become dictionary keys.

````python
with open(csv_path, "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["name"], "->", row["department"], "->", row["salary"])
````

### Why `DictReader` is useful

Instead of:

```python
row[0]
row[1]
row[2]
```

you can write:

```python
row["name"]
row["department"]
row["salary"]
```

This makes code easier to understand and less dependent on column positions.

### Important data science lesson

Human readable column names improve code maintainability. They also reduce the risk of accidentally using the wrong column index.

## 15. Converting CSV Values to Useful Data Types

CSV data is commonly read as text. We can explicitly convert fields.

````python
with open(csv_path, "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        name = row["name"]
        salary = float(row["salary"])
        years = int(row["years_experience"])

        print(f"{name}: salary={salary:,.0f}, experience={years} years")
````

### Line by line breakdown

**`name = row["name"]`**

Keeps the employee name as text.

**`salary = float(row["salary"])`**

Converts salary from a CSV string into a floating point number.

**`years = int(row["years_experience"])`**

Converts years of experience into an integer.

This conversion is essential before performing mathematical operations.

For example:

```python
"85000" + "5000"
```

would produce:

```text
850005000
```

because both values are strings.

But:

```python
85000 + 5000
```

produces:

```text
90000
```

because both values are numbers.

# Part 3: JSON Files

## 16. What is JSON?

JSON means **JavaScript Object Notation**.

Despite its name, JSON is widely used by Python applications, APIs, databases, web applications, and AI systems.

A JSON object can look like:

```json
{
    "name": "Ada",
    "role": "Data Scientist",
    "skills": ["Python", "SQL", "Machine Learning"]
}
```

JSON maps naturally to Python data structures:

| JSON | Python |
|---|---|
| object | dictionary |
| array | list |
| string | string |
| number | int/float |
| true | True |
| false | False |
| null | None |

## 17. Writing JSON

Python's `json` module provides JSON functionality.

````python
import json

employee = {
    "name": "Ada",
    "role": "Data Scientist",
    "skills": ["Python", "SQL", "Machine Learning"],
    "years_experience": 4,
    "active": True
}

json_path = DATA_DIR / "employee.json"

with open(json_path, "w", encoding="utf-8") as file:
    json.dump(employee, file, indent=4)

print(json_path.read_text(encoding="utf-8"))
````

### Line by line breakdown

**`import json`**

Imports Python's built in JSON module.

**`employee = {...}`**

Creates a Python dictionary containing strings, a list, an integer, and a Boolean.

**`json.dump(employee, file, indent=4)`**

Serializes the Python object and writes it directly into the open file.

The `indent=4` argument makes the JSON readable by humans.

### `dump()` vs `dumps()`

This distinction is important:

```python
json.dump(data, file)
```

writes JSON directly to a file.

```python
json.dumps(data)
```

returns JSON as a Python string.

The extra `s` in `dumps` can be remembered as **string**.

## 18. Reading JSON from a File

````python
with open(json_path, "r", encoding="utf-8") as file:
    loaded_employee = json.load(file)

print(loaded_employee)
print(type(loaded_employee))
print(loaded_employee["skills"])
````

### Interpretation

**`json.load(file)`**

Reads JSON from the file and converts it into an appropriate Python object.

Because the JSON document represents an object, Python returns a dictionary.

We can then access values using dictionary syntax:

```python
loaded_employee["skills"]
```

The JSON list becomes a Python list.

### `load()` vs `loads()`

```python
json.load(file)
```

reads JSON from a file.

```python
json.loads(text)
```

reads JSON from a string.

## 19. Writing Multiple Records to JSON

JSON can also store a list of dictionaries.

````python
employees_data = [
    {
        "id": 1,
        "name": "Ada",
        "department": "Data Science",
        "salary": 85000
    },
    {
        "id": 2,
        "name": "Tunde",
        "department": "Analytics",
        "salary": 72000
    },
    {
        "id": 3,
        "name": "Mariam",
        "department": "AI Engineering",
        "salary": 95000
    }
]

employees_json_path = DATA_DIR / "employees.json"

with open(employees_json_path, "w", encoding="utf-8") as file:
    json.dump(employees_data, file, indent=4)

print("JSON dataset created.")
````

## 20. Reading and Processing the JSON Dataset

````python
with open(employees_json_path, "r", encoding="utf-8") as file:
    data = json.load(file)

total_salary = sum(employee["salary"] for employee in data)
average_salary = total_salary / len(data)

print("Total salary:", total_salary)
print("Average salary:", average_salary)
````

### Line by line breakdown

**`data = json.load(file)`**

Loads the list of employee dictionaries.

**`sum(employee["salary"] for employee in data)`**

Uses a generator expression to retrieve each salary and add them together.

**`average_salary = total_salary / len(data)`**

Divides total salary by the number of employees.

### Data science connection

This is the beginning of a typical data pipeline:

```text
Raw file
   ↓
Load
   ↓
Validate
   ↓
Clean
   ↓
Transform
   ↓
Analyze
   ↓
Report / Model
```

File handling is therefore not separate from data science. It is often the first stage of the data workflow.

# Part 4: Error Handling

## 21. Why Error Handling Matters

A program can fail for many reasons:

- A file does not exist.
- A file is corrupted.
- A user provides invalid input.
- A CSV contains unexpected data.
- JSON is malformed.
- A number cannot be converted.
- The program does not have permission to access a file.

If errors are not handled appropriately, a program may stop unexpectedly.

Python provides `try` and `except` for handling exceptions.

## 22. Basic `try/except`

Let's intentionally attempt to open a file that does not exist.

````python
missing_path = DATA_DIR / "does_not_exist.txt"

try:
    with open(missing_path, "r", encoding="utf-8") as file:
        content = file.read()
except FileNotFoundError:
    print("The requested file does not exist.")
````

### Line by line breakdown

**`try:`**

Marks code that may cause an exception.

**`with open(...)`**

Attempts to open the file.

Because the file does not exist, Python raises `FileNotFoundError`.

**`except FileNotFoundError:`**

Catches that specific error.

**`print(...)`**

Provides a useful message instead of allowing the program to terminate with an unhandled exception.

### Best practice

Catch the specific exception you expect whenever practical.

Prefer:

```python
except FileNotFoundError:
```

over immediately writing:

```python
except Exception:
```

Specific exceptions make debugging easier and prevent unrelated problems from being hidden.

## 23. Handling Multiple Exceptions

Different problems may require different responses.

````python
def read_number_from_file(path):
    try:
        with open(path, "r", encoding="utf-8") as file:
            value = file.read().strip()

        number = int(value)
        return number

    except FileNotFoundError:
        print("Error: the file was not found.")
    except ValueError:
        print("Error: the file does not contain a valid integer.")
    except PermissionError:
        print("Error: permission was denied.")

    return None
````

### Detailed interpretation

The function attempts to read a number from a text file.

**`FileNotFoundError`**

Means the path does not point to an accessible existing file.

**`ValueError`**

Means the file was found and read, but the content could not be converted to an integer.

For example:

```text
hello
```

cannot become:

```python
int("hello")
```

**`PermissionError`**

Means the operating system prevented the program from accessing the file.

**`return None`**

Returns a clear Python value indicating that the operation did not produce a valid number.

This is a much stronger pattern than allowing an unexplained crash.

## 24. The `else` Block

`else` runs only when the `try` block completes without raising an exception.

````python
number_path = DATA_DIR / "number.txt"
number_path.write_text("250", encoding="utf-8")

try:
    with open(number_path, "r", encoding="utf-8") as file:
        value = int(file.read().strip())
except (FileNotFoundError, ValueError):
    print("Could not read a valid number.")
else:
    print("Successfully loaded:", value)
````

### Structure

```python
try:
    # risky operation
except:
    # runs if an exception occurs
else:
    # runs if no exception occurs
```

This is useful when you want to keep successful processing separate from error handling.

## 25. The `finally` Block

`finally` runs whether or not an exception occurs.

It is commonly used for cleanup operations.

````python
try:
    print("Performing an operation...")
    result = 100 / 5
except ZeroDivisionError:
    print("Cannot divide by zero.")
finally:
    print("This cleanup or final message always runs.")
````

### Structure

```python
try:
    ...
except SomeError:
    ...
finally:
    ...
```

The `finally` section is useful when something must happen regardless of success or failure.

For file handling, the `with open(...)` pattern usually handles closing the file for you, which is one reason it is preferred.

## 26. Raising Your Own Error

Sometimes a program should deliberately stop an operation when a business rule is violated.

````python
def validate_salary(salary):
    if salary < 0:
        raise ValueError("Salary cannot be negative.")

    return salary

print(validate_salary(85000))
````

### Interpretation

**`if salary < 0:`**

Checks whether the salary violates the expected rule.

**`raise ValueError(...)`**

Creates an exception intentionally.

This is different from an accidental error. We are telling Python:

> This value is invalid for the rules of our application.

Custom validation is extremely important in data cleaning and production data pipelines.

# Part 5: Combining File Handling and Error Handling

## 27. A Robust Text File Reader

Let's create a reusable function.

````python
def read_text_file(path):
    try:
        with open(path, "r", encoding="utf-8") as file:
            return file.read()

    except FileNotFoundError:
        print(f"File not found: {path}")
    except PermissionError:
        print(f"Permission denied: {path}")
    except UnicodeDecodeError:
        print(f"Could not decode the file as UTF-8: {path}")

    return None

text = read_text_file(file_path)

if text is not None:
    print(text)
````

### Why this is a stronger design

Instead of writing the same `try/except` logic every time, we place it inside a reusable function.

The function handles three realistic file problems:

1. Missing file.
2. Permission problem.
3. Encoding problem.

Returning `None` allows the calling code to decide what to do next.

This separation is a useful software engineering practice:

```text
Function -> handles the operation
Caller   -> decides what to do with the result
```

## 28. A Robust CSV Reader

Now we will build a reusable CSV loading function.

````python
def read_csv_file(path):
    try:
        with open(path, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)
            rows = list(reader)

        if not rows:
            print("Warning: the CSV file contains no data rows.")

        return rows

    except FileNotFoundError:
        print(f"CSV file not found: {path}")
    except PermissionError:
        print(f"Permission denied: {path}")
    except csv.Error as error:
        print(f"CSV parsing error: {error}")

    return None

rows = read_csv_file(csv_path)

if rows:
    print("Number of rows:", len(rows))
    print("First row:", rows[0])
````

### Important concepts in this function

**`csv.DictReader(file)`**

Interprets the first row as column names.

**`list(reader)`**

Consumes the reader and stores all rows in memory.

This is convenient for small and medium datasets.

For very large datasets, you may prefer to process rows one at a time rather than converting the entire reader into a list.

**`csv.Error`**

Handles errors specifically related to CSV parsing.

This is another example of targeted exception handling.

# Part 6: Data Validation and Cleaning

## 29. Validating CSV Records

Real world data is rarely perfect. A CSV may contain missing or invalid values.

Let's create a deliberately messy dataset.

````python
messy_path = DATA_DIR / "messy_employees.csv"

messy_rows = [
    ["name", "department", "salary"],
    ["Ada", "Data Science", "85000"],
    ["Tunde", "Analytics", "72000"],
    ["Mariam", "AI Engineering", "not_available"],
    ["Chika", "Machine Learning", ""],
]

with open(messy_path, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(messy_rows)

print("Messy CSV created.")
````

## 30. Cleaning the Messy CSV Safely

````python
def clean_employee_csv(path):
    cleaned = []

    try:
        with open(path, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            for row_number, row in enumerate(reader, start=2):
                name = row["name"].strip()
                department = row["department"].strip()
                salary_text = row["salary"].strip()

                if not name:
                    print(f"Row {row_number}: missing name. Skipping.")
                    continue

                if not salary_text:
                    print(f"Row {row_number}: missing salary for {name}. Skipping.")
                    continue

                try:
                    salary = float(salary_text)
                except ValueError:
                    print(
                        f"Row {row_number}: invalid salary '{salary_text}' "
                        f"for {name}. Skipping."
                    )
                    continue

                cleaned.append({
                    "name": name,
                    "department": department,
                    "salary": salary
                })

    except FileNotFoundError:
        print("CSV file was not found.")
    except KeyError as error:
        print(f"Expected column is missing: {error}")

    return cleaned

cleaned_employees = clean_employee_csv(messy_path)

for employee in cleaned_employees:
    print(employee)
````

### Detailed line by line breakdown

**`cleaned = []`**

Creates an empty list that will store valid records.

**`with open(...)`**

Safely opens the CSV.

**`reader = csv.DictReader(file)`**

Reads each row as a dictionary.

**`enumerate(reader, start=2)`**

Starts row numbering at 2 because row 1 is the header.

**`.strip()`**

Removes unnecessary whitespace.

**`if not name:`**

Checks whether the name is empty.

**`if not salary_text:`**

Checks for missing salary values.

**`float(salary_text)`**

Attempts to convert salary text into a numeric value.

**Inner `try/except ValueError`**

This is important. A single invalid salary should not necessarily destroy the entire dataset processing operation.

Instead, that particular record is skipped.

**`cleaned.append({...})`**

Adds only validated data to the output list.

### This illustrates an important production principle

Good data pipelines do not simply ask:

> Did the file open?

They also ask:

> Is the data inside the file valid enough to use?

# Part 7: JSON Error Handling

## 31. Handling Invalid JSON

JSON must follow a specific syntax. Invalid JSON can cause `json.JSONDecodeError`.

````python
invalid_json_path = DATA_DIR / "invalid.json"
invalid_json_path.write_text(
    '{"name": "Ada", "salary": 85000,}',
    encoding="utf-8"
)

try:
    with open(invalid_json_path, "r", encoding="utf-8") as file:
        data = json.load(file)
except json.JSONDecodeError as error:
    print("Invalid JSON detected.")
    print("Details:", error)
````

### Interpretation

The JSON contains a trailing comma:

```json
{"salary": 85000,}
```

That syntax is invalid JSON.

`json.load()` detects the malformed structure and raises `JSONDecodeError`.

Catching this error is useful when reading external files or API responses that may not contain valid JSON.

# Part 8: Practical Mini Project

## 32. Employee Data Pipeline

We will now combine the major ideas:

1. Read a CSV file.
2. Validate its records.
3. Calculate summary statistics.
4. Save the cleaned data to JSON.
5. Handle possible errors.

````python
def employee_data_pipeline(input_csv, output_json):
    try:
        with open(input_csv, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            cleaned = []

            for row_number, row in enumerate(reader, start=2):
                name = row.get("name", "").strip()
                department = row.get("department", "").strip()
                salary_text = row.get("salary", "").strip()

                if not name or not department or not salary_text:
                    print(f"Row {row_number}: incomplete record. Skipping.")
                    continue

                try:
                    salary = float(salary_text)
                except ValueError:
                    print(f"Row {row_number}: invalid salary. Skipping.")
                    continue

                cleaned.append({
                    "name": name,
                    "department": department,
                    "salary": salary
                })

        if not cleaned:
            print("No valid records were found.")
            return

        average_salary = sum(row["salary"] for row in cleaned) / len(cleaned)

        summary = {
            "record_count": len(cleaned),
            "average_salary": average_salary,
            "employees": cleaned
        }

        with open(output_json, "w", encoding="utf-8") as file:
            json.dump(summary, file, indent=4)

        print(f"Successfully wrote {len(cleaned)} records to {output_json}")
        print(f"Average salary: {average_salary:,.2f}")

    except FileNotFoundError:
        print(f"Input file not found: {input_csv}")
    except PermissionError:
        print("Permission denied while accessing a file.")
    except OSError as error:
        print(f"File system error: {error}")

employee_data_pipeline(
    messy_path,
    DATA_DIR / "employee_summary.json"
)
````

### Full interpretation of the pipeline

This function demonstrates how file handling appears in a real data workflow.

### Stage 1: Input

```python
input_csv
```

is the source dataset.

### Stage 2: Extraction

`csv.DictReader` extracts records from the CSV.

### Stage 3: Validation

The code checks for:

- Missing name
- Missing department
- Missing salary
- Non numeric salary

### Stage 4: Transformation

Salary strings are converted into numbers.

### Stage 5: Analysis

The pipeline calculates average salary.

### Stage 6: Output

A JSON document is produced containing:

- Number of valid records
- Average salary
- Cleaned employee records

### Stage 7: Error handling

The outer `try/except` handles file system problems while the inner `try/except` handles record level data conversion problems.

This distinction is valuable:

```text
File level errors
        |
        +-- missing file
        +-- permission problem
        +-- operating system problem

Record level errors
        |
        +-- invalid salary
        +-- missing field
        +-- malformed value
```

Separating these categories makes data pipelines easier to debug and maintain.

# Part 11: Interview and Practical Questions

## Question 1

What is the difference between `w` and `a`?

**Answer:**

`w` writes to a file and can replace existing content. `a` preserves existing content and writes new content at the end.

## Question 2

Why is `with open(...)` preferred?

**Answer:**

It manages the file resource automatically and ensures the file is closed after the block finishes, including when an exception occurs.

## Question 3

What is the difference between `json.load()` and `json.loads()`?

**Answer:**

`json.load()` reads JSON from a file-like object. `json.loads()` reads JSON from a string.

## Question 4

What does `DictReader` provide?

**Answer:**

It reads CSV rows as dictionaries, using column names as dictionary keys.

## Question 5

Why should exceptions be specific?

**Answer:**

Specific exception handling makes programs easier to understand and debug. It also avoids accidentally hiding unrelated programming errors.

## Question 6

Why convert CSV values after reading them?

**Answer:**

CSV values are normally read as strings. Numerical operations require appropriate numeric types such as `int` or `float`.

## Question 7

What is `finally` used for?

**Answer:**

It contains code that should run whether the operation succeeds or fails. It is commonly used for cleanup.

## Question 8

What does `raise` do?

**Answer:**

It deliberately creates an exception, often to enforce a validation rule or signal an invalid state.
