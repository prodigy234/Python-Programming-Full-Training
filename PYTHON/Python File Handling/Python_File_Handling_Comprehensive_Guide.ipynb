## 1. Introduction to File Handling

### 1.1 What Is a File?

A file is a named, persistent collection of data stored on a physical storage medium — a hard drive, an SSD, a USB stick, or cloud storage — that continues to exist after the program that created it has finished running and even after the computer has been switched off and back on. This persistence is the single most important property that distinguishes a file from an ordinary Python variable: a variable such as `total = 500` lives only in RAM (Random Access Memory) for the duration of the program's execution, and vanishes completely the instant the program ends. A file, by contrast, is written to durable storage, so the number `500` saved inside a file called `total.txt` is still there tomorrow, next year, or after a power outage.

**Real-world analogy:** Think of a variable as a thought you are holding in your head right now — vivid and immediately accessible, but gone the moment you stop thinking about it or fall asleep. A file is like writing that same thought down in a notebook: slower to create and to read back, but still there when you wake up, and still there if someone else opens the notebook after you have gone home.

### 1.2 Why Programs Need Files

Every useful program eventually needs to interact with the outside world beyond its own temporary memory. Files are the primary mechanism for:

| Need | Example |
|---|---|
| Persisting data between runs | Saving a user's high score so it survives closing and reopening a game |
| Exchanging data between programs | One program writes a CSV that another program (or another person, in Excel) reads |
| Working with data too large for memory | Streaming through a 50 GB log file a few lines at a time |
| Configuration | Reading settings from a `.json` or `.ini` file instead of hard-coding values |
| Recording history | Appending timestamped entries to a log file as a program runs |
| Sharing results | Writing a report, a trained model, or a dataset that a colleague or another system will consume |

### 1.3 What "File Handling" Means

File handling is the umbrella term for everything a program does to interact with files: opening them, reading their contents, writing new content, appending to existing content, moving the read/write position around inside them, and finally closing them to release the underlying operating-system resource. Python treats file handling as a first-class, built-in capability — no special library is required just to open and read a plain text file, unlike some other languages.

### 1.4 What a File System Is

A file system is the software layer, built into the operating system, responsible for organizing how files and directories are physically stored on a device and how programs are allowed to name, locate, create, and delete them. Common file systems include NTFS (Windows), APFS (macOS), and ext4 (Linux). Python's file-handling tools (`open()`, `os`, `pathlib`) do not talk to the physical disk directly; they talk to the operating system, which in turn talks to the file system, which is what actually manages where the bytes live on the storage device. This layering is precisely why the same Python code (`open('data.txt')`) works identically whether the underlying disk is an SSD, a network drive, or a USB stick — Python only needs to understand the file system's *interface*, not its physical implementation.

### 1.5 Files vs Folders (Directories)

A **file** holds actual data — text, numbers, images, program instructions. A **folder**, more precisely called a **directory**, holds no data of its own; it is simply a named container that organizes references to other files and other directories, allowing a file system to group related files together instead of storing every file that has ever been created in one enormous flat list. A directory can contain files, other directories (called **subdirectories**), or both, which is what allows a file system to form the familiar tree-like structure seen in any file explorer.

```python
import os
print(os.path.isfile('report.csv'))   # True if report.csv is a file
print(os.path.isdir('reports'))       # True if reports is a directory
```

| Code Line | Explanation |
|---|---|
| `os.path.isfile('report.csv')` | Returns `True` only if a file (not a directory) named `report.csv` exists at that location. |
| `os.path.isdir('reports')` | Returns `True` only if `reports` exists and is a directory, not a file. |

### 1.6 File Paths (Preview)

A **file path** is the address that tells the operating system, and therefore Python, exactly where a file or directory lives within the overall file-system tree. Paths are covered in complete depth in Part 2; they are previewed here because virtually every file operation in this handbook needs one.

### 1.7 Reading vs Writing

**Reading** means a program asks the file system to hand it the data that already exists inside a file, without changing that data. **Writing** means a program asks the file system to store new data inside a file, which may create a brand-new file, completely replace an existing file's contents, or add on to what is already there, depending on which **mode** the file was opened with (covered fully in Part 3).

### 1.8 Text Files vs Binary Files (Preview)

A **text file** stores data as human-readable characters (letters, digits, punctuation) arranged according to a specific **character encoding** (Part 9). A **binary file** stores data as raw bytes that follow a format specific to whatever created them (an image, an executable, a compressed archive) and are not meant to be interpreted as readable text at all. This distinction shapes almost every decision in file handling, from which mode to open a file in, to which module is appropriate for reading it, and is covered fully in Part 10.

### 1.9 Files as Persistent Storage — A Worked Example

```python
with open('diary.txt', 'w', encoding='utf-8') as f:
    f.write('Today I learned about file handling in Python.')
```

| Code Line | Explanation |
|---|---|
| `open('diary.txt', 'w', encoding='utf-8')` | Asks the operating system to create (or overwrite) a file named `diary.txt` in write mode, using the UTF-8 text encoding. |
| `f.write(...)` | Sends the given string to the file system to be physically stored. |

Even after this Python program finishes running entirely, and even after the computer running it is restarted, `diary.txt` continues to exist with that exact sentence inside it, until something (a program, or a person) explicitly deletes or overwrites it. This is the essential, defining behavior of persistent storage that variables held only in memory can never provide on their own.

## 2. File Paths

### 2.1 What a Path Is

A path is a string that describes the exact location of a file or directory within a file system's hierarchical tree structure, the same way a full street address describes the exact location of a house within a city. Just as "221B Baker Street, London" uniquely identifies one specific building, `/home/ada/documents/report.csv` uniquely identifies one specific file.

### 2.2 Absolute Paths

An **absolute path** specifies a location starting from the very top of the file-system tree — the **root** — and lists every directory that must be entered to reach the target, leaving no ambiguity about where the file actually is, regardless of what directory a program happens to be running from.

```python
# Linux / macOS absolute path
'/home/ada/projects/data/sales.csv'

# Windows absolute path
'C:\\Users\\Ada\\Documents\\sales.csv'
```

### 2.3 Relative Paths

A **relative path** specifies a location in relation to the program's current working directory (Section 2.4) rather than from the root, and therefore points to a genuinely different file depending on where the program happens to be run from.

```python
'data/sales.csv'        # look inside a 'data' folder relative to wherever the program runs
'../reports/summary.txt'  # '..' means: go up one directory first, then into 'reports'
```

| Code Line | Explanation |
|---|---|
| `data/sales.csv` | Looks for a subdirectory named `data` inside the current working directory, then for `sales.csv` inside that. |
| `..` | A special path segment meaning "the parent of the current directory"; `../reports/summary.txt` steps up one level, then down into `reports`. |
| `.` | A special path segment meaning "the current directory itself", occasionally used explicitly, as in `./script.py`. |

### 2.4 The Current Working Directory (CWD)

The current working directory is the directory a running program is considered to be "standing in" at any given moment, and it is the implicit starting point that every relative path is measured from. It is not necessarily the directory the Python script file itself lives in — it is whatever directory the program was *launched* from.

```python
import os
print(os.getcwd())
os.chdir('/home/ada/projects')
print(os.getcwd())
```

| Code Line | Explanation |
|---|---|
| `os.getcwd()` | Returns the absolute path of the current working directory as a string. |
| `os.chdir('/home/ada/projects')` | Changes the current working directory for the remainder of the running program; every relative path used afterward is now measured from this new location. |

**Common Mistake:** Assuming the current working directory is always the folder containing the `.py` file being run. It is not — it is determined by where the *command* to run the program was issued from, which is a frequent source of `FileNotFoundError` when a script is run from a different folder than expected.

### 2.5 Parent and Root Directories

The **root directory** is the single top-level directory that contains every other directory and file on a given file system — `/` on Linux and macOS, and a drive letter such as `C:\` on Windows. A **parent directory** is simply the directory that directly contains a given file or directory; every directory except the root has exactly one parent.

### 2.6 Windows Paths vs Linux/macOS Paths

| Aspect | Windows | Linux / macOS |
|---|---|---|
| Path separator | Backslash `\` | Forward slash `/` |
| Root | Drive letter, e.g. `C:\` | Single root `/` |
| Case sensitivity | Case-insensitive by default | Case-sensitive |
| Example absolute path | `C:\Users\Ada\file.txt` | `/home/ada/file.txt` |

Because the backslash is also Python's string escape character, a raw Windows path typed as a normal string requires doubled backslashes (`'C:\\Users\\Ada'`) or a raw string (`r'C:\Users\Ada'`) to avoid Python misinterpreting sequences like `\U` or `\n` as escape codes.

### 2.7 Why Hard-Coding OS-Specific Paths Causes Problems

```python
# Fragile: only works on Windows, and only for this one user
path = 'C:\\Users\\Ada\\data\\sales.csv'
```

Writing a path like this directly into code creates several serious problems: it breaks immediately on any other operating system, since Linux and macOS have no `C:\` drive at all; it breaks for any other user, since it hard-codes one specific username; and it is fragile even on the original machine, since it hard-codes an assumption about exactly where the project folder happens to live. The professional fix, covered fully in Sections 2.8 and Part 12, is to build paths programmatically using `os.path.join()` or, in modern code, `pathlib.Path`, both of which automatically use the correct separator for whatever operating system the code is actually running on.

### 2.8 Building Portable Paths

```python
import os
path = os.path.join('data', 'raw', 'sales.csv')
print(path)
```

| Code Line | Explanation |
|---|---|
| `os.path.join('data', 'raw', 'sales.csv')` | Joins the given path segments using whatever separator is correct for the current operating system — `data/raw/sales.csv` on Linux/macOS, `data\raw\sales.csv` on Windows — so the same source code produces a correctly formed path everywhere it runs. |

## 3. The open() Function

### 3.1 Syntax

```python
open(file, mode='r', encoding=None, errors=None)
```

`open()` is a built-in function, meaning it never needs to be imported; it is the single entry point Python provides for gaining access to a file, and it returns a **file object** (Part 7) that all subsequent reading or writing happens through.

| Parameter | Explanation |
|---|---|
| `file` | The path to the file, as a string or a `pathlib.Path` object. |
| `mode` | A short string describing how the file should be opened — for reading, writing, appending, and so on. Defaults to `'r'` (read text mode). |
| `encoding` | The character encoding to use for text mode, such as `'utf-8'`. Defaults to the operating system's default encoding if omitted, which is a common source of cross-platform bugs (see Part 9). |
| `errors` | How to handle characters that cannot be encoded or decoded, such as `'strict'` (the default, raises an error) or `'replace'`. |

### 3.2 File Modes

| Mode | Meaning |
|---|---|
| `'r'` | Read (default). File must already exist, or `FileNotFoundError` is raised. |
| `'w'` | Write. Creates the file if it does not exist; **completely erases** existing content if it does. |
| `'a'` | Append. Creates the file if it does not exist; new writes are added after existing content. |
| `'x'` | Exclusive creation. Creates a new file; raises `FileExistsError` if the file already exists. |
| `'b'` | Binary mode modifier, combined with another mode, e.g. `'rb'`. |
| `'t'` | Text mode modifier (the default); rarely written explicitly since it is implied. |
| `'+'` | Update modifier, enabling both reading and writing on the same file object. |

### 3.3 Common Mode Combinations

| Mode | Meaning |
|---|---|
| `'rb'` | Read binary — e.g. opening an image. |
| `'wb'` | Write binary — completely overwrites, in bytes. |
| `'ab'` | Append binary. |
| `'r+'` | Read and write text; file must already exist; does not truncate. |
| `'w+'` | Read and write text; truncates (erases) the file first. |
| `'a+'` | Read and write text; writes always go to the end; reading can happen anywhere. |
| `'rb+'` | Read and write binary; file must already exist. |
| `'wb+'` | Read and write binary; truncates the file first. |

### 3.4 Detailed Mode Comparison Table

| Mode | File Must Exist? | Erases Existing Content? | Can Read? | Can Write? | Cursor Starts At |
|---|---|---|---|---|---|
| `r` | Yes | No | Yes | No | Start |
| `w` | No | Yes | No | Yes | Start |
| `a` | No | No | No | Yes (end only) | End |
| `x` | Must NOT exist | N/A (creates new) | No | Yes | Start |
| `r+` | Yes | No | Yes | Yes | Start |
| `w+` | No | Yes | Yes | Yes | Start |
| `a+` | No | No | Yes | Yes (end only) | End |

### 3.5 Worked Examples of Each Core Mode

```python
# 'x' - exclusive creation
try:
    f = open('new_file.txt', 'x')
    f.write('First and only time this can be created safely.')
    f.close()
except FileExistsError:
    print('That file already exists.')
```

| Code Line | Explanation |
|---|---|
| `open('new_file.txt', 'x')` | Attempts to create `new_file.txt`; if a file by that name already exists anywhere at that path, Python raises `FileExistsError` instead of silently overwriting it, making `'x'` the safest mode for "create this file only if it is genuinely new". |

### 3.6 File Descriptors (Conceptually)

Underneath every open file object, the operating system tracks the open file using a small integer called a **file descriptor** (on Linux/macOS) or a file handle (on Windows). Python's file object is a friendly, high-level wrapper around this low-level OS resource. This matters practically for one reason above all others: the operating system allows only a limited number of file descriptors to be open at once per process, which is precisely why files must always eventually be closed (Part 6) — leaving files open leaks this limited resource until the program is forced to stop working correctly.

### 3.7 Error Handling With open()

```python
try:
    f = open('missing_file.txt', 'r')
except FileNotFoundError:
    print('The file does not exist.')
```

Opening a file can fail for many reasons beyond a missing file — insufficient permissions, the path pointing to a directory instead of a file, or the disk being unavailable — all covered fully with professional handling patterns in Part 19.

## 4. Reading Text Files

### 4.1 read() — Reading the Entire File

```python
with open('notes.txt', 'r', encoding='utf-8') as f:
    content = f.read()
print(content)
```

`read()` with no argument consumes the entire file in one call and returns it as a single string, including any newline characters embedded in it. This is convenient for small files but risky for large ones, since the whole content must fit in memory at once (Part 20).

### 4.2 read(size) — Reading a Limited Number of Characters

```python
with open('notes.txt', 'r', encoding='utf-8') as f:
    chunk = f.read(10)
    print(chunk)
```

| Code Line | Explanation |
|---|---|
| `f.read(10)` | Reads at most the next 10 characters from the file's current cursor position, then stops; the cursor advances by that amount, so a second `f.read(10)` call continues from where the first left off. |

### 4.3 readline() — Reading One Line at a Time

```python
with open('notes.txt', 'r', encoding='utf-8') as f:
    first_line = f.readline()
    second_line = f.readline()
print(first_line)
print(second_line)
```

`readline()` reads and returns a single line, including its trailing newline character (`\n`), advancing the cursor to the start of the next line. Calling it again continues from that new position. At the end of the file, `readline()` returns an empty string `''`, which is how code can detect that there is nothing left to read.

### 4.4 readlines() — Reading Every Line Into a List

```python
with open('notes.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()
print(lines)
```

`readlines()` reads the entire file and returns a list of strings, one per line, with each line's trailing newline character preserved. Like `read()`, it loads the whole file into memory at once, which makes it unsuitable for very large files.

### 4.5 Iterating Over a File Directly (The Preferred Approach)

```python
with open('notes.txt', 'r', encoding='utf-8') as f:
    for line in f:
        print(line.strip())
```

| Code Line | Explanation |
|---|---|
| `for line in f:` | A file object is itself iterable; looping over it directly yields one line at a time without ever loading the entire file into memory simultaneously, making it the most memory-efficient and idiomatic way to process a text file line by line. |
| `line.strip()` | Removes the trailing newline character (and any surrounding whitespace) from each line before printing it. |

### 4.6 File Cursor Position and End of File

Every open file object tracks an internal **cursor** (also called the file position) marking exactly where the next read or write operation will begin. Reading operations advance this cursor automatically; once the cursor reaches the very end of the file's content, any further `read()` call returns an empty string (text mode) or empty bytes object (binary mode), which is the standard, reliable way Python code detects **end of file (EOF)** — there is no separate `EOF` marker character to check for.

### 4.7 Memory Considerations When Reading Large Files

`read()` and `readlines()` both require enough free memory to hold the entire file's content simultaneously. For a 200 MB log file, this means the program's memory usage jumps by roughly 200 MB (or more, accounting for Python's string object overhead) the instant `read()` is called. Iterating with `for line in f:`, by contrast, keeps memory usage roughly constant regardless of file size, since only one line needs to exist in memory at any given moment. This distinction becomes critical and is explored fully with streaming and chunked-processing techniques in Part 20.

### 4.8 Reading Different File Types as Text

```python
with open('server.log', 'r', encoding='utf-8') as f:
    for line in f:
        if 'ERROR' in line:
            print(line.strip())
```

Plain-text-based formats — `.txt`, `.log`, and even `.csv` and `.json` at the raw-text level — can all be opened and read with the exact same `open()` mechanics shown above, since they are fundamentally just text. Parts 14 and 15 introduce the `csv` and `json` modules, which build structured parsing logic on top of this same underlying text-reading foundation, saving the work of manually splitting lines on commas or parsing brackets by hand.

## 5. Writing Files

### 5.1 write() — Writing a Single String

```python
with open('output.txt', 'w', encoding='utf-8') as f:
    f.write('Hello, file handling!')
```

`write()` sends exactly the string given to it to the file — nothing more. Critically, it does **not** automatically add a newline character at the end, unlike `print()`; calling `write()` twice in a row writes both strings directly back to back on the same line unless a `\n` is included explicitly.

### 5.2 writelines() — Writing Multiple Lines

```python
lines = ['First line\n', 'Second line\n', 'Third line\n']
with open('output.txt', 'w', encoding='utf-8') as f:
    f.writelines(lines)
```

| Code Line | Explanation |
|---|---|
| `f.writelines(lines)` | Writes every string in the given iterable to the file, one after another, with **no** newline characters inserted automatically between them; each string in the list above must include its own trailing `\n` to appear as a separate line, which is a frequent point of confusion, since the name suggests it behaves like `print()` across multiple lines, but it does not. |

### 5.3 Overwriting vs Appending — 'w' vs 'a'

```python
with open('log.txt', 'w', encoding='utf-8') as f:
    f.write('First entry\n')

with open('log.txt', 'w', encoding='utf-8') as f:
    f.write('Second entry\n')
# log.txt now contains ONLY 'Second entry', the first line is gone

with open('log.txt', 'a', encoding='utf-8') as f:
    f.write('Third entry\n')
# log.txt now contains 'Second entry' AND 'Third entry'
```

| Code Line | Explanation |
|---|---|
| Second `open(..., 'w')` | `'w'` mode truncates (completely empties) the file the instant it is opened, before any writing even happens; the previous content is unrecoverably gone. |
| `open(..., 'a')` | `'a'` mode never erases existing content; every write is appended after whatever is already there, with the cursor forced to the end of the file regardless of any `seek()` calls. |

**Common Mistake and Data-Loss Risk:** Opening a file with `'w'` inside a loop that runs more than once, expecting each iteration to add a new line, is one of the most common and damaging beginner mistakes in file handling — each pass silently destroys the previous iteration's output, leaving only the very last write. If the goal is genuinely to accumulate content across multiple writes or multiple runs of a program, `'a'` mode (or opening the file once, outside the loop, with `'w'`) is required instead.

### 5.4 Creating New Files

Both `'w'` and `'a'` mode create the target file automatically if it does not already exist yet, requiring no separate file-creation step. `'x'` mode (Part 3) additionally guarantees the file is genuinely new, refusing to proceed if a file by that name is already present.

### 5.5 Newline Handling

```python
with open('output.txt', 'w', encoding='utf-8') as f:
    f.write('Line one\n')
    f.write('Line two\n')
```

Python's text mode performs **universal newline translation** by default: regardless of what operating system the code runs on, writing `\n` produces the correct native line-ending sequence for that platform (`\n` on Linux/macOS, `\r\n` on Windows), and reading a file translates any of `\n`, `\r\n`, or the old Mac `\r` back into a plain `\n` inside Python. This translation can be disabled with `newline=''` when exact byte-for-byte control over line endings is required, which matters most when working with the `csv` module (Part 14).

## 6. The with Statement

### 6.1 Why Files Must Be Closed

Every open file consumes a limited operating-system resource (the file descriptor, Section 3.6) and may also hold data in an internal buffer that has not yet been physically written to disk. Failing to close a file can leave written data stuck in that buffer and never actually saved, and can eventually exhaust the operating system's limit on how many files a single program may have open simultaneously.

### 6.2 The Manual (Fragile) Approach

```python
f = open('data.txt', 'w', encoding='utf-8')
f.write('Some data')
f.close()
```

This works, but it is fragile: if an exception is raised anywhere between `open()` and `close()`, the `close()` line is skipped entirely, and the file is left open indefinitely.

```python
f = open('data.txt', 'w', encoding='utf-8')
f.write(1 / 0)   # raises ZeroDivisionError
f.close()        # never reached!
```

### 6.3 with open(...) — The Professional Standard

```python
with open('data.txt', 'w', encoding='utf-8') as f:
    f.write('Some data')
    result = 1 / 0
```

| Code Line | Explanation |
|---|---|
| `with open(...) as f:` | Opens the file and binds the resulting file object to `f`, exactly like a direct `open()` call, but wraps the entire indented block in a **context manager** that guarantees cleanup. |
| (exception inside the block) | Even though `1 / 0` raises a `ZeroDivisionError` partway through the block, Python still automatically closes the file before the exception propagates further, because the context manager's cleanup step runs regardless of how the block exits. |

### 6.4 What Happens If an Exception Occurs — In Detail

A `with` block's cleanup logic is guaranteed to run under all three possible ways a block can end: normal completion, an explicit `return` or `break` inside the block, and an uncaught exception. This guarantee is precisely what manual `open()`/`close()` pairs cannot provide without extra, error-prone `try`/`finally` boilerplate.

### 6.5 __enter__ and __exit__ — How Context Managers Work

A context manager is any object that defines two special (dunder) methods: `__enter__`, called when the `with` block begins, and `__exit__`, called when the block ends, no matter how it ends. A file object implements both: `__enter__` returns the file object itself (which is what gets bound to `f`), and `__exit__` calls `close()` on it automatically.

```python
class SimpleFileLogger:
    def __init__(self, path):
        self.path = path
    def __enter__(self):
        self.file = open(self.path, 'a', encoding='utf-8')
        return self.file
    def __exit__(self, exc_type, exc_value, traceback):
        self.file.close()
        return False

with SimpleFileLogger('activity.log') as log:
    log.write('Started process\n')
```

| Code Line | Explanation |
|---|---|
| `def __enter__(self):` | Runs when the `with` block starts; here it opens the underlying file and returns the file object, which becomes what `as log` binds to. |
| `def __exit__(self, exc_type, exc_value, traceback):` | Runs automatically when the block ends; here it closes the file. The three parameters describe any exception that occurred, or are all `None` if the block completed normally. |
| `return False` | Signals that `__exit__` has not handled the exception itself, so Python should continue propagating it normally after cleanup; returning `True` would instead suppress the exception entirely. |

### 6.6 Why with open() Is Preferred Over Manual close()

- It guarantees the file is closed even when an exception occurs, without needing manual `try`/`finally` code.
- It is more concise and immediately signals to any reader "this file's lifetime is scoped to this block".
- It eliminates an entire category of resource-leak bugs caused by a `close()` call being forgotten, skipped, or unreachable.
- It is the idiomatic, expected style in professional and reviewed Python code; a `with`-free `open()` is a common code-review flag.

## 7. File Objects

### 7.1 What the File Object Represents

The value returned by `open()` is a **file object** — a Python object that wraps the underlying operating-system file resource and exposes methods (`read()`, `write()`, `seek()`, `close()`, and more) plus a small set of descriptive attributes about the open file itself.

### 7.2 Key File Object Attributes

```python
f = open('data.txt', 'w', encoding='utf-8')
print(f.name)
print(f.mode)
print(f.closed)
f.close()
print(f.closed)
```

| Attribute | Explanation |
|---|---|
| `f.name` | The path that was passed to `open()`, e.g. `'data.txt'`. |
| `f.mode` | The mode string the file was opened with, e.g. `'w'`. |
| `f.closed` | A boolean, `True` once `close()` has been called, `False` while the file is still open; useful for defensive checks before attempting a further operation. |

### 7.3 File State: Open vs Closed

Once a file object's `close()` method has run — whether called manually or automatically by a `with` block — any further attempt to `read()` or `write()` through that same object raises a `ValueError: I/O operation on closed file`. A closed file object cannot be reopened; a fresh call to `open()` is required to work with the file again.

### 7.4 File Metadata vs File Object Attributes

It is worth distinguishing what the file *object* knows about itself (`name`, `mode`, `closed`, covered here) from what the *file system* knows about the file on disk (size, modification time, permissions), which is covered fully in Parts 23 and 24 using `os.stat()` and `pathlib`.

## 8. File Pointers

### 8.1 The File Cursor Revisited

As introduced in Section 4.6, every open file maintains a cursor marking the current read/write position, measured in bytes from the start of the file (even in text mode, internally). Parts 8.2 and 8.3 introduce the two methods that let a program directly inspect and control this position.

### 8.2 tell() — Where Is the Cursor Right Now?

```python
with open('data.txt', 'r', encoding='utf-8') as f:
    print(f.tell())
    f.read(5)
    print(f.tell())
```

| Code Line | Explanation |
|---|---|
| `f.tell()` (first call) | Returns `0`; the cursor starts at the very beginning of the file immediately after opening. |
| `f.tell()` (second call) | Returns `5` after reading 5 characters, reflecting the cursor's new position. |

### 8.3 seek() — Moving the Cursor

```python
with open('data.txt', 'r', encoding='utf-8') as f:
    f.seek(0)      # go to the very start
    f.seek(10)     # jump to byte offset 10
    f.read()
    f.seek(0)      # reset back to the start to read again
```

`seek(offset)` moves the cursor directly to the given byte offset from the start of the file, enabling **random access** — reading or writing at an arbitrary position without having to sequentially read through everything before it. This is especially useful in binary mode (Part 10) for structured binary formats, and for resetting a file object back to the beginning to read it a second time without closing and reopening it.

**Caution with text mode:** `seek()` in text mode should generally only be used with offsets of `0` or values previously returned by `tell()`, because multi-byte encodings such as UTF-8 (Part 9) mean an arbitrary byte offset can land in the *middle* of a multi-byte character, producing a `UnicodeDecodeError`. Binary mode has no such restriction, since every byte is a valid position.

### 8.4 A Complete Random-Access Example

```python
with open('record.bin', 'r+b') as f:
    f.seek(0)
    f.write(b'HEADER')
    f.seek(100)
    f.write(b'DATA_BLOCK_AT_OFFSET_100')
    f.seek(0)
    print(f.read(6))
```

| Code Line | Explanation |
|---|---|
| `f.seek(100)` | Jumps directly to byte offset 100 in the file without needing to read through the first 100 bytes first, then writes there — a technique impossible with purely sequential reading/writing. |

## 9. Character Encoding

### 9.1 Characters vs Bytes

A file, at the most fundamental physical level, only ever stores raw **bytes** — numbers from 0 to 255. A **character encoding** is the agreed-upon rule that maps human-readable characters (letters, digits, emoji, symbols) to and from those raw bytes. Without an encoding, a sequence of bytes is just numbers with no inherent textual meaning at all.

### 9.2 Encoding and Decoding

**Encoding** is the process of converting a Python `str` (text, made of characters) into a `bytes` object (raw numbers), ready to be physically written to a file or sent over a network. **Decoding** is the reverse: converting raw `bytes` back into a readable `str`, using the same encoding scheme that was used to encode it in the first place.

```python
text = 'café'
encoded = text.encode('utf-8')
print(encoded)
decoded = encoded.decode('utf-8')
print(decoded)
```

| Code Line | Explanation |
|---|---|
| `text.encode('utf-8')` | Converts the string into bytes, producing `b'caf\xc3\xa9'`; note that the accented `é` takes two bytes under UTF-8, while the plain ASCII letters each take one. |
| `encoded.decode('utf-8')` | Converts the bytes back into the original string `'café'`, correctly reassembling the two-byte sequence into the single character `é`, provided the same encoding is used for both directions. |

### 9.3 ASCII

ASCII is one of the oldest and simplest encodings, assigning each of 128 characters (the basic unmarked English alphabet, digits, and common punctuation) to a single byte value from 0 to 127. It cannot represent accented letters, non-Latin scripts, or emoji at all.

### 9.4 Unicode

Unicode is not itself an encoding but a universal *standard* that assigns every character from virtually every writing system in the world — Latin, Cyrillic, Han characters, emoji, and more — a unique numeric identifier called a **code point**. Encodings such as UTF-8, UTF-16, and UTF-32 are the different concrete schemes for representing those Unicode code points as actual bytes.

### 9.5 UTF-8, UTF-16, UTF-32

| Encoding | Bytes per Character | Notes |
|---|---|---|
| UTF-8 | Variable: 1 to 4 bytes | ASCII-compatible for the first 128 characters; the dominant encoding on the web and in modern files; Python's recommended default. |
| UTF-16 | Usually 2 bytes, sometimes 4 | Common on Windows internals and in some Java/`.NET` ecosystems. |
| UTF-32 | Always exactly 4 bytes | Simple fixed-width mapping, but noticeably wasteful of space for ordinary text. |

### 9.6 Specifying encoding= Explicitly

```python
with open('notes.txt', 'r', encoding='utf-8') as f:
    content = f.read()
```

When `encoding` is omitted from `open()`, Python falls back to the operating system's **locale-dependent default encoding**, which is commonly UTF-8 on Linux and macOS but has historically been something else (like `cp1252`) on Windows. A file written without an explicit encoding on one machine can therefore fail to read correctly, or even raise a `UnicodeDecodeError`, on a different machine or a different user's Windows configuration. **Always specifying `encoding='utf-8'` explicitly is considered a professional best practice** precisely to eliminate this entire class of platform-dependent bug.

### 9.7 Encoding and Decoding Errors

```python
try:
    'café'.encode('ascii')
except UnicodeEncodeError as e:
    print(e)

try:
    b'\xc3\xa9'.decode('ascii')
except UnicodeDecodeError as e:
    print(e)
```

A `UnicodeEncodeError` occurs when a string contains a character that the target encoding has no representation for (ASCII cannot represent `é`). A `UnicodeDecodeError` occurs when a byte sequence does not form a valid character under the encoding being used to decode it. The optional `errors=` parameter of `open()` controls this behavior — `'strict'` (the default) raises an exception; `'replace'` substitutes a placeholder character (`�`); `'ignore'` silently drops the offending bytes or characters.

### 9.8 Practical Example: English and Non-English Text Together

```python
sentence = 'Hello, world! Привет, мир! 你好，世界！'
with open('multilingual.txt', 'w', encoding='utf-8') as f:
    f.write(sentence)

with open('multilingual.txt', 'r', encoding='utf-8') as f:
    print(f.read())
```

With `encoding='utf-8'` used consistently on both the write and the read, English, Russian (Cyrillic), and Chinese characters all round-trip perfectly, since UTF-8 can represent every character in the Unicode standard.

## 10. Text Files vs Binary Files

### 10.1 Text Files

A text file is interpreted through an encoding (Part 9) as a sequence of characters. In Python, reading a text-mode file (`open(path, 'r')`) always yields `str` objects.

### 10.2 Binary Files

A binary file is treated as a raw sequence of bytes with no character interpretation applied at all. In Python, reading a binary-mode file (`open(path, 'rb')`) always yields `bytes` objects, never `str`.

```python
with open('photo.jpg', 'rb') as f:
    header = f.read(4)
print(header)
```

### 10.3 str vs bytes

```python
text_value = 'Hello'
binary_value = b'Hello'
print(type(text_value))
print(type(binary_value))
print(text_value == binary_value)
```

`str` and `bytes` are entirely distinct types in Python 3, and `==` between them is always `False`, never raising an error but never matching either — a common source of silent bugs when binary data is accidentally compared against a text literal.

### 10.4 When to Use Binary Mode

| File Type | Mode | Reason |
|---|---|---|
| Images (`.jpg`, `.png`) | `'rb'` / `'wb'` | Pixel and header data are not text; decoding them as characters would corrupt the data. |
| PDFs | `'rb'` / `'wb'` | Contain a mix of text streams and binary-encoded objects; must be handled as raw bytes (or through a dedicated library — see the separate PDF handbook). |
| Audio / video | `'rb'` / `'wb'` | Encoded in format-specific binary structures (e.g. MP3 frames, MP4 boxes). |
| ZIP archives | `'rb'` / `'wb'`, or the `zipfile` module (Part 26) | Compressed binary container format. |
| Executables | `'rb'` | Machine code, not human-readable text at all. |

### 10.5 Copying a Binary File

```python
with open('source.jpg', 'rb') as src, open('copy.jpg', 'wb') as dst:
    dst.write(src.read())
```

| Code Line | Explanation |
|---|---|
| `open(...) as src, open(...) as dst` | A single `with` statement can open multiple files at once, separated by commas, each getting its own `as` binding and each guaranteed to be closed when the block ends. |
| `dst.write(src.read())` | Reads the entire binary source file into memory as `bytes`, then writes those exact bytes to the destination, producing an identical byte-for-byte copy — appropriate for small-to-medium files; Part 20 covers chunked copying for very large binary files. |

### 10.6 Mixing Text and Binary Concepts — A Common Mistake

```python
# WRONG - mixing modes
with open('data.bin', 'rb') as f:
    content = f.read()
    print(content.upper())  # bytes has .upper() too, but mixing with str logic elsewhere causes bugs
```

Attempting to write a `str` to a file opened in binary mode, or a `bytes` object to a file opened in text mode, raises a `TypeError` immediately — Python does not silently convert between them, which is a deliberate design choice that forces the encoding decision (Part 9) to be made explicitly rather than guessed.

## 14. CSV Files

### 14.1 What CSV Is

CSV (Comma-Separated Values) is a plain-text tabular format where each line represents one row, and values within a row are separated by a delimiter, most commonly a comma. Despite its simplicity, CSV is one of the most widely used data-interchange formats in Data Analytics, since it is human-readable, universally supported, and far lighter than binary spreadsheet formats.

### 14.2 Reading CSV With csv.reader

```python
import csv

with open('sales.csv', 'r', encoding='utf-8', newline='') as f:
    reader = csv.reader(f)
    header = next(reader)
    for row in reader:
        print(row)
```

| Code Line | Explanation |
|---|---|
| `open(..., newline='')` | The `csv` module's own documentation specifically recommends `newline=''` when opening files for `csv.reader`/`csv.writer`, so that the module can correctly manage line endings itself rather than have Python's own universal-newline translation (Section 5.5) interfere with it. |
| `csv.reader(f)` | Wraps the file object in a reader that automatically splits each line into a list of string fields, correctly handling quoted fields that themselves contain commas. |
| `next(reader)` | Consumes and returns just the first row (typically the header row), advancing the reader so the following loop starts from the second row. |

### 14.3 Writing CSV With csv.writer

```python
import csv

rows = [['Name', 'Score'], ['Ada', 95], ['Grace', 88]]
with open('results.csv', 'w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f)
    writer.writerows(rows)
```

| Code Line | Explanation |
|---|---|
| `writer.writerows(rows)` | Writes every row in the given list of lists in one call, automatically inserting commas between fields and the correct line ending after each row. A single row can also be written with `writer.writerow([...])`. |

### 14.4 csv.DictReader — Reading Rows as Dictionaries

```python
import csv

with open('sales.csv', 'r', encoding='utf-8', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row['Name'], row['Amount'])
```

`DictReader` automatically uses the file's first row as field names and yields every subsequent row as an ordered dictionary mapping those names to the row's values, which is generally far more readable than accessing fields by numeric position with plain `csv.reader`.

### 14.5 csv.DictWriter — Writing Rows From Dictionaries

```python
import csv

data = [{'Name': 'Ada', 'Score': 95}, {'Name': 'Grace', 'Score': 88}]
with open('results.csv', 'w', encoding='utf-8', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['Name', 'Score'])
    writer.writeheader()
    writer.writerows(data)
```

| Code Line | Explanation |
|---|---|
| `csv.DictWriter(f, fieldnames=['Name', 'Score'])` | Requires `fieldnames` to be specified explicitly, since a list of dictionaries has no inherent guaranteed column order on its own. |
| `writer.writeheader()` | Writes the header row using the given `fieldnames`; this step is not automatic and must be called explicitly. |

### 14.6 Delimiters Other Than Comma

```python
import csv
with open('data.tsv', 'r', encoding='utf-8', newline='') as f:
    reader = csv.reader(f, delimiter='\t')
    for row in reader:
        print(row)
```

The `delimiter` parameter allows the `csv` module to correctly parse tab-separated (`.tsv`), semicolon-separated (common in European locales where comma is the decimal separator), or any other consistently delimited text format.

### 14.7 Quoting

```python
import csv
rows = [['Product', 'Description'], ['Widget', 'A "small" gadget, very useful']]
with open('products.csv', 'w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f, quoting=csv.QUOTE_MINIMAL)
    writer.writerows(rows)
```

Quoting rules determine when a field is wrapped in double quotes in the output — necessary whenever a field itself contains the delimiter character, a quote character, or a newline. `csv.QUOTE_MINIMAL` (the default) quotes only fields that actually need it; `csv.QUOTE_ALL` quotes every field regardless.

### 14.8 Handling Missing Values and Malformed CSV

```python
import csv

with open('messy.csv', 'r', encoding='utf-8', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        amount = row.get('Amount') or '0'
        try:
            value = float(amount)
        except ValueError:
            print(f'Skipping malformed row: {row}')
            continue
```

Real-world CSV files frequently have missing fields, inconsistent column counts, or values that fail to convert to the expected type; professional file-processing code anticipates this with `.get()` for possibly-missing dictionary keys, explicit `try`/`except` around type conversions, and a clear policy (skip, default, or flag) for each kind of problem row, rather than assuming the data is always clean.

### 14.9 A Realistic Data Analytics Example

```python
import csv

total_revenue = 0.0
row_count = 0

with open('transactions.csv', 'r', encoding='utf-8', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        try:
            total_revenue += float(row['amount'])
            row_count += 1
        except (ValueError, KeyError):
            continue

print(f'Processed {row_count} transactions')
print(f'Total revenue: {total_revenue:.2f}')
```

|---|---|
| `json.dump(data, f, indent=4)` | Serializes the Python object and writes it directly to the given open file, formatted with 4-space indentation for human readability. |
| `json.dumps(data, indent=4)` | Serializes the Python object into a JSON-formatted **string** held in memory, without touching any file — useful for logging, sending over a network, or further string processing. |

### 15.5 JSON File vs JSON String

The distinction between the `load`/`dump` pair (file-based) and the `loads`/`dumps` pair (string-based) is one of the most common early points of confusion; the extra `s` always means "string", never "save". Reading an API response's raw text and reading a `.json` file's contents both ultimately go through `loads()`/`load()` respectively, once the text is available.

### 15.6 A Data Science / API Example

```python
import json

api_response = '''
{
    "status": "success",
    "records": [
        {"id": 1, "temperature": 21.5},
        {"id": 2, "temperature": 22.1}
    ]
}
'''

data = json.loads(api_response)
average_temp = sum(r['temperature'] for r in data['records']) / len(data['records'])
print(f'Average temperature: {average_temp:.2f}')

with open('processed_temps.json', 'w', encoding='utf-8') as f:
    json.dump({'average_temperature': average_temp}, f, indent=2)
```

## 17. Serialization

### 17.1 What Serialization Means

Serialization is the process of converting an in-memory Python object — which might be a simple dictionary or a complex custom class instance — into a format that can be stored in a file or transmitted over a network, and later reconstructed back into an equivalent object. **Deserialization** is that reverse reconstruction process.

### 17.2 Why Serialization Is Useful

Without serialization, the only things a program can persist are raw strings and bytes; serialization provides a systematic, repeatable way to save and restore structured Python data — configuration, application state, processed datasets, or trained models — without hand-writing custom parsing logic for every single object.

### 17.3 JSON Serialization vs Pickle

| Aspect | JSON | Pickle |
|---|---|---|
| Format | Human-readable text | Binary, Python-specific |
| Cross-language support | Yes (universal) | No (Python only) |
| Supported types | Basic types only (dict, list, str, number, bool, None) | Almost any Python object, including custom classes |
| Security | Safe to load from untrusted sources | **Unsafe** to load from untrusted sources (Part 18) |
| Typical use | Config files, APIs, data interchange | Saving Python-specific objects like trained ML models |

### 17.4 Choosing Between Them

JSON is the appropriate choice whenever the data needs to be read by another program, another language, or a human, or whenever security around untrusted input matters. Pickle is appropriate specifically when the object being saved is a Python-native structure that JSON cannot represent — a trained scikit-learn model, a custom class instance, a set, or a tuple with mixed nested types — and the file will only ever be loaded back by trusted Python code.

### 18.1 pickle.dump() and pickle.load()

```python
import pickle

data = {'model_name': 'baseline', 'weights': [0.2, 0.5, 0.3]}

with open('model.pkl', 'wb') as f:
    pickle.dump(data, f)

with open('model.pkl', 'rb') as f:
    loaded = pickle.load(f)
print(loaded)
```

| `pickle.dump(data, f)` | Serializes `data` and writes the resulting bytes directly to the open file. |
| `pickle.load(f)` | Reads and deserializes the file's content back into an equivalent Python object. |

### 18.2 Saving and Loading Custom Objects

Pickle can serialize entire custom class instances, including their internal state, which JSON fundamentally cannot do without writing custom conversion logic first — this is pickle's core practical advantage.

### 18.3 Model and Data Persistence in Machine Learning

```python
import pickle

with open('trained_model.pkl', 'wb') as f:
    pickle.dump(trained_model, f)

with open('trained_model.pkl', 'rb') as f:
    trained_model = pickle.load(f)
```

### 18.4 Security Risks of Untrusted Pickle Files

Unpickling data is **not** a safe operation to perform on data from an untrusted source, because the pickle format can encode instructions that execute arbitrary code during deserialization — a maliciously crafted `.pkl` file can run harmful code the instant `pickle.load()` is called on it, with no additional trigger required. This is fundamentally different from JSON, which has no such capability at all. **Never call `pickle.load()` on a file downloaded from the internet, received from an untrusted user, or of unknown origin.** This is covered further with defensive practices in Part 32.

## 19. File Handling and Exception Handling

### 19.1 Common File-Related Exceptions

| Exception | Typical Cause |
|---|---|
| `FileNotFoundError` | Attempting to open a nonexistent file in read mode. |
| `PermissionError` | Lacking the operating-system permission required for the attempted operation. |
| `IsADirectoryError` | Passing a directory's path to a function expecting a file. |
| `NotADirectoryError` | Passing a file's path to a function expecting a directory. |
| `UnicodeDecodeError` | Reading bytes that are not valid under the encoding being used to decode them. |
| `UnicodeEncodeError` | Writing a character that has no representation in the target encoding. |
| `OSError` | A general operating-system-level failure; most of the above are actually subclasses of `OSError`. |

### 19.2 Professional Handling Pattern

```python
def load_config(path):
    try:
        with open(path, 'r', encoding='utf-8') as f:
            return f.read()
    except FileNotFoundError:
        print(f'Config file not found: {path}')
    except PermissionError:
        print(f'No permission to read: {path}')
    except UnicodeDecodeError:
        print(f'Could not decode {path} as UTF-8')
    except OSError as e:
        print(f'Unexpected OS error: {e}')
    return None
```

Handling specific exceptions before the general `OSError` fallback (this ordering matters — Python checks `except` clauses top to bottom and uses the first match) allows a program to give a precise, actionable message for each anticipated failure mode, while still catching any other operating-system-level problem gracefully rather than crashing outright. Full coverage of exception-handling mechanics, including `try`/`except`/`else`/`finally` and custom exceptions, is in the existing Exception Handling handbook; this section focuses specifically on the file-related exceptions that arise in practice.

### 19.3 Validating Before Acting

```python
import os

path = 'report.csv'
if not os.path.exists(path):
    print('File does not exist — cannot continue')
elif os.path.isdir(path):
    print('Expected a file but found a directory')
else:
    with open(path, 'r', encoding='utf-8') as f:
        pass
```

Checking conditions before attempting an operation ("look before you leap") can be combined with `try`/`except` ("easier to ask forgiveness than permission") — Python code commonly favors the latter for file operations specifically, since a check-then-act sequence has a small race-condition window where the file's state could change between the check and the actual operation; wrapping the actual operation in `try`/`except` closes that gap.

## 20. Large Files and Memory Efficiency

### 20.1 The Problem With read() on Huge Files

Calling `.read()` or `.readlines()` on a multi-gigabyte file attempts to load the entire content into RAM at once, which can slow a program to a crawl or crash it outright with a `MemoryError` on a machine that does not have enough free memory to hold it.

### 20.2 Reading Line by Line (Streaming)

```python
with open('huge_log.txt', 'r', encoding='utf-8') as f:
    for line in f:
        if 'CRITICAL' in line:
            print(line.strip())
```

As established in Section 4.5, iterating a file object directly never loads more than the current line into memory, making this the standard approach for processing arbitrarily large text files.

### 20.3 Chunk Processing for Binary Files

```python
CHUNK_SIZE = 1024 * 1024  # 1 MB

with open('large_file.bin', 'rb') as src, open('copy.bin', 'wb') as dst:
    while True:
        chunk = src.read(CHUNK_SIZE)
        if not chunk:
            break
        dst.write(chunk)
```

| Code Line | Explanation |
|---|---|
| `src.read(CHUNK_SIZE)` | Reads a fixed, bounded amount of data at a time, rather than the whole file. |
| `if not chunk: break` | An empty `bytes` object (`b''`) signals end of file, exactly as an empty string does in text mode (Section 4.6); the loop exits once nothing more remains. |

### 20.4 Generators for Custom Streaming Logic

```python
def read_in_batches(path, batch_size=1000):
    with open(path, 'r', encoding='utf-8') as f:
        batch = []
        for line in f:
            batch.append(line.strip())
            if len(batch) == batch_size:
                yield batch
                batch = []
        if batch:
            yield batch

for batch in read_in_batches('huge_dataset.txt'):
    process(batch)
```

A generator function that `yield`s batches instead of `return`ing a full list lets calling code process large datasets in manageable groups, combining the memory efficiency of line-by-line reading with the convenience of working on several records at once.

### 20.5 Buffering

Python's file objects are buffered by default, meaning data is not necessarily written to physical disk the instant `write()` is called; it first accumulates in an internal buffer and is flushed (physically written) either when the buffer fills, when the file is closed, or when `f.flush()` is called explicitly. This buffering is what makes repeated small writes efficient, since it avoids an expensive disk operation for every single `write()` call.

```python
with open('log.txt', 'a', encoding='utf-8') as f:
    f.write('Event occurred\n')
    f.flush()  # force this line to disk immediately, without waiting to close
```

## 21. File Processing Workflows

### 21.1 The General Pattern

A professional file-processing workflow typically follows a consistent shape: locate the input files, open and read them, validate the data, transform it into the desired form, write the output, and log what happened — with error handling wrapped around every stage that could plausibly fail.

### 21.2 A Complete Worked Workflow

```python
import csv
import logging
from pathlib import Path

logging.basicConfig(filename='processing.log', level=logging.INFO)

def process_sales_file(path):
    processed = []
    with open(path, 'r', encoding='utf-8', newline='') as f:
        reader = csv.DictReader(f)
        for row_number, row in enumerate(reader, start=2):
            try:
                amount = float(row['amount'])
                if amount < 0:
                    raise ValueError('negative amount')
                processed.append({'id': row['id'], 'amount': amount})
            except (ValueError, KeyError) as e:
                logging.warning(f'Row {row_number} skipped: {e}')
    return processed

def write_output(records, path):
    with open(path, 'w', encoding='utf-8', newline='') as f:
        writer = csv.DictWriter(f, fieldnames=['id', 'amount'])
        writer.writeheader()
        writer.writerows(records)

for source_file in Path('incoming').glob('*.csv'):
    results = process_sales_file(source_file)
    write_output(results, Path('processed') / source_file.name)
    logging.info(f'Processed {source_file.name}: {len(results)} valid rows')
```

This example ties together directory searching (`glob`, Part 22), CSV reading and writing (Part 14), validation with `try`/`except` (Part 19), and logging (Part 27) into a single realistic, production-style pipeline that finds every incoming CSV file, cleans and validates it, and writes a corresponding output file while recording what happened for later review.

### 21.3 Archiving Processed Files

```python
import shutil
from pathlib import Path

source = Path('incoming/sales_january.csv')
archive_dir = Path('archive')
archive_dir.mkdir(exist_ok=True)
shutil.move(str(source), str(archive_dir / source.name))
```

Moving a source file into an archive folder once it has been successfully processed is a common practical step that prevents the same file from being accidentally reprocessed on a later run.

## 24. File Permissions

### 24.1 Read, Write, and Execute Permissions

On Unix-like systems (Linux, macOS), every file carries three permission categories — owner, group, and others — each with independent read, write, and execute permissions, together forming the familiar `rwxr-xr--`-style permission string seen in a terminal's `ls -l` output.

### 24.2 Checking Permissions With os.access()

```python
import os
print(os.access('data.csv', os.R_OK))
print(os.access('data.csv', os.W_OK))
print(os.access('script.sh', os.X_OK))
```

`os.access()` checks whether the *current* process actually has the requested permission for a given path, returning a simple boolean, which is often more directly useful than parsing a raw permission bitmask.

### 24.3 The stat Module

```python
import stat
import os

mode = os.stat('config.json').st_mode
print(oct(stat.S_IMODE(mode)))
print(bool(mode & stat.S_IRUSR))
```

| Code Line | Explanation |
|---|---|
| `stat.S_IMODE(mode)` | Extracts just the permission bits from the full mode value returned by `os.stat()`, discarding the file-type bits. |
| `mode & stat.S_IRUSR` | Uses a bitwise AND (see the Operators handbook, Part 8) with a named constant to test whether one specific permission bit — here, owner-read — is set. |

### 24.4 Changing Permissions

```python
import os
os.chmod('script.sh', 0o755)
```

`os.chmod()` sets a file's permission bits directly using an octal literal (`0o755` is the traditional Unix "owner: read/write/execute, group and others: read/execute" pattern), mirroring the command-line `chmod` utility.

### 24.5 Permission Errors

```python
try:
    with open('/etc/protected_config', 'w') as f:
        f.write('data')
except PermissionError:
    print('Insufficient permission to write to this location')
```

### 24.6 Cross-Platform Differences

Windows uses a fundamentally different, ACL (Access Control List) based permission model rather than the Unix `rwx` bits; `os.chmod()` on Windows has only limited effect (primarily toggling the read-only attribute), so permission-sensitive code intended to run on both platforms needs to be written and tested with this difference explicitly in mind, and should rely on catching `PermissionError` at the point of the actual operation rather than only on `os.access()` pre-checks.

### 24.7 File Ownership (Conceptually)

Every file on a Unix-like system has an owning user and an owning group, which together with the permission bits determine who may read, write, or execute it. Python can inspect ownership through `os.stat().st_uid`/`st_gid`, though changing ownership (`os.chown()`) typically requires elevated administrator privileges and is outside the scope of ordinary application-level file handling.

## 25. Temporary Files and Directories

### 25.1 Why Temporary Files Are Useful

A temporary file provides scratch storage for data that only needs to exist for the duration of a single operation or program run — an intermediate processing stage, a download in progress, or a large computation's working data — without cluttering the file system with files a program is responsible for cleaning up manually.

### 25.2 The tempfile Module

```python
import tempfile

with tempfile.NamedTemporaryFile(mode='w', delete=True, suffix='.txt') as tmp:
    tmp.write('scratch data')
    print(tmp.name)
```

| Code Line | Explanation |
|---|---|
| `tempfile.NamedTemporaryFile(...)` | Creates a genuinely unique, safely named temporary file in the operating system's designated temp directory, avoiding any risk of colliding with another file (or another running process's temp file) by name. |
| `delete=True` | The default; the file is automatically deleted the moment it is closed (including when the `with` block ends), requiring no manual cleanup. |

### 25.3 Temporary Directories

```python
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as tmpdir:
    scratch_path = Path(tmpdir) / 'work.txt'
    scratch_path.write_text('temporary data', encoding='utf-8')
    print(list(Path(tmpdir).iterdir()))
```

`TemporaryDirectory()` creates an entire unique, empty directory for scratch use, and automatically removes it and everything inside it — recursively — once the `with` block ends.

### 25.4 Why Temporary Storage Is Safer Than Manual Naming

Manually constructing a "temporary" file name like `'temp.txt'` risks two other processes (or two runs of the same program) colliding on the same name, and risks the file being left behind if the program crashes before manually deleting it. `tempfile` addresses both problems: names are guaranteed unique, and deletion is tied to the object's lifetime rather than to a manual cleanup step that could be skipped.

### 25.5 Practical Application: Safe Intermediate Processing

```python
import tempfile
import shutil

def safely_rewrite(path, transform_fn):
    with tempfile.NamedTemporaryFile('w', delete=False, dir='.', encoding='utf-8') as tmp:
        with open(path, 'r', encoding='utf-8') as original:
            for line in original:
                tmp.write(transform_fn(line))
        temp_path = tmp.name
    shutil.move(temp_path, path)
```

Writing to a temporary file first, then atomically replacing the original only once the new content is fully and successfully written, is a standard, safe pattern for rewriting a file in place without ever leaving it in a half-written, corrupted state if the program is interrupted partway through.

## 27. Log Files

### 27.1 What Log Files Are

A log file is a text file, typically appended to continuously over time, that records events, status messages, warnings, and errors as a program runs — the primary tool for understanding what a long-running or unattended program actually did, especially after the fact, when something has gone wrong.

### 27.2 Reading and Processing Logs Manually

```python
error_count = 0
with open('app.log', 'r', encoding='utf-8') as f:
    for line in f:
        if 'ERROR' in line:
            error_count += 1
print(f'Found {error_count} error lines')
```

### 27.3 Writing Logs Manually

```python
from datetime import datetime

def log_event(message, path='app.log'):
    timestamp = datetime.now().isoformat()
    with open(path, 'a', encoding='utf-8') as f:
        f.write(f'[{timestamp}] {message}\n')

log_event('Process started')
```

Hand-rolled logging like this works for the simplest cases, but Python's built-in `logging` module (Section 27.4) provides a far more capable, standard, and configurable solution and is strongly preferred for anything beyond a trivial script.

### 27.4 Python's logging Module

```python
import logging

logging.basicConfig(
    filename='app.log',
    level=logging.INFO,
    format='%(asctime)s %(levelname)s %(message)s'
)

logging.info('Process started')
logging.warning('Low disk space')
logging.error('Failed to connect to database')
```

| Code Line | Explanation |
|---|---|
| `logging.basicConfig(filename='app.log', level=logging.INFO, ...)` | Configures the root logger once, at program start, to write to a specific file, at a specific minimum severity level, in a specific text format — every subsequent `logging.info()`/`.warning()`/`.error()` call throughout the program automatically respects this configuration without needing to reopen or manage the file manually. |
| `level=logging.INFO` | Messages below this severity (like `logging.debug()`) are silently ignored; only `INFO` and above (`WARNING`, `ERROR`, `CRITICAL`) are written. |

## 31. Cross-Platform File Handling

### 31.1 Path Separators Recap

Always build paths with `os.path.join()` or `pathlib`'s `/` operator rather than hard-coding `/` or `\` directly, as established in Section 2.7.

### 31.2 File Permissions Recap

Unix-style `rwx` permission bits and `os.chmod()` behave meaningfully on Linux/macOS but only partially on Windows (Section 24.6); code that must run identically on both should not assume full permission-bit control everywhere.

### 31.3 Case Sensitivity

```python
from pathlib import Path
print(Path('Report.CSV') == Path('report.csv'))
```

Linux file systems are case-sensitive, meaning `Report.csv` and `report.csv` are two entirely different files; Windows (and, by default, macOS) file systems are case-*insensitive* (though usually case-*preserving*, keeping the original capitalization for display purposes). Code that hard-codes a specific capitalization and happens to work while developing on Windows can fail with `FileNotFoundError` once deployed to a Linux server — a genuinely common cross-platform deployment bug.

### 31.4 Encoding Differences

As covered in Section 9.6, omitting `encoding=` lets Python fall back to the operating system's default, which differs across platforms; always specifying `encoding='utf-8'` explicitly removes this entire category of cross-platform inconsistency.

### 31.5 Line Ending Differences

Windows traditionally uses `\r\n` for line endings, while Linux and macOS use plain `\n`; Python's universal newline handling in text mode (Section 5.5) normalizes this automatically during reads and writes, so code that always uses text mode (rather than binary mode) for genuinely textual files rarely needs to think about this directly.

### 31.6 Environment Differences

```python
from pathlib import Path
config_dir = Path.home() / '.myapp'
config_dir.mkdir(exist_ok=True)
```

Using `Path.home()` instead of a hard-coded `/home/username` or `C:\Users\username` is a small but important habit that makes code immediately portable across users and operating systems without any conditional logic.

### 31.7 How pathlib Helps Create Portable Code

`pathlib.Path` automatically renders itself using the correct separator for whatever platform the code is actually executing on — a `Path` built with `/` on Linux prints with forward slashes, while the identical code running on Windows prints with backslashes — meaning a single, unmodified code path works correctly everywhere, which is the central practical argument for preferring `pathlib` over manually formatted string paths in new code.

## 32. Security and Safe File Handling

### 32.1 Path Traversal

```python
import os

def unsafe_read(base_dir, user_filename):
    return open(os.path.join(base_dir, user_filename)).read()

# A malicious user_filename like '../../etc/passwd' escapes base_dir entirely
```

**Path traversal** is an attack where a filename supplied by an untrusted source (a web form, an API parameter) contains `..` segments designed to escape the intended directory and access files elsewhere on the file system. Accepting a filename from user input and joining it directly onto a base path without validation is a genuine, serious vulnerability, not a theoretical one.

```python
from pathlib import Path

def safe_read(base_dir, user_filename):
    base = Path(base_dir).resolve()
    target = (base / user_filename).resolve()
    if not target.is_relative_to(base):
        raise ValueError('Invalid path: outside allowed directory')
    return target.read_text(encoding='utf-8')
```

| Code Line | Explanation |
|---|---|
| `target.is_relative_to(base)` | After resolving both paths to their absolute, `..`-collapsed form, checks that the requested target still lives inside the intended base directory, rejecting any path that has escaped it. |

### 32.2 Unsafe Filenames

Filenames supplied by users can also contain characters that are problematic for other reasons — null bytes, unusually long names, or characters with special meaning in shell commands if the filename is ever passed to one; validating and, where appropriate, sanitizing user-supplied filenames before using them is a standard defensive step.

### 32.3 Untrusted File Content

Reading a file's content and treating it as inherently safe to use directly — for instance, feeding it straight into `eval()`, a shell command, or a database query — is dangerous regardless of how the file arrived; content-based validation and safe parsing (using `json.loads()` instead of `eval()` for structured data, for example) should always be applied to any file whose contents are not fully controlled by the program itself.

### 32.4 Unsafe Deserialization — Pickle Security Revisited

As established in Sections 18.4 and 29.8, `pickle.load()` on untrusted data can execute arbitrary code. The defensive rule is simple and absolute: **only unpickle data that the program itself created, or that comes from a fully trusted, verified source.** For data interchange with any external or untrusted party, JSON (or another non-code-executing format) is the appropriate choice instead.

### 32.5 Secure Temporary Files

As covered in Part 25, `tempfile`'s functions create files with restrictive permissions and unpredictable names by design, specifically to prevent other, potentially malicious processes on a shared system from predicting or intercepting temporary file names — a real security consideration on multi-user systems that manually named temp files do not address.

### 32.6 Input Validation Checklist for File Operations

- Never build a file path by directly concatenating untrusted input without validation.
- Always resolve and verify paths stay within an intended base directory when accepting external file names.
- Never call `pickle.load()`, `eval()`, or `exec()` on file content from an untrusted source.
- Validate file size before reading fully into memory, to guard against being sent an unexpectedly enormous file.
- Validate and sanitize file extensions and MIME types when accepting uploaded files, rather than trusting a claimed extension alone.
- Catch and handle `PermissionError` and `OSError` gracefully rather than letting a file operation crash a service.

### 34.1 Project: Student Records File System

**Goal:** Store, update, and query student records using a JSON file as a lightweight database.

```python
import json
from pathlib import Path

DB_PATH = Path('students.json')

def load_records():
    if not DB_PATH.exists():
        return {}
    with open(DB_PATH, 'r', encoding='utf-8') as f:
        return json.load(f)

def save_records(records):
    with open(DB_PATH, 'w', encoding='utf-8') as f:
        json.dump(records, f, indent=2)

def add_student(name, grades):
    records = load_records()
    records[name] = {'grades': grades, 'average': sum(grades) / len(grades)}
    save_records(records)

def get_student(name):
    return load_records().get(name)

add_student('Ada', [95, 88, 91])
add_student('Grace', [82, 79, 85])
print(get_student('Ada'))
```

**Step-by-step:** `load_records()` safely returns an empty dictionary the first time the program runs, before the file exists, avoiding a `FileNotFoundError` on a fresh setup. Every write goes through `save_records()`, which always writes the complete, current state back out — a simple but effective pattern for a small file-backed database where full rewrites are cheap enough to be practical.

### 34.2 Project: Sales CSV Processing System

**Goal:** Read multiple raw sales CSV files, validate and clean the data, and produce a single consolidated summary.

```python
import csv
from pathlib import Path
from collections import defaultdict

def process_all_sales(input_dir):
    totals_by_region = defaultdict(float)
    errors = []
    for file in Path(input_dir).glob('*.csv'):
        with open(file, 'r', encoding='utf-8', newline='') as f:
            reader = csv.DictReader(f)
            for row_num, row in enumerate(reader, start=2):
                try:
                    amount = float(row['amount'])
                    region = row['region'].strip().upper()
                    totals_by_region[region] += amount
                except (ValueError, KeyError) as e:
                    errors.append(f'{file.name} row {row_num}: {e}')
    return dict(totals_by_region), errors

totals, errors = process_all_sales('raw_sales')
for region, total in sorted(totals.items()):
    print(f'{region}: {total:.2f}')
if errors:
    print(f'{len(errors)} rows skipped due to errors')
```

**Step-by-step:** Every CSV file in the input directory is opened and streamed through with `csv.DictReader`, accumulating totals per region into a `defaultdict` (which supplies `0.0` automatically for any region seen for the first time). Malformed rows are collected as descriptive error strings rather than crashing the whole run, so one bad row in one file never prevents the rest of the data from being processed.

### 34.6 Project: Log File Analyzer

**Goal:** Stream a large log file, summarize error frequency by hour, and write a report.

```python
import re
from collections import Counter
from pathlib import Path

LOG_PATTERN = re.compile(r'\[(\d{4}-\d{2}-\d{2} \d{2}):\d{2}:\d{2}\].*ERROR')

def analyze_log(path, report_path):
    hourly_errors = Counter()
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            match = LOG_PATTERN.search(line)
            if match:
                hourly_errors[match.group(1)] += 1
    with open(report_path, 'w', encoding='utf-8') as f:
        for hour, count in sorted(hourly_errors.items()):
            f.write(f'{hour}: {count} errors\n')

analyze_log('server.log', 'error_summary.txt')
```

```python
import csv

REQUIRED_COLUMNS = {'id', 'name', 'email', 'signup_date'}

def validate_dataset(path):
    issues = []
    with open(path, 'r', encoding='utf-8', newline='') as f:
        reader = csv.DictReader(f)
        missing = REQUIRED_COLUMNS - set(reader.fieldnames or [])
        if missing:
            issues.append(f'Missing required columns: {missing}')
            return issues
        for row_num, row in enumerate(reader, start=2):
            if not row['email'] or '@' not in row['email']:
                issues.append(f'Row {row_num}: invalid email')
            if not row['id']:
                issues.append(f'Row {row_num}: missing id')
    return issues

problems = validate_dataset('new_signups.csv')
if problems:
    for p in problems:
        print(p)
else:
    print('Dataset passed validation')
```

### 34.9 Project: Machine Learning Model Persistence System

### 34.10 Project: AI Data Pipeline File Manager

**Goal:** Manage a folder of `.jsonl` training examples: validate, deduplicate, and split into train/validation files.

```python
import json
import random
from pathlib import Path

def load_jsonl(path):
    with open(path, 'r', encoding='utf-8') as f:
        return [json.loads(line) for line in f if line.strip()]

def save_jsonl(records, path):
    with open(path, 'w', encoding='utf-8') as f:
        for record in records:
            f.write(json.dumps(record) + '\n')

def prepare_training_data(input_path, train_path, val_path, val_ratio=0.2, seed=42):
    records = load_jsonl(input_path)
    seen = set()
    deduped = []
    for r in records:
        key = json.dumps(r, sort_keys=True)
        if key not in seen:
            seen.add(key)
            deduped.append(r)

    random.Random(seed).shuffle(deduped)
    split_point = int(len(deduped) * (1 - val_ratio))
    save_jsonl(deduped[:split_point], train_path)
    save_jsonl(deduped[split_point:], val_path)
    return len(deduped[:split_point]), len(deduped[split_point:])

train_count, val_count = prepare_training_data(
    'raw_examples.jsonl', 'train.jsonl', 'val.jsonl'
)
print(f'Train: {train_count}, Validation: {val_count}')
```

**Q1. What is the difference between `'w'` and `'a'` mode?**
Answer: `'w'` truncates (erases) the file's existing content the moment it is opened, before any writing happens; `'a'` preserves existing content and adds all new writes to the end of the file.

**Q2. Why should `with open(...)` be preferred over manual `open()`/`close()`?**
Answer: `with` guarantees the file is closed automatically, even if an exception occurs inside the block, eliminating an entire class of resource-leak bugs that manual `close()` calls are vulnerable to if an error skips over them.

**Q3. What does `read()` return when called at the end of a file?**
Answer: An empty string `''` in text mode, or an empty `bytes` object `b''` in binary mode — this is the standard way Python code detects end of file.

**Q4. What is the difference between `os.path.exists()` and `os.path.isfile()`?**
Answer: `exists()` returns `True` for anything at that path, whether it is a file or a directory; `isfile()` returns `True` only if it is specifically a file.

**Q5. Why is it good practice to specify `encoding='utf-8'` explicitly?**
Answer: Without it, Python falls back to the operating system's default encoding, which can differ between platforms and cause a file written on one machine to fail to read correctly, or raise a `UnicodeDecodeError`, on another.

### 35.2 Intermediate Questions

**Q6. What is the difference between `json.load()` and `json.loads()`?**
Answer: `load()` reads directly from an open file object; `loads()` parses a JSON string that is already held in memory. The trailing `s` signals "string".

**Q7. Why does the `csv` module recommend opening files with `newline=''`?**
Answer: So the `csv` module can manage line endings itself correctly, rather than Python's own universal-newline translation interfering with how the module detects and writes row boundaries, which can otherwise produce extra blank lines on some platforms.

**Q8. What happens if you call `.read()` on a very large file?**
Answer: The entire file's content is loaded into memory at once, which can be slow or exhaust available memory; iterating the file object line by line, or reading in fixed-size chunks, avoids this.

**Q9. What is the practical difference between `os.rename()` and `os.replace()`?**
Answer: `os.replace()` is guaranteed to overwrite the destination atomically if it already exists; `os.rename()`'s behavior in that situation can be platform-inconsistent, making `os.replace()` the safer general choice.

**Q10. Why is `pickle` considered unsafe for untrusted data, while `json` is not?**
Answer: Unpickling data can execute arbitrary code embedded in the pickle stream as part of deserialization; JSON parsing only ever produces basic data structures (dicts, lists, strings, numbers) and has no mechanism to execute code at all.

### 35.3 Advanced Questions

**Q11. Explain why `seek()` can be risky in text mode but safe in binary mode.**
Answer: Text mode files are decoded through a character encoding, and multi-byte encodings like UTF-8 mean an arbitrary byte offset can land in the middle of a multi-byte character, causing a `UnicodeDecodeError`; every byte offset is inherently valid in binary mode, since there is no character decoding involved.

**Q12. What is the difference between `shutil.copy()` and `shutil.copy2()`?**
Answer: Both copy file content and permission bits; `copy2()` additionally attempts to preserve file metadata such as modification time, making it a closer approximation of a "true" copy.

**Q13. How does `os.walk()` differ from `Path.rglob()`?**
Answer: `os.walk()` yields, for every directory in the tree, a tuple of `(dirpath, dirnames, filenames)`, giving fine-grained control over which subdirectories to descend into (by mutating `dirnames` in place); `rglob()` simply yields every matching `Path` recursively, which is more concise but offers less control over traversal.

**Q14. Why might `joblib` be preferred over raw `pickle` for a scikit-learn model?**
Answer: `joblib` is optimized specifically for objects containing large NumPy arrays, which is exactly what most trained ML models consist of internally, generally resulting in faster and more efficient serialization than raw `pickle` for that specific kind of data.

**Q15. What is a path traversal vulnerability, and how can it be prevented?**
Answer: It occurs when untrusted input containing `..` segments is used to build a file path, allowing it to escape an intended base directory; it is prevented by resolving both the base directory and the requested path to their absolute forms and verifying the target still lies within the base directory before using it.

### 35.4 Code-Output Questions

**Q16. What does the following print?**

```python
with open('data.txt', 'w', encoding='utf-8') as f:
    f.write('A')
    f.write('B')
with open('data.txt', 'r', encoding='utf-8') as f:
    print(f.read())
```

Answer: `AB`. `write()` never inserts a newline automatically, so the two calls produce one continuous line.

**Q17. What does the following print, assuming `numbers.txt` does not yet exist?**

```python
try:
    with open('numbers.txt', 'r') as f:
        print(f.read())
except FileNotFoundError:
    print('Not found')
```

Answer: `Not found`, since `'r'` mode requires the file to already exist.

### 35.5 Debugging Questions

**Q18. This code is supposed to append a log entry every time it runs, but it only ever contains the last entry. What is wrong?**

```python
with open('log.txt', 'w', encoding='utf-8') as f:
    f.write('Event happened\n')
```

Answer: It uses `'w'` mode, which truncates the file on every run; it should use `'a'` mode to append instead.

**Q19. This code raises `UnicodeDecodeError` intermittently on different machines. What is the likely cause and fix?**

```python
with open('notes.txt', 'r') as f:
    content = f.read()
```

Answer: No `encoding` was specified, so Python falls back to the platform's default encoding, which can vary; explicitly passing `encoding='utf-8'` (matching how the file was actually written) fixes it.

### 35.6 Scenario-Based Questions

**Q20. You need to process a 40 GB CSV file on a machine with 8 GB of RAM. What approach do you take?**
Answer: Avoid `pandas.read_csv()` without chunking and avoid `.read()`/`.readlines()`; stream the file with `csv.reader`/`csv.DictReader` iterating line by line, or use `pandas.read_csv(path, chunksize=...)`, processing and discarding each chunk before moving to the next, so memory usage stays roughly constant regardless of total file size.

### 35.7 Data Science File-Handling Questions

**Q21. Why might a Data Scientist choose `pandas.read_csv()` over the built-in `csv` module for a large tabular dataset?**
Answer: `pandas` provides vectorized parsing, automatic type inference, missing-value handling, and a `DataFrame` interface suited directly to analysis, whereas the built-in `csv` module requires manually building that structure and handling those concerns row by row.

### 35.8 AI Engineering File-Handling Questions

**Q22. Why is JSON Lines (`.jsonl`) commonly preferred over a single large JSON array for AI training datasets?**
Answer: Each line is an independently parseable JSON object, which allows the dataset to be streamed and processed one example at a time without loading the entire file into memory, and allows new examples to be appended cheaply without rewriting the whole file — both of which a single top-level JSON array cannot support.

**Exercise I1.** Write a function that counts the number of lines in a text file without loading the whole file into memory at once.

```python
def count_lines(path):
    count = 0
    with open(path, 'r', encoding='utf-8') as f:
        for _ in f:
            count += 1
    return count
```

*Explanation:* Iterating the file object directly (Section 4.5) processes one line at a time, keeping memory usage constant regardless of the file's total size.

### 36.3 Advanced Exercises

**Exercise A1.** Write a function that safely copies a large binary file in 1 MB chunks, without loading the entire file into memory.

*Solution:*

```python
def copy_large_file(source, destination, chunk_size=1024 * 1024):
    with open(source, 'rb') as src, open(destination, 'wb') as dst:
        while True:
            chunk = src.read(chunk_size)
            if not chunk:
                break
            dst.write(chunk)
```

**Exercise A2.** Write a function that reads a CSV file and returns only the rows where a given column's value, converted to `float`, exceeds a given threshold, skipping any row that fails to convert without crashing.

*Solution:*

```python
import csv

def filter_rows_above(path, column, threshold):
    results = []
    with open(path, 'r', encoding='utf-8', newline='') as f:
        reader = csv.DictReader(f)
        for row in reader:
            try:
                if float(row[column]) > threshold:
                    results.append(row)
            except (ValueError, KeyError):
                continue
    return results
```

### 36.4 Challenge Exercise

**Exercise C1.** Write a function that takes a directory of `.jsonl` files, merges them into a single deduplicated `.jsonl` file, and returns the count of records written and the count of duplicates removed.

*Solution:*

```python
import json
from pathlib import Path

def merge_jsonl_files(input_dir, output_path):
    seen = set()
    written = 0
    duplicates = 0
    with open(output_path, 'w', encoding='utf-8') as out:
        for file in Path(input_dir).glob('*.jsonl'):
            with open(file, 'r', encoding='utf-8') as f:
                for line in f:
                    line = line.strip()
                    if not line:
                        continue
                    key = json.dumps(json.loads(line), sort_keys=True)
                    if key in seen:
                        duplicates += 1
                        continue
                    seen.add(key)
                    out.write(line + '\n')
                    written += 1
    return written, duplicates
```

*Explanation:* Each record is normalized with `sort_keys=True` before being used as a deduplication key, ensuring that two JSON objects with the same content but differently ordered keys are still correctly recognized as duplicates.

## Conclusion

File handling is where a Python program stops being a self-contained calculation and starts becoming a genuine tool: something that remembers, that exchanges data with other systems, that survives being switched off, and that can process data far larger than any single variable could ever hold in memory at once. The mechanics covered in this handbook — modes, context managers, encodings, cursors, streaming, and the specialized modules for CSV, JSON, pickle, and ZIP — are not simply syntax to memorize, but a coherent set of tools for a single underlying job: moving data reliably and safely between a running program and the durable world of the file system.

The distinction that matters most in practice is rarely which single function to call, but which *pattern* fits the situation: `with open(...)` over manual `close()`, `pathlib` over hard-coded path strings, streaming over `.read()` for anything that might be large, JSON over pickle for anything that might be untrusted, and explicit encodings over silent platform defaults. Internalizing these defaults, and understanding precisely why each one is the safer or more portable choice, is what separates file-handling code that merely works on one developer's machine from file-handling code that is genuinely production-ready — across operating systems, across file sizes, and across the Data Analytics, Data Science, and Machine Learning pipelines this handbook has kept in view throughout.
