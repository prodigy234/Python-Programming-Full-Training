# 1. Introduction to Modules

## 1.1 What Is a Module?

A **module** is simply a file containing Python code - variables, functions, classes, or runnable statements - that can be imported and reused in another file. In practice, almost any `.py` file can act as a module the moment another file imports it. There is no special syntax required to "declare" a file as a module; the moment you write `import something`, Python treats `something.py` as a module.

Modules are the foundation of how Python code is organized. Instead of writing every line of a program inside a single giant file, you split related code into separate files, and then bring the pieces together using imports.

## 1.2 Why Modules Exist

| Reason | Explanation |
|---|---|
| Code organization | Related functionality lives together in one place, instead of one file containing everything. |
| Code reuse | A function written once can be imported and used in many different scripts or projects. |
| Maintainability | Bugs are easier to find and fix when related code is grouped logically rather than scattered. |
| Namespace isolation | Each module has its own namespace, so two modules can define a function with the same name without conflict. |
| Collaboration | Different team members can work on different modules of the same project simultaneously. |
| Testing | Individual modules can be tested in isolation, independent of the rest of the application. |

## 1.3 Code Organization and Namespace Isolation

Without modules, every name (variable, function, class) in a program would compete for space in a single global namespace, and name collisions would be constant. Modules solve this by giving each file its own private namespace. A function called `load_data` defined inside `data_loader.py` does not clash with a different `load_data` defined inside `report_loader.py`, because each lives inside its own module's namespace and must be accessed through that module (`data_loader.load_data` vs `report_loader.load_data`).

## 1.4 Python's Module System and .py Files as Modules

Every `.py` file is a potential module. When Python imports a file, it executes the file's code top to bottom exactly once, and the resulting names (functions, classes, variables) become attributes of a module object. That module object is then bound to a name in the importing file's namespace.

````python
# greetings.py
def say_hello(name):
    return f"Hello, {name}!"

LANGUAGE = "English"
````

````python
# main.py
import greetings

print(greetings.say_hello("Ada"))
print(greetings.LANGUAGE)
````

| Line | Explanation |
|---|---|
| `import greetings` | Python locates `greetings.py`, executes it top to bottom, and binds the resulting module object to the name `greetings` in `main.py`. |
| `greetings.say_hello("Ada")` | Accesses the `say_hello` function through the module's namespace using dot notation. |
| `greetings.LANGUAGE` | Accesses a module-level variable the same way, through the module object. |

## 1.5 Script vs Module vs Package vs Library vs Framework vs Application

These six terms are often used loosely, but each has a precise meaning in professional Python development.

| Term | Definition |
|---|---|
| Script | A `.py` file that is meant to be *run directly* to perform a task, typically top to bottom, from start to finish (e.g. `python cleanup.py`). |
| Module | A `.py` file meant to be *imported* and reused by other code, providing functions, classes, or constants. |
| Package | A directory of related modules, marked as a Python package (traditionally with an `__init__.py` file), that groups modules under a common namespace. |
| Library | A collection of one or more packages/modules, published and versioned together, intended to be reused across many different projects (e.g. `requests`, `numpy`). |
| Framework | A larger, opinionated collection of libraries that dictates the overall structure and flow of your application, calling *your* code rather than the other way around (e.g. Django, FastAPI). |
| Application | A complete, runnable program built by combining your own modules, packages, and third-party libraries to accomplish a specific real-world purpose. |

**Key distinction - "who calls whom":** with a library, *your code* calls the library's functions. With a framework, the *framework* calls your code at the appropriate points (this is often called "inversion of control").

A single `.py` file can be both a script and a module at the same time - runnable directly, and also importable elsewhere. Section 8 (`__name__`) explains exactly how Python code detects which mode it is running in.


# 2. Creating Your Own Modules

## 2.1 Creating a .py File as a Module

Any text file saved with a `.py` extension can serve as a module. There is no special header or declaration required - the file's content (its functions, classes, and variables) becomes importable the moment another file imports it.

````python
# calculations.py
"""Simple arithmetic helper functions for the finance app."""

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

TAX_RATE = 0.075
````

## 2.2 Importing and Accessing Module Members

````python
import calculations

total = calculations.add(100, 50)
after_tax = total * (1 + calculations.TAX_RATE)
print(total, after_tax)
````

Every function, class, and variable defined at the top level of `calculations.py` becomes an attribute reachable through `calculations.<name>`, exactly as an object's attributes are reached through dot notation.

## 2.3 Using Module Aliases

````python
import calculations as calc

print(calc.add(2, 3))
````

Aliasing is covered in complete depth in Part 5, but it is worth introducing here: `as` simply binds the imported module to a different name in the current namespace, which is especially useful for long or frequently used module names.

## 2.4 Organizing Related Functionality

As a project grows, group related functions and classes into the same module, and split unrelated functionality into separate modules. A practical rule of thumb: if you find yourself scrolling through hundreds of lines to find one function, or if a module serves two unrelated purposes, it is time to split it.

```
project/
├── main.py
├── calculations.py     # arithmetic and formulas
├── formatting.py       # text and number formatting
└── validation.py       # input validation
```

## 2.5 A Progressive Example: From Script to Reusable Module

**Beginner version** - everything in one file:

````python
# report.py
sales = [120, 340, 560]
total = sum(sales)
print(f"Total sales: {total}")
````

**Professional version** - logic extracted into a reusable module:

````python
# sales_utils.py
def total_sales(sales):
    return sum(sales)

def average_sales(sales):
    return sum(sales) / len(sales) if sales else 0
````

````python
# report.py
from sales_utils import total_sales, average_sales

sales = [120, 340, 560]
print(f"Total: {total_sales(sales)}")
print(f"Average: {average_sales(sales):.2f}")
````

The second version separates *what is computed* (in `sales_utils.py`, fully reusable and testable) from *how it is reported* (in `report.py`). This separation of concerns is one of the most important habits a professional Python developer builds early.


# 3. The import Statement

## 3.1 import module

````python
import math

print(math.sqrt(16))
print(math.pi)
````

`import math` does three things, in order: it locates the `math` module, executes its code (for a C-implemented module like `math`, this means initializing the module object), and binds the name `math` in the current namespace to that module object. Every subsequent reference must go through `math.` - a plain `sqrt(16)` without the prefix would raise a `NameError`.

## 3.2 import module as alias

````python
import numpy as np

array = np.array([1, 2, 3])
````

Covered fully in Part 5.

## 3.3 Importing Multiple Modules

````python
import os
import sys
import json
````

Each module gets its own `import` line. While `import os, sys, json` is technically legal, PEP 8 (Python's official style guide) recommends one import per line, since it makes diffs cleaner and makes it immediately obvious which line to remove if a dependency is no longer needed.

## 3.4 Module Namespaces and Accessing Members

Every imported module keeps its own namespace, entirely separate from the namespace that imported it. This is precisely why `import` (as opposed to `from ... import`, covered next) is considered the "safest" style of import - it can never silently overwrite a name already defined in your file, because everything stays behind the `module.` prefix.

````python
import datetime

now = datetime.datetime.now()
print(now.year)
````

## 3.5 What Python Actually Does When import Executes

Understanding the exact sequence of steps behind `import module_name` demystifies a huge amount of import-related behavior (including caching, covered in Part 23, and circular imports, covered in Part 14).

| Step | What Happens |
|---|---|
| 1 | Python checks `sys.modules` (a cache of already-imported modules) to see if this module has already been imported in this process. |
| 2 | If it is already cached, Python simply binds the existing module object to the name - the module's code is **not** re-executed. |
| 3 | If it is not cached, Python searches `sys.path` (covered in Part 10) to locate the module's source file. |
| 4 | Python creates a new, empty module object and immediately adds it to `sys.modules`, *before* running any of its code. |
| 5 | Python executes the module's code top to bottom inside that module object's namespace, populating it with functions, classes, and variables. |
| 6 | Python binds the now fully-populated module object to the requested name in the importing file's namespace. |

Step 4 - inserting the module into `sys.modules` *before* executing its code - is precisely the mechanism that makes circular imports behave the way they do, since a second import of an already-in-progress module returns the partially built module object rather than starting over.

## 3.6 Import Execution and Caching (Preview)

````python
import config
import config  # this second import does NOT re-run config.py
````

The second `import config` on the second line is essentially a no-op beyond a namespace lookup; Python finds `config` already in `sys.modules` and skips re-execution entirely. This is explored fully in Part 23.


# 4. from ... import

## 4.1 from module import name

````python
from math import sqrt

print(sqrt(16))
````

Unlike `import math`, this pulls a *specific name* directly into the current namespace. You can now call `sqrt(16)` without any prefix - but you also no longer have access to `math` itself unless you import it separately.

## 4.2 Importing Multiple Names

````python
from math import sqrt, pi, floor

print(sqrt(9), pi, floor(4.7))
````

## 4.3 from module import name as alias

````python
from datetime import datetime as dt

now = dt.now()
````

## 4.4 Importing Classes, Functions, and Constants

````python
from collections import OrderedDict   # a class
from math import sqrt                 # a function
from math import pi                   # a constant
````

`from ... import` works identically regardless of whether the imported name refers to a function, a class, or a plain value - Python does not distinguish between these at the syntax level.

## 4.5 Namespace Differences: import vs from ... import

| Style | What Lands in Your Namespace | Access Pattern |
|---|---|---|
| `import module` | Only the name `module` itself | `module.name` |
| `from module import name` | The name `name` directly | `name` |

This has a real practical consequence: `from module import name` can silently shadow (overwrite) an existing name in your file if it happens to share the same name, whereas plain `import module` never can, since everything stays namespaced behind `module.`.

## 4.6 Why Wildcard Imports Should Generally Be Avoided

````python
from math import *

print(sqrt(16))
print(pi)
````

`from module import *` imports *every public name* from a module directly into the current namespace. While convenient to type, it is discouraged in professional code for several concrete reasons:

| Problem | Explanation |
|---|---|
| Name collisions | Two wildcard-imported modules can silently overwrite each other's names, and the last import wins with no warning. |
| Unclear origin | Reading `result = calculate(x)` gives no clue which module `calculate` came from, making code much harder to navigate. |
| Accidental shadowing | A wildcard import can silently replace a builtin or a name you already defined. |
| Tooling limitations | Linters, IDEs, and static analyzers cannot reliably determine what names exist after a wildcard import. |

Wildcard imports are occasionally acceptable in small, throwaway interactive scripts (or when a module explicitly defines `__all__` to curate exactly what a wildcard import exposes - see Part 22), but explicit imports are the professional standard.

# 7. Module Namespaces

## 7.1 What a Namespace Is

A namespace is a mapping from names (identifiers) to objects - conceptually similar to a dictionary where the keys are variable/function/class names and the values are the actual objects those names refer to. Python maintains several distinct namespaces at once, and understanding how they relate is essential to understanding imports.

## 7.2 The Four Namespace Levels

| Namespace | Scope | Example |
|---|---|---|
| Built-in namespace | Names always available everywhere (`print`, `len`, `Exception`) | Created when the interpreter starts, lives for the whole process. |
| Global (module) namespace | Names defined at the top level of a single module | Each module gets its own; this is what `module.name` accesses from outside. |
| Local (function) namespace | Names defined inside a function while it runs | Created fresh each time the function is called, destroyed when it returns. |
| Enclosing namespace | Names in an outer function, visible to a nested inner function | Relevant to closures; distinct from the global namespace. |

## 7.3 Name Resolution: the LEGB Rule

When Python encounters a name, it searches for it in this order, stopping at the first match: **L**ocal, **E**nclosing, **G**lobal, **B**uilt-in.

````python
x = "global x"

def outer():
    x = "enclosing x"
    def inner():
        x = "local x"
        print(x)
    inner()

outer()
print(x)
````

## 7.4 How Imports Affect Namespaces

````python
import math          # binds "math" in the global namespace
from math import pi  # binds "pi" directly in the global namespace
````

`import module` adds exactly one name (`module`) to the current namespace. `from module import name` adds `name` directly - which is why it carries a higher risk of overwriting an existing name, as covered in Part 4.

## 7.5 Module Namespace as an Object's \_\_dict\_\_

A module's namespace is, under the hood, literally a dictionary accessible as `module.__dict__`.

````python
import math

print("pi" in math.__dict__)   # True
print(math.__dict__["pi"])     # 3.14159...
````

This is the same mechanism that powers every attribute lookup in Python - accessing `math.pi` is equivalent to looking up `"pi"` inside `math.__dict__`.

## 7.6 Attribute Access and Name Resolution Across Modules

````python
# config.py
DEBUG = True

# main.py
import config
print(config.DEBUG)      # attribute access, not a plain name lookup
config.DEBUG = False     # this modifies config's namespace, visible everywhere config is imported
````

Because every import of `config` after the first returns the *same cached module object* (Part 23), mutating an attribute on that module object is visible from every other file that has already imported it - a subtle but important consequence of how module namespaces work.


# 8. \_\_name\_\_

## 8.1 What \_\_name\_\_ Represents

Every module, while running, has an automatically created variable called `__name__` that holds a string identifying that module. Its value depends entirely on *how* the module was started, not on anything you write in the file.

| How the File Runs | Value of `__name__` |
|---|---|
| Run directly (`python myfile.py`) | `"__main__"` |
| Imported by another file (`import myfile`) | `"myfile"` (the module's own name) |

## 8.2 A Practical Demonstration

````python
# greet.py
print(f"greet.py's __name__ is: {__name__}")

def say_hi():
    print("Hi!")
````

Running `python greet.py` directly prints `greet.py's __name__ is: __main__`. But if another file does `import greet`, the same print statement now outputs `greet.py's __name__ is: greet`, because it was imported rather than executed directly.

## 8.3 Why Scripts Use if \_\_name\_\_ == "\_\_main\_\_":

````python
# analysis.py
def clean_data(rows):
    return [r for r in rows if r is not None]

def main():
    raw = [1, None, 3, None, 5]
    print(clean_data(raw))

if __name__ == "__main__":
    main()
````

This guard ensures `main()` only runs when `analysis.py` is executed directly - not when another file imports it purely to reuse `clean_data`. Without the guard, importing `analysis` anywhere else would immediately trigger `main()` to run as a side effect, which is almost always unwanted.

## 8.4 Scripts vs Imported Modules

| Aspect | Script Usage | Module Usage |
|---|---|---|
| Entry point | `python file.py` | `import file` |
| `__name__` value | `"__main__"` | The module's actual name |
| Typical intent | Run a task now | Reuse functions/classes later |
| Best practice | Guard the runnable logic with `if __name__ == "__main__":` | Keep import-time side effects minimal |

## 8.5 Practical Applications

**Dual-purpose files**, both importable and directly runnable for quick testing:

````python
# validators.py
def is_valid_email(text):
    return "@" in text and "." in text

if __name__ == "__main__":
    # quick manual test when run directly: python validators.py
    print(is_valid_email("ada@example.com"))
````

**Command-line entry points** for small tools:

````python
# cli_tool.py
import sys

def run(args):
    print(f"Processing: {args}")

if __name__ == "__main__":
    run(sys.argv[1:])
````

**Avoiding accidental execution on import** - one of the most common real-world bugs in beginner code is a module that fetches data, trains a model, or writes a file as soon as it is imported, simply because that code was never wrapped in the `__name__` guard.


# 9. \_\_main\_\_.py

## 9.1 What \_\_main\_\_.py Is

`__main__.py` is a special filename Python recognizes when a *package* (a directory, not a single file) is executed directly. It defines what happens when the package itself, rather than one specific module inside it, is run.

```
myapp/
├── __init__.py
├── __main__.py
├── core.py
└── utils.py
```

````python
# myapp/__main__.py
from myapp.core import run

if __name__ == "__main__":
    run()
````

## 9.2 Executing Packages With python -m

```
python -m myapp
```

This command tells Python: "treat `myapp` as a package, and run its `__main__.py`." This is the same mechanism behind well-known commands like `python -m venv`, `python -m pip`, and `python -m http.server` - each of those is a package with its own `__main__.py`.

## 9.3 Why python -m Matters

| Benefit | Explanation |
|---|---|
| Correct `sys.path` setup | `python -m` adds the current directory to `sys.path` in a way that matches how the package would behave once installed, avoiding a class of import errors. |
| Package-relative imports work correctly | Running `python myapp/core.py` directly can break relative imports inside `core.py`; `python -m myapp` does not have this problem. |
| Consistent execution across environments | Works identically whether the package is run from source or after being installed via `pip`. |

## 9.4 Package Entry Points and CLI Applications

`__main__.py` is the standard way to make an installable package directly runnable as a command-line tool without requiring users to know the internal module structure.


Combined with `pyproject.toml` entry points (Part 28), this same logic can also be exposed as a standalone command (e.g. typing `myapp` directly in the terminal after installation), which is how most professional Python CLI tools are packaged and distributed.


# 10. Module Search and Import Paths

## 10.1 How Python Finds a Module

When you write `import something`, Python needs to locate the actual file on disk. It does this by searching, in order, through a list of directories called `sys.path`.

````python
import sys
for p in sys.path:
    print(p)
````

## 10.2 What Populates sys.path

| Source | Description |
|---|---|
| The script's own directory | The directory containing the file being run is always searched first (or the current directory, for interactive sessions). |
| `PYTHONPATH` | An optional environment variable listing extra directories to search, set outside of Python. |
| Standard library locations | The directories containing Python's built-in modules (`math`, `os`, etc.), installed alongside the interpreter itself. |
| `site-packages` | The directory where third-party packages installed via `pip` are placed. |

## 10.3 Module Discovery Order

Python searches `sys.path` from first entry to last, and uses the *first* matching module it finds. This has a real practical consequence covered in Part 35: naming your own file `math.py` in a project directory can shadow the standard library's `math` module, since your local directory is typically searched before the standard library.

````python
import sys
print(sys.path[0])   # typically the running script's directory
````

## 10.4 site-packages

```
venv/lib/python3.12/site-packages/
├── numpy/
├── pandas/
└── requests/
```

This is where `pip install` places third-party packages by default. Every package you install becomes importable because its directory lives somewhere on `sys.path`, typically inside `site-packages`.

## 10.5 PYTHONPATH

```
export PYTHONPATH=/home/user/shared_modules
```

Setting the `PYTHONPATH` environment variable adds extra directories to `sys.path` before the interpreter even starts, useful for sharing internal modules across multiple projects without formally packaging them. In modern professional workflows, an editable package install (`pip install -e .`, covered in Part 27) is generally preferred over relying on `PYTHONPATH`, since it is more explicit and reproducible.

## 10.6 Inspecting and Modifying sys.path at Runtime

Appending to `sys.path` directly inside code works but is considered a last resort; it is fragile, hard to reproduce across machines, and bypasses the standard packaging tools covered later in this handbook.

## 12.1 Purpose of \_\_init\_\_.py

`__init__.py` is a file placed inside a directory to mark it (in traditional Python packaging) as a package, and to control what happens when that package is imported. Its code runs exactly once, the first time the package (or any module inside it) is imported.

Anything defined directly inside `__init__.py` becomes an attribute of the package itself, exactly as top-level code in a module becomes an attribute of that module.

## 12.3 Exposing Package Members

A very common, professional use of `__init__.py` is re-exporting selected names from internal submodules, so users of the package get a clean, simple import path instead of needing to know the internal file layout.

````python
# project/utils/__init__.py
from .calculations import add, subtract
from .validation import is_valid_email
````

````python
# project/utils/__init__.py
# (intentionally empty)
````

An empty `__init__.py` is completely valid and extremely common - its mere presence (in traditional packaging) is enough to mark the directory as a package. No initialization logic is required unless you specifically want some.

## 12.5 Modern Namespace Packages: When \_\_init\_\_.py Is (or Isn't) Required

Since Python 3.3, **implicit namespace packages** allow a directory to be treated as a package *without* an `__init__.py` at all, as long as Python can locate modules inside it through the normal import machinery. This is especially useful for splitting a single logical package's code across multiple, separately distributed directories.

| Package Type | Requires `__init__.py`? | Typical Use |
|---|---|---|
| Regular package | Traditionally yes (still the common, explicit choice) | Ordinary application and library code where a clear owner directory exists. |
| Namespace package | No | Plugin systems and large ecosystems where multiple independently-installed distributions contribute modules under one shared top-level name. |

**Practical guidance:** for ordinary application and library code, continue to include an `__init__.py` (even if empty) in every package directory. It is explicit, predictable, and lets you control initialization and exports if you ever need to. Reserve namespace packages for the specific, advanced scenario of a plugin ecosystem spanning multiple installable distributions.


# 13. Absolute vs Relative Imports

## 13.1 Absolute Imports

An absolute import spells out the full path to a module, starting from a top-level package (or the project root).

````python
from project.utils.calculations import add
import project.utils.validation
````

Absolute imports are explicit, unambiguous, and the generally recommended default in professional code, since they read the same no matter which file they appear in.

## 13.2 Relative Imports

A relative import instead specifies a module's location *relative to the current file's position inside the package*, using leading dots.

```
project/
├── main.py
└── utils/
    ├── __init__.py
    ├── calculations.py
    └── validation.py
```

````python
# utils/validation.py
from . import calculations          # from the current package
from .calculations import add       # a specific name from a sibling module
````

## 13.3 from . import and from .. import

| Syntax | Meaning |
|---|---|
| `from . import name` | Import `name` from the *current* package. |
| `from .module import name` | Import `name` from a *sibling* module in the current package. |
| `from .. import name` | Import `name` from the *parent* package. |
| `from ..sibling_package import name` | Import `name` from a package next to the current one, one level up. |

```
project/
├── main.py
└── analytics/
    ├── __init__.py
    ├── loading/
    │   ├── __init__.py
    │   └── csv_loader.py
    └── processing/
        ├── __init__.py
        └── cleaning.py       # needs something from loading/
```

````python
# analytics/processing/cleaning.py
from ..loading.csv_loader import load    # go up one level, then into loading
````

## 13.4 When Relative Imports Work (and When They Don't)

Relative imports only work **inside a package that Python recognizes as such through a proper import** - they rely on the special `__package__` attribute being set correctly, which only happens when the file is imported as part of a package, not when it is run directly as a script.

````python
# This FAILS if you run: python utils/validation.py directly
from . import calculations
````

```
ImportError: attempted relative import with no known parent package
```

**The fix:** run the file as part of the package, using `python -m project.utils.validation`, or restructure so the file is only ever imported, never run directly.

## 13.5 Common Import Errors With Relative Imports

| Error | Typical Cause |
|---|---|
| `ImportError: attempted relative import with no known parent package` | Running a file directly with `python file.py` when it uses relative imports. |
| `ImportError: attempted relative import beyond top-level package` | Using too many leading dots (`...`), trying to go above the project's top-level package. |
| `ModuleNotFoundError` | The parent package is missing an `__init__.py`, or the project root is not on `sys.path`. |

**Professional guidance:** many teams prefer absolute imports throughout an entire codebase, reserving relative imports only for very tightly coupled sibling modules within the same small subpackage. Absolute imports are less error-prone when files are moved, refactored, or occasionally run directly for debugging.


# 14. Circular Imports

## 14.1 What Circular Imports Are

A circular import occurs when two (or more) modules import each other, directly or indirectly, creating a dependency loop: module A needs something from module B, and module B needs something from module A.

## 14.2 Why They Occur

Circular imports usually emerge naturally as a codebase grows: two modules that started independent gradually accumulate a two-way relationship, often through a single shared helper function or class that each side has come to depend on.

## 14.3 A Minimal Example

````python
# a.py
import b

def func_a():
    return b.func_b()
````

````python
# b.py
import a

def func_b():
    return a.func_a()
````

````python
import a
a.func_a()
````

```
ImportError: cannot import name 'func_b' from partially initialized module 'b'
(most likely due to a circular import)
```

## 14.4 Why This Happens: Tracing the Execution

| Step | What Happens |
|---|---|
| 1 | Python starts importing `a`, adds a (currently empty) `a` module object to `sys.modules`. |
| 2 | `a.py`'s first line, `import b`, runs; Python starts importing `b`. |
| 3 | `b.py`'s first line, `import a`, runs; Python finds `a` already in `sys.modules` (Part 3.5) and reuses that *partially built* object instead of re-running `a.py`. |
| 4 | `b.py` continues executing and defines `func_b` successfully. |
| 5 | Back in `a.py`, `func_a` gets defined, and `a.py` finishes normally this time. |

In this particular ordering it happens to work, but the moment `b.py` tries to *use* something from `a` at import time (rather than just defining a function that references it later), the partially initialized module will not yet have that name, and an `ImportError` results - the exact failure mode depends delicately on which module is imported first, which is precisely what makes circular imports so unpredictable and hard to debug.

## 14.5 Why Circular Imports Are Problematic

| Problem | Explanation |
|---|---|
| Order-dependent failures | Code can work when run one way and break when imported a different way, or in a different order. |
| Partial initialization | A module may be only partly populated at the moment the circular import reaches back into it. |
| Hard to trace | The `ImportError` traceback often points at the symptom (a missing name), not the underlying circular structure causing it. |
| Signals a design problem | Circular imports are frequently a sign that two modules are too tightly coupled and should not be separate in their current form. |

## 14.6 Refactoring Strategies

**Move the shared logic into a third module** that both original modules can depend on without depending on each other:

````python
# shared.py
def shared_helper():
    ...
````

````python
# a.py
from shared import shared_helper

# b.py
from shared import shared_helper
````

**Move the import inside the function** (a local import) when the circular dependency is rare and unavoidable:

````python
# a.py
def func_a():
    import b   # imported only when func_a actually runs, not at module load time
    return b.func_b()
````

**Restructure using dependency inversion** - instead of module A directly importing module B, have both depend on a shared abstraction (an interface, base class, or callback), so neither module needs concrete knowledge of the other.

## 14.7 Better Project Architecture to Prevent Circular Imports

| Principle | How It Helps |
|---|---|
| Layered architecture | Design modules in layers (e.g. data access -> business logic -> presentation) where each layer only imports from layers below it, never above. |
| Single responsibility | A module that does one clearly defined thing is far less likely to need something back from a module that depends on it. |
| Shared "core" modules | Constants, shared types, and truly common utilities live in a dependency-free module that everything else can safely import. |


# 15. Import Errors

## 15.1 Overview

Import-related errors are among the most common errors new Python developers encounter, precisely because they can stem from so many different causes: a typo, a missing file, a broken environment, a circular dependency, or an incorrect project structure. This section walks through the most common ones as Error -> Cause -> Diagnosis -> Solution.

## 15.2 ModuleNotFoundError

````python
import pandas
````

```

```

|---|---|
| Cause | The module is not installed in the currently active Python environment, or the name is misspelled. |

| Solution | `pip install pandas`, making sure the correct virtual environment (Part 17) is activated first. |

## 15.3 ImportError

````python
from math import square_root
````

```
ImportError: cannot import name 'square_root' from 'math'
```

| Stage | Detail |
|---|---|
| Cause | The module exists and was found, but the specific name requested does not exist inside it (here, the real function is `sqrt`, not `square_root`). |
| Diagnosis | Check the module's actual public API with `dir(math)`, or consult its documentation. |
| Solution | Correct the imported name: `from math import sqrt`. |

## 15.4 AttributeError Caused by Incorrect Imports

````python
import datetime
print(datetime.now())
````

```
AttributeError: module 'datetime' has no attribute 'now'
```

| Stage | Detail |
|---|---|
| Cause | `now()` belongs to the `datetime` *class*, not the `datetime` *module* of the same name - a very common naming collision. |
| Diagnosis | Check whether the object being accessed is the module itself or a class/attribute nested inside it. |
| Solution | Either `datetime.datetime.now()`, or `from datetime import datetime` followed by `datetime.now()`. |

## 15.5 Circular Import Errors

```

(most likely due to a circular import)
```

| Stage | Detail |
|---|---|
| Cause | Two modules import each other, and one reaches for a name that has not been defined yet at the point the circular import loops back (Part 14). |
| Diagnosis | Trace the import chain: which module was imported first, and what does it import in turn? |
| Solution | Restructure using a shared module, move the import inside a function, or apply dependency inversion (Part 14.6). |

## 15.6 Name Conflicts With the Standard Library or Third-Party Packages

````python
# accidentally named your own file "random.py"
import random
random.randint(1, 10)
````

```
AttributeError: module 'random' has no attribute 'randint'
```

| Stage | Detail |
|---|---|
| Cause | A local file named `random.py` sits earlier on `sys.path` than the standard library's `random` module, so Python imports the wrong one entirely. |
| Diagnosis | Check `random.__file__` to see exactly which file was actually imported. |
| Solution | Rename the local file to something that does not collide with a standard library or installed package name. |

## 15.7 Incorrect Package Structure

```
ModuleNotFoundError: No module named 'utils'
```

| Stage | Detail |
|---|---|
| Cause | The importing script is being run from a directory where `utils` is not reachable on `sys.path` - commonly because the script was run from inside a subdirectory rather than the project root. |
| Diagnosis | Print `sys.path` and compare it against the actual project layout on disk. |
| Solution | Run the script from the project root using `python -m package.module` (Part 9.2), or adjust the project structure/packaging so it installs correctly (Part 27). |

# 17. Virtual Environments

## 17.1 Why Virtual Environments Exist

Every Python project potentially needs different, sometimes conflicting, versions of the same packages. If all packages were installed globally on your machine, one project's requirement (say, `pandas==1.5`) could silently break another project that needs `pandas==2.1`. A virtual environment solves this by giving each project its own isolated, self-contained set of installed packages.

## 17.2 Dependency Isolation

| Without a Virtual Environment | With a Virtual Environment |
|---|---|
| All projects share one global set of packages | Each project has its own independent set of packages |
| Version conflicts between projects are common | Each project can use exactly the versions it needs |
| Hard to reproduce a working setup elsewhere | The exact environment can be captured and recreated (Part 18) |

## 17.3 Creating an Environment With venv

`venv` is the standard library's built-in tool for creating virtual environments - no separate installation required.

```
python -m venv env
```

This creates a new directory named `env/` containing a private copy of the Python interpreter and an empty `site-packages` ready for project-specific installs.

## 17.4 Activating and Deactivating an Environment

| Platform | Activate Command |
|---|---|
| macOS / Linux | `source env/bin/activate` |
| Windows (Command Prompt) | `env\Scripts\activate.bat` |
| Windows (PowerShell) | `env\Scripts\Activate.ps1` |

```
deactivate
```

Once activated, the shell prompt typically changes to show the environment's name, and both `python` and `pip` now refer to the environment's private copies rather than the system-wide installation.

## 17.5 Installing Packages Inside an Environment

```
source env/bin/activate
pip install requests pandas
```

Packages installed while an environment is active are placed inside that environment's own `site-packages`, completely separate from the system Python and from any other project's environment.

## 17.6 Project-Specific Dependencies in Practice

A typical professional workflow:

```
mkdir my_project && cd my_project
python -m venv env
source env/bin/activate
pip install pandas requests
pip freeze > requirements.txt
```

Anyone else cloning the project can then recreate the exact same environment (Part 18) with:

```
python -m venv env
source env/bin/activate
pip install -r requirements.txt
```

## 17.7 Cross-Platform Workflow Notes

| Concern | macOS / Linux | Windows |
|---|---|---|
| Activation script location | `env/bin/activate` | `env\Scripts\activate` |
| Python launcher | `python3` (often required) | `python` (usually sufficient) |
| Path separators in scripts | Forward slash `/` | Backslash `\` (though `pathlib`, Part 6.5, handles this automatically) |

Regardless of platform, the underlying concept and the `pip`/`venv` commands themselves are identical - only the activation script's path and, occasionally, which `python` command is available differ.


# 18. Requirements Files

## 18.1 What requirements.txt Is

A `requirements.txt` file is a plain-text list of the packages (and typically their exact versions) that a project depends on. It is the simplest, most widely supported way to communicate and reproduce a Python project's dependencies.

```
requests==2.31.0
pandas==2.1.4
numpy>=1.26,<2.0
python-dotenv==1.0.0
```

## 18.2 Dependency Pinning

| Style | Example | Meaning |
|---|---|---|
| Exact pin | `pandas==2.1.4` | Installs precisely that version, nothing else. |
| Minimum version | `pandas>=2.0` | Installs any version 2.0 or newer. |
| Range | `pandas>=2.0,<3.0` | Installs any version within the given range. |
| Unpinned | `pandas` | Installs whatever the latest available version is at install time. |

Exact pinning is the professional standard for **applications** (guarantees reproducibility), while looser ranges are often more appropriate for **libraries** you publish (so they remain compatible with a wider range of consumers' environments).

## 18.3 Installing From requirements.txt

```
pip install -r requirements.txt
```

## 18.4 Generating requirements.txt

```
pip freeze > requirements.txt
```

`pip freeze` lists every package currently installed in the active environment, along with its exact version - a quick way to snapshot a working setup. Note that `pip freeze` includes every installed package, including indirect dependencies, which can make hand-maintained files noisy; some teams instead maintain requirements by hand, listing only their direct dependencies.

## 18.5 Reproducing Environments

```

source env/bin/activate
pip install -r requirements.txt
```

This three-command sequence is the standard way a teammate, a CI pipeline, or a production server recreates the exact dependency set a project needs, starting from nothing.

# 19. Dependency Management

## 19.1 Direct vs Transitive Dependencies

| Type | Definition | Example |
|---|---|---|
| Direct dependency | A package your code explicitly imports and relies on. | Your app imports `requests`. |
| Transitive dependency | A dependency of one of your dependencies - you never import it directly, but you still need it installed. | `requests` itself depends on `urllib3` and `certifi`. |

## 19.2 Version Conflicts

A version conflict occurs when two of your dependencies require incompatible versions of a shared transitive dependency - for example, Package A needs `urllib3<2.0` while Package B needs `urllib3>=2.0`. `pip` will either fail outright or, in older versions, silently install a version that satisfies only one of them, causing hard-to-diagnose bugs.

## 19.3 Dependency Compatibility and Reproducible Environments

Reproducibility means that anyone (a teammate, a CI server, a production deployment) who sets up the project from scratch ends up with the *exact same* set of installed packages and versions you were using. Exact-pinned `requirements.txt` files (Part 18) are the simplest way to achieve this; modern tooling, discussed below, achieves it more rigorously through lock files.

## 19.4 Dependency Security

Outdated dependencies are one of the most common real-world sources of security vulnerabilities in software. Professional practice includes:

| Practice | Purpose |
|---|---|
| Regularly running `pip list --outdated` | Surfaces packages with newer versions available. |
| Using tools like `pip-audit` | Scans installed packages against known vulnerability databases. |
| Pinning versions deliberately, not blindly always using "latest" | Balances security patching against unexpected breaking changes. |
| Reviewing a new dependency before adding it | Avoids pulling in poorly maintained or malicious packages. |

## 19.5 Modern Python Dependency-Management Approaches

While `pip` + `venv` + `requirements.txt` remains the most universally understood workflow (and a perfectly reasonable default for smaller projects), several modern tools build on top of this foundation to add automatic dependency resolution, lock files, and unified project configuration:

| Tool | What It Adds |
|---|---|
| `pip-tools` | Compiles a high-level `requirements.in` into a fully pinned `requirements.txt`, including all transitive dependencies, with reproducible hashes. |
| Poetry | Manages dependencies, virtual environments, packaging, and publishing together through a single `pyproject.toml` and a lock file. |
| PDM / Hatch | Similar modern, `pyproject.toml`-centered workflows conforming to current Python packaging standards (Part 28). |

**When each is appropriate:**

| Scenario | Reasonable Choice |
|---|---|
| Learning, small scripts, simple projects | `pip` + `venv` + `requirements.txt` |
| A library you intend to publish, with strict metadata needs | `pyproject.toml` with `pip`/`build`, or Poetry/PDM |
| A team wanting fully reproducible, hash-locked installs | `pip-tools` or Poetry |

Understanding the fundamentals covered in this handbook - `pip`, `venv`, `requirements.txt`, and `pyproject.toml` - gives you everything needed to pick up any of these more advanced tools quickly, since they all build directly on the same underlying concepts.

# 21. Module Design Best Practices

## 21.1 Single Responsibility

A well-designed module should have one clear, describable purpose. If summarizing a module's job requires the word "and" more than once ("this module loads data *and* trains models *and* sends emails"), it is a strong candidate for splitting.

## 21.2 Cohesion and Coupling

| Concept | Definition | Goal |
|---|---|---|
| Cohesion | How closely related the contents of a single module are to each other. | High cohesion - everything in the module belongs together. |
| Coupling | How much one module depends on the internal details of another. | Low coupling - modules interact through clear, minimal interfaces. |

## 21.3 Reusability and Naming Conventions

| Guideline | Example |
|---|---|
| Module names: short, lowercase, underscores if needed | `data_loader.py`, not `DataLoader.py` |
| Avoid names that shadow the standard library | Never `json.py`, `random.py`, or `email.py` |
| Package names: short, lowercase, no underscores if possible | `myproject`, not `my_project` (a common but not universal convention) |
| Function/class names inside modules follow standard Python style | `snake_case` for functions, `PascalCase` for classes |

## 21.4 Avoiding Giant Utility Modules

A single sprawling `utils.py` that accumulates every miscellaneous function in a project is one of the most common signs of eroding module design. It becomes hard to navigate, invites unrelated code to pile up together, and often becomes a magnet for circular imports (Part 14), since everything in the project ends up depending on it.

**Bad:**
```
project/
└── utils.py     # 2,000 lines: date formatting, email sending,
                  #   database queries, string validation, math helpers...
```

**Good:**
```
project/
├── formatting/
│   └── dates.py
├── email/
│   └── sender.py
├── db/
│   └── queries.py
└── validation/
    └── strings.py
```

## 21.5 Avoiding Circular Dependencies Through Design

Structuring modules in clear layers - for example, data access at the bottom, business logic in the middle, presentation/API at the top - so that dependencies only ever point downward, prevents the vast majority of circular import problems before they occur (see also Part 14.7).

## 21.6 Public vs Internal Module APIs

````python
# analytics/processing.py

def clean_dataset(df):          # public: intended for external use
    df = _strip_whitespace(df)
    return _drop_duplicates(df)

def _strip_whitespace(df):      # internal: leading underscore signals "not part of the public API"
    ...

def _drop_duplicates(df):       # internal
    ...
````

A leading underscore is Python's conventional (not enforced) signal that a name is an internal implementation detail, not meant to be relied on from outside the module. This convention, combined with `__all__` (Part 22), is how Python communicates a module's intended public surface.


# 22. \_\_all\_\_

## 22.1 What \_\_all\_\_ Is

`__all__` is a module-level list of strings that explicitly defines which names are exported when another file does `from module import *`.

````python
# shapes.py
__all__ = ["Circle", "Square"]

class Circle:
    ...

class Square:
    ...

def _internal_helper():   # not exported, even without __all__, by underscore convention
    ...

SCALE_FACTOR = 1.0         # exists, but NOT exported, since it's absent from __all__
````

````python
from shapes import *

Circle()      # works - listed in __all__
Square()      # works - listed in __all__
SCALE_FACTOR  # NameError - not exported by the wildcard import, even though it exists in the module
````

## 22.2 The Public API Concept

`__all__` is a module author's explicit, curated statement of "this is the intended public interface of this module" - distinct from everything that merely *exists* in the module's namespace. It complements, but does not replace, the leading-underscore convention (Part 21.6).

## 22.3 Controlling Wildcard Imports

Without an `__all__`, `from module import *` exports every top-level name that does **not** start with an underscore. Defining `__all__` overrides this default entirely, giving the module author precise control regardless of naming.

## 22.4 Practical Use Cases

````python
# analytics/__init__.py
from .loading import load_csv, load_json
from .cleaning import remove_nulls, normalize

__all__ = ["load_csv", "load_json", "remove_nulls", "normalize"]
````

This pattern is especially common inside `__init__.py` files, clearly documenting - in one place - exactly which re-exported names make up the package's public surface.

## 22.5 Limitations of \_\_all\_\_

| Limitation | Explanation |
|---|---|
| Only affects `import *` | `__all__` has *no effect whatsoever* on explicit imports like `from module import _internal_helper` - that still works fine. |
| Not a security mechanism | `__all__` is a documentation and convenience tool, not an access-control mechanism; any name in the module remains fully accessible via `module._internal_helper` or an explicit `from module import _internal_helper`. |
| Easy to forget to update | Adding a new public function without adding it to `__all__` silently excludes it from wildcard imports. |

**Important framing:** `__all__` communicates intent to other developers and controls one specific import style. It should never be relied upon as a way to "hide" or "protect" sensitive code - Python does not have true private members, and `__all__` does not change that.


# 23. Import Caching

## 23.1 How Imports Are Cached

As introduced in Part 3.5, Python caches every module it imports inside a dictionary called `sys.modules`, keyed by the module's fully qualified name. Any subsequent `import` of the same module - anywhere else in the program - reuses the cached module object instead of re-running its code.

````python
import sys
import math

print("math" in sys.modules)   # True, after the first import
````

## 23.2 Why Modules Generally Execute Only Once Per Process

````python
# noisy.py
print("noisy.py is running!")
VALUE = 42
````

````python
import noisy   # prints "noisy.py is running!"
import noisy   # prints nothing - already cached
from noisy import VALUE   # still nothing - same cached module
````

This one-time execution guarantee is intentional and important: it means module-level state (database connections, loaded configuration, compiled regular expressions) is set up exactly once and then safely shared everywhere that module is imported, rather than being wastefully - or dangerously - reinitialized on every import.

## 23.3 Re-importing and importlib.reload()

Occasionally, especially during interactive development, you want a module's code to actually re-run - for example, after editing its source file without restarting the interpreter.

````python
import importlib
import my_module

importlib.reload(my_module)
````

`importlib.reload()` forces the target module's code to execute again *in place*, updating its existing module object, rather than creating a new one from scratch.

## 23.4 When Reloading Is Useful

| Scenario | Why Reload Helps |
|---|---|
| Interactive/REPL development | Test a change to a module's source without restarting the whole session. |
| Jupyter notebook workflows | Iterating on a helper module imported by a notebook. |
| Plugin systems (Part 26) | Reloading a plugin after it has been updated on disk, without restarting the host application. |

## 23.5 Limitations of Reloading

| Limitation | Explanation |
|---|---|
| Existing references are not updated | Objects created *before* the reload (e.g. class instances) still point to the *old* class definition, not the reloaded one. |
| Does not cascade to dependents | Reloading module A does not automatically reload module B, even if B imported names from A. |
| Not a substitute for a restart in production | Reloading is a development convenience; production systems should restart cleanly instead of relying on reload semantics. |
| C extension modules generally cannot be usefully reloaded | Many compiled modules ignore reload or behave unpredictably. |

**Professional guidance:** `importlib.reload()` is a genuinely useful tool during interactive exploration, but production code should never depend on runtime reloading as part of its normal behavior - restart the process instead.


# 24. importlib

## 24.1 What importlib Is

`importlib` is the standard library package that exposes Python's own import machinery as a set of programmable functions and classes. It lets code import modules *dynamically* - deciding what to import while the program is running, rather than writing a fixed `import` statement in advance.

## 24.2 Dynamic Imports With import_module()

````python
import importlib

module_name = "json"   # could come from user input, a config file, etc.
module = importlib.import_module(module_name)

print(module.dumps({"key": "value"}))
````

````python
plugin = importlib.import_module(f"myapp.plugins.{plugin_name}")
plugin.run()
````

## 24.3 Reloading

````python
import importlib
importlib.reload(module)   # covered fully in Part 23.3
````

## 24.4 A Glimpse Into the Import Machinery

`importlib` also exposes the lower-level building blocks Python itself uses to implement `import` - finders (which locate a module) and loaders (which execute it). Most application code never needs to touch these directly, but understanding they exist explains *why* dynamic imports and custom import behavior (like plugin auto-discovery, Part 26) are possible at all.

````python
import importlib.util

spec = importlib.util.spec_from_file_location("my_module", "/path/to/my_module.py")
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
````

This lower-level pattern loads a module directly from an arbitrary file path, bypassing `sys.path` entirely - useful for loading plugin files from a location not otherwise on the import path.

## 24.5 Practical Use Cases

| Use Case | Why importlib Helps |
|---|---|
| Configuration-driven behavior | Load a different implementation module based on a config file's contents. |
| Plugin systems (Part 26) | Discover and import plugin modules whose names are not known until runtime. |
| Optional dependencies | Attempt to import an optional library and gracefully fall back if it is not installed. |
| Testing tools | Some testing frameworks use dynamic imports to discover test modules automatically. |

## 24.6 When Dynamic Importing Is Appropriate vs Unnecessary Complexity

| Appropriate | Unnecessary Complexity |
|---|---|
| The exact module to import genuinely isn't known until runtime (plugin name from config, user selection). | You already know exactly which modules you need - just write a normal `import` statement. |
| Building a plugin or extension system with independently developed add-ons. | Trying to avoid writing a small number of explicit imports "to be clever." |
| Optional/soft dependencies with graceful fallback. | Replacing simple `if/elif` dispatch logic that would be clearer as ordinary code. |

Dynamic importing is a powerful, legitimate tool for the right problem, but it also makes code harder to statically analyze, harder for IDEs to navigate, and harder to reason about at a glance - it should be reached for deliberately, not by default.

Building on `importlib` (Part 24), applications can decide *which* modules to load while they are running, based on outside factors rather than fixed code written in advance. This section focuses on the practical patterns for doing so.

## 25.4 Loading Based on Environment

````python
import os
import importlib

env = os.environ.get("APP_ENV", "development")
settings = importlib.import_module(f"myapp.settings.{env}")
````

```

├── __init__.py
```

This is a widely used pattern (seen in many real web frameworks) for maintaining separate, environment-specific configuration modules that are selected automatically without touching application code.

## 25.5 A Complete Practical Example

````python
# myapp/notifiers/__init__.py
import importlib

_NOTIFIER_MAP = {
    "email": "myapp.notifiers.email_notifier",
    "sms": "myapp.notifiers.sms_notifier",
    "slack": "myapp.notifiers.slack_notifier",
}

def get_notifier(channel):
    if channel not in _NOTIFIER_MAP:
        raise ValueError(f"Unknown notification channel: {channel}")
    module = importlib.import_module(_NOTIFIER_MAP[channel])
    return module.Notifier()
````

````python
notifier = get_notifier("slack")
notifier.send("Deployment finished successfully")
````

This pattern isolates the *decision* of which module to load from the *code* that uses the result, keeping the calling code simple and completely unaware of how many notifier types exist.


# 26. Plugin Architecture

## 26.1 What a Plugin System Is

A plugin system lets a core application be extended with additional functionality - written independently, sometimes by different developers entirely - without modifying the core application's source code. Python's dynamic import capabilities (Parts 24 and 25) make this pattern natural to implement.

## 26.2 Dynamic Discovery

```
myapp/
├── core.py
└── plugins/
    ├── __init__.py
    ├── uppercase_plugin.py
    └── reverse_plugin.py
```

````python
# core.py
import importlib
import pkgutil
import myapp.plugins as plugins_package

def discover_plugins():
    discovered = {}
    for _, name, _ in pkgutil.iter_modules(plugins_package.__path__):
        module = importlib.import_module(f"myapp.plugins.{name}")
        discovered[name] = module
    return discovered
````

`pkgutil.iter_modules` scans a package's directory and finds every module inside it at runtime, without the core application needing to know their names in advance.

## 26.3 Interfaces and Registration

A robust plugin system defines a clear *contract* every plugin must follow, so the core application can rely on consistent behavior regardless of which plugin is loaded.

````python
# plugins/uppercase_plugin.py
NAME = "uppercase"

def process(text):
    return text.upper()
````

````python
# plugins/reverse_plugin.py
NAME = "reverse"

def process(text):
    return text[::-1]
````

````python
# core.py
def run_plugin(module, text):
    return module.process(text)   # every plugin is expected to expose process()
````

## 26.4 A Simple, Complete Plugin Example

````python
# core.py
import importlib
import pkgutil
import myapp.plugins as plugins_package

def load_plugins():
    registry = {}
    for _, name, _ in pkgutil.iter_modules(plugins_package.__path__):
        module = importlib.import_module(f"myapp.plugins.{name}")
        registry[module.NAME] = module
    return registry

def main():
    plugins = load_plugins()
    text = "Hello World"
    for name, plugin in plugins.items():
        print(f"{name}: {plugin.process(text)}")

if __name__ == "__main__":
    main()
````

```
uppercase: HELLO WORLD
reverse: dlroW olleH
```

## 26.5 Entry Points, Conceptually

Larger, installable plugin ecosystems (for example, `pytest` plugins, or Flask extensions) typically go one step further, using **entry points** declared in a package's own `pyproject.toml` metadata (Part 28) to advertise "I provide a plugin for application X" at install time - allowing the host application to discover plugins that were installed via `pip`, without even needing them to live inside its own `plugins/` directory. The `pkgutil`-based approach shown above is the right mental model to start with; entry-point-based discovery is the natural next step once plugins are distributed as separate, independently installable packages.

## 26.6 Extensibility Benefits

| Benefit | Explanation |
|---|---|
| Core stays stable | New functionality is added without touching or risking existing core code. |
| Independent development | Plugin authors can build and release features on their own schedule. |
| Optional functionality | Users only install/enable the plugins they actually need. |
| Clear contracts | A well-defined plugin interface keeps the system predictable even as it grows. |


# 27. Packaging a Python Project

## 27.1 What "Packaging" Means

Packaging is the process of turning a folder of Python code into a distributable, installable unit - something that can be installed with `pip install`, shared with others, or published to PyPI (Part 29), rather than only usable by manually copying files around.

## 27.2 Project Metadata

Every installable Python package needs metadata describing itself: its name, version, author, description, license, and dependencies. In modern Python packaging, this metadata lives in `pyproject.toml` (Part 28), replacing the older `setup.py`-based approach.

## 27.3 A Minimal Installable Package Structure

```
myproject/
├── src/
│   └── myproject/
│       ├── __init__.py
│       └── core.py
├── pyproject.toml
└── README.md
```

## 27.4 Build Systems

A **build system** is the tool that turns your source files into a distributable package archive. `pyproject.toml` declares which build backend to use:

```toml
[build-system]
requires = ["setuptools>=68.0"]
build-backend = "setuptools.build_meta"
```

| Backend | Notes |
|---|---|
| `setuptools` | The long-standing, most widely used build backend. |
| `hatchling` (Hatch) | A modern, actively developed alternative with a simpler configuration model. |
| `poetry-core` | Used automatically when managing a project with Poetry. |

## 27.5 Build Artifacts

Running a build produces distributable archive files, typically placed in a `dist/` directory:

```
python -m build
```

```
dist/
├── myproject-1.0.0.tar.gz          # source distribution ("sdist")
└── myproject-1.0.0-py3-none-any.whl # built distribution ("wheel")
```

| Artifact | Description |
|---|---|
| Source distribution (`.tar.gz`) | Contains the raw source code and metadata; installing it may require a build step. |
| Wheel (`.whl`) | A pre-built, ready-to-install format; installing it is just extracting files - faster and more reliable, and the modern default. |

## 27.6 Installing a Local Package

```
pip install .
```

Installs the package located in the current directory, following the metadata in `pyproject.toml`, exactly as if it had been downloaded from PyPI.

## 27.7 Editable Installs

```
pip install -e .
```

An editable install links the installed package directly back to your source files on disk, rather than copying them into `site-packages`. Any change you make to the source is immediately reflected the next time the package is imported - no reinstall required. This is the standard workflow while actively developing a package you also intend to publish.

## 27.8 Modern vs Legacy Packaging Practices

| Aspect | Legacy (`setup.py`) | Modern (`pyproject.toml`) |
|---|---|---|
| Configuration format | Executable Python script | Declarative TOML file |
| Metadata definition | Arguments passed to `setup()` | `[project]` table |
| Build isolation | Inconsistent across tools | Standardized via PEP 517/518 |
| Current recommendation | Still encountered in older/legacy codebases | The current standard for new projects |

New projects should default to `pyproject.toml`; understanding `setup.py` remains valuable only because a great deal of existing, older code in the wild still uses it.


# 28. pyproject.toml

## 28.1 What It Is

`pyproject.toml` is the single, standardized configuration file for modern Python projects, written in the TOML format. It consolidates project metadata, dependencies, build configuration, and tool-specific settings that used to be scattered across `setup.py`, `setup.cfg`, and various other config files.

## 28.2 Why Modern Python Projects Use It

| Reason | Explanation |
|---|---|
| Standardized | Defined by official Python packaging standards (PEP 517, 518, 621), so every build tool understands the same core format. |
| Declarative | Plain data instead of an executable script, which is easier to parse, validate, and reason about safely. |
| Single source of truth | One file for metadata, dependencies, and tool configuration, instead of several. |
| Tool-agnostic | Works the same whether you use `pip`/`build`, Poetry, Hatch, or PDM underneath. |

## 28.3 A Realistic Example

```toml
[build-system]
requires = ["setuptools>=68.0"]
build-backend = "setuptools.build_meta"

[project]
name = "myproject"
version = "1.0.0"
description = "A data processing toolkit for internal analytics."
readme = "README.md"
requires-python = ">=3.10"
license = { text = "MIT" }
authors = [
    { name = "Kajola Gbenga", email = "kajola@example.com" }
]
dependencies = [
    "pandas>=2.0,<3.0",
    "requests>=2.28",
]

[project.optional-dependencies]
dev = ["pytest>=7.0", "black", "mypy"]

[project.scripts]
myproject-cli = "myproject.cli:main"

[tool.black]
line-length = 88

[tool.mypy]
strict = true
```

## 28.4 Explaining Each Section

| Section | Purpose |
|---|---|
| `[build-system]` | Declares which build backend and tools are required to build the package (Part 27.4). |
| `[project]` | Core metadata: name, version, description, supported Python versions, author, and license. |
| `dependencies` | The list of packages required to run the project - equivalent in purpose to `requirements.txt`, but part of the package's own metadata. |
| `[project.optional-dependencies]` | Named groups of *extra* dependencies (like development or testing tools) that are not required for normal use. |
| `[project.scripts]` | Declares command-line entry points; here, running `myproject-cli` after installation calls `main()` inside `myproject/cli.py`. |
| `[tool.*]` | Configuration for specific third-party tools (`black`, `mypy`, `pytest`, etc.), all living in one shared file. |

## 28.5 Dependencies vs Optional Dependencies

```
pip install myproject                # installs core dependencies only
pip install "myproject[dev]"         # also installs the "dev" extras group
```

Optional dependency groups let a package define add-on functionality (like plotting support, or a specific database driver) that most users do not need, avoiding forcing an unnecessarily heavy install on everyone.

# 30. Semantic Versioning

## 30.1 The Format

Semantic versioning (often called "SemVer") expresses a package's version as three numbers: `MAJOR.MINOR.PATCH` (for example, `2.4.1`).

## 30.2 What Each Number Means

| Segment | Incremented When | Example |
|---|---|---|
| MAJOR | A breaking (backward-incompatible) change is made. | `1.9.0` -> `2.0.0` |
| MINOR | New, backward-compatible functionality is added. | `2.0.0` -> `2.1.0` |
| PATCH | A backward-compatible bug fix is made, with no new features. | `2.1.0` -> `2.1.1` |

## 30.3 Breaking Changes vs Backward Compatibility

A **breaking change** is any change that could cause code depending on the package to stop working - renaming a public function, changing a function's required arguments, or removing a feature entirely. A **backward-compatible** change adds or fixes something without requiring any existing user of the package to change their own code.

## 30.4 Dependency Version Constraints

Semantic versioning is what makes version *ranges* in dependency declarations meaningful and safe:

```
pandas>=2.0,<3.0
```

This says: "any 2.x release of pandas is safe to use, because MINOR and PATCH releases within the same MAJOR version are expected to remain backward compatible; a 3.0 release, however, might break things, so it is deliberately excluded."

| Constraint Style | Meaning |
|---|---|
| `==2.1.4` | Exactly this version - no flexibility. |
| `>=2.1` | This version or newer - any MAJOR version. |
| `>=2.1,<3.0` | This version or newer, but staying within MAJOR version 2. |
| `~=2.1.4` (compatible release) | Equivalent to `>=2.1.4,<2.2.0` - allows PATCH-level updates only. |

## 30.5 Pre-Release and Build Metadata (Brief Note)

SemVer also supports pre-release labels for versions still under development, such as `2.0.0-beta.1` or `2.0.0-rc.1`, signaling to users that the release is not yet considered stable, even though it numerically precedes `2.0.0` itself.

## 30.6 Why This Matters for Both Consumers and Publishers

| Perspective | Why SemVer Matters |
|---|---|
| As a consumer | You can set dependency ranges confidently, trusting that MINOR/PATCH updates will not silently break your code. |
| As a publisher | Your version number itself communicates the nature of each change to every user, without them needing to read a full changelog just to know if it's safe to upgrade. |

# 35. Common Module and Package Mistakes

## 35.1 Naming a File After a Standard Library Module

````python
# random.py  <- a project file with this exact name
import random
random.randint(1, 10)
````

| Stage | Detail |
|---|---|
| Problem | Your own `random.py` shadows the standard library's `random` module, since your project directory is searched before the standard library. |
| Why it happens | Python has no built-in protection against a local file reusing a standard library name. |
| Solution | Rename the file to something unambiguous, such as `random_utils.py`. |
| Best practice | Before naming a new module, briefly check it does not collide with a standard library or already-installed third-party package name. |

## 35.2 Circular Imports

Covered in full depth in Part 14. In summary: two modules depending on each other directly is the problem, and the solution is a shared module, a local import, or dependency inversion.

## 35.3 Incorrect Relative Imports

````python
# Run directly: python utils/validation.py
from . import calculations
````

| Stage | Detail |
|---|---|
| Problem | `ImportError: attempted relative import with no known parent package` |
| Why it happens | Relative imports require the file to be imported as part of a package; running it directly as a script does not set that up. |
| Solution | Run with `python -m package.utils.validation`, or avoid running files with relative imports directly. |
| Best practice | Keep relative imports inside files that are only ever imported, never executed directly as scripts. |

## 35.4 Running Modules Incorrectly

```
cd utils
python validation.py
```

| Stage | Detail |
|---|---|
| Problem | `ModuleNotFoundError`, since the project root is no longer on `sys.path` from inside the subdirectory. |
| Why it happens | `sys.path`'s first entry is based on the location of the file being run, not the project root. |
| Solution | Run scripts from the project root using `python -m package.module`, not by `cd`-ing into subdirectories. |

## 35.5 Missing Package Structure

```
ModuleNotFoundError: No module named 'utils'
```

| Stage | Detail |
|---|---|
| Problem | A directory intended to be a package is missing an `__init__.py` (in traditional packaging), or is simply not reachable on `sys.path`. |
| Why it happens | The directory layout does not match what the import statement expects. |
| Solution | Add the missing `__init__.py`, or run the project via `python -m` from the correct root directory. |

## 35.6 Confusing Scripts With Modules

| Problem | A file both performs side effects at import time (writing files, hitting a network) and is meant to be imported elsewhere for its functions. |
| Why it happens | The `if __name__ == "__main__":` guard (Part 8.3) was never added. |
| Solution | Wrap runnable logic in a `main()` function, and only call it inside the `__name__` guard. |

## 35.7 Installing Packages Globally

| Problem | Installing project dependencies directly into the system-wide Python installation instead of a virtual environment. |
| Why it happens | Skipping `python -m venv` "to save a step," especially early in a project. |
| Consequence | Version conflicts between unrelated projects, and a much harder time reproducing a working setup elsewhere. |
| Solution | Always create and activate a virtual environment (Part 17) before installing project dependencies. |

## 35.8 Dependency Conflicts

| Problem | Two dependencies require incompatible versions of a shared transitive dependency (Part 19.2). |
| Why it happens | Dependencies were added over time without checking compatibility, or without pinned versions. |
| Solution | Use `pip`'s dependency resolver output (or a tool like `pip-tools`) to identify the conflict, and adjust version constraints until a compatible combination exists. |

## 35.9 Poor Project Structure

| Problem | Code, data, notebooks, and configuration all mixed together with no clear organization. |
| Why it happens | The project grew organically without revisiting structure as it expanded. |
| Solution | Adopt one of the structures shown in Part 20, appropriate to the project's domain. |

## 35.10 Giant Utility Modules

Covered in Part 21.4: a single sprawling `utils.py` accumulating unrelated functions. Split by responsibility into clearly named, focused modules instead.

# 37. Interview Questions

## 37.1 Beginner Questions

**Q1. What is the difference between a module and a package?**
A module is a single `.py` file; a package is a directory of related modules (traditionally containing an `__init__.py`).

**Q2. What does `import module_name` actually do?**
It locates the module, executes its code once (unless already cached), and binds the resulting module object to a name in the current namespace.

**Q3. What is the difference between `import module` and `from module import name`?**
`import module` binds only the module itself, requiring `module.name` access. `from module import name` binds `name` directly into the current namespace.

**Q4. What does `__name__` equal when a file is run directly versus imported?**
`"__main__"` when run directly; the module's own name when imported.

**Q5. Why should wildcard imports (`from module import *`) generally be avoided?**
They risk silent name collisions, obscure where a name came from, and hinder static analysis tools.

## 37.2 Intermediate Questions

**Q6. What is the purpose of `__init__.py`?**
It marks a directory as a package (in traditional packaging), runs once on first import, and can control what the package exposes.

**Q7. What is the difference between an absolute and a relative import?**
An absolute import spells out the full path from the project root; a relative import is expressed relative to the current module's position, using leading dots.

**Q8. Why do relative imports fail when a file is run directly?**
Because relative imports rely on the file's `__package__` context, which is only set correctly when the file is imported as part of a package, not run as a standalone script.

**Q9. What is `sys.path` and how does it affect imports?**
A list of directories Python searches, in order, to locate a module being imported; the first match found is used.

**Q10. What is the purpose of a virtual environment?**
To isolate a project's installed dependencies from the system Python and from other projects, avoiding version conflicts.

## 37.3 Advanced Questions

**Q11. Explain exactly what happens the second time a module is imported in the same process.**
Python finds the module already present in `sys.modules` and simply rebinds the existing module object to the requested name - the module's code is not re-executed.

**Q12. Why do circular imports sometimes work and sometimes fail, depending on which module is imported first?**
Because Python inserts a module into `sys.modules` before executing its code; whichever module starts the cycle determines which names are available, and in what partially-initialized state, when the cycle loops back.

**Q13. What is the difference between `__all__` and a leading underscore on a name?**
A leading underscore is an informal convention meaning "internal, not part of the public API," but has no effect on `import *`. `__all__` explicitly and formally controls exactly what `import *` exports, regardless of underscore naming.

**Q14. What is the difference between a wheel and a source distribution?**
A wheel (`.whl`) is a pre-built, ready-to-install format. A source distribution (`.tar.gz`) contains raw source and may require a build step during installation.

**Q15. How does `importlib.import_module()` differ from a standard `import` statement?**
It imports a module whose name is only known at runtime (e.g. from a string variable), whereas a normal `import` statement requires the module name to be written explicitly in the source code.

## 37.4 Import-Troubleshooting Questions

**Q16. You get `ModuleNotFoundError: No module named 'pandas'` even though you installed it. What do you check first?**
Whether the package was installed into the same Python environment/virtual environment currently being used to run the script.

**Q17. You get `ImportError: attempted relative import with no known parent package`. What is the likely cause?**
The file using a relative import was run directly (`python file.py`) rather than as part of its package (`python -m package.file`).

**Q18. Your own `random.py` breaks calls like `random.randint()`. Why?**
Your local file is shadowing the standard library's `random` module because it is found earlier on `sys.path`.

## 37.5 Package-Management Questions

**Q19. What is the difference between a direct and a transitive dependency?**
A direct dependency is one your code explicitly imports; a transitive dependency is a dependency *of* one of your dependencies, installed automatically but never imported by your own code.

**Q20. Why might `pip freeze > requirements.txt` produce a much longer file than expected?**
Because `pip freeze` lists every installed package, including all transitive dependencies, not just the ones you directly added.

## 37.6 Project-Architecture Questions

**Q21. Why is a giant, catch-all `utils.py` considered poor design?**
It tends to accumulate unrelated functionality, becomes hard to navigate, and often becomes an accidental hub for circular imports.

**Q22. What is the benefit of the `src/` project layout over placing the package directly in the project root?**
It forces tools and tests to import the actually-installed version of the package, catching packaging mistakes that would otherwise go unnoticed.

## 37.7 Code-Output Questions

**Q23. What does the following print if run directly, and what does it print if imported?**

````python
# demo.py
print(__name__)
````

Answer: prints `__main__` when run directly (`python demo.py`), and prints `demo` when imported elsewhere (`import demo`).

**Q24. Given `module_a.py` containing `import module_b` and `module_b.py` containing `import module_a`, what determines whether this succeeds?**
Whether the specific names each module needs from the other are already defined by the time the circular import loops back - which depends on which module is imported first and the exact order of statements inside each file.

## 37.8 Scenario-Based Questions

**Q25. You are designing a plugin system where third-party developers can add new functionality without modifying your core application. What Python features would you use?**
`pkgutil.iter_modules` (or entry points) for discovery, `importlib.import_module` for dynamic loading, and a clearly defined interface/contract every plugin must implement (Part 26).

**Q26. Your team wants every developer's local environment to exactly match production. What would you set up?**
A virtual environment per project, combined with an exact-pinned `requirements.txt` (or a lock file from a tool like `pip-tools` or Poetry), so the same versions are installed everywhere.

# 39. Complete Modules & Packages Cheat Sheet

## 39.1 import

| Syntax | Meaning |
|---|---|
| `import module` | Imports the whole module; access via `module.name`. |
| `import module as alias` | Imports the module bound to a different name. |
| `import a, b, c` | Legal, but PEP 8 recommends one import per line. |

## 39.2 from ... import

| Syntax | Meaning |
|---|---|
| `from module import name` | Pulls `name` directly into the current namespace. |
| `from module import a, b` | Pulls multiple names directly. |
| `from module import name as alias` | Pulls a name in under a different local name. |
| `from module import *` | Imports all public names (or exactly `__all__` if defined) - generally avoided. |

## 39.3 as

| Use | Example |
|---|---|
| Module alias | `import pandas as pd` |
| Function/class alias | `from datetime import datetime as dt` |
| Avoiding a name collision | `from a import process as process_a` |

## 39.4 \_\_name\_\_

| Context | Value |
|---|---|
| File run directly | `"__main__"` |
| File imported | The module's own name |
| Guard pattern | `if __name__ == "__main__": main()` |

## 39.5 \_\_main\_\_.py

| Concept | Detail |
|---|---|
| Purpose | Defines what happens when a *package* is executed directly. |
| Run with | `python -m package_name` |
| Common use | CLI entry points for installable packages. |

## 39.6 \_\_init\_\_.py

| Concept | Detail |
|---|---|
| Purpose | Marks a directory as a package; runs once on first import. |
| Empty file | Valid; simply marks the package. |
| Re-exporting | `from .module import name` inside `__init__.py` exposes it at the package level. |
| Modern alternative | Namespace packages (no `__init__.py` needed) for multi-distribution plugin ecosystems. |

## 39.7 sys.path

| Concept | Detail |
|---|---|
| What it is | The ordered list of directories Python searches for modules. |
| First entry | Typically the running script's own directory. |
| Extending it | `PYTHONPATH` environment variable, or `sys.path.append(...)` at runtime (last resort). |

## 39.8 sys.modules

| Concept | Detail |
|---|---|
| What it is | A cache mapping already-imported module names to their module objects. |
| Effect | A module's code executes only once per process; later imports reuse the cached object. |

## 39.9 importlib

| Function | Purpose |
|---|---|
| `importlib.import_module(name)` | Dynamically imports a module whose name is only known at runtime. |
| `importlib.reload(module)` | Re-executes a module's code in place, updating its existing module object. |

## 39.10 pip

| Command | Purpose |
|---|---|
| `pip install pkg` | Install the latest compatible version. |
| `pip install pkg==1.2.3` | Install an exact version. |

## 39.17 Import Troubleshooting Quick-Reference
