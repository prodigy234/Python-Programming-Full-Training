# Part 1. Introduction to Python Data Types

Every program, at its core, is a set of instructions for moving and transforming data. Before any of that transformation can happen, Python needs to know what kind of data it is looking at: a whole number, a piece of text, a collection of items, or something else. That classification is a data type, and it is the single most foundational concept in the entire language.

## 1.1 What Is Data?

In programming, data is any piece of information a program stores, processes, or produces: a person's age, a product name, a list of exam scores, a customer's email address, an image's pixel values. On its own, raw data is just bits in memory; it has no inherent meaning until a program interprets it in a specific way.

Python's job as an interpreter is to take a literal value you write, such as 42 or "Lagos", and store it in memory in a structured way that the rest of the language knows how to work with. That structure is exactly what a data type provides.

## 1.2 What Is a Data Type?

A data type is a classification that tells Python (and you) three things about a piece of data: what kind of values it can hold, what operations are valid on it, and how much memory and what internal representation it needs.

For example, the type int tells Python that a value is a whole number, that it supports arithmetic operators like + and *, and that it should be stored using an internal structure built for arbitrary-precision integers. The type str tells Python that a value is text, that it supports concatenation and indexing, and that it is stored as an immutable sequence of Unicode code points.

## 1.3 Why Data Types Matter in Python

- They determine which operations are legal: you can add two integers, but adding an integer to a list raises a TypeError.
- They determine memory layout and performance: an int, a list and a dict are stored completely differently in memory.
- They enable correctness: knowing a variable is a list rather than a single value tells you (and Python) that it can be iterated, indexed, and appended to.
- They shape program design: choosing a tuple over a list, or a set over a list, is a data-type decision that affects speed, correctness and intent.
- They are the language every Python error message speaks: TypeError, ValueError and AttributeError all exist because of type rules.

## 1.4 How Python Determines Data Types

Unlike languages such as Java or C, where you must declare a variable's type before using it (int x;), Python figures out a value's type automatically, at the moment the value is created, based on the literal syntax or the constructor used. Writing x = 5 creates an int because 5 is an integer literal; writing x = 5.0 creates a float because of the decimal point; writing x = "5" creates a str because of the quotes.

This automatic determination is possible because in Python, types belong to values (objects), not to the variable names that refer to them. A variable name is just a label; the object it points to carries its own type information permanently.

## 1.5 Dynamic Typing

Python is dynamically typed, which means a variable name is not permanently bound to one type. The same name can be reassigned to point to objects of completely different types over the life of a program, and Python will not complain at that reassignment.

```python
x = 42          # x refers to an int
print(type(x))

x = "forty-two"  # x now refers to a str; this is completely legal
print(type(x))

x = [42]         # x now refers to a list
print(type(x))
```

| Code / Item | Explanation |
|---|---|
| `x = 42` | Binds the name x to a new int object. |
| `x = "forty-two"` | Rebinds the same name x to an entirely different object, this time a str; the earlier int object is simply no longer referenced by anything named x. |
| `type(x)` | Asks Python for the type of whatever object x currently refers to, which can change from line to line. |

## 1.6 Strong Typing

Dynamic typing is often confused with weak typing, but Python is dynamically typed and strongly typed at the same time. Strong typing means Python will not silently coerce incompatible types together in an operation; it raises a TypeError instead of guessing what you meant.

Contrast this with a weakly typed language (or even certain operations in JavaScript) where "5" + 3 might silently become "53". Python refuses that ambiguity outright.

```python
try:
    result = "5" + 3
except TypeError as e:
    print("TypeError:", e)

# Python requires an explicit, deliberate conversion instead:
result = "5" + str(3)
print(result)
result2 = int("5") + 3
print(result2)
```

| Code / Item | Explanation |
|---|---|
| `"5" + 3` | Raises TypeError because str and int have no implicit addition rule; Python refuses to guess whether you wanted string concatenation or numeric addition. |
| `"5" + str(3)` | Explicitly converts 3 to the string "3" first, so both operands are str and concatenation succeeds, producing "53". |
| `int("5") + 3` | Explicitly converts "5" to the integer 5 first, so both operands are int and addition succeeds, producing 8. |

## 1.7 Objects and Values

Every single piece of data in Python, without exception, is an object: every int, every string, every function, even every class itself. An object is a region of memory that bundles together a value (the actual data), a type (which operations are valid), and an identity (a unique location Python can reference).

The value is the concrete content the object holds: the number 42, the text "hello", or the elements [1, 2, 3]. Two different objects can hold the same value (two separate int objects that both equal 42) while still being distinct objects in memory.

## 1.8 Variables vs Objects

A common beginner misconception is that a variable 'is' a value, the way a box contains its contents. In Python, it is more accurate to think of a variable as a name tag tied to an object elsewhere in memory. The object holds the value and the type; the variable is just a reference, a way to find that object.

This distinction becomes essential once mutable objects are introduced in Part 16, because multiple variable names can point to the very same object, and changes made through one name are visible through every other name pointing at it.

```python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
print(b)
print(a is b)
```

| Code / Item | Explanation |
|---|---|
| `a = [1, 2, 3]` | Creates a list object and attaches the name a to it. |
| `b = a` | Does not copy the list; it attaches a second name, b, to the exact same object a already refers to. |
| `b.append(4)` | Mutates the single underlying list object through the name b. |
| `print(a)` | Shows [1, 2, 3, 4] too, because a and b were never two different lists to begin with, just two names for one. |

## 1.9 Object References

When you write a = 10, Python creates an integer object 10 somewhere in memory and stores a reference to it (conceptually, a memory address) under the name a. Assignment in Python is always reference-binding, never a low-level byte-for-byte copy of the object's content into the variable itself.

This is true for every type, immutable or mutable; the difference in Part 16 is not about whether references are used (they always are) but about whether the object being referenced can be changed in place after creation.

## 1.10 The type() Function

type(obj) returns the exact class (type) of an object. It is the most direct way to ask Python what kind of data you are holding.

```python
print(type(10))
print(type(3.14))
print(type("data"))
print(type([1, 2]))
print(type((1, 2)))
print(type({1: 2}))
print(type({1, 2}))
print(type(None))
```

| Code / Item | Explanation |
|---|---|
| `type(10)` | Returns <class 'int'>. |
| `type(3.14)` | Returns <class 'float'>. |
| `type(None)` | Returns <class 'NoneType'>, confirming that even the absence-of-value marker None is itself a typed object. |

## 1.11 The isinstance() Function

isinstance(obj, classinfo) checks whether an object belongs to a given type, or any subclass of it, returning True or False. It is generally preferred over comparing type(obj) == SomeType directly, because it correctly accounts for inheritance: a bool, for instance, is technically a subclass of int, and isinstance() reflects that relationship faithfully.

```python
value = True
print(type(value) == int)      # False: exact type is bool, not int
print(isinstance(value, int))  # True: bool is a subclass of int
print(isinstance(value, bool)) # True: exact match too
print(isinstance(3.0, (int, float)))  # True: matches any type in the tuple
```

| Code / Item | Explanation |
|---|---|
| `type(value) == int` | False, because type() returns the exact class bool, and bool is not literally equal to the class int. |
| `isinstance(value, int)` | True, because isinstance() also accepts subclasses, and bool is defined as a subclass of int in Python. |
| `isinstance(3.0, (int, float))` | Accepts a tuple of acceptable types, returning True if the object matches any one of them. |

## 1.12 Type Checking in Practice

Type checking means writing code that verifies a value's type before relying on type-specific behavior, which prevents a whole class of runtime errors. It is common at the boundaries of a program: validating user input, function arguments, or data loaded from a file or an API.

```python
def calculate_total(prices):
    if not isinstance(prices, (list, tuple)):
        raise TypeError("prices must be a list or tuple")
    return sum(prices)

print(calculate_total([10, 20, 30]))
try:
    calculate_total("not a collection")
except TypeError as e:
    print("TypeError:", e)
```

## 1.13 Type Conversion and Type Casting

Type conversion (also called type casting) is the process of turning a value of one type into an equivalent value of another type, using a constructor function such as int(), float(), str(), or list(). Python distinguishes implicit conversion, which the interpreter performs automatically in very limited, safe situations (such as promoting an int to a float inside 3 + 4.0), from explicit conversion, which you request yourself by calling a constructor.

Part 20 covers type conversion and casting in full depth; this introduction only establishes the vocabulary.

```python
# implicit conversion: Python silently widens int to float for the operation
result = 3 + 4.0
print(result, type(result))

# explicit conversion: you request the change yourself
text_number = "100"
as_int = int(text_number)
print(as_int, type(as_int))
```

## 1.14 The Core Relationship: Variable to Object to Value to Data Type

Every one of the preceding ideas connects into a single mental model that underlies all of Python:

- A variable is a name that references an object; it holds no data of its own.
- An object is the actual thing in memory: it has an identity (a unique location), a type, and a value.
- A value is the concrete content the object represents (42, "hello", [1, 2, 3]).
- A data type classifies the object, defining what operations are valid on its value and how it is stored.

> **Note:** Variable -> (references) -> Object -> (has a) -> Value, and every Object -> (belongs to a) -> Data Type. Keep this chain in mind for the rest of this handbook: whenever you see 'x is an int', what is really meant is 'the object that x currently references is of type int'.

## Part 1 Practice Exercises

**Beginner:** What is printed by: x = 7; print(type(x))?

*Answer:* <class 'int'> is printed, because 7 is a whole-number literal with no decimal point, so Python classifies it as int.

**Intermediate:** Explain why isinstance(True, int) returns True even though True looks like a boolean, not a number.

*Answer:* bool is implemented in Python as a subclass of int (True behaves as 1 and False as 0 in numeric contexts). Because isinstance() checks the class and all of its ancestor classes, and int is an ancestor of bool, the check succeeds.

**Advanced:** Given a = [1, 2]; b = a; b.append(3), explain in terms of variables, objects and references why a also shows [1, 2, 3], without using the word 'copy'.

*Answer:* a and b are two separate names that were both bound to reference the exact same single list object in memory. b.append(3) mutates that one shared object in place. Since a references that same object, reading a afterward reflects the mutation too, because a was never a distinct object to begin with, it was always the same object as b.

# Part 3. Numeric Data Types: int, float and complex

Python provides three built-in numeric types, all under the abstract umbrella of 'numbers': int for whole numbers, float for decimal numbers, and complex for numbers with a real and an imaginary part. This part explains how each is created, represented internally, and where their edge cases lie.

## 3.1 Creating Integers

An integer literal is simply digits with no decimal point: 0, 42, -17. Python's int type has no fixed size limit the way a 32-bit integer in C does; it automatically grows to hold arbitrarily large whole numbers, limited only by available memory.

```python
positive = 42
negative = -17
zero = 0
huge = 123456789012345678901234567890
print(positive, negative, zero)
print(huge)
print(type(huge))
```

| Code / Item | Explanation |
|---|---|
| `huge = 123456789012345678901234567890` | Creates an integer far larger than a typical 64-bit machine word could hold; Python's int automatically uses an arbitrary-precision representation internally, so this works without overflow. |

## 3.2 Integer Arithmetic

Integers support the usual arithmetic operators. Two behaviors are worth calling out because they surprise newcomers from other languages: division with / always produces a float (true division), while // performs floor division and always produces a result rounded toward negative infinity.

```python
print(7 / 2)
print(7 // 2)
print(-7 // 2)
print(7 % 2)
print(2 ** 10)
```

| Code / Item | Explanation |
|---|---|
| `7 / 2` | True division always returns a float, 3.5, even though both operands are int. |
| `7 // 2` | Floor division returns 3, the largest integer not greater than 3.5. |
| `-7 // 2` | Returns -4, not -3, because floor division rounds toward negative infinity, not toward zero. |
| `2 ** 10` | Exponentiation, returning the integer 1024. |

## 3.3 Base Conversion: Binary, Octal, Hexadecimal

Integers can be written directly in binary (0b prefix), octal (0o prefix) or hexadecimal (0x prefix), and Python always stores them as ordinary base-10 integers internally; the prefix only affects how the literal is written in source code, not how the value behaves afterward.

```python
b = 0b1010
o = 0o12
h = 0xA
print(b, o, h)
print(bin(10), oct(10), hex(10))
```

| Code / Item | Explanation |
|---|---|
| `0b1010` | A binary literal equal to decimal 10. |
| `0o12` | An octal literal, also equal to decimal 10. |
| `0xA` | A hexadecimal literal, also equal to decimal 10; all three variables hold the identical int value 10, just written differently. |

## 3.4 Underscores in Numeric Literals

Since Python 3.6, underscores can be inserted into numeric literals purely for human readability; Python ignores them completely when parsing the value.

```python
population = 1_400_000_000
print(population)
print(population == 1400000000)
```

## 3.5 Important Integer Limitations and Edge Cases

- Integer division by zero (5 // 0 or 5 % 0) raises ZeroDivisionError; it does not return infinity the way float division by zero can in some contexts.
- True integer overflow does not exist for Python's int; the only practical limit is available memory, though extremely large integers become slow to compute with.
- Comparing an int to a float (5 == 5.0) works correctly and returns True, because Python compares by mathematical value, not by type.

## 3.6 Creating Floats

A float literal includes a decimal point, or uses scientific notation with an 'e'. Internally, Python's float is a C double, following the IEEE 754 double-precision standard, giving about 15 to 17 significant decimal digits of precision.

```python
price = 19.99
scientific = 6.022e23
negative_exp = 1.6e-19
print(price, scientific, negative_exp)
print(type(price))
```

## 3.7 Floating-Point Precision and Errors

Because floats are stored in binary, most decimal fractions cannot be represented exactly, the same way 1/3 cannot be written exactly in base-10 decimal. This produces small, well-known rounding artifacts that surprise many newcomers.

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(round(0.1 + 0.2, 10) == round(0.3, 10))
```

| Code / Item | Explanation |
|---|---|
| `0.1 + 0.2` | Prints 0.30000000000000004, not 0.3, because neither 0.1 nor 0.2 has an exact binary floating-point representation. |
| `0.1 + 0.2 == 0.3` | Returns False for the same reason; comparing floats for exact equality is unreliable. |
| `round(...) == round(...)` | Rounding both sides to a shared, coarser precision before comparing sidesteps the tiny representation error. |

> **Note:** For financial calculations where exact decimal precision matters, use Python's built-in decimal.Decimal type instead of float; float's binary approximation is unsuitable for money.

## 3.8 NaN and Infinity

float supports three special values beyond ordinary numbers: positive infinity, negative infinity, and NaN (Not a Number), which represents an undefined or unrepresentable numeric result.

```python
import math
pos_inf = float("inf")
neg_inf = float("-inf")
not_a_number = float("nan")
print(pos_inf, neg_inf, not_a_number)
print(math.isinf(pos_inf), math.isnan(not_a_number))
print(not_a_number == not_a_number)
```

| Code / Item | Explanation |
|---|---|
| `float("inf") / float("-inf")` | Create the positive and negative infinity float values, useful as sentinel starting points for min/max searches. |
| `not_a_number == not_a_number` | Returns False; NaN is defined by the IEEE 754 standard to never equal anything, including itself, which is why math.isnan() must be used to test for it rather than ==. |

## 3.9 Creating Complex Numbers

A complex number has a real part and an imaginary part, written with a trailing j (or J) on the imaginary component. Complex numbers are used in engineering, signal processing, and certain scientific/mathematical computations.

```python
z1 = 3 + 4j
z2 = complex(1, -2)
print(z1, z2)
print(z1.real, z1.imag)
print(z1 + z2)
print(z1 * z2)
print(abs(z1))
```

| Code / Item | Explanation |
|---|---|
| `3 + 4j` | Creates a complex number with real part 3.0 and imaginary part 4.0 using literal syntax. |
| `complex(1, -2)` | Creates the same kind of object using the complex() constructor instead of literal syntax. |
| `z1.real / z1.imag` | Access the real and imaginary components as separate float attributes. |
| `abs(z1)` | Returns the magnitude (Euclidean length) of the complex number, computed as the square root of real^2 + imag^2, here 5.0. |

## Part 3 Practice Exercises

**Beginner:** What type does 7 / 2 produce in Python 3, and what is the value?

*Answer:* It produces a float, 3.5, because the / operator always performs true division regardless of whether the operands are int.

**Intermediate:** Why does print(0.1 + 0.2) not print exactly 0.3?

*Answer:* Floats are stored in binary (base-2) floating point, and most decimal fractions, including 0.1 and 0.2, cannot be represented exactly in binary, the same way 1/3 cannot be written exactly as a finite decimal. The tiny representation error surfaces as 0.30000000000000004.

**Advanced:** Write an expression that safely checks whether a float variable x holds NaN, and explain why x == float('nan') would not work.

*Answer:* Use math.isnan(x) after import math. The equality check fails because IEEE 754 defines NaN as never equal to any value, including another NaN or itself, so x == float('nan') always evaluates to False regardless of what x holds.

# Part 4. The Boolean Data Type

bool is Python's type for logical truth values, but it has one unusual property compared to booleans in many other languages: it is implemented as a subclass of int, meaning True and False are, underneath, just the integers 1 and 0 with special printing behavior.

## 4.1 True and False as Objects

True and False are the only two instances of bool, and both are singleton objects: every occurrence of True in a program refers to the exact same object in memory.

```python
print(type(True))
print(isinstance(True, int))
print(True == 1)
print(False == 0)
print(True + True)
print(True is True)
```

| Code / Item | Explanation |
|---|---|
| `isinstance(True, int)` | True, because bool inherits from int in Python's type hierarchy. |
| `True == 1` | True, because bool values compare equal to their underlying integer counterparts. |
| `True + True` | Evaluates to 2, an int, because arithmetic on bool values falls back to their integer values. |

## 4.2 Truthy and Falsy Values

Every Python object, not just bool itself, can be evaluated in a boolean context (an if statement, a while loop, or a call to bool()). Objects that evaluate to False are called falsy; everything else is truthy.

- Falsy values: False, None, 0, 0.0, 0j, "" (empty string), [] (empty list), () (empty tuple), {} (empty dict), set() (empty set), and any custom object whose __bool__ or __len__ returns False/0.
- Truthy values: everything else, including any non-empty string, non-empty collection, and any nonzero number.

```python
values = [0, 1, -1, "", "text", [], [0], None, {}, {"a": 1}]
for v in values:
    print(repr(v), "->", bool(v))
```

## 4.3 Boolean Expressions and Conditions

Comparison operators (==, !=, <, >, <=, >=) and membership/identity checks (in, is) all produce bool results, which is why they are used directly as conditions in if and while statements.

```python
age = 20
is_adult = age >= 18
print(is_adult)
if is_adult:
    print("Access granted")
else:
    print("Access denied")
```

## 4.4 Boolean Conversion With bool()

bool(x) explicitly converts any object to True or False by applying Python's truthiness rules, rather than relying on implicit evaluation inside an if statement.

```python
print(bool(""))
print(bool("0"))   # non-empty string, even though it looks like zero
print(bool([]))
print(bool([False]))  # non-empty list, even though its only item is falsy
```

| Code / Item | Explanation |
|---|---|
| `bool("0")` | Returns True. A string containing the character '0' is a non-empty string, and any non-empty string is truthy, regardless of what characters it contains. |
| `bool([False])` | Returns True. A list containing one item, even a falsy item, is still a non-empty list, and non-empty collections are always truthy. |

> **Note:** A very common bug is confusing a falsy-looking value with an actually-falsy value: the string "0" and the string "False" are both truthy, because bool() only checks for emptiness/zero-ness of the container or number itself, not the meaning of its contents.

## Part 4 Practice Exercises

**Beginner:** Is the empty string "" truthy or falsy?

*Answer:* Falsy. bool("") is False, along with every other empty built-in collection.

**Intermediate:** What does True + True + False evaluate to, and why?

*Answer:* It evaluates to 2. Because bool is a subclass of int, True behaves as 1 and False behaves as 0 in arithmetic contexts, so the expression is really 1 + 1 + 0.

**Advanced:** A function reads a form field and stores the text "False" (the literal 4-character string) into a variable called flag. Explain why 'if flag:' will behave unexpectedly, and show the corrected check.

*Answer:* Because flag holds the string "False", not the boolean False, and any non-empty string is truthy, 'if flag:' will evaluate to True even though the text says 'False', which is almost certainly not the intended behavior. The fix is to compare explicitly: 'if flag == "False": is_true = False' or, better, convert the source data to an actual bool earlier in the pipeline rather than leaving it as text.

# Part 5. Strings (str)

A str is an immutable, ordered sequence of Unicode characters. Strings represent virtually all text in Python: names, sentences, file paths, JSON payloads. This part focuses on the str data type itself, how it stores and represents text, and how it behaves; a companion Methods handbook already covers the full catalogue of string methods in detail, so only the interaction between methods and the type is mentioned here where necessary.

## 5.1 Creating Strings

Strings can be written with single quotes, double quotes, or triple quotes (either ''' or """); Python treats all three as equivalent for a single-line string. Triple quotes additionally allow the string literal to span multiple lines directly in source code.

```python
s1 = 'single quoted'
s2 = "double quoted"
s3 = """triple
quoted
multiline"""
print(s1)
print(s2)
print(s3)
```

## 5.2 Escape Characters and Raw Strings

A backslash inside a string introduces an escape sequence: \n for newline, \t for tab, \\ for a literal backslash, \' and \" for quote characters inside a matching-quote string. A raw string, prefixed with r, disables escape processing entirely, which is especially useful for Windows file paths and regular expressions.

```python
normal = "Line1\nLine2\tTabbed"
raw = r"C:\Users\name\Documents"
print(normal)
print(raw)
```

| Code / Item | Explanation |
|---|---|
| `"Line1\nLine2\tTabbed"` | \n and \t are interpreted as a newline and a tab character respectively, not printed literally. |
| `r"C:\Users\name\Documents"` | The r prefix tells Python to treat every backslash as a literal character, so \U, \n and so on are not interpreted as escapes. |

## 5.3 Unicode

Since Python 3, every str is a sequence of Unicode code points natively, meaning a Python string can hold text from virtually any writing system or emoji without special handling.

```python
yoruba = "E kaaro"
emoji = "Python \U0001F40D"
print(yoruba)
print(emoji)
print(len(emoji))
```

## 5.4 String Indexing and Negative Indexing

Because a string is an ordered sequence, each character has a numeric position (index), starting at 0 for the first character. Negative indices count backward from the end, with -1 referring to the last character.

```python
word = "Python"
print(word[0])
print(word[5])
print(word[-1])
print(word[-6])
```

| Code / Item | Explanation |
|---|---|
| `word[0]` | The first character, 'P'. |
| `word[-1]` | The last character, 'n', counting from the end instead of the start. |

## 5.5 String Slicing

Slicing extracts a substring using [start:stop:step]. The start index is included, the stop index is excluded, and step controls the direction and interval; omitting any of the three uses Python's sensible defaults.

```python
word = "Python"
print(word[0:3])
print(word[:3])
print(word[3:])
print(word[::-1])
print(word[::2])
```

| Code / Item | Explanation |
|---|---|
| `word[0:3]` | Characters at index 0, 1 and 2 ('Pyt'); index 3 is excluded. |
| `word[::-1]` | A step of -1 walks the string backward, producing a fully reversed string, 'nohtyP'. |
| `word[::2]` | Every second character starting from index 0, 'Pto'. |

## 5.6 String Immutability

A str object can never be changed in place once created; every operation that appears to 'modify' a string, such as concatenation or a method call, actually creates and returns a brand-new string object, leaving the original untouched.

```python
s = "hello"
try:
    s[0] = "H"
except TypeError as e:
    print("TypeError:", e)

new_s = "H" + s[1:]
print(s, new_s)
```

| Code / Item | Explanation |
|---|---|
| `s[0] = "H"` | Raises TypeError: 'str' object does not support item assignment, because strings cannot be modified in place. |
| `"H" + s[1:]` | Builds an entirely new string by concatenation rather than mutating s; the original s remains 'hello'. |

## 5.7 String Comparison, Concatenation, Repetition and Membership

```python
print("apple" == "apple")
print("apple" < "banana")
print("Py" + "thon")
print("ab" * 3)
print("th" in "python")
```

| Code / Item | Explanation |
|---|---|
| `"apple" < "banana"` | Strings compare lexicographically (dictionary order), character by character using Unicode code point values. |
| `"Py" + "thon"` | Concatenation joins two strings into a new string, 'Python'. |
| `"ab" * 3` | Repetition repeats the string, producing 'ababab'. |
| `"th" in "python"` | Membership checks for a substring anywhere inside the string, returning True. |

## 5.8 f-Strings and Formatting

An f-string, prefixed with f, lets you embed expressions directly inside a string literal using curly braces; the expression is evaluated and converted to text at the point the string is created. Format specifications after a colon inside the braces control decimal places, padding, alignment, and more.

```python
name = "Gbenga"
score = 97.6543
print(f"{name} scored {score:.2f} percent")
print(f"{score:>10.1f}")
print(f"{1_000_000:,}")
```

| Code / Item | Explanation |
|---|---|
| `f"{score:.2f}"` | The .2f format spec rounds the float to two decimal places when converting it to text: '97.65'. |
| `f"{score:>10.1f}"` | Right-aligns the formatted number within a field 10 characters wide. |
| `f"{1_000_000:,}"` | Inserts thousands separators into the formatted integer: '1,000,000'. |

## 5.9 Converting Other Types to Strings

```python
print(str(42))
print(str(3.14))
print(str([1, 2, 3]))
print(str(True))
print(str(None))
```

> **Note:** str() is covered here as it applies specifically to text conversion; the full behavior of type conversion functions across all types is unified in Part 20.

# Part 6. Lists (list)

A list is Python's general-purpose, mutable, ordered sequence type. Conceptually, a list is best understood as an array of references: each slot in the list holds not a value directly, but a reference to an object elsewhere in memory, which is why a single list can freely mix ints, strings, other lists, or any other object type.

## 6.1 Creating Lists

```python
empty = []
numbers = [1, 2, 3, 4, 5]
mixed = [1, "two", 3.0, True, None]
from_constructor = list("abc")
print(empty, numbers, mixed, from_constructor)
```

| Code / Item | Explanation |
|---|---|
| `mixed = [1, "two", 3.0, True, None]` | A single list freely holding five completely different types, which is legal because each slot only stores a reference to its object, not the object's raw bytes inline. |
| `list("abc")` | The list() constructor treats its argument as an iterable and collects each element, producing ['a', 'b', 'c'] from the three characters of the string. |

## 6.2 Indexing, Negative Indexing and Slicing

```python
fruits = ["mango", "pawpaw", "banana", "kiwi"]
print(fruits[0])
print(fruits[-1])
print(fruits[1:3])
print(fruits[::-1])
```

## 6.3 Nested Lists

Because a list can hold any object, including other lists, lists naturally represent grids, matrices, or hierarchical data. Indexing into a nested list requires one set of brackets per level of nesting.

```python
grid = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(grid[1])
print(grid[1][2])
for row in grid:
    print(row)
```

## 6.4 List Mutability

Unlike a string, a list can be changed in place after creation: individual elements can be reassigned, and the list can grow or shrink, all without creating a new list object.

```python
nums = [1, 2, 3]
print(id(nums))
nums[0] = 100
nums.append(4)
print(nums)
print(id(nums))
```

| Code / Item | Explanation |
|---|---|
| `nums[0] = 100` | Overwrites the reference stored in slot 0 with a reference to the int 100, modifying the existing list object rather than creating a new one. |
| `id(nums) before and after` | Both calls return the same identity, proving the list object itself was never replaced, only its contents changed. |

## 6.5 Aliasing vs Copying

Because assignment binds a name to an existing object rather than duplicating it, writing b = a for a list a creates an alias, a second name for the same object, not an independent copy. To get a genuinely separate list, you must copy explicitly.

```python
a = [1, 2, 3]
alias = a
copy1 = a.copy()
copy2 = a[:]
alias.append(99)
print(a)      # changed, because alias IS a
print(copy1)  # unaffected, an independent list
print(copy2)  # also unaffected
```

| Code / Item | Explanation |
|---|---|
| `alias = a` | Creates an alias: alias and a refer to the same list object. |
| `a.copy() / a[:]` | Both create a brand-new list object containing the same top-level elements as a, independent of further changes to a. |

## 6.6 Shallow Copy vs Deep Copy

A shallow copy duplicates the outer list but still shares references to any nested mutable objects inside it; a deep copy recursively duplicates every nested object too, so the two structures share nothing.

```python
import copy
original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)

shallow[0].append(999)
print("original after shallow edit:", original)
print("deep, unaffected:", deep)
```

| Code / Item | Explanation |
|---|---|
| `copy.copy(original)` | Creates a new outer list, but the inner lists inside it are the same shared objects as in original. |
| `shallow[0].append(999)` | Mutates the inner list that both shallow and original still point to, so the change is visible through original too. |
| `copy.deepcopy(original)` | Recursively copies the inner lists as well, so deep shares no nested objects with original at all. |

## 6.7 List Unpacking

```python
point = [10, 20, 30]
x, y, z = point
print(x, y, z)

first, *rest = [1, 2, 3, 4, 5]
print(first, rest)
```

| Code / Item | Explanation |
|---|---|
| `x, y, z = point` | Unpacks each element of the list into its own variable, requiring the exact number of names to match the list length. |
| `first, *rest = [1, 2, 3, 4, 5]` | The * collects all remaining elements into a list, letting you unpack a variable-length sequence without knowing its exact size in advance. |

## 6.8 Lists as Stacks and General Sequences

A list's append() and pop() (from the end) operate in constant time, which makes list a natural, efficient stack (last-in, first-out) structure, in addition to its role as a general-purpose growable sequence.

```python
stack = []
stack.append("a")
stack.append("b")
stack.append("c")
print(stack.pop())
print(stack)
```

*Answer:* [20, 30]: everything from index 1 to the end of the list.

**Intermediate:** Explain the difference in outcome between b = a and b = a.copy() for a list a, in terms of what b refers to afterward.

*Answer:* b = a makes b an alias: both names reference the exact same list object, so mutating through either name affects what the other sees. b = a.copy() creates a distinct new list object with the same elements, so mutating b afterward does not affect a, and vice versa.

**Advanced:** Given original = [[1], [2]], why does copy.copy(original)[0].append(9) also change original[0], while copy.deepcopy(original)[0].append(9) does not?

*Answer:* copy.copy() performs a shallow copy: it creates a new outer list, but the inner list objects inside it are not duplicated, only referenced again, so original and the shallow copy still share the very same inner list objects. Appending to that shared inner list is visible through both. copy.deepcopy() recursively copies every nested object, so the deep copy's inner lists are entirely separate objects from original's, and mutating one has no effect on the other.

# Part 7. Tuples (tuple)

A tuple is an immutable, ordered sequence, essentially a list that can never be changed after creation. Tuples are used wherever a fixed collection of values needs to be treated as a single, hashable, protected unit.

## 7.1 Creating Tuples

Tuples are usually written with parentheses, though the parentheses are technically optional; what actually makes a tuple is the comma.

```python
t1 = (1, 2, 3)
t2 = 1, 2, 3       # parentheses are optional
empty = ()
single = (5,)       # the trailing comma is required
not_a_tuple = (5)   # this is just an int in parentheses
print(t1, t2, empty, single, type(not_a_tuple))
```

| Code / Item | Explanation |
|---|---|
| `t2 = 1, 2, 3` | The comma, not the parentheses, is what actually creates a tuple; Python builds the same tuple object here as t1. |
| `single = (5,)` | The trailing comma is mandatory for a one-element tuple; without it, (5) is just the integer 5 in redundant parentheses. |

## 7.2 Indexing, Slicing and Nested Tuples

```python
point3d = (10, 20, 30)
print(point3d[0])
print(point3d[-1])
print(point3d[0:2])

nested = ((1, 2), (3, 4))
print(nested[1][0])
```

## 7.3 Tuple Unpacking and Extended Unpacking

```python
coordinates = (4, 5)
x, y = coordinates
print(x, y)

first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)
```

| Code / Item | Explanation |
|---|---|
| `x, y = coordinates` | Unpacks a two-item tuple directly into two names in one line. |
| `first, *middle, last = (1, 2, 3, 4, 5)` | The starred name middle absorbs every element that is not claimed by first or last, here becoming [2, 3, 4]. |

## 7.4 Tuple Immutability

```python
point = (1, 2)
try:
    point[0] = 99
except TypeError as e:
    print("TypeError:", e)
```

> **Note:** Immutability applies to the tuple's own slots, not necessarily to what those slots reference: a tuple containing a list, such as ([1, 2], 3), cannot have its list replaced with a different object, but the list inside it can still be mutated in place, because the list itself is a mutable object.

## 7.5 Tuple vs List: When to Use Which

- Use a tuple for a fixed-size, fixed-meaning record, such as (latitude, longitude) or a database row, where the position of each element has a specific, unchanging meaning.
- Use a list for a variable-length, homogeneous collection that you expect to grow, shrink, sort, or otherwise modify.
- Tuples are hashable (when all their elements are hashable) and can therefore be used as dictionary keys or set elements; lists cannot.
- Tuples communicate intent: seeing a tuple in code signals 'this is a fixed, immutable record', which helps prevent accidental modification bugs.

## 7.6 Tuples as Dictionary Keys

```python
distances = {
    ("Lagos", "Abuja"): 756,
    ("Lagos", "Ibadan"): 128,
}
print(distances[("Lagos", "Ibadan")])
```

> **Note:** This works because tuples are immutable and hashable, satisfying dict's requirement that every key be hashable (Part 19 covers hashability in full).

## 7.7 Named Tuples (Advanced)

collections.namedtuple() creates a tuple subclass whose elements can also be accessed by name instead of only by position, combining the memory efficiency and immutability of a tuple with the readability of attribute access.

```python
from collections import namedtuple
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(p)
print(p.x, p.y)
print(p[0], p[1])
print(isinstance(p, tuple))
```

| Code / Item | Explanation |
|---|---|
| `namedtuple("Point", ["x", "y"])` | Dynamically generates a new tuple subclass called Point, with named fields x and y. |
| `p.x, p.y` | Access the same values as p[0] and p[1] would, but with self-documenting attribute names instead of numeric positions. |

## Part 7 Practice Exercises

**Beginner:** How do you write a one-element tuple containing just the number 7?

*Answer:* (7,) - the trailing comma is required; (7) alone is just the integer 7.

**Intermediate:** Why can a tuple be used as a dictionary key while a list cannot?

*Answer:* Dictionary keys must be hashable, and hashability generally requires immutability so the hash value cannot change after insertion. A tuple is immutable (and hashable, provided its elements are also hashable), so it qualifies. A list is mutable, so Python deliberately makes it unhashable to prevent a key's hash from silently becoming invalid after the key object is mutated.

**Advanced:** Given record = ([1, 2], "fixed"), explain why record[0].append(3) succeeds without error even though record is a tuple.

*Answer:* Tuple immutability protects the tuple's own slots, meaning you cannot reassign record[0] to point at a different object. It says nothing about whether the object a slot references is itself mutable. record[0] references a list, and lists remain fully mutable regardless of what container holds a reference to them, so appending to that list in place is entirely legal.

# Part 8. The range Type

range is a distinct, immutable sequence type built specifically to represent an arithmetic progression of integers without storing every single value in memory. It is what powers the vast majority of for loops that need to repeat a fixed number of times.

## 8.1 The Three Forms of range()

```python
r1 = range(5)
r2 = range(2, 8)
r3 = range(0, 20, 5)
print(list(r1))
print(list(r2))
print(list(r3))
```

| Code / Item | Explanation |
|---|---|
| `range(5)` | Produces integers 0, 1, 2, 3, 4: a single argument is treated as the stop value, with start defaulting to 0 and step defaulting to 1. |
| `range(2, 8)` | Produces 2 through 7: start is included, stop is excluded. |
| `range(0, 20, 5)` | Produces 0, 5, 10, 15, stepping by 5 each time. |

## 8.2 Positive Step, Negative Step and Empty Ranges

```python
print(list(range(10, 0, -2)))
print(list(range(5, 5)))
print(list(range(5, 1)))
```

| Code / Item | Explanation |
|---|---|
| `range(10, 0, -2)` | A negative step counts downward: 10, 8, 6, 4, 2. |
| `range(5, 5)` | Start equals stop, producing an empty range with zero elements. |
| `range(5, 1)` | With a default positive step, start is already past stop in the wrong direction, so this also produces an empty range rather than an error. |

## 8.3 Indexing, Slicing and Membership

Despite never materializing its full contents, range fully supports indexing, slicing (which returns another range object) and fast membership testing, all computed arithmetically rather than by scanning.

```python
r = range(0, 100, 10)
print(r[3])
print(r[2:5])
print(50 in r)
print(55 in r)
```

| Code / Item | Explanation |
|---|---|
| `r[3]` | Computed directly as start + 3*step (30), not by walking through the sequence. |
| `r[2:5]` | Slicing a range returns a new range object, range(20, 50, 10), still not an actual list. |
| `50 in r / 55 in r` | Membership is checked with arithmetic (is 55 reachable by start + k*step for some integer k?) in constant time, not by iterating. |

## 8.4 Why range Is Not a List

range(1_000_000) uses a small, fixed amount of memory (just the start, stop and step values), regardless of how large the range is, because it computes each value on demand instead of storing a million integers. A list of a million integers, in contrast, allocates memory for every single element up front.

```python
import sys
big_range = range(1_000_000)
big_list = list(range(1_000_000))
print(sys.getsizeof(big_range))
print(sys.getsizeof(big_list))
```

> **Note:** This memory efficiency is precisely why 'for i in range(n):' is the idiomatic way to loop n times in Python, instead of building an explicit list of numbers first.

## Part 8 Practice Exercises

**Beginner:** What integers does range(3, 12, 3) produce?

*Answer:* 3, 6, 9. Start is included (3), stop is excluded (12), stepping by 3.

**Intermediate:** Why does range(1_000_000) use dramatically less memory than list(range(1_000_000))?

*Answer:* range stores only three integers internally, start, stop and step, and computes any requested value arithmetically on demand. A list must actually allocate storage for every one of its million elements up front, so its memory use scales with the number of elements while range's does not.

**Advanced:** Without converting it to a list, write an expression that checks whether 999999 is a member of range(0, 1_000_000, 3), and explain why this check is fast even for a huge range.

*Answer:* 999999 in range(0, 1_000_000, 3) is the expression. It is fast because range's __contains__ implementation checks membership arithmetically: it verifies the value falls between start and stop, and that (value - start) is evenly divisible by step, all in constant time, rather than iterating through every element to search for a match.

# Part 9. Dictionaries (dict)

A dict is Python's built-in mapping type: a mutable, ordered collection of key-value pairs, where each key must be unique and hashable. Internally, dict is implemented as a hash table, which is what gives key lookups their famous near-constant-time speed regardless of how many items the dictionary holds.

## 9.1 Creating Dictionaries

```python
empty = {}
student = {"name": "Ada", "age": 21, "active": True}
from_pairs = dict([("a", 1), ("b", 2)])
from_kwargs = dict(x=1, y=2)
print(student, from_pairs, from_kwargs)
```

> **Note:** {} creates an empty dictionary, not an empty set; Part 10 explains why, and how to create an empty set correctly.

## 9.2 Accessing, Adding, Updating and Deleting Data

```python
student = {"name": "Ada", "age": 21}
print(student["name"])
print(student.get("gpa", "not recorded"))

student["gpa"] = 3.8       # add
student["age"] = 22          # update
del student["gpa"]           # delete
print(student)
```

| Code / Item | Explanation |
|---|---|
| `student["name"]` | Direct key access; raises KeyError if the key is missing. |
| `student.get("gpa", "not recorded")` | Safe lookup with a fallback default, avoiding a KeyError for a key that may not exist. |
| `student["gpa"] = 3.8` | Assigning to a key that does not yet exist adds a brand-new key-value pair. |
| `student["age"] = 22` | Assigning to a key that already exists overwrites its value in place. |
| `del student["gpa"]` | Removes a key-value pair entirely. |

## 9.3 Dictionary Nesting

Values in a dictionary can be any object, including other dictionaries or lists, which lets dict represent arbitrarily deep, structured, JSON-like data naturally.

```python
company = {
    "name": "Prodigy Training Hub",
    "location": {"city": "Lagos", "country": "Nigeria"},
    "courses": ["Data Science", "AI Engineering", "BI Tools"],
}
print(company["location"]["city"])
print(company["courses"][0])
```

## 9.4 Valid Dictionary Keys and Hashability

A dictionary key must be hashable, which in practice means it must also be immutable: int, float, str, bool, tuple (of hashable elements) and frozenset are all valid keys. list, dict and set are mutable and therefore unhashable, so none of them can ever be used as a dictionary key.

```python
valid = {1: "a", "two": "b", (3, 4): "c", True: "d"}
print(valid)
try:
    invalid = {[1, 2]: "bad key"}
except TypeError as e:
    print("TypeError:", e)
```

> **Note:** Mutable objects generally cannot be dictionary keys because a dict relies on a key's hash value staying constant for the object's entire lifetime, to know which internal bucket to search. If a list's contents (and therefore its notional hash) could change after it was used as a key, the dictionary would no longer be able to find it, silently corrupting lookups; Python avoids this entire class of bug by refusing to hash mutable types at all.

## 9.5 Dictionary Ordering

Since Python 3.7, dictionaries are guaranteed by the language specification to preserve insertion order: iterating over a dict's keys, values, or items always visits them in the order they were first added, and re-assigning an existing key's value does not change its position.

```python
d = {}
d["z"] = 1
d["a"] = 2
d["m"] = 3
print(list(d.keys()))
d["z"] = 99
print(list(d.keys()))
```

## 9.6 Dictionary Copying

```python
import copy
original = {"nums": [1, 2, 3]}
shallow = original.copy()
deep = copy.deepcopy(original)
shallow["nums"].append(4)
print(original["nums"])
print(deep["nums"])
```

> **Note:** Exactly as with lists (Part 6), dict.copy() only performs a shallow copy: nested mutable values are still shared between the original and the copy unless copy.deepcopy() is used instead.

## 9.7 Dictionary Unpacking

```python
defaults = {"theme": "dark", "font_size": 12}
overrides = {"font_size": 16, "language": "en"}
merged = {**defaults, **overrides}
print(merged)
```

| Code / Item | Explanation |
|---|---|
| `{**defaults, **overrides}` | Unpacks both dictionaries' key-value pairs into a new dictionary; when the same key appears in both (font_size here), the value from whichever dictionary was unpacked last wins. |

## 9.8 Dictionary Comprehensions

```python
squares = {n: n ** 2 for n in range(1, 6)}
print(squares)

prices = {"apple": 100, "banana": 50, "mango": 150}
affordable = {k: v for k, v in prices.items() if v <= 100}
print(affordable)
```

| Code / Item | Explanation |
|---|---|
| `{n: n**2 for n in range(1, 6)}` | Builds a new dictionary in one expression, mapping each number to its square. |
| `{k: v for k, v in prices.items() if v <= 100}` | Filters key-value pairs with a condition while building the new dictionary, keeping only affordable items. |

## Part 9 Practice Exercises

**Beginner:** How do you safely look up the key "score" in a dict d, returning 0 if it is missing, without raising an error?

*Answer:* d.get("score", 0)

**Intermediate:** Why does {[1, 2]: "value"} raise a TypeError?

*Answer:* A list is a mutable object and Python deliberately makes mutable built-in types unhashable, since a dictionary key's hash value must never change after the key is inserted. Because [1, 2] cannot be hashed, it cannot be used as a dictionary key, and Python raises TypeError: unhashable type: 'list'.

**Advanced:** Given config = {"a": [1, 2]}, explain why shallow = config.copy(); shallow["a"].append(3) also changes config["a"], and show the fix using the copy module.

*Answer:* dict.copy() performs only a shallow copy: it creates a new top-level dictionary, but every value that is itself a mutable object, such as the list under "a", is not duplicated, only referenced again. Both config and shallow therefore point to the same underlying list, so appending through either name is visible through both. The fix is 'import copy; deep = copy.deepcopy(config)', which recursively duplicates nested mutable values so the two dictionaries share nothing.

# Part 10. Sets (set)

A set is a mutable, unordered collection of unique, hashable elements. Sets are modeled directly on mathematical sets, and Python's set type implements the classic set-theory operations (union, intersection, difference) as fast, built-in operators and methods.

## 10.1 Creating Sets and the {} Trap

Sets are written with curly braces containing comma-separated values, or built with the set() constructor. Critically, empty curly braces {} do NOT create an empty set; they create an empty dictionary, because {} was reserved for dict long before set literals existed in the language. The only correct way to create an empty set is set().

```python
numbers = {1, 2, 3, 2, 1}
print(numbers)

looks_empty = {}
print(type(looks_empty))

truly_empty_set = set()
print(type(truly_empty_set))
```

| Code / Item | Explanation |
|---|---|
| `{1, 2, 3, 2, 1}` | Automatically discards duplicate values, storing only the unique elements {1, 2, 3}. |
| `looks_empty = {}` | Creates an empty dict, not an empty set; this is one of the most common beginner mistakes in Python. |
| `truly_empty_set = set()` | The only syntax that actually produces an empty set. |

## 10.2 Uniqueness and Unordered Nature

A set can never contain duplicate elements; adding an element that already exists is a silent no-op. Sets also have no defined order: printing the same set twice, or on two different runs, is not guaranteed to show elements in the same sequence, and sets cannot be indexed with [0].

```python
s = {3, 1, 2}
s.add(1)   # no effect, 1 is already present
print(s)
try:
    s[0]
except TypeError as e:
    print("TypeError:", e)
```

## 10.3 Fast Membership Testing

```python
large_set = set(range(1_000_000))
print(999_999 in large_set)
```

> **Note:** Membership testing in a set is near-constant time on average, because it works the same way dict lookups do, via hashing, rather than scanning element by element the way 'x in a_list' would for a list.

## 10.4 Set Operations (Mathematical Interpretation)

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
print(a ^ b)   # symmetric difference
print({1, 2}.issubset(a))
print(a.issuperset({1, 2}))
print({1, 2}.isdisjoint({99, 100}))
```

| Code / Item | Explanation |
|---|---|
| `a \| b` | Union: every element that appears in a or b (or both), {1, 2, 3, 4, 5, 6}. |
| `a & b` | Intersection: only elements present in both a and b, {3, 4}. |
| `a - b` | Difference: elements in a that are not in b, {1, 2}. |
| `a ^ b` | Symmetric difference: elements in exactly one of the two sets, not both, {1, 2, 5, 6}. |
| `issubset() / issuperset()` | Check whether one set's elements are entirely contained within (or entirely contain) another set. |
| `isdisjoint()` | Returns True if the two sets share no elements at all. |

## 10.5 Set Comprehensions

```python
words = ["apple", "banana", "apple", "kiwi", "banana"]
unique_lengths = {len(w) for w in words}
print(unique_lengths)
```

> **Note:** A set comprehension uses the same curly-brace syntax as a dict comprehension, but with a single expression instead of a key:value pair, which is what tells Python to build a set rather than a dict.

## 10.6 Practical Applications

- Removing duplicates from a list: unique_items = list(set(original_list)).
- Fast membership checks against a large collection of allowed or blocked values.
- Comparing two collections to find what changed, what is common, or what is missing entirely.
- Tagging systems, where each item has a set of unique labels/categories.

## Part 10 Practice Exercises

**Beginner:** What does {} create in Python: an empty set or an empty dict?

*Answer:* An empty dict. The only way to create an empty set is with the set() constructor.

**Intermediate:** Given a = {1, 2, 3} and b = {2, 3, 4}, what is a ^ b and what does it mean conceptually?

*Answer:* a ^ b is {1, 4}, the symmetric difference: every element that belongs to exactly one of the two sets but not both, i.e. everything except the overlap.

**Advanced:** You have a list of ten thousand user IDs and need to repeatedly check whether incoming IDs already exist in that list. Explain why converting the list to a set first is a significant performance improvement.

*Answer:* Checking 'x in a_list' requires scanning the list element by element in the worst case, an O(n) operation, so repeating that check many times against a ten-thousand-element list costs roughly O(n) per check. Converting the list to a set once costs O(n) total, but afterward, 'x in a_set' uses hashing to jump almost directly to the relevant bucket, an O(1) operation on average. For many repeated membership checks, the one-time conversion cost is quickly repaid by the dramatically faster lookups.

# Part 11. Frozenset (frozenset)

A frozenset is the immutable counterpart to set: same unordered, unique-element behavior, but once created it can never be modified. That immutability is what makes a frozenset hashable, which unlocks two uses an ordinary set can never have: being a dictionary key, or being an element inside another set.

## 11.1 Why frozenset Exists

set is deliberately unhashable because it is mutable (Part 19 explains why mutability and hashability are incompatible). But sometimes you genuinely want set-like behavior, uniqueness and fast membership, for a fixed, unchanging collection that also needs to act as a hashable value elsewhere. frozenset fills exactly that gap.

## 11.2 Creating and Comparing to set

```python
mutable = {1, 2, 3}
frozen = frozenset([1, 2, 3])
print(frozen)
print(type(frozen))
try:
    frozen.add(4)
except AttributeError as e:
    print("AttributeError:", e)
```

| Code / Item | Explanation |
|---|---|
| `frozenset([1, 2, 3])` | Builds an immutable set from any iterable, here a list. |
| `frozen.add(4)` | Raises AttributeError, because frozenset deliberately has no add(), remove(), or any other mutating method. |

## 11.3 frozenset as a Dictionary Key

```python
group_scores = {
    frozenset({"Ada", "Bo"}): 180,
    frozenset({"Chi", "Dumi"}): 165,
}
print(group_scores[frozenset({"Bo", "Ada"})])
```

> **Note:** Because set order does not matter for equality, frozenset({"Ada", "Bo"}) and frozenset({"Bo", "Ada"}) are equal and hash identically, so either order retrieves the same dictionary entry.

## 11.4 frozenset Inside Another Set

```python
combinations = {frozenset({1, 2}), frozenset({2, 3}), frozenset({1, 2})}
print(len(combinations))
```

> **Note:** This works only because frozenset is hashable; attempting {set(), set()} with ordinary sets would raise TypeError, since a set cannot contain other unhashable sets.

## 11.5 Set Operations Still Work

```python
a = frozenset({1, 2, 3})
b = frozenset({2, 3, 4})
print(a | b)
print(a & b)
print(type(a | b))
```

> **Note:** frozenset supports the same union, intersection, difference and symmetric-difference operations as set; the result of combining two frozensets is itself a frozenset.

## Part 11 Practice Exercises

**Beginner:** What is the key difference in capability between set and frozenset?

*Answer:* set is mutable and supports methods like add() and remove(); frozenset is immutable and has no mutating methods at all once created.

**Intermediate:** Why can a frozenset be used as a dictionary key while an ordinary set cannot?

*Answer:* Dictionary keys must be hashable, and Python only allows hashing of immutable objects because a key's hash value must never change after insertion. frozenset is immutable, so it is hashable and valid as a key; set is mutable, so Python deliberately makes it unhashable to prevent that key's hash from becoming stale.

**Advanced:** Design a data structure using frozenset to track which unique pairs of students have already been grouped together for a class project, ensuring {"Ada","Bo"} and {"Bo","Ada"} are treated as the same pair.

*Answer:* used_pairs = set(); to record a pairing, do used_pairs.add(frozenset({name1, name2})); to check if a pairing already happened, test frozenset({name1, name2}) in used_pairs. Wrapping each two-student pairing in a frozenset makes the pairing order-independent (since frozenset equality ignores element order) and hashable, so it can live inside the outer set, giving automatic duplicate-pairing detection regardless of which name was mentioned first.

# Part 12. Bytes (bytes)

bytes is an immutable sequence type whose elements are integers from 0 to 255, representing raw binary data rather than text. Any time Python reads a file in binary mode, receives data over a network socket, or works with images, audio, or any non-text format, that data arrives as bytes.

## 12.1 Bytes vs Strings

A str is a sequence of abstract Unicode characters; a bytes object is a sequence of raw 8-bit integers with no inherent text meaning at all. The bridge between the two is an explicit encoding (turning text into bytes) or decoding (turning bytes back into text) step, using a specific character encoding scheme such as UTF-8 or ASCII.

## 12.2 Creating Bytes

```python
b1 = b"hello"
b2 = bytes([104, 101, 108, 108, 111])
b3 = bytes(5)
print(b1, b2, b3)
print(type(b1))
```

| Code / Item | Explanation |
|---|---|
| `b"hello"` | A bytes literal, prefixed with b; each character must be representable in a single byte (ASCII range). |
| `bytes([104, 101, 108, 108, 111])` | Constructs bytes directly from a list of integer byte values (ASCII codes spelling 'hello'). |
| `bytes(5)` | With a single integer argument, creates a zero-filled bytes object of that length: b'\x00\x00\x00\x00\x00'. |

## 12.3 Encoding and Decoding

```python
text = "Naira: \u20a6"
encoded = text.encode("utf-8")
print(encoded)
print(type(encoded))

decoded = encoded.decode("utf-8")
print(decoded)
print(decoded == text)
```

| Code / Item | Explanation |
|---|---|
| `text.encode("utf-8")` | Converts the str into a bytes object using the UTF-8 encoding rules, the most common encoding on the modern web. |
| `encoded.decode("utf-8")` | Reverses the process, converting the bytes back into a str, using the same encoding that was used to produce them. |

> **Note:** Decoding bytes with the wrong encoding (for instance, decoding UTF-8 bytes as ASCII when they contain non-ASCII characters) raises a UnicodeDecodeError; the encoding used to decode must match the encoding used to encode.

## 12.4 Indexing and Slicing Bytes

```python
data = b"Python"
print(data[0])       # an int, not a single-character bytes object
print(data[0:2])     # a slice IS a bytes object
print(bytes([data[0]]))
```

| Code / Item | Explanation |
|---|---|
| `data[0]` | Indexing a single position returns an int (80, the ASCII code for 'P'), not a one-byte bytes object, which surprises many newcomers. |
| `data[0:2]` | Slicing, unlike single indexing, does return a bytes object, b'Py'. |

## 12.5 Bytes Immutability and Practical Applications

- Reading and writing files opened in binary mode ('rb', 'wb').
- Sending and receiving data over network sockets, which always operate on raw bytes.
- Working with binary file formats: images, audio, compressed archives, serialized data.
- Cryptographic operations, which are defined over raw byte sequences, not text.

```python
data = b"hello"
try:
    data[0] = 72
except TypeError as e:
    print("TypeError:", e)
```

## Part 12 Practice Exercises

**Beginner:** What does b"hi"[0] return: an int or a one-character bytes object?

*Answer:* An int, 104 (the ASCII code for 'h'). Single-index access on bytes always returns an integer; only slicing returns another bytes object.

**Intermediate:** Given a str containing the Naira symbol, why must you call .encode() before writing it to a file opened in binary mode?

*Answer:* A file opened in binary mode ('wb') only accepts bytes-like objects, not str, because it performs no text encoding of its own. Calling .encode("utf-8") (or another chosen encoding) explicitly converts the abstract Unicode text into a concrete sequence of bytes that matches what the binary-mode file write() method requires.

**Advanced:** You receive bytes data over a network socket that was encoded as UTF-8, but you mistakenly call data.decode("ascii"). Explain what happens and why, for non-ASCII content.

*Answer:* ASCII can only represent code points 0 through 127 (one byte, high bit unset), while UTF-8 can encode any Unicode character using one to four bytes, with continuation bytes that have their high bit set. If the bytes contain any character outside the ASCII range (encoded as UTF-8's multi-byte sequences), decoding with 'ascii' encounters a byte value that has no valid ASCII interpretation, and Python raises a UnicodeDecodeError rather than silently producing corrupted or partial text.

# Part 13. Bytearray (bytearray)

bytearray is the mutable counterpart to bytes: the same sequence of raw byte values, but modifiable in place, element by element, without creating a brand-new object for every change.

## 13.1 Bytes vs Bytearray

Everywhere bytes offers read-only binary data, bytearray offers the same capability plus mutation: you can reassign individual byte values, append new bytes, or slice-assign a whole region, all in place.

## 13.2 Creating Bytearrays

```python
ba1 = bytearray(b"hello")
ba2 = bytearray(5)
ba3 = bytearray([72, 105])
print(ba1, ba2, ba3)
```

## 13.3 Indexing, Slicing and In-Place Modification

```python
data = bytearray(b"Hello")
data[0] = 74
print(data)
data[1:3] = b"XY"
print(data)
data.append(33)
print(data)
```

| Code / Item | Explanation |
|---|---|
| `data[0] = 74` | Overwrites the first byte in place, changing 'H' (72) to 'J' (74). |
| `data[1:3] = b"XY"` | Slice assignment replaces a whole region of bytes at once, even with a replacement of a different length. |
| `data.append(33)` | Grows the bytearray by one byte, adding '!' (ASCII 33) to the end. |

## 13.4 Converting Between bytes and bytearray

```python
immutable = bytes(bytearray(b"data"))
mutable = bytearray(bytes(b"data"))
print(type(immutable), type(mutable))
```

> **Note:** Each constructor accepts the other type directly, making it easy to switch between a locked-down, shareable bytes view and a working, editable bytearray buffer.

## 13.5 Practical Applications

- Building up a binary message or file piece by piece without repeatedly allocating new bytes objects.
- In-place decryption/encryption or checksum manipulation on a binary buffer.
- Reading data into a pre-allocated buffer for performance-sensitive I/O, using readinto().

## Part 13 Practice Exercises

**Beginner:** Which of bytes and bytearray supports item assignment like ba[0] = 65?

*Answer:* bytearray. bytes is immutable and raises TypeError on item assignment, exactly like a str would.

**Intermediate:** Write code that converts a bytearray back into an immutable bytes object.

*Answer:* bytes(my_bytearray) - passing a bytearray into the bytes() constructor produces an equivalent, independent, immutable bytes object.

**Advanced:** Explain a practical scenario where using bytearray instead of repeatedly concatenating bytes objects would meaningfully improve performance.

*Answer:* Building up a large binary payload by repeated bytes concatenation (result = result + new_chunk) creates a brand-new bytes object on every single concatenation, since bytes is immutable, which becomes an O(n^2) pattern overall as the payload grows. Using a bytearray and calling extend() or append() in a loop modifies the same underlying buffer in place, avoiding the repeated allocation and copying, which is significantly faster for assembling large binary data such as a file being streamed in chunks over a network.

# Part 14. Memoryview (memoryview)

memoryview is a special object that exposes another object's internal binary buffer directly, without copying its bytes, using Python's buffer protocol. It exists purely for efficiency when working with large binary data: it lets you slice, view, and even edit portions of a buffer while avoiding the cost of duplicating that data in memory.

## 14.1 Why memoryview Exists

Slicing an ordinary bytes or bytearray object (data[1000:2000]) creates a brand-new object containing a full copy of those bytes. For a large buffer, repeatedly slicing like this to inspect different regions wastes both time and memory. memoryview instead creates a lightweight window onto the exact same underlying memory, with no copying at all.

## 14.2 The Buffer Protocol

The buffer protocol is a low-level C-level interface that certain Python objects, including bytes, bytearray and numpy arrays, expose to allow direct access to their raw memory. memoryview is the Python-level tool for working with any object that implements this protocol.

## 14.3 Creating and Using a memoryview

```python
data = bytearray(b"Hello, World!")
view = memoryview(data)
print(view[0])
print(bytes(view[0:5]))

# Editing through the view mutates the original buffer directly
view[0] = 74
print(data)
```

| Code / Item | Explanation |
|---|---|
| `memoryview(data)` | Creates a view object referencing data's existing memory buffer; no bytes are copied. |
| `view[0:5]` | Slicing a memoryview also produces another memoryview (no copy); wrapping it in bytes() materializes those bytes as a concrete, independent object only when you actually need one. |
| `view[0] = 74` | Writes directly into the original bytearray's memory through the view, which is only possible because the underlying object, a bytearray, is itself mutable. |

## 14.4 Relationship With bytes and bytearray

- A memoryview over an immutable bytes object is itself read-only: attempting item assignment through it raises TypeError.
- A memoryview over a mutable bytearray allows in-place writes through the view, which is reflected immediately in the original bytearray.
- Converting a memoryview back to a concrete bytes or bytearray object (bytes(view), bytearray(view)) does copy the data at that point, which is expected and sometimes necessary.

## 14.5 Practical Use Cases

- Processing large files or network buffers in fixed-size chunks without copying each chunk out of the main buffer.
- Interfacing with C extensions and libraries such as numpy, which rely heavily on the buffer protocol for zero-copy array operations.
- Parsing binary protocols (image formats, network packets) where slicing out header/payload regions repeatedly would otherwise be wasteful.

## Part 14 Practice Exercises

**Beginner:** What is the main advantage memoryview offers over simply slicing a bytes object directly?

*Answer:* memoryview avoids copying the underlying data; slicing a bytes object directly always creates a new, independent copy of that region, while a memoryview creates a zero-copy window onto the same memory.

**Intermediate:** Why can you write through a memoryview created from a bytearray, but not through one created from a bytes object?

*Answer:* A memoryview does not change the mutability of the object it wraps; it merely exposes that object's existing buffer. bytes is immutable, so a memoryview over it inherits that read-only restriction and raises TypeError on assignment. bytearray is mutable, so a memoryview over it can write directly into that same mutable buffer.

**Advanced:** Describe, in terms of the buffer protocol, why memoryview is especially valuable when working with a very large binary file that needs to be split into many chunks for processing.

*Answer:* Without memoryview, extracting each chunk with ordinary slicing (large_buffer[i:i+chunk_size]) would allocate a brand-new bytes or bytearray object and copy that chunk's data into it every single time, which for a large file processed in many small chunks adds up to significant redundant memory allocation and copying overhead. A memoryview lets you create lightweight, zero-copy sub-views of specific chunk ranges directly over the original buffer via the buffer protocol, so each chunk can be inspected or passed to another buffer-protocol-aware function without ever duplicating the underlying bytes, which is considerably more memory- and time-efficient for large data.

# Part 15. None and NoneType

None is a single, special object that represents the deliberate absence of a value. Its type, NoneType, has exactly one possible instance: None itself is a singleton, meaning every reference to None throughout an entire Python program points to the exact same object.

## 15.1 Why Python Uses None

Many situations need a way to say 'there is no meaningful value here yet' that is distinct from any real, valid data value like 0, an empty string, or False. None fills that role: it is a explicit, unambiguous marker for missing, unset, or not-yet-computed data.

## 15.2 Functions That Return None

Any Python function without an explicit return statement, or with a bare return, automatically returns None. This is extremely common and is the single most frequent source of the 'I got None back and didn't expect it' bug.

```python
def greet(name):
    print(f"Hello, {name}")
    # no return statement

result = greet("Gbenga")
print(result)
print(type(result))
```

| Code / Item | Explanation |
|---|---|
| `def greet(name): ... (no return)` | The function performs a side effect (printing) but never explicitly returns a value. |
| `result = greet("Gbenga")` | result is assigned None, because that is what a function with no return statement always produces, regardless of what it printed internally. |

## 15.3 None vs False vs 0 vs Empty Collections

All of these are falsy in a boolean context, which leads beginners to treat them as interchangeable, but they carry entirely different meanings and should generally not be confused.

| Code / Item | Explanation |
|---|---|
| `None` | Represents 'no value exists here at all', an absence, not a quantity. |
| `False` | Represents a specific boolean truth value: 'this condition is not true'. |
| `0` | Represents a specific, valid numeric value: zero, a legitimate quantity. |
| `"" / [] / {}` | Represent a specific, valid but empty collection or string; the container exists, it simply has no contents. |

## 15.4 Correctly Checking for None

The idiomatic, recommended way to check whether a variable is None is the identity operator is, not equality (==), because None is a true singleton, and is checks object identity directly, which is both faster and immune to a custom object overriding __eq__ in a way that could make 'x == None' behave unexpectedly.

```python
value = None
print(value is None)       # idiomatic
print(value == None)       # works, but not the convention

if value is None:
    print("No value provided")
```

> **Note:** PEP 8, Python's official style guide, explicitly recommends 'is None' and 'is not None' over '== None' and '!= None'.

## 15.5 Practical Applications

```python
def find_user(user_id, users):
    for user in users:
        if user["id"] == user_id:
            return user
    return None  # explicit signal: no matching user was found

users = [{"id": 1, "name": "Ada"}, {"id": 2, "name": "Bo"}]
result = find_user(3, users)
if result is None:
    print("User not found")
else:
    print(result)
```

> **Note:** Using a default argument value of None (def f(x=None):) instead of a mutable object like [] is also a very common and important pattern, discussed further alongside mutability in Part 16.

## Part 15 Practice Exercises

**Beginner:** What does a function that has no return statement return?

*Answer:* None, automatically.

**Intermediate:** Why is 'if value is None:' generally preferred over 'if value == None:'?

*Answer:* is checks for object identity, and since None is a singleton (only one None object ever exists), identity comparison is both a precise, unambiguous test and typically faster than an equality comparison. Using == also relies on the __eq__ method of value's class behaving correctly, which a custom class could override in a way that makes 'value == None' return something unexpected (like True for a non-None object), whereas 'is None' cannot be fooled that way.

**Advanced:** Explain the conceptual difference between a variable holding None and a variable holding 0 or an empty string, and give a real scenario where confusing the two would cause a bug.

*Answer:* None represents the complete absence of a value, while 0 and "" are specific, valid values that happen to be falsy. Consider a function that looks up a student's exam score and is meant to return None if the student has not yet taken the exam, versus 0 if the student took it and scored zero. Code that writes 'if not score:' to check for a missing exam would incorrectly treat a genuine score of 0 the same as a missing score, silently hiding a real (if unfortunate) data point as though the exam were never taken. The correct check is 'if score is None:', which only matches the true absence-of-value case.

# Part 16. Mutable vs Immutable Types

This is one of the most important conceptual distinctions in all of Python, because it explains behavior that otherwise looks mysterious: why some 'copies' aren't really copies, why some function arguments seem to change unexpectedly, and why some objects can be dictionary keys while others cannot. This part treats the topic in full depth.

## 16.1 What Mutability Means

Mutability describes whether an object's internal state can be changed after it has been created, without creating a new object. A mutable object can be edited in place: its identity (id()) stays the same before and after the change. An immutable object can never be edited in place; any operation that looks like a modification actually builds and returns an entirely new object, leaving the original completely untouched.

## 16.2 Immutable Built-in Types

- int, float, complex, bool - all numeric types.
- str - text.
- tuple - as long as, importantly, this only protects the tuple's own slots (see 16.7).
- frozenset.
- bytes.
- range.
- NoneType (None is a singleton, and there is nothing to mutate).

## 16.3 Mutable Built-in Types

- list.
- dict.
- set.
- bytearray.

## 16.4 Proving Immutability With id()

```python
x = 100
print(id(x))
x = x + 1
print(id(x))   # a DIFFERENT id: a new int object was created
```

| Code / Item | Explanation |
|---|---|
| `x = x + 1` | Does not increment the existing int object; ints cannot be changed in place. Instead, Python computes a brand-new int object with value 101 and rebinds the name x to point at it. The original 100 object, if nothing else references it, becomes eligible for garbage collection. |

## 16.5 Proving Mutability With id()

```python
lst = [1, 2, 3]
print(id(lst))
lst.append(4)
print(id(lst))  # the SAME id: the same list object was modified in place
```

| Code / Item | Explanation |
|---|---|
| `lst.append(4)` | Modifies the existing list object's internal storage directly; no new object is created, which is confirmed by id(lst) being identical before and after. |

## 16.6 Why the Distinction Matters: Assignment and Aliasing

Because assignment always binds a name to an object rather than copying it (Part 1.9), the practical consequences of mutability only become visible when two names reference the same object, an alias. For an immutable object, aliasing is harmless because neither name can ever change the shared object's value; any apparent modification just rebinds one name to a new object, leaving the other name and the original object untouched. For a mutable object, aliasing is a real hazard, because a change made through either name is visible through both, since there was only ever one object.

```python
# Immutable: safe aliasing
a = 10
b = a
b = b + 5
print(a, b)   # a is unaffected: 10 15

# Mutable: aliasing hazard
list_a = [1, 2, 3]
list_b = list_a
list_b.append(4)
print(list_a, list_b)  # BOTH show the change: [1,2,3,4] [1,2,3,4]
```

## 16.7 A Nuance: Immutable Containers Holding Mutable Objects

A tuple is immutable, but that immutability only guarantees that the tuple's own slots cannot be reassigned to different objects. If a slot holds a reference to a mutable object, such as a list, that inner object can still be mutated in place, and the tuple's immutability does nothing to prevent that.

```python
record = ("Ada", [90, 85, 92])
record[1].append(100)   # legal: mutating the list, not the tuple's slots
print(record)
try:
    record[1] = []       # illegal: reassigning a tuple slot
except TypeError as e:
    print("TypeError:", e)
```

| Code / Item | Explanation |
|---|---|
| `record[1].append(100)` | Legal, because it mutates the list object that slot 1 references; the tuple itself never changes which object slot 1 points to. |
| `record[1] = []` | Illegal, because it would require reassigning what the tuple's slot 1 references, which tuple's immutability explicitly forbids. |

## 16.8 Copying Revisited: Shallow vs Deep

Because mutable objects can be shared via aliasing, Python provides explicit copying mechanisms when an independent duplicate is genuinely needed. A shallow copy (list.copy(), dict.copy(), or slicing) duplicates only the outermost container; any mutable objects nested inside are still shared with the original. A deep copy (copy.deepcopy()) recursively duplicates every nested object, producing a structure that shares nothing at all with the original.

> **Note:** See Part 6.6 and Part 9.6 for concrete, worked shallow-vs-deep copy examples on list and dict respectively; the underlying principle here is identical for every mutable built-in type.

## 16.9 A Classic Real-World Bug: Mutable Default Arguments

One of the most notorious Python pitfalls stems directly from mutability: a mutable default argument value (such as []) is created exactly once, when the function is defined, not fresh on every call. Every call that relies on the default therefore shares and mutates the exact same object.

```python
def add_item(item, basket=[]):     # DANGEROUS: mutable default
    basket.append(item)
    return basket

print(add_item("apple"))
print(add_item("banana"))  # unexpectedly contains 'apple' too!

def add_item_fixed(item, basket=None):   # SAFE pattern
    if basket is None:
        basket = []
    basket.append(item)
    return basket

print(add_item_fixed("apple"))
print(add_item_fixed("banana"))  # correctly independent
```

| Code / Item | Explanation |
|---|---|
| `def add_item(item, basket=[]):` | The empty list [] is created a single time, at function-definition time, and reused as the actual default object on every call that omits basket. |
| `add_item("banana") showing ['apple', 'banana']` | Both calls silently mutated the same shared default list, which is almost never the intended behavior. |
| `basket=None ... if basket is None: basket = []` | The safe idiom: use an immutable sentinel (None) as the default, and construct a genuinely fresh mutable object inside the function body on every call. |

## 16.10 Complete Mutability Classification of Built-in Types

| Type | Mutable or Immutable | Notes |
|---|---|---|
| int | Immutable | Every arithmetic result is a new int object. |
| float | Immutable | Same as int. |

**Advanced:** Explain, using mutability and default-argument evaluation timing, exactly why 'def f(x=[]):' is considered dangerous, and describe the correct alternative pattern.

*Answer:* Default argument values in a Python function definition are evaluated exactly once, at the moment the def statement executes, not freshly on each call. Because [] creates a single, specific, mutable list object at that moment, every subsequent call to f that relies on the default (omits x) receives a reference to that same shared list object, not a new empty list. Any mutation performed through one call (such as an append) persists and is visible in every later call that also uses the default, which is almost never the intended behavior and can introduce very hard-to-trace bugs. The correct pattern is to use an immutable sentinel, typically None, as the actual default ('def f(x=None):'), and then explicitly construct a brand-new mutable object inside the function body only when the sentinel is detected ('if x is None: x = []'), guaranteeing a genuinely fresh list on every call that needs one.

# Part 17. Ordered vs Unordered Types

Ordering describes whether a collection type has a defined, meaningful position for each of its elements, one that you can rely on when iterating, indexing, or comparing two collections for equality.

## 17.1 What Ordering Means

An ordered collection remembers the sequence in which its elements were placed (or, for range, follows a defined arithmetic sequence) and will always yield elements in that same, predictable sequence when iterated. An unordered collection makes no such guarantee: internally, elements are organized by hash value for fast lookup, not by insertion sequence, so iteration order is an implementation detail, not a contract.

## 17.2 Ordered Collection Types

- str - character order is exactly the order written.
- list - element order is exactly the order inserted or assigned.
- tuple - same guarantee as list.
- range - follows its defined start/stop/step arithmetic sequence.
- bytes and bytearray - byte order is preserved exactly.
- dict (since Python 3.7) - preserves insertion order as a language guarantee.

## 17.3 Unordered Collection Types

- set - no guaranteed iteration order; internal layout is driven by each element's hash value.
- frozenset - same as set.

## 17.4 dict and Modern Python Ordering Behavior

It is worth emphasizing precisely what dict's ordering guarantee does and does not promise. It guarantees insertion order is preserved, meaning the first key ever inserted is iterated first, and so on, and re-assigning an existing key's value does not move that key's position. It does not sort keys in any particular alphabetical or numeric order; 'ordered' here means 'remembers insertion sequence', not 'sorted'.

```python
d = {"banana": 1, "apple": 2, "cherry": 3}
print(list(d.keys()))   # insertion order, NOT alphabetical

for s1, s2 in [({1,2,3}, {3,2,1})]:
    print(list(s1), list(s2))  # sets may print in different internal orders
```

## 17.5 Why Sets Should Never Be Treated as Ordered

Because set's internal layout is driven by hashing rather than insertion sequence, code that relies on a particular set iteration order is fragile and can behave inconsistently, even across different runs of the same program or different Python versions/builds. If a specific, reliable order is required, convert the set to a sorted list explicitly with sorted(my_set) rather than relying on however the set happens to iterate.

```python
s = {5, 1, 4, 2, 3}
print(sorted(s))   # explicit, reliable order
# print(s)         # do not rely on this order for correctness
```

Relying on it directly for a report would risk inconsistent or unpredictable output. The correct fix is to explicitly call sorted(my_set), which returns a new list containing the set's elements arranged in a well-defined, repeatable, alphabetical (or numeric) order, independent of the set's internal layout.

A sequence is any ordered collection that supports positional access by integer index. Python's six major sequence types (str, list, tuple, range, bytes, bytearray) all share a common set of behaviors defined by the sequence protocol, even though they store very different kinds of data.

## 18.1 The Sequence Family

- str - a sequence of Unicode characters (immutable).
- list - a sequence of arbitrary objects (mutable).
- tuple - a sequence of arbitrary objects (immutable).
- range - a sequence of integers following an arithmetic progression (immutable).
- bytes - a sequence of raw byte values (immutable).
- bytearray - a sequence of raw byte values (mutable).

## 18.2 Common Sequence Behavior: Indexing and Slicing

```python
examples = ["Python", [1,2,3], (1,2,3), range(5), b"abc", bytearray(b"xyz")]
for seq in examples:
    print(type(seq).__name__, seq[0], seq[-1], seq[1:])
```

The + and * operators work consistently across the mutable and immutable sequence types (with the sole, deliberate exception of range, which supports neither, since an arithmetic progression cannot generally be meaningfully concatenated or repeated).

```python
print([1, 2] + [3, 4])
print((1, 2) + (3, 4))
print("ab" + "cd")
print(b"ab" + b"cd")
print([1, 2] * 3)
print("ab" * 3)
```

| Code / Item | Explanation |
|---|---|
| `[1, 2] + [3, 4]` | Concatenates two lists into a new list, [1, 2, 3, 4]. |

## 18.5 Mutable Sequences vs Immutable Sequences

Within the sequence family, list and bytearray are mutable (support item assignment, insertion, deletion), while str, tuple, range and bytes are immutable (support only reading, never in-place modification). This is the same mutable/immutable distinction from Part 16, applied specifically to the ordered sequence category.

## 18.6 Sequence vs Non-Sequence Collections

dict, set and frozenset are collections, but they are not sequences, because they do not support meaningful integer-index positional access (d[0] on a dict looks up the key 0, not 'the first item'; sets cannot be indexed at all). This distinction matters when choosing which container fits a given problem: reach for a sequence type when position/order matters, and a set or dict when uniqueness or key-based lookup matters more than position.

Because dict's indexing is semantically key-based rather than position-based, and it does not support slicing, it is classified as a mapping type, not a sequence type, despite being ordered and iterable.

# Part 19. Hashability

Hashability determines whether an object can be used as a dictionary key or stored inside a set. It is a deceptively deep topic that ties directly back to mutability, and understanding it explains a whole category of TypeErrors that otherwise seem arbitrary.

## 19.1 What Hashability Means

An object is hashable if it has a hash value that never changes during its entire lifetime (implemented via a __hash__ method) and can be meaningfully compared for equality against other objects (via __eq__). dict and set both rely on hashing internally: instead of scanning every existing key to check for a match, they compute a key's hash value and jump almost directly to the right internal bucket, which is what makes lookups so fast.

## 19.2 Why Dictionary Keys and Set Elements Must Be Hashable

If an object's hash value could change after it was inserted as a key (or added to a set), the dict or set would no longer be able to find it: it would still be sitting in the bucket corresponding to its old hash, but any future lookup would compute the new hash and look in the wrong bucket entirely, silently breaking the data structure. To prevent this entire class of corruption, Python requires hashability, and in practice restricts it to objects whose value cannot change after creation, i.e. immutable objects.

## 19.3 Immutable vs Hashable: A Careful Distinction

Immutability and hashability are closely related but not strictly identical. Every hashable built-in type happens to also be immutable, but immutability alone does not automatically guarantee hashability: a tuple is immutable, yet a tuple containing an unhashable element (like a list) is itself unhashable, because Python cannot compute a stable hash for something that contains a component whose value could still change.

```python
safe_tuple = (1, 2, 3)
print(hash(safe_tuple))

unsafe_tuple = (1, [2, 3])
try:
    hash(unsafe_tuple)
except TypeError as e:
    print("TypeError:", e)
```

| Code / Item | Explanation |
|---|---|
| `hash(safe_tuple)` | Succeeds, because every element inside the tuple is itself immutable and hashable. |
| `hash(unsafe_tuple)` | Fails with TypeError: unhashable type: 'list', because the tuple contains a mutable list, even though the tuple itself is immutable. |

## 19.4 Examples of Hashable and Unhashable Objects

| Code / Item | Explanation |
|---|---|
| `int, float, complex, bool` | Hashable |
| `str` | Hashable |
| `tuple (with all-hashable elements)` | Hashable |
| `frozenset` | Hashable |
| `bytes` | Hashable |
| `NoneType (None)` | Hashable |
| `list` | Unhashable |
| `dict` | Unhashable |
| `set` | Unhashable |
| `bytearray` | Unhashable |

## 19.5 Using hash() Directly

```python
print(hash("python"))
print(hash(42))
print(hash((1, 2, 3)))
print(hash("python") == hash("python"))
```

> **Note:** Two calls to hash() on equal, immutable values within the same program run always produce the same hash value, which is exactly the property dict and set rely on to reliably relocate a previously inserted key or element.

## 19.6 Practical Implications

- You cannot use a list, dict, or set as a dictionary key, or as an element inside another set; convert to tuple, frozenset, or a similarly hashable equivalent first, if the underlying data allows it.
- A common workaround for making a mutable-looking record hashable is to represent it as a tuple of its fields, or to use collections.namedtuple, instead of a dict or list.
- Custom classes are hashable by default (using the object's identity), unless you define __eq__ without also defining a compatible __hash__, in which case Python automatically makes the class unhashable to prevent inconsistent behavior.

## Part 19 Practice Exercises

**Beginner:** Is a Python str hashable?

*Answer:* Yes; str is immutable, and Python's built-in immutable types are all hashable.

**Intermediate:** Why is a tuple containing a list, such as (1, [2, 3]), unhashable even though the tuple itself is an immutable type?

*Answer:* Hashability requires that an object's hash value never change for its entire lifetime, which in turn requires every component contributing to that hash to also be stable. Although the tuple's own slots cannot be reassigned, one of those slots references a list, a mutable object whose contents (and therefore its notional contribution to a hash) could change at any time. Because Python cannot guarantee a stable, consistent hash under those conditions, it refuses to hash the tuple at all, raising TypeError.

**Advanced:** Explain, step by step, why dict lookups are described as 'near-constant time' specifically because of hashing, contrasting this with how a plain list would have to perform an equivalent lookup.

*Answer:* When you write d[key], Python first computes hash(key), a fixed-size integer, and uses that hash value to compute directly which internal 'bucket' (a small region of the dictionary's underlying array) the key-value pair should live in, essentially jumping straight to approximately the right location without inspecting any other keys. It then does a quick equality check only against the (typically very small number of) entries that landed in that same bucket, to handle rare hash collisions. This means the average lookup cost stays roughly constant regardless of how many total keys the dictionary holds. A plain list has no such structure: checking whether a given item exists, or finding an item that matches some target, requires a list to compare the target against list elements one by one from the start until a match is found or the list is exhausted, an operation whose cost grows linearly with the list's length. # Part 21. Data Type Compatibility

Not all types can interact with each other directly. This part explains which combinations Python allows automatically, which require an explicit conversion, and which are simply incompatible and will always raise a TypeError.

Python automatically promotes across the numeric tower (int -> float -> complex) when they interact in an expression, always widening to the more general type rather than losing information, and never requiring an explicit conversion call from you.

```python
print(3 + 4.5)        # int + float -> float
print(3 + 4.5j)       # int + complex -> complex
print(True + 1)       # bool + int -> int
print(2.0 + 3j)       # float + complex -> complex
```

| Code / Item | Explanation |
|---|---|

> **Note:** str and int have no automatic promotion rule between them, unlike the purely numeric types; Python always requires you to state explicitly which interpretation (text concatenation, or numeric addition) you intend.

## 21.3 String and Numeric Interactions

## 21.5 Comparing Across Types

```python
print(3 == 3.0)          # True: numeric types compare by value
print(3 == "3")          # False: different categories entirely, never equal
print([1, 2] == (1, 2))  # False: list and tuple are never equal, even with identical contents
```

# Part 22. Object Identity and Data Types

Every object in Python has an identity, a unique marker (in CPython, effectively its memory address) that distinguishes it from every other object for its entire lifetime. Understanding identity, and how it differs from equality, resolves a great deal of confusion around aliasing, mutability, and the is operator.

## 22.1 The id() Function

id(obj) returns an integer that uniquely identifies an object for as long as it exists. In the standard CPython implementation, this integer corresponds to the object's memory address, though that detail is an implementation choice, not a language guarantee.

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(id(a))
print(id(b))
print(id(a) == id(b))
```

## 22.2 Identity (is) vs Equality (==)

is checks identity: do these two names refer to the literal same object in memory? == checks equality: do these two objects have the same value, according to their type's definition of equality, regardless of whether they are the same object? Two distinct objects can absolutely be equal in value while having different identities.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b)   # True: same VALUE
print(a is b)    # False: different OBJECTS
print(a is c)    # True: c is literally the same object as a
```

| Code / Item | Explanation |
|---|---|
| `a == b` | True, because the two lists happen to contain equal elements in the same order. |
| `a is b` | False, because a and b were built from two separate list literals, creating two distinct objects that merely happen to hold equal values. |
| `a is c` | True, because c = a was a direct assignment, an alias, giving c a reference to the exact same object a already referenced. |

## 22.3 Small Integer and String Caching (CPython Detail)

As a memory optimization specific to the CPython implementation, small integers (typically -5 through 256) and certain simple string literals are cached and reused automatically, so is can appear to return True even for separately written literals of these specific, small values. This is an implementation detail, not something to rely on in real code; always use == to check value equality, and reserve is specifically for identity checks such as 'is None'.

```python
x = 100
y = 100
print(x is y)   # commonly True, due to small-int caching

p = 10_000_000
q = 10_000_000
print(p is q)   # commonly False, outside the small-int cache range
```

> **Note:** Never rely on is for general value comparison, even though it happens to 'work' for small integers; this behavior is an implementation detail of CPython and is not guaranteed by the Python language specification.

## 22.4 Aliasing Revisited Through Identity

```python
original = {"score": 90}
alias = original
independent = {"score": 90}

print(original is alias)        # True: same object
print(original is independent)  # False: equal value, different object

alias["score"] = 100
print(original["score"])        # 100: changed, because alias IS original
```

## 22.5 Why Two Variables Can Refer to the Same Object

This is simply a direct consequence of how assignment works in Python (Part 1.9): a variable is a name bound to an object, never a container holding a private copy of that object's bytes. Any assignment of the form b = a, where a already references an existing object, makes b a second name for that identical object; nothing is duplicated unless you explicitly call a copying function.

**Intermediate:** Why is 'if x is None:' recommended, while 'if x is []:' would be a bug even when x actually holds an empty list?

*Answer:* None is a true singleton in Python, guaranteed to be the exact same single object everywhere, so 'is None' reliably checks identity against that one object. Empty list literals, however, are not cached or unified the way None is: every time [] appears in source code, a brand-new, distinct empty list object is created. So 'x is []' checks whether x is identical to this particular newly created empty list object right here, which will almost always be False even if x holds a perfectly valid, equal, but separately created empty list; the correct check for 'is x an empty list' is 'x == []' or, more idiomatically, 'not x' or 'len(x) == 0'.

**Advanced:** Explain why relying on 'x is 200' to check whether an integer variable equals 200 is considered bad practice, even if it happens to return True when tested.

*Answer:* CPython, as a specific implementation optimization, pre-creates and caches small integer objects (conventionally -5 through 256) and reuses the same cached object whenever that value is needed, which is why 'is' comparisons for small integers like 200 often appear to work. This caching behavior is not part of the Python language specification; it is an implementation detail that could differ between Python implementations (such as PyPy or Jython) or even change in future CPython versions, and it explicitly does not extend to larger integers, which are typically created as fresh, distinct objects each time. Relying on 'is' for integer value comparison is therefore fragile and non-portable; == is the semantically correct operator for comparing values, and is should be reserved for genuine identity checks such as comparing against the None singleton.

# Part 25. Advanced Python Type Concepts

Having covered every built-in type in depth, this part zooms out to Python's broader object model: how classes themselves are objects, how user-defined types relate to built-in ones, and how Python's flexible approach to typing (duck typing) differs from the strict, built-in-type rules covered so far.

## 25.1 The Object Model: Classes Are Objects Too

In Python, everything is an object, including classes themselves. A class is an instance of a special built-in type called type, meaning types have types: int is an object whose own type is type.

```python
print(type(42))          # <class 'int'>
print(type(int))         # <class 'type'>
print(type(type))        # <class 'type'>: type is its own type
print(isinstance(int, type))
```

## 25.2 User-Defined Types (Custom Classes)

A class statement defines a brand-new type, exactly as fundamental to Python's type system as int or str, just written in Python instead of built into the interpreter. Once defined, that class can be instantiated to create objects with their own identity, type, and value, following the exact same rules covered throughout this handbook.

```python
class Student:
    def __init__(self, name, gpa):
        self.name = name
        self.gpa = gpa

s = Student("Ada", 3.8)
print(type(s))
print(isinstance(s, Student))
print(s.name, s.gpa)
```

## 25.3 Type Inheritance

A custom class can inherit from another class (including a built-in type), acquiring its behavior while adding or overriding parts of it, exactly like bool inherits from int internally.

```python
class PremiumStudent(Student):
    def __init__(self, name, gpa, discount):
        super().__init__(name, gpa)
        self.discount = discount

ps = PremiumStudent("Bo", 3.5, 0.1)
print(isinstance(ps, Student))       # True: inherited relationship
print(isinstance(ps, PremiumStudent))
```

## 25.4 Duck Typing and Structural Behavior

Duck typing is Python's informal philosophy of caring about what an object can do, not what type it formally is: 'if it walks like a duck and quacks like a duck, treat it as a duck'. Rather than requiring an object to be an instance of a specific class, code often simply calls a method or uses an operator, and lets Python raise an error naturally if the object cannot actually support that behavior.

```python
class FileLikeString:
    def __init__(self, text):
        self.text = text
    def read(self):
        return self.text

def print_contents(source):
    print(source.read())   # works on ANY object with a .read() method

print_contents(FileLikeString("hello from a fake file"))
```

> **Note:** print_contents() never checks isinstance(source, SomeFileClass); it simply trusts that whatever is passed in supports .read(), which is the essence of duck typing, and is a very common pattern throughout Python's standard library and ecosystem.

## 25.5 Protocols (Informal Interfaces)

A protocol is an informal contract describing a set of methods an object must implement to participate in some piece of Python behavior, without requiring formal inheritance from any particular base class. The sequence protocol (Part 18), the iterator protocol (__iter__/__next__), and the buffer protocol (Part 14) are all examples already covered in this handbook.

## 25.6 Generic Collections and Type Hints (Introductory)

Type hints are optional annotations that document a variable's or function's expected type without being enforced by the interpreter at runtime; Python remains dynamically typed even in fully type-hinted code. Generic collection hints (list[int], dict[str, float]) describe not just the container type but the type of elements it is expected to hold, which is useful for tooling (editors, static type checkers like mypy) even though Python itself will not stop you from violating the hint at runtime.

```python
def average_scores(scores: list[float]) -> float:
    return sum(scores) / len(scores)

print(average_scores([88.0, 92.5, 79.0]))
# Python will NOT raise an error even if you break the hint:
print(average_scores([88, 92, 79]))   # ints instead of floats: still runs fine
```

> **Note:** Type hints are read by external tools, not enforced by the Python interpreter itself; passing a list[int] where list[float] is hinted still runs without error, since Python's core execution remains fully dynamically typed regardless of any annotations present.

## 25.7 Distinguishing Built-in Types From User-Defined Types

| Code / Item | Explanation |
|---|---|
| `Implementation` | Built-in types are implemented in C inside the interpreter; user-defined types are implemented in Python via a class statement. |
| `Performance` | Built-in types are typically faster for their core operations, being lower-level. |
| `Extensibility` | User-defined types can add arbitrary custom methods, attributes and behavior; built-in types cannot be modified, only subclassed. |
| `Instantiation` | Both are created by calling the type as if it were a function: int("5"), Student("Ada", 3.8 |
| `Underlying Model` | Both are ultimately instances of type, and both produce objects with identity, type and value, following the exact same object model. |

# Part 26. Common Data Type Errors

Almost every runtime error a Python beginner encounters traces directly back to a data-type misunderstanding. This part walks through the six most common type-related exceptions, why each occurs, incorrect versus corrected code, and how to prevent each one going forward.

## 26.1 TypeError

Raised when an operation or function is applied to an object of an inappropriate type: mixing incompatible types in an operator, calling a function with the wrong argument type, or attempting to mutate an immutable object.

```python
# Incorrect:
try:
    total = "5" + 3
except TypeError as e:
    print("TypeError:", e)

# Corrected:
total = int("5") + 3
print(total)
```

> **Note:** Prevention: validate or convert types explicitly at the boundaries of your code (user input, file/API data) before combining them in operations.

## 26.2 ValueError

Raised when an operation receives an argument of the correct type but an inappropriate or unparseable value.

```python
# Incorrect:
try:
    age = int("twenty")
except ValueError as e:
    print("ValueError:", e)

# Corrected:
raw = "twenty"
if raw.isdigit():
    age = int(raw)
else:
    age = None
print(age)
```

> **Note:** Prevention: validate that a string actually represents the expected kind of value (isdigit(), a try/except block, or a regular expression) before attempting a conversion.

## 26.3 IndexError

Raised when a sequence (list, tuple, str, etc.) is accessed with an index that is out of the valid range for its current length.

```python
items = ["a", "b", "c"]
# Incorrect:
try:
    print(items[5])
except IndexError as e:
    print("IndexError:", e)

# Corrected:
index = 5
if index < len(items):
    print(items[index])
else:
    print("Index out of range")
```

> **Note:** Prevention: check an index against len(sequence) first, or use a safe pattern such as slicing (which never raises IndexError, it simply returns an empty result for an out-of-range slice).

## 26.4 KeyError

Raised when a dictionary is accessed with a key that does not exist in it.

```python
student = {"name": "Ada"}
# Incorrect:
try:
    print(student["gpa"])
except KeyError as e:
    print("KeyError:", e)

# Corrected:
print(student.get("gpa", "Not recorded"))
```

> **Note:** Prevention: use dict.get(key, default) for optional keys, or explicitly check 'if key in my_dict:' before direct [] access when a missing key is a genuinely exceptional condition worth distinguishing.

## 26.5 AttributeError

Raised when code attempts to access a method or attribute that does not exist on a given object's type, very often because the object turned out to be a different type than expected, frequently None.

```python
def find_student(name, students):
    for s in students:
        if s["name"] == name:
            return s
    return None

students = [{"name": "Ada", "gpa": 3.8}]

# Incorrect:
result = find_student("Bo", students)
try:
    print(result["gpa"])
except TypeError as e:
    print("TypeError:", e)

# Corrected:
if result is not None:
    print(result["gpa"])
else:
    print("Student not found")
```

> **Note:** Prevention: after any lookup that might legitimately fail and return None (Part 15), always check 'is None' before using the result as though it were the expected type. This example technically raises TypeError since dicts don't support subscripting None, but the identical root cause (assuming a value's type without checking) is the single most common source of AttributeError as well, when code calls a method on what turns out to be a None result.

## 26.6 NameError

Raised when code references a variable name that has not been defined anywhere in the currently accessible scope, often due to a typo, or using a variable before it has actually been assigned.

```python
# Incorrect:
try:
    print(total_score)
except NameError as e:
    print("NameError:", e)

# Corrected:
total_score = 0
print(total_score)
```

> **Note:** Prevention: always assign a variable an initial value before referencing it, and double-check spelling; NameError is unrelated to data types directly, but it very frequently appears alongside type-related bugs when a variable's expected initialization was skipped due to a logic error.

## 26.7 Quick-Reference Error Summary

| Exception | Typical Trigger | Common Fix |
|---|---|---|
| TypeError | Mixing incompatible types, or mutating an immutable object. | Convert types explicitly before combining them. |
| ValueError | Correct type, but an unparseable or invalid value. | Validate before converting; use try/except. |
| IndexError | Sequence index outside the valid range. | Check against len() first, or use safe slicing. |
| KeyError | Dictionary accessed with a missing key. | Use .get(key, default) or 'in' to check first. |
| AttributeError | Method/attribute called on an object of the wrong type (often None). | Check 'is None' or isinstance() before use. |
| NameError | Referencing an undefined variable name. | Ensure the variable is assigned before use; check spelling. |

*Answer:* tuple, provided all of its own elements are themselves hashable; list can never be used as a dictionary key.

**Intermediate:** Using the Bytes vs Bytearray table, explain why bytes is hashable but bytearray is not.

*Answer:* bytes is immutable, and Python's built-in immutable types are hashable by default, since their value (and therefore their hash) cannot change after creation. bytearray is mutable, and mutable built-in types are deliberately made unhashable, since allowing them to be hashed would risk a stored hash value becoming inconsistent with the object's current contents after a later in-place edit.

**Advanced:** A colleague argues that since both list and set can hold multiple items and both support 'x in collection' membership checks, they are functionally interchangeable for most purposes. Using the comparison tables, construct a rebuttal covering at least three genuine differences.

*Answer:* First, ordering: list preserves a specific, meaningful position for every element and supports indexing/slicing, while set has no defined order and cannot be indexed at all, so any code relying on 'the third element' or on a stable iteration order cannot safely use a set. Second, duplicates: list allows and preserves duplicate values, which matters whenever repeated occurrences carry meaning (such as a list of exam scores where the same score can legitimately appear more than once), while set silently collapses duplicates into a single entry, discarding that information entirely. Third, performance profile: while both support 'in', a list's membership test is O(n), scanning elements one by one, whereas a set's is O(1) on average via hashing, so for a workload dominated by repeated membership checks against a large collection, set is dramatically faster, but for a workload that depends on preserved order or duplicate counts, only list is correct at all, regardless of speed.

**Beginner:** Create a dictionary representing a single product with keys "name" (str), "price" (float), "in_stock" (bool), and "tags" (list of str). Then print whether the product is in stock using its boolean value directly in an if statement.

*Answer:* product = {"name": "Notebook", "price": 500.0, "in_stock": True, "tags": ["stationery", "office"]}

This combines str, float, bool and list as values within a single dict, and relies on bool's direct usability as a condition (Part 4), exactly as it would appear in real inventory code.

**Intermediate:** Given a list of tuples representing (student_name, list_of_scores), write code that builds a new dictionary mapping each student's name to their average score, correctly handling any student with an empty scores list by mapping them to None instead of raising a ZeroDivisionError.

*Answer:* records = [("Ada", [88, 92, 79]), ("Bo", []), ("Chi", [95, 91])]
averages = {}
for name, scores in records:
    averages[name] = sum(scores) / len(scores) if scores else None
print(averages)
This combines tuple unpacking in a for loop (Part 7), list aggregation with sum()/len(), a dict being built incrementally (Part 9), and the same None-for-missing-data pattern from Part 15 and Part 24, all working together, exactly as a real preprocessing step would.

**Advanced:** You receive a raw dataset as a list of dictionaries, where some records have duplicate combinations of (city, category) that should be merged, and each record has a 'sales' float. Using a dictionary keyed by a tuple, and a set to track which unique (city, category) combinations were seen, write code that sums sales per (city, category) pair and prints the number of unique combinations found.

*Answer:* raw_records = [
    {"city": "Lagos", "category": "Electronics", "sales": 1500.0},
    {"city": "Lagos", "category": "Electronics", "sales": 800.0},
    {"city": "Abuja", "category": "Fashion", "sales": 640.0},
]
totals = {}
seen_combinations = set()
for record in raw_records:
    key = (record["city"], record["category"])   # tuple used as a dict key, Part 7.6
    seen_combinations.add(key)                       # set tracks uniqueness, Part 10
    totals[key] = totals.get(key, 0.0) + record["sales"]
print(totals)
print("Unique combinations:", len(seen_combinations))
This exercise deliberately uses a tuple as a dict key specifically because it is hashable and immutable (Part 7.6, Part 19), tracks uniqueness with a set for fast, automatic deduplication (Part 10), and relies on dict.get() with a default to safely accumulate a running total without a KeyError on the first occurrence of any given key (Part 9.2, Part 26.4), combining five separate parts of this handbook into one realistic aggregation task.

Every one of Python's built-in data types, from the humble int to the specialized memoryview, is built on the same small set of foundational ideas covered throughout this handbook: every value is an object, every object has an identity, a type, and a value; variables are references, not containers; mutability determines whether an object can be changed in place or must be replaced; and hashability, ordering, and sequence behavior all follow directly from those first principles rather than being arbitrary, memorized rules.
