# Part 1. Introduction to Web Development

This part establishes the vocabulary and mental model every later part builds on: what happens when a browser or a Python script talks to a server, and the core concepts (HTTP, URLs, status codes, REST, JSON) that every framework in this handbook is built around.

## 1.1 What Web Development Is

Web development is building software that communicates over HTTP, typically between a client (a browser, a mobile app, another program) and a server. It splits into frontend (what runs in the browser: HTML, CSS, JavaScript, rendering UI) and backend (what runs on a server: business logic, databases, APIs). This handbook is entirely about the backend: building the Python programs that receive requests and return responses.

## 1.2 Frontend vs Backend, Client vs Server

| Code / Item | Explanation |
|---|---|
| `Frontend` | Runs in the user's browser/device; renders UI; sends requests to a backend. |
| `Backend` | Runs on a server; processes requests, applies business logic, talks to databases; returns responses. |
| `Client` | Whatever initiates a request: a browser, a mobile app, another server, a Python script using requests. |
| `Server` | A program that listens for requests on a network port and sends back responses. |

## 1.3 HTTP: HyperText Transfer Protocol

HTTP is the text-based protocol that clients and servers speak to exchange requests and responses. It is stateless by default (each request is independent, with no memory of previous requests, unless state is explicitly added via cookies/sessions/tokens), and it is a request-response protocol: the client always initiates, the server always responds.

## 1.4 The Request/Response Cycle

```
Client                          Server
  |  1. sends HTTP request         |
  | ------------------------------>|
  |                                 |  2. server processes it
  |                                 |     (routes, business logic, DB)
  |  3. sends HTTP response        |
  | <------------------------------|
  |  4. client uses the response   |
```

## 1.5 URLs, Domains and Ports

````python
url = "https://api.example.com:443/v1/users?active=true"
# scheme:   https
# domain:   api.example.com
# port:     443 (implicit default for https; 80 is the default for http)
# path:     /v1/users
# query:    active=true
print("Parsed conceptually above; urllib.parse.urlparse can do this programmatically:")
from urllib.parse import urlparse
parsed = urlparse(url)
print(parsed.scheme, parsed.netloc, parsed.path, parsed.query)
````

| Code / Item | Explanation |
|---|---|
| `scheme (http/https)` | Which protocol to use; https adds TLS encryption on top of HTTP. |
| `domain` | The human-readable server address, resolved to an IP address via DNS. |
| `port` | Which 'door' on the server to connect to; 80 for HTTP, 443 for HTTPS by default. |
| `path` | Identifies a specific resource on the server. |
| `query string` | Optional key=value pairs after ?, used for filtering, pagination, options. |

## 1.6 HTTP Methods (Preview; Full Detail in Part 2)

- GET - retrieve a resource, no side effects (safe, idempotent).
- POST - create a resource or trigger an action (not idempotent).
- PUT - replace a resource entirely (idempotent).
- PATCH - partially update a resource.
- DELETE - remove a resource (idempotent).

## 1.7 Status Codes (Preview)

| Code / Item | Explanation |
|---|---|
| `2xx Success` | 200 OK, 201 Created, 204 No Content. |
| `3xx Redirection` | 301 Moved Permanently, 304 Not Modified. |
| `4xx Client Error` | 400 Bad Request, 401 Unauthorized, 403 Forbidden, 404 Not Found, 422 Unprocessable Entity. |
| `5xx Server Error` | 500 Internal Server Error, 503 Service Unavailable. |

## 1.8 Headers, Cookies and Sessions

Headers are metadata attached to a request or response (content type, authentication tokens, caching rules). Cookies are small pieces of data the server asks the browser to store and resend on future requests, most commonly used to maintain a session, a way of remembering a specific user across multiple otherwise-stateless HTTP requests, typically by storing a session ID in a cookie that maps to server-side state.

## 1.9 REST APIs and JSON

A REST (Representational State Transfer) API exposes resources (users, orders, predictions) at URLs, manipulated using standard HTTP methods, with a consistent, predictable structure. JSON (JavaScript Object Notation) is the near-universal data format REST APIs use for request and response bodies, because it is lightweight, human-readable, and maps directly onto Python dicts and lists.

````python
import json
data = {"id": 1, "name": "Ada", "active": True, "scores": [90, 85]}
json_text = json.dumps(data)
print(json_text)
parsed_back = json.loads(json_text)
print(parsed_back == data)
````

## Part 1 Practice Exercises

**Beginner:** In the request/response cycle, who always initiates: the client or the server?

*Answer:* The client always initiates a request; the server responds. HTTP is fundamentally a request-response protocol, never the other way around.

**Intermediate:** Why is HTTP described as 'stateless by default', and what mechanism is commonly used to add state across requests?

*Answer:* Stateless means each HTTP request is handled independently, with the server retaining no memory of previous requests from the same client by default. Cookies (often carrying a session ID) are the common mechanism used to add state, letting the server recognize that several separate requests belong to the same ongoing user session.

**Advanced:** Explain why JSON, rather than a Python-specific format like pickle, is the standard format for REST API bodies, even when both the client and server are written in Python.

*Answer:* JSON is a language-agnostic, text-based format understood natively by virtually every programming language and platform (browsers, mobile apps, other backend languages), which is essential since a REST API's clients are rarely guaranteed to all be Python; pickle, by contrast, is Python-specific, version-sensitive, and can execute arbitrary code during deserialization, making it both incompatible with non-Python clients and a serious security risk if ever used to parse data from an untrusted source. Even in an all-Python system, JSON's simplicity, human-readability for debugging, and universal tooling support make it the practical standard.

## Part 2 Practice Exercises

**Beginner:** Which HTTP method is conventionally used to partially update a resource, versus replacing it entirely?

*Answer:* PATCH partially updates a resource (only the fields provided change); PUT replaces the entire resource.

**Intermediate:** What does it mean for an HTTP method to be 'idempotent', and which of GET, POST, DELETE are idempotent?

*Answer:* Idempotent means making the same request multiple times has the same effect as making it once; repeating it does not cause additional changes. GET and DELETE are idempotent (repeating a DELETE on an already-deleted resource still results in it being gone); POST is generally NOT idempotent, since repeating it typically creates multiple new resources.

**Advanced:** A client sends a POST request to create a resource, but the request body is syntactically valid JSON that is missing a required field. Which status code should the server return, 400 or 422, and why does this distinction matter for API design?

*Answer:* 422 Unprocessable Entity is the more precise, semantically correct choice here: the request body IS valid JSON (syntactically well-formed), but it fails business/schema validation (a required field is missing), which is exactly the distinction 422 exists to capture, as opposed to 400 Bad Request, which more properly indicates the request itself is malformed at a structural level (e.g., invalid JSON syntax, wrong Content-Type). This distinction matters because API clients often branch their error-handling logic on status code: a well-designed API using 422 specifically for validation failures (as FastAPI does by default) lets clients reliably distinguish 'you sent garbage' (400) from 'you sent well-formed but invalid data' (422), enabling more precise, automated error handling.

# Part 3. The Python Web Ecosystem

This part orients you across Python's major web tools before diving into any one of them: Flask, FastAPI, Django for building APIs/servers, and requests/httpx for consuming them.

## 3.1 Flask: The Minimal, Flexible Microframework

Flask provides routing, request/response handling, and templating, with almost everything else (database ORM, validation, auth) left to you or third-party extensions. It is simple to learn, highly flexible, and widely used for small-to-medium APIs, prototypes, and internal tools.

## 3.2 FastAPI: Modern, Async-First, Type-Hint-Driven

FastAPI builds on Python type hints and Pydantic to give you automatic request validation, automatic interactive documentation (Swagger UI/OpenAPI), native async support, and excellent performance, with far less boilerplate than manually validating and documenting an API by hand. This is why FastAPI is prioritized throughout this handbook, especially for ML/AI APIs (Parts 5, 17, 18).

## 3.3 Django: The Full-Featured, Batteries-Included Framework

Django bundles an ORM, admin interface, authentication, templating, and much more into one opinionated, cohesive framework, historically for full websites; Django REST Framework extends it specifically for building APIs. Django suits larger, long-lived applications that benefit from a mature, all-in-one ecosystem and a built-in admin panel.

## 3.4 requests and httpx: Consuming APIs From Python

requests is the long-standing, synchronous standard for making HTTP calls FROM Python (as a client), covered in depth in Part 8. httpx is a modern alternative with a near-identical synchronous API, plus native async support (httpx.AsyncClient), covered in Part 9, making it the natural choice when your own code is already async (e.g., inside a FastAPI endpoint calling another API).

## 3.5 When Each Is Appropriate

- Choose Flask for: small APIs/prototypes, projects needing maximum flexibility over structure, teams already familiar with it, or simple internal tools where FastAPI's extra features are not needed.
- Choose FastAPI for: ML/AI prediction APIs, modern async-first backends, any project that benefits from automatic validation and interactive docs, and as the default recommendation for new professional API projects in this handbook.
- Choose Django (+ DRF) for: large, long-lived applications, projects needing a built-in admin interface and ORM out of the box, and teams that prefer an opinionated, all-in-one structure.
- Use requests for straightforward, synchronous API consumption; use httpx (async) when consuming APIs from within an already-async application, or when you need to make many concurrent external calls efficiently (Part 9).

## 3.6 Framework Comparison

| Feature | Flask | FastAPI | Django |
|---|---|---|---|
| Validation | Manual, or via extensions | Automatic, via Pydantic | Via Django Forms/DRF Serializers |
| Async support | Limited/manual | Native, first-class | Growing, but not the default paradigm |
| Auto-generated docs | No (manual/extension) | Yes, built in (Swagger/OpenAPI) | No (manual/extension, e.g. DRF) |
| ORM included | No (use SQLAlchemy separately) | No (use SQLAlchemy separately) | Yes, Django ORM built in |
| Admin panel | No | No | Yes, built in |
| Learning curve | Low | Low-Medium | Medium-High |
| Best fit here | Small APIs, prototypes | ML/AI APIs, modern backends (primary focus) | Large full-featured apps |

## Part 3 Practice Exercises

**Beginner:** Which of Flask, FastAPI, Django comes with a built-in admin interface and ORM out of the box?

*Answer:* Django.

**Intermediate:** Why does this handbook prioritize FastAPI specifically for ML/AI prediction APIs?

*Answer:* FastAPI's use of Python type hints and Pydantic gives automatic request validation (catching malformed feature inputs before they ever reach a model), native async support (useful for I/O-bound work like calling external model services or databases without blocking), and automatic interactive documentation, all with less boilerplate than manually implementing the same in Flask; these properties directly match the needs of a production ML/AI API, which typically needs strict input validation, good performance under concurrent load, and clear documentation for other engineers or systems consuming it.

**Advanced:** A team is building a small internal script that calls three different public APIs sequentially and does simple data transformation, no server of its own is needed. Which of Flask, FastAPI, Django, requests, httpx is actually relevant here, and why are the other three irrelevant?

*Answer:* Only requests (or httpx used synchronously) is relevant; Flask, FastAPI and Django are all tools for BUILDING a server that RECEIVES requests, which this scenario does not need at all, since the script itself is acting purely as a CLIENT, consuming existing APIs rather than exposing one. requests is the straightforward, synchronous choice for calling a small, fixed number of APIs sequentially with no particular need for concurrency; httpx would only become the more relevant choice if the three calls needed to happen concurrently (Part 9) or if the surrounding code were already async for other reasons, neither of which this scenario describes.

## 4.2 Creating a Minimal Application

````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return "Hello, World!"

# To actually run this as a server (not executed in this notebook):
# if __name__ == "__main__":
#     app.run(debug=True)
print("Flask app object created:", app)
````

## 4.3 Routes and HTTP Methods

````python
from flask import Flask, request

app2 = Flask(__name__)

@app2.route("/items", methods=["GET", "POST"])
def items():
    if request.method == "GET":
        return {"items": ["apple", "banana"]}
    elif request.method == "POST":
        return {"message": "Item created"}, 201

print("Route registered for GET and POST on /items")
````

| Code / Item | Explanation |
|---|---|
| `methods=["GET", "POST"]` | By default a route only accepts GET; listing multiple methods lets one view function handle several, branching internally on request.method. |
| `return {"message": "Item created"}, 201` | Returning a (body, status_code) tuple sets a non-default HTTP status; Flask auto-converts a returned dict to a JSON response. |

## 4.4 URL Parameters (Path Parameters)

````python
from flask import Flask

app3 = Flask(__name__)

@app3.route("/users/<int:user_id>")
def get_user(user_id):
    return {"id": user_id, "name": f"User {user_id}"}

with app3.test_client() as client:
    response = client.get("/users/42")
    print(response.status_code, response.get_json())
````

| Code / Item | Explanation |
|---|---|
| `<int:user_id>` | A URL converter: matches only integer path segments, and passes the parsed int directly as the user_id argument to get_user(). |
| `app3.test_client()` | Flask's built-in test client, used here to simulate an HTTP request without actually running a server, for demonstration inside this notebook. |

## 4.5 Query Parameters

````python
from flask import Flask, request

app4 = Flask(__name__)

@app4.route("/search")
def search():
    query = request.args.get("q", "")
    limit = int(request.args.get("limit", 10))
    return {"query": query, "limit": limit}

with app4.test_client() as client:
    response = client.get("/search?q=python&limit=5")
    print(response.get_json())
````

| Code / Item | Explanation |
|---|---|
| `request.args.get("q", "")` | Reads the query parameter q from the URL, with "" as a default if it is missing entirely. |

## 4.6 The Request Object and JSON Bodies

````python
from flask import Flask, request, jsonify

app5 = Flask(__name__)

@app5.route("/users", methods=["POST"])
def create_user():
    data = request.get_json()
    if not data or "name" not in data:
        return jsonify({"error": "name is required"}), 400
    return jsonify({"id": 1, "name": data["name"]}), 201

with app5.test_client() as client:
    ok = client.post("/users", json={"name": "Ada"})
    print(ok.status_code, ok.get_json())
    bad = client.post("/users", json={})
    print(bad.status_code, bad.get_json())
````

| Code / Item | Explanation |
|---|---|
| `request.get_json()` | Parses the incoming request body as JSON into a Python dict. |
| `jsonify({...}), 400` | jsonify() builds a proper JSON Response object with correct Content-Type headers; pairing it with 400 sets the status code explicitly. |

## 4.7 Error Handlers

````python
from flask import Flask, jsonify

app6 = Flask(__name__)

@app6.errorhandler(404)
def not_found(error):
    return jsonify({"error": "Resource not found"}), 404

with app6.test_client() as client:
    response = client.get("/does-not-exist")
    print(response.status_code, response.get_json())
````

| Code / Item | Explanation |
|---|---|
| `@app6.errorhandler(404)` | Registers a custom handler that runs whenever Flask would otherwise return its default 404 page, letting you return a consistent JSON error format instead. |

## 4.8 Templates (Brief)

## Part 4 Practice Exercises

**Beginner:** What does @app.route("/") do?

*Answer:* It registers the function defined immediately below it as the view function that handles GET requests to the URL path '/', so visiting that path triggers that function and returns its result as the HTTP response.

**Intermediate:** In Flask, what is the difference between request.args and request.get_json()?

*Answer:* request.args provides access to the URL's query string parameters (the key=value pairs after '?'), typically used with GET requests. request.get_json() parses the request BODY as JSON into a Python dict, typically used with POST/PUT/PATCH requests that send structured data.

**Advanced:** Write a Flask route /products/<int:product_id> that returns a 404 JSON error if product_id is not in a predefined dictionary of products, and a 200 JSON response with the product otherwise; explain why using the int converter in the route matters here.

*Answer:* PRODUCTS = {1: "Widget", 2: "Gadget"}

@app.route("/products/<int:product_id>")
def get_product(product_id):
    if product_id not in PRODUCTS:
        return jsonify({"error": "Product not found"}), 404
    return jsonify({"id": product_id, "name": PRODUCTS[product_id]})
Using <int:product_id> instead of a plain <product_id> matters for two reasons: first, Flask automatically converts the matched URL segment to a genuine Python int rather than a str, so PRODUCTS.get(product_id) can correctly match against the integer keys in PRODUCTS without needing a manual int() conversion inside the view function; second, the int converter causes Flask to automatically reject non-numeric path segments (like /products/abc) with its own 404 before the view function even runs, providing basic input validation for free at the routing layer.

## 5.2 Path Parameters (With Type Validation for Free)

````python
@app.get("/users/{user_id}")
def get_user(user_id: int):
    return {"id": user_id, "name": f"User {user_id}"}

response = client.get("/users/42")
print(response.status_code, response.json())

bad_response = client.get("/users/not-a-number")
print(bad_response.status_code, bad_response.json())
````

# Part 6. Building REST APIs

This part builds a complete, realistic in-memory REST API demonstrating full CRUD, basic auth, validation, error handling, pagination, filtering, sorting, search and logging together, in FastAPI.

## 6.1 The Data Model and In-Memory Store

````python
from pydantic import BaseModel, Field
from typing import Optional
import logging

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("products_api")

class Product(BaseModel):
    id: int
    name: str
    price: float = Field(gt=0)
    category: str

class ProductCreate(BaseModel):
    name: str
    price: float = Field(gt=0)
    category: str

PRODUCTS: dict[int, Product] = {
    1: Product(id=1, name="Widget", price=9.99, category="tools"),
    2: Product(id=2, name="Gadget", price=49.99, category="electronics"),
    3: Product(id=3, name="Gizmo", price=19.99, category="electronics"),
}
next_id = 4
print("Initial products:", list(PRODUCTS.keys()))
````

## 6.2 CRUD: Create, Read, Update, Delete

````python
from fastapi import FastAPI, HTTPException

crud_app = FastAPI(title="Products API")

@crud_app.post("/products", response_model=Product, status_code=201)
def create_product(payload: ProductCreate):
    global next_id
    product = Product(id=next_id, **payload.model_dump())
    PRODUCTS[next_id] = product
    logger.info(f"Created product {next_id}")
    next_id += 1
    return product

@crud_app.get("/products/{product_id}", response_model=Product)
def read_product(product_id: int):
    if product_id not in PRODUCTS:
        raise HTTPException(404, "Product not found")
    return PRODUCTS[product_id]

@crud_app.put("/products/{product_id}", response_model=Product)
def update_product(product_id: int, payload: ProductCreate):
    if product_id not in PRODUCTS:
        raise HTTPException(404, "Product not found")
    PRODUCTS[product_id] = Product(id=product_id, **payload.model_dump())
    logger.info(f"Updated product {product_id}")
    return PRODUCTS[product_id]

@crud_app.delete("/products/{product_id}", status_code=204)
def delete_product(product_id: int):
    if product_id not in PRODUCTS:
        raise HTTPException(404, "Product not found")
    del PRODUCTS[product_id]
    logger.info(f"Deleted product {product_id}")

from fastapi.testclient import TestClient
crud_client = TestClient(crud_app)
created = crud_client.post("/products", json={"name": "Doohickey", "price": 5.5, "category": "tools"})
print(created.status_code, created.json())
````

## 6.3 Pagination, Filtering, Sorting and Search

````python
from typing import Optional

@crud_app.get("/products", response_model=list[Product])
def list_products(
    category: Optional[str] = None,
    search: Optional[str] = None,
    sort_by: str = "id",
    page: int = 1,
    limit: int = 10,
):
    results = list(PRODUCTS.values())
    if category:
        results = [p for p in results if p.category == category]
    if search:
        results = [p for p in results if search.lower() in p.name.lower()]
    results.sort(key=lambda p: getattr(p, sort_by))
    start = (page - 1) * limit
    return results[start:start + limit]

filtered = crud_client.get("/products?category=electronics&sort_by=price")
print([p["name"] for p in filtered.json()])

searched = crud_client.get("/products?search=gadget")
print([p["name"] for p in searched.json()])
````

## 6.4 Basic Token Authentication on Write Operations

````python
from fastapi import Header, Depends

VALID_TOKEN = "secret-token-123"

def require_auth(authorization: str = Header(default=None)):
    if authorization != f"Bearer {VALID_TOKEN}":
        raise HTTPException(401, "Invalid or missing token")

@crud_app.get("/secure/products")
def secure_list(authorization: str = Header(default=None)):
    if authorization != f"Bearer {VALID_TOKEN}":
        raise HTTPException(401, "Invalid or missing token")
    return list(PRODUCTS.values())

unauthorized = crud_client.get("/secure/products")
print(unauthorized.status_code)
authorized = crud_client.get("/secure/products", headers={"Authorization": f"Bearer {VALID_TOKEN}"})
print(authorized.status_code)
````

## 8.7 Retries With Backoff

````python
import time

def get_with_retries(client, url, max_retries=3, backoff=0.1):
    for attempt in range(1, max_retries + 1):
        response = client.get(url)
        if response.status_code < 500:
            return response
        time.sleep(backoff * attempt)
    return response

result = get_with_retries(local_client, "/items/1")
print(result.status_code)
````

| Code / Item | Explanation |
|---|---|
| `time.sleep(backoff * attempt)` | A simple linear backoff: waits progressively longer between each retry, reducing load on a struggling server; production code often uses exponential backoff instead, or a library like tenacity. |

## 8.8 Handling Pagination

````python
def fetch_all_items(client):
    all_items = []
    page = 1
    while True:
        response = client.get("/items", params={"page": page, "limit": 10})
        data = response.json()
        all_items.extend(data["items"])
        if len(all_items) >= data["total"]:
            break
        page += 1
    return all_items

result = fetch_all_items(local_client)
print("Total fetched:", len(result))
````

## 8.9 Handling Rate Limits (429)

````python
def get_respecting_rate_limit(client, url, retry_after_default=1):
    response = client.get(url)
    if response.status_code == 429:
        retry_after = int(response.headers.get("Retry-After", retry_after_default))
        print(f"Rate limited; would wait {retry_after}s and retry (simulated)")
    return response

result = get_respecting_rate_limit(local_client, "/items/1")
print(result.status_code)
````

> **Note:** Respecting a real API's Retry-After header (when present on a 429 response) is the correct, well-behaved way to handle rate limiting, rather than immediately retrying and risking a harder block.

        try:
            response = client.get(url)

                raise

# Part 9. Async APIs

This part connects directly to the async/concurrency concepts you already know, applied specifically to HTTP: async request-making with httpx.AsyncClient, concurrent API calls, timeouts, cancellation, connection pooling and rate limiting.

## 9.1 Why Async Matters for API Calls

Making an HTTP request is I/O-bound: most of the time is spent WAITING on the network, not computing. A synchronous call blocks the entire thread for that wait; an async call, using await, lets the event loop run other work during that same wait, which matters enormously when your code needs to make MANY API calls, or serve many concurrent clients each making outbound calls.

## 9.2 httpx.AsyncClient: Basic Usage

````python
import httpx
import asyncio

async def fetch_local(url):
    async with httpx.AsyncClient(transport=httpx.ASGITransport(app=consume_app), base_url="http://test") as client:
        response = await client.get(url)
        return response.json()

result = await fetch_local("/items/1")
print(result)
````

| Code / Item | Explanation |
|---|---|
| `httpx.ASGITransport(app=consume_app)` | Routes requests directly to our local FastAPI app in-process, for reliable, reproducible examples; against a real API you would simply use 'async with httpx.AsyncClient() as client: await client.get("https://api.example.com/...")' with no special transport needed. |

## 9.3 Concurrent API Calls With asyncio.gather

````python
async def fetch_many(item_ids):
    async with httpx.AsyncClient(transport=httpx.ASGITransport(app=consume_app), base_url="http://test") as client:
        tasks = [client.get(f"/items/{i}") for i in item_ids]
        responses = await asyncio.gather(*tasks)
        return [r.json() for r in responses]

results = await fetch_many([1, 2, 3, 4, 5])
print(results)
````

| Code / Item | Explanation |
|---|---|
| `asyncio.gather(*tasks)` | Runs all five requests CONCURRENTLY rather than one after another; total wait time is closer to the SLOWEST single request, not the SUM of all five, exactly the concurrency benefit already covered in your concurrency material, now applied specifically to API calls. |

## 9.4 Timeouts

````python
async def fetch_with_timeout(url, timeout_seconds=2.0):
    async with httpx.AsyncClient(
        transport=httpx.ASGITransport(app=consume_app), base_url="http://test",
        timeout=timeout_seconds,
    ) as client:
        try:
            response = await client.get(url)
            return response.status_code
        except httpx.TimeoutException:
            return "timed out"

result = await fetch_with_timeout("/items/1")
print(result)
````

## 9.5 Cancellation

````python
async def cancellable_task():
    try:
        await asyncio.sleep(10)
        return "completed"
    except asyncio.CancelledError:
        print("Task was cancelled cleanly")
        raise

task = asyncio.create_task(cancellable_task())
await asyncio.sleep(0.05)
task.cancel()
try:
    await task
except asyncio.CancelledError:
    print("Confirmed: task raised CancelledError as expected")
````

> **Note:** Cancelling a long-running outbound API call (e.g., if a client disconnects, or a timeout elapses) is important in production systems to avoid wasting resources on work whose result nobody needs anymore.

## 9.6 Connection Pooling

Creating a NEW httpx.AsyncClient (or requests.Session) for every single call discards TCP connection reuse; a single, shared client instance maintains a connection pool, reusing underlying TCP connections across multiple requests to the same host, which is significantly faster and lower-overhead for making many calls.

````python
async def efficient_many_calls(item_ids):
    async with httpx.AsyncClient(
        transport=httpx.ASGITransport(app=consume_app), base_url="http://test"
    ) as client:   # ONE client, reused connection pool, for ALL calls below
        results = []
        for i in item_ids:
            response = await client.get(f"/items/{i}")
            results.append(response.status_code)
        return results

print(await efficient_many_calls([1, 2, 3]))
````

## 9.7 Client-Side Rate Limiting With asyncio.Semaphore

````python
async def rate_limited_fetch(item_ids, max_concurrent=2):
    semaphore = asyncio.Semaphore(max_concurrent)
    async with httpx.AsyncClient(transport=httpx.ASGITransport(app=consume_app), base_url="http://test") as client:
        async def fetch_one(i):
            async with semaphore:
                response = await client.get(f"/items/{i}")
                return response.status_code
        tasks = [fetch_one(i) for i in item_ids]
        return await asyncio.gather(*tasks)

results = await rate_limited_fetch([1, 2, 3, 4, 5, 6], max_concurrent=2)
print(results)
````

| Code / Item | Explanation |
|---|---|
| `asyncio.Semaphore(max_concurrent)` | Limits how many requests are actually in-flight simultaneously (here, 2 at a time), preventing your own client from overwhelming a rate-limited external API, even though asyncio.gather() has launched all six tasks conceptually at once. |

## Part 9 Practice Exercises

**Beginner:** Why is making an HTTP request considered I/O-bound rather than CPU-bound?

*Answer:* Most of the time spent making an HTTP request is spent WAITING (for the network round trip, for the remote server to process and respond), not performing local computation; the CPU is largely idle during that wait, which is exactly the profile that benefits from async/concurrent execution rather than more CPU parallelism.

**Intermediate:** Why does creating a single, shared httpx.AsyncClient (reused across many requests) outperform creating a brand-new client for every single request?

*Answer:* A client instance maintains a connection pool of already-established TCP (and TLS, for HTTPS) connections to hosts it has talked to; reusing the same client lets subsequent requests to the same host reuse an existing open connection, skipping the overhead of a fresh TCP handshake (and TLS negotiation for HTTPS) for every single request. Creating a new client per request discards this reuse, forcing a full new connection setup each time, which is measurably slower, especially when making many requests to the same API.

**Advanced:** Explain why asyncio.gather(*tasks) alone does not protect an external API from being overwhelmed by too many simultaneous requests, and how asyncio.Semaphore fixes this, using the rate_limited_fetch example.

*Answer:* asyncio.gather(*tasks) simply schedules ALL the given coroutines to run concurrently as soon as possible, with no built-in limit on how many are actually in flight to the remote server at the same instant; if tasks contains, say, 100 requests, gather() would attempt to have all 100 in progress essentially at once, which could easily exceed a real API's rate limit or overwhelm it, resulting in many 429 or 5xx responses. Wrapping each individual request in 'async with semaphore:' (backed by asyncio.Semaphore(max_concurrent)) ensures that no more than max_concurrent coroutines can be inside that block simultaneously; every other task attempting to enter is made to wait (asynchronously, without blocking anything else) until a semaphore slot frees up, effectively THROTTLING how many requests are truly concurrent at any given moment, even though all the tasks were conceptually launched together via gather(), directly enforcing a self-imposed concurrency ceiling that respects the target API's actual capacity.

## 10.5 JWT: Structure and Concepts

A JWT (JSON Web Token) is a self-contained, cryptographically SIGNED token encoding claims (user ID, roles, expiration) as three base64-encoded, dot-separated parts: header.payload.signature. The server can verify the signature to trust the claims WITHOUT a database lookup on every request, which is JWT's main practical advantage over a plain opaque token.

````python
from jose import jwt
import datetime

SECRET_KEY = "demo-secret-change-in-production"
ALGORITHM = "HS256"

payload = {
    "sub": "user-42",
    "role": "admin",
    "exp": datetime.datetime.utcnow() + datetime.timedelta(minutes=30),
}
token = jwt.encode(payload, SECRET_KEY, algorithm=ALGORITHM)
print(token)

decoded = jwt.decode(token, SECRET_KEY, algorithms=[ALGORITHM])
print(decoded)
````

| Code / Item | Explanation |
|---|---|
| `jwt.encode(payload, SECRET_KEY, ...)` | Signs the payload with the secret key, producing a compact, URL-safe token string that any party WITHOUT the secret key cannot forge (though they CAN still read the payload, since it is only encoded, not encrypted). |
| `jwt.decode(token, SECRET_KEY, ...)` | Verifies the signature (raising an error if tampered with or expired) and returns the original payload dict. |

## 10.6 Access Tokens and Refresh Tokens

An access token is short-lived (minutes to a few hours) and sent with every authenticated request. A refresh token is longer-lived and used ONLY to obtain a new access token when the old one expires, without requiring the user to log in again; keeping access tokens short-lived limits the damage window if one is ever stolen, while refresh tokens are stored more carefully and can be revoked server-side.

## 10.7 A Complete JWT Login Flow

````python
from fastapi import FastAPI, HTTPException, Depends
from fastapi.security import HTTPBearer, HTTPAuthorizationCredentials
from jose import jwt, JWTError
import datetime

jwt_app = FastAPI()
USERS_DB = {"ada": {"password": "secret123", "role": "admin"}}

def create_token(username, role):
    payload = {"sub": username, "role": role,
               "exp": datetime.datetime.utcnow() + datetime.timedelta(minutes=30)}
    return jwt.encode(payload, SECRET_KEY, algorithm=ALGORITHM)

@jwt_app.post("/login")
def login(username: str, password: str):
    user = USERS_DB.get(username)
    if not user or user["password"] != password:
        raise HTTPException(401, "Invalid credentials")
    return {"access_token": create_token(username, user["role"]), "token_type": "bearer"}

bearer2 = HTTPBearer()

@jwt_app.get("/me")
def me(creds: HTTPAuthorizationCredentials = Depends(bearer2)):
    try:
        payload = jwt.decode(creds.credentials, SECRET_KEY, algorithms=[ALGORITHM])
    except JWTError:
        raise HTTPException(401, "Invalid or expired token")
    return {"username": payload["sub"], "role": payload["role"]}

jwt_client = TestClient(jwt_app)
login_response = jwt_client.post("/login?username=ada&password=secret123")
token = login_response.json()["access_token"]
me_response = jwt_client.get("/me", headers={"Authorization": f"Bearer {token}"})
print(me_response.json())
````

## 10.8 Password Hashing

````python
import bcrypt

plain_password = "my-plain-password"
hashed = bcrypt.hashpw(plain_password.encode(), bcrypt.gensalt())
print(hashed[:30], b"...")
print(bcrypt.checkpw(plain_password.encode(), hashed))
print(bcrypt.checkpw(b"wrong-password", hashed))
````

> **Note:** NEVER store plain-text passwords. bcrypt is a slow, salted hashing algorithm specifically designed to resist brute-force attacks; each call to hashpw() produces a DIFFERENT output (due to a random salt from gensalt()) even for the same input, and checkpw() checks a plain-text password against a stored hash without ever needing to reverse the hash. In larger applications, the passlib library's CryptContext provides a convenient wrapper supporting multiple hashing schemes and easy migration between them.

## 10.9 Role-Based Access Control (RBAC)

````python
def require_role(required_role):
    def checker(creds: HTTPAuthorizationCredentials = Depends(bearer2)):
        try:
            payload = jwt.decode(creds.credentials, SECRET_KEY, algorithms=[ALGORITHM])
        except JWTError:
            raise HTTPException(401, "Invalid or expired token")
        if payload.get("role") != required_role:
            raise HTTPException(403, "Insufficient permissions")
        return payload
    return checker

@jwt_app.get("/admin-only")
def admin_only(payload: dict = Depends(require_role("admin"))):
    return {"message": f"Welcome, admin {payload['sub']}"}

response = jwt_client.get("/admin-only", headers={"Authorization": f"Bearer {token}"})
print(response.status_code, response.json())
````

| Code / Item | Explanation |
|---|---|
| `require_role("admin")` | A dependency FACTORY: calling it with a specific role returns a customized dependency function, letting the same pattern protect different routes with different required roles. |

## Part 10 Practice Exercises

**Beginner:** What is the difference between authentication and authorization?

*Answer:* Authentication verifies WHO a client is (identity); authorization verifies WHAT that already-authenticated client is permitted to do (permissions/roles).

**Intermediate:** Why must Basic Authentication only ever be used over HTTPS, never plain HTTP?

*Answer:* Basic auth sends the username and password base64-ENCODED, not encrypted, in the Authorization header on every single request; base64 encoding is trivially reversible by anyone who intercepts the traffic, so over plain HTTP, credentials would be exposed in plain text to anyone able to observe the network traffic (a classic man-in-the-middle scenario). HTTPS encrypts the entire request (including headers) via TLS, which is the only thing actually protecting those credentials in transit.

**Advanced:** Explain why JWT access tokens are typically kept short-lived (e.g. 15-30 minutes) and paired with a separate, longer-lived refresh token, rather than simply issuing one long-lived access token, in terms of the security trade-off this design makes.

*Answer:* A JWT's signature can be verified WITHOUT a database lookup, which is exactly its performance advantage, but this also means a server generally CANNOT immediately revoke a specific already-issued JWT before it naturally expires (short of maintaining a separate revocation list, which partially defeats the no-database-lookup benefit); if a long-lived access token were ever stolen (through a compromised device, a leaked log, a network interception), an attacker could use it to impersonate the legitimate user for the ENTIRE remaining validity period, however long that is. Keeping the access token short-lived tightly bounds the damage window of a stolen token, since it will naturally expire soon regardless. The refresh token, in contrast, is used far less frequently (only to mint new access tokens) and can be stored and validated against a server-side record more carefully (and can, unlike a bare JWT, be explicitly revoked in that record if the user logs out or a compromise is detected), so the design trades the performance benefit of stateless verification for access tokens against the security benefit of a short exposure window, while still avoiding the poor user experience of requiring a fresh login every few minutes, by using the refresh token as a bridge.

## 11.1 SQL Injection

## 11.2 XSS (Cross-Site Scripting)

XSS occurs when user-supplied input is rendered back into an HTML page WITHOUT proper escaping, allowing an attacker to inject malicious JavaScript that runs in other users' browsers. Since this handbook focuses on JSON APIs (not HTML rendering), XSS is a smaller direct concern for the API itself, but matters heavily for any frontend consuming API data and rendering it as HTML; APIs should still avoid ever reflecting raw, unescaped input back in any HTML-adjacent response (e.g., error pages).

## 11.3 CSRF (Cross-Site Request Forgery)

CSRF tricks an authenticated user's browser into making an unwanted request to your API, exploiting cookies automatically sent by the browser. Token-based auth (Bearer tokens/JWT sent explicitly in an Authorization header, per Part 10) is inherently far less vulnerable to CSRF than cookie-based session auth, since the attacker's page cannot read/attach the token itself; CSRF is a much larger concern specifically for cookie-authenticated APIs.

## 11.5 Authentication Attacks and Brute Force Protection

````python
import time

failed_attempts = {}

def check_brute_force(username, max_attempts=5, window_seconds=300):
    now = time.time()
    attempts = [t for t in failed_attempts.get(username, []) if now - t < window_seconds]
    failed_attempts[username] = attempts
    if len(attempts) >= max_attempts:
        raise Exception("Too many failed login attempts; account temporarily locked")

def record_failed_attempt(username):
    failed_attempts.setdefault(username, []).append(time.time())

for i in range(6):
    try:
        check_brute_force("ada")
        record_failed_attempt("ada")
        print(f"Attempt {i+1}: allowed")
    except Exception as e:
        print(f"Attempt {i+1}: {e}")
        break
````

## 11.6 Rate Limiting

````python
import time

class SimpleRateLimiter:
    def __init__(self, max_requests, window_seconds):
        self.max_requests = max_requests
        self.window_seconds = window_seconds
        self.requests = {}

    def allow(self, client_id):
        now = time.time()
        history = [t for t in self.requests.get(client_id, []) if now - t < self.window_seconds]
        if len(history) >= self.max_requests:
            return False
        history.append(now)
        self.requests[client_id] = history
        return True

limiter = SimpleRateLimiter(max_requests=3, window_seconds=60)
for i in range(5):
    print(f"Request {i+1} allowed:", limiter.allow("client-a"))
````

> **Note:** In production, rate limiting is typically implemented at a reverse proxy/gateway level, or backed by Redis (Part 15, Part 23) for correctness across multiple server instances, rather than an in-memory dict, which does not work correctly once you have more than one server process.

**Intermediate:** Why is CORS described as a browser-enforced protection rather than a true server-side authorization mechanism?

*Answer:* CORS headers only affect how BROWSERS enforce cross-origin request restrictions on behalf of a webpage's JavaScript; a non-browser client, such as a Python script using requests, curl, or another server, is completely unaffected by CORS headers and can call the API directly regardless of the allow_origins configuration. **Advanced:** Explain why an in-memory rate limiter (as in 11.6) breaks down once an API is deployed across multiple server processes/instances, and what change is needed to fix it.

*Answer:* An in-memory rate limiter stores each client's request history in a plain Python dict that lives entirely within ONE running server process's memory; if the API is deployed behind a load balancer across multiple separate server instances (a common production setup for scalability), each instance has its OWN independent copy of that dict, with no shared visibility into what the others have recorded. A client could then effectively get max_requests allowed PER server instance rather than in total, since requests routed to different instances would each see an emptier history, completely undermining the intended global rate limit. The fix is to move the rate-limiting STATE out of any single process's memory into a shared, centralized store all instances can read and write to, most commonly Redis (Part 23), which supports atomic increment-and-expire operations well suited to implementing a correct, shared rate limit across an arbitrary number of server instances.

## 12.1 SQLAlchemy Setup (SQLite for This Demo, PostgreSQL in Production)

````python
from sqlalchemy import create_engine, Column, Integer, String, Float
from sqlalchemy.orm import sessionmaker, declarative_base
from sqlalchemy.pool import StaticPool

# SQLite in-memory here for a self-contained, dependency-free demo (StaticPool keeps
# ONE shared connection alive so the in-memory DB persists across sessions);
# in production, swap the URL for PostgreSQL, e.g.:
# "postgresql://user:password@localhost:5432/mydb"
DATABASE_URL = "sqlite:///:memory:"
engine = create_engine(
    DATABASE_URL,
    connect_args={"check_same_thread": False},
    poolclass=StaticPool,
)
SessionLocal = sessionmaker(bind=engine, autoflush=False, autocommit=False)
Base = declarative_base()

class AccountDB(Base):
    __tablename__ = "accounts"
    id = Column(Integer, primary_key=True, index=True)
    name = Column(String, nullable=False)
    balance = Column(Float, default=0.0)

Base.metadata.create_all(bind=engine)
print("Table created:", AccountDB.__tablename__)
````

## 12.2 The Database Layer (Business Logic Separated From Routes)

````python
from sqlalchemy.orm import Session

def get_db():
    db = SessionLocal()
    try:
        yield db
    finally:
        db.close()

def create_account(db: Session, name: str, initial_balance: float = 0.0):
    account = AccountDB(name=name, balance=initial_balance)
    db.add(account)
    db.commit()
    db.refresh(account)
    return account

def get_account(db: Session, account_id: int):
    return db.query(AccountDB).filter(AccountDB.id == account_id).first()

def transfer_funds(db: Session, from_id: int, to_id: int, amount: float):
    from_acc = get_account(db, from_id)
    to_acc = get_account(db, to_id)
    if not from_acc or not to_acc:
        raise ValueError("Account not found")
    if from_acc.balance < amount:
        raise ValueError("Insufficient funds")
    from_acc.balance -= amount
    to_acc.balance += amount
    db.commit()   # both changes committed TOGETHER, atomically
    return from_acc, to_acc

print("Database layer functions defined")
````

| Code / Item | Explanation |
|---|---|
| `def get_db(): ... yield db ... finally: db.close()` | A generator-based dependency (Part 5.7): yields a session for use, then guarantees cleanup via finally, exactly the resource-management pattern from your context managers material. |
| `db.commit() inside transfer_funds` | Both balance changes are part of the SAME database transaction; if anything failed before commit(), db.rollback() (implicitly, or explicitly in a broader try/except) would undo BOTH changes together, never leaving the two accounts in an inconsistent, half-transferred state. |

## 12.3 The API Layer: Client -> API -> Business Logic -> Database -> Response

````python
from fastapi import FastAPI, HTTPException, Depends
from pydantic import BaseModel

db_app = FastAPI()

class AccountCreate(BaseModel):
    name: str
    initial_balance: float = 0.0

class AccountOut(BaseModel):
    id: int
    name: str
    balance: float
    model_config = {"from_attributes": True}

class TransferRequest(BaseModel):
    from_id: int
    to_id: int
    amount: float

@db_app.post("/accounts", response_model=AccountOut, status_code=201)
def api_create_account(payload: AccountCreate, db: Session = Depends(get_db)):
    return create_account(db, payload.name, payload.initial_balance)

@db_app.get("/accounts/{account_id}", response_model=AccountOut)
def api_get_account(account_id: int, db: Session = Depends(get_db)):
    account = get_account(db, account_id)
    if not account:
        raise HTTPException(404, "Account not found")
    return account

@db_app.post("/transfer")
def api_transfer(payload: TransferRequest, db: Session = Depends(get_db)):
    try:
        from_acc, to_acc = transfer_funds(db, payload.from_id, payload.to_id, payload.amount)
    except ValueError as e:
        raise HTTPException(400, str(e))
    return {"from_balance": from_acc.balance, "to_balance": to_acc.balance}

from fastapi.testclient import TestClient
db_client = TestClient(db_app)

acc1 = db_client.post("/accounts", json={"name": "Ada", "initial_balance": 100.0}).json()
acc2 = db_client.post("/accounts", json={"name": "Bo", "initial_balance": 20.0}).json()
print(acc1, acc2)

transfer = db_client.post("/transfer", json={"from_id": acc1["id"], "to_id": acc2["id"], "amount": 30.0})
print(transfer.json())

failed_transfer = db_client.post("/transfer", json={"from_id": acc1["id"], "to_id": acc2["id"], "amount": 10000.0})
print(failed_transfer.status_code, failed_transfer.json())
````

| Code / Item | Explanation |
|---|---|
| `model_config = {"from_attributes": True}` | Lets AccountOut be built directly from a SQLAlchemy model INSTANCE (reading its attributes), not just from a dict; required whenever a route returns an ORM object directly, as api_get_account does. |
| `db: Session = Depends(get_db)` | Every route gets its OWN fresh database session, automatically closed afterward, exactly matching the get_db() generator dependency from 12.2. |
| `failed_transfer (400)` | transfer_funds() raises ValueError for insufficient funds, which the API layer catches and translates into a proper 400 HTTPException; the database itself remains untouched, since db.commit() was never reached. |

## 12.4 Transactions and Error Handling in Practice

The transfer_funds() function in 12.2 demonstrates the core transactional principle directly: multiple related database changes (debiting one account, crediting another) must succeed or fail TOGETHER; validating BOTH accounts exist and sufficient funds are available BEFORE making any change, then committing once, ensures the database is never left in a partially-updated, inconsistent state, even if an error occurs partway through in a more complex scenario.

## Part 12 Practice Exercises

**Beginner:** In the layered design shown here, which layer is responsible for raising HTTPException: the database/business logic layer, or the API/route layer?

*Answer:* The API/route layer; the business logic layer (transfer_funds, create_account) raises plain, framework-agnostic exceptions like ValueError, and the route functions translate those into proper HTTP-specific HTTPException responses, keeping the business logic layer independent of FastAPI/HTTP concerns.

**Intermediate:** Why does get_db() use 'yield db' inside a try/finally block rather than simply 'return SessionLocal()'?

*Answer:* Using yield inside a try/finally makes get_db() a generator-based dependency: FastAPI runs the code before the yield (creating the session), hands the yielded session to the route for the duration of the request, and then, CRITICALLY, resumes execution after the yield (running db.close() in the finally block) once the request is finished, GUARANTEEING the session is always properly closed, even if the route function raises an exception partway through. A plain 'return SessionLocal()' would create the session but provide no built-in mechanism to guarantee it gets closed afterward, risking leaked/unclosed database connections over time.

**Advanced:** Explain why transfer_funds() checks 'if from_acc.balance < amount: raise ValueError' BEFORE modifying either account's balance, rather than modifying from_acc.balance first and checking for a negative result afterward.

*Answer:* Validating the precondition (sufficient funds) before making ANY change ensures the function either performs the COMPLETE, correct transfer or makes NO changes at all; there is no intermediate state to accidentally leave behind. If the code instead modified from_acc.balance first (subtracting the amount) and only checked for a negative balance afterward, a check failure would require the code to explicitly UNDO that already-applied change (reverting from_acc.balance) before raising the error, adding complexity and creating a real risk that some future code change might forget to revert it correctly, or that a crash occurs between the modification and the revert, leaving from_acc.balance negative and to_acc's balance unchanged. Validating first and mutating only after every precondition is confirmed is a simpler, more robust pattern that avoids ever needing to unwind a partially-applied change.

# Part 13. API Architecture

This part explains the professional separation of concerns behind any non-trivial API: routes, services, repositories, models, schemas, and configuration, and shows a real project folder structure.

## 13.1 The Layers, Defined

Table 13.1a lists each layer and its single responsibility; the sections that follow explain why keeping them separate matters.

## 13.2 Why Separate These Layers?

- Testability: business logic (services) can be unit-tested without spinning up a real HTTP server or database, by mocking the repository layer.
- Swappability: the repository layer isolates database specifics, so switching from SQLite to PostgreSQL, or even swapping SQLAlchemy for a different data layer, touches ONE layer, not routes scattered across the whole codebase.
- Clarity: each file/module has ONE clear job, making a large codebase navigable; a bug in a validation rule is obviously in a schema, a bug in a calculation is obviously in a service.
- Reuse: the same service function can be called from a route, a background job (Part 15), or a CLI script, without duplicating logic.

## 13.4 A Concrete Example Across All Layers

````python
# repositories/account_repository.py
from fastapi import Depends

class AccountRepository:
    def __init__(self, db):
        self.db = db
    def get(self, account_id):
        return self.db.query(AccountDB).filter(AccountDB.id == account_id).first()
    def create(self, name, balance):
        account = AccountDB(name=name, balance=balance)
        self.db.add(account)
        self.db.commit()
        self.db.refresh(account)
        return account

# services/account_service.py
class AccountService:
    def __init__(self, repository):
        self.repository = repository
    def open_account(self, name, initial_balance):
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative")
        return self.repository.create(name, initial_balance)

# routers/accounts.py (conceptual; using our db_app instance's session dependency)
def make_account_service(db=Depends(get_db)):
    return AccountService(AccountRepository(db))

@db_app.post("/v2/accounts")
def api_open_account(payload: AccountCreate, service: AccountService = Depends(make_account_service)):
    try:
        account = service.open_account(payload.name, payload.initial_balance)
    except ValueError as e:
        raise HTTPException(400, str(e))
    return {"id": account.id, "name": account.name, "balance": account.balance}

response = db_client.post("/v2/accounts", json={"name": "Chi", "initial_balance": 50.0})
print(response.json())
````

| Code / Item | Explanation |
|---|---|
| `AccountRepository` | The ONLY class that constructs SQLAlchemy queries; AccountService never imports AccountDB or writes a query directly. |
| `AccountService.open_account` | Contains the actual BUSINESS RULE (no negative initial balance); this same method could be called from a route, a script, or a test, with identical behavior. |
| `make_account_service (a dependency of a dependency)` | Composes the repository and service together per-request, following the same Depends() pattern from Part 5.7, keeping the ROUTE function itself extremely thin. |

# Part 14. Middleware

Middleware wraps every request/response passing through your application, letting you implement cross-cutting concerns (logging, timing, authentication, CORS, error handling) in ONE place rather than duplicating them in every route.

## 14.1 What Middleware Is

```
Client Request
     |
     v
[ Middleware 1 ] --before-->
     |
     v
[ Middleware 2 ] --before-->
     |
     v
 [ Route Handler ]
     |
     v
[ Middleware 2 ] <--after---
     |
     v
[ Middleware 1 ] <--after---
     |
     v
Client Response
```

## 14.2 Request and Response Middleware in One Function

````python
from fastapi import FastAPI, Request
import time

mw_demo_app = FastAPI()

@mw_demo_app.middleware("http")
async def log_requests(request: Request, call_next):
    print(f"REQUEST: {request.method} {request.url.path}")   # 'before' logic
    start = time.time()
    response = await call_next(request)                        # runs the route
    duration = time.time() - start
    print(f"RESPONSE: {response.status_code} in {duration:.4f}s")  # 'after' logic
    return response

@mw_demo_app.middleware("http")
async def add_process_time_header(request: Request, call_next):
    start = time.time()
    response = await call_next(request)
    response.headers["X-Process-Time-Ms"] = str(round((time.time() - start) * 1000, 2))
    return response

@mw_demo_app.get("/ping")
def ping():
    return {"pong": True}

from fastapi.testclient import TestClient
mw_client = TestClient(mw_demo_app)
response = mw_client.get("/ping")
print(response.json())
print(response.headers.get("x-process-time-ms"))
````

| Code / Item | Explanation |
|---|---|
| `code BEFORE 'await call_next(request)'` | Runs on the way IN, before the route handler executes; can inspect/modify the request, or short-circuit entirely (return early without calling the route). |
| `code AFTER 'await call_next(request)'` | Runs on the way OUT, after the route handler has produced a response; can inspect/modify the response (headers, logging). |

## 14.3 Timing Middleware: How the Header Is Added

The add_process_time_header middleware registered alongside log_requests in 14.2 demonstrates the response-modification half of middleware directly: it records a start time before calling the route, then, once call_next() returns the response, attaches a new X-Process-Time-Ms header to it before that response continues on its way back to the client, exactly as the printed header value above confirms.

## 14.4 Authentication Middleware

````python
from fastapi.responses import JSONResponse

auth_mw_app = FastAPI()
PUBLIC_PATHS = {"/ping", "/docs", "/openapi.json"}

@auth_mw_app.middleware("http")
async def require_auth_header(request: Request, call_next):
    if request.url.path not in PUBLIC_PATHS and "authorization" not in request.headers:
        return JSONResponse(status_code=401, content={"error": "Missing Authorization header"})
    return await call_next(request)

@auth_mw_app.get("/ping")
def ping2():
    return {"pong": True}

@auth_mw_app.get("/secret")
def secret():
    return {"secret": 42}

auth_mw_client = TestClient(auth_mw_app)
print(auth_mw_client.get("/ping").status_code)      # public path: 200
print(auth_mw_client.get("/secret").status_code)    # no auth header: 401
````

> **Note:** Authentication middleware applied GLOBALLY like this is a coarse-grained approach; per-route Depends()-based auth (Part 5.7, Part 10) is usually preferred for fine-grained control over exactly which routes need which kind of auth, but global middleware is useful for blanket rules (e.g., 'require SOME form of auth on everything except health checks and docs').

## 14.5 CORS Middleware

Covered fully in Part 11.4; CORSMiddleware is itself implemented as middleware, following exactly this same before/after request-wrapping pattern, specialized for adding the specific CORS-related response headers browsers check.

## 14.6 Error-Handling Middleware

````python
err_mw_app = FastAPI()

@err_mw_app.middleware("http")
async def catch_all_errors(request: Request, call_next):
    try:
        return await call_next(request)
    except Exception as e:
        print(f"Unhandled error: {e}")
        return JSONResponse(status_code=500, content={"error": "Internal server error"})

@err_mw_app.get("/boom")
def boom():
    raise ValueError("Something unexpected broke")

err_mw_client = TestClient(err_mw_app, raise_server_exceptions=False)
response = err_mw_client.get("/boom")
print(response.status_code, response.json())
````

> **Note:** A catch-all middleware like this is a SAFETY NET for truly unexpected exceptions, ensuring clients always get a clean JSON 500 response rather than a raw traceback or a broken connection; it should NOT replace deliberate, specific error handling (HTTPException, Part 5.6) for expected error conditions.

## 14.7 Middleware Ordering Matters

Middleware registered LAST runs FIRST on the way in (and last on the way out), forming a nested, onion-like structure; ordering matters when one middleware depends on another having already run (e.g., a logging middleware that wants to log the authenticated user's ID needs to run AFTER authentication middleware has attached that ID to the request).

## Part 14 Practice Exercises

**Beginner:** In FastAPI middleware, what does the code written BEFORE 'await call_next(request)' run in relation to the route handler?

*Answer:* It runs before the route handler executes, on the way IN, while the code written after call_next() runs after the route handler has produced its response, on the way OUT.

**Intermediate:** Why is a global, catch-all error-handling middleware (14.6) described as a 'safety net' rather than a replacement for using HTTPException for expected errors?

*Answer:* HTTPException (Part 5.6) is for EXPECTED, anticipated error conditions (a resource not found, invalid input) where the API deliberately chooses a specific, meaningful status code and error message tailored to that exact situation. A catch-all error middleware exists to handle TRULY UNEXPECTED exceptions, bugs, unhandled edge cases, third-party library failures, that were never anticipated by a specific HTTPException call anywhere in the code; without it, such an exception would propagate uncaught and could crash the request handling or leak an unfriendly raw traceback to the client. Relying on the catch-all middleware for expected errors instead of raising a specific HTTPException would lose the ability to return precise, meaningful status codes and messages for known failure cases, collapsing everything into the same generic 500 response regardless of the actual cause.

**Advanced:** A team registers a logging middleware that reads request.state.user_id (expected to be set by a separate authentication middleware) and BOTH middlewares are added to the app, but the log always shows user_id as missing. Using the 'ordering matters' principle from 14.7, diagnose the likely cause and the fix.

*Answer:* Since middleware registered LAST runs FIRST on the incoming side, if the logging middleware was registered AFTER (later in the code than) the authentication middleware, the logging middleware would actually run BEFORE the authentication middleware on the way in, meaning it executes and tries to read request.state.user_id before the authentication middleware has had a chance to set it, explaining why it is always missing. The fix is to register the authentication middleware AFTER the logging middleware in the source code (since FastAPI/Starlette middleware registered later wraps AROUND, and therefore runs before, middleware registered earlier for the incoming request path), ensuring authentication middleware runs first and sets user_id before the logging middleware's 'before' code executes and tries to read it; alternatively, moving the actual logging statement to the 'after call_next()' portion of the logging middleware would also work in some orderings, since by the time the response comes back, the authentication middleware (if it ran at any point before the route handler) would have already set the state.

# Part 15. Background Tasks and Job Queues

This part explains when to make the client wait versus when to hand work off to run separately: FastAPI's built-in BackgroundTasks, worker/queue concepts, Celery, Redis/RabbitMQ, scheduled tasks, and retry mechanisms.

## 15.1 When to Use Background Processing Instead of Making the Client Wait

- The work takes longer than an acceptable HTTP response time (seconds, not milliseconds): generating a large report, processing an uploaded video, retraining a model.
- The work does not need to complete before responding usefully: sending a confirmation email after signup, logging an analytics event, invalidating a cache.
- The work should be retried on failure independently of the original request (a webhook delivery, an external API call that might transiently fail).
- The work needs to be scheduled for later or run periodically (a nightly data sync, a report generated every Monday morning).

## 15.2 FastAPI's Built-in BackgroundTasks (Recap and Extension)

````python
from fastapi import FastAPI, BackgroundTasks
import time

bg_app = FastAPI()

def write_audit_log(action: str, user: str):
    time.sleep(0.01)   # simulates I/O, e.g. writing to a log file/service
    print(f"AUDIT: {user} performed {action}")

@bg_app.post("/orders")
def create_order(user: str, background_tasks: BackgroundTasks):
    background_tasks.add_task(write_audit_log, "create_order", user)
    return {"message": "Order created"}

from fastapi.testclient import TestClient
bg_client = TestClient(bg_app)
response = bg_client.post("/orders?user=ada")
print(response.json())
````

> **Note:** FastAPI's BackgroundTasks run AFTER the response is sent, in the SAME process; they are appropriate for lightweight, fire-and-forget work, but are NOT durable (a server crash/restart loses any pending background task, and there is no retry mechanism built in), which is exactly the gap dedicated job queues fill.

## 15.3 Worker Processes and Queues: The General Pattern

```
API Process                  Queue (e.g. Redis)             Worker Process(es)
    |                              |                                |
    | 1. enqueue a job ----------->|                                |
    |    (returns immediately)     |                                |
    |                              |<---- 2. worker picks up job ---|
    |                              |                                |
    |                              |       3. worker executes job   |
    |                              |          (independently, can   |
    |                              |           retry on failure)    |
```

## 15.4 Celery: Concepts

Celery is Python's most widely used distributed task queue: you define tasks as decorated functions, a message broker (typically Redis or RabbitMQ) carries job messages between your API process and separate worker processes, and workers execute tasks independently, with built-in support for retries, scheduling, and monitoring.

````python
celery_example = '''
# tasks.py
from celery import Celery

celery_app = Celery("myapp", broker="redis://localhost:6379/0")

@celery_app.task(bind=True, max_retries=3)
def send_email_task(self, to_address, subject):
    try:
        # actual email-sending logic here
        print(f"Sending email to {to_address}: {subject}")
    except Exception as exc:
        raise self.retry(exc=exc, countdown=5)

# main.py (FastAPI route)
from tasks import send_email_task

@app.post("/signup")
def signup(email: str):
    send_email_task.delay(email, "Welcome!")   # enqueues, returns IMMEDIATELY
    return {"message": "Signup received"}
'''
print(celery_example)
````

| Code / Item | Explanation |
|---|---|
| `@celery_app.task(bind=True, max_retries=3)` | Declares a function as a Celery task, with automatic retry support (up to 3 attempts) built in. |
| `send_email_task.delay(...)` | Enqueues the task to run on a SEPARATE worker process, returning immediately without waiting for the email to actually be sent. |
| `self.retry(exc=exc, countdown=5)` | Explicitly re-queues the task to run again after 5 seconds if it fails, up to max_retries times. |

## 15.5 Redis and RabbitMQ as Message Brokers (Concepts)

| Code / Item | Explanation |
|---|---|
| `Redis` | An in-memory data store, commonly used as BOTH a message broker (simple, fast) and a cache (Part 23); simpler to operate, good for most use cases. |
| `RabbitMQ` | A dedicated, feature-rich message broker with more advanced routing/delivery guarantees; often chosen for more complex, high-throughput or enterprise messaging needs. |

## 15.6 Scheduled Tasks

````python
scheduling_example = '''
# Using Celery Beat for periodic tasks:
from celery.schedules import crontab

celery_app.conf.beat_schedule = {
    "nightly-report": {
        "task": "tasks.generate_nightly_report",
        "schedule": crontab(hour=2, minute=0),   # every day at 2:00 AM
    },
}
'''
print(scheduling_example)
````

> **Note:** Scheduled/periodic tasks (nightly data syncs, weekly reports, model retraining jobs) are a natural extension of the same task-queue infrastructure, run by a separate scheduler process (Celery Beat) that enqueues jobs on a timer rather than in response to an API request.

## 15.7 Retry Mechanisms: Why They Matter

Background jobs frequently depend on external systems (an email provider, another API, a database) that can fail TRANSIENTLY; a well-designed job queue automatically retries failed jobs (often with exponential backoff, per Part 8.7's pattern applied at the job level), and after exhausting retries, routes the job to a dead-letter queue for manual inspection, rather than silently losing the work.

## Part 15 Practice Exercises

**Beginner:** Name two situations where using a background task is more appropriate than making the client wait for the work to finish.

*Answer:* Any two of: sending a confirmation email after signup, generating a large report, processing an uploaded file, logging an analytics event, or any other work that either takes too long for an acceptable HTTP response time or does not need to complete before a useful response can be returned.

**Intermediate:** What is the key durability limitation of FastAPI's built-in BackgroundTasks compared to a dedicated task queue like Celery?

*Answer:* FastAPI's BackgroundTasks run within the SAME process that handled the original request, and are held only in memory; if the server process crashes or restarts before a background task has finished (or even started), that task is simply lost with no record of it and no automatic retry. A dedicated task queue like Celery persists enqueued jobs in a separate, durable broker (Redis/RabbitMQ), independent of any single API process's lifetime, so jobs survive an API server restart and can be picked up by any available worker, with built-in retry support for failures.

**Advanced:** A team is deciding whether to use FastAPI's BackgroundTasks or Celery for a feature that sends a push notification after a user completes a purchase. Describe the specific conditions under which each choice would be the more appropriate one.

*Answer:* FastAPI's BackgroundTasks would be appropriate if the push notification is a simple, best-effort, non-critical action, brief in execution time, where an occasional lost notification (due to a rare server restart at just the wrong moment) is an acceptable, low-consequence outcome, and the team wants to avoid the operational overhead of running and monitoring a separate message broker and worker fleet for something this lightweight. # Part 17. APIs for Machine Learning

This part builds a complete ML prediction API end to end: Client -> FastAPI -> Validation -> Preprocessing -> Model -> Prediction -> JSON Response, including probability scores, model versioning, and logging.

## 17.1 Training and Saving a Simple Model (Setup)

````python
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification
import joblib

X, y = make_classification(n_samples=200, n_features=4, random_state=42)
model = LogisticRegression().fit(X, y)
joblib.dump(model, "demo_model.joblib")
print("Model trained and saved. Coefficients shape:", model.coef_.shape)
````

## 17.2 Feature Validation With Pydantic

````python
from pydantic import BaseModel, Field

class PredictionInput(BaseModel):
    feature_1: float
    feature_2: float
    feature_3: float
    feature_4: float

class PredictionOutput(BaseModel):
    prediction: int
    probability: float
    model_version: str

print("Input/output schemas defined")
````

## 17.3 Loading the Model and Building the Prediction Endpoint

````python
from fastapi import FastAPI, HTTPException
import numpy as np
import logging

ml_app = FastAPI(title="ML Prediction API")
logger = logging.getLogger("ml_api")
logging.basicConfig(level=logging.INFO)

MODEL_VERSION = "v1.0.0"
loaded_model = joblib.load("demo_model.joblib")

@ml_app.post("/predict", response_model=PredictionOutput)
def predict(payload: PredictionInput):
    features = np.array([[payload.feature_1, payload.feature_2,
                           payload.feature_3, payload.feature_4]])
    try:
        prediction = int(loaded_model.predict(features)[0])
        probability = float(loaded_model.predict_proba(features)[0][prediction])
    except Exception as e:
        logger.error(f"Prediction failed: {e}")
        raise HTTPException(500, "Prediction failed")
    logger.info(f"Prediction made: class={prediction} prob={probability:.3f} version={MODEL_VERSION}")
    return PredictionOutput(prediction=prediction, probability=round(probability, 4), model_version=MODEL_VERSION)

from fastapi.testclient import TestClient
ml_client = TestClient(ml_app)
response = ml_client.post("/predict", json={
    "feature_1": 0.5, "feature_2": -1.2, "feature_3": 0.8, "feature_4": 1.1,
})
print(response.status_code, response.json())
````

| Code / Item | Explanation |
|---|---|
| `payload: PredictionInput` | FastAPI validates every feature is a genuine float BEFORE the model ever sees the data; malformed input (a missing feature, a string that isn't numeric) is rejected automatically with 422, never reaching the model. |
| `loaded_model.predict_proba(features)[0][prediction]` | Retrieves the model's confidence in its OWN predicted class specifically, not just an arbitrary probability. |
| `response_model=PredictionOutput` | Guarantees the API's output contract stays fixed and well-documented, regardless of what the model object itself returns internally. |

## 17.5 Model Versioning

````python
@ml_app.get("/model-info")
def model_info():
    return {"version": MODEL_VERSION, "n_features": loaded_model.coef_.shape[1]}

response = ml_client.get("/model-info")
print(response.json())
````

> **Note:** Returning model_version with every prediction (17.3) lets consumers, and your own monitoring (Part 26), track exactly which model version produced a given prediction, essential when comparing model versions in production (A/B testing, gradual rollouts, debugging a regression after a redeploy).

## 17.6 A Batch Prediction Endpoint

````python
class BatchPredictionInput(BaseModel):
    instances: list[PredictionInput]

@ml_app.post("/predict/batch")
def predict_batch(payload: BatchPredictionInput):
    features = np.array([[inst.feature_1, inst.feature_2, inst.feature_3, inst.feature_4]
                          for inst in payload.instances])
    predictions = loaded_model.predict(features)
    probabilities = loaded_model.predict_proba(features)
    results = [
        {"prediction": int(p), "probability": round(float(probabilities[i][p]), 4)}
        for i, p in enumerate(predictions)
    ]
    return {"results": results, "model_version": MODEL_VERSION}

response = ml_client.post("/predict/batch", json={"instances": [
    {"feature_1": 0.5, "feature_2": -1.2, "feature_3": 0.8, "feature_4": 1.1},
    {"feature_1": -0.3, "feature_2": 0.4, "feature_3": -0.9, "feature_4": 0.2},
]})
print(response.json())
````

> **Note:** Batching many predictions into ONE request/one model call is significantly more efficient than N separate /predict calls, both for network overhead and, for many model types, for the model's own vectorized computation.

# Part 18. APIs for AI Engineering

## 18.1 A Text Classification / Sentiment API

````python
from fastapi import FastAPI
from pydantic import BaseModel

ai_app = FastAPI()

class TextInput(BaseModel):
    text: str

class SentimentOutput(BaseModel):
    sentiment: str
    confidence: float

def simulated_sentiment_model(text: str):
    positive_words = {"good", "great", "excellent", "love", "amazing"}
    negative_words = {"bad", "terrible", "hate", "awful", "poor"}
    words = set(text.lower().split())
    pos = len(words & positive_words)
    neg = len(words & negative_words)
    if pos > neg:
        return "positive", min(0.6 + 0.1 * pos, 0.99)
    elif neg > pos:
        return "negative", min(0.6 + 0.1 * neg, 0.99)
    return "neutral", 0.5

@ai_app.post("/sentiment", response_model=SentimentOutput)
def sentiment(payload: TextInput):
    label, confidence = simulated_sentiment_model(payload.text)
    return SentimentOutput(sentiment=label, confidence=round(confidence, 2))

from fastapi.testclient import TestClient
ai_client = TestClient(ai_app)
print(ai_client.post("/sentiment", json={"text": "This product is amazing and great"}).json())
print(ai_client.post("/sentiment", json={"text": "This is terrible and bad"}).json())
````

> **Note:** In production, simulated_sentiment_model would be replaced by a call to a real model (a local transformer model, or an external LLM API); the API-layer contract (validate input, call the model, shape the output) stays identical.

## 18.2 An Embeddings Endpoint

````python
import hashlib

def simulated_embedding(text: str, dims: int = 8):
    # Deterministic pseudo-embedding for illustration; a real system would
    # call a genuine embedding model (local or via an API).
    h = hashlib.sha256(text.encode()).digest()
    return [b / 255.0 for b in h[:dims]]

class EmbeddingOutput(BaseModel):
    embedding: list[float]
    dims: int

@ai_app.post("/embeddings", response_model=EmbeddingOutput)
def embeddings(payload: TextInput):
    vector = simulated_embedding(payload.text)
    return EmbeddingOutput(embedding=vector, dims=len(vector))

response = ai_client.post("/embeddings", json={"text": "Python web APIs"})
print(response.json())
````

## 18.3 Document Processing (Chunking, Directly Reusing Prior Handbook Material)

````python
def chunk_text(text, chunk_size=50, overlap=10):
    start = 0
    chunks = []
    while start < len(text):
        chunks.append(text[start:start + chunk_size])
        start += chunk_size - overlap
    return chunks

class DocumentInput(BaseModel):
    text: str
    chunk_size: int = 50

@ai_app.post("/documents/chunk")
def chunk_document(payload: DocumentInput):
    chunks = chunk_text(payload.text, payload.chunk_size)
    return {"num_chunks": len(chunks), "chunks": chunks}

long_text = "Python is a versatile language. " * 5
response = ai_client.post("/documents/chunk", json={"text": long_text, "chunk_size": 40})
print(response.json()["num_chunks"])
````

> **Note:** This directly reuses the generator-based chunking approach from the companion Iterators/Generators handbook's Part 18, now exposed as an actual API endpoint.

## 18.4 A Minimal RAG (Retrieval-Augmented Generation) Endpoint

````python
DOCUMENT_STORE = {
    "doc1": "FastAPI is a modern Python web framework built on Starlette and Pydantic.",
    "doc2": "Flask is a minimal, flexible Python web microframework.",
    "doc3": "Redis is an in-memory data store often used for caching and message queues.",
}

def simple_retrieve(query: str, top_k: int = 1):
    scored = [(doc_id, sum(1 for w in query.lower().split() if w in text.lower()))
              for doc_id, text in DOCUMENT_STORE.items()]
    scored.sort(key=lambda x: x[1], reverse=True)
    return [DOCUMENT_STORE[doc_id] for doc_id, score in scored[:top_k] if score > 0]

class RAGInput(BaseModel):
    query: str

@ai_app.post("/rag/query")
def rag_query(payload: RAGInput):
    retrieved = simple_retrieve(payload.query, top_k=1)
    context = " ".join(retrieved)
    simulated_answer = f"Based on retrieved context: {context}" if context else "No relevant context found."
    return {"query": payload.query, "retrieved_context": retrieved, "answer": simulated_answer}

response = ai_client.post("/rag/query", json={"query": "What is FastAPI"})
print(response.json())
````

| Code / Item | Explanation |
|---|---|
| `simple_retrieve()` | A minimal stand-in for a real retrieval step (which, in production, would use vector similarity search over embeddings, Part 18.2, typically backed by a vector database). |
| `simulated_answer` | A stand-in for the GENERATION step, where retrieved context would normally be passed to an actual LLM to produce a grounded answer; the API SHAPE (retrieve, then generate, then respond) is the reusable pattern, independent of which specific models back it. |

## 18.5 A Simple AI Agent Endpoint (Tool-Calling Pattern, Simplified)

````python
def calculator_tool(expression: str):
    import ast, operator
    ops = {ast.Add: operator.add, ast.Sub: operator.sub, ast.Mult: operator.mul, ast.Div: operator.truediv}
    node = ast.parse(expression, mode="eval").body
    def eval_node(n):
        if isinstance(n, ast.BinOp):
            return ops[type(n.op)](eval_node(n.left), eval_node(n.right))
        return n.value
    return eval_node(node)

class AgentInput(BaseModel):
    instruction: str

@ai_app.post("/agent/run")
def agent_run(payload: AgentInput):
    if "calculate" in payload.instruction.lower():
        expr = payload.instruction.lower().replace("calculate", "").strip()
        try:
            result = calculator_tool(expr)
            return {"tool_used": "calculator", "result": result}
        except Exception:
            return {"tool_used": "calculator", "error": "Could not parse expression"}
    return {"tool_used": None, "result": "No matching tool for this instruction"}

response = ai_client.post("/agent/run", json={"instruction": "calculate 4 + 5 * 2"})
print(response.json())
````

> **Note:** This is a deliberately simplified illustration of the AGENT PATTERN (deciding which tool to invoke based on input, then invoking it and returning a structured result); production agent systems use an actual LLM to make the tool-selection decision, but the surrounding API structure, exposing tool-execution as a well-defined, validated endpoint, remains the same shape.

# Part 19. Streaming

This part covers streaming HTTP responses: chunked responses, Server-Sent Events, and async streaming, directly connecting to the companion Iterators/Generators handbook's coverage of Python generators and the distinction between in-process generators and network-level streaming.

## 19.1 Why Streaming Matters for AI Applications

An LLM generating a long response can take many seconds to complete in full; streaming lets the client start displaying tokens/words AS THEY ARE GENERATED, dramatically improving perceived responsiveness, rather than the client staring at a blank screen until the entire response is ready.

## 19.2 A Basic Streaming Response With FastAPI

````python
from fastapi import FastAPI
from fastapi.responses import StreamingResponse
import time

stream_app = FastAPI()

def generate_words():
    words = ["The", "quick", "brown", "fox", "jumps"]
    for word in words:
        yield word + " "

@stream_app.get("/stream-text")
def stream_text():
    return StreamingResponse(generate_words(), media_type="text/plain")

from fastapi.testclient import TestClient
stream_client = TestClient(stream_app)
response = stream_client.get("/stream-text")
print(response.text)
````

| Code / Item | Explanation |
|---|---|
| `StreamingResponse(generate_words(), ...)` | Takes a PYTHON GENERATOR (Part 9 of the Iterators handbook) and streams its yielded values to the client incrementally as HTTP chunks, rather than buffering everything and sending one complete response at the end. |

## 19.3 Simulating Token-by-Token LLM Output

````python
def simulated_llm_stream(prompt: str):
    response_text = f"Here is a response to: {prompt}. This demonstrates streaming output."
    for word in response_text.split():
        yield word + " "

@stream_app.get("/generate")
def generate(prompt: str):
    return StreamingResponse(simulated_llm_stream(prompt), media_type="text/plain")

response = stream_client.get("/generate?prompt=explain+APIs")
print(response.text)
````

> **Note:** In a real system, simulated_llm_stream would instead be an async generator wrapping an actual streaming call to a model API (Part 32 of the Iterators handbook covers async generators in full); the FastAPI/StreamingResponse layer works identically either way.

## 19.4 Server-Sent Events (SSE)

SSE is a standard, simple protocol for a server to push a stream of named, structured text events to a client over a single long-lived HTTP connection, commonly used for streaming chat/LLM responses to a browser, since it is simpler than WebSockets for one-directional (server-to-client) streaming and works over plain HTTP.

````python
import json

def sse_event_stream():
    events = [{"token": "Hello"}, {"token": " world"}, {"token": "!"}, {"done": True}]
    for event in events:
        yield f"data: {json.dumps(event)}\n\n"

@stream_app.get("/sse-demo")
def sse_demo():
    return StreamingResponse(sse_event_stream(), media_type="text/event-stream")

response = stream_client.get("/sse-demo")
print(response.text)
````

| Code / Item | Explanation |
|---|---|
| `media_type="text/event-stream"` | The specific MIME type that tells browsers/clients to interpret the stream using the SSE protocol. |
| `f"data: {json.dumps(event)}\n\n"` | The SSE wire format: each event is prefixed with 'data: ' and terminated with a blank line (two newlines); a JavaScript EventSource client parses this format natively. |

## 19.5 Async Streaming

````python
import asyncio

async def async_llm_stream(prompt: str):
    words = f"Async response to {prompt}".split()
    for word in words:
        await asyncio.sleep(0)   # simulates an async wait per token (e.g. waiting on the model)
        yield word + " "

@stream_app.get("/async-generate")
async def async_generate(prompt: str):
    return StreamingResponse(async_llm_stream(prompt), media_type="text/plain")

response = stream_client.get("/async-generate?prompt=hello")
print(response.text)
````

| Code / Item | Explanation |
|---|---|
| `async def async_llm_stream(...): ... yield ...` | An async generator (Part 32 of the Iterators handbook); StreamingResponse accepts either a plain generator OR an async generator, using the async version specifically when each 'next chunk' genuinely involves an async wait (a real streaming API call), so the server can serve OTHER requests concurrently during those waits. |

## 19.6 Chunked Responses (How Streaming Is Actually Transmitted)

Under the hood, HTTP's 'chunked transfer encoding' is what allows a server to send a response of UNKNOWN total length in successive pieces (chunks), which is exactly the transport mechanism StreamingResponse relies on; the client processes each chunk as it arrives rather than waiting for a Content-Length-bounded, complete response.

## Part 19 Practice Exercises

**Beginner:** What FastAPI response class is used to stream a Python generator's output incrementally to a client?

*Answer:* StreamingResponse (from fastapi.responses).

**Intermediate:** What specific media_type value signals to a client that a response should be interpreted using the Server-Sent Events protocol?

*Answer:* "text/event-stream".

**Advanced:** Explain why async_llm_stream (19.5) is defined as an async generator (async def with yield) rather than a plain synchronous generator, specifically in terms of what happens to the FastAPI server's ability to handle OTHER concurrent requests while this one is streaming.

*Answer:* A plain synchronous generator used inside an async route would, on each iteration, run entirely synchronously; if producing each chunk involved a genuinely blocking wait (e.g., a blocking network call to a model API), that wait would tie up the server's event loop, preventing it from making progress on any OTHER concurrent request during that time, exactly the blocking-versus-non-blocking distinction covered in Part 32 of the Iterators handbook. Defining async_llm_stream as an async generator, using 'await asyncio.sleep(0)' (standing in for a real 'await an async model API call') between yields, means each wait is a genuine async await point: the event loop is explicitly given the opportunity to switch to and make progress on OTHER pending requests during that wait, only resuming this particular stream once its awaited operation completes. This is essential for a production streaming endpoint serving MULTIPLE concurrent clients simultaneously, since a naive synchronous/blocking implementation would effectively serialize all streaming responses, processing them one at a time rather than genuinely concurrently.

## 21.7 Load Testing Concepts

Load testing measures how an API behaves under many CONCURRENT requests: response time percentiles (p50, p95, p99), error rates, and throughput (requests/second) as load increases, typically using a dedicated tool (locust, k6, Apache JMeter) rather than a unit-testing framework. Unlike unit/integration tests (which verify CORRECTNESS), load tests verify PERFORMANCE and STABILITY under realistic or peak traffic, revealing issues like connection pool exhaustion or slow database queries that only appear under concurrent load.

## Part 21 Practice Exercises

**Beginner:** What is the key difference between a unit test and an integration test for an API?

*Answer:* A unit test verifies one function or class in isolation, with its dependencies mocked out; an integration test verifies multiple layers (routing, validation, business logic, database) working together, typically through a real or realistic test client and test database.

**Intermediate:** Why is mocking requests.get() in a test (21.5) preferable to letting the test actually call the real external API?

**Advanced:** Explain why load testing is described as verifying performance and stability rather than correctness, and describe a concrete failure mode that a full, passing suite of unit and integration tests would NOT catch, but a load test would.

*Answer:* Unit and integration tests typically run requests one (or a small, fixed number) at a time, verifying that GIVEN a specific input, the system produces the CORRECT output; they say nothing about how the system behaves when many requests arrive simultaneously and compete for shared, limited resources. A concrete failure mode load testing would catch but ordinary tests would not: a database connection pool configured with, say, 10 maximum connections might work perfectly in every unit/integration test (which typically use one connection at a time), yet under a load test simulating 200 concurrent users, requests could pile up waiting for a free connection, causing response times to balloon or requests to start timing out and failing, entirely due to CONCURRENT resource contention that never manifests when tests run requests sequentially or in small numbers; this is a stability/performance problem, not a correctness problem, since the underlying business logic itself may be completely correct.

## 22.6 Query Optimization: The N+1 Query Problem

````python
n_plus_1_example = '''
# INEFFICIENT: N+1 queries (1 query for orders, then N more for each order's customer)
orders = db.query(Order).all()
for order in orders:
    print(order.customer.name)   # triggers a SEPARATE query per order!

# EFFICIENT: eager loading fetches everything in far fewer queries
from sqlalchemy.orm import joinedload
orders = db.query(Order).options(joinedload(Order.customer)).all()
for order in orders:
    print(order.customer.name)   # no extra query; already loaded
'''
print(n_plus_1_example)
````

**Intermediate:** What is the N+1 query problem, and why does it often go unnoticed during development?

*Answer:* The N+1 problem occurs when code fetches a list of N parent records with one query, then triggers a SEPARATE additional query for each parent's related data inside a loop, resulting in 1 + N total queries instead of a small, fixed number. # Part 23. Caching

This part covers why caching matters, in-memory caching, Redis concepts, cache keys, expiration, invalidation, and caching expensive ML/AI results specifically.

## 23.1 Why Caching Is Useful

Caching stores the RESULT of an expensive operation (a database query, an aggregation, a model inference call) so that a subsequent, IDENTICAL request can be served instantly from the cache instead of repeating the expensive work; this is often the single biggest performance lever available for read-heavy, repeat-request-heavy APIs.

## 23.2 Simple In-Memory Caching

````python
import time
import functools

def timed_cache(ttl_seconds):
    def decorator(func):
        cache = {}
        @functools.wraps(func)
        def wrapper(*args):
            now = time.time()
            if args in cache and now - cache[args][1] < ttl_seconds:
                return cache[args][0]
            result = func(*args)
            cache[args] = (result, now)
            return result
        return wrapper
    return decorator

call_count = 0

@timed_cache(ttl_seconds=5)
def expensive_computation(x):
    global call_count
    call_count += 1
    return x * x

print(expensive_computation(4))
print(expensive_computation(4))   # served from cache, call_count does NOT increase
print("Actual function calls:", call_count)
````

| Code / Item | Explanation |
|---|---|
| `cache[args] = (result, now)` | Stores both the RESULT and the TIME it was computed, keyed by the function's arguments; a later call checks whether the cached entry is still within its TTL (time-to-live) before deciding to reuse it or recompute. |

## 23.3 Redis Concepts

Redis is an in-memory key-value store, commonly used as a shared cache ACROSS multiple server processes/instances (unlike the in-process dict cache in 23.2, which is NOT shared, exactly the same limitation flagged for in-memory rate limiting in Part 11.6). Basic Redis operations map directly onto caching needs: SET key value EX seconds (store with expiration), GET key (retrieve), DEL key (invalidate).

````python
redis_example = '''
import redis
r = redis.Redis(host="localhost", port=6379, db=0)

r.set("user:42:profile", json.dumps({"name": "Ada"}), ex=300)   # cache for 5 minutes
cached = r.get("user:42:profile")
if cached:
    profile = json.loads(cached)
else:
    profile = fetch_from_database(42)
    r.set("user:42:profile", json.dumps(profile), ex=300)
'''
print(redis_example)
````

## 23.4 Cache Keys

A good cache key uniquely and precisely identifies exactly what is cached, typically combining a namespace/resource type, an identifier, and any parameters that affect the result: 'user:42:profile', 'sales:2026-01:by-region', 'prediction:model-v2:hash-of-input-features'. A key that is too COARSE causes incorrect cache hits (returning stale/wrong data for different actual requests); a key that is too FINE-grained defeats the purpose of caching by rarely producing a hit at all.

## 23.5 Expiration

Every cache entry should generally have an expiration (TTL); an entry with NO expiration risks serving increasingly stale data forever if invalidation (23.6) is ever missed or incorrectly implemented, whereas a sensible TTL bounds the maximum staleness automatically, even if explicit invalidation logic has a bug.

## 23.6 Cache Invalidation

Invalidation means proactively removing/updating a cache entry the MOMENT its underlying data changes, rather than waiting for it to expire naturally; this matters when up-to-the-second freshness after a write is important (e.g., a user should see their OWN just-updated profile immediately, not a stale cached copy for the next 5 minutes).

````python
cache_store = {}

def update_user_profile(user_id, new_data):
    # ... perform the actual database update here ...
    cache_key = f"user:{user_id}:profile"
    cache_store.pop(cache_key, None)   # INVALIDATE: remove the now-stale cached entry
    print(f"Invalidated cache for {cache_key}")

update_user_profile(42, {"name": "Ada Updated"})
````

> **Note:** "There are only two hard things in Computer Science: cache invalidation and naming things" is a famous, only-half-joking observation; getting invalidation exactly right (invalidating everything affected, and ONLY what is affected) is a genuinely common source of subtle bugs.

## 23.7 Caching API Responses

````python
from fastapi import FastAPI
import time

cache_app = FastAPI()
response_cache = {}

@cache_app.get("/expensive-report")
def expensive_report():
    cache_key = "expensive-report"
    if cache_key in response_cache:
        cached_result, cached_at = response_cache[cache_key]
        if time.time() - cached_at < 60:
            return {**cached_result, "from_cache": True}
    time.sleep(0.01)   # simulates an expensive computation
    result = {"total": 123456.78, "generated_at": time.time()}
    response_cache[cache_key] = (result, time.time())
    return {**result, "from_cache": False}

from fastapi.testclient import TestClient
cache_client = TestClient(cache_app)
print(cache_client.get("/expensive-report").json()["from_cache"])
print(cache_client.get("/expensive-report").json()["from_cache"])
````

## 23.8 Caching Expensive ML/AI Results

Caching is especially valuable for ML/AI inference: identical inputs to a deterministic model always produce identical outputs, so caching predictions keyed by a hash of the input features (or embedding a query, or an LLM prompt) can eliminate redundant, potentially costly (in time and, for external LLM APIs, in money) model calls entirely for repeated or common inputs.

````python
import hashlib
import json

prediction_cache = {}

def cached_predict(features: dict):
    key = hashlib.sha256(json.dumps(features, sort_keys=True).encode()).hexdigest()
    if key in prediction_cache:
        return prediction_cache[key], True
    # result = real_model.predict(features)  -- the actual, expensive call
    result = {"prediction": 1, "probability": 0.87}   # simulated
    prediction_cache[key] = result
    return result, False

result1, hit1 = cached_predict({"a": 1, "b": 2})
result2, hit2 = cached_predict({"a": 1, "b": 2})
print(result1, "cache hit:", hit1)
print(result2, "cache hit:", hit2)
````

| Code / Item | Explanation |
|---|---|
| `json.dumps(features, sort_keys=True)` | Sorting keys before hashing ensures {"a":1,"b":2} and {"b":2,"a":1} (semantically identical inputs with different key ORDER) produce the SAME cache key, avoiding a missed cache hit due to an irrelevant ordering difference. |

## Part 23 Practice Exercises

**Beginner:** Why does a cache entry generally need an expiration (TTL)?

*Answer:* So that, even if explicit invalidation logic is missed or buggy, the cached data cannot become stale FOREVER; the TTL bounds the maximum possible staleness automatically.

**Intermediate:** Why does an in-process dict-based cache (as in 23.2 and 23.7) fail to work correctly once an API is deployed across multiple server instances, and what is the standard fix?

*Answer:* Exactly like the in-memory rate limiter problem from Part 11.6, an in-process dict lives entirely within ONE server instance's memory; a request served by a different instance has no visibility into another instance's cached entries, so the SAME logical request could be a cache 'miss' on one instance and a 'hit' on another, undermining the cache's effectiveness and potentially causing inconsistent behavior across instances. The standard fix is to use a SHARED, external cache store, most commonly Redis, that every instance reads from and writes to, ensuring a consistent, shared view of what is cached regardless of which specific instance handles any given request.

**Advanced:** Explain why caching is particularly high-leverage for ML/AI prediction endpoints specifically, compared to typical CRUD API endpoints, referencing both cost and latency.

*Answer:* Typical CRUD operations are usually already fast (a simple, indexed database lookup or write), so caching them mainly saves a relatively small amount of database load/latency. ML/AI inference, by contrast, is often SIGNIFICANTLY more expensive per call: a local model's forward pass can be computationally heavy, and calling an external LLM/embedding API adds both real per-call MONETARY cost (many providers charge per request or per token) and meaningfully higher latency (hundreds of milliseconds to several seconds) compared to a typical database query. When the same or a common input recurs (a frequently-asked question to a chatbot, a popular product's feature vector being scored repeatedly), caching the prediction/result keyed by the input eliminates BOTH that latency and that monetary cost entirely for the repeated case, which is a proportionally much larger win than caching a typical fast CRUD read, making caching one of the highest-leverage, lowest-effort optimizations available specifically for ML/AI-serving endpoints.

## 25.2 A Dockerfile for a FastAPI Application

````python
dockerfile_content = '''
FROM python:3.12-slim

WORKDIR /app

COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

COPY . .

EXPOSE 8000

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
'''
print(dockerfile_content)
````

| `EXPOSE 8000` | Documents which port the container listens on (informational; the actual port mapping happens at 'docker run', 25.3). |

## 25.3 Building and Running: Images, Containers, Ports, Environment Variables

|---|---|

## 25.5 Docker Compose: A Complete Python API + Database Example

````python
compose_content = '''
version: "3.9"
services:
  api:
    build: .
    ports:
      - "8000:8000"
    environment:
      DATABASE_URL: postgresql://appuser:secret@db:5432/appdb
    depends_on:
      - db

  db:
    image: postgres:16
    environment:
      POSTGRES_USER: appuser
      POSTGRES_PASSWORD: secret
      POSTGRES_DB: appdb
    volumes:
      - pgdata:/var/lib/postgresql/data
    ports:
      - "5432:5432"

volumes:
  pgdata:
'''
print(compose_content)
````

| Code / Item | Explanation |
|---|---|
| `services: api / db` | Defines TWO separate containers that Compose will build/run TOGETHER, as one coordinated application stack. |
| `DATABASE_URL: postgresql://appuser:secret@db:5432/appdb` | Note the hostname 'db', not 'localhost'; Compose automatically creates an internal network where each SERVICE NAME is resolvable as a hostname by the others. |
| `depends_on: [db]` | Tells Compose to start the db service before the api service (though this only controls START ORDER, not whether the database is actually READY to accept connections yet, which production setups often handle with an explicit healthcheck/retry). |
| `volumes: pgdata:` | Declares a NAMED volume, persisting the database's data across 'docker compose down' and 'docker compose up' cycles. |

## 25.6 Running the Full Stack

````python
compose_commands = '''
docker compose up -d      # build (if needed) and start both services in the background
docker compose logs -f api  # follow logs from just the api service
docker compose down        # stop and remove containers (volumes persist unless -v is added)
'''
print(compose_commands)
````

## Part 25 Practice Exercises

**Beginner:** What is the difference between a Docker image and a Docker container?

*Answer:* An image is a read-only, static blueprint/snapshot (application code, dependencies, base OS layer); a container is a RUNNING instance created FROM an image. Many containers can be run from the same single image, each an independent, isolated instance.

**Intermediate:** In the Dockerfile example, why does it copy requirements.txt and run pip install BEFORE copying the rest of the application code?

*Answer:* Docker builds an image in LAYERS and caches each layer; if a layer's inputs have not changed since the last build, Docker reuses the cached layer instead of re-executing it. By copying ONLY requirements.txt first and installing dependencies before copying the rest of the (more frequently changing) application code, a rebuild triggered by an ordinary code change (which does not touch requirements.txt) can reuse the cached, already-installed-dependencies layer, skipping the often slow pip install step entirely and making rebuilds during development significantly faster.

**Advanced:** In the Docker Compose example, explain why the DATABASE_URL uses the hostname 'db' rather than 'localhost', and what would go wrong if 'localhost' were used instead.

*Answer:* Docker Compose automatically creates an internal, private network shared by all services defined in the same compose file, and registers each service's NAME (here, 'db') as a resolvable hostname on that internal network, pointing to that service's own container. The api service's container and the db service's container are two SEPARATE, isolated containers, each with its OWN 'localhost' that refers only to itself; if the api service's DATABASE_URL used 'localhost', it would be telling the api container to look for a PostgreSQL server running INSIDE ITS OWN container (where none exists), rather than reaching out across the Compose network to the actual db container, causing every database connection attempt to fail with a connection-refused error. Using the service name 'db' correctly routes the connection across Compose's internal network to the actual database container.

## 26.1 Structured Application Logging

````python
import logging
import json

class JSONFormatter(logging.Formatter):
    def format(self, record):
        return json.dumps({
            "level": record.levelname,
            "message": record.getMessage(),
            "logger": record.name,
            "time": self.formatTime(record),
        })

logger = logging.getLogger("api")
logger.setLevel(logging.INFO)
handler = logging.StreamHandler()
handler.setFormatter(JSONFormatter())
logger.addHandler(handler)

logger.info("Application started")
````

## 26.2 Request Logging Middleware

````python
from fastapi import FastAPI, Request
import time

monitor_app = FastAPI()

@monitor_app.middleware("http")
async def log_request_details(request: Request, call_next):
    start = time.time()
    response = await call_next(request)
    duration_ms = round((time.time() - start) * 1000, 2)
    logger.info(f"{request.method} {request.url.path} -> {response.status_code} ({duration_ms}ms)")
    return response

@monitor_app.get("/ping")
def ping():
    return {"pong": True}

from fastapi.testclient import TestClient
monitor_client = TestClient(monitor_app)
monitor_client.get("/ping")
print("Request logged (see above)")
````

## 26.3 Health Checks

````python
@monitor_app.get("/health")
def health():
    checks = {"database": True, "cache": True}   # in real code, actually PING each dependency
    healthy = all(checks.values())
    status_code = 200 if healthy else 503
    from fastapi.responses import JSONResponse
    return JSONResponse(status_code=status_code, content={"status": "ok" if healthy else "degraded", "checks": checks})

response = monitor_client.get("/health")
print(response.status_code, response.json())
````

## 26.4 Metrics: Latency and Error Rates

````python
from collections import defaultdict

request_metrics = defaultdict(lambda: {"count": 0, "total_time": 0.0, "errors": 0})

def record_metric(path, duration, status_code):
    m = request_metrics[path]
    m["count"] += 1
    m["total_time"] += duration
    if status_code >= 500:
        m["errors"] += 1

record_metric("/predict", 0.045, 200)
record_metric("/predict", 0.052, 200)
record_metric("/predict", 0.038, 500)

for path, m in request_metrics.items():
    avg_latency = m["total_time"] / m["count"]
    error_rate = m["errors"] / m["count"]
    print(f"{path}: avg_latency={avg_latency*1000:.1f}ms error_rate={error_rate:.1%}")
````

> **Note:** Production systems typically export metrics like this to a dedicated time-series monitoring system (Prometheus, Datadog, CloudWatch Metrics) rather than an in-process dict, enabling dashboards, alerting (e.g., 'page the on-call engineer if error_rate exceeds 5% for 5 minutes'), and historical trend analysis.

## 26.5 Model Inference Monitoring

For ML APIs (Part 17), monitoring should go beyond generic HTTP metrics to include ML-SPECIFIC signals: prediction latency distribution, prediction/confidence-score distributions over time (to detect DRIFT, a gradual change in the pattern of incoming data or model behavior), and the rate of predictions falling into each output class (a sudden shift can indicate a data pipeline bug or genuine real-world change).

````python
prediction_log = []

def log_prediction(features, prediction, probability, model_version):
    prediction_log.append({
        "prediction": prediction, "probability": probability,
        "model_version": model_version,
    })

log_prediction({}, 1, 0.87, "v1.0.0")
log_prediction({}, 0, 0.62, "v1.0.0")
log_prediction({}, 1, 0.91, "v1.0.0")

avg_confidence = sum(p["probability"] for p in prediction_log) / len(prediction_log)
class_distribution = {c: sum(1 for p in prediction_log if p["prediction"] == c) for c in {0, 1}}
print("Average confidence:", round(avg_confidence, 3))
print("Class distribution:", class_distribution)
````

# Part 27. API Versioning

This part covers why APIs need versions, common versioning schemes, backward compatibility, deprecation, and migration strategies.

## 27.1 Why APIs Need Versions

Once an API has real consumers (a mobile app, a partner's integration, your own frontend), changing its request/response SHAPE can break every client that depends on the old shape; versioning lets you introduce breaking changes in a NEW version while existing clients continue safely using the OLD version until they can migrate on their own schedule.

## 27.2 URL Path Versioning: /v1/, /v2/

````python
from fastapi import FastAPI, APIRouter

versioned_app = FastAPI()

v1_router = APIRouter(prefix="/v1")
v2_router = APIRouter(prefix="/v2")

@v1_router.get("/users/{user_id}")
def get_user_v1(user_id: int):
    return {"id": user_id, "name": "Ada Lovelace"}   # old shape: single 'name' field

@v2_router.get("/users/{user_id}")
def get_user_v2(user_id: int):
    return {"id": user_id, "first_name": "Ada", "last_name": "Lovelace"}   # new shape: split name

versioned_app.include_router(v1_router)
versioned_app.include_router(v2_router)

from fastapi.testclient import TestClient
v_client = TestClient(versioned_app)
print(v_client.get("/v1/users/1").json())
print(v_client.get("/v2/users/1").json())
````

| Code / Item | Explanation |
|---|---|
| `APIRouter(prefix="/v1")` | Groups a set of routes under a common URL prefix; both versions can coexist, be developed, and be DEPRECATED independently, without one breaking the other. |

## 27.3 Other Versioning Schemes (Brief)

| Code / Item | Explanation |
|---|---|
| `URL path (/v1/, /v2/)` | Simple, highly visible, easy to route/cache differently per version; used throughout this handbook's examples. |
| `Header-based (Accept: application/vnd.myapi.v2+json)` | Keeps URLs stable across versions; less visible/discoverable, requires clients to set a specific header. |
| `Query parameter (?version=2)` | Simple to implement; easy to accidentally omit, less conventional than the other two approaches. |

## 27.4 Backward Compatibility

A NON-breaking change (adding a new OPTIONAL field to a response, adding a new endpoint, adding a new optional query parameter) can generally be released WITHOUT bumping the version, since existing clients ignoring the new field/endpoint continue working unaffected. A BREAKING change (removing/renaming a field, changing a field's type or meaning, making a previously-optional parameter required) requires a new version, since it would break clients relying on the old contract.

## 27.5 Deprecation

````python
from fastapi import FastAPI

deprecated_example_app = FastAPI()

@deprecated_example_app.get("/v1/legacy-endpoint", deprecated=True)
def legacy_endpoint():
    return {"message": "This endpoint is deprecated; please migrate to /v2/new-endpoint"}

schema = deprecated_example_app.openapi()
print(schema["paths"]["/v1/legacy-endpoint"]["get"]["deprecated"])
````

| Code / Item | Explanation |
|---|---|
| `deprecated=True` | Marks the endpoint as deprecated directly in the OpenAPI schema, which Swagger UI displays with a visible strikethrough/warning, signaling to consumers that they should migrate away from it. |

## 27.6 Migration Strategies

- Run OLD and NEW versions side by side for an announced transition period (weeks to months, depending on your consumer base), giving clients time to migrate on their own schedule.
- Communicate deprecation clearly: documentation updates, a Deprecation/Sunset HTTP header, direct outreach to known API consumers if feasible.
- Monitor actual USAGE of the deprecated version (via request logging, Part 26) to know when it is safe to actually remove it, rather than guessing.
- Provide a clear migration guide describing exactly what changed between versions and how to adapt client code.

## Part 27 Practice Exercises

**Beginner:** Does adding a new OPTIONAL field to an existing API response generally require a new API version?

*Answer:* No; this is generally considered a non-breaking, backward-compatible change, since existing clients that do not look for the new field are unaffected and continue working exactly as before.

**Intermediate:** Why is /v1/ vs /v2/ URL path versioning described as more 'discoverable' than header-based versioning?

*Answer:* With URL path versioning, the version is directly visible in the URL itself, easy to see in logs, browser address bars, documentation links, and API client code, and easy to test manually (simply visiting a different URL); header-based versioning requires a client to know to set a specific, often less obvious, custom header value, which is easy to overlook, harder to test casually (e.g., a URL cannot simply be pasted into a browser to test a specific version), and less immediately apparent when reading logs or documentation that reference only a path.

**Advanced:** A company's API has a /v1/orders endpoint used by an actively integrated partner, and the team wants to release a breaking change (renaming a field). Describe a complete, responsible migration plan, referencing versioning, deprecation, and monitoring.

*Answer:* First, implement the breaking change as a NEW /v2/orders endpoint (or router, per 27.2), leaving /v1/orders entirely unchanged and fully functional, so the existing partner integration continues working without any disruption at the moment of release. Second, mark /v1/orders as deprecated=True (27.5) in the OpenAPI schema, and separately communicate the deprecation directly to the partner (since a passive schema flag alone may go unnoticed by an already-integrated partner who is not actively re-checking the documentation), including a clear migration guide describing exactly what changed (the field rename) and by when /v1 will be retired. Third, use request logging/monitoring (Part 26) to track actual ongoing usage of /v1/orders over the following weeks/months, watching for the partner's traffic to that endpoint dropping to zero (or near-zero) as they complete their migration to /v2, rather than assuming migration is complete based on a calendar date alone. Only once monitoring confirms /v1/orders usage has genuinely stopped (or after directly confirming with the partner that migration is complete) should the team actually remove the /v1/orders endpoint, minimizing the risk of breaking a partner integration that may not have migrated on the originally assumed schedule.

# Part 28. Webhooks

This part covers webhooks: receiving them, sending them, signature verification, idempotency, retry handling, and event-driven architecture in general.

## 28.1 What Webhooks Are

A webhook is a REVERSED API call: instead of your application POLLING another service repeatedly asking 'has anything happened yet?', that OTHER service proactively sends (POSTs) an HTTP request to a URL YOU provide, the moment a relevant event occurs (a payment succeeded, a file finished processing). This is far more efficient than polling and delivers near-real-time notifications.

## 28.2 Receiving Webhooks

````python
from fastapi import FastAPI, Request, HTTPException

webhook_app = FastAPI()

@webhook_app.post("/webhooks/payment")
async def receive_payment_webhook(request: Request):
    payload = await request.json()
    event_type = payload.get("event")
    if event_type == "payment.succeeded":
        print(f"Payment succeeded: {payload.get('amount')} for order {payload.get('order_id')}")
    elif event_type == "payment.failed":
        print(f"Payment failed for order {payload.get('order_id')}")
    return {"received": True}

from fastapi.testclient import TestClient
webhook_client = TestClient(webhook_app)
response = webhook_client.post("/webhooks/payment", json={"event": "payment.succeeded", "amount": 49.99, "order_id": 101})
print(response.json())
````

> **Note:** A webhook receiver should respond QUICKLY (typically within a few seconds) with a 2xx status to acknowledge receipt; any SLOW processing triggered by the event should be handed off to a background task (Part 15) rather than done inline, since the sending service often has its own timeout and retry logic that assumes a fast acknowledgment.

## 28.3 Signature Verification

Since a webhook endpoint is a PUBLIC URL, anyone could POST a fake payload claiming to be a real event unless the receiver verifies the request genuinely came from the expected sender. This is typically done via an HMAC SIGNATURE: the sender computes a hash of the payload using a SHARED SECRET and includes it in a header; the receiver recomputes the same hash and compares.

````python
import hmac
import hashlib

WEBHOOK_SECRET = "shared-secret-key"

def verify_signature(payload_bytes: bytes, received_signature: str) -> bool:
    expected_signature = hmac.new(WEBHOOK_SECRET.encode(), payload_bytes, hashlib.sha256).hexdigest()
    return hmac.compare_digest(expected_signature, received_signature)

payload_bytes = b'{"event": "payment.succeeded", "amount": 49.99}'
valid_signature = hmac.new(WEBHOOK_SECRET.encode(), payload_bytes, hashlib.sha256).hexdigest()

print(verify_signature(payload_bytes, valid_signature))
print(verify_signature(payload_bytes, "forged-signature"))
````

| Code / Item | Explanation |
|---|---|
| `hmac.compare_digest(...)` | Uses a CONSTANT-TIME comparison rather than a plain '==' string comparison, specifically to prevent a timing attack where an attacker could infer the correct signature byte-by-byte by measuring how long comparisons take. |

## 28.4 Idempotency

Because webhook delivery can be RETRIED by the sender (28.5) if it doesn't receive a timely acknowledgment, a receiver may occasionally receive the SAME event more than once; idempotent handling means processing the same event twice produces the SAME end result as processing it once (e.g., not double-crediting an account for the same payment event).

````python
processed_event_ids = set()

def handle_event_idempotently(event_id, event_data):
    if event_id in processed_event_ids:
        print(f"Event {event_id} already processed; skipping (idempotent)")
        return
    processed_event_ids.add(event_id)
    print(f"Processing event {event_id}: {event_data}")

handle_event_idempotently("evt_123", {"amount": 50})
handle_event_idempotently("evt_123", {"amount": 50})   # duplicate delivery: safely skipped
````

> **Note:** In production, processed_event_ids would be a persistent store (a database table), not an in-memory set, so idempotency survives a server restart.

## 28.5 Sending Webhooks (With Retry Handling)

````python
import time

def send_webhook(url, payload, max_retries=3):
    for attempt in range(1, max_retries + 1):
        try:
            # In real code: response = requests.post(url, json=payload, timeout=5)
            print(f"Attempt {attempt}: sending webhook to {url}")
            success = attempt == 2   # simulate success on the 2nd attempt
            if success:
                print("Webhook delivered successfully")
                return True
            raise ConnectionError("Simulated delivery failure")
        except ConnectionError:
            if attempt == max_retries:
                print("Webhook delivery failed after all retries; giving up")
                return False
            time.sleep(0.01 * attempt)

send_webhook("https://partner.example.com/webhook", {"event": "order.shipped"})
````

## 28.6 Event-Driven Architecture (Brief)

Webhooks are one concrete instance of a broader EVENT-DRIVEN pattern: systems communicate by emitting and reacting to discrete EVENTS rather than through direct, synchronous request/response calls, which decouples producers and consumers (a producer does not need to know WHO is listening, or wait for them to respond), and composes naturally with the background job/queue infrastructure from Part 15 for reliably processing events at scale.

## Part 28 Practice Exercises

**Beginner:** Why should a webhook receiver respond quickly with a 2xx status, even if the actual processing of the event will take longer?

*Answer:* The service SENDING the webhook typically has its own timeout, and will consider the delivery FAILED (potentially triggering a retry, causing duplicate deliveries) if it does not receive a prompt acknowledgment; responding quickly with a 2xx status confirms receipt immediately, while any slower processing should be handed off to a background task so it does not block that prompt acknowledgment.

**Intermediate:** Why is signature verification (28.3) necessary for a webhook endpoint, given that it is typically a public, unauthenticated-looking URL?

*Answer:* Because a webhook endpoint's URL is public and does not, by itself, prove that any given incoming POST request genuinely came from the expected sender, anyone who discovers or guesses the URL could send a FORGED payload claiming to represent a real event (e.g., falsely claiming a payment succeeded), potentially tricking the receiving system into taking an inappropriate action. Signature verification, using a secret shared ONLY between the sender and receiver to compute a verifiable hash of the payload, lets the receiver confirm that a request's payload genuinely originated from (or was at least authorized by) the expected sender, since a forger without the shared secret cannot compute a valid matching signature.

**Advanced:** Explain, using the idempotency example, what could go wrong for a payment-processing webhook receiver that does NOT implement idempotent handling, and describe why simply relying on 'the sender promises to only deliver each event once' would not be a safe design.

*Answer:* Without idempotent handling, if the sender ever redelivers the SAME event (which can legitimately happen due to network issues, a delayed or lost acknowledgment causing the sender to retry per its own retry logic, exactly as illustrated in 28.5), the receiver would process it AGAIN as though it were a brand new, distinct event; for a payment webhook specifically, this could mean crediting a customer's account, or marking an order as paid, or triggering a shipment, TWICE for what was actually a single real payment, a serious, costly business/data-integrity bug. Relying on a sender's promise to deliver 'only once' is unsafe as a design principle because at-least-once delivery (allowing occasional duplicates, but never silently dropping an event) is a MUCH easier and more common reliability guarantee for a distributed sender to actually provide than a true, robust exactly-once guarantee, which is notoriously difficult to achieve correctly across an unreliable network; well-designed webhook systems and their documentation typically explicitly warn receivers to expect occasional duplicate deliveries and to implement idempotent handling themselves, rather than depending on a fragile, hard-to-guarantee promise from the sending side.

## 29.2 Walking Through the Architecture

- API Gateway / Reverse Proxy: terminates HTTPS, load-balances across multiple API instances, may also handle rate limiting (Part 11.6) and basic request validation at the edge.
- Multiple FastAPI instances: horizontal scaling (Part 22.9); any instance can handle any request, since they are STATELESS (no in-process session/cache state that only one instance knows about, per Part 11.6/23.3's lessons).
- Cache (Redis): shared across all instances, absorbing repeat-read load away from the database and expensive computations/model calls (Part 23).
- Job Queue + Workers: offloads slow/background work (Part 15) so API instances stay responsive to new incoming requests.
- Database: the durable source of truth; read replicas can offload read-heavy query load in high-traffic systems, separate from the primary write database.
- ML/AI Services: for heavy models, often deployed as a SEPARATE service (possibly with its own scaling profile, e.g., GPU-backed instances scaled independently from the lightweight API layer).

## 29.3 Scalability Considerations

- Statelessness: API instances should not hold state that only exists in ONE instance's memory (sessions, caches, rate-limit counters, Part 11.6/23.3) so that ANY instance can handle ANY request, enabling horizontal scaling.
- Separate scaling profiles: a lightweight CRUD API and a heavy ML inference service have very different resource needs and should typically scale independently rather than being bundled into the same deployable unit.
- Database as the eventual bottleneck: caching and read replicas delay, but do not eliminate, the need to eventually consider database sharding/partitioning at sufficiently large scale.

## 29.4 Reliability Considerations

- Health checks (Part 26.3) let the load balancer/orchestrator automatically route around unhealthy instances.
- Retries with backoff (Part 8.7) for calls to external dependencies (databases, other APIs, model services) handle TRANSIENT failures gracefully.
- Background job retry mechanisms (Part 15.7) ensure important async work is not silently lost on a transient failure.
- Graceful degradation: e.g., serving slightly stale cached data if the database is briefly unreachable, rather than failing the request entirely, where the business context allows it.

## 29.5 Security Considerations (Holistic Recap)

- HTTPS everywhere (Part 24.6), terminated at the gateway/reverse proxy.
- Authentication and authorization on every non-public endpoint (Part 10).
- Input validation via Pydantic on every endpoint (Part 5, Part 7) as a first line of defense.
- Secrets management (Part 11.9) via environment variables/secrets manager, never hard-coded or committed to version control.
- Rate limiting (Part 11.6) and CORS configuration (Part 11.4) at the gateway or application layer.
- Security event logging (Part 11.10) feeding into monitoring/alerting (Part 26).

## Part 29 Practice Exercises

**Beginner:** In the production architecture diagram, what is the purpose of placing a reverse proxy/API gateway in front of multiple FastAPI instances?

*Answer:* It terminates HTTPS (so individual application instances do not each need to handle TLS directly), and it load-balances incoming requests across the multiple FastAPI instances, distributing traffic and allowing horizontal scaling and resilience (routing around an unhealthy instance).

**Intermediate:** Why must the FastAPI instances in this architecture be STATELESS for the horizontal scaling shown to work correctly?

*Answer:* If an individual instance held state that only IT knew about (an in-memory session store, an in-memory cache, an in-memory rate-limit counter, exactly the pitfalls flagged in Part 11.6 and Part 23.3), then which specific instance a given client's request happened to be routed to would matter, since a request landing on a DIFFERENT instance than a previous request from the same client would not see that state, causing inconsistent behavior (a rate limit not being enforced correctly, a cached value appearing to 'miss' unpredictably, a session appearing lost). For a load balancer to be free to route ANY request to ANY instance interchangeably, which is the entire point of horizontal scaling, every instance must behave identically regardless of which one handles a given request, which requires that any state needed across requests live in a SHARED, external store (the database, Redis) rather than in any individual instance's own memory.

**Advanced:** Explain why the architecture diagram shows ML/AI Services as potentially a SEPARATE deployed service from the main FastAPI application, rather than running heavy model inference directly inside the same API instances that handle ordinary CRUD requests.

*Answer:* Heavy ML model inference often has a fundamentally different resource profile than typical CRUD API work: it may require significant CPU, memory, or even GPU resources per request, and its appropriate SCALING behavior (how many instances/replicas are needed, and when) is driven by inference traffic and computational cost, which can vary independently and very differently from the traffic pattern of ordinary, lightweight CRUD requests (fetching a user profile, listing orders). If heavy model inference ran directly inside the SAME instances handling all other API traffic, a spike in prediction requests could exhaust resources needed for ordinary, unrelated CRUD requests (and vice versa, unrelated CRUD traffic could compete for resources genuinely needed by inference), and scaling decisions would be forced to be a compromise across two very different workloads rather than optimized for either. Deploying ML/AI inference as a SEPARATE service allows it to be scaled independently (e.g., on GPU-backed instances that scale based on inference queue depth or latency), resourced appropriately for its specific computational needs, and potentially deployed/updated (a new model version) on its own release cadence, entirely decoupled from the main API's deployment lifecycle, which is a cleaner, more operationally manageable separation of concerns as a system grows.

## 30.1 Project 1: Customer Management REST API

Business problem: a small business needs to store and manage customer records via an API consumable by an internal tool. Requirements: CRUD on customers, search by name, basic validation. Architecture: single-layer FastAPI + SQLAlchemy (simplest project; later projects add layering).

````python
from fastapi import FastAPI, HTTPException, Depends
from pydantic import BaseModel, EmailStr
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.orm import sessionmaker, declarative_base, Session
from sqlalchemy.pool import StaticPool

p1_engine = create_engine("sqlite:///:memory:", connect_args={"check_same_thread": False}, poolclass=StaticPool)
P1Session = sessionmaker(bind=p1_engine)
P1Base = declarative_base()

class CustomerDB(P1Base):
    __tablename__ = "customers"
    id = Column(Integer, primary_key=True)
    name = Column(String, nullable=False)
    email = Column(String, nullable=False, unique=True)

P1Base.metadata.create_all(p1_engine)

class CustomerIn(BaseModel):
    name: str
    email: str

class CustomerOut(BaseModel):
    id: int
    name: str
    email: str
    model_config = {"from_attributes": True}

def p1_get_db():
    db = P1Session()
    try:
        yield db
    finally:
        db.close()

project1_app = FastAPI(title="Customer Management API")

@project1_app.post("/customers", response_model=CustomerOut, status_code=201)
def create_customer(payload: CustomerIn, db: Session = Depends(p1_get_db)):
    existing = db.query(CustomerDB).filter(CustomerDB.email == payload.email).first()
    if existing:
        raise HTTPException(400, "Email already registered")
    customer = CustomerDB(**payload.model_dump())
    db.add(customer)
    db.commit()
    db.refresh(customer)
    return customer

@project1_app.get("/customers/search", response_model=list[CustomerOut])
def search_customers(q: str, db: Session = Depends(p1_get_db)):
    return db.query(CustomerDB).filter(CustomerDB.name.ilike(f"%{q}%")).all()

from fastapi.testclient import TestClient
p1_client = TestClient(project1_app)
created = p1_client.post("/customers", json={"name": "Ada Lovelace", "email": "ada@example.com"})
print(created.json())
found = p1_client.get("/customers/search?q=ada")
print(found.json())
````

> **Note:** Endpoints: POST /customers, GET /customers/search. Testing: unit tests as in Part 21. Security: input validation via Pydantic; would add auth (Part 10) before real production use. Deployment: Docker + Gunicorn/Uvicorn (Parts 24-25). Performance: index the email column for the uniqueness check at scale (Part 22.5).

## 30.3 Project 3: Financial Transaction API

Business problem: process account transactions with strict correctness guarantees. Requirements: atomic transfers, transaction history, no negative balances. This directly reuses Part 12's transfer_funds() transactional pattern, now with an audit log.

````python
from sqlalchemy import Float, DateTime
import datetime

p3_engine = create_engine("sqlite:///:memory:", connect_args={"check_same_thread": False}, poolclass=StaticPool)
P3Session = sessionmaker(bind=p3_engine)
P3Base = declarative_base()

class AccountP3(P3Base):
    __tablename__ = "accounts_p3"
    id = Column(Integer, primary_key=True)
    balance = Column(Float, default=0.0)

class TransactionLog(P3Base):
    __tablename__ = "transaction_log"
    id = Column(Integer, primary_key=True)
    from_id = Column(Integer)
    to_id = Column(Integer)
    amount = Column(Float)
    timestamp = Column(DateTime, default=datetime.datetime.utcnow)

P3Base.metadata.create_all(p3_engine)
db3 = P3Session()
db3.add_all([AccountP3(id=1, balance=100.0), AccountP3(id=2, balance=0.0)])
db3.commit()

def transfer_with_audit(db, from_id, to_id, amount):
    from_acc = db.query(AccountP3).filter(AccountP3.id == from_id).first()
    to_acc = db.query(AccountP3).filter(AccountP3.id == to_id).first()
    if from_acc.balance < amount:
        raise ValueError("Insufficient funds")
    from_acc.balance -= amount
    to_acc.balance += amount
    db.add(TransactionLog(from_id=from_id, to_id=to_id, amount=amount))
    db.commit()   # balances AND audit log committed together, atomically

transfer_with_audit(db3, 1, 2, 25.0)
print(db3.query(AccountP3).all()[0].balance, db3.query(AccountP3).all()[1].balance)
print(db3.query(TransactionLog).count(), "transaction(s) logged")
````

> **Note:** Endpoints: POST /transfer, GET /accounts/{id}/history. Security: this API MUST have strong authentication/authorization (Part 10) and audit logging (Part 11.10) given its financial nature; every write should be in a database TRANSACTION (Part 12.4). Performance: index from_id/to_id on transaction_log for fast history lookups.

## 30.5 Project 5: Real-Time Fraud Detection API

Business problem: score incoming transactions for fraud risk in real time, low latency, high-stakes decisions. Requirements: sub-100ms scoring, explainable risk factors, rate limiting per merchant.

````python
from fastapi import FastAPI
from pydantic import BaseModel

fraud_app = FastAPI(title="Fraud Detection API")

class TransactionInput(BaseModel):
    amount: float
    merchant_category: str
    is_new_device: bool
    country_mismatch: bool

def score_fraud_risk(txn: TransactionInput) -> tuple[float, list[str]]:
    risk = 0.0
    reasons = []
    if txn.amount > 1000:
        risk += 0.3; reasons.append("high_amount")
    if txn.is_new_device:
        risk += 0.25; reasons.append("new_device")
    if txn.country_mismatch:
        risk += 0.35; reasons.append("country_mismatch")
    return min(risk, 1.0), reasons

@fraud_app.post("/score")
def score_transaction(payload: TransactionInput):
    risk_score, reasons = score_fraud_risk(payload)
    decision = "BLOCK" if risk_score >= 0.6 else ("REVIEW" if risk_score >= 0.3 else "APPROVE")
    return {"risk_score": round(risk_score, 2), "decision": decision, "reasons": reasons}

fraud_client = TestClient(fraud_app)
response = fraud_client.post("/score", json={
    "amount": 1500, "merchant_category": "electronics", "is_new_device": True, "country_mismatch": True,
})
print(response.json())
````

## 30.6 Project 6: AI Document Processing API

Business problem: accept uploaded documents, chunk and process them for downstream search/analysis. Directly builds on Part 18.3's chunking endpoint, adding a processing status workflow via background tasks (Part 15).

````python
from fastapi import FastAPI, BackgroundTasks
from pydantic import BaseModel
import uuid

doc_app2 = FastAPI(title="Document Processing API")
JOB_STATUS = {}

def process_document_job(job_id: str, text: str):
    chunks = [text[i:i+50] for i in range(0, len(text), 50)]
    JOB_STATUS[job_id] = {"status": "complete", "num_chunks": len(chunks)}

class DocumentSubmission(BaseModel):
    text: str

@doc_app2.post("/documents")
def submit_document(payload: DocumentSubmission, background_tasks: BackgroundTasks):
    job_id = str(uuid.uuid4())
    JOB_STATUS[job_id] = {"status": "processing"}
    background_tasks.add_task(process_document_job, job_id, payload.text)
    return {"job_id": job_id, "status": "processing"}

@doc_app2.get("/documents/{job_id}")
def get_job_status(job_id: str):
    return JOB_STATUS.get(job_id, {"status": "not_found"})

doc_client2 = TestClient(doc_app2)
submit_response = doc_client2.post("/documents", json={"text": "Long document text " * 20})
print(submit_response.json())
status_response = doc_client2.get(f"/documents/{submit_response.json()['job_id']}")
print(status_response.json())
````

> **Note:** Endpoints: POST /documents (async submission), GET /documents/{job_id} (poll status). In production, JOB_STATUS would be a database table, and process_document_job would run on a Celery worker (Part 15.4), not FastAPI's in-process BackgroundTasks, for durability at scale.

## 30.7 Project 7: RAG-Based AI API

Business problem: answer questions grounded in a private document collection. Directly extends Part 18.4's minimal RAG endpoint, adding document ingestion and a confidence/no-answer path.

````python
rag_app2 = FastAPI(title="RAG API")
RAG_STORE = {}

class IngestInput(BaseModel):
    doc_id: str
    text: str

@rag_app2.post("/documents/ingest")
def ingest_document(payload: IngestInput):
    RAG_STORE[payload.doc_id] = payload.text
    return {"ingested": payload.doc_id, "total_documents": len(RAG_STORE)}

class QueryInput(BaseModel):
    question: str

@rag_app2.post("/query")
def rag_query2(payload: QueryInput):
    scored = [(doc_id, sum(1 for w in payload.question.lower().split() if w in text.lower()))
              for doc_id, text in RAG_STORE.items()]
    scored.sort(key=lambda x: x[1], reverse=True)
    if not scored or scored[0][1] == 0:
        return {"answer": "I don't have enough information to answer that.", "sources": []}
    best_doc_id = scored[0][0]
    return {"answer": f"Based on {best_doc_id}: {RAG_STORE[best_doc_id][:100]}", "sources": [best_doc_id]}

rag_client2 = TestClient(rag_app2)
rag_client2.post("/documents/ingest", json={"doc_id": "doc1", "text": "FastAPI is great for building ML APIs quickly."})
response = rag_client2.post("/query", json={"question": "What is FastAPI good for?"})
print(response.json())
````

## 30.8 Project 8: Production-Style AI Engineering Backend

This capstone project combines the full production architecture from Part 29: a layered FastAPI application (Part 13) exposing an AI capability (sentiment, Part 18.1), with authentication (Part 10), caching (Part 23.8), rate limiting (Part 11.6), structured logging (Part 26.1), and a health check (Part 26.3), all in one running application.

````python
from fastapi import FastAPI, Depends, HTTPException, Header
import time, hashlib, logging

capstone_app = FastAPI(title="Production AI Backend")
logging.basicConfig(level=logging.INFO)
capstone_logger = logging.getLogger("capstone")

VALID_TOKEN_CAP = "prod-token-abc"
sentiment_cache = {}
rate_limit_tracker = {}

def check_auth(authorization: str = Header(default=None)):
    if authorization != f"Bearer {VALID_TOKEN_CAP}":
        raise HTTPException(401, "Invalid token")

def check_rate_limit(authorization: str = Header(default=None)):
    now = time.time()
    history = [t for t in rate_limit_tracker.get(authorization, []) if now - t < 60]
    if len(history) >= 5:
        raise HTTPException(429, "Rate limit exceeded")
    history.append(now)
    rate_limit_tracker[authorization] = history

class TextInputCap(BaseModel):
    text: str

@capstone_app.post("/v1/sentiment", dependencies=[Depends(check_auth), Depends(check_rate_limit)])
def sentiment_capstone(payload: TextInputCap):
    cache_key = hashlib.sha256(payload.text.encode()).hexdigest()
    if cache_key in sentiment_cache:
        capstone_logger.info("Cache hit for sentiment request")
        return {**sentiment_cache[cache_key], "cached": True}
    label = "positive" if "good" in payload.text.lower() else "neutral"
    result = {"sentiment": label}
    sentiment_cache[cache_key] = result
    capstone_logger.info(f"Computed sentiment: {label}")
    return {**result, "cached": False}

@capstone_app.get("/health")
def health_capstone():
    return {"status": "ok"}

capstone_client = TestClient(capstone_app)
headers = {"Authorization": f"Bearer {VALID_TOKEN_CAP}"}
print(capstone_client.post("/v1/sentiment", json={"text": "good product"}, headers=headers).json())
print(capstone_client.post("/v1/sentiment", json={"text": "good product"}, headers=headers).json())
print(capstone_client.get("/health").json())
````

FastAPI's dependencies=[Depends(check_auth), Depends(check_rate_limit)] list, evaluated in the order given, naturally enforces exactly this correct sequence, since check_auth is listed, and therefore runs, before check_rate_limit, and the route's own business logic (including the cache lookup) only executes after both dependencies have already succeeded.

## Part 31 Practice Exercises

**HTTP:** What is the difference between PUT and PATCH?

*Answer:* PUT replaces a resource ENTIRELY with the provided representation (any fields omitted are typically reset/removed); PATCH applies a PARTIAL update, changing only the fields explicitly provided and leaving the rest unchanged.

**HTTP:** Why is GET considered 'safe' while POST is not?

*Answer:* Safe means the method causes no server-side side effects / data modification; GET is intended purely for retrieval. POST is explicitly intended to create resources or trigger actions, which are side effects, so it is not classified as safe.

**HTTP:** What does the 304 Not Modified status code mean, and when is it used?

*Answer:* It tells the client that the resource has not changed since the client's cached version (as indicated by conditional headers like If-None-Match/If-Modified-Since), so the client should reuse its cached copy; it saves bandwidth by avoiding retransmitting unchanged data.

**REST:** What are the core principles of REST?

*Answer:* Statelessness (each request contains all information needed, no server-side session dependency), a uniform interface (resources identified by URLs, manipulated via standard HTTP methods), client-server separation, cacheability, and a layered system architecture.

**REST:** Why is statelessness considered important for REST API scalability?

*Answer:* Because no request depends on server-side session state tied to a specific server instance, ANY server instance can handle ANY request, which is exactly what enables straightforward horizontal scaling and load balancing (Part 29.3) without needing sticky sessions or shared session state.

**APIs:** What is the difference between a path parameter and a query parameter?

*Answer:* A path parameter identifies a specific resource and is part of the URL's structure itself (e.g., /users/42); a query parameter is an optional key-value pair after '?' used for filtering, sorting, or configuration (e.g., ?active=true).

**APIs:** What is idempotency and which HTTP methods are idempotent?

*Answer:* An idempotent operation produces the same result no matter how many times it is repeated. GET, PUT, DELETE, HEAD, and OPTIONS are idempotent; POST and PATCH generally are not.

**Flask:** How does Flask determine which view function handles a given request?

*Answer:* Via the URL routing table built from @app.route() (or equivalent) decorators, matching the request's URL path (and any URL converters like <int:id>) and HTTP method against the registered routes.

**Flask:** How do you return a JSON response with a custom status code in Flask?

*Answer:* Return a tuple of (body, status_code) from the view function, e.g., 'return jsonify({"error": "..."}), 400'; Flask/jsonify handles serialization and Flask applies the given status code to the response.

**FastAPI:** How does FastAPI achieve automatic request validation without you writing explicit validation code?

*Answer:* By reading the Python type hints (and any Pydantic model types) declared on path operation function parameters, and using Pydantic under the hood to validate and coerce incoming data against those types automatically, raising a 422 error automatically when validation fails.

**FastAPI:** What is the purpose of response_model in a FastAPI route?

*Answer:* It declares the exact shape of data the endpoint promises to return; FastAPI uses it to filter the actual returned data down to only the declared fields, validate the output, and document the response schema in OpenAPI/Swagger.

**FastAPI:** How does FastAPI's dependency injection (Depends()) improve code organization?

*Answer:* It lets you declare reusable, composable pieces of logic (auth checks, database sessions, pagination params) as function parameters that FastAPI resolves and injects automatically, promoting reuse across many routes and making routes easier to test via dependency overrides.

**Django:** What is the main architectural difference between Django and Flask?

*Answer:* Django is a full-featured, opinionated, 'batteries-included' framework bundling an ORM, admin interface, authentication and more; Flask is a minimal microframework providing routing and request handling, leaving most other concerns to your own code or extensions.

**Django:** What does Django REST Framework (DRF) add on top of Django for building APIs?

*Answer:* Serializers (roughly analogous to Pydantic models, for validation/serialization), viewsets and routers for building REST endpoints with less boilerplate, built-in authentication/permission classes, and browsable API documentation.

**Pydantic:** What is the difference between a Pydantic field with a default value and one without?

*Answer:* A field WITHOUT a default value is REQUIRED; omitting it in input raises a validation error. A field WITH a default value is OPTIONAL; if omitted, the default is used automatically.

**Pydantic:** How do you make Pydantic reject any unexpected extra fields in input, rather than silently ignoring them?

*Answer:* Set model_config = {"extra": "forbid"} on the model; by default, Pydantic silently ignores extra fields.

**Authentication:** What is the difference between authentication and authorization?

*Answer:* Authentication verifies WHO a client is; authorization verifies WHAT that already-identified client is permitted to do.

**Authentication:** Why should passwords never be stored in plain text, and what should be used instead?

*Answer:* Storing plain-text passwords means a single database breach exposes every user's actual password directly; a slow, salted hashing algorithm (like bcrypt) should be used instead, storing only the hash, which cannot practically be reversed, and against which a login attempt's plain-text password can be safely checked without ever needing to store or recover the original.

**JWT:** What are the three parts of a JWT, and what does each contain?

*Answer:* Header (algorithm and token type), Payload (the claims: user ID, roles, expiration, etc.), and Signature (a cryptographic signature over the header and payload, used to verify the token has not been tampered with).

**JWT:** Can the payload of a JWT be read by anyone who intercepts it, even without the secret key?

*Answer:* Yes; JWT payloads are only base64-ENCODED, not encrypted, so anyone can decode and read the claims. The secret key is needed only to verify (or forge) the SIGNATURE, not to read the contents, which is why sensitive data should never be placed directly in a JWT payload.

**Async APIs:** Why does async programming benefit I/O-bound API code but not CPU-bound code?

*Answer:* Async's benefit comes from allowing the event loop to run OTHER work while one coroutine is WAITING on I/O (network, disk); CPU-bound code has no such waiting periods to yield during; scaling CPU-bound work requires multiple processes/cores instead.

**Async APIs:** What does asyncio.gather() do, and how does it differ from awaiting several coroutines sequentially?

*Answer:* asyncio.gather() schedules multiple coroutines to run CONCURRENTLY, returning once all have completed, with total wait time closer to the SLOWEST individual coroutine; awaiting them sequentially (one after another) takes the SUM of all their individual durations instead.

**Databases:** What is the N+1 query problem?

*Answer:* A pattern where fetching N parent records triggers ONE additional query PER parent to fetch related data, resulting in 1 + N total queries instead of a small, fixed number; it is fixed using eager loading (e.g., SQLAlchemy's joinedload()).

**Databases:** Why is a database transaction important when performing a multi-step operation like a funds transfer?

*Answer:* A transaction ensures multiple related changes (debiting one account, crediting another) either ALL succeed together or ALL fail together (are rolled back), preventing the database from ever being left in a partially-applied, inconsistent state.

**Deployment:** What is the difference between WSGI and ASGI?

*Answer:* WSGI is the traditional, synchronous interface between Python web apps and servers (used by Flask/traditional Django); ASGI is the modern interface supporting both sync and async code (used by FastAPI/Starlette and modern Django), enabling async request handling, WebSockets, and streaming.

**Deployment:** Why is DEBUG mode always disabled in production?

*Answer:* Debug mode can expose sensitive internal information (stack traces, source snippets, configuration values) directly to users when an error occurs, which is both a security risk and unprofessional; production should log full details server-side only and return generic error responses to clients.

**Docker:** What is the difference between a Docker image and a Docker container?

*Answer:* An image is a read-only, static blueprint; a container is a RUNNING instance created from an image. Many containers can run from one image simultaneously, each isolated.

**Docker:** Why should a Dockerfile copy requirements.txt and install dependencies BEFORE copying the rest of the application code?

*Answer:* Docker caches build layers; installing dependencies before copying frequently-changing application code lets rebuilds reuse the cached dependency-installation layer when only the code (not requirements.txt) has changed, making rebuilds significantly faster.

**ML APIs:** Why should a machine learning prediction endpoint validate input types and ranges before calling the model?

*Answer:* Malformed or nonsensical input can cause the model to crash, throw an unclear internal error, or silently produce a meaningless/garbage prediction; validating input upfront (via Pydantic, Part 17.2) catches these problems with a clear, immediate error before they ever reach the model.

**ML APIs:** Why is it important for a prediction API to return a model_version alongside predictions?

*Answer:* It lets consumers and monitoring systems know exactly which model produced a given prediction, which is essential for debugging a regression after a redeploy, comparing model versions (A/B testing), and auditability.

**AI APIs:** What is the difference between a Python generator being used for streaming and actual network-level streaming?

*Answer:* A Python generator is purely an in-process construct for pausing/resuming a function's execution, producing Python objects; network streaming (chunked transfer encoding, SSE, WebSockets) is a transport-layer mechanism for sending bytes incrementally across a network connection. A generator is commonly used as the Pythonic INTERFACE wrapping an underlying network stream, but the two are conceptually distinct.

**AI APIs:** Why is caching considered especially high-leverage for LLM/embedding API endpoints specifically?

*Answer:* External LLM/embedding calls are typically slower and carry real per-call monetary cost compared to typical database operations; caching results for repeated or common inputs eliminates both that latency and that cost for the repeated case, a proportionally much larger win than caching a fast, free database read.

**API Security:** What is the single most important defense against SQL injection?

*Answer:* Always using parameterized queries (or an ORM that generates them automatically) rather than building SQL strings via direct string formatting/concatenation with user input.

**API Security:** Why is CORS described as a browser-enforced mechanism rather than true server-side authorization?

*Answer:* CORS headers only affect how BROWSERS restrict cross-origin JavaScript requests; non-browser clients (scripts, other servers) are entirely unaffected by CORS and can call an API directly regardless of its CORS configuration, so CORS must never be relied upon as a substitute for real authentication/authorization.

**System Design:** How would you design an API architecture to handle both lightweight CRUD traffic and heavy ML inference traffic?

*Answer:* Deploy them as SEPARATE services with independent scaling profiles (Part 29.2, 29.5's advanced answer): a lightweight FastAPI service for CRUD, horizontally scaled based on request volume, and a separate ML inference service (potentially GPU-backed) scaled based on inference load, communicating over an internal API; this avoids one workload's resource needs interfering with the other's, and allows independent deployment/versioning.

**System Design:** How would you design a rate limiter that works correctly across multiple horizontally-scaled API instances?

*Answer:* Store rate-limit counters in a SHARED, external store (Redis) rather than in any individual instance's local memory, using atomic increment-and-expire operations, so every instance consults and updates the same shared state regardless of which instance handles a given request (Part 11.6's advanced answer).

**Data Scientist Scenario:** You need to expose a pandas-based analytics pipeline as an API for a dashboard team. What are the key design considerations?

*Answer:* Return clean, well-shaped JSON (typically via df.to_dict(orient="records")) rather than raw DataFrame objects; support filtering/date-range query parameters so the dashboard can request exactly the slice it needs rather than always fetching everything; avoid recomputing expensive aggregations on every single request by pre-aggregating on a schedule or caching results (Part 16.7, Part 23); and keep the API read-only/low-risk from a security standpoint, but still consider basic auth/rate limiting if it will ever be reachable outside a trusted internal network.

**ML Engineer Scenario:** A deployed model's prediction endpoint is showing a gradual drop in average confidence scores over the past two weeks. What would you investigate?

*Answer:* This is a classic signal of potential DATA or CONCEPT DRIFT (Part 26.5): investigate whether the DISTRIBUTION of incoming feature values has shifted compared to the training data or historical production traffic (a genuine change in the real-world population, or a bug in an upstream data pipeline feeding malformed/differently-scaled features); check whether a recent upstream change (a schema change in a data source, a new feature engineering step) coincides with when the drop began; and consider whether the underlying real-world relationship the model was trained to predict has itself changed, which would indicate the model genuinely needs retraining on more recent data rather than there being a bug.

**AI Engineer Scenario:** You are building an API that calls an external LLM provider, and users complain about slow response times. What are your main options for improving perceived and actual performance?

*Answer:* Implement streaming (Part 19) so users see tokens appearing progressively rather than waiting for the complete response, dramatically improving PERCEIVED responsiveness even if total generation time is unchanged; cache responses for identical or near-identical prompts where appropriate (Part 23.8) to skip the external call entirely for repeated queries; ensure appropriate timeouts and consider making independent, unrelated LLM calls CONCURRENTLY via async (Part 9) rather than sequentially if a single user-facing request requires multiple model calls; and monitor the external provider's own latency (Part 26.6) to distinguish 'our system is slow' from 'the provider is slow', which require very different fixes.

**HTTP:** What does the Content-Type header indicate, and what happens if it is set incorrectly?

*Answer:* It tells the receiver how to interpret the body's format (e.g., application/json, text/html, multipart/form-data). If set incorrectly (e.g., sending JSON but declaring Content-Type: text/plain), the receiving framework may fail to parse the body correctly, often resulting in the body being treated as raw text or a parsing/validation error.

**HTTP:** What is the purpose of the Accept header?

*Answer:* It tells the server what response format(s) the client can handle (e.g., Accept: application/json), allowing a server that supports multiple formats to choose the one the client prefers via content negotiation.

**REST:** What does HATEOAS mean in the context of REST?

*Answer:* Hypermedia As The Engine Of Application State: a stricter REST principle where responses include links to related actions/resources, letting a client navigate an API dynamically via provided links rather than needing prior, out-of-band knowledge of every URL structure; many practical 'REST' APIs relax this principle for simplicity.

**REST:** Why do REST APIs typically use plural nouns for resource collection endpoints (e.g., /users rather than /user)?

*Answer:* It is a widely adopted convention that reflects the endpoint representing a COLLECTION of resources (GET /users lists many users) versus a specific member of that collection (GET /users/42 for one specific user), making the API's structure more predictable and self-explanatory.

**Flask:** What does the g object in Flask provide?

*Answer:* A per-request namespace for storing data that needs to be shared across different parts of handling a SINGLE request (e.g., a database connection opened once and reused within that request), automatically reset for each new request.

**Flask:** How would you structure a larger Flask application to avoid putting all routes in one file?

*Answer:* Using Flask Blueprints, which let you group related routes into separate modules/files and register them onto the main app, similar in spirit to FastAPI's APIRouter (Part 27.2).

**FastAPI:** What is the difference between a path operation function's parameter with no default and Query(...) explicitly?

*Answer:* A plain parameter with a simple type hint and no default (e.g., 'q: str') is treated as required by default if it is a query parameter for a GET request; Query(...) makes this explicit and additionally allows attaching extra validation constraints (min_length, regex patterns, descriptions) directly to a query parameter.

**FastAPI:** How can you version an OpenAPI-documented FastAPI application without duplicating your entire codebase?

*Answer:* Using separate APIRouter instances with different prefixes (/v1, /v2, Part 27.2), sharing common underlying service/business logic where the versions' behavior has not actually changed, and only duplicating the specific routes/schemas where the versions genuinely differ.

**Django:** What is Django's ORM, and how does it compare conceptually to SQLAlchemy?

*Answer:* Django's ORM is a built-in object-relational mapper, tightly integrated with the rest of the framework (models double as the basis for admin interface and forms); SQLAlchemy is a separate, standalone ORM/toolkit commonly paired with Flask or FastAPI, offering more flexibility and lower-level control at the cost of needing more manual wiring.

**Pydantic:** How does Pydantic handle a field typed as Optional[str] with no default value, versus one with a default of None?

*Answer:* Optional[str] with NO default is still REQUIRED in input, but its value may explicitly be null/None; Optional[str] = None is both optional (can be omitted, defaulting to None) AND nullable. This is a common point of confusion: 'Optional' in the type hint controls what VALUES are acceptable, not whether the field itself can be omitted.

**Pydantic:** What does model_validate() do, and how does it differ from directly calling the model's constructor?

*Answer:* model_validate() validates and constructs a model instance from an arbitrary object (commonly a dict), which is the more explicit, conventional entry point for deserializing external data; calling the constructor directly (Model(**data)) works similarly for a dict unpacked as keyword arguments, but model_validate() is often preferred for clarity and consistency, especially alongside model_validate_json() for raw JSON strings.

**Authentication:** What is the difference between an API key and a JWT for authenticating API requests?

*Answer:* An API key is typically a simple, opaque, long-lived string checked against a stored list/database record (revocable by deleting that record, but requiring a lookup on every request); a JWT is a self-contained, signed token that can be verified WITHOUT a database lookup (faster), but is harder to revoke immediately before its natural expiration, which is why JWTs are typically kept short-lived and paired with refresh tokens (Part 10.6).

**Authentication:** What is role-based access control (RBAC)?

*Answer:* An authorization model where permissions are granted based on a user's assigned ROLE (e.g., admin, editor, viewer) rather than being assigned individually per user, simplifying permission management by letting many users share a role's permission set.

**JWT:** Why should a JWT's expiration (exp claim) always be set, even for an internal, trusted service-to-service token?

*Answer:* Without an expiration, a leaked or compromised token remains valid FOREVER, giving an attacker indefinite access; setting a reasonable expiration bounds the damage window of any token that is ever leaked, regardless of how it happened, which is a basic defense-in-depth practice even for tokens believed to be well-protected.

**Async APIs:** What is a semaphore, and how is it used to limit concurrent API calls?

*Answer:* An asyncio.Semaphore is a synchronization primitive that allows at most N concurrent holders of it at once; wrapping each outbound API call in 'async with semaphore:' limits how many calls are genuinely in-flight simultaneously, protecting a rate-limited external API from being overwhelmed even when many tasks are conceptually launched together via asyncio.gather() (Part 9.7).

**Async APIs:** Why is it important to reuse a single httpx.AsyncClient instance rather than creating a new one for every request?

*Answer:* A client instance maintains a connection pool of already-established connections to hosts it has talked to; reusing it lets subsequent requests to the same host skip the overhead of a fresh TCP/TLS handshake, which is measurably faster when making many requests (Part 9.6).

**Databases:** What is the purpose of an index, and what is its trade-off?

*Answer:* An index dramatically speeds up lookups/filters/joins on the indexed column(s) (closer to O(log n) rather than a full O(n) table scan); the trade-off is that indexes consume additional storage and SLOW DOWN writes (INSERT/UPDATE/DELETE), since the index itself must also be updated, so indexes should be added deliberately based on actual query patterns, not on every column.

**Databases:** Why might a production system use a read replica in addition to its primary database?

*Answer:* A read replica is a copy of the primary database kept in sync, dedicated to serving READ queries, offloading read-heavy traffic away from the primary database so it can focus on handling writes; this improves overall scalability for read-heavy workloads, at the cost of replicas potentially being very slightly behind the primary (replication lag).

**Deployment:** What is the role of a reverse proxy like Nginx in a production deployment?

*Answer:* It sits in front of the application server(s), typically handling TLS/HTTPS termination, serving static files efficiently, load balancing across multiple backend instances, and providing a buffering/security layer between raw internet traffic and the Python application process.

**Deployment:** Why would you run multiple worker processes (e.g., via Gunicorn) rather than a single process for a production API?

*Answer:* A single Python process is limited by the GIL for CPU-bound work and represents a single point of failure; multiple worker processes let the API utilize multiple CPU cores and continue serving requests via other workers even if one worker crashes or is being restarted (Part 22.8, Part 24.4).

**Docker:** What is the purpose of a Docker volume?

*Answer:* It provides persistent storage that survives beyond an individual container's lifetime, essential for data (like a database's files) that must NOT be lost when a container is stopped, removed, or replaced (Part 25.4).

**Docker:** In a Docker Compose file with services 'api' and 'db', why does the api service refer to the database using the hostname 'db' rather than 'localhost'?

*Answer:* Docker Compose creates an internal network where each service's NAME is resolvable as a hostname by other services on that same network; 'localhost' inside the api container refers only to the api container itself, not the separate db container, so using 'db' correctly routes the connection across Compose's internal network (Part 25.5's advanced answer).

**ML APIs:** What is the benefit of a batch prediction endpoint over requiring N separate single-prediction requests?

*Answer:* Batching amortizes network/request overhead across many predictions in one round trip, and for many model types, allows the model's own computation to be VECTORIZED efficiently across the whole batch at once, both of which are typically significantly more efficient than N separate individual calls (Part 17.6).

**AI APIs:** Why might a RAG system return 'I don't have enough information to answer that' instead of always attempting an answer?

*Answer:* If no retrieved document is genuinely relevant to the query, forcing the generation step to answer anyway risks the model HALLUCINATING a plausible-sounding but ungrounded, potentially incorrect answer; explicitly detecting low/no retrieval relevance and returning an honest 'insufficient information' response is a key reliability practice for RAG systems (Part 30.7).

**API Security:** What is a brute-force attack, and what is one practical mitigation?

*Answer:* A brute-force attack repeatedly guesses credentials (passwords, tokens) until one succeeds; a practical mitigation is tracking failed attempts per account/IP and temporarily locking out or introducing increasing delay after a threshold of failures (Part 11.5), alongside strong password hashing and rate limiting.

**API Security:** Why should error messages returned to API clients avoid revealing overly specific internal details?

*Answer:* Overly specific error details (exact database error messages, internal file paths, stack traces) can help an attacker understand your system's internals, technology stack, and potential vulnerabilities; production error responses should be generic and safe for external consumption, while full details are logged server-side only (Part 14.6, Part 24.1).

**System Design:** How would you design an ML API to support safely rolling out a new model version to only a small percentage of traffic first?

*Answer:* Implement a routing layer (in the API or an upstream gateway) that directs a configurable small percentage of incoming prediction requests to the NEW model version while the majority continue to the current, proven version (a canary/gradual rollout), monitoring the new version's prediction distribution, latency, and any available quality/error signals (Part 26.5-26.6) closely before gradually increasing its traffic share, with the ability to instantly roll back to 100% on the old version if problems are detected.

**System Design:** How would you design idempotent handling for a payment webhook receiver to avoid double-processing?

*Answer:* Assign (or extract from the webhook payload) a unique event ID for each event, and before processing, check a persistent store (a database table, not just an in-memory set) for whether that event ID has already been processed; if so, skip processing and simply return a success acknowledgment again; if not, process it and record the event ID as processed as part of the SAME transaction as the processing itself, ensuring the record-as-processed step cannot be lost even if the server crashes immediately after (Part 28.4).

**Data Scientist Scenario:** A stakeholder wants a new KPI added to an existing analytics API endpoint. What steps would you take before deploying the change?

*Answer:* Confirm whether adding the new KPI field is a backward-compatible change (Part 27.4) - adding a new field to a response is generally non-breaking - verify the underlying data/calculation is correct against known values, add or update tests covering the new field (Part 21), and consider whether the added computation meaningfully increases the endpoint's response time, in which case caching or pre-aggregation (Part 16.7, Part 23) might need revisiting.

**ML Engineer Scenario:** Your model inference API's latency has recently doubled. Walk through your diagnostic approach.

*Answer:* Check monitoring/metrics (Part 26.4) to see whether the latency increase correlates with a traffic volume increase (suggesting a scaling/capacity issue) or occurred independent of traffic (suggesting a code, model, or dependency change); check whether a new model version was recently deployed that might simply be more computationally expensive; check for the N+1 query problem or a newly-introduced blocking/synchronous call inside what should be an async request path; and check whether an external dependency (a feature store, a downstream API) the endpoint calls has itself become slower, distinguishing an issue in your own code from a degraded upstream dependency.

**AI Engineer Scenario:** You need to expose both a fast, cheap sentiment-classification model and a slow, expensive LLM-based summarization feature from the same backend. How would you architect this?

*Answer:* Following Part 29's layered production architecture, expose both as separate, clearly-versioned endpoints (e.g., /v1/sentiment and /v1/summarize) but consider deploying their underlying compute with independent scaling profiles: the fast, cheap sentiment model can likely run comfortably within the same lightweight API instances handling ordinary traffic, while the slow, expensive LLM-based summarization endpoint may benefit from being routed to a separate service (Part 29.2), potentially using background processing/streaming (Parts 15, 19) given its longer expected latency, so that heavy summarization load does not degrade the responsiveness of the fast sentiment endpoint sharing the same infrastructure.

**Advanced - HTTP:** Design the request/response shape for an API that supports both cursor-based and offset-based pagination, and explain when you'd choose each.

*Answer:* Offset-based: GET /items?page=2&limit=10, simple, allows jumping to an arbitrary page, but can shift/duplicate results if items are inserted/deleted between requests. Cursor-based: GET /items?after=item_123&limit=10, returning an opaque cursor pointing to the last-seen item, avoids that inconsistency for continuously-changing data (e.g., a live feed) and scales better for very large offsets (a large offset value can be slow to compute in some databases), at the cost of not allowing arbitrary page jumps. Choose offset-based for simple, relatively static or admin-facing paginated views; choose cursor-based for large, frequently-changing datasets or infinite-scroll style consumption.

**Advanced - API Consumption:** Design a resilient API client wrapper class that combines timeouts, retries with exponential backoff, and connection pooling via a single reusable httpx.Client instance.

*Answer:* import httpx, time

class ResilientAPIClient:
    def __init__(self, base_url, timeout=5, max_retries=3):
        self.client = httpx.Client(base_url=base_url, timeout=timeout)
        self.max_retries = max_retries

    def get(self, path, **kwargs):
        for attempt in range(1, self.max_retries + 1):
            try:
                response = self.client.get(path, **kwargs)
                if response.status_code < 500:
                    return response
            except httpx.TransportError:
                pass
            time.sleep(2 ** attempt * 0.1)   # exponential backoff
        raise RuntimeError(f"Failed after {self.max_retries} retries")

    def close(self):
        self.client.close()
The single, reused self.client instance provides connection pooling (Part 9.6) across every call made through this wrapper, while get() layers in bounded retries with exponential backoff (Part 8.7) and respects the per-instance timeout, combining three separate resilience concerns into one reusable abstraction.

**Advanced - Authentication:** Design a complete access-token/refresh-token flow: describe the two endpoints needed and what each returns.

*Answer:* POST /login: accepts username/password, verifies against a hashed password (Part 10.8), and on success returns BOTH a short-lived access_token (e.g., 15-30 min expiry) and a longer-lived refresh_token (stored server-side, associated with the user, so it can be revoked). POST /refresh: accepts a valid, non-revoked refresh_token, verifies it, and if valid, issues a NEW access_token (and often a new refresh_token, rotating it) without requiring the user to re-enter credentials; if the refresh_token is invalid, expired, or has been revoked (e.g., due to logout), it returns 401, forcing a fresh login. This two-endpoint design implements the short-access/long-refresh trade-off from Part 10.6.

**Advanced - Database APIs:** Design a paginated, filterable, sortable SQLAlchemy-backed endpoint for a large 'orders' table, considering the N+1 problem if each order has a related customer.

*Answer:* from sqlalchemy.orm import joinedload

@app.get("/orders")
def list_orders(customer_id: int = None, sort_by: str = "id", page: int = 1, limit: int = 20, db: Session = Depends(get_db)):
    query = db.query(Order).options(joinedload(Order.customer))   # avoids N+1 (Part 22.6)
    if customer_id:
        query = query.filter(Order.customer_id == customer_id)
    query = query.order_by(getattr(Order, sort_by))
    offset = (page - 1) * limit
    return query.offset(offset).limit(limit).all()
joinedload(Order.customer) ensures the related customer for every order in the page is fetched in the SAME query (via a SQL JOIN) rather than one extra query per order, directly avoiding the N+1 problem for what would otherwise be a common, easy-to-miss performance bug at scale.

**Advanced - Async APIs:** Design an async function that fetches data from 3 different external APIs concurrently, but ensures the overall operation fails gracefully (returns partial results) if ONE of them fails.

*Answer:* import httpx, asyncio

async def fetch_safe(client, url):
    try:
        response = await client.get(url, timeout=5)
        return {"url": url, "data": response.json(), "error": None}
    except Exception as e:
        return {"url": url, "data": None, "error": str(e)}

async def fetch_all_safe(urls):
    async with httpx.AsyncClient() as client:
        tasks = [fetch_safe(client, u) for u in urls]
        return await asyncio.gather(*tasks)   # gather() here never raises; each result carries its own error
Wrapping EACH individual fetch in its own try/except (rather than letting asyncio.gather() propagate the first exception it encounters, which is its default behavior) ensures one failing API call does not abort the entire concurrent batch; the caller receives a complete list of results, with failures clearly marked, rather than losing the successful results alongside the one failure.

**Advanced - Debugging:** A production API works correctly under light load but starts returning intermittent 500 errors and timeouts only under heavy concurrent load. Outline your debugging approach.

*Answer:* First, check whether the errors correlate with database connection pool exhaustion (a common cause of exactly this symptom, Part 22.9's related discussion): if the pool's maximum size is smaller than the number of concurrent requests needing a connection, requests queue and eventually time out. Second, check for the N+1 query problem (Part 22.6) becoming severe specifically under load if concurrent requests are each generating many more queries than expected. Third, check whether any SHARED, IN-PROCESS state (an in-memory cache or rate limiter, Part 11.6/23.3) is causing contention or incorrect behavior specifically when many requests hit it concurrently. Fourth, check infrastructure-level resource limits (CPU, memory) on the server instances, since heavy concurrent load might simply be exceeding available capacity for a single instance, pointing toward needing horizontal scaling (Part 22.9, Part 29) rather than a pure code fix.

**Advanced - Security:** Design a comprehensive input-validation and rate-limiting strategy for a public-facing password reset endpoint, considering it is a common target for abuse.

*Answer:* Validate the submitted email format strictly with Pydantic before any processing; apply STRICT rate limiting keyed by BOTH the submitted email address AND the requesting IP address (to prevent both 'many attempts against one account' and 'one IP hammering many accounts'), since password reset endpoints are a classic target for both account-enumeration and abuse; always return the SAME generic response ('If that email exists, a reset link has been sent') regardless of whether the email actually exists in the system, to avoid leaking which emails are registered (an account-enumeration vulnerability); ensure any reset token generated is cryptographically random, single-use, and short-lived; and log all reset attempts (Part 11.10) for security monitoring without logging the plain-text email/token combination in an easily-searchable way that could itself become a liability if logs are compromised.
