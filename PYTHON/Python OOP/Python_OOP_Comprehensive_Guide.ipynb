# 1. Introduction to Object-Oriented Programming

## 1.1 What Is a Programming Paradigm?

A **programming paradigm** is a general style or approach to structuring code and thinking about a problem. It is not a specific language feature, but a way of organizing logic and data. The same problem can be solved in different paradigms, and most modern languages, including Python, let you mix them.

## 1.2 Procedural Programming

Procedural programming organizes code as a sequence of steps and functions that operate on data passed between them. Data and the functions that act on it are kept separate.

````python
def create_account(name, balance):
    return {"name": name, "balance": balance}

def deposit(account, amount):
    account["balance"] += amount
    return account

acc = create_account("Ada", 100)
acc = deposit(acc, 50)
print(acc)
````

**Output:**
```
{'name': 'Ada', 'balance': 150}
```

Here, the account is a plain dictionary, and any function anywhere in the program could reach in and change `"balance"` directly, with no guarantee that changes go through `deposit()`. As a program grows, this becomes hard to control.

## 1.3 Functional Programming

Functional programming emphasizes pure functions (no side effects, same input always gives same output) and treats functions as first-class values. Python supports this style well (it is covered in depth in the separate Functions reference), but it is not the focus of this document.

````python
def apply_interest(balance, rate):
    return balance * (1 + rate)   # pure function: no side effects, returns a new value

print(apply_interest(100, 0.05))
````

**Output:**
```
105.00000000000001
```

## 1.4 Object-Oriented Programming (OOP)

Object-oriented programming organizes code around **objects** that bundle data (**attributes**) together with the behavior that operates on that data (**methods**), into a single unit defined by a **class**.

````python
class Account:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

acc = Account("Ada", 100)
acc.deposit(50)
print(acc.balance)
````

**Output:**
```
150
```

Now `balance` and the logic that is allowed to change it (`deposit`) travel together as one thing. Nothing outside the class needs to know *how* balance is stored internally, only that `acc.deposit(50)` is the correct way to change it.

## 1.5 Why OOP Exists - Problems It Solves

| Problem in Procedural Code | How OOP Addresses It |
|---|---|
| Data and the functions that operate on it are scattered and easy to misuse | Bundling data and behavior into one object (encapsulation) |
| Duplicated logic across similar entities (a "Student" dict here, an "Employee" dict there, both handled by near-identical functions) | Classes let you define shared behavior once and reuse it via inheritance |
| Hard to model real-world entities that have both state and behavior | Objects map naturally onto real things: a `Car`, a `BankAccount`, a `MLModel` |
| Difficult to swap implementations without breaking calling code | Polymorphism lets different objects respond to the same method call in their own way |
| Growing codebases become tangled as more functions operate on the same raw data | Classes provide a controlled, named interface (methods) instead of direct data manipulation everywhere |

## 1.6 Objects and Classes - The Real-World Analogy

Think of a **class** as a *blueprint* and an **object** as a *physical thing built from that blueprint*.

- A blueprint for a house describes what every house of that design will have: a certain number of rooms, a roof, a door. It is not itself a house you can live in.
- Every actual house built from that blueprint is a separate house: painting one blue does not paint the others blue. Each house is a distinct **instance** of the blueprint.

In Python:

````python
class House:              # the blueprint
    def __init__(self, color, rooms):
        self.color = color
        self.rooms = rooms

house1 = House("blue", 3)   # an actual house, built from the blueprint
house2 = House("red", 5)    # a completely separate house, same blueprint

house1.color = "green"       # painting house1 does not affect house2
print(house1.color, house2.color)
````

**Output:**
```
green red
```

## 1.7 Attributes and Methods - State and Behavior

Every object combines two things:

- **State** (data) - stored in **attributes**, e.g. `house1.color`, `house1.rooms`.
- **Behavior** (actions) - implemented as **methods**, e.g. a hypothetical `house1.repaint("yellow")`.

````python
class House:
    def __init__(self, color, rooms):
        self.color = color       # state
        self.rooms = rooms        # state

    def repaint(self, new_color):  # behavior
        print(f"Repainting from {self.color} to {new_color}")
        self.color = new_color

house1 = House("blue", 3)
house1.repaint("yellow")
print(house1.color)
````

**Output:**
```
Repainting from blue to yellow
yellow
```

## 1.8 Real-World Representation Using Objects

OOP is powerful because it maps naturally onto how humans already think about the world: things (nouns) that have properties (adjectives/measurements) and can perform or receive actions (verbs).

| Real-World Entity | Attributes (State) | Methods (Behavior) |
|---|---|---|
| Bank Account | owner, balance, account_number | deposit(), withdraw(), get_statement() |
| Car | make, model, fuel_level, speed | accelerate(), brake(), refuel() |
| Employee | name, salary, department | give_raise(), transfer_department() |
| Machine Learning Model | parameters, training_data, is_trained | train(), predict(), evaluate() |

This "nouns become classes, verbs become methods" heuristic is a useful starting point for designing classes, refined further in Part 27 (Class Design).

# 2. Classes and Objects

## 2.1 What a Class Is

A **class** is a blueprint that defines what attributes and methods its objects will have. Defining a class does not create any object by itself - it only describes the shape objects of that type will take.

````python
class Dog:
    pass    # the simplest possible class - defines a type, but no attributes or methods yet

print(type(Dog))
````

**Output:**
```
<class 'type'>
```

`type(Dog)` shows that `Dog` itself is an instance of `type` - in Python, classes are themselves objects, created by the built-in metaclass `type`. This detail is not needed for everyday use, but it explains why classes can be passed around, stored in variables, and inspected like any other object.

## 2.2 What an Object Is

An **object** (also called an **instance**) is a concrete thing created from a class. Creating an object from a class is called **instantiation**.

````python
class Dog:
    pass

my_dog = Dog()          # instantiation: creates one Dog object
your_dog = Dog()         # a second, completely separate Dog object

print(my_dog)
print(your_dog)
print(my_dog is your_dog)   # different objects in memory
````

**Output:**
```
<__main__.Dog object at 0x1023a4d90>
<__main__.Dog object at 0x1023a4e10>
False
```

Each call to `Dog()` produces a distinct object at a distinct memory address, even though both come from the identical blueprint.

## 2.3 The Class Body

Everything indented under `class ClassName:` is the **class body** - it defines the attributes and methods every object of this class will have access to.

````python
class Dog:
    species = "Canis familiaris"     # a class attribute, shared by all dogs (Part 5)

    def bark(self):                    # an instance method
        return "Woof!"

print(Dog.species)
d = Dog()
print(d.bark())
````

**Output:**
```
Canis familiaris
Woof!
```

## 2.4 Adding Attributes to Objects

The most explicit (though not usually recommended, see `__init__` in Part 3) way to give an object attributes is to assign them directly after creation.

````python
class Dog:
    pass

d = Dog()
d.name = "Rex"
d.age = 3

print(d.name, d.age)
````

**Output:**
```
Rex 3
```

**Important:** attributes assigned this way exist only on that specific object (`d`), not on the `Dog` class itself, and not automatically on any other `Dog` instance.

````python
d2 = Dog()
print(hasattr(d2, "name"))    # False - d2 never got a "name" attribute
````

**Output:**
```
False
```

This is precisely the problem `__init__()` solves, covered next in Part 3: guaranteeing every object of a class starts out with the attributes it needs, instead of relying on someone remembering to set them manually afterward.

## 2.5 Defining and Calling Methods

A method is simply a function defined inside a class body. Its first parameter, by convention named `self`, refers to the specific object the method was called on (explained fully in Part 4).

````python
class Dog:
    def bark(self):
        return "Woof!"

    def introduce(self, name):
        return f"My name is {name} and I say: {self.bark()}"

d = Dog()
print(d.bark())
print(d.introduce("Rex"))
````

**Output:**
```
Woof!
My name is Rex and I say: Woof!
```

| Code Line | Explanation |
|---|---|
| `def bark(self):` | Defines a method; `self` is required as the first parameter even though it is not passed explicitly when calling `d.bark()`. |
| `d.bark()` | Python automatically supplies `d` as `self`, so only the remaining arguments (none, here) need to be given explicitly. |
| `self.bark()` inside `introduce` | A method calling another method on the same object must still go through `self`, since methods are not automatically available as bare names inside the class body. |

## 2.6 Multiple Objects from One Class

The entire value of a class is that many independent objects can be created from it, each with its own attribute values but sharing the same defined behavior.

````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def describe(self):
        return f"{self.name} is {self.age} years old"

dogs = [Dog("Rex", 3), Dog("Bella", 5), Dog("Max", 1)]

for dog in dogs:
    print(dog.describe())
````

**Output:**
```
Rex is 3 years old
Bella is 5 years old
Max is 1 years old
```

## 2.7 Object State Is Independent

Changing one object's attributes never affects another object's attributes, even if both were created from the same class at the same time with the same initial values.

````python
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1

a = Counter()
b = Counter()

a.increment()
a.increment()
b.increment()

print("a:", a.count)
print("b:", b.count)
````

**Output:**
```
a: 2
b: 1
```

## 2.8 Summary Table - Class vs Object Terminology

| Term | Meaning | Example |
|---|---|---|
| Class | The blueprint/template that defines attributes and methods | `class Dog:` |
| Object / Instance | A concrete thing created from a class | `d = Dog()` |
| Instantiation | The act of creating an object from a class | `Dog()` |
| Attribute | A piece of data stored on an object or class | `d.name` |
| Method | A function defined in a class, called on an object | `d.bark()` |
| Class body | The indented block defining a class's contents | everything under `class Dog:` |

# 3. The `__init__()` Method

## 3.1 What `__init__()` Is

`__init__()` is a special ("dunder") method that Python automatically calls immediately after a new object has been created, giving you one guaranteed place to set up that object's initial attributes.

````python
class Dog:
    def __init__(self, name, age):
        print(f"Setting up a new Dog: {name}")
        self.name = name
        self.age = age

d = Dog("Rex", 3)
print(d.name, d.age)
````

**Output:**
```
Setting up a new Dog: Rex
d.name, d.age
```

```
Setting up a new Dog: Rex
Rex 3
```

## 3.2 Important Precision: `__init__()` Is Not the Object Allocator

It is extremely common, and technically inaccurate, to call `__init__()` "the constructor." In Python, object creation is actually a two-step process:

1. **`__new__()`** actually **creates** the raw object in memory and returns it. This is the true allocator.
2. **`__init__()`** then **initializes** that already-existing object - it receives the object as `self` and sets up its attributes. It does not create anything and does not need to return a value (returning anything other than `None` from `__init__` raises a `TypeError`).

````python
class Dog:
    def __new__(cls, *args, **kwargs):
        print("1. __new__ allocates the raw object")
        instance = super().__new__(cls)
        return instance

    def __init__(self, name):
        print("2. __init__ initializes the already-created object")
        self.name = name

d = Dog("Rex")
````

**Output:**
```
1. __new__ allocates the raw object
2. __init__ initializes the already-created object
```

**Precise terminology used throughout this guide:**
- **Instantiation** = the overall process of producing a new object, i.e. `Dog("Rex")`.
- **`__new__`** = allocation of the raw object.
- **`__init__`** = initialization of that object's attributes.
- Calling `__init__` "the constructor" is common informal shorthand, but the accurate statement is: *`__init__` is the initializer; `__new__` is the true constructor, and you rarely need to override it.*

## 3.3 Parameters in `__init__()`, Including Defaults

`__init__()` is an ordinary method in terms of parameter handling: it can take required parameters, default values, `*args`, and `**kwargs`, exactly like any other function (covered in depth in the Functions reference).

````python
class Employee:
    def __init__(self, name, salary=50000, department="General"):
        self.name = name
        self.salary = salary
        self.department = department

e1 = Employee("Ada")
e2 = Employee("Grace", 65000, "Engineering")

print(e1.name, e1.salary, e1.department)
print(e2.name, e2.salary, e2.department)
````

**Output:**
```
Ada 50000 General
Grace 65000 Engineering
```

## 3.4 Multiple Attributes and Validation During Initialization

`__init__()` is the natural place to validate incoming data before it is ever stored on the object, preventing invalid objects from existing in the first place.

````python
class Employee:
    def __init__(self, name, salary):
        if salary < 0:
            raise ValueError("Salary cannot be negative")
        self.name = name
        self.salary = salary

try:
    bad = Employee("Ada", -500)
except ValueError as e:
    print("Error:", e)

good = Employee("Ada", 50000)
print(good.name, good.salary)
````

**Output:**
```
Error: Salary cannot be negative
Ada 50000
```

## 3.5 Common Mistakes with `__init__()`

````python
# MISTAKE 1: forgetting self as the first parameter
class Broken:
    def __init__(name):     # missing self!
        pass

try:
    Broken("Ada")
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
Error: Broken.__init__() takes 1 positional argument but 2 were given
```

````python
# MISTAKE 2: forgetting to assign a parameter to self
class Forgetful:
    def __init__(self, name):
        name    # does nothing useful - never stored!

f = Forgetful("Ada")
try:
    print(f.name)
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
Error: 'Forgetful' object has no attribute 'name'
```

````python
# MISTAKE 3: returning a value from __init__ (not allowed)
class BadReturn:
    def __init__(self):
        return "oops"

try:
    BadReturn()
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
Error: __init__() should return None, not 'str'
```

## 3.6 Summary Table - `__init__()` at a Glance

| Aspect | Detail |
|---|---|
| Type | Dunder (special) method |
| Called automatically? | Yes, immediately after `__new__` creates the object |
| Purpose | Initialize the object's starting attributes |
| Must return | `None` (implicitly, by not using `return` with a value) |
| First parameter | `self` - the newly created object |
| Common name for this concept | "Constructor" (informal); technically the *initializer* |
| True allocator | `__new__` (rarely overridden) |

# 4. Instance Attributes and Instance Methods

## 4.1 Instance Attributes

**Instance attributes** are variables that belong to one specific object, typically set inside `__init__()` using `self.attribute_name = value`. Each object keeps its own independent copy.

````python
class Car:
    def __init__(self, make, model):
        self.make = make          # instance attribute
        self.model = model         # instance attribute

car1 = Car("Toyota", "Corolla")
car2 = Car("Honda", "Civic")

print(car1.make, car1.model)
print(car2.make, car2.model)
````

**Output:**
```
Toyota Corolla
Honda Civic
```

## 4.2 Instance Methods and `self`

An **instance method** is a function defined in the class body that takes `self` as its first parameter, giving it access to the specific object it was called on.

````python
class Car:
    def __init__(self, make, model, fuel=100):
        self.make = make
        self.model = model
        self.fuel = fuel

    def drive(self, distance):
        fuel_needed = distance * 0.1
        if fuel_needed > self.fuel:
            print("Not enough fuel!")
            return
        self.fuel -= fuel_needed
        print(f"Drove {distance} km. Fuel remaining: {self.fuel:.1f}")

car1 = Car("Toyota", "Corolla")
car1.drive(200)
car1.drive(900)
````

**Output:**
```
Drove 200 km. Fuel remaining: 80.0
Not enough fuel!
```

## 4.3 Why `self` Is Used - And Why It Is Not a Keyword

**Precise terminology:** `self` is a **naming convention**, not a reserved word in Python. You could technically name the first parameter of an instance method anything - Python does not enforce the name `self` - but every Python programmer, style guide, and library uses `self`, and deviating from it makes code confusing and non-idiomatic. It is a very strong convention, not a language rule.

````python
class Demo:
    def show(this_object_right_here):        # works, but never do this
        return f"I am {this_object_right_here}"

d = Demo()
print(d.show())
````

**Output:**
```
I am <__main__.Demo object at 0x1023a5250>
```

`self` exists because Python needs an explicit way to know *which* object's attributes a method should read or modify - unlike some languages where "the current object" is implicit. When you write `car1.drive(200)`, Python translates this internally to `Car.drive(car1, 200)`, automatically supplying `car1` as `self`.

````python
# proving self is just the object, passed automatically
Car.drive(car1, 50)     # exactly equivalent to car1.drive(50)
````

## 4.4 Accessing and Modifying Instance Attributes

Attributes are read and written with dot notation, both from outside the class and from methods within it (via `self`).

````python
class Car:
    def __init__(self, make, fuel=100):
        self.make = make
        self.fuel = fuel

    def refuel(self, amount):
        self.fuel += amount

car1 = Car("Toyota")
print(car1.fuel)          # read from outside

car1.refuel(20)            # modify through a method
print(car1.fuel)

car1.fuel = 999            # direct external modification (possible, but often discouraged - see Part 13/14)
print(car1.fuel)
````

**Output:**
```
100
120
999
```

## 4.5 Methods Interacting with Multiple Attributes

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def scale(self, factor):
        self.width *= factor
        self.height *= factor

r = Rectangle(4, 5)
print(r.area(), r.perimeter())

r.scale(2)
print(r.area(), r.perimeter())
````

**Output:**
```
20 18
80 36
```

## 4.6 Independent State Across Multiple Instances

This reinforces Section 2.7 with a more complete example: every instance method call operates only on the `self` it was given, never accidentally reaching into another object.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Insufficient funds for {self.owner}")
            return
        self.balance -= amount

alice = BankAccount("Alice", 100)
bob = BankAccount("Bob", 50)

alice.deposit(200)
bob.withdraw(100)      # fails - Bob doesn't have enough, unaffected by Alice's balance

print(alice.balance, bob.balance)
````

**Output:**
```
Insufficient funds for Bob
300 50
```

| `self` | Refers to the specific object a method was called on; a convention, not a keyword |
| `obj.method(args)` | Python's shorthand for `ClassName.method(obj, args)` |
| Independent state | Every instance keeps its own separate copy of instance attributes |

# 5. Class Attributes and Class Methods

## 5.1 Class Attributes

A **class attribute** is defined directly in the class body (not inside `__init__`) and is shared by every instance of that class, rather than each object keeping its own separate copy.

````python
class Dog:
    species = "Canis familiaris"     # class attribute - shared by ALL dogs

    def __init__(self, name):
        self.name = name              # instance attribute - unique per dog

d1 = Dog("Rex")
d2 = Dog("Bella")

print(d1.species, d2.species)          # both see the same shared value
print(Dog.species)                      # accessible on the class itself, too
````

**Output:**
```
Canis familiaris Canis familiaris
Canis familiaris
```

## 5.2 Instance Attributes vs Class Attributes - Lookup Order

When you access `obj.name`, Python first looks for `name` on the object itself; only if it is not found there does it fall back to looking on the class.

````python
class Dog:
    species = "Canis familiaris"

d1 = Dog()
print(d1.species)              # found on the class - "Canis familiaris"

d1.species = "Wolf (mutant)"   # this creates a NEW instance attribute, shadowing the class one
print(d1.species)               # found on the instance now
print(Dog.species)              # the class attribute itself is untouched

d2 = Dog()
print(d2.species)                # d2 still sees the original class attribute
````

**Output:**
```
Canis familiaris
Wolf (mutant)
Canis familiaris
Canis familiaris
```

| Code Line | Explanation |
|---|---|
| `d1.species = "Wolf (mutant)"` | This does **not** modify `Dog.species`. It creates a brand-new instance attribute on `d1` that happens to share the same name, which then shadows the class attribute for `d1` only. |
| `Dog.species` (after the assignment) | Still shows the original value, proving the class attribute itself was never touched. |
| `d2.species` | A different object with no instance attribute of that name, so it still falls back to the class attribute. |

**The one case where this bites people:** mutable class attributes.

````python
class Team:
    members = []          # DANGER: a single list shared by every Team instance

    def add_member(self, name):
        self.members.append(name)     # mutates the SHARED list, not a per-instance one

t1 = Team()
t2 = Team()

t1.add_member("Ada")
print(t1.members, t2.members)          # t2 is affected too - not what most people expect!
````

**Output:**
```
['Ada'] ['Ada']
```

**Fix:** initialize mutable attributes inside `__init__` so each instance gets its own.

````python
class Team:
    def __init__(self):
        self.members = []       # now a fresh list per instance

    def add_member(self, name):
        self.members.append(name)

t1, t2 = Team(), Team()
t1.add_member("Ada")
print(t1.members, t2.members)
````

**Output:**
```
['Ada'] []
```

## 5.3 Class Methods and `cls`

A **class method** receives the class itself (conventionally `cls`) instead of an instance, and is defined with the `@classmethod` decorator. Like `self`, `cls` is a convention, not a reserved word.

````python
class Dog:
    population = 0

    def __init__(self, name):
        self.name = name
        Dog.population += 1

    @classmethod
    def get_population(cls):
        return cls.population

Dog("Rex")
Dog("Bella")
print(Dog.get_population())
````

**Output:**
```
2
```

## 5.4 Class Methods as Alternative Constructors

The most common and practically important use of `@classmethod` is providing alternative ways to build an object, beyond the default `__init__` signature.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def from_csv_row(cls, row):
        name, salary = row.split(",")
        return cls(name, float(salary))       # cls(...) works correctly even for subclasses

    @classmethod
    def entry_level(cls, name):
        return cls(name, 40000)

e1 = Employee.from_csv_row("Ada,55000")
e2 = Employee.entry_level("Grace")

print(e1.name, e1.salary)
print(e2.name, e2.salary)
````

**Output:**
```
Ada 55000.0
Grace 40000
```

**Why `cls(...)` instead of `Employee(...)`:** using `cls` means that if a subclass inherits `from_csv_row`, it will correctly build an instance of the subclass, not always an `Employee`, demonstrated in Part 8.

## 5.5 Class Methods That Modify Class State

````python
class Configuration:
    debug_mode = False

    @classmethod
    def enable_debug(cls):
        cls.debug_mode = True

Configuration.enable_debug()
print(Configuration.debug_mode)

c = Configuration()
print(c.debug_mode)      # instances see the shared class-level change too
````

**Output:**
```
True
True
```

## 5.6 Class Methods vs Instance Methods - Practical Use Cases

| Use Case | Instance Method | Class Method |
|---|---|---|
| Operate on one specific object's data | `account.deposit(50)` | - |
| Build an object in a non-default way | - | `Employee.from_csv_row(row)` |
| Track/modify data shared across all instances | Possible but awkward | `Dog.get_population()` |
| Provide a "factory" of preset configurations | - | `Pizza.margherita()`, `Pizza.pepperoni()` |

## 5.7 Summary Table

| Concept | Syntax | First Parameter | Shared Across Instances? | Typical Use |
|---|---|---|---|---|
| Instance attribute | `self.x = value` in `__init__` | n/a | No, one copy per object | Object-specific data |
| Class attribute | `x = value` in class body | n/a | Yes, one copy for the whole class | Shared defaults/counters |
| Instance method | `def method(self):` | `self` (the instance) | n/a | Behavior tied to one object |
| Class method | `@classmethod` `def method(cls):` | `cls` (the class) | n/a | Alternative constructors, class-wide operations |

# 6. Static Methods

## 6.1 What `@staticmethod` Is

A **static method** is defined inside a class using the `@staticmethod` decorator, and receives **neither** `self` nor `cls` automatically. It behaves exactly like a plain, ordinary function, but is grouped inside the class's namespace because it is logically related to that class.

````python
class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(c):
        return c * 9 / 5 + 32

    @staticmethod
    def fahrenheit_to_celsius(f):
        return (f - 32) * 5 / 9

print(TemperatureConverter.celsius_to_fahrenheit(100))
print(TemperatureConverter.fahrenheit_to_celsius(32))
````

**Output:**
```
212.0
0.0
```

## 6.2 Why Static Methods Exist

Static methods exist for **organizational grouping**: logic that is conceptually related to a class but does not need to read or modify any instance state (`self`) or class state (`cls`). Without `@staticmethod`, such a function would either need to live outside the class (losing the logical grouping/namespacing) or would need to accept an unused `self`/`cls` parameter purely to satisfy Python's method syntax.

````python
class Validator:
    @staticmethod
    def is_valid_email(email):
        return "@" in email and "." in email.split("@")[-1]

print(Validator.is_valid_email("ada@example.com"))
print(Validator.is_valid_email("not-an-email"))
````

**Output:**
```
True
False
```

## 6.3 Static Methods vs Instance Methods vs Class Methods

````python
class Demo:
    class_data = "shared"

    def __init__(self, value):
        self.value = value

    def instance_method(self):
        return f"self={self.value}, class_data={self.class_data}"

    @classmethod
    def class_method(cls):
        return f"cls={cls.__name__}, class_data={cls.class_data}"

    @staticmethod
    def static_method(x, y):
        return x + y     # no access to self or cls at all

d = Demo("hello")
print(d.instance_method())     # needs an instance to run meaningfully
print(Demo.class_method())      # can run on the class directly
print(Demo.static_method(2, 3)) # just a regular function, namespaced under Demo
````

**Output:**
```
self=hello, class_data=shared
cls=Demo, class_data=shared
5
```

## 6.4 When to Use Static Methods

- The logic is clearly related to the class's purpose (e.g. a validation helper, a unit conversion, a formatting utility).
- The function does **not** need any instance-specific data (`self`) and does **not** need to build instances or read/modify class-level data (`cls`).
- You want the function to be callable both on the class itself and on any instance, without needing an instance to exist first.

## 6.5 When Not to Use Static Methods

- If the logic needs `self`, it should be an instance method instead.
- If the logic needs to build new instances (an alternative constructor) or read/modify class-wide state, it should be a `@classmethod` instead.
- If the function has nothing conceptually to do with the class, it likely belongs as a standalone function in a module, not namespaced inside the class at all - forcing everything into classes when it does not need to be there is a common overuse, discussed further in Part 29 (Common Mistakes).

````python
# Slight overuse example - this doesn't need to be in a class at all:
class MathHelpers:
    @staticmethod
    def add(a, b):
        return a + b

# Just as valid, arguably clearer, as a plain module-level function:
def add(a, b):
    return a + b
````

## 6.6 Comparison Table

| Feature | Instance Method | Class Method | Static Method |
|---|---|---|---|
| Decorator | none | `@classmethod` | `@staticmethod` |
| First parameter | `self` | `cls` | none |
| Access to instance state | Yes | No | No |
| Access to class state | Yes (indirectly) | Yes | No (unless the class is named explicitly) |
| Can be called on the class without an instance | No (needs an instance, or an explicit instance argument) | Yes | Yes |
| Typical use | Object-specific behavior | Alternative constructors, class-wide operations | Grouped utility logic unrelated to instance/class state |
| Example | `acc.deposit(50)` | `Employee.from_csv_row(row)` | `Validator.is_valid_email(e)` |

# 7. The Four Core Principles of OOP

Object-oriented programming rests on four foundational principles. Python supports all four, though - true to its philosophy of flexibility over strict enforcement - it implements some of them (especially encapsulation) through convention rather than hard language restrictions.

## 7.1 Encapsulation

**Meaning:** bundling data (attributes) and the behavior that operates on it (methods) into a single unit (a class), and restricting direct, uncontrolled access to some of that data from outside the class.

### Public Attributes

By default, every attribute and method in Python is **public** - accessible from anywhere, with no restriction.

````python
class Account:
    def __init__(self, balance):
        self.balance = balance     # public - anyone can read or overwrite it directly

a = Account(100)
a.balance = -500                    # nothing stops this, even though it's nonsensical
print(a.balance)
````

**Output:**
```
-500
```

### Naming Conventions: `_attribute` and `__attribute`

Python has no true "private" keyword like `private` in Java or C++. Instead, it uses **naming conventions** backed by one real language mechanism (name mangling, for double underscores only):

| Convention | Example | Meaning | Enforced by Python? |
|---|---|---|---|
| No underscore | `self.balance` | Fully public; part of the intended external interface | Not enforced at all |
| Single leading underscore | `self._balance` | "Protected by convention" - a signal to other developers: "internal, please don't touch this directly, even though you technically can" | Not enforced at all - purely a convention |
| Double leading underscore | `self.__balance` | Triggers **name mangling**, making accidental external access and accidental subclass collisions much less likely | Partially enforced via name mangling (see below) |
| Leading and trailing double underscore | `self.__init__`, `__len__` | Reserved for Python's own special/dunder methods and attributes - **never** invent your own `__name__`-style identifiers | Not a privacy mechanism at all - a naming reservation |

````python
class Account:
    def __init__(self, balance):
        self._balance = balance          # protected by convention

    def get_balance(self):
        return self._balance

a = Account(100)
print(a._balance)          # still technically accessible - Python trusts the developer
````

**Output:**
```
100
```

### Name Mangling

A double-leading-underscore attribute (e.g. `self.__balance`) is automatically renamed by Python to `_ClassName__balance` internally. This is called **name mangling**, and its real purpose is to avoid accidental attribute collisions in inheritance hierarchies, not to create bulletproof privacy.

````python
class Account:
    def __init__(self, balance):
        self.__balance = balance     # name-mangled to _Account__balance

    def get_balance(self):
        return self.__balance

a = Account(100)

try:
    print(a.__balance)               # fails - this exact name doesn't exist
except AttributeError as e:
    print("Error:", e)

print(a._Account__balance)            # but it's still reachable via the mangled name
print(a.get_balance())
````

**Output:**
```
Error: 'Account' object has no attribute '__balance'
100
100
```

**Precise takeaway:** Python does not have traditional strict private variables the way Java or C++ do. Name mangling makes accidental access harder and prevents naming clashes in subclasses, but any sufficiently determined caller can still reach `_ClassName__attribute` directly. Python's culture favors *trusting developers with clearly signposted conventions* over the compiler-enforced privacy found in some other languages.

### Encapsulation Best Practices in Python

- Use plain public attributes for simple data with no invariants to protect.
- Use a single leading underscore (`_attribute`) to signal "internal use only" for attributes/methods not meant to be part of the public API.
- Use double leading underscore (`__attribute`) sparingly, mainly to avoid subclass naming collisions, not as your primary privacy mechanism.
- Use `@property` (Part 14) when you need validation or computed values while keeping a clean, attribute-like public interface.

## 7.2 Abstraction

**Meaning:** exposing only the essential, relevant details of an object's interface while hiding the complexity of its internal implementation.

### Why Abstraction Matters

A driver uses a car's steering wheel and pedals (the *interface*) without needing to understand the engine's internal combustion process (the *implementation*). Good class design works the same way: callers of a class should be able to use it correctly through a small, clear set of public methods, without needing to understand what happens inside them.

````python
class EmailService:
    def send(self, to, message):
        self._connect_to_server()
        self._authenticate()
        self._transmit(to, message)
        self._disconnect()

    def _connect_to_server(self):
        print("Connecting...")

    def _authenticate(self):
        print("Authenticating...")

    def _transmit(self, to, message):
        print(f"Sending to {to}: {message}")

    def _disconnect(self):
        print("Disconnecting...")

service = EmailService()
service.send("ada@example.com", "Hello!")   # caller only needs to know about send()
````

**Output:**
```
Connecting...
Authenticating...
Sending to ada@example.com: Hello!
Disconnecting...
```

### Abstract Classes and `@abstractmethod`

Python formalizes abstraction with the `abc` module: an **abstract class** cannot be instantiated directly, and can declare **abstract methods** that every concrete subclass is required to implement.

````python
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    @abstractmethod
    def process_payment(self, amount):
        """Every subclass MUST implement this."""

    def receipt(self, amount):              # a normal, non-abstract shared method
        return f"Processed payment of ${amount}"

class CreditCardProcessor(PaymentProcessor):
    def process_payment(self, amount):
        return f"Charging ${amount} to credit card"

try:
    p = PaymentProcessor()          # cannot instantiate - has an unimplemented abstract method
except TypeError as e:
    print("Error:", e)

cc = CreditCardProcessor()
print(cc.process_payment(100))
print(cc.receipt(100))
````

**Output:**
```
Error: Can't instantiate abstract class PaymentProcessor with abstract method process_payment
Charging $100 to credit card
Processed payment of $100
```

Abstraction and abstract classes are treated in full depth in Part 17.

## 7.3 Inheritance

**Meaning:** a class (the **child** or **derived** class) can acquire the attributes and methods of another class (the **parent** or **base** class), reusing and optionally extending or overriding its behavior.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating"

class Dog(Animal):             # Dog inherits from Animal
    def bark(self):
        return f"{self.name} says Woof!"

d = Dog("Rex")
print(d.eat())     # inherited from Animal, works with no extra code
print(d.bark())     # new behavior defined on Dog itself
````

**Output:**
```
Rex is eating
Rex says Woof!
```

## 7.4 Polymorphism

**Meaning:** "many forms" - different classes can respond to the *same* method call, each in its own appropriate way, letting calling code treat different objects uniformly.

### Method Overriding as Polymorphism

````python
class Animal:
    def speak(self):
        return "Some generic sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.speak())     # same method call, different behavior per actual type
````

**Output:**
```
Woof!
Meow!
Some generic sound
```

### Duck Typing

Python's polymorphism goes further than requiring a shared parent class: because Python checks behavior at runtime rather than enforcing strict types at compile time, **any** object with the right method will work, regardless of its class hierarchy. This is called **duck typing**: "if it walks like a duck and quacks like a duck, treat it as a duck."

````python
class Dog:
    def speak(self):
        return "Woof!"

class Car:
    def speak(self):           # completely unrelated to Dog, no shared parent class
        return "Vroom!"

def make_it_speak(thing):
    print(thing.speak())        # works on ANY object that happens to have .speak()

make_it_speak(Dog())
make_it_speak(Car())
````

**Output:**
```
Woof!
Vroom!
```

**Precise terminology:** "compile-time polymorphism" (like Java/C++ method overloading, choosing between multiple same-named methods based on argument types before the program runs) does not meaningfully exist in Python, because Python is dynamically typed and resolves method calls at runtime. Python supports **runtime, duck-typed polymorphism** through method overriding and shared interfaces, not compile-time overload resolution. This distinction is expanded in Part 21.

## 7.5 Summary Table - The Four Pillars

| Principle | Core Idea | Primary Python Mechanism |
|---|---|---|
| Encapsulation | Bundle data + behavior; control access to internal state | Classes, `_`/`__` naming conventions, `@property` |
| Abstraction | Expose essential interface, hide implementation detail | Public methods, `abc.ABC`, `@abstractmethod` |
| Inheritance | Reuse and extend behavior from a parent class | `class Child(Parent):`, `super()` |
| Polymorphism | Same call, different behavior depending on the actual object | Method overriding, duck typing |

# 8. Inheritance

## 8.1 Basic Inheritance

A child class inherits every attribute and method its parent defines, unless the child overrides them.

````python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def start(self):
        return f"{self.make} {self.model} is starting"

class Car(Vehicle):
    pass          # inherits everything, adds nothing new yet

c = Car("Toyota", "Corolla")
print(c.start())
````

**Output:**
```
Toyota Corolla is starting
```

## 8.2 Extending a Parent Class

A child class typically adds new attributes and methods on top of what it inherits.

````python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def start(self):
        return f"{self.make} {self.model} is starting"

class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)     # reuse parent's setup (Part 9 covers super() fully)
        self.num_doors = num_doors

    def honk(self):                        # new method, not present on Vehicle
        return "Beep beep!"

c = Car("Toyota", "Corolla", 4)
print(c.start())     # inherited
print(c.honk())       # new
print(c.num_doors)    # new attribute
````

**Output:**
```
Toyota Corolla is starting
Beep beep!
4
```

**Hierarchical inheritance**: multiple child classes inherit from the *same* single parent class.

````python
class Vehicle:
    def start(self):
        return "Vehicle starting"

class Car(Vehicle):
    pass

class Truck(Vehicle):
    pass

class Motorcycle(Vehicle):
    pass

for v in [Car(), Truck(), Motorcycle()]:
    print(type(v).__name__, "->", v.start())
````

# 9. `super()`

## 9.1 What `super()` Does

`super()` gives you a proxy object that lets you call methods defined on the **next class in the Method Resolution Order** (usually the parent class), without hardcoding that parent's name directly.

````python
class Animal:
    def speak(self):
        return "Some generic sound"

class Dog(Animal):
    def speak(self):
        original = super().speak()      # call Animal's version of speak()
        return f"{original}, specifically: Woof!"

print(Dog().speak())
````

**Output:**
```
Some generic sound, specifically: Woof!
```

## 9.2 Calling the Parent's `__init__()`

The single most common use of `super()` is reusing a parent class's `__init__` logic instead of duplicating it.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)      # reuse Employee's setup
        self.team_size = team_size            # add Manager-specific setup

m = Manager("Ada", 90000, 5)
print(m.name, m.salary, m.team_size)
````

**Output:**
```
Ada 90000 5
```

**Without `super()`**, you would have to duplicate the parent's initialization logic, which then has to be kept manually in sync if the parent ever changes:

````python
class Manager(Employee):
    def __init__(self, name, salary, team_size):
        self.name = name          # duplicated logic - fragile if Employee.__init__ changes later
        self.salary = salary
        self.team_size = team_size
````

## 9.3 Extending, Not Just Replacing, Inherited Behavior

`super()` lets a child class **extend** a parent's method - running the parent's version and then adding more - rather than completely replacing it (which plain overriding without `super()` would do).

````python
class Shape:
    def describe(self):
        return "This is a shape"

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def describe(self):
        base = super().describe()                 # extend, don't just replace
        return f"{base}, specifically a circle with radius {self.radius}"

print(Circle(5).describe())
````

**Output:**
```
This is a shape, specifically a circle with radius 5
```

## 9.4 `super()` with Multiple Inheritance and MRO

In multiple inheritance, `super()` does not necessarily mean "my direct parent" - it means "the next class after me in the MRO," which can be a sibling class rather than a traditional parent. This is explored fully in Part 11, but a preview:

````python
class Base:
    def greet(self):
        return "Base"

class Left(Base):
    def greet(self):
        return f"Left -> {super().greet()}"

class Right(Base):
    def greet(self):
        return f"Right -> {super().greet()}"

class Child(Left, Right):
    def greet(self):
        return f"Child -> {super().greet()}"

print(Child().greet())
print(Child.__mro__)
````

**Output:**
```
Child -> Left -> Right -> Base
(<class '__main__.Child'>, <class '__main__.Left'>, <class '__main__.Right'>, <class '__main__.Base'>, <class 'object'>)
```

Notice that `Left.greet`'s `super().greet()` call reaches **`Right`**, not `Base` directly - because `super()` always follows the MRO of the *actual* object's class (`Child`), not a fixed relationship hardcoded into `Left`. This cooperative behavior is what makes `super()` work correctly across complex multiple-inheritance hierarchies.

## 9.5 Common Mistakes with `super()`

````python
# MISTAKE 1: calling the parent class directly instead of using super()
class Employee:
    def __init__(self, name):
        self.name = name

class Manager(Employee):
    def __init__(self, name, team_size):
        Employee.__init__(self, name)   # works, but bypasses MRO - fragile with multiple inheritance
        self.team_size = team_size
````

````python
# MISTAKE 2: forgetting to call super().__init__() at all
class Employee:
    def __init__(self, name):
        self.name = name

class Manager(Employee):
    def __init__(self, name, team_size):
        self.team_size = team_size    # forgot super().__init__(name) !

m = Manager("Ada", 5)
try:
    print(m.name)          # never set - AttributeError
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
Error: 'Manager' object has no attribute 'name'
```

````python
# MISTAKE 3: passing self explicitly to super() (unnecessary in Python 3)
class Employee:
    def __init__(self, name):
        self.name = name

class Manager(Employee):
    def __init__(self, name, team_size):
        super(Manager, self).__init__(name)   # valid, but the Python-2-style long form is unnecessary
        # super().__init__(name)                # this Python-3 shorthand is preferred
        self.team_size = team_size
````

## 9.6 Summary Table

| Use | Syntax | Notes |
|---|---|---|
| Call parent's `__init__` | `super().__init__(args)` | Most common use; avoids duplicating setup logic |
| Call parent's other method | `super().method_name(args)` | Extends rather than fully replaces inherited behavior |
| Python 3 shorthand | `super()` | Preferred; automatically infers the class and instance |
| Python 2 style (still valid) | `super(ClassName, self)` | Verbose, rarely needed in modern Python |
| In multiple inheritance | Follows the MRO, not just "my direct parent" | Enables cooperative multi-class method chains |

# 10. Method Overriding

## 10.1 What Overriding Means

**Method overriding** happens when a child class defines a method with the same name as one it inherited from a parent class. The child's version takes precedence for instances of the child class.

````python
class Employee:
    def calculate_bonus(self):
        return 1000

class SalesEmployee(Employee):
    def calculate_bonus(self):        # overrides Employee's version
        return 2500

print(Employee().calculate_bonus())
print(SalesEmployee().calculate_bonus())
````

**Output:**
```
1000
2500
```

## 10.2 Why Overriding Is Useful

Overriding lets different subclasses share a common interface (the same method name and general purpose) while each providing behavior appropriate to its specific case - this is precisely what enables polymorphism (Part 7.4, Part 21).

````python
class Employee:
    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary

    def calculate_bonus(self):
        return self.base_salary * 0.05

class SalesEmployee(Employee):
    def calculate_bonus(self):
        return self.base_salary * 0.15     # sales gets a bigger bonus percentage

class Intern(Employee):
    def calculate_bonus(self):
        return 0                             # interns get no bonus

employees = [
    Employee("Ada", 60000),
    SalesEmployee("Grace", 55000),
    Intern("Sam", 30000),
]

for e in employees:
    print(f"{e.name}: {e.calculate_bonus()}")
````

**Output:**
```
Ada: 3000.0
Grace: 8250.0
Sam: 0
```

## 10.3 Overriding Completely vs Extending with `super()`

Overriding a method can either **completely replace** the parent's behavior, or **extend** it by calling `super()` first (Part 9). Both are valid depending on intent.

````python
class Logger:
    def log(self, message):
        print(f"[LOG] {message}")

class TimestampedLogger(Logger):
    def log(self, message):                  # EXTENDS - still uses the parent's formatting
        super().log(f"2026-01-01 - {message}")

class SilentLogger(Logger):
    def log(self, message):                   # REPLACES completely - ignores parent behavior
        pass

TimestampedLogger().log("Server started")
SilentLogger().log("This will not print anything")
````

**Output:**
```
[LOG] 2026-01-01 - Server started
```

## 10.4 Overriding `__init__()` - A Special, Very Common Case

Overriding `__init__()` is so common it deserves explicit attention: without `super().__init__()`, the parent's initialization logic is silently skipped (as shown as a mistake in Part 9.5).

````python
class Shape:
    def __init__(self, color):
        self.color = color

class Square(Shape):
    def __init__(self, color, side):
        super().__init__(color)          # ensures Shape's setup still runs
        self.side = side

s = Square("red", 4)
print(s.color, s.side)
````

**Output:**
```
red 4
```

## 10.5 Overriding and Signature Changes

Python does not enforce that an overriding method must keep the exact same parameter signature as the parent's version - this is a source of subtle bugs if calling code assumes a consistent interface across a class hierarchy.

````python
class Shape:
    def area(self):
        raise NotImplementedError

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self, unit="sq_m"):       # added an extra parameter - still technically "overriding" area()
        value = self.width * self.height
        return f"{value} {unit}"

print(Circle(3).area())
print(Rectangle(4, 5).area())
````

**Output:**
```
28.27431
20 sq_m
```

**Best practice:** keep overridden method signatures consistent with the parent's, unless there is a strong, deliberate reason not to - inconsistent signatures undermine the polymorphic guarantee that "any `Shape` can be treated the same way."

## 10.6 Common Mistakes with Overriding

````python
# MISTAKE: typo in the method name creates a NEW method instead of overriding
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def Speak(self):        # capital S - this does NOT override speak()!
        return "Woof!"

d = Dog()
print(d.speak())    # still calls Animal's version - the typo silently created a sibling method
print(d.Speak())
````

**Output:**
```
Some sound
Woof!
```

| Concept | Description |
|---|---|
| Overriding | Redefining a method with the identical name in a subclass |
| Complete override | New method body has no call to `super()`; entirely replaces parent behavior |
| Extending override | New method body calls `super().method()`, then adds more |
| Enables polymorphism | Same method name, different behavior per subclass, callable uniformly |
| Common mistake | A typo or case mismatch in the method name creates an unrelated new method rather than overriding |

# 11. Method Resolution Order (MRO)

## 11.1 What MRO Is and Why Python Needs It

The **Method Resolution Order** is the precise, well-defined sequence Python follows when searching for a method or attribute across a class hierarchy. It matters most whenever a class has more than one ancestor that could plausibly supply the same method name - Python needs one unambiguous rule to decide which version actually runs.

````python
class A:
    def greet(self):
        return "A"

class B(A):
    def greet(self):
        return "B"

class C(A):
    def greet(self):
        return "C"

class D(B, C):
    pass          # doesn't define greet itself - which ancestor's greet() does it get?

print(D().greet())
````

**Output:**
```
B
```

## 11.2 C3 Linearization

Python computes MRO using an algorithm called **C3 linearization**. Without needing to implement it by hand, the practical rules it guarantees are:

1. A subclass is always checked **before** any of its parents.
2. If a class inherits from multiple parents, those parents are checked in the **order they are listed** in the class definition (left to right).
3. A class only ever appears **once** in the final MRO, no matter how many paths lead to it (this is what correctly resolves the "diamond problem," covered in Part 12).
4. The resulting order is consistent for every class in the hierarchy - no class can be linearized in a way that contradicts its own parents' individual orderings.

## 11.3 Inspecting MRO: `ClassName.__mro__` and `ClassName.mro()`

````python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

print(D.__mro__)          # a tuple - the canonical way to view MRO
print(D.mro())              # a list - equivalent information, callable as a method
````

**Output:**
```
(<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)
[<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>]
```

## 11.4 MRO with the Diamond Shape


```
      A
     / \
    B   C
     \ /
      D
```

````python
class A:
    def identify(self):
        return "A"

class B(A):
    def identify(self):
        return "B"

class C(A):
    def identify(self):
        return "C"

class D(B, C):
    pass

print(D.__mro__)
print(D().identify())     # follows MRO: D has none, so checks B next -> found
````

**Output:**
```
(<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)
B
```

Crucially, `A` appears only **once** in the MRO, even though both `B` and `C` inherit from it - this is exactly what C3 linearization guarantees, and exactly what prevents `A`'s method (or `__init__`) from accidentally running twice in a cooperative `super()` chain.

## 11.5 How `super()` Follows MRO

`super()` does not mean "my direct parent class" - it means "whatever comes immediately after the current class in the *actual instance's* MRO." This is why cooperative multiple inheritance with `super()` works correctly even though each class was written without knowing exactly what it would eventually be combined with.

````python
class A:
    def show(self):
        print("A.show")

class B(A):
    def show(self):
        print("B.show start")
        super().show()
        print("B.show end")

class C(A):
    def show(self):
        print("C.show start")
        super().show()
        print("C.show end")

class D(B, C):
    def show(self):
        print("D.show start")
        super().show()
        print("D.show end")

D().show()
print(D.__mro__)
````

**Output:**
```
D.show start
B.show start
C.show start
A.show
C.show end
B.show end
D.show end
(<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)
```

Notice `B.show`'s `super().show()` call reaches `C`, not `A` directly - because in `D`'s MRO, `C` comes right after `B`. Every class's `super()` call is resolved relative to the **full MRO of the actual object's class**, not relative to that class's own, narrower view of its ancestors. This cooperative chaining, where every class in the chain calls `super()` exactly once, is what allows all four classes' logic to run exactly once, in a well-defined order.

## 11.6 When MRO Cannot Be Computed - `TypeError`

If the parent classes are listed in an order that creates a contradiction (an inconsistency C3 linearization cannot resolve), Python raises a `TypeError` at class definition time rather than guessing.

````python
class X:
    pass

class Y(X):
    pass

try:
    class Z(X, Y):     # contradiction: X is listed before Y, but Y already requires X to come after it
        pass
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
Error: Cannot create a consistent method resolution order (MRO) for bases X, Y
```

## 11.7 Summary Table

| Concept | Detail |
|---|---|
| MRO | The precise order Python searches classes for a method/attribute |
| Algorithm | C3 linearization |
| Guarantee 1 | Subclasses are checked before their parents |
| Guarantee 2 | Multiple parents are checked left to right, as listed |
| Guarantee 3 | Each class appears exactly once, even in diamond hierarchies |
| Inspect via | `ClassName.__mro__` (tuple) or `ClassName.mro()` (list) |
| `super()` relationship | Always resolves to "the next class after the current one in the actual object's MRO" |
| Unsolvable case | Raises `TypeError` at class definition time |

# 12. Multiple Inheritance

## 12.1 Multiple Parent Classes

A Python class can inherit from more than one parent class simultaneously, gaining attributes and methods from all of them.

````python
class Flyable:
    def fly(self):
        return "Flying through the air"

class Swimmable:
    def swim(self):
        return "Swimming through water"

class Duck(Flyable, Swimmable):
    def quack(self):
        return "Quack!"

d = Duck()
print(d.fly())
print(d.swim())
print(d.quack())
````

**Output:**
```
Flying through the air
Swimming through water
Quack!
```

## 12.2 Advantages of Multiple Inheritance

- Combines independent, reusable behaviors ("mixins") without forcing an artificial single-parent hierarchy.
- Avoids duplicating shared logic across otherwise unrelated classes.

````python
class JSONSerializableMixin:
    def to_json(self):
        import json
        return json.dumps(self.__dict__)

class LoggableMixin:
    def log(self, message):
        print(f"[{type(self).__name__}] {message}")

class User(JSONSerializableMixin, LoggableMixin):
    def __init__(self, name, age):
        self.name = name
        self.age = age

u = User("Ada", 28)
print(u.to_json())
u.log("User created")
````

**Output:**
```
{"name": "Ada", "age": 28}
[User] User created
```

## 12.3 Risks of Multiple Inheritance

- **Ambiguity**: if two parents define the same method name, MRO decides which one wins, which may not be immediately obvious to readers of the code (Part 11).
- **Fragility**: changes to one parent class can unexpectedly affect any class that mixes it in.
- **Complexity**: deep or wide multiple-inheritance hierarchies become difficult to reason about and debug.
- **The diamond problem**: covered next.

## 12.4 The Diamond Inheritance Problem

The **diamond problem** occurs when two parent classes share a common ancestor, and a child class inherits from both parents.


```
        Animal
       /      \
     Bird     Mammal
       \      /
       Platypus
```

````python
class Animal:
    def __init__(self, name):
        self.name = name
        print(f"Animal.__init__ for {name}")

class Bird(Animal):
    def __init__(self, name):
        super().__init__(name)
        print("Bird.__init__")

class Mammal(Animal):
    def __init__(self, name):
        super().__init__(name)
        print("Mammal.__init__")

class Platypus(Bird, Mammal):
    def __init__(self, name):
        super().__init__(name)
        print("Platypus.__init__")

p = Platypus("Perry")
print(Platypus.__mro__)
````

**Output:**
```
Animal.__init__ for Perry
Mammal.__init__
Bird.__init__
Platypus.__init__
(<class '__main__.Platypus'>, <class '__main__.Bird'>, <class '__main__.Mammal'>, <class '__main__.Animal'>, <class 'object'>)
```

Because every class in this chain calls `super().__init__()` exactly once, and Python's MRO ensures `Animal.__init__` runs only **once** (not once via `Bird` and once via `Mammal`), the diamond is resolved cleanly with no duplicated initialization. This cooperative pattern - every `__init__` calling `super().__init__()` - is what makes diamond inheritance safe in Python.

**What would go wrong without cooperative `super()` calls:**

````python
class Animal:
    def __init__(self, name):
        self.name = name
        print("Animal.__init__")

class Bird(Animal):
    def __init__(self, name):
        Animal.__init__(self, name)     # hardcoded parent call, bypasses MRO
        print("Bird.__init__")

class Mammal(Animal):
    def __init__(self, name):
        Animal.__init__(self, name)     # hardcoded parent call, bypasses MRO
        print("Mammal.__init__")

class Platypus(Bird, Mammal):
    def __init__(self, name):
        Bird.__init__(self, name)
        Mammal.__init__(self, name)      # Animal.__init__ now runs TWICE
        print("Platypus.__init__")

Platypus("Perry")
````

**Output:**
```
Animal.__init__
Bird.__init__
Animal.__init__
Mammal.__init__
Platypus.__init__
```

## 12.5 Best Practices for Multiple Inheritance

- Prefer small, focused **mixin classes** (like `JSONSerializableMixin` above) that add one specific, self-contained piece of behavior, rather than deep hierarchies of "real" classes with heavy state.
- Always use `super()` cooperatively (never call a parent's method directly by name) in any class that might participate in multiple inheritance.
- Keep an eye on `__mro__` when designing non-trivial hierarchies, and write a quick test confirming it matches your expectations.
- Consider whether composition (Part 18) would express the relationship more simply before reaching for multiple inheritance - "has-a" is often easier to reason about than a complex "is-a-and-a" hierarchy.

## 12.6 Summary Table

| Concept | Detail |
|---|---|
| Multiple inheritance | `class Child(ParentA, ParentB):` |
| Advantage | Combine independent behaviors (mixins) without an artificial single hierarchy |
| Risk | Ambiguous method resolution, fragile coupling, harder-to-follow hierarchies |
| Diamond problem | Two parents share a common ancestor; naive direct calls can run the ancestor's method twice |
| Python's solution | MRO (C3 linearization) plus cooperative `super()` calls ensure each class runs exactly once |
| Best practice | Use focused mixins, always call `super()`, inspect `__mro__`, consider composition as an alternative |

# 13. Encapsulation and Access Control

This section consolidates and extends the naming-convention material introduced in Part 7.1 into a complete, precise reference on Python's approach to access control.

## 13.1 Python's Philosophy: Convention Over Enforcement

Many languages (Java, C++, C#) enforce access control at compile time with keywords like `private`, `protected`, and `public`. **Python does not have these keywords at all.** Every attribute is technically accessible from anywhere. Instead, Python relies on naming conventions, backed by one genuine (but limited) mechanism - name mangling - and a strong cultural norm: "we are all consenting adults here."

## 13.2 The Three Naming Patterns, Precisely Distinguished

````python
class Example:
    def __init__(self):
        self.public = "anyone can use this"
        self._protected = "internal use - please don't touch, but you technically can"
        self.__private = "name-mangled - harder to reach by accident"

e = Example()
print(e.public)
print(e._protected)
print(e._Example__private)     # the real, mangled name
````

**Output:**
```
anyone can use this
internal use - please don't touch, but you technically can
name-mangled - harder to reach by accident
```

| Pattern | Example | Meaning | Real Enforcement? |
|---|---|---|---|
| `name` | `self.balance` | Fully public - part of the intended external API | None |
| `_name` | `self._balance` | "Protected by convention" - internal implementation detail, but still directly reachable | None - purely a signal to other developers |
| `__name` | `self.__balance` | Triggers name mangling to `_ClassName__balance` | Partial - avoids accidental access/collisions, not true privacy |
| `__name__` | `self.__init__`, `obj.__dict__` | Reserved exclusively for Python's own dunder methods/attributes | N/A - not a privacy mechanism; never invent your own |

**Critical distinction - `__name` vs `__name__`:** a leading-and-trailing double underscore (`__init__`, `__len__`, `__str__`) is Python's own reserved dunder syntax for special methods and attributes; it triggers **no** name mangling. Only a **leading-only** double underscore (`__balance`, with no trailing double underscore) triggers name mangling. Confusing these two is a common beginner mistake.

````python
class Demo:
    def __init__(self):
        self.__x = 1        # name-mangled to _Demo__x (leading double underscore ONLY)
        self.__y__ = 2        # NOT mangled - looks dunder-like, but Python leaves it alone

d = Demo()
print(dir(d))
````

**Output:**
```
['_Demo__x', '__class__', ..., '__y__', ...]
```

## 13.3 Name Mangling in Depth

Name mangling exists primarily to prevent **accidental collisions** in inheritance hierarchies - not to create airtight privacy.

````python
class Base:
    def __init__(self):
        self.__secret = "Base's secret"      # mangled to _Base__secret

    def reveal(self):
        return self.__secret                  # Python rewrites this to self._Base__secret too

class Derived(Base):
    def __init__(self):
        super().__init__()
        self.__secret = "Derived's secret"     # mangled to _Derived__secret - a DIFFERENT attribute!

d = Derived()
print(d.reveal())              # still returns Base's secret - no collision occurred
print(d._Base__secret)
print(d._Derived__secret)
````

**Output:**
```
Base's secret
Base's secret
Derived's secret
```

Without name mangling, `Derived.__secret` would silently overwrite `Base.__secret` (since both would just be named `self.__secret` in memory), corrupting `Base`'s internal state. Name mangling automatically keeps them as two separate attributes, `_Base__secret` and `_Derived__secret`.

## 13.4 Why Python Does Not Have Traditional Strict Private Variables

This is a deliberate design philosophy, not an oversight. Python's community strongly favors trusting developers, with clear signals (`_`/`__`) about intent, over compiler-enforced restrictions that can be worked around anyway in most languages (e.g. via reflection) and that add friction during legitimate use cases like testing, debugging, and monkey-patching.

````python
class Account:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

a = Account(100)
# "Private" in name only - still reachable if you really want to (and know you probably shouldn't):
a._Account__balance = 99999
print(a.get_balance())
````

**Output:**
```
99999
```

## 13.5 Practical Best Practices

- Default to public attributes for simple data with no invariants worth protecting.
- Use `_protected` to mark internal implementation details not meant for external use - this is by far the most common and idiomatic convention in real Python code.
- Reserve `__private` (name-mangled) mainly for attributes at risk of being accidentally overwritten in a subclass, not as a general-purpose privacy tool.
- When you need validation, computed values, or read-only external access, prefer `@property` (Part 14) over manual `get_x()`/`set_x()` methods - it keeps the clean `obj.attribute` syntax while still enforcing rules.

## 13.6 Summary Table

| Convention | Syntax | Enforcement | Typical Use |
|---|---|---|---|
| Public | `self.name` | None | Default choice; simple, unrestricted data |
| Protected (by convention) | `self._name` | None (developer trust only) | Internal implementation detail |
| Private (name-mangled) | `self.__name` | Partial (name mangling) | Avoiding subclass attribute collisions |
| Dunder (reserved) | `self.__name__` | N/A - reserved for Python itself | Never define your own; only implement Python's existing dunders |

# 14. Properties

## 14.1 The Problem `@property` Solves

Without properties, enforcing validation or computing a derived value requires explicit getter/setter methods, which breaks Python's natural, clean `obj.attribute` access style.

````python
# Without @property - the "manual getter/setter" style
class TemperatureManual:
    def __init__(self, celsius):
        self._celsius = celsius

    def get_celsius(self):
        return self._celsius

    def set_celsius(self, value):
        if value < -273.15:
            raise ValueError("Below absolute zero")
        self._celsius = value

t = TemperatureManual(25)
t.set_celsius(30)          # awkward - not plain attribute syntax
print(t.get_celsius())
````

**Output:**
```
30
```

## 14.2 `@property`, Getter, Setter, and Deleter

````python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):                 # the GETTER
        return self._celsius

    @celsius.setter
    def celsius(self, value):           # the SETTER
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is not possible")
        self._celsius = value

    @celsius.deleter
    def celsius(self):                   # the DELETER
        print("Deleting celsius value")
        del self._celsius

t = Temperature(25)
print(t.celsius)          # looks like plain attribute access - calls the getter

t.celsius = 30              # looks like plain assignment - calls the setter
print(t.celsius)

try:
    t.celsius = -300          # setter validation runs automatically
except ValueError as e:
    print("Error:", e)

del t.celsius                # calls the deleter
````

**Output:**
```
25
30
Error: Temperature below absolute zero is not possible
Deleting celsius value
```

## 14.3 Read-Only Properties (Computed Attributes)

A property with only a getter defined (no `@x.setter`) is effectively **read-only** from outside the class - attempting to assign to it raises `AttributeError`.

````python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):                     # computed attribute - no stored value at all
        return 3.14159 * self.radius ** 2

    @property
    def diameter(self):
        return self.radius * 2

c = Circle(5)
print(c.area)          # computed on the fly, every time it's accessed
print(c.diameter)

try:
    c.area = 100         # no setter defined - fails
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
78.53975
10
Error: can't set attribute
```

## 14.4 Validation Using Properties

Properties are the idiomatic Python way to enforce invariants while keeping a clean public interface.

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance      # goes through the setter below, even in __init__

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative")
        self._balance = value

acc = BankAccount(100)
acc.balance = 200
print(acc.balance)

try:
    acc.balance = -50
except ValueError as e:
    print("Error:", e)

try:
    bad = BankAccount(-10)     # validation runs even during construction
except ValueError as e:
    print("Error:", e)
````

**Output:**
```
200
Error: Balance cannot be negative
Error: Balance cannot be negative
```

| Code Line | Explanation |
|---|---|
| `self.balance = balance` in `__init__` | Because `balance` is a property, this line actually calls the **setter**, not a plain attribute assignment - so validation applies from the very first moment the object is created. |
| `@balance.setter` | Must use the same method name (`balance`) as the `@property` getter; the decorator `@balance.setter` only exists after `balance` has already been defined as a property. |

## 14.5 Encapsulation with Properties

Properties let you change how data is stored *internally* at any time, without breaking any code that already uses the public attribute-style interface - a core encapsulation benefit.

````python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):                    # computed from two other attributes
        return f"{self.first_name} {self.last_name}"

    @full_name.setter
    def full_name(self, value):               # allows "reverse" assignment too
        first, last = value.split(" ", 1)
        self.first_name = first
        self.last_name = last

p = Person("Ada", "Lovelace")
print(p.full_name)

p.full_name = "Grace Hopper"       # sets first_name and last_name via the setter
print(p.first_name, p.last_name)
````

**Output:**
```
Ada Lovelace
Grace Hopper
```

## 14.6 Summary Table

| Decorator | Purpose | Triggered By |
|---|---|---|
| `@property` | Defines the getter | `obj.name` (read access) |
| `@name.setter` | Defines the setter | `obj.name = value` (write access) |
| `@name.deleter` | Defines the deleter | `del obj.name` |
| Getter only, no setter | Read-only computed attribute | Assignment raises `AttributeError` |

# 15. Dunder / Special Methods

Dunder ("double underscore") methods let a custom class plug into Python's own built-in syntax and functions. You almost never call them directly with dot notation; instead, you define them, and Python calls them automatically when its syntax is used on your object. Throughout this section, the emphasis is on the **connection between everyday syntax and the dunder method Python invokes underneath it**.

## 15.1 Object Creation and Initialization: `__new__`, `__init__`

````python
class Point:
    def __new__(cls, *args, **kwargs):
        print("__new__: allocating the raw object")
        return super().__new__(cls)

    def __init__(self, x, y):
        print("__init__: initializing attributes")
        self.x = x
        self.y = y

p = Point(1, 2)     # Point(1, 2) triggers __new__ first, then __init__
````

**Output:**
```
__new__: allocating the raw object
__init__: initializing attributes
```

## 15.2 String Representation: `__str__` and `__repr__`

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __str__(self):
        return f"({self.x}, {self.y})"                # triggered by str(obj), print(obj)

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"         # triggered by repr(obj), the REPL, containers

p = Point(1, 2)
print(p)             # -> calls __str__
print(str(p))         # -> calls __str__
print(repr(p))         # -> calls __repr__
print([p])              # -> lists always use __repr__ for their elements
````

**Output:**
```
(1, 2)
(1, 2)
Point(x=1, y=2)
[Point(x=1, y=2)]
```

## 15.3 Length: `__len__`

````python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)          # triggered by len(obj)

p = Playlist(["Song A", "Song B", "Song C"])
print(len(p))
````

**Output:**
```
3
```

## 15.4 Truthiness: `__bool__`

````python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __bool__(self):
        return len(self.songs) > 0        # triggered by bool(obj), if obj:, while obj:

empty = Playlist([])
full = Playlist(["Song A"])

print(bool(empty), bool(full))
if full:
    print("Playlist has songs")
if not empty:
    print("Playlist is empty")
````

**Output:**
```
False True
Playlist has songs
Playlist is empty
```

**Note:** if `__bool__` is not defined, Python falls back to `__len__` (an object with `len() == 0` is falsy); if neither is defined, every instance is considered truthy by default.

## 15.5 Equality and Ordering: `__eq__`, `__ne__`, `__lt__`, `__le__`, `__gt__`, `__ge__`

````python
class Money:
    def __init__(self, cents):
        self.cents = cents

    def __eq__(self, other):        # triggered by ==
        return self.cents == other.cents

    def __ne__(self, other):         # triggered by != (Python derives this automatically from __eq__ since 3.x, but can be defined explicitly)
        return not self.__eq__(other)

    def __lt__(self, other):          # triggered by <
        return self.cents < other.cents

    def __le__(self, other):           # triggered by <=
        return self.cents <= other.cents

    def __gt__(self, other):            # triggered by >
        return self.cents > other.cents

    def __ge__(self, other):             # triggered by >=
        return self.cents >= other.cents

    def __repr__(self):
        return f"Money({self.cents})"

a, b = Money(500), Money(750)
print(a == b, a != b, a < b, a <= b, a > b, a >= b)
print(sorted([b, a]))    # sorted() relies on __lt__ internally
````

**Output:**
```
False True True True False False
[Money(500), Money(750)]
```

**Note on `functools.total_ordering`:** if you define `__eq__` plus just one of the ordering methods (commonly `__lt__`), the `@functools.total_ordering` class decorator can generate the rest automatically, reducing boilerplate.

## 15.6 Arithmetic Operators: `__add__`, `__sub__`, `__mul__`, `__truediv__`, `__floordiv__`, `__mod__`, `__pow__`

````python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):        # triggered by +
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):         # triggered by -
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):         # triggered by *
        return Vector(self.x * scalar, self.y * scalar)

    def __truediv__(self, scalar):      # triggered by /
        return Vector(self.x / scalar, self.y / scalar)

    def __floordiv__(self, scalar):      # triggered by //
        return Vector(self.x // scalar, self.y // scalar)

    def __mod__(self, scalar):            # triggered by %
        return Vector(self.x % scalar, self.y % scalar)

    def __pow__(self, power):              # triggered by **
        return Vector(self.x ** power, self.y ** power)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1, v2 = Vector(10, 20), Vector(3, 4)
print(v1 + v2)
print(v1 - v2)
print(v1 * 2)
print(v1 / 2)
print(v1 // 3)
print(v1 % 3)
print(v2 ** 2)
````

**Output:**
```
Vector(13, 24)
Vector(7, 16)
Vector(20, 40)
Vector(5.0, 10.0)
Vector(3, 6)
Vector(1, 2)
Vector(9, 16)
```

## 15.7 Container Behavior: `__getitem__`, `__setitem__`, `__delitem__`, `__contains__`

````python
class Roster:
    def __init__(self):
        self._students = []

    def add(self, name):
        self._students.append(name)

    def __getitem__(self, index):         # triggered by obj[index]
        return self._students[index]

    def __setitem__(self, index, value):    # triggered by obj[index] = value
        self._students[index] = value

    def __delitem__(self, index):            # triggered by del obj[index]
        del self._students[index]

    def __contains__(self, name):             # triggered by "x in obj"
        return name in self._students

r = Roster()
r.add("Ada")
r.add("Grace")

print(r[0])
r[0] = "Katherine"
print(r[0])
print("Grace" in r)
del r[1]
print(r[0])
````

**Output:**
```
Ada
Katherine
True
Katherine
```

## 15.8 Iteration: `__iter__` and `__next__`

````python
class Countdown:
    def __init__(self, start):
        self.start = start

    def __iter__(self):                # triggered at the start of "for x in obj:"
        self.current = self.start
        return self

    def __next__(self):                  # triggered on each loop iteration
        if self.current <= 0:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

for n in Countdown(3):
    print(n)
````

**Output:**
```
3
2
1
```

## 15.9 Callable Objects: `__call__`

````python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):        # triggered by obj(...)
        return value * self.factor

double = Multiplier(2)
print(double(10))               # calling the INSTANCE directly, like a function
print(callable(double))
````

**Output:**
```
20
True
```

## 15.10 Context Managers: `__enter__` and `__exit__`

````python
class ManagedFile:
    def __init__(self, name):
        self.name = name

    def __enter__(self):                # triggered when entering a "with" block
        print(f"Opening {self.name}")
        return self

    def __exit__(self, exc_type, exc_value, traceback):    # triggered when leaving the block
        print(f"Closing {self.name}")
        return False       # False means: don't suppress any exception that occurred

with ManagedFile("data.txt") as f:
    print("Using the file")
````

**Output:**
```
Opening data.txt
Using the file
Closing data.txt
```

## 15.11 Hashing: `__hash__`

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        return (self.x, self.y) == (other.x, other.y)

    def __hash__(self):                  # triggered by hash(obj); required for set/dict-key use
        return hash((self.x, self.y))

p1 = Point(1, 2)
p2 = Point(1, 2)

print(p1 == p2)             # True - same coordinates
print(hash(p1) == hash(p2))   # True - consistent with __eq__

points = {p1, p2}              # a set - relies on both __hash__ and __eq__
print(len(points))              # 1, because p1 and p2 are considered the same
````

**Output:**
```
True
True
1
```

**Critical rule:** if you define `__eq__`, Python automatically sets `__hash__` to `None` unless you also explicitly define `__hash__` - this is because two objects considered equal (`==`) must produce the same hash, or dictionaries and sets will behave incorrectly. Defining `__eq__` without `__hash__` makes instances **unhashable**, discussed further with common mistakes in Part 26 and Part 29.

## 16.1 What Operator Overloading Means

**Operator overloading** is defining how a custom class responds to Python's built-in operators (`+`, `-`, `==`, `[]`, `in`, etc.) by implementing the corresponding dunder methods, covered individually in Part 15. This section focuses specifically on the design perspective: *why and how* to give operators sensible, intuitive meaning for your own classes.

## 16.2 Arithmetic Operators on a Custom Type

````python
class Money:
    def __init__(self, dollars, cents=0):
        self.total_cents = dollars * 100 + cents

    def __add__(self, other):
        result = Money(0)
        result.total_cents = self.total_cents + other.total_cents
        return result

    def __sub__(self, other):
        result = Money(0)
        result.total_cents = self.total_cents - other.total_cents
        return result

    def __repr__(self):
        return f"${self.total_cents // 100}.{self.total_cents % 100:02d}"

price = Money(19, 99)
tax = Money(1, 60)
total = price + tax          # reads naturally, thanks to __add__
print(total)

change = Money(20) - total
print(change)
````

## 16.7 Calling an Object Like a Function: `__call__`

````python
class Discount:
    def __init__(self, percent):
        self.percent = percent

    def __call__(self, price):        # obj(...) -> __call__
        return price * (1 - self.percent / 100)

ten_percent_off = Discount(10)
print(ten_percent_off(200))       # the object itself behaves like a function
````

**Output:**
```

```

## 16.8 Design Guidance: When Operator Overloading Helps vs Hurts

| Good Use of Operator Overloading | Poor Use of Operator Overloading |
|---|---|
| `Vector + Vector` -> combines vectors (matches mathematical intuition) | `Employee + Employee` -> unclear what this should mean |
| `Money - Money` -> difference in value (intuitive) | `User - Permission` -> ambiguous, surprising |
| `"x" in Roster` -> membership check (matches container intuition) | Overloading `<` on a `Logger` class to mean something unrelated to ordering |
| `len(Playlist)` -> number of songs (matches "size" intuition) | Overloading `__len__` to return something that isn't actually a count |

**Rule of thumb:** only overload an operator when its meaning for your class is the same, unsurprising concept the operator already represents for built-in types. If callers would have to guess or read documentation to understand what `a + b` does for your class, a clearly named method (`a.combine_with(b)`) is almost always a better choice than operator overloading.

# 17. Abstract Base Classes

## 17.1 What Abstract Classes Are

An **abstract class** defines a common interface that subclasses are required to implement, but cannot itself be instantiated directly. It exists purely to be inherited from.

````python
from abc import ABC, abstractmethod

class Shape(ABC):                     # inheriting from ABC marks this as abstract
    @abstractmethod
    def area(self):
        """Subclasses MUST implement this."""

    @abstractmethod
    def perimeter(self):
        """Subclasses MUST implement this too."""

try:
    s = Shape()
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
Error: Can't instantiate abstract class Shape with abstract methods area, perimeter
```

## 17.2 `abc`, `ABC`, and `@abstractmethod`

- `abc` is the standard library module providing abstract base class support.
- `ABC` is a convenience base class (built on the `ABCMeta` metaclass) that you inherit from to make your class abstract.
- `@abstractmethod` marks individual methods as required for any concrete (non-abstract) subclass.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):                  # implements the abstract method
        return 3.14159 * self.radius ** 2

class Incomplete(Shape):
    pass                              # does NOT implement area()

c = Circle(5)
print(c.area())

try:
    i = Incomplete()                  # still fails - abstract method not implemented
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
78.53975
Error: Can't instantiate abstract class Incomplete with abstract method area
```

## 17.3 Why Abstraction Matters - Preventing Direct Instantiation

Abstract classes formalize a design intention: "this class describes a *category* of things, not a concrete thing itself." Instantiating `Shape()` directly would be meaningless - a shape with no defined `area()` logic represents an incomplete concept. Making this a `TypeError` at the language level, rather than a documentation comment, prevents entire categories of bugs where a caller accidentally uses an incomplete implementation.

## 17.4 Mixing Abstract and Concrete Methods

An abstract class can define ordinary, fully-implemented methods alongside its abstract ones - subclasses inherit the concrete methods for free while still being required to fill in the abstract ones.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    def describe(self):                    # concrete method - shared by all subclasses
        return f"This shape has an area of {self.area():.2f}"

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

print(Square(4).describe())
````

**Output:**
```
This shape has an area of 16.00
```

## 17.5 Designing Interfaces with Abstract Base Classes

Abstract base classes are Python's primary tool for expressing "interfaces" - a guaranteed contract of methods any implementation must provide, useful when different concrete classes need to be used interchangeably (polymorphism, Part 21).

````python
from abc import ABC, abstractmethod

class NotificationChannel(ABC):
    @abstractmethod
    def send(self, message):
        pass

class EmailChannel(NotificationChannel):
    def send(self, message):
        return f"Emailing: {message}"

class SMSChannel(NotificationChannel):
    def send(self, message):
        return f"Texting: {message}"

def notify_all(channels, message):
    for channel in channels:
        print(channel.send(message))

notify_all([EmailChannel(), SMSChannel()], "Server is down!")
````

**Output:**
```
Emailing: Server is down!
Texting: Server is down!
```

## 17.6 Practical Example - An AI/Software Engineering Interface

Abstract base classes are especially common in professional codebases to guarantee that every model, data source, or processing stage in a pipeline exposes a consistent interface, regardless of its internal implementation.

````python
from abc import ABC, abstractmethod

class MLModel(ABC):
    @abstractmethod
    def train(self, data):
        pass

    @abstractmethod
    def predict(self, data):
        pass

class LinearRegressionModel(MLModel):
    def __init__(self):
        self.coefficients = None

    def train(self, data):
        self.coefficients = "fitted_coefficients"
        print("Training linear regression model...")

    def predict(self, data):
        return f"Prediction using {self.coefficients}"

class DecisionTreeModel(MLModel):
    def __init__(self):
        self.tree = None

    def train(self, data):
        self.tree = "fitted_tree"
        print("Training decision tree model...")

    def predict(self, data):
        return f"Prediction using {self.tree}"

def run_pipeline(model: MLModel, training_data, new_data):
    model.train(training_data)
    return model.predict(new_data)

print(run_pipeline(LinearRegressionModel(), "train_set", "new_row"))
print(run_pipeline(DecisionTreeModel(), "train_set", "new_row"))
````

**Output:**
```
Training linear regression model...
Prediction using fitted_coefficients
Training decision tree model...
Prediction using fitted_tree
```

Because `run_pipeline` only depends on the `MLModel` interface (`train`/`predict`), any new model type can be plugged in without changing `run_pipeline` at all - a direct application of abstraction in a realistic AI engineering context, expanded further in Part 31.

## 17.7 Abstract Properties

Properties (Part 14) can also be marked abstract, requiring subclasses to provide a specific attribute-like interface.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @property
    @abstractmethod
    def name(self):
        pass

class Triangle(Shape):
    @property
    def name(self):
        return "Triangle"

print(Triangle().name)
````

**Output:**
```
Triangle
```

## 17.8 Summary Table

| Concept | Syntax | Purpose |
|---|---|---|
| Abstract base class | `class X(ABC):` | Marks a class as non-instantiable, meant only to be subclassed |
| Abstract method | `@abstractmethod` | Requires every concrete subclass to implement this method |
| Abstract property | `@property` + `@abstractmethod` | Requires subclasses to provide a specific property |
| Mixed abstract/concrete | Abstract methods alongside normal methods in the same class | Shares common logic while still enforcing a required interface |
| Enforcement | `TypeError` at instantiation time | If any abstract method is left unimplemented |

# 18. Composition

## 18.1 "Has-a" Relationships

**Composition** models a **"has-a"** relationship: an object contains, and typically owns, one or more other objects as part of its own state, rather than inheriting behavior from them.

````python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f"Engine starting with {self.horsepower} HP"

class Car:
    def __init__(self, make, horsepower):
        self.make = make
        self.engine = Engine(horsepower)     # Car HAS-A Engine

    def start(self):
        return f"{self.make}: {self.engine.start()}"

car = Car("Toyota", 180)
print(car.start())
````

**Output:**
```
Toyota: Engine starting with 180 HP
```

## 18.2 Composition vs Inheritance - The Basic Distinction

Inheritance models **"is-a"** (`ElectricCar` **is a** `Car`); composition models **"has-a"** (`Car` **has an** `Engine`). A `Car` is not a kind of `Engine` - it *contains* one. Using inheritance here (`class Car(Engine)`) would be semantically wrong, even though it might work mechanically.

````python
# WRONG - Car is not a type of Engine, this misuses inheritance
class Car(Engine):
    pass

# RIGHT - Car has an Engine, composition models the relationship correctly
class Car:
    def __init__(self, horsepower):
        self.engine = Engine(horsepower)
````

## 18.3 Objects Containing Other Objects - Practical Examples

### Example 1: Company -> Employees

````python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def describe(self):
        return f"{self.name} ({self.role})"

class Company:
    def __init__(self, name):
        self.name = name
        self.employees = []        # Company HAS-A list of Employees

    def hire(self, employee):
        self.employees.append(employee)

    def list_staff(self):
        return [e.describe() for e in self.employees]

acme = Company("Acme Corp")
acme.hire(Employee("Ada", "Engineer"))
acme.hire(Employee("Grace", "Manager"))
print(acme.list_staff())
````

**Output:**
```
['Ada (Engineer)', 'Grace (Manager)']
```

### Example 2: AI System -> Model

````python
class Model:
    def __init__(self, name):
        self.name = name

    def predict(self, data):
        return f"Prediction from {self.name} for {data}"

class AISystem:
    def __init__(self, model_name):
        self.model = Model(model_name)       # AISystem HAS-A Model

    def run(self, data):
        return self.model.predict(data)

system = AISystem("FraudDetector-v2")
print(system.run("transaction_123"))
````

**Output:**
```
Prediction from FraudDetector-v2 for transaction_123
```

### Example 3: Banking System -> Accounts

````python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

class BankingSystem:
    def __init__(self):
        self.accounts = {}          # BankingSystem HAS-A collection of Accounts

    def open_account(self, owner, initial_balance=0):
        self.accounts[owner] = Account(owner, initial_balance)

    def total_deposits(self):
        return sum(acc.balance for acc in self.accounts.values())

bank = BankingSystem()
bank.open_account("Ada", 1000)
bank.open_account("Grace", 2500)
print(bank.total_deposits())
````

**Output:**
```
3500
```

## 18.4 Dependency Relationships

A related but weaker form of connection is a **dependency**: one object temporarily uses another (often passed as a method argument) without permanently storing or owning it.

````python
class EmailSender:
    def send(self, message):
        print(f"Sending: {message}")

class OrderProcessor:
    def process(self, order, sender: EmailSender):   # depends on EmailSender only during this call
        print(f"Processing order: {order}")
        sender.send(f"Your order '{order}' has shipped")

processor = OrderProcessor()
processor.process("Book", EmailSender())    # EmailSender is used, not owned/stored
````

**Output:**
```
Processing order: Book
Sending: Your order 'Book' has shipped
```

## 18.5 Practical Software Design Guidance

- Prefer composition when a class needs to **use the capabilities** of another object without being a specialized version of it.
- Composed objects can be swapped out at runtime (e.g., a `Car` could be given a different `Engine` instance), offering more flexibility than a fixed inheritance relationship decided once at class-definition time.
- Composition often leads to smaller, more focused classes, each responsible for one thing - directly supporting the Single Responsibility Principle (Part 28).

## 18.6 Summary Table

| Aspect | Composition |
|---|---|
| Relationship modeled | "Has-a" |
| Mechanism | One object stores another as an attribute |
| Flexibility | High - the contained object can be swapped or reconfigured at runtime |
| Coupling | Looser than inheritance - depends only on the contained object's interface |
| Example | `Car` has an `Engine`; `Company` has `Employees` |

# 19. Aggregation

## 19.1 What Aggregation Is

**Aggregation** is a special, looser form of "has-a" relationship where the contained object can exist **independently** of the container, and is typically not created or destroyed by the container itself. Composition, by contrast, usually implies the container **owns** and manages the lifecycle of the contained object.

````python
class Author:
    def __init__(self, name):
        self.name = name

class Book:
    def __init__(self, title, author: Author):    # aggregation - Book does not create or own the Author
        self.title = title
        self.author = author                        # the same Author object could belong to many Books

tolkien = Author("J.R.R. Tolkien")

book1 = Book("The Hobbit", tolkien)
book2 = Book("The Fellowship of the Ring", tolkien)   # tolkien exists independently, shared across books

print(book1.author.name, book2.author.name)
print(book1.author is book2.author)     # same Author object, not a copy
````

**Output:**
```
J.R.R. Tolkien J.R.R. Tolkien
True
```

## 19.2 Composition vs Aggregation - The Key Difference

| Aspect | Composition | Aggregation |
|---|---|---|
| Ownership | Container creates and owns the contained object | Contained object exists independently, often created elsewhere and passed in |
| Lifecycle | Contained object typically dies with the container | Contained object can outlive, or exist separately from, the container |
| Example | `Car` creates its own `Engine` internally | `Book` references an `Author` created elsewhere and possibly shared |
| Typical construction | `self.engine = Engine(...)` inside `__init__` | `self.author = author` where `author` is passed in as a parameter |

````python
# COMPOSITION - Car creates and owns its Engine; the Engine has no life outside this Car
class Car:
    def __init__(self, horsepower):
        self.engine = Engine(horsepower)     # created HERE, owned by this Car

# AGGREGATION - Book references an Author that exists independently
class Book:
    def __init__(self, title, author):
        self.author = author                  # passed IN, not created here, may be shared
````

## 19.3 When to Use Each

- Use **composition** when the contained object has no meaningful existence outside its container, and should be created and destroyed alongside it (an `Engine` without a `Car` around it is a strange, incomplete concept in this design).
- Use **aggregation** when the contained object is a legitimate, independent entity that can be shared, reused, or outlive any one particular container (an `Author` clearly exists as a person independent of any specific `Book`).

## 19.4 Practical Example - Aggregation in a Course/Student System

````python
class Student:
    def __init__(self, name):
        self.name = name

class Course:
    def __init__(self, title):
        self.title = title
        self.students = []       # aggregation - Students exist independently of any one Course

    def enroll(self, student):
        self.students.append(student)

ada = Student("Ada")     # Ada exists on her own, independent of any course

math = Course("Mathematics")
physics = Course("Physics")

math.enroll(ada)            # the SAME Ada is enrolled in multiple, independent courses
physics.enroll(ada)

print(math.students[0].name, physics.students[0].name)
print(math.students[0] is physics.students[0])
````

**Output:**
```
Ada Ada
True
```

## 19.5 Summary Table

| Concept | Relationship | Ownership | Example |
|---|---|---|---|
| Composition | Strong "has-a" | Container owns and manages lifecycle | `Car` -> `Engine` |
| Aggregation | Weak "has-a" | Contained object is independent, possibly shared | `Book` -> `Author`, `Course` -> `Student` |

# 20. Inheritance vs Composition

## 20.1 "Is-a" vs "Has-a" - The Core Distinction

| Question to Ask | If "Yes" |
|---|---|
| Is this new class fundamentally **a specific kind of** the other class? | Inheritance ("is-a") - e.g. `ElectricCar` **is a** `Car` |
| Does this new class simply **need to use** the other class's capability as one of its parts? | Composition ("has-a") - e.g. `Car` **has an** `Engine` |

````python
# "is-a" - inheritance is appropriate
class Car:
    def start(self):
        return "Car starting"

class ElectricCar(Car):        # an ElectricCar genuinely IS a specialized kind of Car
    def start(self):
        return "Electric car starting silently"

# "has-a" - composition is appropriate
class Engine:
    def ignite(self):
        return "Engine igniting"

class Car:
    def __init__(self):
        self.engine = Engine()    # a Car HAS an Engine, it is not a kind of Engine
````

## 20.2 Code Reuse: Two Different Mechanisms

Both inheritance and composition let you reuse code, but in structurally different ways.

````python
# Reuse via INHERITANCE
class Vehicle:
    def start(self):
        return "Starting"

class Car(Vehicle):
    pass    # reuses start() automatically through the class hierarchy

# Reuse via COMPOSITION
class Starter:
    def start(self):
        return "Starting"

class Car:
    def __init__(self):
        self.starter = Starter()

    def start(self):
        return self.starter.start()    # reuses start() by delegating to a contained object
````

## 20.3 Coupling and Maintainability

**Inheritance creates tight coupling**: a subclass depends directly on its parent's internal implementation details, and changes to the parent can unexpectedly ripple into every subclass (this is sometimes called the "fragile base class problem").

**Composition creates looser coupling**: a class depends only on the *public interface* of the object it contains, not its internal implementation, making it easier to change or replace the contained object without breaking the containing class.

````python
class Engine:
    def start(self):
        return "V8 engine roaring to life"

class ElectricEngine:
    def start(self):
        return "Electric motor humming"

class Car:
    def __init__(self, engine):
        self.engine = engine        # composition - depends only on engine.start() existing

    def start(self):
        return self.engine.start()

gas_car = Car(Engine())
electric_car = Car(ElectricEngine())    # swapped in an entirely different engine implementation

print(gas_car.start())
print(electric_car.start())
````

**Output:**
```
V8 engine roaring to life
Electric motor humming
```

This flexibility - swapping the `engine` at runtime, without touching `Car`'s class definition at all - is difficult to achieve as cleanly with inheritance, since a class's parent is fixed at class-definition time.

## 20.4 Flexibility and Scalability

| Aspect | Inheritance | Composition |
|---|---|---|
| Relationship fixed at | Class definition time (can't change a class's parent at runtime) | Runtime (the contained object can be swapped or reconfigured) |
| Depth of hierarchy | Can grow deep and unwieldy as more specialization is added | Stays flat; complexity grows by adding/replacing parts, not by extending a chain |
| Multiple behaviors | Requires multiple inheritance (Part 12), with its own risks | Straightforward - just compose multiple objects together |
| Testing | Testing a subclass often requires understanding its whole ancestor chain | Testing is easier - mock or stub the contained object's interface |

## 20.5 "Favor Composition Over Inheritance" - A Widely Cited Design Guideline

This well-known principle from software engineering does **not** mean inheritance is bad - it means: reach for composition by default when modeling a relationship, and use inheritance specifically when there is a genuine, stable "is-a" relationship where subclasses truly are specialized versions of the parent and are expected to be usable anywhere the parent is expected (this expectation is formalized as the Liskov Substitution Principle, Part 28).

## 20.6 When Inheritance Is Appropriate

- There is a genuine, conceptually stable "is-a" relationship (`Dog` is an `Animal`; `SavingsAccount` is a `BankAccount`).
- Subclasses should be usable anywhere the parent class is expected, without surprising behavior differences.
- You want to share a common interface and let polymorphism (Part 21) work naturally across the hierarchy.

## 20.7 When Composition Is Better

- The relationship is really "uses" or "has," not "is."
- You want to swap implementations at runtime (different `Engine` types, different `PaymentGateway` types).
- You want to avoid deep, fragile hierarchies and keep classes independently testable.
- You want to combine multiple independent behaviors without the risks of multiple inheritance (Part 12).

## 20.8 Summary Comparison Table

| Criterion | Inheritance | Composition |
|---|---|---|
| Relationship | "is-a" | "has-a" |
| Coupling | Tighter (depends on parent's internals) | Looser (depends only on public interface) |
| Flexibility | Fixed at class definition | Swappable at runtime |
| Risk of fragility | Higher, especially with deep/multiple inheritance | Lower |
| Best suited for | Genuine specialization, shared polymorphic interface | Reuse of a capability, flexible internal parts |
| Guiding principle | Use when Liskov Substitution genuinely holds | "Favor composition over inheritance" as a general default |

# 21. Polymorphism and Duck Typing

## 21.1 Python's Dynamic Nature

Python is **dynamically typed**: variable types are determined at runtime, not declared and checked ahead of time by a compiler. This dynamic nature is what makes Python's style of polymorphism possible and distinct from statically typed languages.

## 21.2 Compile-Time vs Runtime Concepts

In languages like Java or C++, "method overloading" typically means defining multiple methods with the **same name but different parameter lists**, and the compiler decides at **compile time** which version to call based on argument types. Python has no direct equivalent of this:

````python
class Calculator:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):        # this does NOT overload - it simply REPLACES the earlier add()
        return a + b + c

calc = Calculator()
try:
    calc.add(1, 2)          # fails - only the LAST defined add() exists at all
except TypeError as e:
    print("Error:", e)

print(calc.add(1, 2, 3))
````

**Output:**
```
Error: Calculator.add() missing 1 required positional argument: 'c'
Error: Calculator.add() missing 1 required positional argument: 'c'
6
```

**Precise terminology:** Python does not support traditional compile-time method overloading. Defining two methods with the same name simply makes the second definition overwrite the first in the class namespace - there is only ever one `add` method at any given time. The common workaround is default arguments, `*args`/`**kwargs` (see the separate Functions reference), or `functools.singledispatch` for type-based dispatch.

````python
class Calculator:
    def add(self, a, b, c=0):        # default parameter mimics "optional overload" behavior
        return a + b + c

calc = Calculator()
print(calc.add(1, 2))
print(calc.add(1, 2, 3))
````

**Output:**
```
3
6
```

What Python **does** support fully is **runtime polymorphism**: the same method call (`shape.area()`) invokes different actual code depending on the real, dynamic type of the object, decided while the program is running - this is method overriding (Part 10) in action.

## 21.3 Method-Based Polymorphism

````python
class Shape:
    def area(self):
        raise NotImplementedError

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width, self.height = width, height

    def area(self):
        return self.width * self.height

shapes = [Circle(3), Rectangle(4, 5)]
for shape in shapes:
    print(f"{type(shape).__name__}: {shape.area():.2f}")
````

**Output:**
```
Circle: 28.27
Rectangle: 20.00
```

## 21.4 Duck Typing

**Duck typing** takes polymorphism a step further: Python does not require objects to share a common parent class at all - it only cares whether the object has the method or attribute being used, checked at the moment it's actually accessed. "If it walks like a duck and quacks like a duck, it's treated as a duck," regardless of its actual class or ancestry.

````python
class Duck:
    def make_sound(self):
        return "Quack!"

class Dog:
    def make_sound(self):           # completely unrelated class, no shared parent
        return "Woof!"

class Robot:
    def make_sound(self):            # not even a living thing, still works
        return "Beep boop!"

def trigger_sound(entity):
    print(entity.make_sound())        # works on ANYTHING with a make_sound() method

for entity in [Duck(), Dog(), Robot()]:
    trigger_sound(entity)
````

**Output:**
```
Quack!
Woof!
Beep boop!
```

**Note:** if an object passed to `trigger_sound` does **not** have a `make_sound` method, Python raises `AttributeError` only when the call is actually attempted - duck typing is checked at the point of use, not ahead of time.

````python
class Rock:
    pass       # no make_sound method at all

try:
    trigger_sound(Rock())
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
Error: 'Rock' object has no attribute 'make_sound'
```

## 21.5 Common Interfaces Without Formal Inheritance

Duck typing is why Python code often relies on **informal protocols** - a shared expectation about which methods an object should provide - rather than requiring formal inheritance from a shared abstract base class (though `abc`, Part 17, is available when a stricter, enforced contract is wanted).

````python
class FileWriter:
    def write(self, text):
        print(f"Writing to file: {text}")

class ConsoleWriter:
    def write(self, text):
        print(f"Console output: {text}")

class NetworkWriter:
    def write(self, text):
        print(f"Sending over network: {text}")

def log_message(writer, message):
    writer.write(message)          # works with ANY object providing .write(), by convention

for writer in [FileWriter(), ConsoleWriter(), NetworkWriter()]:
    log_message(writer, "System started")
````

**Output:**
```
Writing to file: System started
Console output: System started
Sending over network: System started
```

This is exactly the same informal protocol that makes Python's own built-in `open(...)` objects, `io.StringIO`, and many third-party logging handlers all interchangeable wherever code expects "something with a `.write()` method."

## 21.6 Practical Example - Polymorphic Functions Across Unrelated Types

````python
def total_length(items):
    return sum(len(item) for item in items)     # works because ALL of these support len()

print(total_length(["hello", "hi", "hey"]))       # strings
print(total_length([[1, 2], [3, 4, 5]]))            # lists
print(total_length([{1, 2, 3}, {4, 5}]))              # sets
````

**Output:**
```
10
5
5
```

`total_length` never checks `isinstance` at all - it simply calls `len()` and trusts that whatever it's given supports it, which is duck typing applied to Python's own built-in protocol for `len()`.

## 21.7 Summary Table

| Concept | Description | Example |
|---|---|---|
| Method overriding polymorphism | Subclasses share a method name, each implementing it differently | `Circle.area()` vs `Rectangle.area()` |
| Duck typing | No shared parent class required; only the needed method/attribute matters | `Duck`, `Dog`, `Robot` all have `make_sound()` |
| Informal protocol | A convention (not enforced by inheritance) that certain methods exist | Anything with `.write()` works with `log_message()` |
| Compile-time overloading | Not supported in Python; same-name methods simply overwrite each other | Use default args, `*args`, or `functools.singledispatch` instead |
| Runtime polymorphism | Fully supported - the actual method called depends on the object's real type at runtime | `shape.area()` across different `Shape` subclasses |

# 22. Dataclasses

## 22.1 The Problem Dataclasses Solve

Classes that primarily exist to hold data traditionally require a lot of repetitive boilerplate: `__init__`, `__repr__`, and `__eq__` all have to be written by hand for even simple data containers.

````python
# WITHOUT @dataclass - lots of repetitive boilerplate
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

    def __eq__(self, other):
        return (self.x, self.y) == (other.x, other.y)

p1 = Point(1, 2)
p2 = Point(1, 2)
print(p1)
print(p1 == p2)
````

**Output:**
```
Point(x=1, y=2)
True
```

## 22.2 `@dataclass` - The Same Class, Far Less Code

````python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

p1 = Point(1, 2)
p2 = Point(1, 2)
print(p1)              # __repr__ generated automatically
print(p1 == p2)          # __eq__ generated automatically
````

**Output:**
```
Point(x=1, y=2)
True
```

`@dataclass` automatically generates `__init__`, `__repr__`, and `__eq__` based on the class's type-annotated fields, eliminating the boilerplate from Section 22.1 entirely.

## 22.3 Fields and Default Values

````python
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    salary: float = 50000.0        # default value
    department: str = "General"      # another default value

e1 = Employee("Ada")
e2 = Employee("Grace", 70000, "Engineering")

print(e1)
print(e2)
````

**Output:**
```
Employee(name='Ada', salary=50000.0, department='General')
Employee(name='Grace', salary=70000, department='Engineering')
```

**Rule:** just like regular function parameters, fields with default values must come after fields without defaults.

## 22.4 `field()` and Mutable Defaults

Using a mutable object (like a `list`) directly as a default value is disallowed in dataclasses - this actually protects you from the exact shared-mutable-default bug described in Part 5.2 and the Functions reference. Use `field(default_factory=...)` instead.

````python
from dataclasses import dataclass, field

@dataclass
class Team:
    name: str
    members: list = field(default_factory=list)     # a NEW list per instance, not shared

t1 = Team("Backend")
t2 = Team("Frontend")

t1.members.append("Ada")
print(t1.members, t2.members)     # t2 is correctly unaffected
````

**Output:**
```
['Ada'] []
```

````python
# Trying the naive (and disallowed) approach:
try:
    @dataclass
    class BadTeam:
        members: list = []       # dataclasses explicitly forbid mutable defaults like this
except ValueError as e:
    print("Error:", e)
````

**Output:**
```
Error: mutable default <class 'list'> for field members is not allowed: use default_factory
```

## 22.5 `__post_init__`

`__post_init__` runs automatically right after the generated `__init__` finishes, and is the idiomatic place for validation or computed fields in a dataclass.

````python
from dataclasses import dataclass

@dataclass
class Rectangle:
    width: float
    height: float
    area: float = None       # will be computed, not provided by the caller

    def __post_init__(self):
        if self.width <= 0 or self.height <= 0:
            raise ValueError("Dimensions must be positive")
        self.area = self.width * self.height

r = Rectangle(4, 5)
print(r)

try:
    Rectangle(-1, 5)
except ValueError as e:
    print("Error:", e)
````

**Output:**
```
Rectangle(width=4, height=5, area=20)
Error: Dimensions must be positive
```

## 22.6 Frozen Dataclasses (Immutability)

`frozen=True` makes instances immutable after creation - attempting to reassign any field raises an error, similar in spirit to a named tuple.

````python
from dataclasses import dataclass

@dataclass(frozen=True)
class Point:
    x: int
    y: int

p = Point(1, 2)
try:
    p.x = 99
except Exception as e:
    print("Error:", type(e).__name__, "-", e)
````

**Output:**

# 23. `__slots__`

## 23.1 What `__slots__` Is

By default, every Python object stores its instance attributes in a per-instance dictionary (`__dict__`), which is flexible (you can add new attributes at any time) but uses noticeably more memory than a fixed, predefined layout would need. `__slots__` lets a class declare, in advance, the *exact* set of attribute names its instances are allowed to have, and Python then uses a more compact internal storage layout instead of a `__dict__` for each instance.

````python
class PointWithDict:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class PointWithSlots:
    __slots__ = ("x", "y")            # only these two attributes are allowed

    def __init__(self, x, y):
        self.x = x
        self.y = y

p1 = PointWithDict(1, 2)
p1.z = 99                              # allowed - __dict__ accepts any new attribute
print(p1.__dict__)

p2 = PointWithSlots(1, 2)
try:
    p2.z = 99                          # NOT allowed - z was never declared in __slots__
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
{'x': 1, 'y': 2}
Error: 'PointWithSlots' object has no attribute 'z'
```

## 23.2 Why `__slots__` Exists - Memory Considerations

For classes that will be instantiated many thousands or millions of times (a very common situation in data-heavy or numerically intensive applications), the per-instance `__dict__` overhead adds up significantly. `__slots__` removes this overhead by allocating only fixed, precisely-sized storage for the declared attributes.

````python
import sys

class WithDict:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class WithSlots:
    __slots__ = ("x", "y")
    def __init__(self, x, y):
        self.x = x
        self.y = y

a = WithDict(1, 2)
b = WithSlots(1, 2)

print(sys.getsizeof(a.__dict__))    # the dictionary itself has overhead
print(hasattr(b, "__dict__"))         # slotted instances have NO __dict__ at all
````

**Output:**
```
104
False
```

## 23.3 Attribute Restrictions

Once `__slots__` is declared, instances of that class **cannot** have any attribute outside the declared list, unless `__dict__` is explicitly re-added to `__slots__` itself (which defeats much of the memory benefit).

````python
class Config:
    __slots__ = ("debug", "verbose")

    def __init__(self, debug, verbose):
        self.debug = debug
        self.verbose = verbose

c = Config(True, False)
print(c.debug, c.verbose)

try:
    c.extra_setting = "oops"    # not declared in __slots__
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
True False
Error: 'Config' object has no attribute 'extra_setting'
```

## 23.4 Performance Considerations

Alongside memory savings, attribute access on slotted classes is often marginally faster, because Python can use a direct, fixed-offset lookup instead of a dictionary hash lookup. In most everyday code this difference is negligible; it becomes meaningful primarily in tight loops over very large numbers of small objects.

## 23.5 Advantages of `__slots__`

- Significantly reduced per-instance memory usage, especially valuable when creating very large numbers of instances (e.g., simulation objects, data records, lightweight value objects).
- Slightly faster attribute access.
- Enforces a fixed attribute set, which can catch typos (`self.nmae = ...`) immediately as an `AttributeError` rather than silently creating an unintended new attribute.

## 23.6 Limitations of `__slots__`

- No `__dict__` by default, so arbitrary dynamic attributes cannot be added - this can break code (including some third-party libraries) that expects to freely set new attributes on any object.
- `__slots__` and multiple inheritance interact awkwardly: only one base class in a hierarchy is generally allowed to define non-empty `__slots__` with actual attribute storage; combining several slotted base classes with overlapping or additional attribute storage often raises errors or forces a `__dict__` to appear anyway.
- Class attributes with the same name as a slot are not allowed (slots and class-level default values for the same name conflict).
- Weak references require explicitly adding `"__weakref__"` to `__slots__` if needed, since slotted classes don't support them by default.

````python
class Base:
    __slots__ = ("a",)

class Child(Base):
    __slots__ = ("b",)          # this works - each level adds its own new slots

c = Child()
c.a = 1
c.b = 2
print(c.a, c.b)
````

**Output:**
```
1 2
```

## 23.7 When to Use `__slots__`

- Classes that will be instantiated in very large quantities (thousands to millions), where memory footprint genuinely matters (common in data processing, simulations, and certain performance-critical AI/ML pipeline components).
- Simple, well-defined data containers where the full, final set of attributes is known upfront and unlikely to need dynamic extension.

## 23.8 When Not to Use `__slots__`

- Small numbers of instances, where memory savings are negligible compared to the added rigidity.
- Classes that rely on dynamic attribute assignment, monkey-patching, or need to support arbitrary extension by users of the class.
- Complex multiple-inheritance hierarchies, where `__slots__` interactions become error-prone.
- As a default habit on every class "just in case" - it adds a real constraint and should be a deliberate, justified choice, not a reflex.

## 23.9 Summary Table

| Aspect | With `__slots__` | Without (default `__dict__`) |
|---|---|---|
| Memory per instance | Lower | Higher |
| Attribute access speed | Slightly faster | Standard |
| Dynamic new attributes | Not allowed (unless `__dict__` re-added) | Freely allowed |
| Typo protection | Yes - unknown attributes raise `AttributeError` immediately | No - a typo silently creates a new attribute |
| Multiple inheritance | Requires care; only one base with real slot storage generally recommended | Works normally |
| Best for | Many instances of simple, fixed-shape objects | General-purpose classes needing flexibility |

# 24. Custom Containers and Iterable Objects

## 24.1 Why Build Custom Containers

Python's built-in `list`, `dict`, `set`, and `tuple` cover most needs, but sometimes a domain-specific collection benefits from its own type: enforcing invariants, adding domain methods, or controlling exactly how items are stored, validated, or iterated. OOP lets you build such a type that still feels like a natural, built-in-style Python collection by implementing the right dunder methods (introduced individually in Part 15).

## 24.2 Building a Custom Iterable

An object is **iterable** if it implements `__iter__`, which must return an **iterator**.

````python
class NumberRange:
    def __init__(self, start, end):
        self.start = start
        self.end = end

    def __iter__(self):
        current = self.start
        while current < self.end:
            yield current                # using yield makes __iter__ itself a generator
            current += 1

for n in NumberRange(1, 5):
    print(n)
````

**Output:**
```
1
2
3
4
```

## 24.3 Building a Custom Iterator (Separating Iterable from Iterator)

A more explicit pattern separates the **iterable** (which can be iterated multiple times, producing a fresh iterator each time) from the **iterator** itself (which tracks progress through a single pass).

````python
class NumberRangeIterator:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self                     # an iterator is its own iterator

    def __next__(self):
        if self.current >= self.end:
            raise StopIteration
        value = self.current
        self.current += 1
        return value

class NumberRange:
    def __init__(self, start, end):
        self.start = start
        self.end = end

    def __iter__(self):
        return NumberRangeIterator(self.start, self.end)    # fresh iterator every time

r = NumberRange(1, 4)
print(list(r))       # first full pass
print(list(r))       # second full pass - works again, because __iter__ makes a NEW iterator each time
````

**Output:**
```
[1, 2, 3]
[1, 2, 3]
```

**Why this separation matters:** if `NumberRange` itself were the iterator (tracking `current` directly on itself), it could only be iterated once - a second `for` loop would find it already exhausted. Separating the two roles allows the same `NumberRange` object to be iterated repeatedly, just like a `list` can be.

## 24.4 Building a Custom Sequence

A **sequence** supports integer indexing (`obj[0]`), slicing, `len()`, and iteration (which Python can derive automatically from `__getitem__` alone, even without `__iter__`).

````python
class ImmutableStack:
    def __init__(self, items):
        self._items = tuple(items)

    def __len__(self):
        return len(self._items)

    def __getitem__(self, index):
        return self._items[index]

stack = ImmutableStack([10, 20, 30])
print(len(stack))
print(stack[0])
print(stack[-1])
print(stack[1:])          # slicing works automatically, delegated to the underlying tuple

for item in stack:          # iteration works automatically too, via repeated __getitem__ calls
    print(item)
````

**Output:**
```
3
10
30
(20, 30)
10
20
30
```

## 24.5 Building a Custom Mapping

A **mapping** supports key-based access (`obj[key]`), similar to a `dict`, implemented with `__getitem__`, `__setitem__`, `__delitem__`, `__contains__`, `__len__`, and `__iter__`.

````python
class CaseInsensitiveDict:
    def __init__(self):
        self._data = {}

    def __setitem__(self, key, value):
        self._data[key.lower()] = value

    def __getitem__(self, key):
        return self._data[key.lower()]

    def __delitem__(self, key):
        del self._data[key.lower()]

    def __contains__(self, key):
        return key.lower() in self._data

    def __len__(self):
        return len(self._data)

    def __iter__(self):
        return iter(self._data)

headers = CaseInsensitiveDict()
headers["Content-Type"] = "application/json"

print(headers["content-type"])       # different case, same key internally
print("CONTENT-TYPE" in headers)
print(len(headers))

for key in headers:
    print(key)
````

**Output:**
```
application/json
True
1
content-type
```

## 24.6 Full Collection-Like Object - Combining Everything

````python
class Playlist:
    def __init__(self, songs=None):
        self._songs = list(songs) if songs else []

    def __len__(self):
        return len(self._songs)

    def __getitem__(self, index):
        return self._songs[index]

    def __setitem__(self, index, value):
        self._songs[index] = value

    def __contains__(self, song):
        return song in self._songs

    def __iter__(self):
        return iter(self._songs)

    def __repr__(self):
        return f"Playlist({self._songs})"

    def add(self, song):
        self._songs.append(song)

p = Playlist(["Song A", "Song B"])
p.add("Song C")

print(len(p))
print(p[0])
print("Song B" in p)
for song in p:
    print(song)
print(p)
````

**Output:**
```
3
Song A
True
Song A
Song B
Song C
Playlist(['Song A', 'Song B', 'Song C'])
```

| Iterator | `__iter__` (returns self) + `__next__` | Manual `next(obj)` stepping, single-pass iteration |

# 25. Object Lifecycle

## 25.1 The Full Lifecycle, Step by Step

Every Python object moves through a predictable lifecycle: creation, use, and eventual destruction.


```
  Dog("Rex")
      |
      v
 [1] __new__      -> allocates raw memory for the object
      |
      v
 [2] __init__     -> initializes attributes on the object
      |
      v
 [3] Object is used -> methods called, attributes read/written, references passed around
      |
      v
 [4] Reference count drops to zero (no variables/containers refer to it anymore)
      |
      v
 [5] __del__      -> called just before the object is destroyed (if defined)
      |
      v
 [6] Garbage collected -> memory reclaimed
```

## 25.2 Creation: `__new__` and `__init__`

````python
class Dog:
    def __new__(cls, *args, **kwargs):
        print("[1] __new__: allocating memory")
        return super().__new__(cls)

    def __init__(self, name):
        print("[2] __init__: setting up attributes")
        self.name = name

d = Dog("Rex")
````

**Output:**
```
[1] __new__: allocating memory
[2] __init__: setting up attributes
```

## 25.3 Object Usage and References

While an object exists, any number of variables can **reference** it - they all point to the same underlying object in memory, rather than each holding a separate copy.

````python
class Dog:
    def __init__(self, name):
        self.name = name

d1 = Dog("Rex")
d2 = d1                     # d2 is just another NAME for the same object, not a copy

d2.name = "Max"
print(d1.name)                 # changed too - d1 and d2 refer to the identical object
print(d1 is d2)
````

**Output:**
```
Max
True
```

## 25.4 Reference Counting and Garbage Collection

CPython (the standard Python implementation) primarily uses **reference counting**: every object tracks how many references point to it, and once that count reaches zero, the memory is reclaimed almost immediately. A secondary **cyclic garbage collector** also runs periodically to catch reference cycles (e.g., two objects referencing each other) that reference counting alone cannot free.

````python
import sys

class Dog:
    def __init__(self, name):
        self.name = name

d = Dog("Rex")
print(sys.getrefcount(d))    # includes the temporary reference getrefcount's own argument creates

d2 = d
print(sys.getrefcount(d))     # one more reference now exists

del d2
print(sys.getrefcount(d))      # back down after removing one reference
````

**Output:**
```
2
3
2
```

## 25.5 Destruction: `__del__`

`__del__` is called just before an object is garbage-collected, and is sometimes described (imprecisely) as a "destructor." It can be used for optional cleanup, logging, or releasing certain resources, but it has real limitations.

````python
class Connection:
    def __init__(self, name):
        self.name = name
        print(f"Opening connection: {self.name}")

    def __del__(self):
        print(f"Closing connection: {self.name}")

def use_connection():
    conn = Connection("DB1")
    print("Using connection")
    # conn goes out of scope at the end of this function

use_connection()
print("Function finished")
````

**Output:**
```
Opening connection: DB1
Using connection
Closing connection: DB1
Function finished
```

## 25.6 Why Relying Heavily on `__del__` Is Discouraged

- **Timing is not guaranteed.** `__del__` runs when the reference count hits zero (in CPython) or at some unpredictable point during garbage collection in other implementations (like PyPy) - it may run much later than expected, or not at all before the program exits.
- **Reference cycles delay or skip cleanup.** Two objects referencing each other can prevent prompt reference-count-based collection, deferring `__del__` until the cyclic collector runs (or never, if the interpreter exits first).
- **Exceptions inside `__del__` are effectively swallowed**, only printed as a warning, since there is no meaningful place for them to propagate to.
- **Critical resource cleanup should not depend on it.** Files, network connections, and locks should be released deterministically using **context managers** (`__enter__`/`__exit__`, Part 15.10) via the `with` statement, not left to `__del__`'s unpredictable timing.

````python
# BETTER PATTERN - deterministic cleanup via a context manager, not __del__
class Connection:
    def __enter__(self):
        print("Opening connection")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Closing connection")     # guaranteed to run when the with-block ends
        return False

with Connection() as conn:
    print("Using connection")
````

**Output:**
```
Opening connection
Using connection
Closing connection
```

## 25.7 Summary Table

| Stage | Method/Mechanism | Notes |
|---|---|---|
| Allocation | `__new__` | Rarely overridden directly; creates the raw object |
| Initialization | `__init__` | Sets up instance attributes |
| Usage | Method calls, attribute access | Multiple references can point to the same object |
| Reference tracking | Reference counting (+ cyclic GC) | CPython-specific implementation detail |
| Destruction | `__del__` | Timing not guaranteed; avoid relying on it for critical cleanup |
| Deterministic cleanup | `__enter__` / `__exit__` via `with` | Preferred for resources like files, connections, locks |

# 26. Object Identity, Equality, and Hashing

## 26.1 Identity vs Equality - `is` vs `==`

**Identity** asks: "are these two names referring to the exact same object in memory?" **Equality** asks: "do these two objects (which may be different objects) represent the same value?" Python has a separate operator for each.

````python
a = [1, 2, 3]
b = [1, 2, 3]           # a DIFFERENT list object, with equal contents
c = a                     # the SAME object as a, just a second name for it

print(a == b)     # True - equal VALUE
print(a is b)      # False - different objects in memory
print(a is c)       # True - literally the same object
````

**Output:**
```
True
False
True
```

## 26.2 `id()` - Inspecting Identity Directly

`id()` returns a unique integer identifying an object for its lifetime (in CPython, this is typically its memory address). `is` is effectively shorthand for `id(a) == id(b)`.

````python
a = [1, 2, 3]
b = a

print(id(a))
print(id(b))
print(id(a) == id(b))
print(a is b)
````

**Output:**
```
140234567891200
140234567891200
True
True
```

## 26.3 The Default `__eq__` - Identity-Based Equality

If a class does not define `__eq__`, Python falls back to the default from `object`, which compares by identity - meaning two separate instances with identical attribute values are **not** considered equal unless `__eq__` is explicitly defined to compare their actual content.

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

p1 = Point(1, 2)
p2 = Point(1, 2)

print(p1 == p2)        # False - default __eq__ compares identity, and these are different objects
print(p1 == p1)          # True - same object
````

**Output:**
```
False
True
```

## 26.4 Custom `__eq__` - Value-Based Equality

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        if not isinstance(other, Point):
            return NotImplemented          # signal "I don't know how to compare to this type"
        return self.x == other.x and self.y == other.y

p1 = Point(1, 2)
p2 = Point(1, 2)

print(p1 == p2)        # True now - compares actual coordinate values
````

**Output:**
```
True
```

**Note on `NotImplemented`:** returning `NotImplemented` (not the same as raising `NotImplementedError`) from a comparison method tells Python "I can't compare with this type - try the other object's reflected method, or fall back to a default." This is the correct way to handle unexpected comparison types, rather than raising an exception directly.

## 26.5 `__hash__` - The Relationship Between Equality and Hashing

**The fundamental rule Python enforces by convention:** if two objects are equal (`a == b` is `True`), they **must** produce the same hash value (`hash(a) == hash(b)`). This is required for dictionaries and sets to work correctly, since they use an object's hash to quickly locate it, and then use `__eq__` to confirm an actual match.

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        return isinstance(other, Point) and (self.x, self.y) == (other.x, other.y)

    def __hash__(self):
        return hash((self.x, self.y))          # consistent with __eq__: same coordinates -> same hash

p1 = Point(1, 2)
p2 = Point(1, 2)

print(p1 == p2, hash(p1) == hash(p2))

points = {p1, p2}          # a set relies on both __hash__ and __eq__ together
print(len(points))            # 1 - correctly recognized as duplicates
````

**Output:**
```
True True
1
```

## 26.6 The Critical Trap: Defining `__eq__` Without `__hash__`

When a class defines `__eq__` but does **not** define `__hash__`, Python automatically sets `__hash__` to `None` on that class, making its instances **unhashable** - they can no longer be used as dictionary keys or set elements at all. This is a deliberate safety measure, since Python cannot safely guess a hash consistent with a custom `__eq__`.

````python
class BrokenPoint:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        return (self.x, self.y) == (other.x, other.y)
    # no __hash__ defined - Python sets __hash__ to None automatically here

p = BrokenPoint(1, 2)
try:
    {p}                  # attempting to put it in a set fails
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
Error: unhashable type: 'BrokenPoint'
```

**Fix:** always define `__hash__` alongside a custom `__eq__` if instances need to be hashable, ensuring the hash is computed from the same fields used in the equality check.

## 26.7 Dictionary and Set Behavior with Custom Objects

````python
class Currency:
    def __init__(self, code):
        self.code = code

    def __eq__(self, other):
        return isinstance(other, Currency) and self.code == other.code

    def __hash__(self):
        return hash(self.code)

    def __repr__(self):
        return f"Currency({self.code})"

usd1 = Currency("USD")
usd2 = Currency("USD")     # a different object, same logical currency
eur = Currency("EUR")

exchange_rates = {usd1: 1.0, eur: 0.92}
print(exchange_rates[usd2])          # usd2 successfully looks up usd1's entry - same hash and equal

unique_currencies = {usd1, usd2, eur}
print(unique_currencies)               # usd1 and usd2 collapse into one, since they're equal
````

**Output:**
```
1.0
{Currency(USD), Currency(EUR)}
```

## 26.8 Frozen Dataclasses and Hashing

As covered in Part 22, `@dataclass(frozen=True)` automatically generates a correct, consistent `__hash__` alongside `__eq__`, precisely because immutability guarantees the fields used for hashing cannot change after the hash is computed - a mutable object's hash changing after being placed in a set or dict would corrupt that collection's internal structure.

````python
from dataclasses import dataclass

@dataclass(frozen=True)
class Point:
    x: int
    y: int

points = {Point(1, 2), Point(1, 2), Point(3, 4)}
print(points)
````

**Output:**
```
{Point(x=1, y=2), Point(x=3, y=4)}
```

## 26.9 Summary Table

| Concept | Operator/Method | Meaning |
|---|---|---|
| Identity | `is` | Same object in memory? |
| Identity value | `id(obj)` | Unique identifier for an object's lifetime |
| Equality | `==` | Same logical value? (calls `__eq__`) |
| Default equality | `object.__eq__` | Falls back to identity comparison if not overridden |
| Hashing | `hash(obj)` | Calls `__hash__`; required for dict keys / set elements |
| Consistency rule | - | If `a == b`, then `hash(a) == hash(b)` must also hold |
| Common trap | Defining `__eq__` without `__hash__` | Makes instances unhashable (`__hash__` becomes `None`) |
| Safe default | `@dataclass(frozen=True)` | Correctly generates a consistent `__eq__` and `__hash__` together |

# 27. Class Design

## 27.1 Single Responsibility - A Class Should Do One Thing Well

A well-designed class has one clear, cohesive purpose. When a class tries to do too much, it becomes hard to understand, test, and modify safely.

````python
# POOR DESIGN - one class doing far too much
class UserManager:
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def save_to_database(self):
        print(f"Saving {self.name} to database")

    def send_welcome_email(self):
        print(f"Emailing {self.email}")

    def generate_pdf_report(self):
        print("Generating PDF report")

    def validate_email_format(self):
        return "@" in self.email
````

This single class mixes user data, persistence, email delivery, and reporting - four unrelated responsibilities. A change to how emails are sent risks breaking something entirely unrelated to database logic, and the class is hard to test in isolation.

````python
# BETTER DESIGN - responsibilities separated into focused, cohesive classes
class User:
    def __init__(self, name, email):
        self.name = name
        self.email = email

class UserRepository:
    def save(self, user: User):
        print(f"Saving {user.name} to database")

class EmailService:
    def send_welcome(self, user: User):
        print(f"Emailing {user.email}")

class ReportGenerator:
    def generate_pdf(self, user: User):
        print("Generating PDF report")

user = User("Ada", "ada@example.com")
UserRepository().save(user)
EmailService().send_welcome(user)
````

**Output:**
```
Saving Ada to database
Emailing ada@example.com
```

## 27.2 Cohesion and Coupling

**Cohesion** measures how closely related the responsibilities *within* a single class are - high cohesion means everything in the class works together toward one clear purpose. **Coupling** measures how much classes *depend* on each other's internal details - low coupling means classes can change independently without breaking one another.

| Goal | What to Aim For |
|---|---|
| High cohesion | Every attribute and method in a class relates to its single, clear purpose |
| Low coupling | Classes interact only through clear public interfaces, not by reaching into each other's internals |

````python
# LOW COUPLING - EmailService depends only on User's public attributes (name, email)
# not on how UserRepository stores data internally
class EmailService:
    def send_welcome(self, user):
        print(f"Emailing {user.email}: Welcome, {user.name}!")
````

## 27.3 Naming

Good class, attribute, and method names communicate intent without needing comments. Classes should generally be named with nouns (`Invoice`, `Customer`); methods with verbs (`calculate_total`, `send_reminder`); boolean-returning methods often read naturally as questions (`is_valid`, `has_permission`).

````python
# POOR NAMING
class Mgr:
    def proc(self, d):
        return d * 1.1

# CLEAR NAMING
class PriceManager:
    def apply_tax(self, price):
        return price * 1.1
````

## 27.4 Attributes: Keep Them Minimal and Meaningful

Only store what the object genuinely needs to represent its state. Derived values that can be computed from existing attributes are usually better expressed as properties (Part 14) than as separately stored, potentially-out-of-sync attributes.

````python
# RISK - stored total can drift out of sync with the actual items
class Order:
    def __init__(self, items):
        self.items = items
        self.total = sum(item.price for item in items)   # computed once, can go stale

    def add_item(self, item):
        self.items.append(item)
        # if we forget to update self.total here, it silently becomes wrong

# BETTER - total is always correct, computed on demand
class Order:
    def __init__(self, items):
        self.items = items

    @property
    def total(self):
        return sum(item.price for item in self.items)

    def add_item(self, item):
        self.items.append(item)     # total is automatically correct next time it's read
````

## 27.5 Methods and Interfaces

A class's public methods form its **interface** - the contract other code relies on. Keep this interface small, clear, and stable; push internal helper logic into `_protected` methods (Part 13) so it can change freely without affecting callers.

````python
class InvoiceGenerator:
    def generate(self, order):                 # small, clear public interface
        self._validate(order)
        lines = self._build_line_items(order)
        return self._format_invoice(lines)

    def _validate(self, order):                 # internal detail - can change freely
        if not order.items:
            raise ValueError("Cannot generate invoice for empty order")

    def _build_line_items(self, order):
        return [f"{item.name}: ${item.price}" for item in order.items]

    def _format_invoice(self, lines):
        return "\n".join(lines)
````

## 27.6 Validation

Validate data as early as possible - ideally in `__init__` or a property setter (Part 14) - so invalid objects can never exist in the first place, rather than discovering invalid state deep inside unrelated logic later.

## 27.7 Reusability and Maintainability

Favor small, focused, well-named classes with clear responsibilities and stable interfaces. This directly supports:
- **Reusability**: a focused `EmailService` can be reused anywhere email needs to be sent, unlike a giant `UserManager` tangled with unrelated logic.
- **Maintainability**: changes to one responsibility (e.g., switching database libraries inside `UserRepository`) do not risk breaking unrelated responsibilities (e.g., email sending).

## 27.8 Bad vs Good Class Design - Side-by-Side Example

````python
# BAD DESIGN
class Report:
    def __init__(self, data):
        self.data = data
        self.formatted = None

    def do_everything(self, fmt):
        cleaned = [d for d in self.data if d is not None]     # cleaning logic
        total = sum(cleaned)                                     # calculation logic
        if fmt == "pdf":
            self.formatted = f"PDF Report: {total}"               # formatting logic
        elif fmt == "csv":
            self.formatted = f"total,{total}"
        print(self.formatted)                                       # output logic, all mixed together
````

````python
# GOOD DESIGN
class DataCleaner:
    @staticmethod
    def clean(data):
        return [d for d in data if d is not None]

class ReportCalculator:
    @staticmethod
    def total(cleaned_data):
        return sum(cleaned_data)

class ReportFormatter:
    @staticmethod
    def to_pdf(total):
        return f"PDF Report: {total}"

    @staticmethod
    def to_csv(total):
        return f"total,{total}"

data = [10, None, 20, 30]
cleaned = DataCleaner.clean(data)
total = ReportCalculator.total(cleaned)
print(ReportFormatter.to_pdf(total))
print(ReportFormatter.to_csv(total))
````

**Output:**
```
PDF Report: 60
total,60
```

Each class in the "good design" version can be tested, reused, and modified completely independently of the others.

# 28. OOP Design Principles

## 28.1 SOLID - Overview

**SOLID** is an acronym for five foundational object-oriented design principles that guide maintainable, flexible class design.

## 28.2 S - Single Responsibility Principle (SRP)

*A class should have only one reason to change.* Already demonstrated in depth in Part 27; the `UserManager` -> `User`/`UserRepository`/`EmailService`/`ReportGenerator` refactor is a direct application of SRP.

## 28.3 O - Open/Closed Principle (OCP)

*Classes should be open for extension, but closed for modification* - you should be able to add new behavior without changing existing, already-tested code.

````python
# VIOLATES OCP - adding a new discount type requires modifying this existing method
class DiscountCalculator:
    def calculate(self, order_type, amount):
        if order_type == "regular":
            return amount
        elif order_type == "vip":
            return amount * 0.9
        # adding "student" discount means editing this method again, risking existing logic

# FOLLOWS OCP - new discount types are added by EXTENSION, not modification
from abc import ABC, abstractmethod

class DiscountStrategy(ABC):
    @abstractmethod
    def apply(self, amount):
        pass

class RegularDiscount(DiscountStrategy):
    def apply(self, amount):
        return amount

class VIPDiscount(DiscountStrategy):
    def apply(self, amount):
        return amount * 0.9

class StudentDiscount(DiscountStrategy):        # NEW behavior added without touching existing classes
    def apply(self, amount):
        return amount * 0.85

def checkout(amount, strategy: DiscountStrategy):
    return strategy.apply(amount)

print(checkout(100, VIPDiscount()))
print(checkout(100, StudentDiscount()))
````

**Output:**
```
90.0
85.0
```

## 28.4 L - Liskov Substitution Principle (LSP)

*Subclasses should be usable anywhere their parent class is expected, without breaking correctness.* This formalizes exactly when inheritance is appropriate (Part 20.6).

````python
# VIOLATES LSP - Square changes the expected behavior of setting width/height independently
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)

    def set_width(self, width):
        self.width = width
        self.height = width          # surprising side effect - breaks caller's assumption

def resize_and_check(rect: Rectangle):
    rect.width = 5
    rect.height = 4
    assert rect.area() == 20, "LSP violated - Square broke this assumption!"

try:
    resize_and_check(Square(3))    # Square secretly links width and height, breaking the assumption
except AssertionError as e:
    print("Error:", e)
````

**Output:**
```
Error: LSP violated - Square broke this assumption!
```

Any code written to work with `Rectangle` reasonably assumes width and height can be set independently. `Square` violates that expectation, even though "a square is a rectangle" sounds intuitively correct - this is a classic, widely cited example of why "is-a" in natural language does not always translate cleanly into "is-a" in code.

## 28.5 I - Interface Segregation Principle (ISP)

*Classes should not be forced to depend on methods they do not use - prefer several small, specific interfaces over one large, general-purpose one.*

````python
# VIOLATES ISP - forces ALL printers to implement scan() and fax(), even simple ones
from abc import ABC, abstractmethod

class MultiFunctionDevice(ABC):
    @abstractmethod
    def print_doc(self, doc):
        pass

    @abstractmethod
    def scan(self, doc):
        pass

    @abstractmethod
    def fax(self, doc):
        pass

class SimplePrinter(MultiFunctionDevice):
    def print_doc(self, doc):
        print(f"Printing {doc}")

    def scan(self, doc):
        raise NotImplementedError("This printer cannot scan")   # forced to implement something it can't do

    def fax(self, doc):
        raise NotImplementedError("This printer cannot fax")
````

````python
# FOLLOWS ISP - small, focused interfaces; classes implement only what they actually support
from abc import ABC, abstractmethod

class Printable(ABC):
    @abstractmethod
    def print_doc(self, doc):
        pass

class Scannable(ABC):
    @abstractmethod
    def scan(self, doc):
        pass

class SimplePrinter(Printable):                # only implements what it actually needs to
    def print_doc(self, doc):
        print(f"Printing {doc}")

class AllInOnePrinter(Printable, Scannable):
    def print_doc(self, doc):
        print(f"Printing {doc}")

    def scan(self, doc):
        print(f"Scanning {doc}")

SimplePrinter().print_doc("Report.pdf")
AllInOnePrinter().scan("Report.pdf")
````

**Output:**
```
Printing Report.pdf
Scanning Report.pdf
```

## 28.6 D - Dependency Inversion Principle (DIP)

*High-level modules should not depend directly on low-level implementation details - both should depend on a shared abstraction.*

````python
# VIOLATES DIP - OrderProcessor is tightly coupled to a SPECIFIC email implementation
class GmailSender:
    def send(self, message):
        print(f"Sending via Gmail: {message}")

class OrderProcessor:
    def __init__(self):
        self.sender = GmailSender()    # hardcoded dependency on one specific implementation

    def complete_order(self):
        self.sender.send("Order confirmed")
````

````python
# FOLLOWS DIP - OrderProcessor depends only on an abstraction (NotificationSender)
from abc import ABC, abstractmethod

class NotificationSender(ABC):
    @abstractmethod
    def send(self, message):
        pass

class GmailSender(NotificationSender):
    def send(self, message):
        print(f"Sending via Gmail: {message}")

class SlackSender(NotificationSender):
    def send(self, message):
        print(f"Sending via Slack: {message}")

class OrderProcessor:
    def __init__(self, sender: NotificationSender):    # depends on the ABSTRACTION, not a specific class
        self.sender = sender

    def complete_order(self):
        self.sender.send("Order confirmed")

OrderProcessor(GmailSender()).complete_order()
OrderProcessor(SlackSender()).complete_order()
````

**Output:**
```
Sending via Gmail: Order confirmed
Sending via Slack: Order confirmed
```

`OrderProcessor` can now work with any current or future `NotificationSender` implementation without any changes to its own code - directly enabled by the abstraction (Part 17) and duck-typed polymorphism (Part 21) covered earlier.

## 28.7 DRY - Don't Repeat Yourself

*Every piece of knowledge or logic should have a single, authoritative representation in the codebase.* Duplicated logic (e.g., the same validation rule copy-pasted in three places) means every future change must remember to update all copies correctly.

````python
# VIOLATES DRY - the same validation logic duplicated in two places
class User:
    def __init__(self, email):
        if "@" not in email:
            raise ValueError("Invalid email")
        self.email = email

class Subscriber:
    def __init__(self, email):
        if "@" not in email:               # duplicated validation logic
            raise ValueError("Invalid email")
        self.email = email
````

````python
# FOLLOWS DRY - shared validation logic defined once
class EmailValidator:
    @staticmethod
    def validate(email):
        if "@" not in email:
            raise ValueError("Invalid email")

class User:
    def __init__(self, email):
        EmailValidator.validate(email)
        self.email = email

class Subscriber:
    def __init__(self, email):
        EmailValidator.validate(email)
        self.email = email
````

## 28.8 KISS - Keep It Simple, Stupid

*Prefer the simplest design that correctly solves the problem - avoid unnecessary complexity, cleverness, or abstraction that isn't actually needed yet.*

````python
# UNNECESSARILY COMPLEX for a simple need
class AbstractGreetingStrategyFactoryProvider:
    def get_strategy_factory(self):
        return GreetingStrategyFactory()
# ... several more layers of indirection just to print "Hello"

# KISS - simple, direct, and perfectly sufficient
def greet(name):
    return f"Hello, {name}!"
````

## 28.9 YAGNI - You Aren't Gonna Need It

*Do not build functionality speculatively for hypothetical future requirements that do not exist yet.* Add abstraction and flexibility when an actual, concrete need arises, not preemptively "just in case."

````python
# YAGNI VIOLATION - building a whole plugin system for a feature that has exactly one implementation
class ReportPluginRegistry:
    def __init__(self):
        self.plugins = {}

    def register(self, name, plugin):
        self.plugins[name] = plugin

    def generate(self, name, data):
        return self.plugins[name].generate(data)

# If there is genuinely only ONE report type today, this is premature complexity.
# A simple function or class suffices until a second report type is ACTUALLY needed:
def generate_report(data):
    return f"Report: {sum(data)}"
````

# 29. Common OOP Mistakes

Each mistake below follows the same structure: incorrect code, the underlying problem, corrected code, and the explanation of the fix.

## 29.1 Confusing Classes and Objects

**Incorrect:**

````python
class Dog:
    pass

Dog.name = "Rex"       # setting an attribute on the CLASS, thinking it's for one dog
d = Dog()
print(d.name)
````

**Output:**
```
Rex
```
**Problem:** this sets `name` as a *class attribute*, shared by every `Dog` instance, not a per-object value - a second `Dog` would also show `"Rex"` even without being told to.

**Corrected:**

````python
class Dog:
    def __init__(self, name):
        self.name = name       # instance attribute - correctly per-object

d1 = Dog("Rex")
d2 = Dog("Bella")
print(d1.name, d2.name)
````

**Output:**
```
Rex Bella
```
**Fix explanation:** attributes meant to vary per object must be set on `self` inside `__init__`, not directly on the class.

## 29.2 Forgetting `self`

**Incorrect:**

````python
class Dog:
    def bark():                # missing self!
        return "Woof!"

d = Dog()
try:
    d.bark()
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
Error: Dog.bark() takes 0 positional arguments but 1 was given
```
**Problem:** Python automatically passes the instance as the first argument to any instance method call; without a `self` parameter to receive it, the call fails.

**Corrected:**

````python
class Dog:
    def bark(self):
        return "Woof!"

print(Dog().bark())
````

**Output:**
```
Woof!
```

## 29.3 Incorrect Use of Mutable Class Attributes

**Incorrect:**

````python
class ShoppingCart:
    items = []          # shared by EVERY instance - a common, serious bug

    def add(self, item):
        self.items.append(item)

cart1 = ShoppingCart()
cart2 = ShoppingCart()
cart1.add("Book")
print(cart2.items)      # WRONG - shows ['Book'] even though we only added to cart1
````

**Output:**
```
['Book']
```
**Problem:** `items` is a class attribute (a single shared list), so mutating it through any instance affects every instance.

**Corrected:**

````python
class ShoppingCart:
    def __init__(self):
        self.items = []       # a fresh list per instance

    def add(self, item):
        self.items.append(item)

cart1 = ShoppingCart()
cart2 = ShoppingCart()
cart1.add("Book")
print(cart1.items, cart2.items)
````

**Output:**
```
['Book'] []
```

## 29.4 Incorrect Inheritance ("is-a" vs "has-a" Confusion)

**Incorrect:**

````python
class Engine:
    def start(self):
        return "Engine starting"

class Car(Engine):        # a Car is NOT a kind of Engine - misuses inheritance
    pass
````

**Problem:** this models a "has-a" relationship using inheritance, creating a semantically incorrect and confusing hierarchy (see Part 18, Part 20).

**Corrected:**

````python
class Engine:
    def start(self):
        return "Engine starting"

class Car:
    def __init__(self):
        self.engine = Engine()      # composition - Car HAS an Engine

    def start(self):
        return self.engine.start()

print(Car().start())
````

**Output:**
```
Engine starting
```

## 29.5 Misusing `super()`

**Incorrect:**

````python
class Employee:
    def __init__(self, name):
        self.name = name

class Manager(Employee):
    def __init__(self, name, team_size):
        self.team_size = team_size     # FORGOT to call super().__init__(name)!

m = Manager("Ada", 5)
try:
    print(m.name)
except AttributeError as e:
    print("Error:", e)
````

**Output:**
```
Error: 'Manager' object has no attribute 'name'
```
**Corrected:**

````python
class Manager(Employee):
    def __init__(self, name, team_size):
        super().__init__(name)          # ensures Employee's setup actually runs
        self.team_size = team_size

m = Manager("Ada", 5)
print(m.name, m.team_size)
````

**Output:**
```
Ada 5
```

## 29.6 Excessive Inheritance (Deep, Fragile Hierarchies)

**Incorrect (conceptually):**

````python
class Animal: pass
class Mammal(Animal): pass
class Carnivore(Mammal): pass
class Canine(Carnivore): pass
class Domesticated(Canine): pass
class Dog(Domesticated): pass
class Puppy(Dog): pass          # six levels deep just to represent a puppy
````

**Problem:** this hierarchy is fragile - a change to `Animal` can ripple unpredictably through six levels of subclasses, and understanding `Puppy` requires understanding the entire chain.

**Corrected (favor composition and flatter hierarchies, per Part 20):**

````python
class Animal:
    def __init__(self, species, diet, is_domesticated, age_group):
        self.species = species
        self.diet = diet
        self.is_domesticated = is_domesticated
        self.age_group = age_group

puppy = Animal("Canine", "Carnivore", True, "Puppy")
print(puppy.species, puppy.age_group)
````

**Output:**
```
Canine Puppy
```

## 29.7 Overusing Static Methods

**Incorrect (conceptually):**

````python
class OrderProcessor:
    @staticmethod
    def process(order, tax_rate, discount, customer_type, shipping_zone):
        # ... a large static method doing everything, unable to hold any state or be extended cleanly
        pass
````

**Problem:** forcing complex, evolving logic into static methods loses the benefits of instance state, inheritance, and polymorphism - every new variation requires adding more parameters rather than a clean subclass or object composition.

**Corrected:**

````python
class OrderProcessor:
    def __init__(self, tax_rate, discount):
        self.tax_rate = tax_rate
        self.discount = discount

    def process(self, order):
        subtotal = sum(item.price for item in order.items)
        return subtotal * (1 + self.tax_rate) * (1 - self.discount)
````

## 29.8 Poor Encapsulation

**Incorrect:**

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance      # fully public, no validation at all

a = BankAccount(100)
a.balance = -99999                    # nothing prevents this nonsensical state
````

**Corrected (using a property, Part 14):**

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative")
        self._balance = value

a = BankAccount(100)
try:
    a.balance = -99999
except ValueError as e:
    print("Error:", e)
````

**Output:**
```
Error: Balance cannot be negative
```

## 29.9 Incorrect `__eq__` Implementation

**Incorrect:**

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        return self.x == other.x        # forgot to compare y at all!

p1 = Point(1, 2)
p2 = Point(1, 99)
print(p1 == p2)      # WRONG - considered equal despite different y values
````

**Output:**
```
True
```
**Corrected:**

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return self.x == other.x and self.y == other.y

p1 = Point(1, 2)
p2 = Point(1, 99)
print(p1 == p2)
````

**Output:**
```
False
```

## 29.10 Incorrect `__hash__` (Or Missing It Entirely)

**Incorrect:**

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        return isinstance(other, Point) and self.x == other.x and self.y == other.y
    # no __hash__ defined - instances are now unhashable

try:
    {Point(1, 2)}
except TypeError as e:
    print("Error:", e)
````

**Output:**
```
Error: unhashable type: 'Point'
```
**Corrected:**

````python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        return isinstance(other, Point) and self.x == other.x and self.y == other.y

    def __hash__(self):
        return hash((self.x, self.y))       # consistent with __eq__

print({Point(1, 2), Point(1, 2)})
````

```
{<__main__.Point object at 0x...>}
```

## 29.11 Misusing `__del__`

**Incorrect (relying on `__del__` for critical cleanup):**

````python
class DatabaseConnection:
    def __init__(self):
        print("Opening connection")

    def __del__(self):
        print("Closing connection")     # timing not guaranteed - risky for critical cleanup

def risky_usage():
    conn = DatabaseConnection()
    raise RuntimeError("Something went wrong")     # exception - when exactly does __del__ run?
````

**Corrected (deterministic cleanup with a context manager, per Part 25.6):**

````python
class DatabaseConnection:
    def __enter__(self):
        print("Opening connection")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Closing connection")      # GUARANTEED to run, even if an exception occurs
        return False

try:
    with DatabaseConnection() as conn:
        raise RuntimeError("Something went wrong")
except RuntimeError as e:
    print("Handled:", e)
````

## 30.1 Project 1 - Student Management System

````python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self._grades = {}

    def add_grade(self, subject, grade):
        if not 0 <= grade <= 100:
            raise ValueError("Grade must be between 0 and 100")
        self._grades[subject] = grade

    @property
    def average(self):
        if not self._grades:
            return 0
        return sum(self._grades.values()) / len(self._grades)

    def __repr__(self):
        return f"Student({self.name}, avg={self.average:.1f})"

class StudentManagementSystem:
    def __init__(self):
        self._students = {}

    def enroll(self, student: Student):
        self._students[student.student_id] = student

    def get_student(self, student_id):
        return self._students.get(student_id)

    def top_student(self):
        return max(self._students.values(), key=lambda s: s.average, default=None)

sms = StudentManagementSystem()
ada = Student("Ada", "S001")
ada.add_grade("Math", 95)
ada.add_grade("Science", 88)

grace = Student("Grace", "S002")
grace.add_grade("Math", 99)
grace.add_grade("Science", 97)

sms.enroll(ada)
sms.enroll(grace)

print(sms.get_student("S001"))
print("Top student:", sms.top_student())
````

## 30.2 Project 2 - Bank Account System

**Concepts used:** encapsulation, properties, inheritance, method overriding.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount

    def __repr__(self):
        return f"{type(self).__name__}({self.owner}, ${self._balance:.2f})"

class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, interest_rate=0.02):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def apply_interest(self):
        self._balance += self._balance * self.interest_rate

class CheckingAccount(BankAccount):
    def __init__(self, owner, balance=0, overdraft_limit=100):
        super().__init__(owner, balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):                # OVERRIDES to allow overdraft
        if amount > self._balance + self.overdraft_limit:
            raise ValueError("Exceeds overdraft limit")
        self._balance -= amount

savings = SavingsAccount("Ada", 1000)
savings.apply_interest()
print(savings)

checking = CheckingAccount("Grace", 50)
checking.withdraw(100)     # allowed thanks to overdraft
print(checking)
````

## 30.3 Project 3 - Employee Management System

**Concepts used:** inheritance, polymorphism, class methods, abstract base classes.

````python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary

    @abstractmethod
    def calculate_pay(self):
        pass

    def __repr__(self):
        return f"{type(self).__name__}({self.name}, ${self.calculate_pay():.2f})"

class SalariedEmployee(Employee):
    def calculate_pay(self):
        return self.base_salary / 12

class HourlyEmployee(Employee):
    def __init__(self, name, hourly_rate, hours_worked):
        super().__init__(name, 0)
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_pay(self):
        return self.hourly_rate * self.hours_worked

class CommissionEmployee(Employee):
    def __init__(self, name, base_salary, sales, commission_rate=0.05):
        super().__init__(name, base_salary)
        self.sales = sales
        self.commission_rate = commission_rate

    def calculate_pay(self):
        return self.base_salary / 12 + self.sales * self.commission_rate

class Payroll:
    def __init__(self):
        self.employees = []

    def add(self, employee: Employee):
        self.employees.append(employee)

    def total_payout(self):
        return sum(e.calculate_pay() for e in self.employees)

payroll = Payroll()
payroll.add(SalariedEmployee("Ada", 96000))
payroll.add(HourlyEmployee("Grace", 25, 160))
payroll.add(CommissionEmployee("Katherine", 60000, 20000))

for e in payroll.employees:
    print(e)
print(f"Total payout: ${payroll.total_payout():.2f}")
````

## 30.4 Project 4 - Library Management System

**Concepts used:** composition, aggregation, encapsulation, custom `__str__`/`__repr__`.

````python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_checked_out = False

    def __repr__(self):
        status = "checked out" if self.is_checked_out else "available"
        return f"'{self.title}' by {self.author} ({status})"

class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []          # aggregation - Books exist independently of Member

class Library:
    def __init__(self):
        self.books = {}          # composition - the catalog owns these Book records
        self.members = {}

    def add_book(self, book: Book):
        self.books[book.isbn] = book

    def register_member(self, member: Member):
        self.members[member.member_id] = member

    def checkout(self, isbn, member_id):
        book = self.books[isbn]
        member = self.members[member_id]
        if book.is_checked_out:
            raise ValueError(f"{book.title} is already checked out")
        book.is_checked_out = True
        member.borrowed_books.append(book)

    def return_book(self, isbn, member_id):
        book = self.books[isbn]
        member = self.members[member_id]
        book.is_checked_out = False
        member.borrowed_books.remove(book)

library = Library()
library.add_book(Book("Dune", "Frank Herbert", "111"))
library.register_member(Member("Ada", "M1"))

library.checkout("111", "M1")
print(library.books["111"])

library.return_book("111", "M1")
print(library.books["111"])
````

**Output:**
```
'Dune' by Frank Herbert (checked out)
'Dune' by Frank Herbert (available)
```

## 30.6 Project 6 - Inventory Management System

**Concepts used:** class methods (alternative constructors), custom exceptions via classes, encapsulation.

````python
class InsufficientStockError(Exception):
    pass

class InventoryItem:
    def __init__(self, sku, name, quantity, reorder_level=10):
        self.sku = sku
        self.name = name
        self.quantity = quantity
        self.reorder_level = reorder_level

    @classmethod
    def from_dict(cls, data):
        return cls(data["sku"], data["name"], data["quantity"])

    def needs_reorder(self):
        return self.quantity <= self.reorder_level

class Inventory:
    def __init__(self):
        self._items = {}

    def add_item(self, item: InventoryItem):
        self._items[item.sku] = item

    def sell(self, sku, quantity):
        item = self._items[sku]
        if quantity > item.quantity:
            raise InsufficientStockError(f"Not enough stock for {item.name}")
        item.quantity -= quantity

    def items_needing_reorder(self):
        return [i for i in self._items.values() if i.needs_reorder()]

inv = Inventory()
inv.add_item(InventoryItem.from_dict({"sku": "A1", "name": "Widget", "quantity": 15}))

inv.sell("A1", 10)
print(inv.items_needing_reorder())

try:
    inv.sell("A1", 100)
except InsufficientStockError as e:
    print("Error:", e)
````

## 30.8 Project 8 - Banking Transaction System

**Concepts used:** composition, `__enter__`/`__exit__`, custom exceptions, operator overloading.

````python
class TransactionError(Exception):
    pass

class Transaction:
    def __init__(self, account, amount, kind):
        self.account = account
        self.amount = amount
        self.kind = kind

    def __enter__(self):
        print(f"Starting {self.kind} transaction of ${self.amount}")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type is not None:
            print(f"Transaction failed: {exc_value}")
            return True     # suppress the exception - transaction was safely rolled back
        print("Transaction committed")
        return False

class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        with Transaction(self, amount, "withdrawal"):
            if amount > self.balance:
                raise TransactionError("Insufficient funds")
            self.balance -= amount

acc = Account("Ada", 100)
acc.withdraw(50)
print(acc.balance)

acc.withdraw(1000)      # fails gracefully - handled inside the Transaction context manager
print(acc.balance)
````

```
50
Starting withdrawal transaction of $1000
Transaction failed: Insufficient funds
50
```

## 30.9 Project 9 - Machine Learning Model Manager

**Concepts used:** abstract base classes, composition, class methods, polymorphism. (Expanded further in Part 31.)

````python
from abc import ABC, abstractmethod

class BaseModel(ABC):
    @abstractmethod
    def train(self, X, y):
        pass

    @abstractmethod
    def predict(self, X):
        pass

class LinearModel(BaseModel):
    def __init__(self):
        self.weights = None

    def train(self, X, y):
        self.weights = "fitted"
        print("LinearModel trained")

    def predict(self, X):
        return f"linear prediction for {X}"

class ModelManager:
    def __init__(self):
        self._models = {}

    def register(self, name, model: BaseModel):
        self._models[name] = model

    def train_all(self, X, y):
        for model in self._models.values():
            model.train(X, y)

    def predict_with(self, name, X):
        return self._models[name].predict(X)

manager = ModelManager()
manager.register("linear", LinearModel())
manager.train_all("features", "labels")
print(manager.predict_with("linear", "new_data"))
````

## 30.10 Project 10 - AI Model Pipeline Using OOP

**Concepts used:** composition of pipeline stages, abstraction, dependency inversion. (Expanded further in Part 31.)

````python
from abc import ABC, abstractmethod

class PipelineStage(ABC):
    @abstractmethod
    def process(self, data):
        pass

class CleaningStage(PipelineStage):
    def process(self, data):
        cleaned = [d for d in data if d is not None]
        print(f"Cleaned data: {cleaned}")
        return cleaned

class NormalizationStage(PipelineStage):
    def process(self, data):
        max_val = max(data)
        normalized = [d / max_val for d in data]
        print(f"Normalized data: {normalized}")
        return normalized

class ModelStage(PipelineStage):
    def process(self, data):
        prediction = sum(data) / len(data)
        print(f"Prediction: {prediction:.3f}")
        return prediction

class Pipeline:
    def __init__(self, stages):
        self.stages = stages          # composition - a list of independent, swappable stages

    def run(self, data):
        for stage in self.stages:
            data = stage.process(data)
        return data

pipeline = Pipeline([CleaningStage(), NormalizationStage(), ModelStage()])
result = pipeline.run([10, None, 20, 30, None])
print("Final result:", result)
````

**Output:**

# 31. OOP for Data Science and AI Engineering

## 31.1 Why Professional ML/AI Codebases Rely Heavily on Classes

Machine learning workflows involve multiple interacting stages - data loading, preprocessing, feature engineering, training, prediction, evaluation - each with configuration, internal state, and behavior that benefits directly from encapsulation, abstraction, and polymorphism. Plain scripts with global variables and standalone functions quickly become unmanageable once a project needs to support multiple models, multiple data sources, or repeatable, testable pipelines. OOP gives ML/AI code:

- A consistent, swappable interface across different model types (abstraction + polymorphism, Parts 17, 21).
- Encapsulated state (learned parameters, configuration, fitted transformers) that travels with the object rather than being scattered across global variables.
- Reusable, composable pipeline stages (composition, Part 18) that can be assembled, reordered, or replaced.

## 31.2 `DataProcessor` - Encapsulating Preprocessing Logic

````python
class DataProcessor:
    def __init__(self):
        self._mean = None
        self._std = None
        self._fitted = False

    def fit(self, data):
        self._mean = sum(data) / len(data)
        variance = sum((x - self._mean) ** 2 for x in data) / len(data)
        self._std = variance ** 0.5
        self._fitted = True
        return self

    def transform(self, data):
        if not self._fitted:
            raise RuntimeError("DataProcessor must be fitted before calling transform()")
        return [(x - self._mean) / self._std for x in data]

    def fit_transform(self, data):
        return self.fit(data).transform(data)

processor = DataProcessor()
scaled = processor.fit_transform([10, 20, 30, 40, 50])
print([round(x, 3) for x in scaled])
````

**Output:**
```
[-1.414, -0.707, 0.0, 0.707, 1.414]
```

Note the `fit()` method returning `self` - this is method chaining (Part 21, and the separate Methods guide), a very common pattern in ML libraries like scikit-learn, allowing `fit(...).transform(...)` in one expression.

## 31.3 `FeatureEngineer` - Composable Feature Transformations

````python
class FeatureEngineer:
    def __init__(self):
        self._steps = []          # composition - a sequence of independent transformation functions

    def add_step(self, name, func):
        self._steps.append((name, func))
        return self

    def transform(self, row):
        result = dict(row)
        for name, func in self._steps:
            result[name] = func(row)
        return result

engineer = FeatureEngineer()
engineer.add_step("bmi", lambda r: r["weight_kg"] / (r["height_m"] ** 2))
engineer.add_step("is_adult", lambda r: r["age"] >= 18)

row = {"weight_kg": 70, "height_m": 1.75, "age": 25}
print(engineer.transform(row))
````

**Output:**
```
{'weight_kg': 70, 'height_m': 1.75, 'age': 25, 'bmi': 22.857142857142858, 'is_adult': True}
```

## 31.4 `ModelTrainer` and `Predictor` - Separating Training from Inference

Separating training responsibility from prediction responsibility follows Single Responsibility (Part 27) and is common in production ML systems, where training may run on a completely different infrastructure/schedule than serving predictions.

````python
class ModelTrainer:
    def __init__(self, model):
        self.model = model

    def train(self, X, y):
        print("Training model...")
        self.model.train(X, y)
        return self.model

class Predictor:
    def __init__(self, trained_model):
        self.model = trained_model

    def predict_one(self, x):
        return self.model.predict(x)

    def predict_batch(self, xs):
        return [self.model.predict(x) for x in xs]

class SimpleModel:
    def __init__(self):
        self.fitted = False

    def train(self, X, y):
        self.fitted = True

    def predict(self, x):
        return "positive" if self.fitted else "unfitted"

trainer = ModelTrainer(SimpleModel())
trained_model = trainer.train("X_train", "y_train")

predictor = Predictor(trained_model)
print(predictor.predict_batch(["row1", "row2"]))
````

**Output:**
```
Training model...
['positive', 'positive']
```

## 31.5 `ModelPipeline` - Composing an End-to-End Pipeline

````python
class ModelPipeline:
    def __init__(self, processor: DataProcessor, model):
        self.processor = processor          # composition of independent, reusable parts
        self.model = model

    def train(self, X, y):
        X_scaled = self.processor.fit_transform(X)
        self.model.train(X_scaled, y)
        return self

    def predict(self, X):
        X_scaled = self.processor.transform(X)
        return self.model.predict(X_scaled)

pipeline = ModelPipeline(DataProcessor(), SimpleModel())
pipeline.train([10, 20, 30, 40], [0, 0, 1, 1])
print(pipeline.predict([15, 25]))
````

**Output:**
```
positive
```

## 31.6 AI Agents Modeled with OOP

````python
class Tool:
    def __init__(self, name, func):
        self.name = name
        self.func = func

    def run(self, *args):
        return self.func(*args)

class Agent:
    def __init__(self, name):
        self.name = name
        self.tools = {}          # composition - an agent HAS a set of tools

    def register_tool(self, tool: Tool):
        self.tools[tool.name] = tool

    def act(self, tool_name, *args):
        if tool_name not in self.tools:
            raise ValueError(f"{self.name} has no tool named '{tool_name}'")
        return self.tools[tool_name].run(*args)

agent = Agent("ResearchAgent")
agent.register_tool(Tool("search", lambda q: f"Results for: {q}"))
agent.register_tool(Tool("summarize", lambda text: f"Summary of: {text}"))

print(agent.act("search", "OOP in Python"))
print(agent.act("summarize", "a long article"))
````

**Output:**
```

Summary of: a long article
```

## 31.7 Configuration Systems for ML Experiments

````python
from dataclasses import dataclass, field

@dataclass(frozen=True)
class ExperimentConfig:
    learning_rate: float = 0.001
    batch_size: int = 32
    epochs: int = 10
    model_type: str = "linear"

class Experiment:
    def __init__(self, config: ExperimentConfig):
        self.config = config
        self.results = []

    def run(self):
        print(f"Running {self.config.model_type} model for {self.config.epochs} epochs")
        self.results.append({"epochs": self.config.epochs, "final_loss": 0.05})

exp = Experiment(ExperimentConfig(learning_rate=0.01, epochs=5, model_type="random_forest"))
exp.run()
print(exp.results)
````

**Output:**
```
Running random_forest model for 5 epochs
[{'epochs': 5, 'final_loss': 0.05}]
```

Using an immutable, `frozen=True` dataclass for configuration (Part 22.6) prevents an experiment's configuration from being accidentally mutated mid-run, which would make results impossible to reproduce reliably.

# 32. OOP Interview Questions

## 32.1 Beginner Questions

**Q1. What is the difference between a class and an object?**
A class is a blueprint defining attributes and methods; an object is a concrete instance created from that blueprint. `class Dog:` defines the blueprint; `d = Dog()` creates one actual object.

**Q2. What is `self` in Python, and is it a keyword?**
`self` refers to the specific instance a method is called on. It is **not** a keyword - it is a strong convention. Any name would work syntactically, but every Python codebase uses `self` for instance methods.

**Q3. What is the purpose of `__init__()`?**
`__init__()` is the initializer, called automatically right after an object is created, used to set up its starting instance attributes. It is often informally called "the constructor," though technically `__new__` is the true allocator (see Q13).

**Q4. What is the difference between an instance attribute and a class attribute?**
An instance attribute is unique to one object, usually set via `self.x = value` in `__init__`. A class attribute is shared by every instance of the class, defined directly in the class body.

**Q5. What is method overriding?**
When a subclass defines a method with the same name as one in its parent class, replacing (or, with `super()`, extending) the parent's version for instances of the subclass.

**Q6. What does `super()` do?**
It provides access to methods defined on the next class in the Method Resolution Order (usually the parent class), most commonly used to call the parent's `__init__()`.

**Q7. What is encapsulation?**
Bundling data and the methods that operate on it into one unit (a class), and controlling access to some of that data using naming conventions (`_protected`, `__private`) or properties.

**Q8. What is inheritance, and give an example.**
A mechanism where a child class acquires attributes and methods from a parent class. Example: `class Dog(Animal):` - `Dog` inherits everything `Animal` defines.

## 32.2 Intermediate Questions

**Q9. What is the difference between `@staticmethod` and `@classmethod`?**
A `@staticmethod` receives neither `self` nor `cls` and behaves like a plain function namespaced under the class. A `@classmethod` receives `cls` (the class itself) and is commonly used for alternative constructors or operations on class-level state.

**Q10. What is polymorphism, and how does duck typing relate to it?**
Polymorphism means the same method call produces different behavior depending on the actual object's type. Duck typing is Python's flexible take on this: an object just needs the right method or attribute to be used correctly, regardless of its class hierarchy - "if it walks like a duck and quacks like a duck..."

**Q11. What is the difference between `__str__` and `__repr__`?**
`__str__` provides a readable, human-facing string (used by `print()` and `str()`). `__repr__` provides an unambiguous, developer-facing string (used by `repr()`, the interactive interpreter, and containers displaying their elements). If `__str__` is missing, Python falls back to `__repr__`.

**Q12. Explain the difference between `is` and `==`.**
`is` checks identity - whether two names refer to the exact same object in memory. `==` checks equality - whether two objects (possibly different ones) represent the same value, via `__eq__`.

**Q13. What is the difference between `__new__` and `__init__`?**
`__new__` is the actual object allocator - it creates and returns the raw object. `__init__` then initializes that already-created object's attributes. `__init__` must return `None`; `__new__` must return the new instance (or occasionally another object).

**Q14. What is name mangling, and why does Python use it?**
A double-leading-underscore attribute (`self.__x`) is automatically renamed internally to `_ClassName__x`. This primarily prevents accidental attribute collisions between a parent class and subclasses that happen to use the same private-style attribute name, not to provide true privacy.

**Q15. Why can defining `__eq__` without `__hash__` cause problems?**
If a class defines `__eq__`, Python automatically sets `__hash__` to `None` unless `__hash__` is also explicitly defined, making instances unhashable - they can no longer be used in a `set` or as `dict` keys. A consistent `__hash__` should be defined whenever a custom `__eq__` is added, if hashability is needed.

## 32.3 Advanced Questions

**Q16. What is the Method Resolution Order (MRO), and how is it computed?**
MRO is the precise order Python searches a class hierarchy for a method or attribute. It is computed using the C3 linearization algorithm, which guarantees subclasses are checked before parents, multiple parents are checked in the order listed, and each class appears exactly once, even in diamond-shaped hierarchies.

**Q17. Explain the diamond inheritance problem and how Python resolves it.**
It occurs when two parent classes share a common ancestor, and a child inherits from both. Without care, the shared ancestor's logic could run twice. Python's MRO ensures the shared ancestor appears only once in the resolution order, and if every class in the chain calls `super()` cooperatively (rather than naming a parent class directly), the ancestor's logic runs exactly once.

**Q18. What is the Liskov Substitution Principle, and give a Python example of violating it.**
It states subclasses should be usable anywhere their parent class is expected, without breaking correctness. A classic violation: making `Square` inherit from `Rectangle` and linking `width`/`height` together, which breaks code that assumes they can be set independently on any `Rectangle`.

**Q19. Why does Python not have true compile-time method overloading?**
Because Python is dynamically typed and resolves method calls at runtime, not by inspecting argument types before the program runs. Defining two methods with the same name in a class simply causes the second definition to overwrite the first. Python instead achieves similar flexibility via default arguments, `*args`/`**kwargs`, or `functools.singledispatch` for type-based dispatch.

**Q20. When would you choose composition over inheritance?**
When the relationship is genuinely "has-a" or "uses," not "is-a"; when you want to swap implementations at runtime; or when a deep/multiple inheritance hierarchy would introduce more fragility and coupling than the shared behavior is worth. "Favor composition over inheritance" is a general default, not an absolute rule.

**Q21. What are `__slots__`, and what trade-off do they introduce?**
`__slots__` declares a fixed set of allowed attributes for a class, replacing the default per-instance `__dict__` with a more compact storage layout, saving memory and slightly speeding up attribute access. The trade-off is losing the ability to freely add new attributes and increased complexity when combined with multiple inheritance.

**Q22. Why should `__del__` not be relied on for critical resource cleanup?**
Its execution timing is not guaranteed - reference cycles or interpreter shutdown order can delay or skip it, and exceptions raised inside it are effectively swallowed. Deterministic cleanup should use `__enter__`/`__exit__` with the `with` statement instead.

## 32.4 Coding Questions

**Q23. Write a class hierarchy where `Shape` is abstract with an abstract `area()` method, and `Circle`/`Rectangle` implement it.**

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width, self.height = width, height

    def area(self):
        return self.width * self.height

shapes = [Circle(3), Rectangle(4, 5)]
for s in shapes:
    print(f"{type(s).__name__}: {s.area():.2f}")
````

**Output:**
```
Circle: 28.27
Rectangle: 20.00
```

**Q24. Implement a class that supports `+` between two instances, returning a new instance.**

````python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

print(Vector(1, 2) + Vector(3, 4))
````

**Output:**
```
Vector(4, 6)
```

## 32.5 Debugging Questions

**Q25. Why does this code print the same list for both objects, and how would you fix it?**

````python
class Team:
    members = []

    def add(self, name):
        self.members.append(name)

t1, t2 = Team(), Team()
t1.add("Ada")
print(t2.members)
````

**Answer:** `members` is a class attribute - a single list shared by every `Team` instance. Fix: move it into `__init__` as `self.members = []` so each instance gets its own list.

**Q26. Why does this raise `AttributeError`, and how would you fix it?**

````python
class Employee:
    def __init__(self, name):
        self.name = name

class Manager(Employee):
    def __init__(self, name, team_size):
        self.team_size = team_size

m = Manager("Ada", 5)
print(m.name)
````

**Answer:** `Manager.__init__` never calls `super().__init__(name)`, so `Employee`'s setup (which sets `self.name`) never runs. Fix: add `super().__init__(name)` as the first line.

## 32.6 Scenario-Based Questions

**Q27. You're designing a notification system that must support email, SMS, and push notifications, with more channels likely added later. How would you design this with OOP, and which SOLID principle does your design satisfy?**
**Answer:** Define an abstract `NotificationChannel` base class with an abstract `send(message)` method; each channel type is a concrete subclass. Code that triggers notifications depends only on the abstraction, not specific channel classes - satisfying the Open/Closed Principle (new channels can be added by extension, without modifying existing code) and the Dependency Inversion Principle (high-level code depends on the abstraction, not concrete channel implementations).

**Q28. A colleague has a `Square` class inheriting from `Rectangle`, and a bug appears whenever code that resizes rectangles is run on squares. What's happening, and what would you recommend?**
**Answer:** This is a Liskov Substitution Principle violation - `Square` changes the meaning of setting width/height independently, breaking assumptions code makes about any `Rectangle`. Recommend either not modeling `Square` as a `Rectangle` subclass at all, or redesigning both to share a common `Shape` interface with independent, non-coupled implementations of `area()`.

# 34. Comprehensive OOP Reference Tables

## 34.1 Class vs Object

| Aspect | Class | Object |
|---|---|---|
| Definition | Blueprint/template | Concrete instance created from a class |
| Syntax | `class Dog:` | `d = Dog()` |
| Memory | No memory allocated for instance data | Occupies memory for its own attributes |
| Quantity | Usually one definition | Many independent objects can exist |
| Example | `Dog` | `d1 = Dog()`, `d2 = Dog()` |

## 34.2 Instance Attributes vs Class Attributes

| Aspect | Instance Attribute | Class Attribute |
|---|---|---|
| Defined | `self.x = value`, typically in `__init__` | `x = value` directly in the class body |
| Storage | One copy per object | One copy shared by the whole class |
| Access | `obj.x` | `obj.x` or `ClassName.x` |
| Risk | None special | Mutable class attributes shared unexpectedly across instances |

## 34.3 Instance Methods vs Class Methods vs Static Methods

| Aspect | Instance Method | Class Method | Static Method |
|---|---|---|---|
| Decorator | none | `@classmethod` | `@staticmethod` |
| First parameter | `self` | `cls` | none |
| Access to instance state | Yes | No | No |
| Access to class state | Yes (indirectly) | Yes | No |
| Typical use | Object-specific behavior | Alternative constructors | Grouped utility logic |

## 34.4 Encapsulation vs Abstraction

| Aspect | Encapsulation | Abstraction |
|---|---|---|
| Focus | Controlling access to internal data | Hiding implementation complexity, exposing essential interface |
| Mechanism | `_protected`, `__private` (name mangling), `@property` | Public methods, abstract base classes (`ABC`, `@abstractmethod`) |
| Question answered | "Who can touch this data, and how?" | "What does the caller need to know to use this?" |
| Example | `self.__balance` with a `get_balance()` method | `Shape.area()` without exposing the formula's internals |

## 34.5 Inheritance vs Composition

| Aspect | Inheritance | Composition |
|---|---|---|
| Relationship | "is-a" | "has-a" |
| Coupling | Tighter | Looser |
| Flexibility | Fixed at class definition | Swappable at runtime |
| Example | `ElectricCar(Car)` | `Car` contains an `Engine` |

## 34.6 Aggregation vs Composition

| Aspect | Aggregation | Composition |
|---|---|---|
| Ownership | Contained object exists independently | Container owns and manages the contained object's lifecycle |
| Sharing | Can be shared across multiple containers | Typically exclusive to one container |
| Example | `Book` references an independently-existing `Author` | `Car` creates and owns its own `Engine` |

## 34.7 Overriding vs Overloading

| Aspect | Overriding | Overloading |
|---|---|---|
| Definition | Subclass redefines a method with the same name as the parent's | Multiple methods with the same name but different parameters (in languages that support it) |
| Python support | Fully supported | **Not supported** in the traditional compile-time sense; same-name methods simply overwrite each other |
| Python alternative | `super()` to extend, or fully replace | Default arguments, `*args`/`**kwargs`, or `functools.singledispatch` |
| Resolved at | Runtime, based on the actual object's type | Compile time, in languages that support it (e.g. Java, C++) |

## 34.8 `is` vs `==`

| Aspect | `is` | `==` |
|---|---|---|
| Checks | Identity - same object in memory? | Equality - same logical value? |
| Backed by | `id(a) == id(b)` | `a.__eq__(b)` |
| Default for custom classes | N/A (always identity-based) | Falls back to identity comparison unless `__eq__` is defined |
| Typical use | `x is None`, checking for the exact same object | Comparing values, e.g. `price == 9.99` |

## 34.9 `__str__` vs `__repr__`

| Aspect | `__str__` | `__repr__` |
|---|---|---|
| Audience | End users | Developers |
| Triggered by | `str(obj)`, `print(obj)` | `repr(obj)`, the interactive interpreter, containers displaying elements |
| Goal | Readable | Unambiguous, ideally valid Python to recreate the object |
| Fallback | Falls back to `__repr__` if undefined | No fallback to `__str__` |

## 34.10 `__new__` vs `__init__`

| Aspect | `__new__` | `__init__` |
|---|---|---|
| Role | Allocates the raw object (true constructor) | Initializes the already-created object's attributes |
| First parameter | `cls` | `self` |
| Return value | Must return the new instance | Must return `None` |
| How often overridden | Rarely | Very commonly |
| Runs | First | Second, immediately after `__new__` |

## 34.11 Public vs Protected (Convention) vs Private (Name-Mangled)

| Aspect | Public (`name`) | Protected (`_name`) | Private (`__name`) |
|---|---|---|---|
| Signal to other developers | Fully intended for external use | "Internal - please don't touch directly" | "Internal, and structurally harder to access accidentally" |
| Enforcement | None | None | Partial, via name mangling to `_ClassName__name` |
| Common use | Everyday data with no invariants | Implementation details, helper methods | Avoiding subclass attribute name collisions |

## 34.12 SOLID Principles at a Glance

| Letter | Principle | One-Line Summary |
|---|---|---|
| S | Single Responsibility | A class should have one reason to change |
| O | Open/Closed | Open for extension, closed for modification |
| L | Liskov Substitution | Subclasses must be safely substitutable for their parent |
| I | Interface Segregation | Prefer small, specific interfaces over large general ones |
| D | Dependency Inversion | Depend on abstractions, not concrete implementations |

## 34.13 Inheritance Types at a Glance

| Type | Structure |
|---|---|
| Single | One parent, one child |
| Multilevel | Parent -> child -> grandchild chain |
| Hierarchical | One parent, multiple independent children |
| Multiple | One child, multiple direct parents |
| Hybrid | Any combination of the above |

# 35. Conclusion

Object-oriented programming is fundamentally a way of organizing code around the things a program models - bundling their data and behavior together, and describing how those things relate to and specialize one another. Python's implementation of OOP is distinctively pragmatic: it gives you classes, inheritance, encapsulation, and polymorphism, but consistently favors clear conventions and developer trust over rigid, compiler-enforced restrictions - there is no true `private` keyword, no compile-time method overloading, and even `self` is a convention rather than a language rule.

This handbook moved from the foundational vocabulary of classes and objects, through the mechanics of `__init__`, instance and class-level state, and the three kinds of methods, into the four core principles - encapsulation, abstraction, inheritance, and polymorphism - and then into the more advanced machinery that makes real-world Python OOP work correctly: `super()` and the Method Resolution Order, dunder methods that connect your classes to Python's own syntax, properties for controlled attribute access, dataclasses and `__slots__` for reducing boilerplate and memory overhead, and the object lifecycle governing creation and destruction.

The design-focused parts - composition versus inheritance, SOLID, DRY/KISS/YAGNI, and the catalog of common mistakes - exist because writing *working* OOP code is a much lower bar than writing *good* OOP code. A class hierarchy can run without errors and still be fragile, hard to test, or painful to extend. The projects and interview questions in the final parts are meant to bridge that gap: real, complete systems that combine multiple principles at once, and the precise vocabulary needed to discuss and defend design decisions clearly, whether in a classroom, a code review, or a technical interview.

The single most valuable habit to take from this handbook is to always ask, before writing a new class or relationship: *is this genuinely "is-a," or is it really "has-a"?* Answering that question honestly, and then applying the right mechanism - inheritance for real specialization, composition for everything else - resolves the large majority of design problems this handbook covers.
