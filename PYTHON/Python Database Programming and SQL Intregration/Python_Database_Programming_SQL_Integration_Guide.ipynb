# 1. Introduction to Databases

## 1.1 What a Database Is

A **database** is an organized collection of data, structured so it can be efficiently stored, retrieved, updated, and managed. Rather than scattering data across loose files, a database enforces structure and rules that keep data consistent, searchable, and safe from accidental corruption.

## 1.2 Why Databases Are Important

| Reason | Explanation |
|---|---|
| Structured storage | Data is organized into a predictable, queryable shape rather than ad hoc files. |
| Concurrent access | Many users or programs can safely read and write at the same time. |
| Data integrity | Constraints (Part 1.9) prevent invalid or contradictory data from being stored. |
| Scalability | Databases are built to handle millions or billions of records efficiently. |
| Querying power | SQL lets you ask precise, complex questions of the data directly, without writing custom code to search through it. |

## 1.3 Database vs Spreadsheet

| | Spreadsheet | Database |
|---|---|---|
| Scale | Comfortable up to tens of thousands of rows | Designed for millions/billions of rows |
| Concurrent access | Difficult, prone to conflicts | Built-in, safe concurrent access |
| Data integrity | Manual, easy to break (e.g. inconsistent formats) | Enforced via constraints and types |
| Querying | Formulas, filters, manual sorting | SQL - a purpose-built query language |
| Relationships | Awkward, usually via manual lookups (VLOOKUP) | Native support via foreign keys and joins |

## 1.4 Relational vs Non-Relational Databases

| | Relational (SQL) | Non-Relational (NoSQL) |
|---|---|---|
| Structure | Fixed tables with defined columns and types | Flexible - documents, key-value pairs, graphs, wide columns |
| Schema | Enforced upfront (schema-on-write) | Often flexible/dynamic (schema-on-read) |
| Relationships | First-class, via foreign keys and joins | Often denormalized or handled at the application level |
| Examples | PostgreSQL, MySQL, SQLite, SQL Server | MongoDB, Redis, Cassandra, DynamoDB |
| Best fit | Structured, relational data with strong consistency needs | High-scale, flexible-schema, or specialized access patterns (e.g. caching, document storage) |

This handbook focuses specifically on **relational databases and SQL**, given their central role in Data Analytics, Data Science, and most professional backend systems.

## 1.5 Tables, Rows, and Columns

A relational database organizes data into **tables** (also called relations). Each table has a fixed set of **columns** (fields, each with a defined data type), and each **row** (record) represents one entity - one customer, one order, one sensor reading.

```
customers
+----+----------+------------------+
| id | name     | email            |
+----+----------+------------------+
| 1  | Ada      | ada@example.com  |
| 2  | Alan     | alan@example.com |
+----+----------+------------------+
```

## 1.6 Primary Keys

A **primary key** is a column (or combination of columns) that uniquely identifies each row in a table. No two rows can share the same primary key value, and it cannot be `NULL`.

```sql
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL
);
```

## 1.7 Foreign Keys and Relationships

A **foreign key** is a column in one table that references the primary key of another table, establishing a relationship between the two.

```sql
CREATE TABLE orders (
    id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    amount REAL,
    FOREIGN KEY (customer_id) REFERENCES customers(id)
);
```

Here, `orders.customer_id` points back to `customers.id` - every order "belongs to" exactly one customer, a relationship explored fully in Part 16 (joins) and Part 17 (design).

## 1.8 Relationships, Conceptually

| Relationship Type | Example |
|---|---|
| One-to-many | One customer has many orders |
| Many-to-many | Many students enroll in many courses (via a junction table) |
| One-to-one | One user has exactly one profile record |

## 1.9 Constraints

Constraints are rules the database enforces automatically to protect data integrity.

| Constraint | Purpose |
|---|---|
| `PRIMARY KEY` | Uniquely identifies each row |
| `FOREIGN KEY` | Enforces a valid relationship to another table |
| `NOT NULL` | Disallows missing values in a column |
| `UNIQUE` | Disallows duplicate values in a column |
| `CHECK` | Enforces a custom condition (e.g. `price >= 0`) |
| `DEFAULT` | Supplies a value automatically when none is given |

## 1.10 Transactions (Preview)

A **transaction** is a group of database operations treated as a single, indivisible unit - either every operation in it succeeds, or none of them take effect. Fully explored in Part 15.

## 1.11 What SQL Is

**SQL** (Structured Query Language) is the standard language used to define, query, and manipulate data in a relational database. It is declarative - you describe *what* data you want, and the database engine determines *how* to retrieve it efficiently.

## 1.12 DBMS vs RDBMS

| Term | Meaning |
|---|---|
| DBMS (Database Management System) | Any software system that manages databases - relational or not. |
| RDBMS (Relational DBMS) | A DBMS specifically built around the relational model - tables, rows, columns, SQL. PostgreSQL, MySQL, SQLite, and SQL Server are all RDBMSs. |

# 3. SQLite with Python

## 3.1 Why Start With sqlite3

Python's `sqlite3` module is built into the standard library - no installation required - making it the ideal starting point for learning database programming in Python before moving to production databases like PostgreSQL.

## 3.2 Creating and Connecting to a Database

````python
import sqlite3

connection = sqlite3.connect("company.db")
````

| Line | Explanation |
|---|---|
| `import sqlite3` | Loads the standard library's SQLite driver. |
| `sqlite3.connect("company.db")` | Opens a connection to `company.db`, creating the file if it does not already exist. Passing `":memory:"` instead creates a temporary, in-memory-only database, useful for tests (Part 26). |

## 3.3 Creating Tables

````python
cursor = connection.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS customers (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        email TEXT UNIQUE NOT NULL
    )
""")

connection.commit()
````

| Line | Explanation |
|---|---|
| `connection.cursor()` | Creates a **cursor** - the object used to execute SQL statements and fetch results. |
| `cursor.execute(...)` | Runs the given SQL statement. |
| `CREATE TABLE IF NOT EXISTS` | Creates the table only if it doesn't already exist, avoiding an error on repeated runs. |
| `connection.commit()` | Persists the change to disk (Part 6.4) - without this, the table creation could be lost. |

## 3.4 Inserting Records

````python
cursor.execute(
    "INSERT INTO customers (name, email) VALUES (?, ?)",
    ("Ada Lovelace", "ada@example.com")
)
connection.commit()
print(f"Inserted row id: {cursor.lastrowid}")
````

```
Inserted row id: 1
```

The `?` placeholders are **parameters** - never build SQL with plain string formatting (Part 5.2, Part 7).

## 3.5 Reading Records

````python
cursor.execute("SELECT id, name, email FROM customers")
rows = cursor.fetchall()
for row in rows:
    print(row)
````

```
(1, 'Ada Lovelace', 'ada@example.com')
```

| Method | Behavior |
|---|---|
| `fetchone()` | Returns the next single row, or `None` if there are no more. |
| `fetchall()` | Returns every remaining row as a list of tuples. |
| `fetchmany(n)` | Returns up to `n` remaining rows. |

## 3.6 Updating Records

````python
cursor.execute(
    "UPDATE customers SET email = ? WHERE id = ?",
    ("ada.lovelace@example.com", 1)
)
connection.commit()
print(f"Rows updated: {cursor.rowcount}")
````

```
Rows updated: 1
```

## 3.7 Deleting Records

````python
cursor.execute("DELETE FROM customers WHERE id = ?", (1,))
connection.commit()
print(f"Rows deleted: {cursor.rowcount}")
````

## 3.8 Executing SQL and Parameters (Recap)

````python
cursor.execute("SELECT * FROM customers WHERE name = ?", ("Ada Lovelace",))
cursor.executemany(
    "INSERT INTO customers (name, email) VALUES (?, ?)",
    [("Alan", "alan@example.com"), ("Grace", "grace@example.com")]
)
````

`executemany` runs the same parameterized statement once per tuple in the given list - efficient for bulk inserts.

## 3.9 Transactions, Commit, and Rollback

````python
try:
    cursor.execute("UPDATE accounts SET balance = balance - 100 WHERE id = 1")
    cursor.execute("UPDATE accounts SET balance = balance + 100 WHERE id = 2")
    connection.commit()
except sqlite3.Error as e:
    connection.rollback()
    print(f"Transaction failed, rolled back: {e}")
````

Both updates succeed together, or neither takes effect - explored fully in Part 15.

## 3.10 Closing Connections

````python
cursor.close()
connection.close()
````

Failing to close connections can leak resources (open file handles, locks) - especially important in long-running applications.

## 3.11 Using Context Managers

````python
import sqlite3

with sqlite3.connect("company.db") as connection:
    cursor = connection.cursor()
    cursor.execute("INSERT INTO customers (name, email) VALUES (?, ?)", ("Grace", "grace@example.com"))
    # commit happens automatically on successful exit of the "with" block
````

**Important nuance:** `sqlite3`'s connection context manager automatically commits (on success) or rolls back (on exception) - but it does **not** automatically close the connection. Closing it explicitly, or using a nested `with contextlib.closing(connection):`, remains good practice.

````python
from contextlib import closing

with closing(sqlite3.connect("company.db")) as connection:
    with connection:
        cursor = connection.cursor()
        cursor.execute("SELECT * FROM customers")
        print(cursor.fetchall())
# connection is closed here, automatically
````

# 4. SQL Fundamentals from Python

## 4.1 How SQL Written Inside Python Is Executed

Every example below follows the same pattern: a Python string containing SQL is passed to `cursor.execute()`, the database engine parses and runs it, and results (if any) are retrieved via `fetchall()`/`fetchone()`. This section focuses on the SQL itself; the surrounding Python is intentionally minimal.

````python
cursor.execute(sql_string, parameters)
results = cursor.fetchall()
````

## 4.2 SELECT and WHERE

````python
cursor.execute("SELECT name, email FROM customers WHERE name LIKE ?", ("A%",))
print(cursor.fetchall())
````

`SELECT` chooses which columns to return; `WHERE` filters which rows qualify. `LIKE 'A%'` matches names starting with "A".

## 4.3 ORDER BY

````python
cursor.execute("SELECT name FROM customers ORDER BY name DESC")
````

Sorts results; `ASC` (default) for ascending, `DESC` for descending.

## 4.4 GROUP BY and HAVING

````python
cursor.execute("""
    SELECT customer_id, COUNT(*) AS order_count
    FROM orders
    GROUP BY customer_id
    HAVING COUNT(*) > 2
""")
````

`GROUP BY` collapses rows sharing the same value into groups, typically combined with an aggregate function (Part 4.6). `HAVING` filters *groups* (after aggregation), whereas `WHERE` filters individual rows (before aggregation) - a frequently confused distinction.

## 4.5 DISTINCT and LIMIT

````python
cursor.execute("SELECT DISTINCT customer_id FROM orders")
cursor.execute("SELECT * FROM customers ORDER BY id LIMIT 10")
````

`DISTINCT` removes duplicate rows from the result. `LIMIT` caps the number of rows returned - essential for paging through large result sets (Part 12).

## 4.6 Aggregate Functions

````python
cursor.execute("""
    SELECT COUNT(*), SUM(amount), AVG(amount), MIN(amount), MAX(amount)
    FROM orders
""")
````

| Function | Purpose |
|---|---|
| `COUNT(*)` | Number of rows |
| `SUM(column)` | Total of a numeric column |
| `AVG(column)` | Average value |
| `MIN(column)` / `MAX(column)` | Smallest / largest value |

## 4.7 CASE Expressions

````python
cursor.execute("""
    SELECT name,
           CASE
               WHEN amount >= 1000 THEN 'High Value'
               WHEN amount >= 100 THEN 'Medium Value'
               ELSE 'Low Value'
           END AS tier
    FROM orders
""")
````

`CASE` performs conditional logic directly inside SQL, similar to an `if/elif/else` chain - useful for categorizing data without pulling it into Python first.

## 4.8 Subqueries

````python
cursor.execute("""
    SELECT name FROM customers
    WHERE id IN (SELECT customer_id FROM orders WHERE amount > 500)
""")
````

A **subquery** is a query nested inside another - here, the inner query finds customer IDs with large orders, and the outer query looks up their names.

## 4.9 Common Table Expressions (CTEs)

````python
cursor.execute("""
    WITH big_orders AS (
        SELECT customer_id, amount FROM orders WHERE amount > 500
    )
    SELECT customers.name, big_orders.amount
    FROM customers
    JOIN big_orders ON customers.id = big_orders.customer_id
""")
````

A **CTE** (`WITH ... AS (...)`) defines a named, temporary result set that can be referenced later in the same query - improving readability over deeply nested subqueries, especially for multi-step analytical logic.

## 4.10 Window Functions

````python
cursor.execute("""
    SELECT name, amount,
           RANK() OVER (ORDER BY amount DESC) AS rank,
           SUM(amount) OVER (PARTITION BY customer_id) AS customer_total
    FROM orders
""")
````

A **window function** computes a value across a set of related rows (a "window") without collapsing them into a single row the way `GROUP BY` does - here, ranking every order by amount while also computing each customer's running total, all in one query, with every original row preserved.


# 5. CRUD Operations

## 5.1 Overview

CRUD - **C**reate, **R**ead, **U**pdate, **D**elete - are the four fundamental operations any application performs against a database. This section builds complete, parameterized examples for each.

## 5.2 Why Parameterized Queries Are Preferable to String Concatenation

**Unsafe:**

````python
name = "Ada"
cursor.execute(f"SELECT * FROM customers WHERE name = '{name}'")
````

**Safe:**

````python
name = "Ada"
cursor.execute("SELECT * FROM customers WHERE name = ?", (name,))
````

| Reason | Explanation |
|---|---|
| Security | String concatenation opens the door to SQL injection (Part 7) - a malicious value could alter the query's actual structure. |
| Correctness | Parameters correctly handle special characters (quotes, apostrophes) automatically; manual string building easily breaks on real-world data like `O'Brien`. |
| Performance | The database can cache and reuse the parsed query plan across multiple calls with different parameter values. |

## 5.3 Create

````python
def create_customer(connection, name, email):
    cursor = connection.cursor()
    cursor.execute(
        "INSERT INTO customers (name, email) VALUES (?, ?)",
        (name, email)
    )
    connection.commit()
    return cursor.lastrowid

new_id = create_customer(connection, "Grace Hopper", "grace@example.com")
print(f"Created customer with id {new_id}")
````

## 5.4 Read

````python
def get_customer_by_id(connection, customer_id):
    cursor = connection.cursor()
    cursor.execute("SELECT id, name, email FROM customers WHERE id = ?", (customer_id,))
    return cursor.fetchone()

def get_all_customers(connection):
    cursor = connection.cursor()
    cursor.execute("SELECT id, name, email FROM customers")
    return cursor.fetchall()

print(get_customer_by_id(connection, 1))
print(get_all_customers(connection))
````

## 5.5 Update

````python
def update_customer_email(connection, customer_id, new_email):
    cursor = connection.cursor()
    cursor.execute(
        "UPDATE customers SET email = ? WHERE id = ?",
        (new_email, customer_id)
    )
    connection.commit()
    return cursor.rowcount

rows_changed = update_customer_email(connection, 1, "grace.hopper@example.com")
print(f"Updated {rows_changed} row(s)")
````

## 5.6 Delete

````python
def delete_customer(connection, customer_id):
    cursor = connection.cursor()
    cursor.execute("DELETE FROM customers WHERE id = ?", (customer_id,))
    connection.commit()
    return cursor.rowcount

deleted = delete_customer(connection, 1)
print(f"Deleted {deleted} row(s)")
````

## 5.7 A Complete CRUD Module

````python
import sqlite3

class CustomerRepository:
    def __init__(self, connection):
        self.connection = connection

    def create(self, name, email):
        cursor = self.connection.cursor()
        cursor.execute("INSERT INTO customers (name, email) VALUES (?, ?)", (name, email))
        self.connection.commit()
        return cursor.lastrowid

    def get(self, customer_id):
        cursor = self.connection.cursor()
        cursor.execute("SELECT id, name, email FROM customers WHERE id = ?", (customer_id,))
        return cursor.fetchone()

    def list_all(self):
        cursor = self.connection.cursor()
        cursor.execute("SELECT id, name, email FROM customers")
        return cursor.fetchall()

    def update_email(self, customer_id, new_email):
        cursor = self.connection.cursor()
        cursor.execute("UPDATE customers SET email = ? WHERE id = ?", (new_email, customer_id))
        self.connection.commit()
        return cursor.rowcount

    def delete(self, customer_id):
        cursor = self.connection.cursor()
        cursor.execute("DELETE FROM customers WHERE id = ?", (customer_id,))
        self.connection.commit()
        return cursor.rowcount
````

Wrapping CRUD logic inside a dedicated class (a **repository**, a common professional pattern) keeps SQL details isolated from the rest of an application's business logic.


# 6. Database Connections

## 6.1 Connection Objects

A **connection object** represents an active link between your Python program and the database - through it, transactions are managed and cursors are created.

````python
connection = sqlite3.connect("company.db")
````

## 6.2 Cursor Objects

A **cursor object**, created from a connection, is what actually executes SQL statements and retrieves results. A single connection can have multiple cursors.

## 6.3 Connection Lifecycle

| Stage | What Happens |
|---|---|
| Open | `sqlite3.connect(...)` (or the equivalent for other drivers, Part 8) establishes the connection. |
| Use | Cursors execute statements; transactions begin implicitly on the first data-modifying statement. |
| Commit / Rollback | Changes are made permanent, or discarded (Part 6.5-6.6). |
| Close | The connection is released, freeing underlying resources. |

## 6.4 Commit

````python
cursor.execute("INSERT INTO customers (name, email) VALUES (?, ?)", ("Ada", "ada@example.com"))
connection.commit()
````

`commit()` makes all changes made since the last commit permanent and visible to other connections.

## 6.5 Rollback

````python
try:
    cursor.execute("INSERT INTO customers (name, email) VALUES (?, ?)", ("Ada", "ada@example.com"))
    cursor.execute("INSERT INTO invalid_table VALUES (1)")  # will raise an error
    connection.commit()
except sqlite3.Error:
    connection.rollback()
    print("Changes rolled back")
````

`rollback()` discards all uncommitted changes since the last commit, restoring the database to its last known-good state.

## 6.6 Autocommit

````python
connection = sqlite3.connect("company.db", isolation_level=None)   # autocommit mode
````

With `isolation_level=None`, every statement is committed immediately - no explicit `commit()` needed. This trades transactional control for simplicity, and is generally discouraged for multi-statement operations that need atomicity (Part 15).

## 6.7 Connection Errors

````python
try:
    connection = sqlite3.connect("/invalid/path/company.db")
except sqlite3.OperationalError as e:
    print(f"Could not connect: {e}")
````

For server-based databases (PostgreSQL, MySQL), connection errors more commonly arise from unreachable hosts, incorrect credentials, or the server being down - handled the same way, with driver-specific exception types (Part 8).

## 6.8 Connection Timeouts

````python
connection = sqlite3.connect("company.db", timeout=10)
````

If the database is locked (e.g. another process holds a write lock), `timeout` controls how many seconds to wait before raising an error, rather than blocking indefinitely.

## 6.9 Resource Management

````python
from contextlib import closing

def get_customers():
    with closing(sqlite3.connect("company.db")) as connection:
        with connection:
            cursor = connection.cursor()
            cursor.execute("SELECT * FROM customers")
            return cursor.fetchall()
````

Professional code should ensure connections (and cursors, where relevant) are always closed - even when an exception occurs - which is precisely what context managers guarantee (Part 3.11).


# 7. SQL Injection and Security

## 7.1 What SQL Injection Is

**SQL injection** is a security vulnerability where an attacker manipulates a query's actual structure by injecting malicious input into a string that gets concatenated directly into SQL - potentially reading, modifying, or destroying data the application was never meant to expose.

## 7.2 How Unsafe Queries Are Constructed

````python
username = input("Username: ")
query = f"SELECT * FROM users WHERE username = '{username}'"
cursor.execute(query)
````

If a user enters:
```
' OR '1'='1
```

The resulting query becomes:
```sql
SELECT * FROM users WHERE username = '' OR '1'='1'
```

Since `'1'='1'` is always true, this returns **every row in the table** - potentially bypassing an authentication check entirely. A more destructive input could append `; DROP TABLE users; --`, deleting the entire table.

## 7.3 Parameterized Queries: The Fix

````python
username = input("Username: ")
cursor.execute("SELECT * FROM users WHERE username = ?", (username,))
````

With parameters, the database driver treats `username` strictly as a **data value**, never as part of the SQL structure - no matter what characters it contains, it cannot alter the query's meaning.

## 7.4 Prepared Statements

Parameterized queries are built on **prepared statements** - the database parses and plans the query structure once, with placeholders for values, then substitutes actual values in a separate, safe step at execution time. This is precisely why parameters close off SQL injection: the value substitution happens *after* the query structure is already fixed.

## 7.5 Input Validation

Parameterization prevents SQL injection, but does not replace ordinary input validation for correctness:

````python
def create_user(connection, username, age):
    if not username or len(username) > 50:
        raise ValueError("Invalid username")
    if not isinstance(age, int) or age < 0:
        raise ValueError("Invalid age")
    cursor = connection.cursor()
    cursor.execute("INSERT INTO users (username, age) VALUES (?, ?)", (username, age))
    connection.commit()
````

## 7.6 Least Privilege

A database user/role used by an application should have **only** the permissions it actually needs - a reporting service should typically have read-only access; an application handling user signups needs `INSERT`/`UPDATE` on specific tables, not full administrative rights. This limits the damage possible if application credentials are ever compromised.

```sql
-- Example (PostgreSQL): a read-only role for an analytics service
CREATE ROLE analytics_reader WITH LOGIN PASSWORD 'secure_password';
GRANT SELECT ON ALL TABLES IN SCHEMA public TO analytics_reader;
```

## 7.7 Credentials Management and Environment Variables

**Wrong - hardcoded credentials:**

````python
connection = psycopg2.connect(
    host="db.example.com",
    user="admin",
    password="SuperSecret123"   # never do this
)
````

**Correct - credentials from environment variables:**

````python
import os
import psycopg2

connection = psycopg2.connect(
    host=os.environ["DB_HOST"],
    user=os.environ["DB_USER"],
    password=os.environ["DB_PASSWORD"],
)
````

Hardcoded credentials risk being committed to version control and exposed permanently in a repository's history. Environment variables (often loaded from a `.env` file excluded from version control) keep secrets out of source code entirely.

## 7.8 Safe vs Unsafe: Summary Table

| Practice | Unsafe | Safe |
|---|---|---|
| Building queries | String concatenation / f-strings with user input | Parameterized queries (`?` or `%s` placeholders) |
| Storing credentials | Hardcoded in source code | Environment variables / secrets manager |
| Database permissions | Application uses an admin/superuser account | Application uses a scoped, least-privilege role |
| Error messages shown to users | Raw database error text (can leak schema details) | Generic, user-facing error messages; detailed errors only in server-side logs |

| | Core | ORM |
|---|---|---|

## 9.6 Declarative Mapping: Models

````python
from sqlalchemy.orm import DeclarativeBase, Mapped, mapped_column

class Base(DeclarativeBase):
    pass

class Customer(Base):
    __tablename__ = "customers"

    id: Mapped[int] = mapped_column(primary_key=True)
    name: Mapped[str]
    email: Mapped[str] = mapped_column(unique=True)
````

| Line | Explanation |
|---|---|
| `class Base(DeclarativeBase)` | The shared base class every mapped model inherits from. |
| `__tablename__` | The actual SQL table name this class maps to. |
| `Mapped[int]` / `mapped_column(...)` | Modern (SQLAlchemy 2.0-style) type-annotated column definitions. |

## 9.7 Creating Tables From Models

````python
Base.metadata.create_all(engine)
````

Generates and executes the `CREATE TABLE` statements for every model registered against `Base`.

## 9.8 Columns and Constraints

````python
from sqlalchemy import ForeignKey

class Order(Base):
    __tablename__ = "orders"

    id: Mapped[int] = mapped_column(primary_key=True)
    amount: Mapped[float]
    customer_id: Mapped[int] = mapped_column(ForeignKey("customers.id"))
````

## 9.9 Relationships

````python
from sqlalchemy.orm import relationship
from typing import List

class Customer(Base):
    __tablename__ = "customers"
    id: Mapped[int] = mapped_column(primary_key=True)
    name: Mapped[str]
    orders: Mapped[List["Order"]] = relationship(back_populates="customer")

class Order(Base):
    __tablename__ = "orders"
    id: Mapped[int] = mapped_column(primary_key=True)
    amount: Mapped[float]
    customer_id: Mapped[int] = mapped_column(ForeignKey("customers.id"))
    customer: Mapped["Customer"] = relationship(back_populates="orders")
````

`relationship()` lets you navigate between related objects in Python (`customer.orders`, `order.customer`) without writing joins by hand - SQLAlchemy generates the appropriate SQL underneath.

## 9.10 Queries (ORM)

````python
with Session(engine) as session:
    ada = session.query(Customer).filter(Customer.name == "Ada").first()
    print(ada.email)
    print([order.amount for order in ada.orders])

    # Modern 2.0-style query syntax
    from sqlalchemy import select
    stmt = select(Customer).where(Customer.name == "Ada")
    result = session.execute(stmt).scalar_one()
````

````python
with Session(engine) as session:
    new_customer = Customer(name="Grace Hopper", email="grace@example.com")
    session.add(new_customer)
    session.commit()
    print(new_customer.id)   # populated automatically after commit
````

## 9.12 Transactions in SQLAlchemy

````python
with Session(engine) as session:
    try:
        session.add(Customer(name="Ada", email="ada@example.com"))
        session.add(Customer(name="Ada", email="ada@example.com"))  # violates UNIQUE constraint
        session.commit()
    except Exception:
        session.rollback()
        print("Transaction rolled back")
````

# 10. Object-Relational Mapping

## 10.1 What ORM Means

**Object-Relational Mapping (ORM)** is a technique that maps database tables to Python classes, and individual rows to instances of those classes - letting you interact with the database using familiar object-oriented Python, rather than writing raw SQL for every operation.

## 10.2 Python Classes Representing Database Tables

````python
class Customer(Base):
    __tablename__ = "customers"
    id: Mapped[int] = mapped_column(primary_key=True)
    name: Mapped[str]
````

The `Customer` class *is* the `customers` table, from the ORM's perspective - its attributes (`id`, `name`) correspond directly to columns.

## 10.3 Objects Representing Rows

````python
ada = Customer(name="Ada Lovelace")
session.add(ada)
session.commit()
print(ada.id)   # the actual row's primary key, now populated
````

Each `Customer` instance corresponds to exactly one row in the `customers` table.

## 10.4 One-to-One Relationships

````python
class Profile(Base):
    __tablename__ = "profiles"
    id: Mapped[int] = mapped_column(primary_key=True)
    bio: Mapped[str]
    customer_id: Mapped[int] = mapped_column(ForeignKey("customers.id"), unique=True)
    customer: Mapped["Customer"] = relationship(back_populates="profile")

class Customer(Base):
    __tablename__ = "customers"
    id: Mapped[int] = mapped_column(primary_key=True)
    profile: Mapped["Profile"] = relationship(back_populates="customer", uselist=False)
````

A `UNIQUE` constraint on the foreign key, combined with `uselist=False`, enforces that each customer has **at most one** profile.

## 10.5 One-to-Many Relationships (Recap)

Covered in Part 9.9: one `Customer` can have many `Order` records, expressed with `relationship()` on both sides and a plain (non-unique) foreign key.

## 10.6 Many-to-Many Relationships

````python
from sqlalchemy import Table, Column

student_course = Table(
    "student_course",
    Base.metadata,
    Column("student_id", ForeignKey("students.id"), primary_key=True),
    Column("course_id", ForeignKey("courses.id"), primary_key=True),
)

class Student(Base):
    __tablename__ = "students"
    id: Mapped[int] = mapped_column(primary_key=True)
    name: Mapped[str]
    courses: Mapped[List["Course"]] = relationship(secondary=student_course, back_populates="students")

class Course(Base):
    __tablename__ = "courses"
    id: Mapped[int] = mapped_column(primary_key=True)
    title: Mapped[str]
    students: Mapped[List["Student"]] = relationship(secondary=student_course, back_populates="courses")
````

A many-to-many relationship requires a **junction table** (`student_course`) holding pairs of foreign keys - `secondary=student_course` tells SQLAlchemy to route through it automatically.

````python
ada = Student(name="Ada")
python_course = Course(title="Python 101")
ada.courses.append(python_course)
session.add(ada)
session.commit()
````

## 10.7 Advantages of ORMs

| Advantage | Explanation |
|---|---|
| Productivity | Less boilerplate SQL for common operations. |
| Type safety and IDE support | Attribute access is checked and autocompleted, unlike raw SQL strings. |
| Portability | The same Python code often works across different database backends with minimal changes. |
| Relationship navigation | Traversing related objects (`customer.orders`) is natural, without manually writing joins. |
| Security | Query building through the ORM is inherently parameterized, reducing SQL injection risk (Part 7). |

## 10.8 Disadvantages of ORMs

| Disadvantage | Explanation |
|---|---|
| Hidden query cost | It's easy to accidentally trigger many small queries (the "N+1 query problem," Part 14) without realizing it. |
| Learning curve | The ORM's own abstractions (sessions, relationship loading strategies) add real complexity to learn. |
| Complex queries can be awkward | Highly specialized SQL (some window functions, database-specific features) can be harder to express cleanly through an ORM. |
| Performance overhead | Object construction and change-tracking add some overhead versus raw SQL for very high-throughput operations. |

## 10.9 When Raw SQL Is Preferable

| Scenario | Why |
|---|---|
| Complex analytical queries (heavy aggregation, window functions) | Often clearer and more directly optimizable as raw SQL. |
| Bulk data operations (large ETL loads) | Raw SQL or SQLAlchemy Core typically outperforms ORM object-by-object operations. |
| Database-specific features | Some engine-specific SQL extensions have no clean ORM equivalent. |
| Performance-critical hot paths | Avoiding ORM overhead can matter measurably at very high query volumes. |

**Professional guidance:** many production systems use the ORM for typical application CRUD logic, and drop down to Core or raw SQL specifically for reporting, analytics, and bulk operations - the two approaches are not mutually exclusive within a single codebase.

## 11.7 Filtering: SQL vs Pandas

**Filtering in SQL (preferred for large tables):**

````python
df = pd.read_sql_query("SELECT * FROM orders WHERE amount > 500", connection)
````

**Filtering in pandas (after loading everything):**

````python
df = pd.read_sql_query("SELECT * FROM orders", connection)
df = df[df["amount"] > 500]
````

**Why SQL filtering is generally preferable:** the database only sends the rows you actually need over the wire, reducing memory usage and transfer time - especially significant for large tables (Part 12). Filtering in pandas after loading everything wastes both time and memory on rows immediately discarded.

## 11.8 Aggregation: SQL vs Pandas

````python
# In SQL - computed by the database engine
df = pd.read_sql_query("""
    SELECT customer_id, SUM(amount) AS total, COUNT(*) AS order_count
    FROM orders
    GROUP BY customer_id
""", connection)

# In pandas - computed after loading raw rows
raw = pd.read_sql_query("SELECT customer_id, amount FROM orders", connection)
summary = raw.groupby("customer_id").agg(total=("amount", "sum"), order_count=("amount", "count"))
````

Both are valid; SQL aggregation reduces the data transferred (only summary rows come back), while pandas aggregation is convenient when the raw rows are also needed for other analysis in the same session.

## 11.9 Joining: SQL vs Pandas

````python
# Joining in SQL
df = pd.read_sql_query("""
    SELECT customers.name, orders.amount
    FROM customers
    JOIN orders ON customers.id = orders.customer_id
""", connection)

# Joining in pandas
customers_df = pd.read_sql_query("SELECT * FROM customers", connection)
orders_df = pd.read_sql_query("SELECT * FROM orders", connection)
merged = pd.merge(customers_df, orders_df, left_on="id", right_on="customer_id")
````

Joining in SQL is almost always more efficient for large tables (the database has indexes and optimized join algorithms, Part 13-14); joining in pandas is convenient when data originates from multiple different sources (e.g. one table plus an API response) that cannot both be expressed in a single SQL query.

## 11.10 Date Handling

````python
df = pd.read_sql_query(
    "SELECT * FROM orders",
    connection,
    parse_dates=["order_date"]
)
print(df["order_date"].dt.month)
````

`parse_dates` tells pandas to convert the given column(s) directly into proper `datetime64` dtype during the read, enabling `.dt` accessor operations immediately, without a separate `pd.to_datetime()` conversion step afterward.

## 11.11 Chunked Reads

````python
chunks = pd.read_sql_query("SELECT * FROM large_table", connection, chunksize=10_000)

total_amount = 0
for chunk in chunks:
    total_amount += chunk["amount"].sum()

print(total_amount)
````

`chunksize` returns an iterator of DataFrames instead of one giant DataFrame, allowing you to process a table far larger than available memory, one manageable piece at a time - explored fully in Part 12.

## 11.12 A Complete ETL Workflow Example

````python
import pandas as pd
import sqlite3

connection = sqlite3.connect("company.db")

# EXTRACT
raw = pd.read_sql_query("SELECT * FROM raw_sales", connection)

# TRANSFORM
raw["sale_date"] = pd.to_datetime(raw["sale_date"])
raw = raw.dropna(subset=["amount"])
raw["amount"] = raw["amount"].round(2)
monthly_summary = (
    raw.groupby(raw["sale_date"].dt.to_period("M"))["amount"]
       .sum()
       .reset_index(name="total_sales")
)
monthly_summary["sale_date"] = monthly_summary["sale_date"].astype(str)

# LOAD
monthly_summary.to_sql("monthly_sales_summary", connection, if_exists="replace", index=False)

print("ETL complete")
````

This mirrors a genuine, professional analytics workflow: pulling raw transactional data out of a database, cleaning and aggregating it in pandas, and writing a reusable summary table back - fully expanded in Part 18.


# 12. Working with Large Datasets

## 12.1 The Problem: Datasets Larger Than Memory

A table with tens of millions of rows may simply not fit into your machine's available RAM if loaded entirely into a single DataFrame - attempting `pd.read_sql_query("SELECT * FROM huge_table", connection)` on such a table can crash the program or the entire machine.

## 12.2 Chunking and chunksize (Recap)

````python
for chunk in pd.read_sql_query("SELECT * FROM huge_table", connection, chunksize=50_000):
    process(chunk)   # only one chunk is ever in memory at a time
````

## 12.3 Streaming Queries at the Cursor Level

````python
cursor = connection.cursor()
cursor.execute("SELECT * FROM huge_table")

while True:
    rows = cursor.fetchmany(1000)
    if not rows:
        break
    process(rows)
````

`fetchmany(n)` retrieves results in controlled batches directly from the underlying DB-API cursor, without pandas involved at all - useful for lower-level streaming processing.

## 12.4 Pagination

````python
page_size = 100
offset = 0

while True:
    cursor.execute("SELECT * FROM huge_table ORDER BY id LIMIT ? OFFSET ?", (page_size, offset))
    rows = cursor.fetchall()
    if not rows:
        break
    process(rows)
    offset += page_size
````

**Performance caveat:** `OFFSET` on very large tables becomes progressively slower, since the database still has to scan through and discard all the skipped rows. **Keyset pagination** (using the last seen ID instead of an offset) scales far better:

````python
last_id = 0
while True:
    cursor.execute("SELECT * FROM huge_table WHERE id > ? ORDER BY id LIMIT ?", (last_id, page_size))
    rows = cursor.fetchall()
    if not rows:
        break
    process(rows)
    last_id = rows[-1][0]   # assumes id is the first column
````

## 12.5 Server-Side Filtering

````python
# Inefficient: fetch everything, filter in Python
df = pd.read_sql_query("SELECT * FROM huge_table", connection)
df = df[df["status"] == "active"]

# Efficient: filter at the database
df = pd.read_sql_query("SELECT * FROM huge_table WHERE status = 'active'", connection)
````

Always push filtering as close to the data as possible - the database can use indexes (Part 13) to skip irrelevant rows entirely, rather than transferring them just to discard them in Python.

## 12.6 Selecting Only Required Columns

````python
# Inefficient
df = pd.read_sql_query("SELECT * FROM huge_table", connection)

# Efficient
df = pd.read_sql_query("SELECT id, amount, order_date FROM huge_table", connection)
````

`SELECT *` transfers every column, even ones never used downstream - explicitly naming only the needed columns reduces both memory usage and network/disk I/O.

## 12.7 Aggregating Inside SQL

````python
# Inefficient: pull millions of raw rows just to sum them in pandas
df = pd.read_sql_query("SELECT amount FROM huge_table", connection)
total = df["amount"].sum()

# Efficient: let the database do the aggregation
result = pd.read_sql_query("SELECT SUM(amount) AS total FROM huge_table", connection)
total = result["total"][0]
````

## 12.8 Indexing (Preview)

Proper indexes (Part 13) dramatically speed up the filtering and pagination patterns above - without one, the database must scan every row to find matches, regardless of how well-written the surrounding Python is.

## 12.9 Inefficient vs Efficient: Summary Comparison

| Task | Inefficient Approach | Efficient Approach |
|---|---|---|
| Filtering | Load everything, filter in pandas | Filter with `WHERE` in SQL |
| Column selection | `SELECT *` | Name only the needed columns |
| Aggregation | Load raw rows, aggregate in pandas | `SUM`/`AVG`/`COUNT` in SQL |
| Large table export | One giant `read_sql_query()` call | `chunksize`-based iteration |
| Deep pagination | `OFFSET`-based paging | Keyset (id-based) pagination |
| Repeated lookups by a column | No index on that column | A proper index on that column |


# 13. Database Indexing

## 13.1 What an Index Is

A **database index** is a separate, ordered data structure (typically a B-tree) that the database maintains alongside a table, allowing it to locate matching rows quickly without scanning the entire table - conceptually similar to the index at the back of a textbook, which lets you jump directly to relevant pages instead of reading the whole book.

## 13.2 Why Indexes Improve Query Performance

```sql
SELECT * FROM customers WHERE email = 'ada@example.com';
```

Without an index on `email`, the database must perform a **full table scan** - checking every single row. With an index on `email`, the database can jump almost directly to the matching row(s), similar in spirit to a binary search.

## 13.3 Creating an Index

```sql
CREATE INDEX idx_customers_email ON customers(email);
```

````python
cursor.execute("CREATE INDEX idx_customers_email ON customers(email)")
connection.commit()
````

## 13.4 Single-Column Indexes

```sql
CREATE INDEX idx_orders_customer_id ON orders(customer_id);
```

Speeds up queries filtering or joining on `customer_id` specifically.

## 13.5 Composite (Multi-Column) Indexes

```sql
CREATE INDEX idx_orders_customer_date ON orders(customer_id, order_date);
```

A composite index is most effective for queries filtering on the *leading* column(s) of the index, in order - this index helps `WHERE customer_id = ?` and `WHERE customer_id = ? AND order_date = ?`, but is far less useful for a query filtering only on `order_date` alone.

## 13.6 Unique Indexes

```sql
CREATE UNIQUE INDEX idx_customers_email_unique ON customers(email);
```

Enforces that no two rows can share the same value in the indexed column(s), in addition to providing the same lookup speed benefit as a regular index. Note: declaring a column `UNIQUE` (Part 1.9) typically creates a corresponding unique index automatically.

## 13.7 When Indexes Hurt Performance

| Cost | Explanation |
|---|---|
| Slower writes | Every `INSERT`/`UPDATE`/`DELETE` must also update every index on the affected table, adding overhead. |
| Storage space | Each index consumes additional disk space, sometimes substantial for large tables. |
| Diminishing returns on low-selectivity columns | Indexing a column with few distinct values (e.g. a boolean flag) often provides little benefit (Part 13.8). |
| Over-indexing | Adding an index for every conceivable query can slow down writes significantly without proportional read benefit. |

## 13.8 Index Selectivity

**Selectivity** measures how effectively an index narrows down the rows: a column with many distinct values (like `email`, nearly unique per row) has high selectivity and benefits greatly from an index. A column with few distinct values (like `is_active`, only `True`/`False`) has low selectivity - an index there often provides little advantage, since the database still ends up examining a large fraction of the table.

## 13.9 Primary-Key Indexes

Most relational databases automatically create an index on a table's primary key - this is precisely why lookups by primary key (`WHERE id = ?`) are typically very fast without any additional setup required.

## 13.10 Query Execution Plans (Preview)

```sql
EXPLAIN QUERY PLAN SELECT * FROM customers WHERE email = 'ada@example.com';
```

The execution plan reveals whether a query actually used an available index, or fell back to a full table scan - the definitive way to confirm an index is genuinely helping, rather than assuming. Fully explored in Part 14.


# 14. Database Performance

## 14.1 Query Optimization: The General Approach

Query optimization means restructuring a query, its indexes, or the schema so the database can retrieve the same correct result while doing meaningfully less work.

## 14.2 EXPLAIN and Execution Plans

```sql
EXPLAIN QUERY PLAN
SELECT * FROM orders WHERE customer_id = 42;
```

```
SEARCH orders USING INDEX idx_orders_customer_id (customer_id=?)
```

`EXPLAIN` (or `EXPLAIN QUERY PLAN` in SQLite; `EXPLAIN ANALYZE` in PostgreSQL) shows exactly how the database intends to execute a query - which indexes it will use, whether it will scan the whole table, and the order operations happen in. This is the primary diagnostic tool for understanding *why* a query is slow.

````python
cursor.execute("EXPLAIN QUERY PLAN SELECT * FROM orders WHERE customer_id = 42")
for row in cursor.fetchall():
    print(row)
````

## 14.3 Avoiding Unnecessary Columns

```sql
-- Wasteful
SELECT * FROM orders;

-- Targeted
SELECT id, amount FROM orders;
```

Fewer columns means less data read from disk and transferred over the network - meaningful at scale (Part 12.6).

## 14.4 Avoiding Unnecessary Joins

```sql
-- Unnecessary: joining a table whose columns are never used
SELECT orders.amount
FROM orders
JOIN customers ON orders.customer_id = customers.id;

-- Better, if customer data isn't actually needed
SELECT amount FROM orders;
```

Every join adds computational cost - only join tables whose data the query actually needs.

## 14.5 Filtering Early

```sql
-- Filters after an expensive join across the whole table
SELECT customers.name, orders.amount
FROM customers
JOIN orders ON customers.id = orders.customer_id
WHERE orders.order_date > '2026-01-01';
```

Modern query planners are often smart enough to reorder this automatically, but explicitly filtering on indexed columns (Part 13) as early and precisely as possible remains a reliable habit, and is essential for the planner to have any chance of using an index effectively.

## 14.6 Index Usage (Recap)

Confirm indexes are actually being used with `EXPLAIN` (Part 14.2) rather than assuming - a poorly written query (e.g. applying a function to an indexed column, like `WHERE UPPER(email) = ?`) can silently prevent the database from using an available index at all.

## 14.7 Batch Inserts

````python
# Inefficient: one round trip per row
for row in data:
    cursor.execute("INSERT INTO orders (amount) VALUES (?)", (row,))
    connection.commit()

# Efficient: one round trip for many rows
cursor.executemany("INSERT INTO orders (amount) VALUES (?)", [(row,) for row in data])
connection.commit()
````

Committing after every single row multiplies transactional overhead unnecessarily; batching inserts (and committing once at the end, or in reasonably sized batches) is significantly faster for bulk loads.

## 14.8 Transactions and Performance

Wrapping many related writes inside a single transaction (Part 15) - rather than auto-committing each one individually - reduces overhead, since the database does not need to durably flush changes to disk after every individual statement.

## 14.9 Connection Reuse

# 15. Database Transactions

## 15.1 What a Transaction Is

A **transaction** groups one or more database operations into a single, all-or-nothing unit of work: either every operation succeeds and is committed, or, if anything fails partway through, every operation is rolled back as if none of them had ever happened.

## 15.2 ACID Properties

| Property | Meaning |
|---|---|
| **A**tomicity | A transaction's operations all succeed together, or all fail together - no partial results. |
| **C**onsistency | A transaction moves the database from one valid state to another, never violating defined constraints. |
| **I**solation | Concurrent transactions do not interfere with each other's intermediate, uncommitted state. |
| **D**urability | Once committed, a transaction's changes survive even a subsequent crash or power loss. |

## 15.3 A Classic Example: Bank Transfer

````python
def transfer_funds(connection, from_id, to_id, amount):
    cursor = connection.cursor()
    try:
        cursor.execute("UPDATE accounts SET balance = balance - ? WHERE id = ?", (amount, from_id))
        cursor.execute("UPDATE accounts SET balance = balance + ? WHERE id = ?", (amount, to_id))
        connection.commit()
    except Exception as e:
        connection.rollback()
        raise
````

**Why atomicity matters here:** if the first update succeeds but the second fails (e.g. a crash, a constraint violation), rolling back ensures money is never subtracted from one account without correspondingly being added to the other - a partial transfer would corrupt the data.

## 15.4 Commit and Rollback (Recap)

Covered in Part 6.4-6.5 - `commit()` makes changes permanent; `rollback()` discards uncommitted changes since the last commit.

## 15.5 Savepoints

A **savepoint** marks an intermediate point within a larger transaction that you can roll back to, without discarding the entire transaction.

````python
cursor.execute("BEGIN")
cursor.execute("INSERT INTO orders (amount) VALUES (100)")
cursor.execute("SAVEPOINT before_risky_step")

try:
    cursor.execute("INSERT INTO orders (amount) VALUES (?)", (invalid_value,))
except sqlite3.Error:
    cursor.execute("ROLLBACK TO SAVEPOINT before_risky_step")

cursor.execute("INSERT INTO orders (amount) VALUES (200)")
connection.commit()
````

This allows recovering from a failure in one part of a larger transaction, without losing the valid work already done earlier in the same transaction.

## 15.6 Transaction Isolation Levels

Isolation levels control how much concurrent transactions can "see" of each other's uncommitted changes, trading off strictness against concurrency/performance.

| Level | Behavior |
|---|---|
| Read Uncommitted | Transactions may see other transactions' uncommitted changes ("dirty reads") - rarely used in practice. |
| Read Committed | Only ever sees data that has been committed by other transactions (PostgreSQL's default). |
| Repeatable Read | Guarantees that if you read the same row twice within a transaction, you get the same result both times. |
| Serializable | The strictest level - transactions behave as if executed one at a time, in some serial order, fully preventing concurrency anomalies at the cost of the most contention. |

````python
# PostgreSQL example via psycopg2
connection.set_isolation_level(psycopg2.extensions.ISOLATION_LEVEL_SERIALIZABLE)
````

## 15.7 Choosing an Isolation Level in Practice

| Scenario | Reasonable Choice |
|---|---|
| Typical web application CRUD | Read Committed (usually the database's sensible default) |
| Financial transfers, inventory counts | Repeatable Read or Serializable, to prevent subtle concurrency bugs |
| Read-only analytics queries | Read Committed is generally sufficient |

Stricter isolation levels reduce the risk of subtle concurrency bugs but can increase contention (transactions waiting on each other) under heavy concurrent load - a genuine tradeoff to make deliberately, not by default.


# 16. Database Relationships and Joins

## 16.1 Setup for Examples

```sql
CREATE TABLE customers (id INTEGER PRIMARY KEY, name TEXT);
CREATE TABLE orders (id INTEGER PRIMARY KEY, customer_id INTEGER, amount REAL);
```

## 16.2 INNER JOIN

```sql
SELECT customers.name, orders.amount
FROM customers
INNER JOIN orders ON customers.id = orders.customer_id;
```

Returns only rows where a match exists in **both** tables - customers with no orders, and orders with no matching customer, are excluded.

## 16.3 LEFT JOIN

```sql
SELECT customers.name, orders.amount
FROM customers
LEFT JOIN orders ON customers.id = orders.customer_id;
```

Returns **every** row from the left table (`customers`), with matching data from `orders` where it exists, and `NULL` in the order columns for customers with no orders at all - useful for "find customers who have never ordered" style questions (`WHERE orders.id IS NULL`).

## 16.4 RIGHT JOIN

```sql
SELECT customers.name, orders.amount
FROM customers
RIGHT JOIN orders ON customers.id = orders.customer_id;
```

The mirror image of `LEFT JOIN` - every row from `orders` is kept, with `NULL` customer data for any order lacking a valid matching customer. **Note:** SQLite does not support `RIGHT JOIN` directly; the equivalent is written by swapping the table order and using `LEFT JOIN` instead.

## 16.5 FULL OUTER JOIN

```sql
SELECT customers.name, orders.amount
FROM customers
FULL OUTER JOIN orders ON customers.id = orders.customer_id;
```

Returns every row from **both** tables, matching where possible, with `NULL`s on whichever side lacks a match. Supported in PostgreSQL and SQL Server; SQLite requires emulating it with a `UNION` of a `LEFT JOIN` and a `RIGHT`-equivalent query.

## 16.6 CROSS JOIN

```sql
SELECT sizes.label, colors.name
FROM sizes
CROSS JOIN colors;
```

Produces the **Cartesian product** - every possible combination of rows from both tables. Genuinely useful for generating combinations (e.g. every size paired with every color for a product catalog), but easy to use accidentally (by forgetting a join condition), producing an explosively large, incorrect result set.

## 16.7 SELF JOIN

```sql
SELECT e.name AS employee, m.name AS manager
FROM employees e
JOIN employees m ON e.manager_id = m.id;
```

A **self join** joins a table to itself, typically used for hierarchical data (an `employees` table where each row references its own manager, also a row in the same table).

## 16.8 Joins Comparison Summary

| Join Type | Keeps |
|---|---|
| `INNER JOIN` | Only matching rows from both tables |
| `LEFT JOIN` | All rows from the left table, matched or not |
| `RIGHT JOIN` | All rows from the right table, matched or not |
| `FULL OUTER JOIN` | All rows from both tables, matched or not |
| `CROSS JOIN` | Every combination of rows from both tables |
| `SELF JOIN` | Rows from one table, related to other rows in the same table |

# 17. Database Design

## 17.1 Normalization

**Normalization** is the process of structuring a database's tables to minimize redundancy and avoid data inconsistencies - achieved by progressively applying a series of "normal forms."

## 17.2 First Normal Form (1NF)

A table is in 1NF if every column holds a single, atomic value (no lists or repeated groups packed into one field), and each row is uniquely identifiable.

**Violates 1NF:**
```
| customer | phone_numbers        |
|----------|-----------------------|
| Ada      | 555-1111, 555-2222   |
```

**Satisfies 1NF (separate rows, or a separate table):**
```
| customer | phone_number |
|----------|--------------|
| Ada      | 555-1111     |
| Ada      | 555-2222     |
```

## 17.3 Second Normal Form (2NF)

A table is in 2NF if it is in 1NF, and every non-key column depends on the **entire** primary key (relevant specifically for tables with composite primary keys).

**Violates 2NF** (assume the primary key is `[order_id, product_id]`, but `product_name` depends only on `product_id`, not the full key):
```
| order_id | product_id | product_name | quantity |
```

**Satisfies 2NF** (split into two tables):
```
products: | product_id | product_name |
order_items: | order_id | product_id | quantity |
```

## 17.4 Third Normal Form (3NF)

A table is in 3NF if it is in 2NF, and no non-key column depends on another non-key column (no "transitive" dependencies).

**Violates 3NF** (`city` depends on `zip_code`, not directly on the primary key `customer_id`):
```
| customer_id | zip_code | city |
```

**Satisfies 3NF** (split `zip_code`/`city` into their own table):
```
customers: | customer_id | zip_code |
zip_codes: | zip_code | city |
```

## 17.5 Denormalization

**Denormalization** deliberately reintroduces some redundancy - often by combining normalized tables back together, or duplicating a value - to optimize for read performance at the cost of some write complexity and potential inconsistency risk. Common in analytical/reporting tables (data warehouses, Part 18) where query speed matters more than eliminating every last bit of redundancy.

```sql
-- Denormalized reporting table: order_date and customer_name duplicated
-- directly on each order row, avoiding a join for every report query
CREATE TABLE order_report (
    order_id INTEGER,
    customer_name TEXT,
    order_date DATE,
    amount REAL
);
```

## 17.6 Entity Relationships and Schema Design

Designing a schema starts by identifying the core **entities** (customers, orders, products), their **attributes** (columns), and the **relationships** between them (Part 1.8) - typically sketched as an entity-relationship diagram before any SQL is written.

```
Customer (1) ----< (many) Order (many) >---- (1) Product
                     via order_items junction table
```

## 17.7 Surrogate vs Natural Keys

| | Surrogate Key | Natural Key |
|---|---|---|
| Definition | An artificial identifier with no business meaning (e.g. an auto-incrementing integer) | A real-world attribute used as the key (e.g. an email address, a national ID number) |
| Stability | Never changes | Can change (e.g. an email address is updated) |
| Example | `id INTEGER PRIMARY KEY AUTOINCREMENT` | `email TEXT PRIMARY KEY` |

**Professional guidance:** surrogate keys are generally preferred as primary keys, since natural keys can change over time or turn out to not be as unique as initially assumed - natural keys are better expressed as a separate `UNIQUE` constraint instead.

## 17.8 Referential Integrity

**Referential integrity** means foreign key values always point to a row that actually exists - the database enforces this automatically once a foreign key constraint is defined, preventing "orphaned" records (e.g. an order referencing a `customer_id` that was never a real customer, or has since been deleted without updating dependent rows).

```sql
CREATE TABLE orders (
    id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    FOREIGN KEY (customer_id) REFERENCES customers(id) ON DELETE CASCADE
);
```

`ON DELETE CASCADE` automatically deletes an order when its associated customer is deleted, keeping the data consistent; alternatives include `ON DELETE RESTRICT` (prevent the delete entirely if dependent rows exist) or `ON DELETE SET NULL`.

## 17.9 A Realistic Business Database Design

```sql
CREATE TABLE customers (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL,
    created_at TEXT DEFAULT CURRENT_TIMESTAMP
);

CREATE TABLE products (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    price REAL NOT NULL CHECK (price >= 0)
);

CREATE TABLE orders (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_id INTEGER NOT NULL,
    order_date TEXT DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (customer_id) REFERENCES customers(id)
);

CREATE TABLE order_items (
    order_id INTEGER NOT NULL,
    product_id INTEGER NOT NULL,
    quantity INTEGER NOT NULL CHECK (quantity > 0),
    unit_price REAL NOT NULL,
    PRIMARY KEY (order_id, product_id),
    FOREIGN KEY (order_id) REFERENCES orders(id),
    FOREIGN KEY (product_id) REFERENCES products(id)
);
```

This design applies 3NF principles throughout: `order_items` acts as the many-to-many junction between `orders` and `products` (Part 10.6), pricing at the time of sale is captured on `order_items` rather than only on `products` (since prices change over time), and every relationship is protected by a foreign key.

````python
def run_batch_scoring(connection, model, feature_columns, model_version):
    new_data = pd.read_sql_query("""
        SELECT customer_id, age, income, tenure_months
        FROM customer_snapshot
        WHERE customer_id NOT IN (SELECT customer_id FROM churn_predictions)
    """, connection)

    if new_data.empty:
        return 0

    new_data["tenure_years"] = new_data["tenure_months"] / 12
    new_data["income_per_year_tenure"] = new_data["income"] / (new_data["tenure_years"] + 1)

    probabilities = model.predict_proba(new_data[feature_columns])[:, 1]

    results = pd.DataFrame({
        "customer_id": new_data["customer_id"],
        "churn_probability": probabilities,
        "model_version": model_version,
        "predicted_at": pd.Timestamp.now(),
    })
    results.to_sql("churn_predictions", connection, if_exists="append", index=False)
    return len(results)

count = run_batch_scoring(connection, model, feature_columns, "v1.2")
print(f"Scored {count} new customers")
````

# 23. Async Database Programming

## 23.1 Why Async Databases Are Useful

In an application handling many concurrent requests (a web API, for example), a traditional synchronous database call blocks the entire thread while waiting for the database to respond. Async database access lets that same thread handle *other* work while waiting - directly connecting to the concurrency concepts already covered in your Concurrency handbook, applied specifically to database I/O.

## 23.2 Async Connections and Queries

````python
import asyncio
import asyncpg   # an async PostgreSQL driver

async def get_customer(customer_id):
    connection = await asyncpg.connect(
        host="localhost", database="mydb", user="myuser", password="mypassword"
    )
    row = await connection.fetchrow("SELECT * FROM customers WHERE id = $1", customer_id)
    await connection.close()
    return row

customer = asyncio.run(get_customer(1))
````

Notice `await` appears before both the connection and the query - both are genuinely asynchronous I/O operations, allowing other coroutines to run while each one is in progress.

## 23.3 Async SQLAlchemy

````python
from sqlalchemy.ext.asyncio import create_async_engine, AsyncSession

engine = create_async_engine("postgresql+asyncpg://user:password@localhost/mydb")

async def get_customer(customer_id):
    async with AsyncSession(engine) as session:
        from sqlalchemy import select
        result = await session.execute(select(Customer).where(Customer.id == customer_id))
        return result.scalar_one_or_none()
````

## 23.4 Async Transactions

````python
async def transfer_funds(engine, from_id, to_id, amount):
    async with AsyncSession(engine) as session:
        async with session.begin():
            await session.execute(
                text("UPDATE accounts SET balance = balance - :amt WHERE id = :id"),
                {"amt": amount, "id": from_id}
            )
            await session.execute(
                text("UPDATE accounts SET balance = balance + :amt WHERE id = :id"),
                {"amt": amount, "id": to_id}
            )
        # commits automatically on successful exit of session.begin(); rolls back on exception
````

## 23.5 Connection Pools in Async Contexts

Async database libraries (like `asyncpg`) provide their own connection pool implementations, since a pool must itself be async-aware to avoid blocking while acquiring a connection:

````python
async def main():
    pool = await asyncpg.create_pool(host="localhost", database="mydb", user="myuser", password="mypassword")
    async with pool.acquire() as connection:
        row = await connection.fetchrow("SELECT * FROM customers WHERE id = $1", 1)
    await pool.close()
````

## 23.6 Integration With asyncio

All the async database calls above must run inside an event loop, exactly like any other coroutine - typically started with `asyncio.run(...)` at the top level of a script, or managed automatically by an async web framework.

## 23.7 Integration With Asynchronous Web Applications

````python
from fastapi import FastAPI

app = FastAPI()

@app.get("/customers/{customer_id}")
async def read_customer(customer_id: int):
    async with AsyncSession(engine) as session:
        result = await session.execute(select(Customer).where(Customer.id == customer_id))
        customer = result.scalar_one_or_none()
        return {"id": customer.id, "name": customer.name}
````

FastAPI (and similar async frameworks) can serve many concurrent requests on a single process efficiently precisely because database calls like this one `await` rather than block - directly connecting back to the decorator-based route patterns from Part 21 of your Decorators handbook, now made async.

## 23.8 When to Use Async Database Access

| Scenario | Recommendation |
|---|---|
| A synchronous script, batch job, or data analysis notebook | Synchronous access (this handbook's default) is simpler and entirely sufficient. |
| A high-concurrency async web API or service | Async database access avoids blocking the event loop, improving throughput under load. |
| Mixing sync ORM code into an async framework | Avoid - blocking calls inside an async application defeat the purpose and can stall the entire event loop; use the async driver/session consistently instead. |


# 24. Connection Pooling

## 24.1 What Connection Pooling Is

A **connection pool** is a cache of already-open database connections, maintained and reused across many requests, rather than opening and closing a brand-new connection for every single database interaction.

## 24.2 Why Repeatedly Opening Connections Is Expensive

Establishing a new database connection typically involves a network round trip, authentication, and session setup on the server side - real, measurable overhead that adds up quickly under load if paid on every single request, particularly for server-based databases like PostgreSQL or MySQL accessed over a network.

## 24.3 Pool Size

````python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://user:password@localhost/mydb",
    pool_size=10,          # number of connections kept open, ready to use
    max_overflow=5,        # extra connections allowed temporarily under heavy load
)
````

Pool size should be tuned based on expected concurrent load and the target database's own connection limits - too small a pool causes requests to queue waiting for a connection; too large a pool can overwhelm the database server itself.

## 24.4 Connection Reuse

````python
with engine.connect() as connection:
    result = connection.execute(text("SELECT * FROM customers"))
````

Using `engine.connect()` via SQLAlchemy does not necessarily open a brand-new physical connection each time - it typically borrows one from the pool, and returns it to the pool (rather than truly closing it) when the `with` block exits.

## 24.5 Pool Exhaustion

If every pooled connection is currently in use and the pool has reached its maximum size (including `max_overflow`), further requests must wait - or, past a configured timeout, fail outright with a pool timeout error.

````python
engine = create_engine(
    "postgresql+psycopg2://user:password@localhost/mydb",
    pool_size=10,
    max_overflow=5,
    pool_timeout=30,   # seconds to wait for an available connection before raising an error
)
````

**Common cause of pool exhaustion:** connections being checked out (e.g. via `engine.connect()`) but never properly returned - typically from forgetting to close a connection or session, or holding one open across a long-running, unrelated operation.

## 24.6 Timeouts

````python
engine = create_engine(
    "postgresql+psycopg2://user:password@localhost/mydb",
    pool_recycle=1800,   # recycle (close and reopen) connections older than 30 minutes
)
````

`pool_recycle` guards against connections becoming stale or being silently dropped by the database server or an intermediate network device after sitting idle too long.

## 24.7 SQLAlchemy Connection Pooling: Summary of Key Settings

| Setting | Purpose |
|---|---|
| `pool_size` | Number of connections kept open and ready in the pool. |
| `max_overflow` | Additional connections allowed temporarily beyond `pool_size` under load. |
| `pool_timeout` | How long to wait for an available connection before raising an error. |
| `pool_recycle` | Maximum connection age (seconds) before it is transparently replaced. |
| `pool_pre_ping` | Test a connection's validity before handing it out, avoiding errors from silently dropped connections. |

````python
engine = create_engine(
    "postgresql+psycopg2://user:password@localhost/mydb",
    pool_size=10,
    max_overflow=5,
    pool_timeout=30,
    pool_recycle=1800,
    pool_pre_ping=True,
)
````

# 25. Database Migrations

## 25.1 What Migrations Are

A **migration** is a versioned, incremental change to a database's schema (adding a table, adding a column, changing a constraint) - tracked as code, applied in a defined order, and reversible where possible.

## 25.2 Why Schemas Change

Applications evolve: new features need new tables or columns, requirements change existing constraints, and performance work adds or removes indexes. Migrations let this evolution happen safely and repeatably, instead of applying ad hoc, undocumented changes directly to a production database.

## 25.3 Migration Files and Schema Versioning

Each migration is typically a small file describing one schema change, applied in sequence - the database itself usually tracks which migrations have already been applied in a dedicated internal table, so re-running the migration tool only applies what's new.

## 25.4 Alembic

**Alembic** is SQLAlchemy's companion migration tool.

```
pip install alembic
alembic init migrations
```

````python
# migrations/versions/0001_add_customers_table.py
from alembic import op
import sqlalchemy as sa

def upgrade():
    op.create_table(
        "customers",
        sa.Column("id", sa.Integer, primary_key=True),
        sa.Column("name", sa.String, nullable=False),
        sa.Column("email", sa.String, unique=True, nullable=False),
    )

def downgrade():
    op.drop_table("customers")
````

```
alembic upgrade head    # apply all pending migrations
alembic downgrade -1    # revert the most recent migration
```

Every migration defines both `upgrade()` (apply the change) and `downgrade()` (reverse it) - the reversal path is what makes migrations safe to roll back if a deployment needs to be undone.

## 25.5 Adding and Removing Columns

````python
def upgrade():
    op.add_column("customers", sa.Column("phone", sa.String, nullable=True))

def downgrade():
    op.drop_column("customers", "phone")
````

## 25.6 Changing Constraints

````python
def upgrade():
    op.alter_column("customers", "email", nullable=False)

def downgrade():
    op.alter_column("customers", "email", nullable=True)
````

## 25.7 Production Migration Practices

| Practice | Why It Matters |
|---|---|
| Always write a `downgrade()` | Provides a safe rollback path if a migration causes problems in production. |
| Test migrations on a staging copy first | Catches issues (locking, long-running alterations) before they affect production. |
| Avoid destructive changes in the same deploy as the code that depends on them | Removing a column the old code still reads from can break a live application mid-deployment. |
| Keep migrations small and focused | Easier to review, test, and roll back individually. |
| Never edit an already-applied migration file | Once applied in any environment, treat it as immutable history; create a new migration for further changes instead. |
| Back up before major schema changes | A safety net for irreversible or high-risk migrations (Part 28.9). |

## 27.6 Logging Performance Issues

````python
import time

def timed_query(connection, sql, params=(), threshold_seconds=1.0):
    start = time.perf_counter()
    cursor = connection.cursor()
    cursor.execute(sql, params)
    result = cursor.fetchall()
    elapsed = time.perf_counter() - start
    if elapsed > threshold_seconds:
        logger.warning(f"Slow query ({elapsed:.2f}s): {sql}")
    return result
````

## 30.1 Python Database Programming

**Q1. What is the DB-API, and why does it matter?**
A standard interface (PEP 249) that Python database drivers implement, so code written against one driver's connection/cursor pattern transfers with minimal changes to another (Part 8.2).

**Q2. What is the difference between `fetchone()`, `fetchmany(n)`, and `fetchall()`?**
`fetchone()` returns the next single row or `None`; `fetchmany(n)` returns up to `n` rows; `fetchall()` returns every remaining row at once.

**Q3. Why should you close database connections explicitly or use a context manager?**
To release underlying resources (file handles, network sockets, locks) promptly rather than leaking them, especially in long-running applications.

**Q4. What does `cursor.lastrowid` return?**
The primary key (row id) of the most recently inserted row, for drivers/tables that support it.

**Q5. What does `cursor.rowcount` return after an UPDATE or DELETE?**
The number of rows affected by the statement.

**Q6. What is the difference between `execute()` and `executemany()`?**
`execute()` runs one SQL statement once; `executemany()` runs the same parameterized statement once per item in a sequence of parameter tuples, efficient for bulk operations.

**Q7. Why does SQLite's `sqlite3.connect(":memory:")` matter for testing?**
It creates a fresh, isolated, in-memory database with no disk I/O, ideal for fast, fully isolated tests (Part 26.2).

## 30.2 SQL

**Q8. What is the difference between `WHERE` and `HAVING`?**
`WHERE` filters individual rows before aggregation; `HAVING` filters groups after aggregation (Part 4.4).

**Q9. What does `GROUP BY` do?**
Collapses rows sharing the same value(s) in specified columns into groups, typically combined with aggregate functions.

**Q10. What is a CTE, and why use one over a subquery?**
A named, temporary result set defined with `WITH ... AS (...)`; improves readability over deeply nested subqueries, especially for multi-step logic (Part 4.9).

**Q11. What does a window function do differently from `GROUP BY`?**
It computes aggregate-style values across a set of related rows without collapsing them into fewer output rows - every original row is preserved (Part 4.10).

**Q12. What's the difference between `UNION` and `UNION ALL`?**
`UNION` combines result sets and removes duplicate rows; `UNION ALL` combines them without removing duplicates, and is faster since it skips the deduplication step.

**Q13. What does `DISTINCT` do?**
Removes duplicate rows from a query's result set.

How long (in seconds) to wait for a lock to be released before raising an error, if the database is locked by another connection (Part 6.8).

Object-Relational Mapping; it maps database tables to Python classes and rows to instances, letting you interact with data using object-oriented code instead of raw SQL (Part 10.1).

Fetching a list of parent objects, then triggering one additional query per object to load a related collection - resulting in far more database round trips than necessary; solved with eager loading strategies.

Every additional index slows down writes (since each index must be updated on `INSERT`/`UPDATE`/`DELETE`) and consumes extra storage - over-indexing a write-heavy table can be counterproductive (Part 13.7).

## 32.12 Common Errors Quick Reference

| Error | Typical Cause |
|---|---|
| `sqlite3.OperationalError: database is locked` | Another connection holds a write lock; consider `timeout` (Part 6.8). |

| `psycopg2.OperationalError` | Cannot reach the PostgreSQL server, or bad credentials. |
