## Part 1 — Introduction to Python Keywords

### 1.1 What Is a Keyword?

A keyword is a word that is built into the syntax of the Python language itself. Keywords are not variables, not functions, and not something you import - they are part of the grammar that the Python interpreter uses to understand the structure of your program. When Python's parser reads your source code, it treats keywords as fixed, non-negotiable tokens that trigger specific parsing rules. The word if, for example, always signals the start of a conditional branch; the word for always signals the start of a loop. This is fundamentally different from a name like print or len, which is just a regular object living in a namespace that could, in principle, be reassigned.

Because keywords are part of the grammar, they carry meaning that the interpreter itself understands structurally, not just by looking up a value in a dictionary of names. This is the single most important idea to hold onto throughout this handbook: keywords shape the sentence structure of Python, while everything else (functions, methods, variables, objects) fills in the content of that structure.

### 1.2 What Is a Reserved Word?

"Reserved word" and "keyword" are used interchangeably in Python. A word is "reserved" because the language reserves it exclusively for syntactic use, meaning you are not permitted to repurpose it as an identifier - you cannot name a variable, function, class, or module using a reserved word. Attempting to do so raises a SyntaxError before your program ever runs, because the parser cannot tell whether you mean the keyword's built-in meaning or a new name.

```python
class = "Physics 101"
# SyntaxError: invalid syntax

for = 10
# SyntaxError: invalid syntax
```

Both examples fail at parse time, before Python even begins executing statements. This is different from a NameError or TypeError, which happen at runtime. A reserved-word conflict is caught immediately because the interpreter cannot build a valid parse tree in the first place.

### 1.3 Why Python Has Keywords

Every programming language needs some fixed vocabulary to describe control flow (branching, looping), structure (defining functions and classes), and relationships between pieces of code (importing, scoping, exception handling). Keywords give the parser unambiguous anchors to recognize these structures. If words like if, while, and def could be reassigned by a programmer, the parser would have no reliable way to distinguish "the language feature" from "a variable someone happened to name the same thing," and the grammar of the language would become ambiguous or even impossible to parse deterministically.

- Keywords make the grammar unambiguous - the parser always knows what a keyword means.

- Keywords make code universally readable - if, for, and class mean the same thing in every Python program ever written.

- Keywords enable the interpreter to optimize parsing and bytecode generation, since their behavior is fixed and known in advance.

- Keywords communicate intent instantly to human readers, which is a large part of why Python is considered a readable language.

### 1.4 Keywords vs Variables

A variable is a name you choose that is bound to a value or object. You can create, rename, reassign, and delete variables freely. A keyword is fixed by the language specification and cannot be used as a variable name at all. The core distinction is ownership: you own your variable names, but you do not own keywords - Python owns them.

```python
total = 100        # 'total' is a variable you defined
is_valid = True    # 'is_valid' is a variable; 'True' is a keyword
```

### 1.5 Keywords vs Functions

A function, whether built-in like print() or user-defined like def greet():, is an object that can be called with parentheses, passed around, stored in a variable, and even reassigned. A keyword can never be called, stored, or reassigned - it has no independent existence as an object at all. You cannot write x = if, and you cannot pass return as an argument to another function.

```python
my_print = print   # legal: functions are ordinary objects
my_print("hi")     # works fine, prints 'hi'

my_if = if         # SyntaxError: keywords cannot be assigned
```

### 1.6 Keywords vs Methods

A method is a function that is attached to an object and accessed through dot notation, such as "hello".upper() or my_list.append(3). Methods belong to a class and operate on an instance of that class. Keywords belong to the language itself and are never attached to objects - there is no such thing as some_object.if() or 5.for().

### 1.7 Keywords vs Built-in Functions

Built-in functions such as print(), len(), range(), and isinstance() are always available without an import statement, just like keywords, and this is exactly why beginners frequently confuse the two. However, built-in functions are ordinary callable objects that live in the builtins module. You can inspect them, reassign them (though it is bad practice), and pass them as arguments. Keywords have none of these properties.

| Aspect | Keyword (e.g. if) | Built-in Function (e.g. len) |
| --- | --- | --- |
| Can be called with () | No | Yes - len(x) |
| Can be reassigned | No - SyntaxError | Technically yes (not recommended) |
| Has a type()  | No - not an object | Yes - it is a builtin_function_or_method |
| Requires import | No | No (lives in builtins) |
| Part of the grammar | Yes | No - ordinary name lookup |

### 1.8 Why Keywords Cannot Normally Be Used as Variable Names

If Python allowed class = 5, the parser would face a genuine ambiguity every time it encountered the token class: is this the start of a class definition, or is it a reference to a variable named class? Rather than trying to resolve this ambiguity with complicated look-ahead rules, Python's grammar simply forbids it outright. This keeps the grammar simple, fast to parse, and predictable for every Python programmer, in every codebase, everywhere.

### 1.9 How Python Recognizes Keywords

Keyword recognition happens at the tokenizer stage, before the parser builds any tree structure. As the source code is scanned character by character, the tokenizer groups characters into identifiers (names) and then checks each identifier against a fixed internal list of reserved words. If a match is found, the token is emitted as a keyword token rather than a NAME token, and the grammar rules for that keyword apply from that point forward. This lookup is essentially instantaneous because CPython stores keywords in a highly efficient internal set.

### 1.10 Viewing Python's Keywords With the keyword Module

Python ships a standard library module, appropriately named keyword, that exposes the exact list of reserved words for the running interpreter version. This is the authoritative, programmatic way to check whether a given word is reserved, rather than relying on memory or an outdated table found online.

```python
import keyword

print(keyword.kwlist)
# ['False', 'None', 'True', 'and', 'as', 'assert', 'async',
#  'await', 'break', 'class', 'continue', 'def', 'del', 'elif',
#  'else', 'except', 'finally', 'for', 'from', 'global', 'if',
#  'import', 'in', 'is', 'lambda', 'nonlocal', 'not', 'or',
#  'pass', 'raise', 'return', 'try', 'while', 'with', 'yield']

print(keyword.iskeyword("class"))   # True
print(keyword.iskeyword("print"))   # False - print is a builtin, not a keyword

print(keyword.softkwlist)
# ['_', 'case', 'match', 'type']

print(keyword.issoftkeyword("match"))  # True
```

> **Tip:** keyword.iskeyword(name) is the single most reliable way to settle any argument about whether something is 'really' a Python keyword. If it returns False, the word is a built-in function, a method, or ordinary convention (like self) - not a keyword.

### 1.12 Hard Keywords vs Soft Keywords - a First Look

Python distinguishes two categories of reserved words. Hard keywords are reserved everywhere and can never be used as identifiers under any circumstance. Soft keywords are reserved only in the specific grammatical position where they carry special meaning; outside that position, they remain perfectly legal variable names. This distinction is covered in full depth in Part 14, but it is worth previewing here because several later parts (especially Part 11 on match/case) depend on understanding it.

```python
match = 10          # Legal! 'match' is only special as a statement keyword
print(match)        # 10

match match:        # Legal: using the soft keyword AND the variable together
    case 10:
        print("ten")
```

## Part 2 — Conditional Keywords

Conditional keywords let a program choose which block of code to execute based on whether a condition evaluates to True or False. Python has exactly three conditional keywords: if, elif, and else. Together they form a single connected construct - elif and else cannot exist without a preceding if.

### 2.1 if

The if keyword begins a conditional statement. It evaluates the expression that follows it, and if that expression is truthy, the indented block beneath it runs. If the expression is falsy, the block is skipped entirely.

#### Syntax

```python
if condition:
    statement(s)
```

#### Conditions and Boolean Expressions

The condition after if does not have to be a literal True or False - it can be any expression, and Python converts the result to a boolean using its truthiness rules. Numbers other than zero, non-empty strings, non-empty collections, and most objects are truthy; zero, empty strings, empty collections, and None are falsy.

```python
age = 20
if age >= 18:
    print("Eligible to vote")

name = ""
if name:
    print("Name provided")
else:
    print("Name is empty")   # this branch runs, because '' is falsy
```

#### Nested if

An if block can contain another if block inside it. Nesting lets you express conditions that depend on other conditions already being true, but deep nesting quickly hurts readability, so it is usually better to combine conditions with and/or (see Part 8) once you go beyond two or three levels.

```python
score = 85
passed = True
if passed:
    if score >= 90:
        print("Grade: A")
    else:
        print("Passed, but below A")
```

#### Multiple Conditions

```python
temperature = 30
humidity = 80
if temperature > 25 and humidity > 70:
    print("Hot and humid")
```

#### Common Mistakes

- Using = instead of == inside a condition (assignment vs comparison) - Python actually catches this at parse time and raises a SyntaxError, unlike some other languages.

- Forgetting the colon at the end of the if line.

- Inconsistent indentation between the if block and sibling code, which raises an IndentationError.

- Writing if x == True: instead of the simpler and more idiomatic if x:.

### 2.2 elif

elif (a contraction of 'else if') lets you test additional conditions after an initial if has failed. Python evaluates elif branches strictly in order, from top to bottom, and stops at the first one whose condition is true - it never checks later branches once a match is found.

#### Purpose and Multiple Branches

```python
grade = 72
if grade >= 90:
    letter = "A"
elif grade >= 80:
    letter = "B"
elif grade >= 70:
    letter = "C"
else:
    letter = "F"
print(letter)   # C
```

#### Evaluation Order Matters

Because elif branches are evaluated top-to-bottom and stop at the first success, ordering conditions from most specific to least specific (or vice versa, depending on intent) is essential. Placing a broad condition too early can silently swallow cases meant for a later, narrower branch.

```python
# BUGGY: the first branch catches everything grade >= 60,
# so 'A' and 'B' branches below it are unreachable.
if grade >= 60:
    letter = "Pass"
elif grade >= 90:
    letter = "A"      # unreachable for any passing grade
```

### 2.3 else

else defines the fallback branch that runs only when every preceding if and elif condition was false. It takes no condition of its own - it is unconditional by definition, which is exactly what makes it the 'catch-all'.

#### Relationship With if and elif

else is optional. A conditional chain can be a bare if, an if/else pair, an if/elif chain with no else, or a full if/elif/.../else chain. There is at most one else per conditional chain, and it must be the last branch.

#### Nested Usage

```python
def classify(n):
    if n > 0:
        if n % 2 == 0:
            return "positive even"
        else:
            return "positive odd"
    else:
        return "zero or negative"
```

> **Cross-reference:** else also appears after for/while loops and inside try/except (Parts 3 and 6). In every one of those positions its meaning is 'run this only if nothing else interrupted the flow' - a consistent theme worth noticing across the whole language.

### 2.4 Progressively Harder Examples

#### Beginner

```python
num = int(input("Enter a number: "))
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

#### Intermediate

```python
def categorize_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"
```

## Part 3 — Loop Keywords

Loop keywords let a block of code run repeatedly. Python provides two looping keywords - for and while - plus three keywords that modify loop behavior from the inside: break, continue, and pass.

### 3.1 for

The for keyword iterates over the items of an iterable (a sequence such as a list, tuple, or string, or any object implementing the iterator protocol) one at a time, binding each item to a loop variable in turn.

#### Syntax

```python
for item in iterable:
    statement(s)
```

#### Iterating Over Sequences

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

#### Iterating Over Dictionaries

Iterating directly over a dictionary yields its keys. To get keys and values together, use .items(); for values alone, use .values().

```python
prices = {"apple": 1.2, "banana": 0.5}
for key in prices:
    print(key)                       # apple, banana

for key, value in prices.items():
    print(key, "->", value)
```

#### Nested Loops

```python
for row in range(3):
    for col in range(3):
        print(f"({row},{col})", end=" ")
    print()
```

#### Loop Variables and Practical Applications

The loop variable retains whatever value it held after the last completed iteration, even after the loop finishes - a detail that surprises many beginners and is worth testing deliberately. for is the workhorse of data processing: reading rows from a CSV, iterating over API results, building lists with accumulation, and driving nearly every piece of batch computation in data science.

```python
total = 0
for value in [10, 20, 30]:
    total += value
print(total)   # 60
```

### 3.2 while

The while keyword repeats a block for as long as its condition remains truthy. Unlike for, which iterates a known collection, while is condition-controlled - it is the natural choice when you don't know in advance how many iterations you need.

#### Syntax

```python
while condition:
    statement(s)
```

#### Counters

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

#### User Input Loops

```python
user_input = ""
while user_input.lower() != "quit":
    user_input = input("Type 'quit' to exit: ")
```

#### Infinite Loops and Loop Termination

while True: creates a deliberately infinite loop that only ends via an internal break, a return, an exception, or an external interrupt. This pattern is common for servers, game loops, and menus, but every infinite loop must have a clearly reachable exit condition, or the program will hang forever.

```python
while True:
    command = input("Command (or 'exit'): ")
    if command == "exit":
        break
    print("Running:", command)
```

> **Warning:** A while loop whose condition never becomes False, and that has no break, raise, or return inside it, will freeze the program. Always double-check that the condition variable is actually updated inside the loop body.

### 3.3 break

break immediately terminates the nearest enclosing for or while loop, skipping any remaining iterations and any code left in the loop body, and control resumes at the first statement after the loop.

```python
for number in range(1, 100):
    if number == 5:
        break
    print(number)
# prints 1, 2, 3, 4, then stops
```

#### break in Nested Loops

break only exits the innermost loop that directly contains it - it does not cascade upward to outer loops. To exit multiple nested loops at once, common techniques include using a flag variable, restructuring the loops into a function and using return, or raising and catching an exception.

```python
found = False
for i in range(3):
    for j in range(3):
        if i == j == 1:
            found = True
            break          # only exits the inner loop
    if found:
        break              # exits the outer loop too
```

#### Practical Applications

break is essential for search loops (stop once the target is found), for validating input (stop retrying once valid input arrives), and for implementing early termination conditions in simulations.

### 3.4 continue

continue skips the rest of the current iteration's body and jumps straight to the next iteration's condition check (for while) or next item (for for). Unlike break, it does not exit the loop - the loop keeps running.

```python
for number in range(1, 10):
    if number % 2 != 0:
        continue           # skip odd numbers
    print(number)          # prints 2, 4, 6, 8
```

#### Filtering Logic

continue is most often used to filter out unwanted cases at the top of a loop body, keeping the 'main path' of the loop unindented and easy to read, instead of wrapping the entire remaining body in a large if block.

```python
for row in dataset:
    if row["status"] != "active":
        continue
    process(row)   # only reached for active rows
```

### 3.5 pass

pass is a null operation - it does absolutely nothing when executed. Its sole purpose is syntactic: Python requires every block introduced by a colon (if, for, while, def, class, try, with, and so on) to contain at least one statement, and pass satisfies that requirement when you have no code to write yet.

#### Empty Blocks and Placeholder Code

```python
def not_implemented_yet():
    pass

class Placeholder:
    pass

if debug_mode:
    pass   # TODO: add logging later
```

#### Why pass Is Different From continue and break

This is one of the most common points of confusion for beginners, so it deserves a direct comparison: pass means 'do nothing at all, and keep going to the next line as normal.' continue means 'stop this iteration early and jump to the next one.' break means 'stop the loop altogether.' pass has no effect on loop control flow whatsoever - it is purely a filler statement satisfying Python's grammar.

| Keyword | Effect on the current iteration | Effect on the loop overall |
| --- | --- | --- |
| pass | None - execution continues to the next line normally | None - loop behaves as if pass were not there |
| continue | Ends immediately, skipping remaining code in the body | Loop proceeds to the next iteration |
| break | Ends immediately | Loop terminates completely |

## Part 4 — Function-Related Keywords

This part covers the keywords that define and control functions at the language level: def, return, yield, and lambda. Since a full Functions handbook already exists separately, the focus here stays tightly on what each keyword itself does, not on general function design.

### 4.1 def

def introduces a function definition. It binds a name to a new function object, which can then be called later using parentheses. When Python executes a def statement, it does not run the function body - it simply creates the function object and stores it in the current namespace.

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Ada"))   # Hello, Ada!
```

def works hand in hand with return (below) to produce a value, and with yield (below) to produce a generator instead of a normal function. Beyond that core behavior, parameter defaults, *args/**kwargs, and decorators are the subject of the dedicated Functions handbook.

### 4.2 return

return ends the execution of the current function immediately and sends a value back to the caller. Once return executes, no further code in that function call runs, even if it appears before other statements in the function body.

#### Returning Values

```python
def square(n):
    return n * n

print(square(4))   # 16
```

#### Returning Multiple Values

Python does not literally return more than one object - what looks like multiple return values is actually a single tuple being returned, which is then unpacked by the caller.

```python
def min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = min_max([4, 9, 1, 7])
print(lowest, highest)   # 1 9
```

#### Returning None

A function with no return statement, or a bare return with no expression, implicitly returns None. This is a frequent source of subtle bugs when a function is expected to return a computed value but the programmer forgot the return keyword entirely.

```python
def add(a, b):
    result = a + b   # missing 'return result'

value = add(2, 3)
print(value)   # None - not 5!
```

#### Early Return

```python
def safe_divide(a, b):
    if b == 0:
        return None      # early return guards against division by zero
    return a / b
```

#### Return Behavior in Nested Calls

return only exits the function it is directly written in - it never exits an outer caller. Each function call has its own independent return, which is what makes recursion possible.

### 4.3 yield

yield is what turns an ordinary-looking function into a generator function. Instead of computing a full result and returning it all at once, a generator function produces a stream of values one at a time, pausing its execution state between each one.

#### Generators and Generator Functions

Any function that contains at least one yield statement anywhere in its body becomes a generator function. Calling it does not run the body at all - it returns a generator object immediately. The body only executes, up to the next yield, when something asks the generator for its next value (via next() or a for loop).

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

counter = count_up_to(3)
print(next(counter))   # 1
print(next(counter))   # 2
print(next(counter))   # 3
print(next(counter))   # raises StopIteration
```

#### Pausing and Resuming Execution

Each yield suspends the function exactly at that line, preserving every local variable's value in memory. The next call to next() resumes execution immediately after that yield, rather than starting the function over from the top - this is what fundamentally distinguishes a generator from a normal function call.

#### yield vs return

| Aspect | return | yield |
| --- | --- | --- |
| What it produces | A single final value, then the function ends | A stream of values, one per resumption |
| Function state | Discarded once return executes | Preserved (paused) between each yield |
| Calling the function | Runs the whole body immediately | Returns a generator object; body runs lazily |
| Memory use for large data | Must build the whole result in memory | Produces values on demand - memory efficient |

#### Practical Applications

Generators built with yield are ideal for processing files line by line without loading the whole file into memory, streaming data from an API, generating infinite sequences (like an ID counter), and building efficient data pipelines in data science workflows where datasets may not fit comfortably in RAM.

```python
def read_large_file(path):
    with open(path) as f:
        for line in f:
            yield line.strip()

for line in read_large_file("data.txt"):
    print(line)
```

### 4.4 lambda

lambda creates a small, unnamed (anonymous) function inline, in a single expression, without using def. A lambda expression can take any number of arguments but is restricted to exactly one expression, whose result is automatically returned.

#### Syntax

```python
lambda arguments: expression
```

```python
square = lambda x: x * x
print(square(5))   # 25

add = lambda a, b: a + b
print(add(3, 4))   # 7
```

#### Limitations

- A lambda body must be a single expression - it cannot contain statements like if/else blocks (though a conditional expression is allowed), loops, assignments, or multiple lines.

- A lambda has no name of its own unless you assign it to a variable, which somewhat defeats its purpose as an 'anonymous' function.

- Debugging a lambda is harder because tracebacks show '<lambda>' instead of a descriptive function name.

#### Practical Use and Functional Programming

lambda shines when a short function is needed only once, typically as an argument to another function such as sorted(), map(), or filter(). This connects lambda to Python's functional-programming style, where functions are passed around as values rather than always being defined with a formal name.

```python
people = [("Ada", 34), ("Bo", 22), ("Cy", 45)]
people.sort(key=lambda person: person[1])
print(people)   # sorted by age: [('Bo', 22), ('Ada', 34), ('Cy', 45)]

numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda n: n % 2 == 0, numbers))
print(evens)   # [2, 4, 6]
```

> **Best Practice:** If a lambda starts feeling complicated enough that you need to explain it with a comment, that is usually the signal to write a proper named function with def instead.

## Part 5 — Class and Object-Oriented Keywords

This part is unusually important for beginners because it deliberately mixes true keywords (class, is) with things that merely look like keywords but are not (self, super, __init__). Learning to tell these apart precisely is one of the most valuable habits a Python learner can build.

### 5.1 class

class is the only true keyword in this part that defines object-oriented structure. It introduces a new class definition, creating a blueprint from which objects (instances) can be created.

#### Defining Classes and the Class Body

```python
class Dog:
    species = "Canis familiaris"   # class attribute, shared by all instances

    def bark(self):
        return "Woof!"
```

#### Class Attributes vs Instance Attributes

A class attribute is defined directly in the class body and shared by every instance unless overridden. An instance attribute is typically set inside __init__ (see 5.4) and belongs to one specific object.

#### Methods and Objects

```python
fido = Dog()          # fido is an object (an instance of Dog)
print(fido.bark())    # Woof!
print(fido.species)   # Canis familiaris
```

### 5.2 self — Not a Keyword

It must be stated plainly: self is not a Python keyword. keyword.iskeyword("self") returns False. self is simply the conventional name Python programmers use for the first parameter of an instance method, which Python automatically fills in with a reference to the specific object the method was called on. You could technically name that parameter anything - this_object, me, s - and it would work identically, but every Python style guide and virtually every codebase in existence uses self, so deviating from it will confuse other readers even though it will not confuse the interpreter.

#### Why self Is Used

```python
class Counter:
    def increment(self):
        self.count += 1

# 'self' lets the method know WHICH object's 'count' to update
```

#### Instance Attributes and Instance Methods

self.attribute_name = value inside a method creates or updates an attribute that belongs specifically to that one instance. An instance method is any function defined in a class body that takes self as its first parameter - it is the mechanism by which a method can read and modify the state of the particular object it was called on.

> **Critical Distinction:** self is not reserved. self is convention. class, by contrast, genuinely is a reserved keyword. This is precisely the kind of distinction the original request for this handbook specifically asked to be made crystal clear.

### 5.3 super — Not a Keyword (a Built-in Function)

super() is also not a keyword - it is a built-in function that returns a temporary proxy object representing the parent class, allowing a subclass to call methods defined on its parent without naming the parent class explicitly.

#### Parent Classes and Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return f"{self.name} makes a sound."

class Cat(Animal):
    def speak(self):
        base = super().speak()
        return base + " Specifically, a meow."

print(Cat("Whiskers").speak())
# Whiskers makes a sound. Specifically, a meow.
```

#### Method Resolution and Calling Parent Implementations

super() follows Python's Method Resolution Order (MRO), a well-defined algorithm (C3 linearization) that determines the order in which base classes are searched. This matters most in multiple inheritance, where a class inherits from more than one parent, and Python needs a consistent, predictable rule for which parent's method 'wins' or gets called first.

```python
class A:
    def greet(self):
        print("A")

class B(A):
    def greet(self):
        print("B")
        super().greet()

class C(A):
    def greet(self):
        print("C")
        super().greet()

class D(B, C):
    def greet(self):
        print("D")
        super().greet()

D().greet()   # D, B, C, A - following the MRO, not a naive parent lookup
```

### 5.4 __init__ — Not a Keyword (a Dunder Method)

__init__ is a 'dunder' (double-underscore) method, not a keyword. It is a special method that Python automatically calls right after a new object has been created, and its job is to initialize that object's instance attributes.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p1 = Point(3, 4)
print(p1.x, p1.y)   # 3 4
```

__init__ relies on class to exist at all (a class definition is what __init__ lives inside of) and on self to know which specific instance it is initializing - but neither self nor __init__ is itself part of Python's reserved grammar. Only class is.

### 5.5 is — Identity Comparison

is genuinely is a reserved keyword, and it belongs conceptually to this part because it is the correct way to check whether two names refer to the exact same object in memory - a question that comes up constantly in object-oriented code.

#### is vs ==

== asks 'do these two objects have equal value?' is asks 'are these two names pointing at the very same object in memory?' Two different objects can be equal in value while being different objects entirely.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)   # True  - same contents
print(a is b)   # False - two distinct list objects
print(a is c)   # True  - c is literally the same object as a
```

#### The Special Case of None

Because there is only ever one None object in a running Python program, checking against None should always use is, not ==. This is such an established convention that most linters and style guides (including PEP 8) flag == None as a mistake.

```python
value = None
if value is None:      # correct, idiomatic style
    print("No value")
```

## Part 6 — Exception Handling Keywords

Six keywords work together to manage errors gracefully: try, except, else, finally, raise, and assert. Together they let a program anticipate failure, respond to it deliberately, and guarantee that cleanup code still runs even when something goes wrong.

### 6.1 The Exception Handling Flow, at a Glance

- try: the block of code that might raise an exception.

- except: runs only if a matching exception was raised inside try.

- else: runs only if the try block completed with no exception at all.

- finally: always runs, whether an exception occurred or not, and even if a return happened inside try or except.

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Division succeeded:", result)
finally:
    print("This always runs")
```

### 6.2 try

try marks the beginning of a block whose statements Python will attempt to execute, while watching for exceptions. A try block cannot stand alone - it must be followed by at least one except, or a finally, or both.

### 6.3 except

except catches and handles a specific type of exception (or several types) raised inside the corresponding try block. If no exception occurs, every except clause is skipped entirely.

#### Catching Specific Exceptions

```python
try:
    value = int("not a number")
except ValueError:
    print("That wasn't a valid integer.")
```

#### Multiple except Blocks

```python
try:
    numbers = [1, 2, 3]
    print(numbers[10])
except IndexError:
    print("Index out of range")
except ValueError:
    print("Bad value")
```

#### Catching Multiple Exception Types in One Clause

```python
try:
    risky_operation()
except (ValueError, TypeError) as error:
    print("Something went wrong:", error)
```

#### The Exception Hierarchy

Python's built-in exceptions form a class hierarchy rooted at BaseException, with the far more commonly used Exception as its direct subclass. Catching a broad exception like Exception will also catch every one of its subclasses (ValueError, TypeError, KeyError, and so on), which is why except Exception: is sometimes used as a deliberately broad catch-all, but should generally be as narrow as the situation allows.

| Exception | Typical Cause |
| --- | --- |
| ValueError | Right type, wrong value (e.g. int('abc')) |
| TypeError | Operation applied to an incompatible type |
| KeyError | Dictionary key does not exist |
| IndexError | Sequence index out of range |
| ZeroDivisionError | Division or modulo by zero |
| FileNotFoundError | Attempting to open a file that does not exist |
| AttributeError | Accessing an attribute or method that does not exist on an object |

### 6.4 else (in try/except)

else here follows the exact same underlying philosophy as the else you saw in Part 2 and Part 3: 'run this only if nothing interrupted the normal path.' In a try/except, the else block runs only when the try block completes successfully with zero exceptions raised.

```python
try:
    file = open("data.txt")
except FileNotFoundError:
    print("File missing")
else:
    content = file.read()
    file.close()
```

> **Best Practice:** Code that does not need exception protection belongs in else, not in try. Keeping the try block as small as possible makes it far easier to know exactly which line an except clause is actually guarding.

### 6.5 finally

finally guarantees cleanup. Its block executes no matter what happens above it - whether the try block succeeded, an exception was caught, an exception was not caught, or even if a return statement fired inside try or except. This makes it the natural home for releasing resources: closing files, releasing locks, or closing network connections.

```python
def read_config():
    file = open("config.txt")
    try:
        return file.read()
    finally:
        file.close()   # runs even though 'return' fired above it
```

### 6.6 raise

raise deliberately triggers an exception, either one of Python's built-in exception types or a custom exception class you define yourself. Execution stops at that point and control transfers to the nearest matching except (or, if none exists, the program terminates with a traceback).

#### Raising Built-in Exceptions

```python
def withdraw(balance, amount):
    if amount > balance:
        raise ValueError("Insufficient funds")
    return balance - amount
```

#### Custom Exceptions

```python
class InsufficientFundsError(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError("Cannot withdraw more than balance")
    return balance - amount
```

#### Re-raising

```python
try:
    risky_operation()
except ValueError:
    log_error()
    raise      # re-raises the same exception after logging it
```

### 6.7 assert

assert checks that a condition is true and, if it is not, raises an AssertionError immediately, optionally with a custom message. assert statements are primarily a debugging and internal-invariant tool - they are meant to catch programmer errors during development, not to validate ordinary user input in production, because assertions can be globally disabled with the -O interpreter flag.

```python
def calculate_average(numbers):
    assert len(numbers) > 0, "List must not be empty"
    return sum(numbers) / len(numbers)

calculate_average([])   # AssertionError: List must not be empty
```

> **Warning:** Never use assert to validate untrusted user input or enforce business logic that must always run - running Python with the -O flag strips out every assert statement in the program, silently.

### 6.8 raise vs assert

| Aspect | raise | assert |
| --- | --- | --- |
| Purpose | Signal any exceptional condition, deliberately | Verify an internal invariant during development |
| Can be disabled | No - always active | Yes - stripped out with python -O |
| Appropriate for user input validation | Yes | No |
| Exception type | Any exception class you choose | Always AssertionError |

### 6.9 Common Mistakes and Best Practices

- Using a bare except: with no exception type, which silently swallows every possible error, including ones you did not anticipate (like KeyboardInterrupt).

- Putting too much code inside try, making it unclear which line could actually raise the caught exception.

- Forgetting that finally still runs after a return inside try, which can be surprising if finally also contains a return (the finally return silently overrides the try return).

- Catching Exception broadly as a habit rather than catching the specific exception you expect.

- Using assert for validating data that comes from users or external systems.

import loads a module (a .py file, or a package) and makes its contents accessible via dotted attribute access. Python only executes a module's top-level code the first time it is imported in a given process; subsequent imports reuse the already-loaded module object.

### 7.3 as

as creates an alias - a second, usually shorter or clearer, name for something. It appears in several distinct grammatical positions across the language, all unified by the same idea: 'call this thing by another name in the code that follows.'

### 7.4 global

Inside a function, an assignment to a name normally creates a new local variable, even if a variable of the same name exists at module level. global overrides that default, telling Python explicitly that a particular name inside this function refers to the module-level (global) variable, so assignments inside the function actually modify the global value rather than shadowing it.

#### Global Scope and Modifying Global Variables

```python
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
print(counter)   # 2
```

#### Scope Rules Without global

```python
counter = 0

def increment_broken():
    counter += 1   # UnboundLocalError: 'counter' referenced before assignment
                   # because assigning to it anywhere in this function makes
                   # Python treat it as local for the ENTIRE function body
```

#### Why Excessive Use of global Should Be Avoided

Functions that rely on global variables are harder to test in isolation, harder to reason about (their behavior depends on hidden external state), and more prone to bugs when multiple functions modify the same global from different places. Passing values as arguments and returning results explicitly is almost always the more maintainable design.

### 7.5 nonlocal

nonlocal is global's counterpart for nested functions. It tells Python that a name inside an inner function refers to a variable in the nearest enclosing (but not global) scope - typically a variable belonging to an outer function that the inner function is defined within.

#### Enclosing Scopes, Nested Functions, and Closures

```python
def make_counter():
    count = 0
    def increment():
        nonlocal count
        count += 1
        return count
    return increment

counter = make_counter()
print(counter())   # 1
print(counter())   # 2
print(counter())   # 3
```

This pattern - a function returning another function that remembers variables from its enclosing scope - is called a closure, and nonlocal is precisely the keyword that allows the inner function to mutate that remembered state rather than merely read it.

#### global vs nonlocal

| Aspect | global | nonlocal |
| --- | --- | --- |
| Targets | The module-level (global) scope | The nearest enclosing function scope (not global) |
| Requires nested functions | No | Yes - only meaningful inside a nested function |
| Typical use case | Modifying a true module-level variable | Implementing closures / stateful inner functions |
| Works at module top level | N/A (already global) | SyntaxError if there is no enclosing function scope |

## Part 8 — Logical and Special Keywords

This part covers and, or, not, in, and is - keywords that combine or test boolean and membership conditions. Since a full Operators handbook already exists separately, this section stays focused specifically on these five keywords rather than the broader operator system.

### 8.1 and

and evaluates to True only if both operands are truthy. Python's and is short-circuiting: if the left operand is falsy, Python never even evaluates the right operand, because the overall result is already determined.

```python
age = 25
has_id = True
if age >= 18 and has_id:
    print("Entry allowed")

# Short-circuit example - right side never runs:
def risky():
    print("called!")
    return True

result = False and risky()   # 'called!' is never printed
```

Technically, and does not always return a plain True/False - it returns whichever operand actually determined the result, which is a subtle but useful detail: False and "anything" returns False (the left value), while True and "anything" returns "anything" (the right value).

### 8.2 or

or evaluates to True if at least one operand is truthy, and it also short-circuits: if the left operand is truthy, the right operand is never evaluated.

```python
username = ""
display_name = username or "Guest"
print(display_name)   # Guest - because '' is falsy, so the right side is used
```

This idiom, value or default, is extremely common in Python for supplying a fallback when a variable might be empty, zero, or None.

### 8.3 not

not is a unary keyword that inverts a truth value: it turns a truthy expression into False and a falsy expression into True.

```python
is_logged_in = False
if not is_logged_in:
    print("Please log in")

print(not 0)        # True  - 0 is falsy
print(not "hello")  # False - non-empty strings are truthy
```

### 8.4 in

in tests membership: whether a value exists within a sequence, collection, or any object that implements the __contains__ protocol. in is also reused as part of the for loop's grammar (for item in iterable:), which is why it appears in both Part 3 and here - the underlying idea, 'this belongs to that collection,' is the same in both places.

#### Membership Testing

```python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)      # True
print("grape" in fruits)       # False
print("grape" not in fruits)   # True

text = "Hello World"
print("World" in text)   # True - substring membership for strings

prices = {"apple": 1.2}
print("apple" in prices)   # True - checks KEYS by default for dicts
```

### 8.5 is (Revisited)

is was introduced fully in Part 5 as an identity comparison keyword, since it comes up so naturally alongside object-oriented code. It belongs equally to this part conceptually, because logically it sits alongside and, or, and not as a comparison-producing keyword. See section 5.5 for the complete explanation and the None special case.

### 8.6 in vs is

These two are easy to confuse for beginners purely because they are both short, two-or-three-letter keywords used in comparisons - but they test entirely different things.

| Keyword | Question It Answers | Example |
| --- | --- | --- |
| in | Is this value contained within that collection? | 3 in [1, 2, 3]  ->  True |
| is | Are these two names the same object in memory? | a is b  ->  True only if identical object |

### 8.7 and vs &, or vs |

and/or are keywords that operate on truthiness and short-circuit; & and | are operators, covered fully in the separate Operators handbook, that perform bitwise operations on integers and, for pandas/numpy objects, element-wise boolean operations. A very common bug for people coming from pandas is using and/or on Series objects, which raises a ValueError, when & and | (with careful parenthesization) were required instead.

```python
import pandas as pd
df = pd.DataFrame({"age": [15, 25, 35]})

# WRONG: raises ValueError - Python's 'and' can't handle a whole Series
# mask = (df['age'] > 18) and (df['age'] < 30)

# CORRECT: use & for element-wise boolean logic on Series
mask = (df['age'] > 18) & (df['age'] < 30)
print(df[mask])
```

## Part 9 — Asynchronous Programming Keywords

async and await work together to enable asynchronous programming: writing code that can pause while waiting on a slow operation (like a network request), letting other work happen in the meantime, without using traditional multi-threading.

### 9.1 Why Asynchronous Programming Exists

Many real-world operations - reading from a network socket, querying a database, waiting on a file download - spend most of their time waiting, not computing. In traditional synchronous code, the entire program sits idle during that wait. Asynchronous programming lets a single thread juggle many such waiting operations concurrently: while one task is waiting on the network, the program can make progress on another task, dramatically improving throughput for I/O-heavy workloads without the complexity and overhead of full multi-threading.

### 9.2 Synchronous vs Asynchronous Execution

| Aspect | Synchronous | Asynchronous |
| --- | --- | --- |
| Execution model | One operation at a time, in strict sequence | Many operations can be 'in flight', cooperatively |
| Waiting on I/O | Blocks the entire program | Yields control so other tasks can run |
| Best suited for | CPU-bound work | I/O-bound work (network, disk, APIs) |
| Keywords involved | None special | async, await |

### 9.3 async

async is placed before def to declare a coroutine function - a special kind of function that, when called, does not run immediately but instead returns a coroutine object that must be driven by an event loop (typically via asyncio.run() or by being awaited from within another coroutine).

#### async def

```python
import asyncio

async def say_hello():
    print("Hello")
    await asyncio.sleep(1)
    print("World")

asyncio.run(say_hello())
```

async can also appear before with and for (async with, async for) to work with asynchronous context managers and asynchronous iterators, which follow the same underlying philosophy but for resource management and iteration respectively.

### 9.4 await

await can only be used inside a function declared with async def. It suspends the execution of the current coroutine until the awaited operation (another coroutine, or an awaitable object such as an asyncio.sleep() call or a network request) completes, and while suspended, the event loop is free to run other coroutines.

#### Coroutines and the Event Loop

A coroutine is the object created by calling an async def function; it does nothing on its own until it is awaited or scheduled onto an event loop. The event loop is the underlying engine, typically provided by the asyncio module, that keeps track of every pending coroutine and resumes each one when its awaited operation finishes.

```python
import asyncio

async def fetch_data(name, delay):
    print(f"Start fetching {name}")
    await asyncio.sleep(delay)
    print(f"Finished fetching {name}")
    return f"{name} result"

async def main():
    results = await asyncio.gather(
        fetch_data("A", 2),
        fetch_data("B", 1),
    )
    print(results)

asyncio.run(main())
# Both fetches run concurrently; total time is ~2s, not 3s.
```

### 9.5 Practical Applications: API and Network Programming

async/await are heavily used in web frameworks (FastAPI, aiohttp), database drivers with async support, and any application that issues many simultaneous network calls - such as calling several external APIs at once and waiting for all of them to respond, or building a chat server that must handle thousands of open connections without spawning thousands of threads.

### 9.6 Common Mistakes

- Calling an async def function directly and expecting the body to run - it only creates a coroutine object; you must await it or run it via asyncio.run().

- Using await outside of an async def function, which raises a SyntaxError.

- Mixing blocking, synchronous calls (like time.sleep() or a regular requests.get()) inside async code, which defeats the entire purpose by blocking the whole event loop.

- Forgetting to await a coroutine, leaving a 'coroutine was never awaited' RuntimeWarning and the code inside it never actually executing.

```python
async def main():
    fetch_data("A", 2)   # BUG: missing 'await' - this just creates
                         # a coroutine object and does nothing with it
```

## Part 10 — Context Manager Keywords

with (together with as, revisited from Part 7) provides a clean, reliable way to acquire a resource, use it, and guarantee it is released afterward - even if an exception occurs in between.

### 10.1 with

with wraps a block of code in a context manager, an object that defines setup logic (running when the block is entered) and teardown/cleanup logic (running when the block is exited, no matter how it is exited).

#### Resource Management and File Handling

```python
with open("notes.txt", "w") as file:
    file.write("Hello, file!")
# file is automatically closed here, even if write() had raised an exception
```

#### Automatic Cleanup

Without with, cleanup must be handled manually with try/finally, which is more verbose and easier to get wrong. with condenses that entire pattern into a single, readable line.

```python
# Equivalent, manual version using try/finally (Part 6):
file = open("notes.txt", "w")
try:
    file.write("Hello, file!")
finally:
    file.close()
```

#### Multiple Context Managers

```python
with open("input.txt") as infile, open("output.txt", "w") as outfile:
    outfile.write(infile.read().upper())
```

### 10.2 Custom Context Managers and __enter__ / __exit__

Any object can be used with with as long as it implements two dunder methods: __enter__, which runs when the block is entered and returns the value bound by as, and __exit__, which runs when the block is exited and receives details of any exception that occurred (or None if none did).

```python
class Timer:
    def __enter__(self):
        import time
        self.start = time.time()
        return self
    def __exit__(self, exc_type, exc_value, traceback):
        import time
        print(f"Elapsed: {time.time() - self.start:.4f}s")
        return False   # do not suppress exceptions

with Timer():
    total = sum(range(1_000_000))
```

> **Cross-reference:** __enter__ and __exit__ are dunder methods, not keywords, exactly as __init__ is not a keyword. with is the keyword; the dunder methods are simply what a custom class must implement to cooperate with it.

### 10.3 as (Revisited for with)

The as here follows the same unifying idea introduced in Part 7: bind the value produced by __enter__() to a name for use inside the block. Not every context manager returns something meaningful, so as is optional - with lock: is perfectly valid if you never need to reference the lock object again inside the block.

## Part 11 — match / case: Structural Pattern Matching

match and case, introduced in Python 3.10, bring structural pattern matching to the language - a more expressive alternative to long if/elif chains when a value needs to be compared against several possible shapes, not just several possible equality checks. Both are soft keywords (see Part 1.12 and Part 14), meaning they are only special in this specific statement position.

### 11.1 Basic Pattern Matching

```python
def describe_day(day):
    match day:
        case "Saturday":
            return "Weekend"
        case "Sunday":
            return "Weekend"
        case _:
            return "Weekday"

print(describe_day("Sunday"))   # Weekend
```

### 11.2 Multiple Cases and the Wildcard _

Several literal values can be grouped in a single case using |. The wildcard pattern _ (itself a soft keyword) matches absolutely anything and is conventionally placed last, as the default case - exactly parallel to else at the end of an if/elif chain.

```python
def describe_day(day):
    match day:
        case "Saturday" | "Sunday":
            return "Weekend"
        case _:
            return "Weekday"
```

### 11.3 Literal Patterns and Variable Patterns

A literal pattern (like "Saturday" or 404) matches only that exact value. A variable pattern (a plain, lowercase, undotted name) always matches, and additionally binds the matched value to that name for use inside the case body.

```python
def http_status(code):
    match code:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case unknown:              # variable pattern - always matches
            return f"Unhandled code: {unknown}"
```

### 11.4 Sequence Patterns

A sequence pattern destructures lists or tuples by shape, optionally capturing individual elements or the remainder with *rest.

```python
def handle_command(parts):
    match parts:
        case ["move", direction]:
            return f"Moving {direction}"
        case ["attack", target, *extra]:
            return f"Attacking {target}, extras={extra}"
        case []:
            return "Empty command"
        case _:
            return "Unknown command"

print(handle_command(["move", "north"]))          # Moving north
print(handle_command(["attack", "dragon", "fire"])) # Attacking dragon, extras=['fire']
```

### 11.5 Mapping Patterns

A mapping pattern destructures dictionaries by key, and can use **rest to capture any remaining key-value pairs. Unlike sequence patterns, mapping patterns do not need to match every key - extra keys in the actual dictionary that are not mentioned in the pattern are simply ignored (unless **rest is used, in which case they are collected).

```python
def handle_event(event):
    match event:
        case {"type": "click", "x": x, "y": y}:
            return f"Click at ({x}, {y})"
        case {"type": "keypress", "key": key}:
            return f"Key pressed: {key}"
        case {"type": event_type}:
            return f"Unhandled event type: {event_type}"
```

### 11.6 Class Patterns

A class pattern matches an object based on its type and, optionally, the values of its attributes - extremely useful for dispatching logic based on the shape of custom objects, such as parsed AST nodes or domain model classes.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

def describe_point(point):
    match point:
        case Point(x=0, y=0):
            return "Origin"
        case Point(x=0, y=y):
            return f"On the y-axis at {y}"
        case Point(x=x, y=0):
            return f"On the x-axis at {x}"
        case Point():
            return "Somewhere else"
        case _:
            return "Not a point"
```

### 11.7 Guards

A guard is an additional if condition attached to a case, which must also be true for that case to match. Guards let you combine structural matching with arbitrary boolean logic.

```python
def classify(point):
    match point:
        case Point(x=x, y=y) if x == y:
            return "On the diagonal"
        case Point(x=x, y=y) if x > 0 and y > 0:
            return "First quadrant"
        case _:
            return "Other"
```

### 11.8 match/case vs if/elif/else

| Aspect | if/elif/else | match/case |
| --- | --- | --- |
| Best suited for | Arbitrary boolean conditions | Comparing a value's shape/structure against patterns |
| Destructuring support | Manual (via unpacking elsewhere) | Built directly into the pattern syntax |
| Readability with many literal branches | Can get repetitive (many == checks) | Very concise for many literal/shape branches |
| Availability | All Python versions | Python 3.10 and later only |

> **When to Choose Which:** match/case does not replace if/elif/else - it complements it. Simple boolean conditions are usually still clearer with if. Reach for match/case specifically when you are branching on the shape or type of a value.

## Part 12 — Type-Related Keywords: True, False, None

True, False, and None are unusual among Python keywords: unlike if or for, they do not control structure - they are literal values, and yet they are still full hard keywords, meaning they cannot be reassigned. (In Python 2, True, False, and None were merely built-in names and technically could be reassigned - Python 3 fixed this by promoting all three to genuine keywords.)

### 12.1 True and False

True and False are the two values of Python's bool type. bool is actually a subclass of int, which is why True behaves like 1 and False behaves like 0 in arithmetic contexts - a quirk worth knowing but rarely worth relying on in real code.

```python
print(type(True))     # <class 'bool'>
print(True + True)    # 2   - bool is a subclass of int
print(True == 1)      # True
print(False == 0)     # True
print(isinstance(True, int))   # True
```

#### Boolean Behavior and Truthiness

Every object in Python has an inherent truthiness, tested implicitly wherever a boolean is expected (if, while, and, or, not). bool(x) makes this explicit. As a rule: zero-like values (0, 0.0, empty string, empty list/dict/set/tuple) and None are falsy; almost everything else, including custom objects by default, is truthy.

| Value | bool(value) |
| --- | --- |
| 0, 0.0, 0j | False |
| "" (empty string) | False |
| [], (), {}, set() | False |
| None | False |
| Any non-zero number | True |
| Any non-empty string/collection | True |

### 12.2 None

None represents the deliberate absence of a value. It is the sole instance of NoneType - there is only ever one None object in the entire running program, which is precisely why identity comparison with is is the correct and idiomatic way to check for it (see section 5.5).

```python
def find_user(user_id, users):
    for user in users:
        if user["id"] == user_id:
            return user
    return None   # explicitly signals 'not found'

result = find_user(99, [])
if result is None:
    print("User not found")
```

#### Correct Comparisons

- Use if value is None: - correct, idiomatic, and unambiguous.

- Avoid if value == None: - works in practice but is not idiomatic and can theoretically be fooled by a custom __eq__.

- Avoid if not value: when you specifically mean 'is None', because that also catches 0, '', [], and other falsy-but-not-None values.

```python
count = 0
if not count:
    print("This runs even though count is 0, not None!")

if count is None:
    print("This correctly does NOT run, because 0 is not None")
```

### 12.3 Practical Applications

- None is the universal 'no default provided yet' sentinel for optional function parameters: def greet(name=None):.

- None is what a function implicitly returns when it has no explicit return statement (see section 4.2).

- True/False drive every conditional and loop-control decision throughout the entire language.

- In data science, None (and NaN in numeric contexts) represents missing data - a concept treated with great care in libraries like pandas.

### 13.1 del

del removes a name binding, a list item, a dictionary key, or a slice, immediately and directly - it is the explicit deletion keyword of the language. Once del removes a name, referencing that name again raises a NameError (for variables) or the appropriate lookup error (IndexError/KeyError) for collection items.

#### Deleting Variables

```python
x = 10
del x
print(x)   # NameError: name 'x' is not defined
```

#### Deleting List Items

```python
numbers = [10, 20, 30, 40]
del numbers[1]
print(numbers)   # [10, 30, 40]

del numbers[0:2]   # slice deletion
print(numbers)     # [40]
```

#### Deleting Dictionary Keys

```python
user = {"name": "Ada", "age": 34}
del user["age"]
print(user)   # {'name': 'Ada'}
```

#### del vs Assigning None

del removes the binding entirely, so the name no longer exists at all. Assigning None to a variable, by contrast, keeps the name bound - it simply points it at the None object instead of its previous value. These are meaningfully different: after x = None, x is still a perfectly valid, usable name; after del x, it is not.

#### Deleting Object Attributes

```python
class Config:
    def __init__(self):
        self.debug = True

cfg = Config()
del cfg.debug
print(cfg.debug)   # AttributeError: 'Config' object has no attribute 'debug'
```

### 14.1 Hard Keywords vs Soft Keywords, In Depth

A hard keyword is reserved unconditionally, in every position, for the entire program. You cannot use False, class, for, or any of the other 35 hard keywords as an identifier anywhere, under any circumstances - the parser rejects it immediately, regardless of context.

A soft keyword is only reserved in the exact grammatical position where it introduces a specific piece of syntax. Outside that position, it is a completely ordinary, available identifier. Python currently has four soft keywords: match, case, _, and type.

```python
import keyword
print(keyword.kwlist)       # the 35 hard keywords
print(keyword.softkwlist)   # ['_', 'case', 'match', 'type']

# 'match' used as an ordinary variable name - perfectly legal:
match = re.search(pattern, text)
if match:
    print(match.group())

# 'type' used as an ordinary variable name - also legal:
type = "admin"
print(type)
```

The reason Python could introduce match, case, and type without breaking millions of existing programs that already used those words as variable names (match is an extremely common variable name for regular-expression results, and type is a common field name) is precisely this soft-keyword design: the new syntax was added without hard-reserving the words everywhere.

#### The type Soft Keyword (Python 3.12+)

Python 3.12 introduced a dedicated statement for declaring type aliases using the soft keyword type, distinct from the long-standing type() built-in function used for type conversion/inspection (see the built-in Functions handbook).

```python
type Vector = list[float]     # 'type' soft keyword: declares a type alias

def scale(v: Vector, factor: float) -> Vector:
    return [x * factor for x in v]
```

### 15.9 try/except vs Conditional Logic

A common design question is whether to prevent an error with an if check beforehand ("Look Before You Leap", LBYL) or attempt the operation and handle failure afterward with try/except ("Easier to Ask Forgiveness than Permission", EAFP). Python culture generally favors EAFP, especially when the failure case is rare, because it avoids a redundant check-then-act race condition and often reads more directly.

```python
# LBYL style
if "age" in data:
    value = data["age"]
else:
    value = None

# EAFP style (more idiomatic Python)
try:
    value = data["age"]
except KeyError:
    value = None
```

This part draws a firm boundary around what a keyword actually is by contrasting it against every other kind of name a Python program can contain. Confusing these categories is one of the most common sources of both syntax errors and simple misunderstanding for people learning the language, so each distinction below includes a concrete, memorable example.

### 16.1 The Nine Categories, Defined

| Category | Definition | Example |

| Built-in function | Always-available callable object, no import needed | print(), len(), isinstance() |
| Method | A function attached to an object, called via dot notation | "abc".upper(), list.append() |
| Variable | A programmer-chosen name bound to a value | total, user_name, is_valid |

### 16.2 Keyword vs Built-in Function

Already introduced in Part 1, section 1.7 - repeated here for completeness of this comparison-focused part. print, len, and range are not keywords; they are ordinary callable objects that happen to be preloaded into every namespace.

```python
print(keyword.iskeyword("print"))   # False - it's a built-in function
print(keyword.iskeyword("len"))     # False - also a built-in function
print(keyword.iskeyword("if"))      # True  - genuine keyword
```

### 16.3 Keyword vs Method

```python
"hello".upper()   # 'upper' is a method - bound to the string object
if True:           # 'if' is a keyword - part of the grammar, never dotted
    pass
```

A reliable test: if a name is accessed with a dot after an object (object.name), it is a method or attribute, never a keyword. Keywords never follow a dot.

### 16.4 Keyword vs Variable

A variable name is entirely up to the programmer and can be reassigned freely; a keyword is fixed by the language and can never be assigned to at all. This was demonstrated concretely in Part 1, section 1.2.

### 16.5 Keyword vs Attribute

An attribute is a value stored on an object, accessed with dot notation (person.name, config.debug). Attributes are created by assignment inside classes or instances, entirely under programmer control, unlike keywords, which the language itself defines and protects.

### 16.6 Keyword vs Operator

Some logical operations in Python are spelled as keywords (and, or, not, in, is) rather than symbols, which sometimes leads learners to lump them in with symbolic operators like + or ==. Functionally, and/or/not/in/is do behave as operators (they combine operands into a result), but grammatically they are still classified as keywords because they are reserved words, not punctuation symbols. Symbolic operators like +, -, ==, and % are never reserved words - you could not make + do something else, but this is enforced by the tokenizer treating it as a fixed symbol, not by the reserved-word mechanism described in Part 1.

### 16.7 Keyword vs Dunder Method

Dunder methods (__init__, __str__, __len__, __enter__, __exit__, and dozens more) are special methods with a fixed naming convention that Python calls automatically in specific situations (object creation, string conversion, the len() call, entering/exiting a with block). They are not keywords - keyword.iskeyword("__init__") returns False - they are simply methods that Python's data model recognizes by name and calls automatically.

### 16.8 Keyword vs Module and Keyword vs Class

Modules (math, os, random) and classes (list, dict, or a programmer-defined Dog) are both ordinary objects that live in a namespace and can be imported, inspected, subclassed, and in principle reassigned (though reassigning built-in names is bad practice). class the keyword defines classes, but a specific class like Dog is not itself a keyword - only the word class used to introduce the definition is reserved.

Adding consistent indentation (4 spaces is the PEP 8 standard) satisfies the grammar.

#### 17.3 Incorrect if/elif/else Structure

```python
if x > 10:
    print("big")
elif:
    print("small")   # SyntaxError: missing condition
```

elif always requires a condition expression - it cannot be used as a bare fallback the way else can.

```python
if x > 10:
    print("big")
else:
    print("small")   # else needs no condition
```

Using else for the unconditional fallback, and reserving elif for cases that do need their own condition, fixes the structure.

#### 17.4 Misusing break Expecting It to Exit Outer Loops

```python
for i in range(3):
    for j in range(3):
        if j == 1:
            break   # only exits the INNER loop
    print("outer still running, i =", i)
```

break only terminates the single nearest enclosing loop; many beginners expect it to exit every level of nesting at once.

```python
done = False
for i in range(3):
    for j in range(3):
        if j == 1:
            done = True
            break
    if done:
        break   # now the outer loop exits too
```

A flag variable (or refactoring into a function that uses return) is required to escape multiple nested loops.

#### 17.5 Misusing continue Inside a while Loop (Skipping the Update)

```python
i = 0
while i < 5:
    if i == 2:
        continue   # BUG: skips i += 1, causing an infinite loop
    print(i)
    i += 1
```

continue jumps straight back to the condition check, skipping everything below it in the loop body - including the line that updates the loop variable, which here creates an infinite loop once i reaches 2.

```python
i = 0
while i < 5:
    if i == 2:
        i += 1
        continue
    print(i)
    i += 1
```

Updating the loop variable before continue (or restructuring the condition update) prevents the infinite loop.

#### 17.6 Incorrect try/except (Catching Nothing, or the Wrong Type)

```python
try:
    value = int("abc")
except TypeError:   # wrong exception type
    print("Failed")
```

int("abc") raises a ValueError, not a TypeError, so this except clause never actually matches, and the exception propagates uncaught.

```python
try:
    value = int("abc")
except ValueError:
    print("Failed")
```

Catching the exception type that is actually raised (verified by reading the traceback, or the exception hierarchy table in Part 6) fixes the handler.

#### 17.7 Incorrect return (Forgetting It Entirely)

```python
def add(a, b):
    total = a + b

result = add(2, 3)
print(result)   # None, not 5
```

Without an explicit return, a function always returns None, regardless of what values were computed inside it.

```python
def add(a, b):
    total = a + b
    return total
```

Adding the return statement sends the computed value back to the caller as intended.

#### 17.8 Incorrect yield (Mixing It With return of a Value)

```python
def broken_generator():
    yield 1
    return 2   # this value is NOT yielded; it becomes StopIteration's value
    yield 3     # unreachable
```

In a generator, return with a value ends iteration and attaches that value to the raised StopIteration exception rather than yielding it as a normal item - and any code after return never executes.

```python
def fixed_generator():
    yield 1
    yield 2
    yield 3
```

Using yield consistently for every value the generator should produce avoids the confusing early-termination behavior of return.

#### 17.9 Incorrect global (Forgetting It When Reassigning)

```python
counter = 0
def increment():
    counter += 1   # UnboundLocalError
increment()
```

Because counter is assigned to inside the function, Python treats it as a local variable for the entire function body, so the read on the right-hand side fails before the assignment ever happens.

```python
counter = 0
def increment():
    global counter
    counter += 1
increment()
```

Declaring global counter tells Python explicitly to use the module-level variable instead of creating a local one.

#### 17.10 Incorrect nonlocal (Using It Outside a Nested Function)

```python
def outer():
    nonlocal x   # SyntaxError: no binding for nonlocal 'x' found
    x = 5
```

nonlocal requires an existing variable of that name in an enclosing FUNCTION scope - it cannot be used in a top-level function with no enclosing function, and it cannot reach all the way to the module/global scope (that is global's job).

```python
def outer():
    x = 5
    def inner():
        nonlocal x
        x += 1
    inner()
    return x
```

nonlocal only works when there is a genuine enclosing function scope already defining that variable.

#### 17.11 Incorrect with (Ignoring That the Block Still Needs Valid Code)

```python
with open("data.txt") as file:
pass
content = file.read()   # IndentationError, and file is likely already unusable outside the block
```

The with block's body must be indented, and relying on the file object outside the with block is fragile - some context managers actively close/invalidate the resource on exit.

```python
with open("data.txt") as file:
    content = file.read()   # do all file work INSIDE the block
```

Keeping all resource-dependent operations inside the with block, correctly indented, is both syntactically required and best practice.

#### 17.12 Incorrect async/await (Forgetting to await a Coroutine)

```python
async def main():
    fetch_data()   # BUG: creates a coroutine object but never runs its body
asyncio.run(main())
```

Calling a coroutine function without await only constructs the coroutine object; Python even emits a 'coroutine was never awaited' warning, and none of the code inside fetch_data actually executes.

```python
async def main():
    await fetch_data()
asyncio.run(main())
```

Adding await drives the coroutine to completion as part of the event loop.

#### 17.13 Incorrect match/case (Accidentally Shadowing With a Variable Pattern)

```python
status = "error"
match status:
    case success:      # BUG: this is a VARIABLE pattern, matches ANYTHING
        print("matched:", success)
    case "error":
        print("This is unreachable!")
```

A lowercase, undotted name in a case pattern is a variable pattern that matches unconditionally - it is not compared against an existing variable of the same name, so it silently swallows every case beneath it.

```python
status = "error"
match status:
    case "success":     # a quoted literal pattern
        print("matched success")
    case "error":
        print("matched error")
```

Using quoted literal patterns (or dotted constant references like Color.RED) instead of bare names prevents the accidental catch-all.

### 18.3 Banking System

```python
class InsufficientFundsError(Exception):
    pass

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(f"{self.owner} has insufficient funds")
        self.balance -= amount
        return self.balance

account = BankAccount("Ada", 100)
try:
    account.withdraw(500)
except InsufficientFundsError as e:
    print("Transaction denied:", e)
finally:
    print("Current balance:", account.balance)
```

### 18.4 Data Validation

### 19.2 Intermediate Capstone

Build a small inventory system: a class Inventory with methods to add stock, remove stock (raising a custom exception if insufficient), and a generator method that yields items below a reorder threshold.

#### Solution

```python
class OutOfStockError(Exception):
    pass

class Inventory:
    def __init__(self):
        self.stock = {}

    def add(self, item, quantity):
        self.stock[item] = self.stock.get(item, 0) + quantity

    def remove(self, item, quantity):
        if item not in self.stock or self.stock[item] < quantity:
            raise OutOfStockError(f"Not enough {item} in stock")
        self.stock[item] -= quantity

    def low_stock(self, threshold):
        for item, quantity in self.stock.items():
            if quantity < threshold:
                yield item, quantity

inv = Inventory()
inv.add("widget", 5)
inv.add("gadget", 20)

try:
    inv.remove("widget", 10)
except OutOfStockError as e:
    print("Error:", e)

for item, qty in inv.low_stock(threshold=10):
    print(f"Reorder needed: {item} ({qty} left)")
```

Keywords exercised: class, def, self, raise, if, not, in, or, and, for, yield, try, except, as.

### 19.3 Advanced / Challenge Capstone

Build an asynchronous data pipeline that fetches data for several IDs concurrently, uses match/case to route each result based on its status field, and uses a generator to lazily filter the successful results before printing a final summary.

#### Solution

```python
import asyncio

async def fetch(item_id):
    await asyncio.sleep(0.2)
    if item_id % 3 == 0:
        return {"id": item_id, "status": "error", "detail": "timeout"}
    return {"id": item_id, "status": "ok", "value": item_id * 10}

def route_result(result):
    match result:
        case {"status": "ok", "value": value}:
            return ("success", value)
        case {"status": "error", "detail": detail}:
            return ("failure", detail)
        case _:
            return ("unknown", None)

def filter_successes(routed_results):
    for kind, payload in routed_results:
        if kind != "success":
            continue
        yield payload

async def main():
    ids = range(1, 7)
    results = await asyncio.gather(*(fetch(i) for i in ids))
    routed = [route_result(r) for r in results]
    successes = list(filter_successes(routed))
    print(f"Successful values: {successes}")
    print(f"Total success value: {sum(successes)}")

asyncio.run(main())
```

Keywords exercised: async, def, await, if, return, match, case, for, in, continue, yield, import (asyncio).
